## 1.Vibe Coding氛围编程

#### 代码优化提示词

请检查当前项目或本次改动，做一次代码简化：
1. 保持现有功能和对外行为不变；
2. 删除重复代码和不必要的抽象；
3. 优先复用已有函数、组件和工具；
4. 简化复杂的条件、状态和数据流；
5. 不要为了“看起来简洁”而牺牲可读性；
6. 修改后运行相关测试，并总结改了什么。

## 3. AI项目初体验：先想清楚Zesay是什么

### 3.1 你的实际使用过程

新建人物“女生A” → 添加你们的关系背景 → 导入聊天文字或截图 → 核对谁说了什么及发生时间 → 补充线下见面等现实事件 → 问Zesay你的具体困惑 → 查看有依据的解读与回复候选 → 你自行选择是否发送 → 后续补充实际发生的结果。

你可以问的不止是“她喜欢我吗”：

- 她明确说了什么，哪些只是我的猜测？
- 这段话可能表达什么态度或情绪，还有哪些合理解释？
- 我和她是不是有误解，哪里需要直接问清楚？
- 我的回复是不是太长、太急、没有接住她说的话？
- 我想约她见面、道歉、设定边界或自然结束话题，怎样说更合适？
- 和前一段时间相比，互动有什么可观察的变化？我还缺少什么信息？

结论应区分**事实、用户转述、推测和未知**。没有足够证据时给出澄清方式，比假装准确判断心情更有用。


### 3.2 第一版页面与技术方案

| 页面／区域 | 第一版职责 |
| --- | --- |
| 左侧人物列表 | 新建、搜索、切换、归档人物，清楚显示当前对象 |
| 中间咨询区 | 向Zesay提问，查看分析、回复候选与历史咨询 |
| 右侧背景／时间线 | 已确认资料、聊天批次、线下事件、证据与时间范围 |
| 导入区 | 粘贴文字、上传截图、预览校对后入库 |
| 设置区 | 模拟／真实模型状态、配置状态与数据管理 |

入门方案采用 **原生HTML/CSS/JavaScript + FastAPI + SQLite + uv**。前后端由同一服务提供页面和API，先做单用户本地应用。先普通完整响应，再按需要添加流式显示。这样能把注意力放在人物数据、证据和功能闭环上。

第一版不自动登录微信、不主动抓取社交账号、不替你发送消息。导入材料和复制建议就能验证核心价值。将来若要多设备、多人或公网访问，需要另做认证、权限与部署设计。

```text
我要在当前本地项目开发一个名为Zesay的私人沟通助手。请先确认工作目录，检查现有文件与环境，不要把本轮当成已经有完整应用。

产品背景：我会为每位女性建立一个独立人物对话空间，提交我与她的聊天记录，并补充现实中相处的事件。Zesay结合这些资料，帮助我理解互动、判断可能的态度和心情、练习网聊表达、生成回复建议，并指出我需要澄清的信息。人物资料必须隔离，不能把推测当事实。

本轮只做开发准备：
1. 检查当前目录、是否已有代码、Git状态和相关项目规则。保留已有有效内容与未提交修改。
2. 检查Python、uv及浏览器验证能力。说明已存在和缺少什么；按环境正式授权流程处理依赖配置，不重复安装无关工具。
3. 建立简短的AGENTS.md：中文解释、依据实际文件、按阶段开发、人物数据隔离、事实与推测区分、只记录真实验证结果。
4. 建立HANDOFF.md，记录目标、当前状态、环境、下一阶段与未验证事项。
5. 采用原生HTML/CSS/JavaScript、FastAPI、SQLite和uv作为初始方案；如现有项目不同，先报告实际差异。

我只用Codex桌面版操作。需要命令行工作时请在你可用的终端工具中执行并解释结果；需要我手动处理的事给出最少且明确的步骤。
本轮不要实现业务、不调用真实模型、不导入真实聊天资料。完成后给出实际工作目录、生成的文件与下一阶段起点。
```

## 阶段1：写出Zesay的需求和设计

**发送时机：**阶段0完成。此阶段把含糊的想法变成以后能验收的规则。

```text
继续当前Zesay项目。先读取AGENTS.md、HANDOFF.md和已有文件，本轮完成需求与设计，不实现业务代码。

产品目标：每个独立对话空间对应一位女性。我可以导入与她实际发生的聊天，记录线下相处事件，向Zesay询问如何理解互动、她可能的态度和情绪、我的表达问题以及合适的下一步。Zesay是沟通分析与练习助手，结论应有证据、允许未知，不承诺读懂内心。

请形成以下明确需求：
1. 人物档案：独立稳定ID、昵称、关系背景、用户希望改善的目标、已知偏好和边界。事实字段需来源，未知可留空。
2. 三类内容严格区分：真实聊天记录、我转述的线下事件、我与Zesay的咨询及AI草稿。草稿不能自动计为已发送消息。
3. 支持粘贴文字及上传聊天截图；导入要先预览、校对说话人、时间和内容，再确认保存。不确定识别不得悄悄变为事实。
4. 线下事件保留发生时间、录入时间、用户原文、可观察行为和用户自己的感受；用户转述不是独立核实过的客观记录。
5. 分析当前片段与一段时期的变化：明确表达、观察到的互动、可能解释、反向证据、未知信息、沟通建议。所有关键判断引用本人物证据。
6. 回复助手按我的目的给自然的候选话术，说明适用条件和可能误解；只提供草稿与复制，不自动外发。
7. 每个人物独立时间线、咨询历史与长期摘要；切换、查询、分析、导出都不能混入其他人物数据。
8. 单用户本地运行，以SQLite持久化；模型配置在后端。源码、测试与Git不能包含真实聊天及密钥。默认仅本机访问。
9. 支持编辑纠错、删除导入批次、删除人物、导出备份；明确相关摘要与分析如何失效、删除或重算。
10. 模拟模式和真实模式必须清楚区分，失败时不伪装成功。没有可用模型时仍能管理资料和测试页面。

技术约束：原生HTML/CSS/JavaScript、FastAPI、SQLite、uv；先非流式，后端同源托管前端。暂不做多人账号、公网部署、社交平台抓取或自动发消息。

请输出docs/PRD.md、docs/data-design.md、docs/api-design.md、docs/acceptance.md，并更新HANDOFF.md。
数据设计必须覆盖：人物、导入批次、消息、线下事件、咨询消息、分析快照、人物摘要版本、附件及其关联。区分发生时间与录入时间，保留原始材料、来源与修订信息；所有关系按人物归属校验。
为核心需求编号并配具体验收例，至少使用两个虚构人物验证隔离。列出合理的默认取舍与真正阻碍开发的未决问题，不用大量不影响实现的小问题中断。
完成后向我说明最小产品范围、页面结构和关键数据流，让我能阅读并修正设计。
```

**你来核对：**每位女性独立、文字和截图都能导入、线下经历可补充、回复只作建议、没有把“心情判断”写成确定结论。确实想改的产品方向，在做原型前告诉Codex。


## 阶段2：做看得见、点得动的界面原型

```text
继续Zesay，读取AGENTS.md、HANDOFF.md和docs中的已确认需求。本轮只实现可操作的界面原型，用虚构演示数据，不接真实模型和真实个人资料。

界面目标：一个安静、清晰、适合长时间阅读聊天记录的私人沟通工作台。品牌名Zesay。浅色中性背景、单一柔和强调色、清楚的文字层级，避免浮夸装饰和密集评分仪表。

桌面布局：左侧人物列表，中间与Zesay的咨询区，右侧当前人物背景及事件时间线。窄屏时改为可切换区域，当前人物姓名始终清楚可见。
必须演示：
1. 新建、搜索、切换、归档两位虚构人物；切换后背景、聊天材料、咨询全部更新。
2. 粘贴聊天和上传截图的入口；导入预览能编辑说话人、时间、文本，显示“尚未保存”。截图先用本地预览占位，不假装识别成功。
3. “记录线下事件”表单：发生日期、时间是否确定、事件内容、我的感受、备注。
4. 咨询区提供“分析这段对话”“帮我回复”“复盘见面”“最近互动变化”“我该问清什么”等入口。
5. 分析卡片展示：证据、可能解释、其他解释、未知、下一步；点击证据能看到原消息或事件。
6. 回复候选支持复制与标记“我实际发送了什么”，默认是未发送草稿。
7. 无人物、无记录、加载、分析失败、导入不确定、待校对、删除确认等状态。

把原型放在prototype目录，清楚标注“演示数据／模拟分析”。主要按钮必须能切换状态，不要只有静态截图。
在可用工具中启动本地预览，提供访问地址并说明如何在桌面版预览或普通浏览器打开。按两个虚构人物走一遍流程；实际无法进行浏览器操作时给出我需要验证的准确清单。
完成后更新HANDOFF.md，列出这轮只演示、尚未实现的功能。
```

## 阶段3：实现人物档案、线下事件和持久化

```text
继续Zesay，先读AGENTS.md、HANDOFF.md和docs。本轮把已确认原型接成真实本地应用，完成基础数据层；暂不调用真实模型。

技术：FastAPI托管API及前端，uv管理Python依赖，SQLite保存数据。数据库路径基于项目文件位置稳定解析，不随启动目录变化。使用可重复且不破坏已有数据的初始化或迁移方式。

完成：
1. 人物新建、列表、详情、编辑、归档和删除。昵称可重复，关联关系使用独立人物ID。
2. 每个人物的背景、沟通目标、已知偏好及其来源，不能把模型猜测自动写入事实档案。
3. 线下事件增删改查：发生时间可不完整，保留原文、我观察到的行为、我自己的感受、录入时间。缺失日期显示未知，不捏造。
4. 咨询消息与材料的存储结构，为后续聊天导入、模型分析预留经过设计的接口。
5. 页面替换假数据，保留明确的开发演示模式与测试数据来源。
6. 后端对人物、事件和相关子资源交叉校验归属；SQLite外键约束与事务正确启用。
7. 新增数据立即可读，刷新和重启后仍在；当前人物切换时清理无关页面状态，避免较早请求覆盖新人物。
8. 本地数据和附件放在明确的数据目录，加入gitignore；不上传到外部服务，不记录原始聊天到普通日志。

只使用两个虚构人物和虚构线下事件进行验证。测试按ID访问另一人物的子资源必须失败；删除或归档的行为与需求一致，归档不等于删除。
在你可用的终端中完成相关依赖与检查，启动服务并给出访问地址。更新README、数据设计、接口说明与HANDOFF，报告实际验证结果，不扩展到公网或多人账号。
```

## 阶段4：导入文本聊天，先校对再入库

```text
继续Zesay，读取现有规则、交接、需求和代码。本轮完成文本聊天导入，使用本地解析与人工校对；不需要真实模型调用。

需求：我在某位女性的人物空间里粘贴一段聊天记录，可能包含时间、我/对方昵称、连续多条消息、引用回复、表情或缺失时间。系统先生成候选记录，我核对后才保存。

请实现：
1. 导入批次绑定当前人物；保存原始粘贴文本和导入时间，提供预览、取消、确认。
2. 说话人映射：我、当前人物、未知。不能仅凭名字或位置强行判断，不允许其他人物内容被自动归入当前人物。
3. 每条记录保留原文、顺序、可识别时间、时间精度、来源批次、校对状态。没有日期不补成今天；引用内容不要重复当成新发送消息。
4. 可编辑、拆分、合并候选条目，必要时手工增加消息。解析失败时保留原文并允许手工整理，不吞掉资料。
5. 展示重复或重叠提示。完全重复导入不能悄悄多算；同一句“好的”发生在不同时刻也不能被误删。去重依据和不确定情况要可查看、可决定。
6. 用户确认后事务写入；批次详情可回到原始文本和对应消息。未确认批次不参与正式分析。
7. 支持纠错与撤销本批次，不能误删另一个批次复用的材料；关联旧分析按来源版本标记需要更新。
8. 咨询内容、AI候选回复及演示文本不会自动进入真实聊天记录。

请用虚构材料验证：正常格式、缺日期、未知说话人、多行消息、引用回复、重复导入、部分重叠、同句不同时间、取消导入、人物切换。
补齐相关API、界面、测试和说明。完成后给出可直接粘贴的虚构测试文本、实际验证结果，更新HANDOFF。
```

### 7.3 结构化分析怎样理解

下面是**教学样例**，用于帮助你检查输出形态，不代表真实人物的判断。实际代码可扩展字段，但必须保留证据与不确定性。

```json
{
  "overview": "她明确说今天加班，并提出明天再聊；仅凭这条记录不能判断长期好感。",
  "observations": [
    {
      "kind": "direct_statement",
      "text": "对方说明今天加班，建议明天继续。",
      "evidence_ids": ["msg_demo_01"]
    }
  ],
  "interpretations": [
    {
      "text": "当前更像暂时没有聊天精力，但真实心情仍未知。",
      "support_level": "medium",
      "reason": "有明确忙碌说明，但只有一条消息。",
      "evidence_ids": ["msg_demo_01"],
      "alternatives": ["也可能希望暂时结束当前话题，需结合后续交流。"]
    }
  ],
  "unknowns": ["她当时的真实情绪", "最近是否持续出现类似情况"],
  "next_step": "可以简短回应并结束本轮，明天再根据实际情况交流。"
}
```

真正应用中的 `msg_demo_01` 要换成数据库中本次输入的有效记录。结构合法只说明能解析，不等于分析正确，还要验证证据归属与语义是否支持结论。

## 6. 桌面版里的项目规则、Memory和Skill

### 6.1 项目规则写什么

`AGENTS.md` 适合记录Zesay的稳定约定：技术栈、目录、启动检查、人物隔离、资料分类、密钥位置和分析证据规则。具体的产品行为写PRD，近期进度写HANDOFF，不把全部聊天历史堆在规则里。

### 6.2 两种Memory不要混

| 记忆 | 内容 | 放在哪里 |
| --- | --- | --- |

### 6.3 Skill在本项目里怎样使用

Skill是可复用的工作流程或技能包，可以帮助做界面检查、接口验证、资料解析等。桌面版若提供技能列表或安装入口，先查看当前已经识别的技能；不同安装方式和版本的目录机制不必相同，不照搬Claude的路径。

```text
请先检查当前环境是否有适用于前端审查或界面设计的已识别技能。
如果有，读取对应说明，再检查Zesay人物切换、聊天导入校对、证据引用和移动端界面。
如果没有，不要假装加载；直接按上述功能清单检查即可。
保持数据隔离和业务行为不变，修复有依据的问题并记录验证结果。
```

Skill能减少遗漏，不代表创建一个技能就自动拥有OCR、模型权限或浏览器工具。技能中的操作也要服从本次任务范围和实际权限。

```text
请总结本次Zesay开发中已经验证解决的问题，整理为可复用的工作手册。
每项包含适用条件、现象、根因、修复方法和验证方式；删掉未证实猜测。
人物隔离、截图校对、证据校验、异步切换和摘要失效分别归类。
通用步骤与当前项目实现示例分开，示例不能包含真实人物材料和密钥。
如果当前客户端支持技能包，按它实际支持的方式组织；否则保存为docs中的普通手册，并在AGENTS.md中说明何时参考。
```
