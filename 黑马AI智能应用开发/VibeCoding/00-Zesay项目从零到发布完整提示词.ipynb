## 3. AI项目初体验：先想清楚Zesay是什么

### 3.1 最终目的与实际使用过程

**Zesay 是一个私人沟通助手：结合你与某位女性的聊天和现实经历，帮助你理解互动、改善自己的表达，并决定下一步怎么做。** 每个对话空间对应一位女性。记录、检索、分析、回复和练习都服务于这个目的；提出问题只是获取必要信息的方法，不是独立追求的目标。

新建人物 → 补充背景 → 导入或手动添加聊天、事件 → 核对保存 → 提出困惑或请求建议 → 查看有依据的分析 → 需要时补充重要背景 → 自行选择行动 → 记录真实结果 → 随时回看、修改和复盘。

你可以问：“她明确表达了什么？”“哪些只是我的猜测？”“我的回复有没有让人难接话？”“我想邀约、道歉或表达边界，怎样说合适？”“最近的互动有什么有证据支持的变化？”

结论区分**直接记录、用户转述、用户感受、推测和未知**。你没有主动删除的已录入历史持续保留，可以手动新增和修改。资料足够就直接帮助你；缺少的背景确实可能改变建议时，再问少量具体问题。

### 3.1.2 澄清的尺度：帮助决策，不为了提问而提问

资料足够时直接给有用的分析或建议，问题数量可以为 **0**。先利用当前人物已有资料和有效回答；只有答案可能明显改变本次判断或下一步时，才提问。每轮优先 1 个关键问题，最多 3 个，不凑数量、不预设动机，也不为了填满档案追问私人细节。

例如你问“见面后为什么交流变少”，但没有提供见面经过，可以问“你说上次见面后交流变少，那次有没有留下没说清楚的事情？”并说明它与判断的关系。如果你只是想回复“今天加班，明天再聊”，通常可直接给一句自然回应，不必追问整段关系史。

问题要说明依据和可能影响；用户可回答、不知道、跳过、不再询问，并能修改或删除补充说明。有效答案复用，同一事项不因刷新、重启或换页重复询问。只有确有新证据冲突时才说明变化并澄清；用户关闭的同一问题不自动重开。

### 3.2 第一版页面与技术方案

| 页面／区域 | 第一版职责 |
| --- | --- |
| 左侧人物列表 | 新建、搜索、切换、归档人物，清楚显示当前对象 |
| 中间咨询区 | 自由咨询、分析、回复、历史咨询；有必要才显示可跳过的澄清问题 |
| 右侧背景／时间线 | 已确认资料、聊天与事件、完整历史搜索、来源和修订查看 |
| 资料入口 | 粘贴文字、上传截图、独立添加聊天和事件、预览校对、编辑已存记录 |
| 设置区 | 模拟／真实模型状态、配置状态与数据管理 |

入门方案采用 **Vite + React（JavaScript）+ FastAPI + SQLite + uv**。前端由 Vite 开发服务器运行，后端由 FastAPI 提供 API；`vite.config.js` 在开发时把 `/api` 代理到后端，本地发布时由 FastAPI 托管构建后的 `frontend/dist`。先做单用户本地应用，先普通完整响应，再按需要添加流式显示。这样能把注意力放在人物数据、证据和功能闭环上。

建议目录边界如下，已有项目不符合时先检查再迁移，不要盲目覆盖：

```text
Zesay/
├─ frontend/              # Vite + React（JavaScript）
│  ├─ src/
│  ├─ public/
│  ├─ package.json
│  ├─ vite.config.js
│  └─ index.html
├─ backend/               # FastAPI
├─ data/                  # 本地数据库和附件
├─ docs/
├─ tests/
├─ AGENTS.md
├─ HANDOFF.md             # Markdown 交接文档
├─ README.md
└─ .gitignore
```

前端启动 `npm install`、`npm run dev`

后端启动 `uv sync --default-index https://pypi.tuna.tsinghua.edu.cn/simple`

`uv run fastapi dev`。

### 7.3 结构化分析怎样理解

下面是**教学样例**，用于帮助你检查输出形态，不代表真实人物的判断。实际代码可扩展字段，但必须保留证据与不确定性。

```json
{
  "overview": "她明确说今天加班，并提出明天再聊；仅凭这条记录不能判断长期好感。",
  "observations": [
    {
      "kind": "direct_statement",
      "text": "对方说明今天加班，建议明天继续。",
      "evidence_ids": ["msg_demo_01"]
    }
  ],
  "interpretations": [
    {
      "text": "当前更像暂时没有聊天精力，但真实心情仍未知。",
      "support_level": "medium",
      "reason": "有明确忙碌说明，但只有一条消息。",
      "evidence_ids": ["msg_demo_01"],
      "alternatives": ["也可能希望暂时结束当前话题，需结合后续交流。"]
    }
  ],
  "unknowns": ["她当时的真实情绪", "最近是否持续出现类似情况"],
  "next_step": "可以简短回应并结束本轮，明天再根据实际情况交流。",
  "clarification_questions": []
}
```

真正应用中的 `msg_demo_01` 要换成数据库中本次输入的有效记录。结构合法只说明能解析，不等于分析正确，还要验证证据归属与语义是否支持结论。

这里 `clarification_questions: []` 是正常结果：本次目标只是自然接话，已有资料足够，不需要为了知道她“真正的心情”继续追问。未知信息可以保留在 `unknowns`。

若用户明确说“见面后交流减少”却未提供见面经过，可有一项候选问题：

```json
{
  "question": "你说上次见面后交流变少，那次有没有留下没说清楚的事情？",
  "reason_to_ask": "当前补充说明提到了见面后的变化，但没有见面经过。",
  "decision_impact": "有具体误会时可考虑澄清；没有时，应保留其他解释。",
  "evidence_ids": ["report_demo_02"]
}
```

`report_demo_02` 必须是本次实际输入的用户补充记录。候选先经过字段、人物归属和问题历史状态检查，应用再分配ID并保存；不是模型列出问题就全部强制展示。回答是否改变建议、有无重复或诱导，仍需人工评审。

```text
案例1：跨年历史与手动修订
人物A小岚。手动补录2024-03-09的消息“我：周五见”，后来发现日期录错，改为2024-03-08并备注“校正日期”。另加一条真实新事件“她后来把见面改到下周”，保留两者。
预期：无文件也能录入；同一消息ID有两个修订，默认仅最新版参与分析；真实新进展是新事件。关键词、日期可定位旧记录，不被今天的列表覆盖。
Codex用虚构数据扩展到至少三页、跨多个年份并超过模型窗口，加入一张可追溯虚构截图。生成摘要、归档、重启、迁移及备份恢复后比对ID/版本/数量与附件，不只检查首页看得见。

案例2：充分资料不追问
已确认消息“今天加班，明天再聊”。我的目标：“给我一句简短自然的回应，不继续展开”。
预期：直接给合适建议，clarification_questions为空。可以承认情绪未知，但不为此追问工作、家庭或整段关系史。

案例3：缺失背景值得问
用户补充：“上次见面后交流变少了，我不知道要不要解释。”没有见面经过。
预期：先给有限建议，必要时问一个关于那次是否有没说清事情的问题，解释原因；不暗示她一定生气或另有对象。
用户回答：“她说最近备考，希望少发消息。”明确这是用户转述。
预期：复用此回答，建议关注已表达的空间需求；不能自动伪造她的聊天原话。重启再问时不重复询问相同事项。

案例4：不强求回答和去重
对案例3分别操作“不知道”“跳过”“不再询问”，刷新并再次提交相同目标与资料。
预期：不知道无新线索不追问；跳过不弹回；关闭的同一事项不改写措辞再问，只有用户主动重开。仍可给有限建议。

案例5：感受、修改和删除
用户回答：“我觉得她不开心，但她没有这样说。”
预期：保存为用户感受，不升级为对方明确表达。改答“后来想想，也许只是累”后旧分析标记待更新；删除该补充后，修订/摘要/问题/缓存不能继续暴露或引用原回答。
切到人物B，不能看到A的任何问题、回答或搜索结果。普通保存/换页不额外调用模型，回答一次不触发自动连续追问。

案例6：完整恢复
备份包含虚构聊天、事件、附件、原文、修订、咨询、草稿、分析及已回答/跳过/关闭的问题。恢复到独立测试目录，按清单比较ID、版本、数量和附件校验值，再查最早记录和问题状态。
预期：完整恢复，不覆盖当前数据，不把已关闭问题重置为待回答；模型密钥不包含在备份里。
```

这些用例检查实际行为。澄清样例的语义好坏由人工评审，不能只靠“返回了1个问题”判定智能；有帮助且不打扰才算达标。

## 4. Zesay建好后，我应该怎样问它

下面的提示词发给 **Zesay某个人物空间里的咨询框**，不是开发线程。先确认你选中了正确人物，材料已经校对保存。

### 4.1 分析一段网聊

```text
请只分析我本次选中的聊天和该人物已确认的相关背景。
我的困惑是：[例如她最近回复变短，我不知道是忙还是不想聊]。
请先列直接证据，再说明可能的态度或情绪、其他合理解释和目前无法确定的事。
也请指出我自己的表达有没有让人难接话的地方。
最后给一个最自然的下一步；如果更适合暂时不发消息，也可以直接说。
不要给好感百分比，每个重要判断让我能看到对应原句或事件。先利用已有背景和有效回答；资料够用直接分析，只有答案可能改变判断或下一步时才问最关键的问题，不为了提问而提问。
```

### 4.2 结合线下经历理解互动

```text
我补充了一条线下事件，请与事件前后的相关聊天一起看。
我观察到的行为：[具体说了什么、做了什么]。
我当时的感受：[你的感受]。
我想确认的是：[具体困惑]。
请区分我的观察和我的解释，不把一件小事直接推断成她整体的想法。
如果线上线下信息不同，指出具体冲突和可能解释；先看已存资料能否解释，只有必要时再向我澄清，或建议我怎样自然与她问清楚。
```


### 4.3 帮我回复，尽量像我本人

```text
请针对我选中的最后一条消息帮我回复。
我的目的：[接话/邀约/表达好感/道歉/澄清/结束话题/表达边界]。
我本来想说：[自己的草稿]。
我希望的风格：[简短自然、略带幽默等]。
请给两到三条有实际差异的候选；一个简单回复足够时不必凑数。尽量保留我的表达习惯，不虚构经历。
每条说明适合什么情况、可能有什么误解，并推荐你认为最稳妥的一条及原因。
资料够用就直接给候选；只有缺失信息会影响建议时才问，复用已有回答。我不知道或跳过时仍可给有限建议，不编造缺失事实；现在不适合继续发时直接说明。
```

### 4.4 复盘最近一段时间

```text
请复盘我选择的这段时间内的互动，说明实际覆盖了哪些材料。
重点看双方明确表达、主动发起的话题、具体约定、未解决的误会和边界。
和前一段有可比材料时再比较变化；样本不足就不要强行做趋势判断。
区分已经确认的事实和仍需问清的问题，并给我下一阶段可以改进的一个沟通习惯。
```

### 4.5 教我思路，不只给答案

```text
这次先不要替我写完回复。请告诉我应该接住她话里的哪一部分、可以表达什么真实信息、什么地方别过度解释。
让我先写一版，再帮我逐句点评并给一版更自然的修改。
如果进行角色练习，请明确这是模拟，不代表她实际会如何回应。
```

### 4.6 回看与纠正长期记录

先在当前人物空间用日期、关键词找到记录。手动新增、编辑和查看修订使用对应界面；在咨询框表达新信息，默认只是补充说明，只有你确认后才转成事件或聊天。

```text
我刚在资料页面纠正了[记录/发生日期/说话人]，请使用最新有效版本重新看[具体问题]。
说明哪些建议因此改变、哪些没有改变；不要把旧版本和新版当成两件事。
若相关旧记录不在这次分析范围，请说明实际覆盖范围，不要声称已经看过全部历史。
```

### 4.7 对必要的问题补充或暂不回答

```text
关于你刚才提出的[问题]，我补充：[具体信息]。
我直接观察到的是：[内容/未知]；我的感受或猜测是：[内容/没有]。
请先把这作为我的补充说明，结合现有资料更新建议，告诉我什么发生了变化。
如果没有值得继续问的信息，就直接给当前最有用的下一步。
```

不知道可直接选择“不知道”，暂时不想答选“跳过”，不想继续讨论同一事项选“不再询问”。这些按钮负责保存状态，不能仅凭模型回复“我记住了”判定操作成功。

## 6. 桌面版里的项目规则、Memory和Skill

### 6.1 项目规则写什么

`AGENTS.md` 适合记录Zesay的稳定约定：技术栈、目录、启动检查、人物隔离、资料分类、密钥位置、完整历史不自动清理、手动修订与删除规则、分析证据及有必要才澄清的约定。具体的产品行为写PRD，近期进度写HANDOFF，不把全部聊天历史堆在规则里。

桌面版加载规则的范围受客户端与环境影响，实际应让Codex说明读过哪些文件。自然语言规则能指导行为，但不等于数据库权限或操作系统的访问控制。

### 6.2 两种Memory不要混

| 记忆 | 内容 | 放在哪里 |
| --- | --- | --- |
| 开发交接 | Zesay做到哪一步、怎么运行、已知问题 | 项目说明、HANDOFF和验证报告 |
| Zesay人物资料与分析上下文 | 完整聊天/事件/附件/修订、用户补充、问题及回答状态；分析选取相关有效资料 | 数据库与附件长期保存；摘要保留来源且可失效，不代替原始历史 |

不要把女生的私人资料写入Codex全局偏好，也不要把Codex“我记住了”的回复当成Zesay已经持久化成功。数据保存要通过刷新、重启、查询和备份验证。


### 6.3 Skill在本项目里怎样使用

Skill是可复用的工作流程或技能包，可以帮助做界面检查、接口验证、资料解析等。桌面版若提供技能列表或安装入口，先查看当前已经识别的技能；不同安装方式和版本的目录机制不必相同，不照搬Claude的路径。

**使用已有技能的提示词：**

```text
请先检查当前环境是否有适用于前端审查或界面设计的已识别技能。
如果有，读取对应说明，再检查Zesay人物切换、聊天导入校对、证据引用和移动端界面。
如果没有，不要假装加载；直接按上述功能清单检查即可。
保持数据隔离和业务行为不变，修复有依据的问题并记录验证结果。
```

Skill能减少遗漏，不代表创建一个技能就自动拥有OCR、模型权限或浏览器工具。技能中的操作也要服从本次任务范围和实际权限。

### 6.4 把Zesay开发经验沉淀下来

```text
请总结本次Zesay开发中已经验证解决的问题，整理为可复用的工作手册。
每项包含适用条件、现象、根因、修复方法和验证方式；删掉未证实猜测。
人物隔离、手动维护与完整历史、截图校对、证据校验、异步切换、摘要失效以及必要澄清与去重分别归类。
通用步骤与当前项目实现示例分开，示例不能包含真实人物材料和密钥。
如果当前客户端支持技能包，按它实际支持的方式组织；否则保存为docs中的普通手册，并在AGENTS.md中说明何时参考。
```

这里的“沉淀／蒸馏”是整理经验，不是训练模型。先有真实修复和验证，再写手册，避免把一次偶然成功写成通用规则。


## 7. 每阶段结束前的固定交接提示词

```text
请整理当前Zesay阶段的结果并更新HANDOFF.md。
必须包含：
1. 本次目标和验收标准。
2. 已修改的文件、已完成的具体行为。
3. 实际运行过的检查、结果及未验证内容。
4. 当前启动方法、访问地址、数据库与附件位置；敏感配置只写变量名。
5. 关键产品决定、数据规则和剩余问题；涉及记录维护时写明保存/修订/删除与备份范围，涉及澄清时写明触发点、状态复用、0问题路径和未通过样例。
6. 下一阶段要读的文件及第一项可执行动作。
区分“实现了接口”“模拟验证通过”“真实模型验证通过”，不要混写。
核对文档与实际代码一致，不重复粘贴全部聊天历史。
```

交接文件可以让你隔天换线程继续，但不是“永久记忆”。重要版本、私人资料与附件仍需要适当备份；恢复时先核对现状。
