# 04 · 数据持久化

## 数据库


### 理解补充：数据库、引擎和查看工具

| 名称 | 在本章中的含义 |
| --- | --- |
| 数据库文件 | 保存表结构与记录的文件，如 `mydb.db` |
| SQLite 引擎 | 负责解析 SQL、读写数据与管理事务 |

| 数据库图形工具 | 帮助查看表、编辑记录、执行 SQL 的应用 |

### 数据类型

sqlite五大数据类型：

1. NULL：空值，没有任何数据

2. Integer：整数

3. Real：浮点数，也就是小数

4. text：文本，也叫做字符串

5. blob：二进制大文件，比如图片，音频，视频等

数据类型的注意点：

1. 如果没有存储数据，默认就是NULL

2. 数据类型不是强制性的，是亲和性。如果数据类型是Integer，但是实际上存入的是文本"abc"，sqlite会先尝试把"abc"转成整数，转换失败还是按照Text文本类型存储，但是最好让数据和类型保持一致。

## 持久化案例


### 案例一：

给AI女友项目添加以下业务：

1. 登录、注册业务

1. 聊天记录持久化存储业务

> 温馨提示：
> 
> 登录用户名 + 女友的性格 + 女友的名字一致，才会加载历史聊天记录，否则不加载


### 项目拆解：按用户和角色隔离会话

本题要求用户名、角色性格、角色名字都匹配才加载历史。可以把它转换成明确的数据关系：

```text
users：账号ID、用户名、密码哈希
conversations：会话ID、用户ID、角色性格、角色名字
messages：消息ID、会话ID、角色、内容、创建时间
```

登录后由服务端验证会话身份，用可信用户 ID 查询；不能仅相信前端提交的用户名，就返回该用户名的历史消息。

**跟练提示词：**

```text
为现有角色聊天项目增加注册、登录和SQLite聊天记录存储。
先检查现有结构并给出数据表、接口和迁移方案。
同一已登录用户、同一角色性格、同一角色名才恢复同一会话历史。
用户身份从经过验证的会话取得，查询要限制到该用户。
密码使用成熟库进行哈希与验证，SQL使用参数绑定。
每次成功交互保存用户和助手消息；明确失败请求的保存规则。
刷新页面和重启后端后重新查询历史，按时间和稳定ID排序。
保留已有数据，完成后说明如何验证不同账号和角色不会串记录。
```

验收：同账号同角色能恢复；改名字或性格不加载原会话；另一账号不能看到历史；重启后数据仍在；失败请求不会伪装成成功回答。
