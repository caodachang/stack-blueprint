# 05 · 工程化开发 Spec Coding


### 工程化开发（Spec Coding / SDD / 规范驱动开发）

流程：

先撰写规范文档（需求、设计、API等）→ AI严格按照文档生成代码 → 逐条验收。

核心理念：

在编写代码之前，先定好页面，开发的规则，最后让AI逐条实现。

优势：

- 代码质量稳定，符合预期

- 便于团队协作，文档即契约

- 可追溯、可维护

- Token消耗可控，避免无效对话

弊端：

- 前期投入较大，需要编写详细文档

- 对开发者的业务分析和文档能力有要求

- AI生成的文档，需人工校正

适用场景：

- 企业级项目

- 产品有明确需求，周期较长

- 需要多人协作的团队项目


### 混合模式（实际项目中的最佳实践）

在实际开发中，我们不会完全依赖某一种模式，而是根据阶段灵活切换：

| 项目阶段 | 推荐模式 | 理由 |
| --- | --- | --- |
| 原型设计 | Vibecoding | 快速出界面，让客户或团队看到效果，收集反馈。 |
| 设计阶段 | SDD（规范驱动） | 确定数据库设计、API接口、模块边界，确保架构不返工。 |
| 传统业务开发 | SDD为主 | 业务逻辑相对固定，按规范生成更可靠。 |
| AI模型/算法探索 | 混合（规则+探索） | Prompt需要大量实验，可先Vibe再固化 |

## 工程化开发Spec Coding/SDD

### MVP 版的企业级的开发流程

#### 先在 README.md 中生成项目目录结构：

```text
项目名/
├─ backend/        - 后端代码
├─ frontend/       - 前端代码（参考 HTML 原型实现 React 页面和交互）
├─ docs/spec/MVP/  - 需求、设计和接口文档，用户故事与验收标准文档，原型HTML
├─ README.md       - 给人看的
├─ .gitignore
└─ AGENTS.md       - 给 AI Agent 看的
```

需求分析 → 原型图设计 → 业务设计 → 编码 → 测试 → 运行

- 需求分析：明确这个项目，你要做什么？（注重功能） --- 罗列功能模块，给每一个功能模块写对应的需求文档

    根据用户故事（作为什么用户需要什么功能），修改需求文档。注意不是所有的要求我都满足，是把合理的要求进行满足

- 原型图设计：设计前端页面，你要把项目写成什么样子？（注重页面） 

    可借助于其他 AI 去设计，一旦设计完毕，以后就按照这个方式进行开发

- 业务设计：设计开发中的细节

- 正式开发

- 测试：MVP 整体联调、回归测试与验收

- 运行：本地运行与服务器部署

| 阶段 | 做什么 | 核心产出 |
| --- | --- | --- |
| 需求分析 | 明确目标、功能列表 | 需求文档（PRD）、用户故事 |
| 原型图设计 | 可视化页面和交互（效果） | HTML原型文件 |
| 业务设计 | 技术选型、数据库、API设计 | AGENTS.md、数据库设计、API设计、业务术语 |
| 编码 | 逐功能实现代码 | 完整项目代码 |
| 测试 | MVP 整体联调、回归测试与验收 | 测试用例、测试报告 |
| 运行 | 验证本地启动，并准备服务器部署 | 本地运行说明、部署文档 |

### 阶段一：该版本的需求分析

目标：

搞清楚“为谁做、解决什么问题、做什么、不做什么”，明确业务目标、功能边界，以及怎样判断需求已完成。

产出：

- 产品需求文档（PRD.md）：包含目标用户、业务目标、功能范围、优先级、约束和非功能需求。
- 用户故事与验收标准（user-stories.md）：每条用户故事附上可验证的验收标准。

PRD和用户故事的区别：

- 需求文档PRD：产品视角，描述"系统要做什么"

- 用户故事：用户视角，描述"用户想要什么"

完成后要根据用户故事再一次修改需求文档

将两个文件放在项目根目录的 /docs/spec/MVP 下 

### 阶段二：该版本的原型图设计

目标：

在写代码之前，先用HTML画出所有页面，确认最终的效果和交互过程

为什么这一步很重要？

- 让需求方（或你自己）提前看到页面长什么样

- 前后端对页面功能达成一致理解

- 编码前发现问题，成本远低于编码后返工

产出：

所有页面的可交互的HTML原型文件，放在项目根目录的 /docs/spec/MVP/prototype 下 

### 阶段三：该版本的业务设计

目标：

将需求和原型转化为可执行的技术方案。

产出：

1. 项目宪法（根目录下的 AGENTS.md 还有MVP下的 AGENTS.md）

1. 业务术语表（glossary.md）

1. 数据库设计（db_design.md）

1. 接口设计（api_design.md）

#### 编写项目宪法（AGENTS.md）

##### 根目录下的 AGENTS.md

交互提示词示例：

```text
请为"项目名"项目生成根目录下的 AGENTS.md，内容包括：

技术栈：
- 前端：vite + react框架（参考 HTML 原型实现 React 页面和交互）
- 后端：Python + FastAPI
- 数据库：SQLite
- AI模型：DeepSeek API （模型版本 deepseek-v4-pro，接口兼容OpenAI调用格式）
- 依赖管理：uv + pnpm

目录结构：

项目名/
├─ backend/        - 后端代码
├─ frontend/       - 前端代码（参考 HTML 原型实现 React 页面和交互）
├─ docs/spec/MVP/       - 需求、设计和接口文档，用户故事与验收标准文档，原型HTML
├─ tests/MVP/           - 可运行测试代码
├─ README.md       - 给人看的
├─ .gitignore
└─ AGENTS.md       - 给 AI Agent 看的

接口响应格式：统一为 {code, message, data}

请生成完整的AGENTS.md文件，放在项目根目录下
```

##### MVP 目录下的 AGENTS.md

交互提示词示例：

#### API接口设计

交互提示词示例：

```text
请根据所有的需求文档、用户故事、数据库设计文档，生成"项目名"的所有的接口，并把接口文档保存到开发文档文件夹当中

需要注意的是：
1. 接口名不要太复杂，简单一些即可
2. 仅设计当前 MVP 范围内的接口，覆盖原型中的实际操作，不要遗漏
3. 明确请求方法、路径、参数及校验、成功与错误响应、HTTP 状态码、业务 code、认证方式和数据访问权限

请生成完整的API设计文档，放在项目根目录的 /docs/spec/MVP 下
```

### 阶段四：编码

#### SDD 开发单元拆解

交互提示词示例：

```text
请阅读项目根目录下的 AGENTS.md、docs/spec/MVP 下的所有文档，同时检查现有代码和已完成的开发单元；项目尚未初始化时，按新项目处理。把 MVP 阶段拆成适合在 SDD 开发中逐次交给 AI 完成的开发单元。

一个开发单元必须围绕一个明确目标，能够在一次开发中完成必要的前端、后端、数据库、AI 调用、权限处理、异常状态和测试，并且完成后可以独立运行、演示和验收。

不要只按页面或 API 拆分，也不要把多个无关功能合并。如果一个单元过大，请继续拆分；如果多个小任务必须一起才能形成完整功能，请合并。请按照前置依赖安排顺序。

输出以下内容：

1. 开发单元编号和名称；
2. 主要目标；
3. 前置依赖；
4. 本次包含的范围；
5. 预计涉及的页面、接口和数据；
6. 验收标准；
7. 测试方式；
8. 优先级和复杂度。

请使用具体名称，不要使用“功能一”“功能二”等占位名称，并指出哪些内容属于基础能力、业务功能或公共能力。

生成 SDD-开发单元拆解.md 文件放在项目根目录的 /docs/spec/MVP 下
```

#### 添加文档清单

将 SDD-开发单元拆解.md 添加到 /docs/spec/MVP/AGENTS.md 的"MVP 版本全部产出文档清单"中

#### 为开发单元生成执行提示词

交互提示词示例：

```text
请阅读项目根目录下的 AGENTS.md、docs/spec/MVP 下的所有文档，同时检查现有代码和已完成的开发单元；项目尚未初始化时，按新项目处理。为指定的开发单元生成一份可以直接交给编程 AI 执行的开发提示词。

指定开发单元：
【填写开发单元编号和名称】

生成的提示词必须包含：

1. 开发目标；
2. 本次必须完成的内容；
3. 本次明确不处理的内容；
4. 需要先阅读的项目文件；
5. 涉及的前端、后端、数据库和 AI 服务；
6. 接口、数据保存和权限要求；
7. 加载、空数据、失败、超时和重复提交处理；
8. 测试要求：要求执行者将测试代码放在项目根目录的 tests/MVP/ 下，运行相关测试并修复问题；
9. 验收标准：明确本开发单元满足哪些可验证的条件才算完成；
10. 汇报要求：要求执行者在完成后汇报实现内容、测试命令与实际结果、验收情况，以及未完成或未验证的事项。

请严格限制在指定开发单元范围内。如果发现文档冲突、前置依赖未完成或需求信息不足，请在提示词中明确列出，不要擅自扩大范围。

每个开发单元都应尽量形成完整闭环：用户操作、前端交互、后端处理、数据保存、结果展示、异常处理和测试。
只生成开发提示词，不要直接写代码。
```

### 阶段五：MVP 整体联调、回归测试与验收

目标：

验证各开发单元组合后是否按预期工作，确保当前版本符合需求文档、用户故事和验收标准。

交互提示词示例：

```text
请根据 docs/spec/MVP 下的文档，对当前版本进行整体联调、回归测试与验收。

运行已有测试，并补充缺失的跨功能流程测试；
新增测试代码放在项目根目录的 tests/MVP/ 下。

使用独立测试环境或测试数据库，只清理本次测试创建的数据。

修复发现的问题，并重新运行受影响的测试。
对未实现的需求、文档冲突和无法验证的事项，列出清单及原因；
不要自行增加 MVP 范围之外的功能。

最后汇报测试命令、实际结果，以及各项验收标准的满足情况。
```


### 阶段六：运行与部署

目标：

验证项目能够按文档在本地运行，并具备部署到服务器的说明和条件。

交互提示词示例：

```text
请根据项目实际配置，完善 README.md 中的运行与部署说明，包括：

1. 依赖安装；
2. 环境变量配置；
3. 数据库初始化；
4. 本地前后端启动命令、访问地址和停止方式；
5. 服务器部署所需的环境、启动命令、访问配置和停止或重启方式。

实际验证本地启动流程；如果具备服务器环境，也验证服务器部署流程。
如发现启动或部署问题，请排查并修复。
将需要长期遵守的开发注意事项补充到 AGENTS.md，
将面向使用者的运行和部署说明补充到 README.md。
```
