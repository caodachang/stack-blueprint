# 网络基础


## 协议分层

TCP/IP 四层模型：

| 层 | 做什么 | 例子 |
|------|------|------|
| 应用层 | 定义数据格式和用途 | HTTP（网页请求）、DNS（域名查询）、SSH（远程登录） |
| 传输层 | 把数据切成包 | TCP（可靠但慢）、UDP（快但可能丢） |
| 网络层 | 在无数路由器之间找到目标 | IP 地址、路由 |
| 网络接口层 | 实际传输电信号或光信号 | 网线、WiFi、光纤 |

> OSI 七层模型是教学版本。TCP/IP 四层是实际标准。

你发一个 HTTP 请求时：应用层拼好报文 → 传输层切成 TCP 包 → 网络层标上 IP 地址 → 接口层转成电信号发出去。

## IP 地址与端口

IP 地址找到网络中的计算机，端口号找到这台计算机上负责通信的程序。端口号是 TCP 或 UDP 报文中的一个 16 位数字，范围是 `0～65535`；客户端连接服务时，需要同时知道服务器的 IP 地址和端口号。

| 默认端口 | 常见服务 | 用途 |
|------|------|------|
| `80` | HTTP | 访问未加密的网页或 API |
| `443` | HTTPS | 通过 TLS 加密访问网页或 API |
| `22` | SSH | 加密地远程登录服务器，也常用于 SFTP 和 Git 的 SSH 连接 |

这些是约定俗成的**默认端口**，不是不可改变的绑定。HTTP 服务也可以监听 `8080`，SSH 服务也可以改到其他端口；连接非默认端口时，客户端必须明确指定端口。

```text
http://example.com       # 省略端口，浏览器默认连接 80
https://example.com      # 省略端口，浏览器默认连接 443
http://localhost:8080    # 明确连接 8080
ssh -p 22 user@server    # 明确使用 SSH 的 22 端口
```

端口由监听中的服务占用。同一台计算机可以让 Web 服务监听 `80` 和 `443`，同时让 SSH 服务监听 `22`，因此一个 IP 地址能够对外提供多种服务。防火墙还可以决定哪些端口允许外部访问。

## 传输层 TCP vs UDP

| | TCP | UDP |
|------|------|------|
| 连接 | 先握手建立连接，再传数据 | 不需要连接，直接发 |
| 可靠性 | 确认机制，丢包重发，保证顺序 | 不保证送到，不保证顺序 |
| 速度 | 慢，握手和确认有开销 | 快，没任何额外操作 |
| 适用 | 网页、文件、邮件——不能丢一字节 | 视频通话、游戏——丢几帧无所谓，但延迟不能高 |

## HTTP

浏览器和服务器之间的通信协议。

### 请求方法

| 方法 | 做什么 |
|------|------|
| `GET` | 获取数据（打开网页、查 API） |
| `POST` | 提交数据（登录、发评论、上传文件） |
| `PUT` | 替换整个资源 |
| `PATCH` | 修改部分数据 |
| `DELETE` | 删除 |

### 状态码（由服务器返回给客户端）

| 范围 | 含义 | 常见 |
|------|------|------|
| 2xx | 成功 | `200 OK`、`201 Created` |
| 3xx | 重定向 | `301 永久搬家`、`302 临时搬家` |
| 4xx | 客户端错误 | `404 找不到`、`403 没权限`、`400 请求格式错误` |
| 5xx | 服务器错误 | `500 服务器内部错误`、`502 网关错误` |

### HTTPS

HTTP 是明文传输。HTTPS 加了一层 TLS 加密，现在所有正经网站都用 HTTPS，证书由 CA（证书颁发机构）签发。

### 一个请求的完整旅程

你在浏览器输入 `github.com`：

```
1. DNS 解析 → github.com → 140.82.112.4
2. TCP 三次握手 → 浏览器和服务器建立连接
3. TLS 握手 → 协商加密密钥
4. HTTP 请求 → GET / HTTP/1.1
5. 服务器处理 → 生成 HTML
6. HTTP 响应 → 200 OK + HTML 内容
7. 浏览器渲染 → 解析 HTML、CSS、JS → 你看到页面
```
