# C盘目录


## 一级目录

| 名称 | 是什么 |
|------|------|
| `用户` | 用户文件 |
| `Windows` | 操作系统，只需要磁盘清理（设置 → 系统 → 存储 → 临时文件） |
| `tmp` | 临时文件（可删） |
| `ProgramData` | 软件共享数据 |
| `Program Files (x86)` | 32位软件，各软件的组件和系统自带 |
| `Program Files` | 64位软件，各软件的组件和系统自带 |
| `PerfLogs` | 用 perfmon 之类的性能诊断工具时产生的日志写在这里（没用过是空的） |
| `OneDriveTemp` | 云盘的临时缓存，可删 |
| `Intel` | 驱动安装时的解压临时目录（是空的），可删 |
| `ESD` | Windows 的更新包（都是受保护的操作系统文件） |
| `c` | Git Bash / MSYS2 路径映射，把 /c/Users/23224 翻译成 C:\Users\23224 |
| `.zf_index` | 搜索索引（可再生文件），可删 |
| `$Windows.~WS` | 系统安装临时文件，没在装系统时可删 |
| `$WINDOWS.~BT` | 系统升级临时文件，没在升级 Windows 时可删 |
| `$SysReset` | 系统重置残留，可删 |
| `$GetCurrent` | 系统日常更新临时文件，没在日常更新时可删 |

## 用户

| 名称 | 是什么 |
|------|------|
| `23224` | 你的账户，所有个人文件都在这里 |
| `公用 / Sangfor` | 公共文件夹 / 有 aTrust 的证书 |
| `Default` | 新用户模板，隐藏且不能删 |
| `WsiAccount` | WSL 系统服务账户，不能删 |
