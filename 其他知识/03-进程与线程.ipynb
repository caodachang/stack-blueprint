# 进程与线程


## 进程与线程

程序是躺在硬盘上的文件，进程是跑在内存里的程序。同一个程序可以跑多个进程，每个有自己的 PID（进程号）。

一个进程可以包含多个线程，进程之间内存隔离，线程之间共享内存。

## 前台与后台

命令默认跑在前台，会占用终端，结尾加 `&` 会把命令放在后台跑。

## 信号与终止

信号是操作系统发给进程的短消息，`kill` 命令就是发送退出信号

```bash
kill 1234        # 默认发 SIGTERM（15），请进程号 1234 的进程退出
kill -9 1234     # 发 SIGKILL（9），强制杀掉，进程无法忽略
Ctrl+C           # 发 SIGINT（2），中断当前前台进程
Ctrl+Z           # 发 SIGTSTP，冻结当前前台进程，扔到后台
```

## 退出码

每个进程结束时都会返回一个数字（退出码）给启动它的进程（Shell），`0` 表示成功，非 `0` 表示出错。

脚本里用 `&&`（短路与）和 `||`（短路或）决定下一步

```bash
make && ./deploy     
make || echo "构建失败"  
```

## 守护进程

守护进程（daemon）：脱离终端，开机自启，在后台一直运行
