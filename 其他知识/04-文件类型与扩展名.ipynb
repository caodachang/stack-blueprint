# 文件类型与扩展名


## 文件类型的本质

扩展名只是给操作系统和人的提示。真正的类型判断靠文件头（magic number）决定

## 代码文件

| 扩展名 | 语言 |
|------|------|
| `.py` | Python |
| `.js` | JavaScript |
| `.ts` | TypeScript（JavaScript 加类型标注） |
| `.jsx` | React 组件（JavaScript 里写 HTML） |
| `.tsx` | React 组件（TypeScript 版） |
| `.go` | Go |
| `.rs` | Rust |
| `.java` | Java |
| `.c` | C |
| `.cpp` | C++ |
| `.h` | C/C++ 头文件（声明函数和类型） |
| `.rb` | Ruby |
| `.php` | PHP |
| `.swift` | Swift（Apple 生态） |
| `.kt` | Kotlin（Android 首选） |
| `.html` | 网页结构 |
| `.css` | 网页样式 |
| `.sql` | 数据库查询语句 |


## 配置文件

| 扩展名 | 用途 | 特点 |
|------|------|------|
| `.json` | 数据交换、配置 | 严格语法，双引号必加，不能写注释 |
| `.yaml` `.yml` | 配置 | 缩进表示层级，比 JSON 可读，能写注释 |
| `.toml` | 配置 | 比 YAML 简洁，Rust/Python 项目新宠 |
| `.xml` | 配置、数据交换 | 老格式，标签包裹数据，冗余但严谨 |
| `.env` | 环境变量 | `KEY=VALUE`，每行一个，不提交到 Git |
| `.ini` | 简单配置 | Windows 风格的 `[section]` 格式 |
| `.cfg` | 通用配置 | 没有固定格式，看项目约定 |
| `Makefile` | 构建规则 | 无扩展名，`make` 命令读取 |
| `Dockerfile` | 容器镜像定义 | 无扩展名，`docker build` 读取 |

## 文档

| 扩展名 | 类型 |
|------|------|
| `.md` | Markdown——纯文本写文档，（README.md、笔记） |
| `.mdx` | Markdown 加 JSX（文档站点） |
| `.txt` | 纯文本 |
| `.pdf` | 便携文档，只读，跨平台不变形 |
| `.docx` | Word 文档 |
| `.pptx` | PPT |
| `.xlsx` | Excel 表格 |
| `.csv` | 逗号分隔的表格数据，纯文本，任何语言都能读写 |
| `.rst` | Python 技术文档老标准 |
| `.log` | 日志文件，纯文本，通常不停追加 |

## 图片

| 扩展名 | 类型 | 特点 |
|------|------|------|
| `.png` | 无损 | 支持透明通道，截图和图标首选 |
| `.jpg` `.jpeg` | 有损 | 压缩率高，每次保存损失画质 |
| `.gif` | 无损（256 色） | 动图，颜色少所以体积小 |
| `.svg` | 矢量 | 数学描述图形，放大不失真，图标和插图 |
| `.webp` | 有损/无损 | Google 的格式，同质量比 jpg 小 30% |
| `.bmp` | 无损无压缩 | Windows 老格式，体积巨大 |
| `.ico` | 图标 | 网站 favicon 和 Windows 快捷方式 |
| `.avif` | 有损/无损 | 新一代格式，比 webp 压缩率更高 |


## 视频与音频

| 扩展名 | 类型 |
|------|------|
| `.mp4` | 最通用的视频格式 |
| `.avi` | 老视频格式，体积大 |
| `.mov` | Apple QuickTime 视频 |
| `.webm` | 网页视频 |
| `.mp3` | 最通用的有损音频 |
| `.wav` | 无损音频，体积大 |
| `.flac` | 无损压缩音频 |


## 压缩与打包

| 扩展名 | 类型 |
|------|------|
| `.zip` | 通用压缩包（Windows 自带） |
| `.tar` | 打包不压缩，把多个文件合成一个 |
| `.gz` | Gzip 压缩，常跟 tar 组合成 `.tar.gz` |
| `.bz2` | 比 gzip 压缩率高，但更慢 |
| `.xz` | 更高压缩比 |
| `.7z` | 极高压缩比 |
| `.rar` | 商业压缩格式 |

## 可执行与安装

| 扩展名 | 类型 |
|------|------|
| `.exe` | Windows 可执行程序 |
| `.msi` | Windows 安装包 |
| `.sh` | Shell 脚本（Linux/macOS） |
| `.bat` | Windows 批处理脚本 |
| `.ps1` | PowerShell 脚本 |
| `.apk` | Android 安装包 |
| `.dmg` | macOS 安装包镜像 |

Linux 可执行文件通常没有扩展名，靠文件头的 magic number 和权限位 `x` 来识别。

## 其他开发相关

| 文件 | 类型 |
|------|------|
| `.db` `.sqlite` `.sqlite3` | SQLite 数据库文件 |
| `.dump` | 数据库导出的 SQL |
| `.ipynb` | Jupyter Notebook |
| `*.lock` | 依赖锁文件（`package-lock.json`、`uv.lock`、`Cargo.lock`），锁定依赖的精确版本 |
| `.gitignore` | 告诉 Git 哪些文件不跟踪 |
| `.dockerignore` | 告诉 Docker 哪些文件不复制 |
| `.editorconfig` | 编辑器通用配置（缩进、换行符），跨 IDE 生效 |
| `.prettierrc` | Prettier 格式化规则 |
| `eslint.config.*` | ESLint 代码检查规则 |
