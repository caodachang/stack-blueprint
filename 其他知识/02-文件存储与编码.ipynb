# 文件存储与编码


## 文件是怎么存的

文件名只是标签，文件真正的身份证是 inode。每个文件对应一个 inode，里面存着文件各种消息。一个 inode 可以有多个文件名指向它。

**硬链接**是给同一个 inode 起多个名字。只要不把名字删完，数据就还在。

**软链接**（符号链接）有自己的 inode，存的是原文件的路径，快捷方式文件

## 权限

```
-  rwx  r-x  r-x
│  ─┬─  ─┬─  ─┬─
│   │    │    └── 其他人的权限
│   │    └─────── 用户组的权限
│   └──────────── 文件拥有者的权限
└── 文件类型（- 普通文件，d 目录，l 软链接）
```

`r` 读，`w` 写，`x` 执行。

chmod 添加权限

```bash
chmod 755 script.sh    # 拥有者 rwx，其他人 rx
chmod 644 file.txt     # 拥有者 rw，其他人 r
```

数字是二进制：`7` = `rwx`（111），`5` = `r-x`（101），`4` = `r--`（100）。


## 字符编码

**ASCII** 是最早的编码标准，8 位表示一个字符，1 位用作校验，表示 128 个字符。

**Unicode** 给全世界每个字符一个唯一的编号（码点）

**UTF-8** 是现在最通用的存储方式兼容 ASCII。是一种变长编码

**Base64** 把任意二进制数据（图片、加密结果、签名）转成 64 个可打印的 ASCII 字符
