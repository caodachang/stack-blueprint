# Git与GitHub协作


Git 是分布式版本控制系统。每个开发者的电脑上都有完整的仓库副本——不只是最新版，是整个历史。没有网络也能提交、看历史、建分支。

GitHub 是托管 Git 仓库的网站。你把本地仓库推上去，别人就能克隆、修改、提交回来。GitHub 在 Git 之上加了 Pull Request、Issue、Actions 等协作功能。

### 怎么用

把一个文件夹变成 Git 仓库：

```bash
git init
```

这会创建隐藏的 `.git` 目录——Git 的全部数据都在里面。删除 `.git` 等于删掉所有版本历史，代码变回普通文件夹。

### 配置提交身份

Git 会把提交者姓名和邮箱写入每次提交。先检查当前全局配置：

```bash
git config --global user.name
git config --global user.email
```

如果没有输出，或者内容需要修改，执行：

```bash
git config --global user.name "你的名字"
git config --global user.email "你的邮箱"
```

`--global` 表示配置应用于当前电脑用户，以后创建其他仓库时也会使用，一般每台电脑配置一次即可。

> 姓名和邮箱是提交历史的一部分，不要在公开仓库中使用不希望公开的个人信息。

## Git 基本操作

Git 有三个区域：

| 区域 | 是什么 | 什么时候变 |
|------|--------|-----------|
| 工作区 | 你硬盘上的文件 | 每次保存文件 |
| 暂存区 | 标记了"下次提交"的修改 | `git add` |
| 仓库 | 永久保存的快照 | `git commit` |

### 怎么用

每次提交走三步：

```bash
git status                # 看哪些文件改过
git add app.py            # 把指定文件加入暂存区
git add .                 # 把所有修改加入暂存区
git commit -m "添加登录功能"  # 提交到仓库
```

查看历史：

```bash
git log                   # 提交记录
git log --oneline         # 每条一行，简洁版
```

切换版本：

```bash
git checkout oooooo         # 000000 是 log 查看的版本号

```


## 分支与合并

### 分支是什么

分支是一个指向某次提交的指针。创建新分支等于给当前状态贴标签，之后的修改不影响其他分支。

### 怎么用

```bash
git switch -c login       # 创建并切换到 login 分支
# 开发登录功能，提交...
git switch main           # 切回 main 分支
git merge login           # 把 login 的修改合并进来
```

如果两个分支改了同一行，Git 报冲突。文件里会这样标记：

```
<<<<<<< HEAD
你的版本
=======
别人的版本
>>>>>>> login
```

手动编辑决定保留哪个，然后 `git add` 和 `git commit` 通知 Git 冲突已解决。

### merge 和 rebase

合并分支有两种方式：

| | merge | rebase |
|------|--------|--------|
| 做了什么 | 把两个分支的历史合在一起，生成一个新提交 | 把你的提交"搬"到目标分支的最新位置，重写历史 |
| 提交历史 | 保留真实的分支时间线，有分叉 | 变成一条直线，看起来像是顺序开发的 |
| 什么时候用 | 合并功能分支到共享分支（main） | 拉取远程更新时保持历史整洁 |

```bash
git merge login       # 合并，保留分支历史
git rebase main       # 变基，把你的提交搬到 main 最新位置
```

> rebase 会改写提交历史。不要对已经推送到远程的提交做 rebase——别人的仓库基于旧历史，你一改写就全乱了。


## Pull Request

### 它是什么

Pull Request（PR）是 GitHub 的核心协作机制。

PR 的本质是一份"合并请求"——你把代码推到一个分支，然后发起一个 PR，告诉仓库维护者："我的代码在 xx 分支上，请审查后合并到 main"。

### 怎么用

一个完整的 PR 流程：

```bash
git pull origin main          # 1. 拉最新代码
git switch -c feature-xxx     # 2. 创建功能分支
# 开发、git add、git commit...
git push origin feature-xxx   # 3. 推送分支到 GitHub
```

4. 在 GitHub 仓库页面点「New Pull Request」，选 main 作为目标分支，feature-xxx 作为源分支
5. 写 PR 标题和描述——说明改了什么、为什么改
6. 其他人审查代码，在 PR 页面逐行评论、提修改建议
7. 根据反馈改代码，后续`git push` 会自动更新 PR
8. 全部通过后，仓库原拥有者点「Merge pull request」合并到 main
