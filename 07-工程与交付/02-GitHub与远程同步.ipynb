# GitHub 与远程同步

## 第一步：在 GitHub 创建空仓库 

登录 GitHub，创建一个新仓库 ( new repository )。建议填写：

| 选项 | 建议 |
|------|------|
| Repository name | 项目名 |
| Public / Private | 按自己的公开需求选择 |
| Add a README file | **不要勾选** |
| Add .gitignore | 不选择 |
| Choose a license | 暂时不选择 |

### 有两种连接方式

```text
HTTPS：https://github.com/你的用户名/项目名.git
SSH : git@github.com:你的用户名/项目名.git
```

## 方式一：使用 HTTPS

### 1. 复制 HTTPS 地址

```text
https://github.com/你的用户名/项目名.git
```

### 2. 关联本地仓库

```bash
git remote add origin https://github.com/你的用户名/项目名.git
git remote -v  # 查看远程仓库
```

### 3. 第一次推送

先用 `git branch` 确认分支名。当前分支为 `main` 时执行：

```bash
git branch
git push -u origin main

git push    # 后续只需要这样
```

## 方式二：使用 SSH

## 生成并添加 SSH 公钥

### 生成密钥对

在自己的电脑上执行：

```bash
ssh-keygen -t ed25519 -C "你的邮箱"
```

- `-t ed25519`：选择现代的 Ed25519 密钥类型。
- `-C`：为密钥添加备注，通常填写 GitHub 邮箱。

终端询问保存位置时，初次配置可直接回车使用默认路径；询问 passphrase 时，可根据需要设置密钥口令或直接回车。

生成后通常得到：

```text
~/.ssh/id_ed25519
~/.ssh/id_ed25519.pub  # 有.pub的是公钥
```

> 如果终端提示目标文件已经存在，不要直接覆盖。已有密钥可能正在被其他服务使用。

### 查看并复制公钥

```bash
cat ~/.ssh/id_ed25519.pub
```

复制终端输出的完整一行。内容一般以 `ssh-ed25519` 开头。

### 添加到 GitHub

依次进入：

```text
GitHub → Settings → SSH and GPG keys → New SSH key
```

为密钥填写便于识别的标题，把公钥完整粘贴到 Key 中并保存。每台需要连接 GitHub 的电脑通常都要单独添加自己的公钥。

## 验证 GitHub SSH 连接

执行：

```bash
ssh -T git@github.com
```

第一次连接时可能会询问是否信任 GitHub 主机。确认目标是 `github.com` 后输入：

```text
yes
```

认证成功时通常会看到类似提示：

```text
Hi 用户名! You've successfully authenticated...
```

GitHub 不提供普通 SSH Shell，因此提示中即使包含“不提供 shell access”，只要出现 `successfully authenticated` 就表示连接成功。

验证成功后，就可以使用以 `git@github.com:` 开头的 SSH 仓库地址。


### 1. 复制 SSH 地址

```text
git@github.com:你的用户名/项目名.git
```

### 2. 关联本地仓库

```bash
git remote add origin git@github.com:你的用户名/项目名.git
git remote -v  # 查看远程仓库
```

### 3. 第一次推送

先用 `git branch` 确认分支名。当前分支为 `main` 时执行：

```bash
git branch
git push -u origin main

git push    # 后续只需要这样
```

## HTTPS 与 SSH 怎么选择

### 切换已有 origin 的连接方式

仓库已经配置 `origin` 时，不要再次执行 `git remote add origin`，而是修改地址。

切换到 HTTPS：

```bash
git remote set-url origin https://github.com/你的用户名/项目名.git
```

切换到 SSH：

```bash
git remote set-url origin git@github.com:你的用户名/项目名.git
```

修改后检查：

```bash
git remote -v
```
