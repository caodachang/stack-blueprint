# CI-CD、GitHubActions、监控告警与灰度发布

## CI

CI 是持续集成，重点是让代码频繁合并，并尽快发现问题。

一次常见 CI 会做：安装依赖、运行格式检查、运行测试、构建项目。它的价值是把“这次提交会不会破坏项目”尽早暴露出来。

## CD

CD 可以指持续交付，也可以指持续部署。

持续交付强调代码随时可以发布，但最终上线可能需要人工确认。持续部署更进一步，流水线通过后自动发布到生产环境。

## GitHub Actions

GitHub Actions 是 GitHub 提供的自动化流水线工具。

## 灰度、金丝雀与蓝绿部署

灰度发布是先让一小部分用户使用新版本，观察稳定后再扩大范围。

金丝雀发布是灰度发布的一种说法，强调先用少量流量验证新版本风险。

蓝绿部署是同时准备两套环境：蓝色环境运行旧版本，绿色环境运行新版本。切流量即可完成发布，如果新版本出问题，可以快速切回旧环境。
