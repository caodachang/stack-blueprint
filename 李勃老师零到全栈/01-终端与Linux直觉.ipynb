# 终端与 Linux 直觉

## 命令

| 命令 | 作用 | 示例 |
|------|------|------|
| `pwd` | 查看当前目录的完整路径 | `pwd` |
| `ls` | 查看当前目录中的文件和目录 | `ls` |
| `cd` | 切换当前目录 | `cd ~/project` |
| `mkdir` | 创建目录 | `mkdir practice` |
| `mkdir -p` | 连同缺少的上级目录一起创建 | `mkdir -p project/css` |
| `touch` | 创建空文件 | `touch hello.txt` |
| `cat` | 查看文本文件内容 | `cat hello.txt` |
| `rm` | 删除文件，通常不进入回收站 | `rm hello.txt` |
| `rmdir` | 删除空目录 | `rmdir practice` |
| `clear` | 清空终端画面 | `clear` |

| 符号 | 含义 | 示例 |
|------|------|------|
| `~` | 当前用户的家目录 | `cd ~` |
| `/` | 根目录；也用于分隔路径层级 | `cd /`、`project/css` |
| `.` | 当前目录 | `code .`用 vscode 打开 |
| `..` | 当前目录的上一级 | `cd ..` |

输入文件名或路径时可以按 `Tab` 自动补全。删除文件前，应先执行 `pwd` 和 `ls`，确认当前位置和目标。

## Vim

Vim 是终端中的文本编辑器。第一次使用只需要理解两种状态：**输入文字**和**执行命令**。

```text
                    按 i
普通模式  ─────────────────────→  插入模式
执行命令                         输入文字
    ↑                              │
    └──────────── 按 Esc ──────────┘

普通模式下按 : 进入命令行，输入保存或退出命令。
```

### 第一次编辑文件

**1. 打开文件**

```bash
vim hello.txt
```

**2. 按 `i`，开始输入**

屏幕底部出现 `-- INSERT --`，表示已经进入插入模式，可以正常输入和修改文字。

**3. 按 `Esc`，停止输入**

回到普通模式后，才能执行保存和退出命令。

**4. 输入 `:wq`，按回车**

文件会被保存，Vim 随后退出并回到终端。

> 最常用流程：`i` → 输入内容 → `Esc` → `:wq` → 回车

### 保存与退出速查

| 想做什么 | 输入 |
|------|------|
| 保存，继续编辑 | `:w` |
| 没有修改时直接退出 | `:q` |
| 保存并退出 | `:wq` |
| 放弃本次修改并退出 | `:q!` |

> 不知道当前是什么模式时，先按 `Esc`。无法输入文字时，按 `i`。无法退出时，按 `Esc` 后输入 `:wq` 或 `:q!`。
