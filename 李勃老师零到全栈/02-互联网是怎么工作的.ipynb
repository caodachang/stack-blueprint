# 互联网是怎么工作的

## 从 HTML 文件到网页

网页首先可以只是电脑中的一个 HTML 文件。创建项目目录和文件：

```bash
cd ~
mkdir zero-to-tech
cd zero-to-tech
touch index.html
```

在 `index.html` 中写入：

```html
<!DOCTYPE html>
<html lang="zh-CN">
<head>
  <meta charset="UTF-8">
  <title>我的第一个网页</title>
</head>
<body>
  <h1>你好，互联网</h1>
  <p>这是我的第一个网页。</p>
</body>
</html>
```

HTML 文件是网页的**源文件**，浏览器读取并解释其中的标签，然后把它显示成可见页面。源文件和浏览器中的页面不是两份独立内容，而是同一份信息的代码形态与显示结果。

```text
index.html 源文件  ──浏览器读取和解释──→  屏幕上的网页
```

本地打开文件时，浏览器地址通常以 `file://` 开头。这表示内容直接来自本机文件系统，没有通过互联网向远程服务器发请求。

## 从本地网页到远程网站

浏览器既能读取本机的 HTML 文件，也能通过网络获取另一台电脑提供的内容。网站不一定只有 HTML，还可能包含 CSS、JavaScript、图片、视频和动态数据。

持续联网、接收请求并提供内容或能力的电脑，通常称为**服务器**。服务器不是某种特殊形状的机器，而是它在当前关系中扮演的角色。

```text
本地文件
浏览器 ─────────────→ 自己电脑里的 index.html

远程网站
浏览器 ──互联网请求──→ 服务器 ──返回内容──→ 浏览器
```

浏览器是发起请求的一方，称为**客户端**；持续等待请求并作出回应的一方称为**服务端**。这里的“返回”表示服务端发送回应，不表示原来的请求原封不动地回来。

## IP、域名、DNS 与端口

浏览器访问远程网站时，要先找到服务器，再找到服务器上的具体服务。

| 概念 | 作用 | 可以怎样理解 |
|------|------|------|
| IP 地址 | 定位网络中的设备 | 服务器的网络地址 |
| 域名 | 提供便于人记忆的名称 | IP 地址的可读名字 |
| DNS | 把域名查询为 IP 地址 | 域名到 IP 的解析系统 |
| 端口 | 定位设备上的具体网络服务 | 同一台服务器上的不同入口 |

例如，用户输入 `github.com`，浏览器先通过 DNS 查询对应的 IP 地址，再连接该服务器上提供网页服务的端口。

一台服务器可以同时运行多个服务，因此只有 IP 还不够。端口号用于区分目标服务：

| 默认端口 | 服务 | 用途 |
|------|------|------|
| `80` | HTTP | 访问未加密的网页 |
| `443` | HTTPS | 通过 TLS 加密访问网页 |

```text
http://example.com   → 默认连接 80 端口
https://example.com  → 默认连接 443 端口
```

这些是默认约定，不是不可修改的绑定。如果服务监听非默认端口，网址中需要明确写出，例如 `http://localhost:8000`。

## 输入网址后发生了什么

当用户在浏览器中输入网址并按下回车时，可以先用下面这条主线理解：

```text
输入域名
   ↓
DNS 查询 IP 地址
   ↓
浏览器找到服务器
   ↓
连接目标端口
   ↓
发送请求 Request
   ↓
服务器处理请求
   ↓
返回响应 Response
   ↓
浏览器解析并显示页面
```

整个通信有两个方向：

| 方向 | 名称 | 内容 |
|------|------|------|
| 浏览器 → 服务器 | 请求（Request） | 想访问什么、提交了什么信息 |
| 服务器 → 浏览器 | 响应（Response） | 网页文件、图片、数据或错误信息 |

因此，访问网站可以概括为：**浏览器通过 DNS、IP 和端口找到服务器上的服务，发送请求；服务器返回响应，浏览器再把响应内容显示出来。**
