# 准备好云服务器

## 租用一台云服务器

购买时重点确认：

| 项目 | 建议 |
|------|------|
| 产品类型 | 云服务器或轻量应用服务器，不要误买对象存储、数据库或 CDN |
| 操作系统 | Ubuntu，优先选择 `24.04 LTS`  LTS 表示长期支持版本 |
| 硬件配置 | 学习阶段从最低档开始即可 |
| 网络 | 必须确认分配了公网 IP |

最后进入云平台的**防火墙**或**安全组**页面，确认服务器已经允许 TCP `22` 端口的入站连接。SSH 默认使用 22 端口，没有放行就无法远程登录。

## 使用 SSH 登录服务器

SSH 是通过网络安全登录远程 Linux 服务器的标准方式。在自己电脑的终端中输入：

```bash
ssh 用户名@服务器公网IP
```

输入 `exit` 断开 SSH 连接。

## 安装并启动 Nginx

Nginx 是常用的 Web 服务器软件。安装后，它可以持续监听 HTTP 请求，并把网页内容返回给浏览器。

先更新 Ubuntu 的软件包列表，再安装 Nginx：

```bash
sudo apt update
sudo apt install nginx -y
```

| 名称 | 作用 |
|------|------|
| `sudo` | 临时以管理员权限执行命令 |
| `apt` | Ubuntu 的软件包管理工具 |
| `update` | 刷新可安装软件的信息 |
| `install nginx` | 安装 Nginx |
| `-y` | 自动确认安装提示 |

检查 Nginx 状态：

```bash
systemctl status nginx
```

看到下面的状态表示服务已经启动：

```text
Active: active (running)
```

状态页面占据终端时，按 `q` 返回命令提示符。若 Nginx 已安装但没有运行，可以执行：

```bash
sudo systemctl start nginx
```

## 开放 80 端口并访问网页

Nginx 默认监听 HTTP 的 `80` 端口，在云平台的安全组或防火墙中放行入站流量。

| 端口 | 服务 | 用途 |
|------|------|------|
| `22/TCP` | SSH | 从自己的电脑远程登录服务器 |
| `80/TCP` | HTTP | 让所有人的浏览器访问普通网页 |

然后在本地浏览器访问：

```text
http://你的服务器公网IP
```

## 找到 Nginx 默认网页文件

浏览器看到的默认页面来自服务器上的真实 HTML 文件。先查看 Nginx 的默认站点配置：

```bash
cat /etc/nginx/sites-available/default
```

在输出中找到：

```nginx
root /var/www/html;
```

这表示 Nginx 会到 `/var/www/html` 目录中寻找网页文件。进入目录并查看内容：

```bash
cd /var/www/html
ls
```

Ubuntu 上常见的默认文件名是：

```text
index.nginx-debian.html
```

查看它的源代码：

```bash
cat index.nginx-debian.html
```

## 使用 Vim 修改公网页面

默认网页位于系统目录，普通用户没有直接写入权限，因此使用 `sudo` 打开：

```bash
sudo vim /var/www/html/index.nginx-debian.html
```

操作顺序：

1. 按 `i` 进入插入模式。
2. 找到页面中醒目的英文，改成自己的文字。
3. 按 `Esc` 返回普通模式。
4. 输入 `:wq`，再按回车，保存并退出。
5. 回到本地浏览器并刷新页面。
