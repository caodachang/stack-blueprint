# 终端与 Linux 直觉

## 命令与重要符号

| 命令 | 作用 | 示例 |
|------|------|------|
| `pwd` | 显示当前目录的完整路径 | `pwd` |
| `ls` | 列出当前目录中的内容 | `ls` |
| `cd` | 切换当前目录 | `cd terminal-practice` |
| `mkdir` | 创建目录 | `mkdir terminal-practice` |
| `mkdir -p` | 连同缺少的上级目录一起创建 | `mkdir -p project/css` |
| `touch` | 创建空文件 | `touch hello.txt` |
| `cat` | 把文本文件内容显示在终端中 | `cat hello.txt` |
| `rm` | 删除文件 （ `-rf`：强制删除整个目录 ） | `rm hello.txt` |
| `rmdir` | 删除空目录 | `rmdir terminal-practice` |
| `clear` | 清空终端当前画面 | `clear` |

| 符号 | 含义 | 示例 |
|------|------|------|
| `~` | 当前用户的家目录 | `cd ~` |
| `/` | 文件系统根目录；也用于分隔路径层级 | `cd /`、`project/css` |
| `..` | 当前目录的上一级 | `cd ..` |
| `.` | 当前目录本身 | `code .` |


### 创建用户

`adduser 用户名`：创建一个普通用户，自动生成 home 目录

### 添加 sudo 权限

`usermod -aG sudo 用户名`:在 root 下执行

### 切换用户

`su - 用户名`

## Vim 编辑文件

```text
                    按 i
普通模式  ─────────────────────→  插入模式
执行命令                         输入文字
    ↑                              │
    └──────────── 按 Esc ──────────┘

普通模式下按 : 进入命令行，输入保存或退出命令。
```

### 保存并退出

```bash
vim hello.txt
```

1. 按 `i` 进入插入模式。
2. 输入 `hello terminal`。
3. 按 `Esc` 回到普通模式。
4. 输入 `:wq`，按回车保存并退出。
5. 执行 `cat hello.txt` 验证保存结果。

| 想做什么 | 命令 |
|------|------|
| 保存但继续编辑 | `:w` |
| 没有未保存修改时退出 | `:q` |
| 保存并退出 | `:wq` |
| 放弃修改并强制退出 | `:q!` |
