# 编程语言概述


## 执行方式（现在界限越来越模糊）

| 方式 | 怎么工作 | 代表语言 |
|------|---------|---------|
| 编译型 | 先完整翻译成机器码，再执行。 | C、C++、Go、Rust |
| 解释型 | 一行一行读、实时翻译执行。 | Python、JavaScript、Ruby |
| JIT 编译 | 运行时把热点代码编译成机器码。 | Java、C#、JavaScript（V8） |

## 类型系统

| | 静态类型 | 动态类型 |
|------|---------|---------|
| 类型检查时机 | 编译时，写错立刻标红 | 运行时，代码跑到那一行才报错 |
| 要不要写类型 | 必须写：`int x = 1` | 不写：`x = 1` |
| 代表 | Java、C++、Go、Rust | Python、JS、Ruby |

## 内存管理

程序运行时变量存在内存里。谁负责清理不再用的内存？

| 方式 | 谁管 | 代表语言 |
|------|------|---------|
| 手动管理 | 程序员自己 `malloc` / `free` | C、C++ |
| 垃圾回收（GC） | 运行时自动扫描、回收没用的内存 | Java、Go、Python、JS |
| 所有权系统 | 编译时分析，谁用谁负责，用完自动释放 | Rust |

> 游戏引擎用 C++ 因为能手动管理内存分配，GC 突然停下来扫一遍内存会卡

## 包管理

| 语言 | 包管理器 | 安装示例 |
|------|---------|---------|
| Python | pip / uv | `pip install requests` |
| JavaScript | npm / pnpm | `npm install lodash` |
| Rust | cargo | `cargo add serde` |
| Go | go mod | `go get github.com/gin-gonic/gin` |
| Java | Maven / Gradle | 写在 pom.xml 里 |

## 错误处理

程序出错时怎么通知调用者？

| 方式 | 怎么用 | 代表语言 |
|------|--------|---------|
| 返回值 | 函数返回一个错误码，调用者自己判断 | C、Go |
| 异常 | `try/catch`，出错了抛出去，不处理就往上冒 | Python、Java、JS |
| Result 类型 | 返回值里同时包含成功和失败两种情况，编译器强制你处理 | Rust、Go（多返回值） |

## 并发模型

程序同时做多件事，不同语言有不同方案：

| 模型 | 怎么工作 | 代表语言 |
|------|---------|---------|
| 多线程 | 操作系统提供多个执行流，共享内存 | Java、C++、Rust |
| 协程 | 语言层面提供轻量级线程，切换成本极低 | Go（goroutine）、Python（asyncio）、Kotlin |
| 事件循环 | 单线程，靠异步回调：有了结果通知我 | JavaScript（Node.js） |
