# HTTP、HTTPS、URL、状态与浏览器安全基础

## HTTP 

超文本传输协议。它最初是为了传输网页文档而出现，后来逐渐成为 Web 应用、接口调用、前后端通信最常见的协议。

## 请求方法

| 方法 | 常见含义 |
| --- | --- |
| `GET` | 获取资源 |
| `POST` | 创建资源或提交操作 |
| `PUT` | 整体更新资源 |
| `PATCH` | 部分更新资源 |
| `DELETE` | 删除资源 |

## URL 统一资源定位符

用来定位网络上的资源

一个 URL 可以拆成几部分：

`https://example.com:443/users?page=1#profile`

| 部分 | 例子 | 含义 |
| --- | --- | --- |
| 协议 | `https` | 使用什么协议 |
| 域名 | `example.com` | 访问哪个主机名 |
| 端口 | `443` | 访问哪个服务入口 |
| 路径 | `/users` | 访问哪个资源路径 |
| 查询参数 | `?page=1` | `?` 后面的附加筛选或分页条件 |
| 片段 | `#profile` | `#` 后面的页面内部位置，通常不发送给服务器 |

## 状态码,对处理结果的简短数字说明

> `404 Not Found`

## TLS 和 HTTPS

HTTP 本身不负责加密， TLS 是传输层安全协议。

HTTPS 可以理解为 HTTP 加上 TLS

## HTTP 是无状态的

协议本身不会自动记住上一次请求是谁发的、做过什么。

## Cookie 

Cookie 是浏览器保存的一小段数据，用来保存一些信息，由服务器通过响应 Header 设置。

## Session 

Session 通常指服务器端保存的一段会话状态。

登录成功 -> 服务器创建 Session -> 浏览器保存 session_id Cookie -> 后续请求带上 Cookie -> 服务器识别用户

## 同源策略

如果两个 URL 的协议、域名、端口都相同，就叫同源

同源策略是浏览器为保护用户数据安全设立的安全限制
- 同源的页面之间不受访问限制
- 跨域场景（非同源）会受到访问限制

## CORS 

CORS 是跨源资源共享，用来让服务器主动声明允许哪些跨域网站绕过同源策略访问自身接口
