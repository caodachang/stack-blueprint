# REST、GraphQL与gRPC：API设计风格

## API

API 是应用程序编程接口

REST、GraphQL、gRPC 是三种主流的API风格

## REST

REST 设计风格：用 URL 表示资源，用 HTTP 方法表达操作

| 操作 | 方法和路径 | 含义 |
| --- | --- | --- |
| 查询列表 | `GET /users` | 获取用户列表 |
| 查询单个 | `GET /users/123` | 获取一个用户 |
| 创建 | `POST /users` | 创建用户 |
| 更新 | `PATCH /users/123` | 更新一个用户 |
| 删除 | `DELETE /users/123` | 删除一个用户 |  


`/users`所有用户资源， `/users/123` ID 为 123 的单个用户资源

## GraphQL

GraphQL 解决 REST 接口里“要么拿太多，要么拿太少”的问题。

GraphQL 允许客户端声明自己需要哪些字段，服务器按这个查询返回对应结构。

查询示例：

```graphql
query {
  user(id: "123") {
    id
    name
    email
  }
}
```

客户端明确写出自己要 `id`、`name`、`email`。服务器返回的结构通常和查询结构接近

## gRPC 

gRPC 是一种远程过程调用框架，常和 Protocol Buffers 搭配

**RPC** 意思是远程过程调用：像调用本地函数一样调用远程服务

Protocol Buffers 简称 **protobuf**：是一种接口描述和数据序列化格式

**序列化**是把数据结构转换成适合传输或存储的格式。

## 三者怎么选

| 风格 | 更常见场景 | 主要特点 |
| --- | --- | --- |
| REST | 通用 Web API、前后端接口、开放平台 | 简单、直观、贴近 HTTP |
| GraphQL | 前端页面复杂、字段组合灵活、聚合多个资源 | 客户端声明需要什么数据 |
| gRPC | 服务间通信、内部微服务、高性能调用 | 强契约、高性能、适合多语言生成代码 |
