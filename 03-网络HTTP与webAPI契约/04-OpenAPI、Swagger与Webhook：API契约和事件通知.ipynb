# OpenAPI、Swagger与Webhook：API契约和事件通知

## API 契约

API 契约指客户端和服务器之间对接口的明确约定

## OpenAPI 

OpenAPI 是一种描述 HTTP API 的规范。它规定了如何用结构化文件描述一个 API 服务。

OpenAPI 文档通常能描述：

- 有哪些接口路径。
- 每个接口支持什么 HTTP 方法。
- 需要哪些参数。
- 请求 Body 是什么结构。
- 响应 Body 是什么结构。
- 可能返回哪些状态码。
- 如何认证。

## Swagger 

Swagger 泛指 OpenAPI 相关工具生态，规范部分发展成 OpenAPI

## OpenAPI 能怎么用

OpenAPI 的价值不只是写文档。

它可以用于：

- 给前端、后端、测试统一接口理解。
- 生成可交互接口文档。
- 生成客户端 SDK。
- 生成接口类型。
- 做接口 Mock。
- 做契约测试。

Mock 是模拟的意思。接口 Mock 指后端还没完成时，先根据契约模拟响应，让前端或测试能继续工作。

## Webhook 是什么

Webhook 是一种事件通知机制，提供一个 URL 给对方（通常是服务器），对方在事件发生时向这个 URL 发送 HTTP 请求
