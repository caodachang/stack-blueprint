# 第7章：Skills 的含义、作用与运行机制

## 7.1 与相近概念区分

| 内容 | 解决的问题 |
| --- | --- |
| 本次提示词 | 处理哪份材料、输出哪里 |
| 项目规则 | 当前项目始终遵守哪些约定 |
| 记忆 | 后续可用的偏好和背景，取决于产品设置 |
| 参考资料或知识库 | 可查哪些事实和定义 |
| Skill | 此类工作如何完成 |
| 工具 | 真正执行什么动作 |
| MCP | 如何连接外部能力 |
| 插件 | 如何打包分发一组技能、工具或其他能力 |

## 7.2 最小文件与可选资源

```text
incident-notes/
├── SKILL.md       技能用途、工作步骤
├── references/    规则、示例等参考资料（可选）
├── assets/        成果模板等资源（可选）
└── scripts/       可重复执行的计算或检查，脚本（可选）
```
