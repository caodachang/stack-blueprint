# 第6章：MCP 服务器介绍与接入

## 6.1 什么是MCP

MCP（Model Context Protocol）为发现外部能力、描述参数、调用并取回结果提供共同协议。

## 6.2 组件与业务接口

| 角色 | 责任 |
| --- | --- |
| 模型 | 分析问题、提出调用并解释结果 |
| Host | Codex、WorkBuddy 等宿主，组织执行 |
| Client | 宿主内部连接某个 MCP 服务的组件 |
| Server | 提供工具、资源或提示模板的程序 |

调用链为用户目标 → 选择工具 → 客户端发送参数 → 服务执行 → 返回结果 → 整理回答。

## 6.3 与模型、Skills的关系

切换模型决定谁推理，MCP决定能接哪些外部能力，Skill决定这类任务按什么步骤完成。

## 6.4 两种传输方式

| 方式 | 配置与运行条件 |
| --- | --- |
| STDIO | 客户端启动程序，标准输入输出通信；需要命令、解释器、参数和环境变量 |
| Streamable HTTP | 访问已经运行的协议端点；需要地址、网络和认证 |
