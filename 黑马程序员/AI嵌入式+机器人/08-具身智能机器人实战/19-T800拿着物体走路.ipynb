## 2. 机器人二次开发的三个空间问题



*原课件第 2 页：依次提出身体部件的位置、目标物体的位置，以及两者的空间关系。*

课件把空间理解拆成三个问题：

| 原图中的问题 | 在本课任务中的理解 |
|---|---|
| “我的身体部件在哪里？” | 先明确参与任务的机器人部件，理解它当前处于什么位置 |
| “目标物体在哪里？” | 明确要携带的物体，理解它当前处于什么位置 |
| “它们俩的空间关系是什么？” | 把部件与物体联系起来，描述物体相对于该部件的位置关系 |

第一个问题关注机器人自身。联系第 04、05 节，找到部件名称之后，还需要知道所描述的是哪个部件、哪个参考点，以及它的位置采用什么坐标系。

第二个问题关注任务对象。机器人部件与物体各有自己的位置描述，比较两者时需要统一参照。例如，两者都采用世界坐标，才能直接比较它们在同一场景中的位置。

第三个问题把前两个问题连接起来。知道部件和物体各自在哪里，才能进一步描述物体在部件的哪个方向、相隔多远。这也是理解首页“相对坐标绑定”的基础。

## 3. 工程补充：用相对关系理解物体随部件移动

以下是对本课主题的概念解释，沿用第 05 节的坐标关系，不代表已核对课程中的绑定实现。

假设以手部作为参考，物体相对于手部保持一个固定偏移。手部移动时，物体的世界位置也需要随之更新；手部转向时，这个偏移的世界方向也会变化。

设手部的世界位置为 $\mathbf p_{WH}$，手部坐标系到世界坐标系的旋转矩阵为 $R_{WH}$，物体参考点在手部坐标系中的固定位置为 $\mathbf p_{HO}$，则物体参考点的世界位置为：

$$
\mathbf p_{WO} = \mathbf p_{WH} + R_{WH}\mathbf p_{HO}
$$

其中，`W` 表示世界，`H` 表示假设选用的手部参考，`O` 表示物体。固定的是按手部坐标轴表达的相对偏移，物体的世界位置会随着手部运动而变化。两组世界位置直接相减，得到的仍是按世界坐标轴表达的位移，需要结合手部朝向才能换成局部坐标。

如果还要求物体朝向随手部一起变化，就需要描述相对姿态。**位姿**包括位置与姿态，单独保持位置偏移不能完整描述两者的固定空间关系。这里的手部只是解释用的参考，课件没有指定实际绑定部件。

坐标关系说明物体怎样跟随参考部件，实际承载效果还取决于物理实现。通过更新显示位置让物体跟随、通过约束连接刚体、通过接触实现抓持，对动力学的影响各不相同。仅凭画面中物体跟随机器人，不能认定其重量和接触力已经作用于机器人。课件未说明采用哪种方式，本篇不补写具体接口或抓取流程。
