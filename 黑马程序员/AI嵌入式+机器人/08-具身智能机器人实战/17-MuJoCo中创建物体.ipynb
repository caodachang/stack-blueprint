## 2. 场景效果：T800 与红色木块同处一个世界



*原课件第 2 页：左侧写明“动态创建一个红色木块”，右侧展示 T800、棋盘格地面与红色方块。*

本课要添加的对象是一个红色木块。示例中，它出现在机器人附近的地面区域，与 T800 同处一个仿真场景。学习关注点由机器人自身，扩展到世界中的其他物体。

课件用“动态创建”描述这一目标，但没有展开采用哪种接口、何时创建以及怎样更新模型。图中可以确认的是木块已经出现在场景里；单张画面不能说明它是否正在运动、是否参与物理碰撞，或机器人是否已经与它交互。

整理或复现本课时，首先应核对物体是否成功加入正确场景，显示位置、大小和颜色是否符合预期。课件没有给出这些数值，不能从画面比例推定木块尺寸或世界坐标。


## 3. 工程补充：区分形状、外观与运动自由度

联系第 02 节的方块仿真和后面的模型结构，理解“创建木块”时可以分别看以下内容。这里是概念补充，不代表课件使用了下表中的具体实现。

| 内容 | 在 MuJoCo 模型中的常见表达 | 需要分清的含义 |
|---|---|---|
| 刚体与位置 | `body` 及其位姿 | 决定物体所属层级与初始放置关系，位置需要结合父坐标系理解 |
| 方块形状 | `geom`，类型为 `box` | `size` 的三个分量是沿各轴的半尺寸，完整边长需乘以 2 |
| 红色外观 | 几何体的 `rgba` 或材质设置 | 颜色影响显示，不能仅凭红色判断质量、摩擦等物理属性 |
| 可自由运动 | 独立顶层刚体上的 `freejoint` 等自由度设置 | 自由关节允许刚体平移和转动；未配置关节的刚体相对父级固定 |

“动态创建”描述创建过程，“物体能否自由运动”取决于模型的自由度，二者需要分别确认。在已有场景中增加参与物理计算的刚体，也涉及模型结构；不能只凭画面出现一个方块，就认定物理模型已同步增加。具体创建和编译方式，应以课程实际代码为准。

木块的名称同样不能代替物理配置。若要验证它的下落、接触或推动效果，还需核对自由度、质量与碰撞设置，并推进仿真观察。这些操作未在本课已读取的页面中展开，本篇不补写成原课步骤。
