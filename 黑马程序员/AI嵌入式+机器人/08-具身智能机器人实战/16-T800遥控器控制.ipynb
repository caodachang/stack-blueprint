## 3. 工程补充：把键盘输入接到用户命令

结合第 12 节的观测方案，遥控输入可以通过用户命令（Command）表达期望运动。该方案中的 3 维命令分别描述前后移动、左右移动和转向的期望速度，随后与身体状态历史一起提供给策略。

在这条控制过程中，操作者给出运动意图，策略结合身体反馈生成关节 Action，再按第 13 节的 `DEFAULT_Q + action × ACTION_SCALE` 转换为目标角，由后续控制环节跟踪。这样可以理解键盘输入与关节动作之间的联系。这里是对前面概念的串联，尚未核对第 16 节的实际程序是否按这一方式实现。

接入或检查键盘控制时，先核对以下约定：

| 待核对项 | 需要确认的内容 |
|---|---|
| 按键映射 | 每个键修改哪个命令分量，正负方向是什么 |
| 速度与范围 | 使用固定速度、逐次增减还是其他规则，是否限制命令范围 |
| 按下与松开 | 按住时怎样更新，松开后是否归零，停止命令怎样触发 |
| 输入是否到达策略 | 键盘事件更新后，策略实际收到的 Command 是否同步变化 |

可以沿用第 14 节的日志思路，把按键事件、对应命令及机器人的实际运动对照起来。命令中的期望速度与机器人实际速度需要分别观察；收到按键或命令数值改变，本身还不足以证明机器人已经按要求运动。课件未提供上述约定的具体值，因此这里不补写按键表或控制代码。
