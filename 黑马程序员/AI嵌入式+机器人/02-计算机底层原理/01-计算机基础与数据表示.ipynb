## 2. 课程要做什么，为什么从底层学起

计算机是一种能够按规则执行计算和处理信息的设备。CPU 是其中执行指令和处理数据的核心部件，内存保存程序与数据，输入输出设备负责与外部交换信息。

课程选择从简单的组件开始，一层层组合：先认识数字电路，再搭建运算和存储部件，最后组合成一个简易的 **8 位 CPU 与计算机系统**。观察每一步的数据变化，比只记住“CPU 是计算机的大脑”更有帮助。

## 3. 计算机的发展：从计算工具到电子系统

课程先介绍算盘、机械计算设备、打孔卡和早期电子计算机。这些例子共同说明：人们不断把计算规则交给机器，以提高计算速度、重复性和自动化程度。

| 发展阶段 | 代表性思路 | 对理解本课的作用 |
|---|---|---|
| 人工与辅助计算工具 | 算盘、计算表等帮助人计算 | 数值可以用不同符号和工具表达 |
| 机械计算 | 齿轮、机械传动实现计算步骤 | 运算规则可以由物理结构实现 |
| 通用计算构想 | 差分机、分析机等设计 | 把运算、存储和控制分工组合起来 |
| 打孔卡处理 | 用孔的位置记录信息 | 数据可以用有限的物理状态编码 |
| 早期电子计算机 | 继电器、真空管实现电路状态变化 | 电子器件替代缓慢的机械动作 |
| 晶体管与集成电路 | 在更小空间内集成大量器件 | 相同基础功能可以做得更小、更快、更可靠 |

ENIAC 于 1946 年公开亮相，是早期通用电子数字计算机的重要代表。理解历史时，要区分“电子”“可编程”“通用”等条件，不能把不同类型设备都简化成同一种“第一台计算机”。

**本课最重要的历史结论是：复杂计算机可以由简单功能逐层组合出来。** 制造工艺和系统结构都在发展，但运算、存储、控制、输入输出这些基础问题一直存在。

> 资料说明：讲义中的部分历史日期和“第一台”表述采用了简化讲法；本篇保留发展脉络，不把这些简化说法当作精确的计算机史年表。“器件变小”也不等于现代 CPU 的结构完全没有变化。


## 4. 数值、数字与位值计数法

**数值**是数量本身；**数字**是表示数值的符号。中文“五”、阿拉伯数字 `5`、罗马数字 `V` 可以表达相同的数值。换一种写法，并没有改变数量。

位值计数法把一个数拆成“每一位的数字 × 这一位的权重”。同一个数字放在不同位置，贡献的数值不同。

### 4.1 十进制：位置决定 1、10、100……

讲义以 `263` 为例：

$$
(263)_{10}=2\times10^2+6\times10^1+3\times10^0=200+60+3
$$

另一个例子 `1045`：

$$
(1045)_{10}=1\times10^3+0\times10^2+4\times10^1+5\times10^0
$$

十进制有 `0—9` 十个数字，每向左移动一位，位权乘以 10。中间的 `0` 不能随意省略：它占住一个位置，确保其他数字的位权正确。

### 4.2 一般形式

对于基数为 $r$ 的非负整数：

$$
(a_{n-1}\cdots a_1a_0)_r=\sum_{i=0}^{n-1}a_i r^i
$$

右端第一位的下标是 0，因为它的位权是 $r^0=1$。这一篇先讨论非负整数；负数的补码表示在第 03 篇处理。

位值计数法的好处是：只需要一组有限的数字符号和一致的进位规则，就能表达很大的数，也方便按列计算。


## 5. 计算机为什么采用二进制

二进制只有 `0`、`1` 两个数字，**逢二进一**。电路比较容易可靠地区分两种状态，因此可以用这两种状态表示一个二进制位。

讲义使用“接近 0 V 表示 0，接近 5 V 表示 1”的例子。关键是把一定范围内的电压解释为逻辑状态，而不是要求每次测到的电压都恰好是 0 V 或 5 V。

这样做的好处是：轻微的电压波动仍可能落在同一个逻辑范围内；而如果要在同样电压范围内精确区分十种状态，可靠识别会更困难。

> 补充边界：5 V 是课程中的示例。实际数字电路也可能使用 3.3 V 等电平，识别范围由器件规定；0、1 是逻辑含义，不是所有器件统一的电压值。

一个二进制位称为 **bit（位）**。一个 bit 只能取两种值，但多个 bit 组合起来能表达更多状态：

| 位数 | 可用组合 | 组合数量 |
|---|---|---|
| 1 位 | `0`、`1` | 2 |
| 2 位 | `00`、`01`、`10`、`11` | 4 |
| 3 位 | `000` 到 `111` | 8 |
| $n$ 位 | 全 0 到全 1 | $2^n$ |

**能表达多少种状态，和这些状态分别代表什么，是两个问题。** `01000001` 可以作为一个整数，也可以按 ASCII 码表解释为字母 `A`。


## 6. 二进制与十进制怎样互相转换

### 6.1 二进制转十进制：相加所有为 1 的位权

从右向左，二进制的位权依次为：

```text
位权：128  64  32  16   8   4   2   1
下标：  7   6   5   4   3   2   1   0
```

讲义以 `1011` 为例：

$$
(1011)_2=1\times8+0\times4+1\times2+1\times1=(11)_{10}
$$

讲义的思考题 `10110111` 可以按同样方法计算：

$$
(10110111)_2=128+32+16+4+2+1=(183)_{10}
$$

不要把 `1011` 直接当成十进制的一千零一十一。写成 $(1011)_2$ 或明确说明“二进制”，可以避免歧义。

### 6.2 十进制转二进制：分解位权

把 `183` 分解为二进制位权之和：

```text
183 = 128 + 32 + 16 + 4 + 2 + 1

位权：128  64  32  16   8   4   2   1
取值：  1   0   1   1   0   1   1   1
结果：10110111
```

每个位置只能放 0 或 1，不需要这个位权就填 0。

### 6.3 补充方法：除以 2，余数逆序读

以 `13` 为例：

| 除法 | 商 | 余数 |
|---|---:|---:|
| 13 ÷ 2 | 6 | 1 |
| 6 ÷ 2 | 3 | 0 |
| 3 ÷ 2 | 1 | 1 |
| 1 ÷ 2 | 0 | 1 |

从最后一个余数向上读，得到 `1101`。反向检查：$8+4+1=13$。

### 6.4 左侧补 0 与整数范围

对于本篇的无符号整数，`1011` 和 `00001011` 的数值都等于 11，但后者明确占用 8 位。

$$
n\text{ 位无符号整数范围：}\quad0\le x\le2^n-1
$$

8 位有 **256 种组合**，最大值是 **255**。数量和最大编号相差 1，因为计数从 0 开始。


## 7. 8421 位权与 8421 BCD 码

课程视频标题包含“二进制与 8421 码”，讲义用 `8、4、2、1` 说明四个二进制位置的权重。这里要分清两个相关概念。

### 7.1 四位二进制的位权

四位二进制从左到右的位权是 `8、4、2、1`。例如：

$$
(1011)_2=8+2+1=11
$$

四位二进制可以表示 `0—15`，`1010—1111` 都是有效的二进制整数。

### 7.2 补充区分：8421 BCD 是逐位编码十进制数字

**BCD（二进制编码的十进制数）**为每个十进制数字单独分配四个 bit。8421 BCD 中，一个四位组只用于表示 `0—9`。

以十进制 `39` 为例：

| 表示方法 | 写法 | 解释 |
|---|---|---|
| 普通二进制 | `00100111` | $32+4+2+1=39$ |
| 8421 BCD | `0011 1001` | 前一组是十进制数字 3，后一组是数字 9 |

两种表示对应同一个数值，但编码规则不同。如果把 `0011 1001` 当普通二进制整数解释，就会得到 57。

> `1011` 是有效的四位二进制整数 11，但不是一个有效的 8421 BCD 十进制数字。不要看到“8421”就默认整串数据使用 BCD。


## 8. 二进制加法：规则相同，进位基数不同

十进制相加是逢十进一，二进制相加是逢二进一。按列加法时，既要加这一列的两个数字，也要加右边一列传来的进位。

| 计算 | 当前位结果 | 向左进位 |
|---|---:|---:|
| 0＋0 | 0 | 0 |
| 0＋1 / 1＋0 | 1 | 0 |
| 1＋1 | 0 | 1 |
| 1＋1＋1 | 1 | 1 |

### 8.1 原课程例题：183＋19



上图来自讲义第 24 页。用 8 位写成：

```text
    10110111    183
  + 00010011     19
  ----------
    11001010    202
```

从最右列开始：

1. `1＋1＝10`：当前位写 0，向左进 1。
2. 下一列 `1＋1＋进位 1＝11`：当前位写 1，继续进 1。
3. 再下一列 `1＋0＋进位 1＝10`：当前位写 0，继续进 1。
4. 按相同规则继续，得到 `11001010`。

检查位权：$128+64+8+2=202$，与十进制的 $183+19$ 一致。

### 8.2 固定位数会限制结果

数学上可以继续增加位数，但电路和存储空间通常有固定宽度。例如 `11111111＋1` 需要九位 `1 00000000`；如果只保留低八位，会得到 `00000000`，还需要单独处理进位。

这说明结果的位宽必须明确。加法器、进位和有符号溢出的具体电路在第 03 篇再分析。


## 9. bit、byte、容量与传输速率

一个 **byte（字节，B）**通常由八个 bit 组成：

$$
1\,B=8\,bit
$$

| 单位 | 常见写法 | 表示什么 |
|---|---|---|
| 位 | bit、b | 一个二进制位置 |
| 字节 | byte、B | 八个 bit |
| 比特每秒 | bit/s、b/s、Mbps | 每秒传输多少 bit |
| 字节每秒 | B/s、MB/s | 每秒传输多少字节 |

### 9.1 容量换算：先明确 1000 还是 1024

讲义用 `KB、MB、GB` 按 1024 递进，这是常见的传统写法。为了避免混淆，本篇把十进制与二进制单位分开：

| 十进制单位 | 数量 | 二进制单位 | 数量 |
|---|---:|---|---:|
| kB | 1000 B | KiB | 1024 B |
| MB | $10^6$ B | MiB | $2^{20}$ B |
| GB | $10^9$ B | GiB | $2^{30}$ B |
| TB | $10^{12}$ B | TiB | $2^{40}$ B |

后面还可以继续到 PB/EB 或 PiB/EiB。实际软件可能仍把二进制容量标成 `MB`、`GB`，比较数值时要先查清它使用的换算规则。

例如，标称十进制容量 `1 TB` 的存储设备，换算成二进制单位约为 `931.3 GiB`；单位不同就会出现数值差异。

### 9.2 100 Mbps 不等于 100 MB/s

如果标称带宽是 `100 Mbps`：

$$
\frac{100\times10^6\,bit/s}{8}=12.5\times10^6\,B/s=12.5\,MB/s
$$

这只是按单位换算的理论上限，实际有效数据速率还受传输开销和网络状态影响。

同理，`10 Mbit/s＝1.25 MB/s`；但 **10 MB/s 本身已经是每秒 10 兆字节，不能再除以 8**。

> 讲义第 25 页把部分下载软件的“10 M/S”统一解释为 bit/s，表述过于笼统。实际必须看界面标注；下载软件显示 MB/s 时就是字节速率，不能仅凭软件名称判断单位。


## 10. 32 位与 64 位：先区分数值范围和地址数量

一个 $n$ 位二进制字段有 $2^n$ 种组合。如果把这些组合用作地址，就可以区分 $2^n$ 个地址；如果每个地址对应一个字节，理论寻址容量就是 $2^n$ 字节。

### 10.1 32 位地址的教学模型

$$
2^{32}=4,294,967,296\text{ 个地址}
$$

按字节寻址：

$$
2^{32}\,B=4\,GiB
$$

地址范围是 `0—4,294,967,295`。**地址数量是 $2^{32}$，最大地址编号是 $2^{32}-1$。**

### 10.2 64 位地址的理论模型

$$
2^{64}=18,446,744,073,709,551,616\text{ 个地址}
$$

在同样的字节寻址假设下，理论容量为 $2^{64}\,B=16\,EiB$。

| 问题 | 对应数量 |
|---|---|
| n 位字段有多少种组合？ | $2^n$ |
| n 位无符号整数最大值是多少？ | $2^n-1$ |
| n 位地址按字节寻址，理论容量多大？ | $2^n$ B |
| 64 位有符号整数的最大正数是多少？ | $2^{63}-1$，约 $9.22\times10^{18}$ |

最后一行只用于解释讲义第 26 页的数字：讲义给出的“64 位最大约 $9.2\times10^{18}$”接近有符号整数的最大正数，**不能直接用作 64 位地址数量**。有符号整数的补码规则在第 03 篇展开。

### 10.3 理论地址空间不等于实际装机内存

地址位宽、每个地址对应的存储单位、处理器实际支持的地址范围、操作系统和硬件配置，共同决定实际可用空间。不能由“64 位电脑”推出机器具有 $2^{64}$ 字节内存，也不能仅凭这几个字断定每次操作都快两倍。

后续搭建简易计算机时，要具体问：地址有几位？一个地址选中多大的一份数据？而不是把地址数量和数据位宽混为一谈。


## 11. 二进制怎样存储，为什么还需要编码规则

纸上的 `0` 和 `1` 只是符号。机器实际保存的是可以被可靠区分的物理状态，再把读到的状态解释成 bit。

例如，可以用电路状态表示 0、1，用电荷、磁性等不同状态保存数据。**物理实现可以不同，读出的逻辑信息仍然可以是一串二进制位。** 锁存器、寄存器和内存的具体实现留到第 04 篇。

课程用亮灭和摩斯码说明“有限符号可以表达丰富信息”：亮、灭本身没有文字含义，需要发送者和接收者事先约定对应规则。摩斯码只是帮助理解编码约定的例子，不表示所有计算机字符都直接使用摩斯码。

对计算机来说，还要区分下面三层：

| 层次 | 例子 | 要回答的问题 |
|---|---|---|
| 物理状态 | 高低电平、存储状态 | 能不能稳定读出 0、1？ |
| 二进制数据 | `01000001` | 有多少位，每一位是什么？ |
| 数据含义 | 整数 65、字母 A、像素数值等 | 按什么规则解释这些位？ |

因此，一个文件不仅需要保存数据，还需要有明确的格式、编码或约定。没有解释规则，即使把每个 bit 都读对，也未必能理解内容。


## 12. 字符码表：把文字对应到数值

码表规定“哪个数字代表哪个字符”。存储时把字符转换成数字和字节；显示时按相同规则把字节解释回字符。

### 12.1 ASCII

**ASCII（美国信息交换标准代码）**用 7 位表示 128 个编码位置，包括英文字母、数字字符、标点和控制字符。常见存储中使用一个字节承载一个 ASCII 字符。

| 字符 | 十进制编码 | 用 8 位展示 |
|---|---:|---|
| 空格 | 32 | `00100000` |
| `'0'` | 48 | `00110000` |
| `'3'` | 51 | `00110011` |
| `'A'` | 65 | `01000001` |
| `'a'` | 97 | `01100001` |
| 换行 LF | 10 | `00001010` |

ASCII 的 128 个位置并非都用于可见字形，其中有控制字符；它也不能直接覆盖全部汉字。

### 12.2 数字字符与整数不是同一个对象

在本篇的无符号 8 位整数模型中，数值 `3` 可以表示为 `00000011`。而字符 `'3'` 在 ASCII 中的编码是 51，即 `00110011`。

屏幕上同样看起来是“3”，保存的数据仍可能不同。处理输入时，需要判断它是文本字符还是已经转换过的数值。

### 12.3 汉字编码为什么会有多个标准

课程介绍 GB2312、Big5、GBK 等编码方案。这些方案覆盖的字符和字符对应的字节规则不同：GB2312 主要面向简体中文，Big5 常用于繁体中文，GBK 扩展了中文字符覆盖范围。

当写入时采用一种规则、读取时采用另一种规则，就可能出现乱码。课程这部分需要掌握“码表必须一致”的原则，而不是把不同码表当成完全相同的映射。


## 13. Unicode 与 UTF-8：字符编号和字节写法

不同地区分别制定码表会造成互通问题。Unicode 用统一的**码点**为字符编号，例如：

| 字符 | Unicode 码点 | UTF-8 字节（十六进制） | UTF-8 字节数 |
|---|---|---|---:|
| A | `U+0041` | `41` | 1 |
| 中 | `U+4E2D` | `E4 B8 AD` | 3 |
| 😀 | `U+1F600` | `F0 9F 98 80` | 4 |

表中的十六进制仅用于紧凑地展示字节：两位十六进制数字可以表示一个字节，例如 `FF` 对应十进制 255，即二进制 `11111111`。

### 13.1 两个步骤不能混在一起

- **Unicode 码点**回答：这个字符的统一编号是什么？
- **UTF-8、UTF-16 等编码形式**回答：这个编号怎样写成可以存储的字节？

UTF-8 对一个有效 Unicode 码点使用 1—4 字节；ASCII 范围的字符保持原来的一个字节表示。常用汉字在 UTF-8 中通常占三字节，但不能推广为“所有字符都占三字节”。UTF-16 对一个码点可能使用一个或两个 16 位编码单元。

> 讲义第 31 页把 Unicode 概括为“16 位数字”，又把 $2^{16}$ 说成“超过 100 万个位置”。这里需要订正：$2^{16}=65,536$；现代 Unicode 码点范围是 `U+0000—U+10FFFF`，并不是每个位置都已分配字符，也不是所有字符都固定占 16 位。码点、编码单元和实际字节数要分开看。

### 13.2 补充验证：文字长度与字节长度

下面的 Python 示例用于阅读和验证编码，不是课程原始代码：

```python
text = "A中"
data = text.encode("utf-8")

print(len(text))                 # 2：这个简单字符串有两个字符
print(len(data))                 # 4：UTF-8 实际占四个字节
print(data.hex(" "))             # 41 e4 b8 ad
print(data.decode("utf-8"))      # A中
```

读取时应使用匹配的编码；不能靠“一个汉字总占两字节”去计算 UTF-8 文本容量。


## 14. 黑白图像：把像素按顺序写成二进制

**像素**是图像网格中的一个位置。课程先用一个 `6×6` 的黑白图示范：约定 `0＝黑，1＝白`，按从左到右、从上到下的顺序保存每个像素。



上图来自讲义第 32 页，像素数据为：

```text
111111
101101
111111
101101
100001
111111
```

把它恢复成图像时，需要知道四件事：宽度为 6、高度为 6、读取顺序以及黑白与 0/1 的对应规则。只有 36 个 bit 而没有这些约定，不能唯一确定显示结果。

### 14.1 理论像素数据量

每个像素只需要 1 bit：

$$
6\times6\times1=36\,bit
$$

紧密打包到整字节存储时，至少需要：

$$
\left\lceil\frac{36}{8}\right\rceil=5\,B
$$

36 bit 在算术上等于 4.5 字节，但常规存储分配不能分配“半个字节”；最后一个字节的未使用位需要补齐。

### 14.2 实际格式可能有额外开销

如果每行单独对齐到一个字节，六行会需要 6 B；文件还可能有宽高、调色信息、对齐或其他头部数据。上面的 5 B 是紧密打包情况下的像素载荷，不是任意图片文件的实际大小。


## 15. 灰度图像：每个像素保存一个亮度值

黑白图只有两种状态。灰度图用更多数值表示从黑到白的不同亮度。课程采用常见的 **8 位灰度**：

- 0 表示黑。
- 255 表示白。
- 中间数值表示不同灰度。

八个 bit 有 $2^8=256$ 种组合，因此这里是 **256 级灰度**，不是 255 级。



图来自讲义第 33 页。对于课程中的 `28×28` 灰度图，每个像素一个字节：

$$
28\times28\times1\,B=784\,B
$$

把每个像素的亮度依次写下来，就得到一份数值矩阵。图片与矩阵是同一批数据的两种呈现方式。

讲义还用 LED 的 PWM 调亮度帮助理解“数值对应亮度”。要区分两个步骤：**图像数据记录亮度值，显示电路负责把数值转换为可见亮度**；PWM 不是所有灰度图片文件的存储格式。


## 16. 彩色图像：每个像素保存 R、G、B

课程用 **RGB** 表示彩色像素：R 是红色分量，G 是绿色分量，B 是蓝色分量。这里讨论每个分量占 8 位的 RGB 数据。

| 颜色 | RGB 数值 |
|---|---|
| 黑 | `(0, 0, 0)` |
| 白 | `(255, 255, 255)` |
| 红 | `(255, 0, 0)` |
| 绿 | `(0, 255, 0)` |
| 蓝 | `(0, 0, 255)` |

### 16.1 为什么每个像素占 3 B

$$
3\text{ 个分量}\times8\,bit=24\,bit=3\,B
$$

三个分量都可以取 `0—255`，每个分量有 256 种取值，理论组合数为：

$$
256^3=16,777,216
$$

> 讲义第 34 页用 `0—255` 说明颜色取值。计算数量时必须把 0 算进去，因此是 $256^3$，不能写成 $255^3$。

### 16.2 原始像素容量

假设采用紧密排列、无额外通道的 8 位 RGB：

$$
\text{像素数据量}=\text{宽}\times\text{高}\times3\,B
$$

补充例题：一张 `640×480` RGB 图像需要：

$$
640\times480\times3=921,600\,B=900\,KiB
$$

这个公式只估算原始像素载荷，不包括文件头、行对齐、透明度通道或压缩。因此不能只凭图片分辨率，直接推断 JPEG、PNG 等文件的磁盘大小。


## 17. 声音表示：采样、量化与编码

声音信号随时间变化。课程用“在每个时刻记录振动幅度”解释声音数字化：在一定的时间间隔取样，把每个取样结果转换成数值，再写成二进制数据。



图来自讲义第 35 页。理解它时要区分时间方向和幅度方向：

| 步骤 | 处理什么 | 相关参数 |
|---|---|---|
| 采样 | 在时间轴上选择取样时刻 | 采样率，单位 Hz |
| 量化 | 把幅度映射到有限的数值等级 | 每个样本的位数 |
| 编码 | 把量化结果写成二进制 | 具体数据表示格式 |

### 17.1 采样率、位深、声道数

- 采样率 `44,100 Hz` 表示每个声道每秒有 44,100 个采样时刻，不是声音“响了 44,100 次”。
- 16 位样本可使用 $2^{16}$ 个编码组合；24 位可以提供更多量化等级。
- 单声道有一组样本，双声道通常需要保存左右两组样本。

在其他条件相同的情况下，采样率或位深增加，会增加数据量。提高这些参数也不自动保证音质一定改善，还要看源信号与采集条件。

### 17.2 补充计算：未压缩 PCM 数据量

对紧密排列、无额外填充的 PCM 样本：

$$
S=\frac{f_s\times b\times C\times T}{8}\,B
$$

其中 $f_s$ 是每秒每声道的样本数，$b$ 是每个样本的位数，$C$ 是声道数，$T$ 是秒数。

例如 `44,100 Hz、16 bit、双声道、60 s`：

$$
S=44,100\times16\times2\times60\div8=10,584,000\,B
$$

即约 `10.584 MB` 的样本数据。音频文件可能有文件头、样本填充或压缩，这个结果不能直接套到所有音频格式上。


## 18. 视频表示：图像帧、声音和时间顺序

课程把视频概括为“图片＋声音”：把一张张图片按时间顺序组织起来，再与声音配合播放，就可以表达连续变化。更准确地说，视频画面可以没有声音；有声音的视频还要保存音轨并保持时间同步。

**帧**是一幅画面；**帧率**表示每秒多少帧，例如 30 fps 表示每秒 30 帧。每一帧都可以按本篇讲过的像素规则理解。

### 18.1 课程示例：1920×1080 的一帧

讲义第 36 页使用 `1920×1080` 分辨率：

$$
1920\times1080=2,073,600\text{ 个像素}
$$

假设每帧采用紧密排列的 8 位 RGB，每个像素 3 B：

$$
S_{\text{帧}}=2,073,600\times3=6,220,800\,B
$$

这是 `6.2208 MB` 的原始画面数据。

### 18.2 补充估算：30 fps 的原始数据

$$
\text{每秒画面数据}=6,220,800\times30=186,624,000\,B/s
$$

一分钟约为：

$$
186,624,000\times60=11,197,440,000\,B\approx11.197\,GB
$$

这还没有加入声音、时间信息等开销。计算说明：未经压缩的画面数据量很大。

### 18.3 原始 RGB 模型与实际视频文件

上面的公式只针对明确约定的未压缩 RGB 模型。实际视频可以采用不同像素格式和压缩方法，文件大小往往更适合用**平均总码率 × 时长 ÷ 8**估算。

因此，“1080p、30 fps”只能告诉我们分辨率与帧率，还不能唯一确定文件大小；还要知道像素格式、压缩情况和码率等条件。


## 19. 一页复习表

| 问题 | 应先想到什么 |
|---|---|
| 数值与数字有什么区别？ | 数值是数量，数字是表示数量的符号 |
| 一位二进制为什么只有两种值？ | 二进制只有 0 和 1，每位用两种状态表示 |
| 二进制怎样转成十进制？ | 从右端 $2^0$ 开始，相加所有为 1 的位权 |
| n 位有多少种组合？ | $2^n$；无符号最大值是 $2^n-1$ |
| 8421 BCD 与普通二进制一样吗？ | BCD 逐个编码十进制数字，解释规则不同 |
| 1 B 有多少 bit？ | 8 bit |
| 100 Mbps 理论上对应多少 MB/s？ | 12.5 MB/s，尚未扣除传输开销 |
| n 位地址等于多大容量？ | 先问每个地址对应什么单位；字节寻址时是 $2^n$ B |
| 为什么会乱码？ | 写入与读取的字符编码不匹配等原因 |
| Unicode 等不等于 UTF-8？ | Unicode 提供码点，UTF-8 规定码点的字节写法 |
| 黑白与 8 位灰度有什么区别？ | 每像素 1 bit 与 8 bit，两种状态与 256 级灰度 |
| 8 位 RGB 每像素多大？ | R、G、B 各一字节，总共 3 B |
| 声音数据量取决于什么？ | 采样率、样本位数、声道数、时长与格式 |
| 只知道视频分辨率能算文件大小吗？ | 不能，还缺帧率、像素格式、压缩与码率等条件 |

常用公式及适用条件：

```text
n 位组合数量              = 2^n
n 位无符号最大值          = 2^n - 1
byte                      = bit / 8
字节寻址的理论地址容量    = 2^(地址位数) B
紧密排列的原始像素数据量  = 宽 × 高 × 每像素位数 / 8
未压缩 PCM 样本数据量     = 采样率 × 每样本位数 × 声道数 × 秒数 / 8
未压缩画面数据量          = 每帧数据量 × 帧率 × 秒数
```

需要整字节存储时向上取整；文件头、行对齐、样本填充和压缩另行考虑。计算前先写清“单位”和“表示规则”，可以避免大部分本篇错误。

### 参考答案

1. `183`；`00001101`，位权之和为 $8+4+1=13$。

8. 4 B：A 占 1 B，中占 3 B。字符数量与编码后的字节数量是不同的量。
9. 36 bit；紧密打包后最少 5 B。若逐行对齐，需按格式另算。
10. 784 B；921,600 B，即 900 KiB。两者均不包括文件格式开销或压缩。
11. $8000\times16\times1\times10\div8=160,000$ B。
12. 186,624,000 B/s。实际文件还取决于帧率、像素格式、码率、压缩和是否带声音等因素。
