# 12-大模型基础与部署方式


## 1. 本篇范围与课程背景

本篇依据课程《大模型介绍》的 **14 页 PPT 在线预览**整理，依次介绍人工智能的发展、算力与数据等术语、生成与推理模型、DeepSeek、深度学习的基本原理，以及云端调用和本地部署。

前面的视觉课程讨论了如何从图像中获得物体信息。这一篇转向大模型基础，重点理解模型是什么、通过什么方式使用，以及模型规模怎样影响部署资源。

课件使用的模型名单、测评和价格主要来自 **2024—2025 年前后的资料**。本文按课程内容保留这些历史示例，并就容易误解的地方作简短说明，不将它们视为当前排行榜、价格或采购建议。

资料中没有完整 API 请求代码、Ollama 安装命令、模型启动参数或机械臂接入程序。本篇整理的是概念与部署方式。

## 3. 算力、算法与数据

课件第 2 页提出三个基础要素：**算力、算法、数据**，并用人的能力作类比。

| 要素 | 课件类比 | 在模型中的含义 |
|---|---|---|
| 算力 | 体力、脑力 | 执行训练与模型计算所需的计算资源 |
| 算法 | 思维方式 | 模型结构、学习方法和计算过程 |
| 数据 | 经验和知识库 | 供模型学习或处理的样本与信息 |

类比帮助区分它们的作用：硬件提供计算能力，算法规定怎样计算，数据提供学习材料。三者共同影响模型能学到什么，以及训练和运行需要多少资源。


## 4. 通用模型、多模态、行业模型与 AI 工具

课件第 3 页把常见概念分为以下几类：

| 概念 | 课件中的说明或例子 |
|---|---|
| 通用大模型 | 大语言模型 LLM、视频或语音模型、多模态大模型 |
| 多模态 | 文本、图片、音频等不同类型的信息 |
| 行业模型，也称垂直模型 | 面向教育、医疗、金融等领域 |
| AI 工具 | 豆包、Kimi、腾讯元宝等面向用户的应用 |

LLM 是 **Large Language Model，大语言模型**。多模态描述模型处理的信息类型，行业模型描述应用领域，这两个维度可以同时存在。



原图进一步列出文本模型、实时交互、文生视频、视觉理解、文生图、语音合成，以及医疗、汽车、教育、金融、工业等领域示例。图中品牌和模型用于说明当时的生态分类。

**补充辨析：**模型是提供能力的核心，AI 工具是用户使用这些能力的产品入口。一个工具可能调用多种模型。“大模型”也不只包括语言模型，还可以涉及图像、语音和视频等任务。

第 4 页用“准备期、跃进期、繁荣期、深化期”组织约 `2022 年末至 2025 年初` 的进展。图中注明参考《中文大模型基准测评 2024 年度报告》。这些阶段是课件的概括，下面按其顺序保留主要内容。

| 阶段 | 课件列出的代表进展 | 课程强调的变化 |
|---|---|---|
| 准备期 | ChatGPT、GPT-4；文心一言、通义千问、讯飞星火、360 智脑、ChatGLM | 通用大模型受到广泛关注，各厂商开始集中投入 |
| 跃进期 | Llama 2、GPT-4 Turbo、Gemini、Midjourney 5.2、Stable Diffusion XL；Baichuan、Qwen、InternLM、ChatGLM3、Yi-34B | 开放模型生态发展，文本和图像生成能力持续扩展 |
| 繁荣期 | Sora、GPT-4o、Claude 3.5、Gemini 1.5、Llama 3；可灵、海螺、Vidu、PixVerse；Qwen2.5、文心 4.0、GLM-4、商量 5.5 | 多模态、视频生成和通用模型同步发展 |
| 深化期 | OpenAI o1、Claude 3.5 Sonnet、ChatGPT 实时视频；DeepSeek-R1、QwQ-32B-Preview、Kimi-k1.5、GLM-Zero、Skywork-o1、StepR-mini、讯飞星火 X1 | 推理、代码、Agent 与实时多模态交互受到更多关注 |

课件也列出豆包、混元、盘古、AndesGPT、BlueLM、Kimi 等产品与模型的更新，以及 DeepSeek-V3、豆包-Pro、混元-Turbo、GLM-4-Plus 等系列的能力提升。

这里要抓住的课程主线是：模型应用从通用内容生成，继续扩展到复杂问题求解和多模态交互。阶段时间是概略划分，不应把每个模型名称所在的区间直接当作其精确发布日期。


## 6. 生成模型与推理模型的比较

课件第 5 页以 GPT-4o 和 DeepSeek-R1 为例，比较通用生成与专门强化推理的模型。下表保留原课的五个比较维度，并将绝对化表述整理为课程中的能力侧重。

| 比较项 | 课件中的 GPT-4o 示例 | 课件中的 DeepSeek-R1 示例 |
|---|---|---|
| 模型定位 | 通用自然语言与多模态任务，侧重对话、内容生成、翻译 | 复杂推理与逻辑任务，侧重数学、编程和问题求解 |
| 推理能力 | 日常语言任务较均衡，复杂推理存在局限 | 课件认为其在部分数学、代码和科学问答基准上更有优势 |
| 多模态支持 | 课件从产品体验层面描述文本、图像、音频、视频相关能力 | 课件所讨论的原始 R1 主要接受文本输入 |
| 应用场景 | 日常对话、内容生成、翻译和多模态信息处理 | 数学竞赛、编程、逻辑分析和科学问题 |
| 用户交互 | 侧重流畅交互与多种输入方式 | 可能展示推理说明，整体回答等待时间可能较长 |

**概念补充：**推理模型同样会生成文本，这两种叫法表达的是能力与训练重点，不是互斥的模型分类。模型运行并产生输出的“推理（inference）”，也不同于这里强调的逻辑求解能力“推理（reasoning）”。

> **接口能力修正：**PPT 将图文、音频、视频处理与生成合并在 GPT-4o 的描述中，容易混淆产品功能和某个 API 模型。OpenAI 官方的 [`gpt-4o` 模型文档](https://developers.openai.com/api/docs/models/gpt-4o)列出的是文本、图像输入与文本输出，不能据此认为调用这个模型就能直接生成音频或视频。

课件关于“展示链式思考”的说法，应理解为产品可能呈现推理说明。可见说明不能当作完整、可靠的内部计算记录，也不能用说明长短判断答案是否正确。

第 6 页列出的推理模型示例包括 **DeepSeek R1、OpenAI o1、OpenAI o3-mini、Gemini 2.0、Grok 3**。这是课程当时的名单，其中系列名称并没有指定全部子型号。


## 7. DeepSeek、V3 与 R1

课件第 7 页介绍深度求索（DeepSeek），并以 V3 对话模型和 R1 推理模型为重点。课件记录的发布时间分别是：

- **DeepSeek-V3：2024 年 12 月 26 日。**采用混合专家架构 MoE，面向通用对话、代码、数学等任务。
- **DeepSeek-R1：2025 年 1 月 20 日。**强调通过强化学习提升推理能力，面向复杂问题求解，并可通过蒸馏将能力迁移到较小模型。

课件将 DeepSeek 公司介绍为 2023 年 7 月成立、与幻方量化有关联、专注于通用人工智能研究的公司。日常表述中的“DeepSeek”可能指公司、模型家族或应用，具体讨论时应指明 V3、R1 等型号。

| 术语 | 本课中的理解 |
|---|---|
| MoE，Mixture of Experts | 混合专家架构，由多个专家子网络参与模型计算 |
| 强化学习 | 通过反馈或奖励优化模型行为，课件用它解释 R1 推理能力的提升 |
| 蒸馏 | 利用较强模型提供的结果或训练信号，训练较小模型 |

**补充说明：**蒸馏是在训练另一个模型，不等于把原模型文件直接压缩。强化学习也是 R1 训练过程中的重要部分，不能据此认为训练只包含一种方法。

## 9. 大模型怎样根据输入生成输出

第 10 页把大语言模型描述为通过深度学习训练、主要用于处理和生成人类自然语言的系统。

课件强调两个特点：

- **模型规模：**常见模型包含数十亿到数千亿参数。
- **训练数据：**通过书籍、文章、网页等大量文本学习语言规律与知识关联。

使用时，用户给出提示 `prompt`，模型根据输入生成相应输出。提示可以包含问题、任务要求或上下文。

**术语补充：**参数是训练中学习的数值，例如权重。课件把参数类比为“大脑神经元”，这只是便于理解的说法，参数数量不等于神经元数量。`B` 表示 billion，即十亿参数，因此 `7B` 约为 70 亿参数。

学习得到的关联也不保证每次回答都正确。课件这里讲的是输入和输出的总体关系，没有展开 Transformer 结构、分词或解码算法。


## 10. 深度神经网络与成语补全

课件第 11 页介绍深度学习：使用层数较多的人工神经网络，从数据中学习输入和输出之间的映射关系。人工神经网络受到生物神经网络启发，但它本身是一个计算模型。



课件用成语补全作例子：

| 输入 | 期望输出 | 补全结果 |
|---|---|---|
| 朝思暮 | 想 | 朝思暮想 |
| 一心一 | 意 | 一心一意 |
| 脱颖而 | 出 | 脱颖而出 |

把网络理解为一个带有已学习参数的函数，就容易理解“给定输入，计算输出”的过程。这张图用于讲解映射关系，并不是 DeepSeek 的实际网络结构图。

> **课件表述修正：**PPT 写网络“能够完成任何输入到输出的转换”。实际能否学到并可靠完成某项任务，仍取决于网络结构、数据、训练与任务条件，不能理解为任意网络训练后都能解决所有问题。


## 11. 云端调用：通过 API 使用模型

第 12 页开始介绍部署方式，先讨论云端调用。课件给出两条路径：

1. 直接使用官方或第三方已经提供的模型 API，将模型服务接入业务。
2. 在云平台创建、部署或微调模型，再通过 API 调用部署后的服务。

API 是程序调用服务的接口。课件列出的接口服务例子有 DeepSeek 官方、硅基流动，云平台例子有腾讯云、华为云、百度智能云、阿里云、火山引擎、无问芯穹。

云端方式的主要优势是：用户不必先购置用于运行模型的计算硬件，可以按需使用远程服务。

原表标题为“第三方平台 DeepSeek-R1 稳定性测评总榜（网页版）”，日期为 `2025 年 2 月 12—13 日`，来源标为 SuperCLUE。比较项包括版本、完整回复率、截断率、无回复率、准确率和推理耗时。

## 12. 本地部署与完整、蒸馏版本

课件第 13 页介绍本地部署：下载 DeepSeek-R1 完整版或蒸馏模型，使用 **Ollama、vLLM** 等工具加载、运行模型，再借助可视化界面与用户交互。

课件称完整版为“满血版”。蒸馏版本则以 Llama 或 Qwen 为基座训练，列出 `1.5B、7B、8B、14B、32B、70B` 等参数规模。

### 12.1 本地部署的含义

课件强调本地运行的自主性，以及在数据需要留在自有环境中的场景下的价值。相应代价是需要准备运行模型的硬件并维护环境。

**补充说明：**模型和依赖准备完成、推理流程没有远程调用时，才可以离线使用。是否需要高性能 GPU 取决于模型规模、量化方式和速度要求，不能把大型完整模型的资源需求套用到所有小模型。

### 12.2 完整版 R1 的课件估算

课件给出以下数据：

- 参数量为 `671B`。
- 16 位完整版约需 `1300G` 显存。
- 硬件举例为 `4 台 8 卡昇腾 910B 服务器及以上`，或 `2 台 NVIDIA H800 服务器及以上`。

**补充计算：**只估算 16 位权重本身的存储，`671 × 10⁹` 个参数，每个参数 `2` 字节，约为 `1.342 × 10¹²` 字节，即十进制 `1342 GB`。这个量级帮助理解课件中的 `1300G`，但运行还需要缓存与其他内存开销，不能把权重大小当作整套服务的显存需求。

课件没有说明这些服务器的完整显存规格、并行方式、上下文长度或并发要求，所列配置只能作为原课的资源量级示例。


## 13. 蒸馏模型的量化显存与硬件示例

第 13 页以 **Ollama 4 位量化版本**为例，给出如下表格。数值和硬件名称保留课件原意：

| 参数规模 | 课件标注的显存需求 | 课件硬件建议 |
|---|---|---|
| 1.5B | 1.1G | CPU 或 GPU |
| 7B | 4.7G | RTX 4050 |
| 8B | 4.9G | RTX 4050 |
| 14B | 9.0G | RTX 4070 Ti |
| 32B | 20G | RTX 4090 |
| 70B | 43G | 多块高性能 GPU |

**量化说明：**量化使用更低精度表示权重等数值，通常能减少存储和显存占用，但可能影响质量或速度，具体取决于实现。蒸馏改变所训练的模型，量化改变数值表示，二者可以结合使用。

这张表没有交代完整测试条件。实际占用还受到具体模型文件、量化格式、上下文长度、并发和推理框架影响；使用 CPU 时占用的是系统内存，不能把 `1.1G` 一概理解为必须配置独立显存。

课件的作用是展示规模增大时资源需求通常会上升，没有提供足够信息保证表中每种配置在任意负载下都能稳定运行。
