## 2. 从电源到负载：一张总图串起 01—08

任何一个小电路都可以先按下面的路径拆解：

```text
能量来源 → 保护与供电 → 控制/处理 → 定时或放大 → 输出负载 → 返回参考点
```

把课程项目放进去：

| 项目 | 能量来源 | 控制/处理 | 输出负载 | 重点检查 |
| --- | --- | --- | --- | --- |
| 5 V LED 小电路 | 电源 | 电阻限制电流 | LED | 极性、功率、回路完整 |
| 继电器开关 | 电源 | 三极管或 MOS 管 | 继电器线圈 | 驱动电流、续流保护、触点边界 |
| 非接触式电笔 | 电池 | J1 耦合、Q1—Q3 级联 | LED | 感应区、集电极连接、限流和局限 |
| NE555 电子琴 | 低压电源 | RC 振荡、按键选 RB | 喇叭或其他发声器件 | 2/6 脚、7 脚、电源去耦、负载 |
| PCB 项目 | 外部电源 | 原理图和网络 | 板上器件及接口 | 封装、板框、走线、丝印、制造资料 |

先问“电流从哪里来、经过什么、回到哪里”，可以快速发现悬空电源、断路、短路和负载放错位置等问题。


## 3. 四个基本量与单位换算

| 量 | 符号 | 常用单位 | 直观含义 |
| --- | --- | --- | --- |
| 电压 | $V$ | V、mV | 两点之间的电势差，是推动电流的条件 |
| 电流 | $I$ | A、mA、μA | 单位时间通过的电荷量 |
| 电阻 | $R$ | Ω、kΩ、MΩ | 对电流的限制程度 |
| 功率 | $P$ | W、mW | 单位时间消耗或传递的能量 |
| 电荷容量 | $Q$ | C、Ah、mAh | 能够提供的电荷数量 |
| 能量 | $E$ | J、Wh、mWh | 电源实际储存或输出的能量 |

必须熟练使用：

$$V=IR$$

$$P=VI=I^2R=\frac{V^2}{R}$$

$$E(\mathrm{Wh})\approx V(\mathrm{V})\times Q(\mathrm{Ah})$$

`mAh` 是电荷容量，`mWh` 是能量。两个电池即使标称 mAh 相同，电压不同，能量也可能不同。

换算时先统一单位：$1\text{ kΩ}=1000\text{ Ω}$，$1\text{ mA}=0.001\text{ A}$，$1\text{ μF}=10^{-6}\text{ F}$。数量级错一个前缀，电流和功率计算就会错三个数量级。


## 4. 串联、并联、分压与节点检查

### 串联

串联元件通过同一电流：

$$R_{总}=R_1+R_2+\cdots$$

电压按阻值分配。一个电阻断路，整条串联路径通常失效。

### 并联

并联元件两端电压相同：

$$\frac1{R_{总}}=\frac1{R_1}+\frac1{R_2}+\cdots$$

两个电阻并联时：

$$R_{总}=\frac{R_1R_2}{R_1+R_2}$$

并联支路的一条短路，可能把另一条支路或电源拉入过大电流。

### 分压

对于 $R_1$ 接电源、$R_2$ 接地的分压器：

$$V_{out}=V_{in}\frac{R_2}{R_1+R_2}$$

如果输出端接入负载，负载会与 $R_2$ 形成并联，不能继续使用未带负载的简单公式。分压器适合参考和测量，不能默认当作稳定电源。

### 节点法快速检查

1. 选一个参考地。
2. 给每个独立网络命名。
3. 沿网络检查是否闭合、是否意外相连。
4. 在关键节点写出流入电流和流出电流的关系。

基尔霍夫电流定律：节点处流入电流代数和等于流出电流代数和。基尔霍夫电压定律：闭合回路的电压代数和为零。它们是排查“图上相连但数值不对”的基础工具。


## 5. 储能元件：电容、电感和时间变化

电容不喜欢两端电压瞬间改变：

$$i_C=C\frac{dv_C}{dt}$$

电感不喜欢通过它的电流瞬间改变：

$$v_L=L\frac{di_L}{dt}$$

因此：

- 电容常用于去耦、耦合、滤波和定时。
- 电感常用于滤波、储能、磁场转换和 LRC 振荡。
- 有极性电容必须按极性、耐压和纹波条件使用。
- 继电器线圈断电时会产生反向感应电压，驱动器件两端通常需要合适的续流通路。

RC 时间常数：

$$\tau_{RC}=RC$$

RL 时间常数：

$$\tau_{RL}=\frac{L}{R}$$

“经过约一个时间常数就充满”只是粗略说法。电容充电到约 63.2% 是 $1\tau$，到约 95% 约为 $3\tau$，到约 99.3% 约为 $5\tau$。

仿真和实测都要观察时间轴、初始条件、元件模型和探头负载，不能只看最后一个时刻的电压。


## 6. 电阻与负载选型：从目标反算，不从外观猜

以 LED 限流为例，串联电阻的初步计算为：

$$R\approx\frac{V_{CC}-V_F-V_{SW}}{I_{LED}}$$

其中 $V_F$ 是 LED 正向压降，$V_{SW}$ 是三极管、MOS 管或其他开关的压降近似值。

例：5 V、LED 约 2.0 V、开关压降约 0.2 V、目标 4 mA：

$$R\approx\frac{5-2.0-0.2}{0.004}=700\,\Omega$$

可选接近的标准值 680 Ω，再检查：

$$P_R=I^2R\approx0.004^2\times680=10.9\text{ mW}$$

阻值不能只看“算出来的最接近值”，还要检查最大电压、最坏情况电流、功率余量、温升、公差和可购性。

对电阻、电容、三极管、MOS 管、继电器和喇叭，统一使用以下选型顺序：

1. 明确工作电压、电流、频率和环境条件。
2. 查完整型号手册中的额定值和推荐工作范围。
3. 计算最坏情况，而不是只算标称情况。
4. 选封装并核对引脚、散热、机械尺寸。
5. 把实际型号写进 BOM、原理图和装配说明。


## 7. 三极管与 MOS 管：先判断用途，再判断工作区

### 三极管

NPN 低端开关的典型路径：

```text
VCC → 负载 → 集电极 C
                 发射极 E → GND
GPIO → 基极电阻 → 基极 B
```

基极电阻的初步估算可以从目标集电极电流和强制放大倍数出发：

$$I_B\ge\frac{I_C}{\beta_{forced}}$$

$$R_B\approx\frac{V_{GPIO}-V_{BE}}{I_B}$$

这是开关设计的估算，不等于任何型号在所有温度和电流下都能满足的保证。若基极电流超过 GPIO 能力，应改变器件或增加驱动级。

NPN 与 PNP 的判断核心是基极相对发射极的电压方向，不是只看符号朝向。截止、放大、饱和三种状态要结合端电压和电流判断。

### MOS 管

MOS 管主要由栅源电压控制。$V_{GS(th)}$ 只是开始有小电流的阈值，不等于在目标电流下充分导通。选型要查看目标 $V_{GS}$ 下的 $R_{DS(on)}$、电流、功耗、封装和热阻。

低端 N 沟道和高端 P 沟道是常见入门结构。高侧 N 沟道方案通常需要额外的栅极驱动，不能只把栅极接 GPIO 就认为可用。

### 三极管与 MOS 管的共同检查

- 控制端电压是否达到工作条件。
- 负载电流和启动电流是否超出能力。
- 关断时是否存在感性反冲。
- 共享地和回流是否造成控制信号抖动。
- 器件是否在允许的功耗和温度范围内。


## 8. 继电器驱动：控制回路与触点回路分开看

继电器至少包含两部分：线圈控制回路，以及 COM、NO、NC 组成的触点回路。线圈一侧的低压控制不会自动决定触点另一侧的电压等级和安全性。

典型低端驱动：

```text
VCOIL → 继电器线圈 → Q/MOS 管 → GND
                         ↑
                       GPIO 控制
```

检查清单：

- 线圈额定电压与电源匹配。
- 驱动器件的电流和功耗有余量。
- 线圈两端有合适的反向保护路径，二极管方向与正常供电方向相反。
- GPIO 上电、复位、悬空时不会误吸合。
- 触点额定电压、电流、负载类型和间距满足实际用途。
- 控制侧与触点侧的隔离要求已经明确。

继电器的触点不是“无限大的开关”，感性负载、浪涌和频繁动作会缩短寿命。若只是低压高速开关，MOS 管通常更合适；若需要机械隔离或多路触点，继电器更方便。


## 9. 测量与仿真的统一方法

### 万用表

| 要测什么 | 表笔与档位 | 电路状态 | 核心禁忌 |
| --- | --- | --- | --- |
| 电阻 | 电阻档，两表笔跨接 | 断电，电容先放电 | 不要在带电电路测阻值 |
| 电压 | 电压档，并联被测两点 | 通电 | 不要把电压档当电流档串入 |
| 电流 | 电流档，串入回路 | 通电，限流范围正确 | 测完及时移回电压插孔 |
| 通断 | 蜂鸣档 | 断电 | 大电容未放电时读数可能异常 |

测量前后都做一次“电源、档位、插孔、表笔、量程”五项检查。测量结果要连同参考点、供电值、负载和时间记录，单独写一个“5 V”没有复现价值。

### 仿真

仿真用于验证趋势和关系，例如：

- 电阻增大，分压点如何变化。
- 电容充放电时间如何变化。
- 继电器线圈断电时的保护效果。
- NE555 的 2/6 节点和 3 脚 OUT 是否同步翻转。

仿真模型不是实物。模型可能没有包含引脚寄生、温升、接触电阻、噪声、负载变化和器件批次差异。应先提出预期，再运行仿真，最后写出偏差原因。


## 10. NE555：一套公式覆盖电子琴和定时练习

标准无稳态接法中，$R_A$ 位于 VCC 与 7 脚之间，$R_B$ 位于 7 脚与 2/6 定时节点之间，定时电容接到地：

$$t_H=\ln(2)(R_A+R_B)C$$

$$t_L=\ln(2)R_BC$$

$$f=\frac1{\ln(2)(R_A+2R_B)C}$$

$$D_H=\frac{R_A+R_B}{R_A+2R_B}$$

必须把公式中的 C 指向定时电容。电子琴原图中 C2=104=100 nF；C1 是控制端电容，C3 是电源去耦，C4 是输出耦合，它们不能互换。

NE555 的引脚复核：1=GND、2=TRIG、3=OUT、4=RESET、5=CONT、6=THRES、7=DISCH、8=VCC。2 与 6 共接定时节点；RESET 低电平会强制输出关闭；7 脚负责放电路径。

以 $R_A=1$ kΩ、$C=100$ nF、$R_B=15$ kΩ 为例：

$$f\approx\frac1{0.693(1+30)\times10^3\times100\times10^{-9}}\approx465.4\text{ Hz}$$

实际频率受阈值、放电开关、元件公差、电源和负载影响。计算值是预期值，不是自动获得的实测值。


## 11. 音频、耦合和负载：看见波形还不等于声音正确

方波输出经过串联电容送入发声器件时，电容主要完成直流隔离；它不会提高 NE555 的输出电流能力。

若负载近似为电阻，耦合电容和负载形成的转折频率可粗略写成：

$$f_c\approx\frac1{2\pi R_{load}C_{couple}}$$

真实喇叭是随频率变化的阻抗，不能只用一个 8 Ω 数值预测响度和音色。检查声音问题时按顺序看：

1. OUT 是否有正确频率和幅度的波形。
2. 耦合电容是否容量、极性和耐压合适。
3. 发声器件是有源蜂鸣器、无源蜂鸣器还是喇叭。
4. 负载电流是否超过芯片、供电和驱动级能力。
5. 供电去耦和回流是否把定时节点拉乱。

有源蜂鸣器内部带振荡器，不能用来任意复现每个 NE555 音调。换发声器件必须同步修改器件说明、封装、BOM 和验收标准。


## 12. 电笔项目的综合复核

电笔的核心逻辑是：变化电场通过微小电容耦合到 J1，输入信号经过 Q1、Q2、Q3 的级联增强，LED 支路由电池供能。

第 69 页网络表：

| 检查点 | 正确关系 |
| --- | --- |
| 输入 | J1 → Q1 基极 |
| 级间 1 | Q1 发射极 → Q2 基极 |
| 级间 2 | Q2 发射极 → Q3 基极 |
| 前级集电极 | Q1、Q2 集电极 → 电池正端 |
| 末级集电极 | Q3 集电极 → LED 支路 → 电池正端 |
| 返回路径 | Q3 发射极 → 电池负端 |

验收时特别检查：

- 三个三极管的 B/C/E 引脚和 PCB 焊盘映射正确。
- LED 支路的限流方式符合实际器件条件。
- J1 到 Q1 的输入线短而清楚，感应区没有无意屏蔽。
- 不把“LED 不亮”解释为安全断电证明。
- 测试只使用受控低压条件，记录环境、距离、方向和负载。

电笔响应会受距离、姿态、污染、湿度、人体和周围耦合影响。它是学习用的感应指示项目，不应替代合规验电器和断电验电流程。


## 13. 电子琴项目的综合复核

原图的核心关系：

- R1=1 kΩ，位于 VCC—7 脚，作为 $R_A$。
- S1—S8 选择 7 脚接到电阻链的不同抽头。
- 抽头到 C2 定时节点的串联电阻是 $R_B$。
- C2=104=100 nF，接 2、6 脚共用节点。
- RESET 接 VCC，CONT 通过 C1 接地，C3 做电源去耦。
- OUT 经 C4=4.7 μF 接 8 Ω、0.5 W 标注的 LS1。
- J3 标注 3—9 V，但实际 NE555 型号和发声负载必须单独核对。

课程参数下：

| 按键 | RB | 理论频率 |
| --- | ---: | ---: |
| S1 | 23 kΩ | 307.0 Hz |
| S2 | 21 kΩ | 335.5 Hz |
| S3 | 19 kΩ | 369.9 Hz |
| S4 | 17 kΩ | 412.2 Hz |
| S5 | 15 kΩ | 465.4 Hz |
| S6 | 13 kΩ | 534.3 Hz |
| S7 | 11 kΩ | 627.3 Hz |
| S8 | 10 kΩ | 687.0 Hz |

八个频率不同，只能证明参数产生不同振荡周期，不能自动证明是准确标准音阶。若目标是标准音高，应先反算总 RB，再确定链段、阻值、公差和校准方法。


## 14. 原理图阅读：从“看图”变成“列网络表”

面对一张陌生原理图，按照下面的顺序读：

1. **找电源入口和参考地。**标出正、负端，确认每个有源器件是否得到供电。
2. **找输入与输出。**输入可能是传感片、按键、GPIO 或接口；输出可能是 LED、继电器、喇叭或另一模块。
3. **按网络名追踪。**沿导线读，不按元件在页面上的距离读。
4. **区分功能网络。**控制信号、功率路径、定时节点、复位、去耦和保护分别列出。
5. **核对脚号和极性。**符号图形不能代替引脚手册。
6. **把每条线翻译成一句话。**例如“7 脚通过 R5 接 C2 定时节点”“Q3 控制 LED 支路回流”。
7. **最后检查未连接端。**确认悬空是设计意图还是漏线。

推荐表格：

| 网络名 | 起点 | 终点 | 中间器件 | 作用 | 需测量的量 |
| --- | --- | --- | --- | --- | --- |
| VCC | 电源正端 | IC/负载供电脚 | 去耦、电源开关 | 提供能量 | 对 GND 的电压 |
| GND | 电源负端 | 各参考点 | 地线/铜区 | 返回电流、参考 | 与各节点的电压 |
| SIGNAL | 输入 | 控制/处理端 | 电阻、电容、晶体管 | 传递信息 | 幅度、频率、偏置 |
| LOAD | 驱动端 | 负载 | 限流、耦合、保护 | 传递功率 | 电流、温升、波形 |


## 15. EDA 到制造：五个结果要分别验收

一个 PCB 项目至少有五个不同结果：

| 结果 | 验收问题 |
| --- | --- |
| 原理图正确 | 网络、脚号、参数、供电和保护是否合理？ |
| 封装正确 | 符号脚号是否对应真实焊盘、方向和尺寸？ |
| PCB 连通 | 所有必需网络是否连接，是否存在短路或漏线？ |
| 规则通过 | 间距、线宽、孔径、板边和工艺规则是否满足厂家能力？ |
| 功能验证 | 通电、测量、负载、温升和目标行为是否达到预期？ |

推荐流程：

```text
原理图 → 电气检查 → 绑定封装 → 转换 PCB → 定板框与摆放
       → 关键回路布线 → 铺铜 → 连通检查 → DRC
       → 3D/丝印检查 → Gerber/钻孔预览 → BOM/坐标核对 → 低压上电
```

改原理图后要导入变更，检查旧走线和新增网络；移动元件或改线后要重新铺铜；导出制造资料后再看一次板框、丝印、孔位和极性。

DRC 通过只说明几何和规则层面没有发现问题，不代表频率、灵敏度、音量、供电或安全性正确。


## 16. PCB 布局的优先级

布局顺序决定返工量。建议按以下优先级：

1. **机械与人机：**板框、安装孔、按键、连接器、喇叭和外壳开孔。
2. **安全与边界：**高低压分区、可触及位置、板边、隔离距离。
3. **高影响小回路：**去耦、定时节点、传感输入、晶体管基极或 MOS 栅极。
4. **功率与回流：**负载电流、继电器线圈、喇叭回流和电源入口。
5. **普通信号：**按网络完成连通和可制造走线。
6. **丝印与外观：**位号、极性、接口、版本和操作标记。

电笔 J1 附近的铜皮会影响耦合，电子琴 C2 与 2/6 脚需要短而清楚的定时回路；这两个项目不能套用同一套“整板铺铜”经验。

线宽要按实际电流和厂家能力设置。短而窄不一定是问题，长而窄、回流绕远以及负载电流穿过敏感参考点才更容易造成故障。


## 17. 故障排查：从现象追到最早异常点

排查不从“换元件”开始，而从现象、供电和关键节点开始。

### 通用顺序

1. 断电检查极性、焊桥、虚焊、连接器方向和封装脚号。
2. 限流上电，观察供电电压、静态电流和温升。
3. 选择一个最靠近输入的关键节点，再选择一个输出节点。
4. 判断异常首次出现在哪一级：输入、偏置、放大/定时、驱动还是负载。
5. 每次只改一个变量，记录改动前后结果。

### 典型现象到优先项

| 现象 | 先查什么 |
| --- | --- |
| 全板无反应 | 电源极性、GND、保险/保护、主开关、供电脚 |
| LED 始终亮 | 输入区漏电、焊桥、晶体管脚号、控制端偏置 |
| 继电器不吸合 | 线圈电压、驱动电流、续流二极管方向、GPIO 状态 |
| NE555 不振荡 | 2/6 节点、7 脚、RESET、C2、R1、供电下限 |
| 电子琴各键同音 | 按键端组、抽头焊桥、电阻链顺序、C2 |
| 所有频率一起偏移 | C2 实际值、共同电阻、电源、测量方法 |
| OUT 正常但无声 | C4、发声器件类型、负载、电源电流、驱动能力 |
| DRC 通过但实物异常 | 引脚映射、封装方向、BOM 型号、焊接和功能条件 |

测量到“某点是 0 V”只是事实，不是原因。还要问该点本来应处于什么状态、上一级是否正常、仪器是否改变了电路。


## 18. 安全边界与上电流程

### 上电前

- 确认电源电压、极性、最大电流和器件手册要求。
- 检查是否有外部市电、充电器、继电器触点或大电容储能。
- 断电测量关键电源轨是否短路。
- 有极性元件、三极管、MOS 管、IC 缺口和连接器方向逐一确认。
- 电源串入限流或使用受控电源，先从低风险状态开始。

### 上电后

- 先看静态电流和温升，再接入大负载。
- 任何器件快速发热、冒烟、异味或电流异常时立即断电。
- 用单手、绝缘探头和合适量程处理需要特别防护的测量；学习项目优先使用隔离低压电源。
- 不把低压验证结果外推为市电安全结论。

### 断电后

- 等待并测量储能元件电压，确认安全后再触碰和改线。
- 记录故障前状态，避免在不清楚改了什么的情况下反复上电。

非接触式电笔项目只做低压功能学习，不安排裸露市电操作。继电器触点的另一侧可能连接危险电压，低压控制侧安全并不等于触点侧安全。

## 21. 自测题与参考答案

### 先独立作答

1. 为什么分压器接入负载后，原来的分压公式可能不再准确？
2. 电容和电感分别反对哪一种瞬间变化？
3. 选 MOS 管时，为什么不能只看 $V_{GS(th)}$？
4. 继电器线圈为什么需要反向保护路径？
5. 测量电阻前为什么必须断电并释放电容储能？
6. NE555 无稳态电路中，哪个电容代入频率公式？电子琴图中它的值是多少？
7. 电子琴所有按键频率一起偏高，优先检查哪些共同参数？
8. DRC 通过但实物不振荡，说明 DRC 能保证功能正确吗？
9. 电笔的三个集电极是否全部接电池正端？
10. 为什么电笔 J1 区域不能机械照搬电子琴的整板铺铜？
11. 电子琴 OUT 有波形但没有声音，至少列出三项检查。
12. 项目修改原理图后，为什么不能直接导出旧 PCB 的制造资料？

### 参考答案

1. 负载会与下方电阻并联，改变等效电阻和输出电压；分压器不是理想电压源。
2. 电容反对两端电压瞬间改变，电感反对通过它的电流瞬间改变。
3. 阈值只表示小电流开始导通；还要查看目标栅源电压下的 $R_{DS(on)}$、电流、功耗和热条件。
4. 线圈断电时的磁场塌缩会产生反向感应电压，可能击穿或扰乱驱动器件；保护路径为能量提供受控回路。
5. 带电测量会让仪表或人体承担不应有的电流；电容残压还可能造成触电、打火或读数异常。
6. 是定时电容 C2，课程电子琴图标为 `104`，即 100 nF。
7. 先查 C2 实际值和单位、R1/R5 等共用电阻、电源、器件型号和测量方法，再查每个按键独有的链段。
8. 不能。DRC 主要检查几何和规则；引脚映射、供电下限、参数、焊接和功能仍需独立验证。
9. 不是。Q1、Q2 集电极接正端，Q3 集电极在 LED 支路下方，由 LED 支路接到正端。
10. 接地铜可能屏蔽或改变微小电场耦合，影响 J1 输入灵敏度；感应区需结合结构验证。
11. 查 C4 的数值、极性和连接，发声器件类型与阻抗，NE555 输出条件，供电能力，驱动级和回流/去耦。
12. 旧 PCB 可能保留旧网络和走线；应导入变更、复查连通、重建铺铜、更新丝印和 BOM，再重新导出同版本资料。
