# 01-电学与电路基础


## 1. 本篇学习范围

本篇对应课程视频 **01—10**：背景介绍、电流、电路和电流、直流与交流、电压、电压电流小实验、欧姆定律、电阻选型和安全电压。

学习目标是先建立一条完整的理解链：

```text
电源提供能量 → 电路形成闭合路径 → 电压推动电荷运动 → 产生电流
                                      ↓
                           电阻限制电流并消耗能量
```

这篇先解决“电路中各个量是什么、怎样测量和计算”。电阻、电容、电感、三极管和 NE555 等器件会在后续笔记中展开。


## 2. 电路由什么组成

**电路**是电流能够沿着闭合路径流动的系统。最基本的电路通常包括四部分：

| 部分 | 作用 | 常见例子 |
|---|---|---|
| 电源 | 提供电能，形成电压差 | 电池、直流电源 |
| 负载 | 把电能转成光、热、声或机械能 | 灯泡、电阻、电机、蜂鸣器 |
| 导线 | 提供电流通路 | 铜线、PCB走线 |
| 开关 | 接通或断开电路 | 按键、继电器触点 |

电路能正常工作通常要同时满足：电源存在、路径闭合、负载连接正确。开关断开或导线断裂时，电路成为**开路**，电流无法形成连续通路；两点被低电阻路径直接连在一起时，可能形成**短路**，电流会异常增大。

> 判断一个电路时，先沿着电源正端、负载和回到电源负端的路径走一遍，比只看元件名称更可靠。


## 3. 电流：电荷的定向移动

**电流**表示电荷定向移动的强弱，用符号 `I` 表示，单位是安培（`A`）。常用换算为：

- `1 A = 1000 mA`
- `1 mA = 1000 μA`

电路分析中默认使用**约定电流方向**：从电源正极经外部电路流向负极。金属导体中的自由电子实际运动方向与约定电流相反，但画原理图、列公式和读仪表时使用的是约定电流方向。

电流不是“凭空产生”的。只有在存在闭合路径，并且路径两端有电压差时，电荷才会形成有组织的运动。电流大小还会受到电路阻碍程度的影响，这个关系由欧姆定律描述。

### 3.1 电流的直观判断

- 导线中没有闭合回路：通常没有持续电流。
- 负载电阻变小而电源电压不变：电流可能增大。
- 多个负载串联：各处电流相同，但电压会按负载情况分配。
- 多个支路并联：电流会分到各支路，总电流等于支路电流之和。

串并联的具体计算放到后续“电路测量与仿真”笔记中，本篇先掌握这些方向性的判断。


## 4. 电压：推动电流的原因

**电压**是两点之间的电势差，用符号 `U` 表示，单位是伏特（`V`）。电源的作用之一，就是在两个端点之间建立电压差。

可以把电压理解成推动电荷运动的条件，把电流理解成电荷实际运动的结果：

- 有电压但回路断开：可能有电压，不能形成持续电流。
- 回路闭合但没有有效电压差：通常没有持续电流。
- 同一条导线上理想情况下电压降很小，负载两端通常存在主要电压降。

### 4.1 电压的参考点

电压是“两点之差”，不能脱离参考点单独谈某一点的电压。电路图中的 `GND` 常作为零电位参考点；说“某点是 5 V”，通常是指该点相对于 `GND` 为 5 V。

### 4.2 电压怎样测量

电压表或万用表的电压档要**并联**在被测元件或两测试点之间。测量前要确认档位、量程和表笔插孔，不能把电压档直接串到电源回路中。

电压有方向性。若红表笔接在电势较高的一端、黑表笔接在较低的一端，读数为正；表笔反接时，数字万用表通常显示负号。


## 5. 直流电与交流电

### 5.1 直流电（DC）

直流电的电流方向保持不变，电压极性也保持不变。实际电源的电压可能有波动，但方向通常不随时间反复改变。电池、USB供电和许多嵌入式系统的电源属于直流供电。

### 5.2 交流电（AC）

交流电的大小和方向随时间变化，常见波形包括正弦波、方波和三角波。家庭插座提供的是交流电，频率和电压等级由所在地区的电力系统规定。

| 对比项 | 直流电 | 交流电 |
|---|---|---|
| 方向 | 通常保持不变 | 随时间变化 |
| 常见来源 | 电池、USB、电源模块 | 电网、交流信号源 |
| 典型用途 | 芯片、传感器、单片机供电 | 输电、交流电机、信号处理 |

直流和交流不是“安全”和“危险”的同义词。安全性取决于电压、电流、频率、接触条件、持续时间和环境等因素。


## 6. 电压、电流小实验应该观察什么

课程中的小实验用于把抽象的电压和电流变成可观察的现象。观察时建议固定下面的顺序：

1. **先看电路是否闭合**：开关、导线和负载是否形成完整回路。
2. **再确认电源极性和量程**：直流电源的正负端不能接反，仪表量程先选较大档。
3. **分别观察电压和电流**：电压测量看两点之间的差值，电流测量需要让电流经过仪表。
4. **改变一个条件**：例如改变电源电压或负载阻值，再记录读数变化。
5. **用公式解释结果**：读数变化应能与 `U`、`I`、`R` 的关系对应起来。

### 6.1 电流表和电压表的接法

| 仪表 | 接入方式 | 目的 | 易错点 |
|---|---|---|---|
| 电压表 | 并联 | 测两点电势差 | 不要把电压档直接串入电源回路 |
| 电流表 | 串联 | 测流过某支路的电流 | 不要把电流档直接并到电源两端 |

如果不确定被测量的大小，优先使用较大量程，再逐步换到合适量程；换挡前应先断开或停止测量，避免超量程损坏仪表。


## 7. 欧姆定律：把电压、电流和电阻联系起来

在电阻性负载、工作条件稳定的范围内，欧姆定律为：

$$
U = IR
$$

三个常用变形是：

$$
I = \frac{U}{R}, \qquad R = \frac{U}{I}
$$

| 量 | 符号 | 单位 | 含义 |
|---|---|---|---|
| 电压 | `U` | `V` | 负载两端的电势差 |
| 电流 | `I` | `A` | 流过负载的电流 |
| 电阻 | `R` | `Ω` | 对电流的阻碍程度 |

### 7.1 例题：根据电压和电阻求电流

已知一个电阻两端电压为 `12 V`，阻值为 `1 kΩ`：

$$
I = \frac{12V}{1kΩ} = 0.012A = 12mA
$$

注意单位要统一：`1 kΩ = 1000 Ω`。计算得到的电流是 12 mA，而不是 12 A。

### 7.2 适用边界

欧姆定律适合描述电阻或近似电阻性元件在一定工作范围内的关系。二极管、三极管、灯丝等元件的电压电流关系可能是非线性的，不能不加判断地把任意元件都当作固定电阻。


## 8. 用欧姆定律进行电阻选型

当已知电源电压和希望通过负载的电流时，可以先用下面的式子估算限流电阻：

$$
R = \frac{U_R}{I}
$$

其中 `U_R` 是电阻两端实际承担的电压，不一定等于整个电源电压。例如负载本身已经占用一部分电压时，应先扣除负载压降，再计算电阻值。

### 8.1 选型步骤

1. 明确电源电压和负载需要的工作电流。
2. 确定电阻需要承担的电压。
3. 用 `R = U_R / I` 算出理论阻值。
4. 从常见标准阻值中选择接近值，并重新估算实际电流。
5. 检查电阻功耗是否有余量。

电阻功率可以用以下任一形式计算：

$$
P = UI = I^2R = \frac{U^2}{R}
$$

例如上面的 `12 V、1 kΩ` 电阻：

$$
P = \frac{12^2}{1000} = 0.144W
$$

实际选型不能只看阻值，还要看额定功率、允许误差和工作环境。额定功率应高于实际功耗，并留出可靠余量。


## 9. 安全电压与用电边界

课程中的“安全电压”用于提醒：电压越高，人体接触时形成危险电流的可能性通常越大；但不能把某个固定数值简单当成所有环境下的绝对安全线。

人体是否受到伤害，还与以下因素有关：

- 电流是否经过胸部、心脏等关键部位；
- 接触时间长短；
- 皮肤干湿、接触面积和接触压力；
- 交流或直流的类型与频率；
- 环境是否潮湿，以及是否有良好绝缘和保护措施。

### 9.1 实际操作原则

- 不带电拆装电路，先断电、放电，再确认无电。
- 不用身体去“试电”，不要用手同时接触两个不同电位点。
- 测量前检查表笔插孔、档位和量程，电流档尤其要谨慎。
- 面对市电、高压电容、未知电源和大功率设备时，停止实验并交给具备资质的人员处理。
- 低压电路也可能因短路产生大电流、发热和起火，不能只看电压数值。

> 安全规则优先于读数和进度。无法确认电路状态时，先断电并查清楚。

| 电流是什么？ | 电荷的定向移动，电流表串联测量 |
| 怎样由电压和电阻求电流？ | `I = U / R`，先统一单位 |
| 怎样估算限流电阻？ | `R = U_R / I`，再检查标准值和功率 |
| 直流和交流差异？ | 主要看电流方向和大小是否随时间变化 |
| 为什么不能把电流档并到电源两端？ | 仪表内阻很小，可能形成近似短路 |
| 低压是不是绝对安全？ | 不是，还要看电流、路径、时间和环境 |

### 关键公式

```text
U = I × R
I = U / R
R = U / I
P = U × I = I²R = U²/R
```


## 11. 自测题

### 题目

1. 一个 `5 V` 电源接在 `1 kΩ` 电阻两端，电流是多少？
2. 一个电阻中流过 `20 mA` 电流，两端电压为 `10 V`，阻值是多少？
3. 电压表和电流表分别应该怎样接入被测电路？
4. 为什么开关断开时，电路两端可能仍有电压，但负载中没有持续电流？
5. 选择电阻时，为什么除了阻值还要检查功率？

### 参考答案

1. `I = 5 V / 1000 Ω = 5 mA`。
2. `R = 10 V / 0.02 A = 500 Ω`。
3. 电压表并联，电流表串联；换挡和接线前先确认量程与表笔插孔。
4. 断开后回路不闭合，负载中没有连续的电荷运动；但电源仍可能在断点两侧建立电势差。
5. 电阻会消耗功率并发热，实际功耗超过额定功率会导致温升过高、阻值漂移甚至损坏。
