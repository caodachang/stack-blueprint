## 2. PCB 是什么：提供连接，也提供安装位置

PCB 是 Printed Circuit Board，即**印制电路板**。常见电路板由绝缘基材与导电铜层组成，按设计加工出走线、焊盘和孔，为器件提供电气连接与机械支撑。

| 对象 | 主要表达什么 |
|---|---|
| 原理图 | 哪些器件端子连接在一起，电路怎样工作 |
| 网表/网络信息 | 引脚、焊盘属于哪些电气网络 |
| PCB 布局与布线 | 器件放在哪里，实际铜线和孔怎样实现连接 |
| 裸 PCB | 加工后的板子，尚未完成元件装配 |
| PCBA | 装有元件的电路板组件，通常还需相应检测 |

### 单层、双层与多层

- 单面板通常只有一层导电铜层。
- 双面板有顶层和底层铜层，可通过合适的金属化孔连接。
- 多层板在内部还包含铜层，用于更多信号、电源或地的连接。

铜层数与元件安装面数是两个概念。双层板可以只在一面安装器件；也不能因为两面都有元件，就直接认定它只有两层铜。

对本课程的低压基础电路，先理解简单单面或双面板的设计方法即可，复杂高速与大功率布板不在本篇展开。


## 3. 常见结构与图层：每一层都有具体用途

| 结构或图层 | 作用 | 容易混淆的地方 |
|---|---|---|
| 基材 | 支撑、绝缘，FR-4 是常见类型之一 | 材料和厚度影响机械与电气特性 |
| 顶层/底层铜 | 形成走线、焊盘与铜区 | 屏幕上看见一条线，不代表它就是铜线 |
| 焊盘 Pad | 焊接器件端子，或连接通孔器件 | 形状、尺寸、脚距和孔径要匹配实物 |
| 过孔 Via | 让同一网络在不同铜层间连接 | 普通机械孔不自动提供电气连接 |
| 阻焊 Solder Mask | 覆盖需要保护的区域，并在焊接位置开窗 | 开窗区域表示相应位置露出铜面 |
| 锡膏层 Solder Paste | 指定贴片焊接时钢网开口等信息 | 主要服务于贴片装配，不能代替铜层或阻焊层 |
| 丝印 Silkscreen | 标注编号、极性、接口和版本 | 通常不承担电气连接 |
| 板框 Board Outline | 指定外形、尺寸与相关切割轮廓 | 在普通绘图层画矩形，未必会成为制板边框 |
| 装配/机械信息 | 表达器件轮廓、结构与装配说明 | 是否参与制造要看导出规则和厂商要求 |

### 通孔与贴片

通孔器件的引脚穿过板子，再与焊盘连接；贴片器件的端子焊在表面焊盘上。入门手工焊接时，应结合器件尺寸和工具选择封装。

金属化孔 PTH 的孔壁有导电金属，可连接相应铜层；非金属化孔 NPTH 的孔壁不提供该导电连接。用于固定螺丝的孔，还需检查螺丝头、垫片与周边铜的空间。


## 4. PCB 怎样制造：从覆铜板到成品裸板

下面概括常见双面 PCB 的制造环节，帮助理解工艺限制。不同板型与制造工艺的具体顺序会有差异。

| 环节 | 主要目的 |
|---|---|
| 工程资料与 CAM 处理 | 核对铜层、板框、孔和生产参数，准备加工数据 |
| 开料与钻孔 | 准备板材，形成通孔、过孔或其他孔结构 |
| 孔金属化与电镀等处理 | 让需要导电的孔壁形成可靠连接 |
| 图形成像、图形电镀/蚀刻 | 按工艺配合，保留所需铜图形并去除多余铜 |
| 阻焊与丝印 | 形成保护层、焊接开窗和文字标识 |
| 表面处理 | 保护外露铜面并满足后续焊接需求 |
| 外形加工 | 按板框加工板子的最终形状 |
| 电气与外观检查 | 检查相关开路、短路和制造缺陷 |

多层板还涉及内层图形、层间对准、压合等环节，本篇先建立整体认识。

### 为什么绘图要有规则

- 线太细或间距太小，可能难以稳定制造。
- 孔径、焊盘与环宽不匹配，可能影响孔壁连接和焊接可靠性。
- 铜太靠近板边、丝印压住焊盘、板框不闭合，都可能影响加工或装配。

制造商公布的“最小加工能力”是工艺边界，不代表每种电流、耐压或手工焊接场景都适合采用最小尺寸。设计还应留合理余量，并满足电路自己的要求。


## 5. EDA 软件负责什么：从电气设计到生产资料

EDA 是 Electronic Design Automation，即电子设计自动化。立创/嘉立创 EDA 提供原理图、元件库、PCB、规则检查和制造文件等功能；不同工具也可完成类似工作。

| 阶段 | 软件中的主要对象 | 要检查的结果 |
|---|---|---|
| 电路定义 | 原理图、符号、网络 | 连接关系符合电路原理 |
| 器件关联 | 型号、封装、引脚映射 | 与采购实物一致 |
| 板级实现 | PCB、焊盘、走线、过孔 | 网络和物理位置正确 |
| 制造导出 | Gerber、钻孔、BOM、坐标 | 与最终设计版本一致 |

### 一开始就建立工程与版本

给工程和文件明确命名，例如 `LED_5V_V1`；原理图、PCB、BOM 和输出文件应能对应到同一个版本。

修改元件型号、网络或板形后，重新保存、更新 PCB 并导出资料。旧的 Gerber 不会因为源工程后来被修改，就自动变成新版本。



## 6. 符号、器件与封装：三者需要一起核对

| 名称 | 含义 | 例子 |
|---|---|---|
| 原理图符号 Symbol | 电气功能、引脚号和引脚名 | NE555 的 1 脚 GND、8 脚 VCC |
| 实际器件/物料 | 要采购和焊接的具体型号 | 某厂商的某种封装 NE555 |
| PCB 封装 Footprint | 焊盘位置、尺寸、编号及相关轮廓 | DIP-8 或 SOIC-8 的焊盘图形 |
| 引脚映射 | 符号引脚号对应哪个焊盘号 | 符号 1 → 焊盘 1 → 实物 1 脚 |

### 引脚“名称”与“编号”不同

名称 GND 表示功能，编号 1 表示该器件的脚号。不能仅因为某个焊盘靠近地线，就把它视为 GND。

同样，一个三极管符号标有 B/C/E，仍需按具体器件的数据手册确认哪个实物引脚对应哪个焊盘。第 05 篇强调的“不同 TO-92 器件脚序可能不同”，在 PCB 上同样适用。

### 相同引脚数不保证可以替换

DIP-8 与 SOIC-8 都可能有八个脚，但焊盘尺寸、脚距和整体形状不同。符号显示的外观也不是按比例的实际安装尺寸。

选择库元件时，同时检查：完整型号、参数、封装、脚号映射和可获取的实物，不只看搜索结果的名称或图片。


## 7. 原理图绘制：先把电气关系画对

课程视频 55 对应立创 EDA 原理图绘制。下面按功能顺序整理，菜单位置以所用版本为准。

1. 创建工程与原理图文件，确定供电、接口和功能模块。
2. 从库中选择实际会使用的器件，查看型号、参数和封装信息。
3. 放置器件并分配唯一编号，例如 R1、C1、U1、J1。
4. 旋转、移动符号，使供电、信号和回流关系容易阅读。
5. 用电气导线、网络标签和电源标识连接端子。
6. 检查数值、极性、未连接引脚和重复编号。
7. 保存，再检查封装和引脚映射，准备转 PCB。

| 常见编号 | 通常表示 |
|---|---|
| R | 电阻 |
| C | 电容 |
| Q | 三极管或其他晶体管 |
| U | 集成电路 |
| D / LED | 二极管或发光二极管 |
| J / P | 接口、连接器，具体命名习惯可能不同 |

编号与器件数值是不同属性：R1 是哪个元件，330 Ω 是它的阻值。不要把同一个编号重复分配给两个独立元件。

### 原理图要方便复核

给接口标注电压和信号名称，把关键电阻值、器件型号与极性显示清楚。供电符号只是网络标识，真实电源仍需通过相应接口接入；仿真时也要有正确的供电模型。


## 8. 导线与网络：看起来碰到，还要确认真的连接

官方文档把电气工具与绘图工具分开：电气导线参与网络连接，普通线段或矩形可能只表示图形。

### 连接的四个检查

- 导线端点是否吸附到引脚连接点，而不是停在旁边？
- 需要相连的交汇处是否形成正确节点？
- 只是交叉而不应相连的两条线，有没有被误连？
- 网络高亮或设计管理器显示的端子集合，是否符合预期？

实际连接以编辑器的网络数据为准，不能只凭缩小后的视觉印象判断。

### 网络标签

同作用域内、具有相同有效网络名称的标签，可以把相隔较远的导线连接到同一网络。标准版文档说明其标签支持跨图页连接；其他工具或层次化工程仍应确认标签作用域。

例如，在电源接口和芯片供电脚分别放置 `VCC_5V`，可以表达它们属于同一网络。要留意名称拼写和软件的命名规则，避免误把两个独立电源标成同名。

### 有意未连接与遗漏连接

未使用且允许悬空的引脚，可用正确的非连接标志说明设计意图。芯片要求连接的供电、复位或控制脚，不能通过添加非连接标志来掩盖遗漏。

看网络与看电路功能要结合起来：软件识别出一组连接，不代表这组连接就是正确的电路。


## 9. 封装核对：先看实物尺寸，再看编号映射

| 核对项 | 要回答的问题 |
|---|---|
| 脚号 | 符号、封装与实物 1 脚是否一致？ |
| 脚距 | 焊盘中心距是否匹配引脚？ |
| 焊盘/孔 | 孔径、焊盘大小是否适合引脚与加工要求？ |
| 器件轮廓 | 本体会不会挡住邻近器件或板边？ |
| 极性与方向 | LED、电解电容、二极管、IC、连接器朝向是否正确？ |
| 安装空间 | 是否留出插线、螺丝、工具和焊接空间？ |

### 尺寸与单位

```text
1 inch = 25.4 mm
1 mil = 0.001 inch = 0.0254 mm
100 mil = 2.54 mm
10 mil = 0.254 mm
```

“mil”是千分之一英寸，不是毫米。常见 DIP 器件或排针会使用 2.54 mm 脚距，但要以所选器件的尺寸图为准。

贴片封装编号也要核对公制、英制。例如英制 0603 常对应约 `1.6×0.8 mm` 的器件，不要只凭“0603”四个字符判断所有命名体系的尺寸。

### 在封装管理器中绑定

嘉立创 EDA 标准版官方文档建议通过封装管理器关联封装，而不是只在属性里手动填一个封装名称。管理器会检查封装是否存在、引脚编号与焊盘编号是否能匹配。

它能发现部分映射错误，但无法单独证明库图形与采购实物完全一致。仍要结合数据手册、尺寸图和实物核对；必要时可做 1:1 打印辅助比对，并确认打印没有缩放。


## 10. 原理图转 PCB 与导入变更

首次生成 PCB 前，先检查原理图和封装。标准版官方文档给出的入口是“设计 → 原理图转 PCB”，其他版本寻找相应生成/更新功能。

转换后通常可以看到：元件封装、网络信息、初始板框和飞线。

### 飞线是什么

飞线提示哪些同网络焊盘还需要建立物理导电连接。它是连接需求的显示，不是已经加工出来的铜线。

- 放好器件后，按网络完成走线、过孔或铺铜连接。
- 隐藏飞线只改变显示，不会完成布线。
- 不应只用“画面里没有飞线”作为完成依据，还要运行网络/未布线检查。

### 修改原理图后怎样处理

```text
修改原理图 → 保存 → 更新 PCB/导入变更
                         ↓
检查新增、删除、封装与网络变更
                         ↓
复核已有走线 → 重建铺铜 → 重新检查
```

官方文档提示，导入变更后已有走线可能保留，某些网络名称或对应关系可能改变；不能认为软件会自动把所有旧布线修成新电路。

重大修改前保留可恢复的工程版本。修改元件型号后，还要重新确认封装和脚序；只更新阻值或标题，也要核对 BOM 与生产输出是否来自最终版本。


## 11. 定板框与摆元件：先满足结构，再缩短关键路径

PCB 设计中的“布局”主要指确定元件位置和方向，“布线”主要指用铜实现网络连接。先完成合理布局，再布线更容易。

### 先确定机械约束

- 板子的长宽、外形和安装位置。
- 电源接口、插座、按键、指示灯需要面向哪里。
- 固定孔与周边器件、走线、铜区的距离。
- 外壳、螺丝、线缆、器件本体和工具需要的空间。

板框要在工具识别的边框层中形成正确闭合轮廓，避免重叠线段、断口或多余边框。板子有槽或开孔时，按工具及厂商要求表达，不只画一个看起来像孔的圆。

### 建议的布局顺序

```text
板框和固定孔
      ↓
接口、按键、指示灯等位置受约束的器件
      ↓
主芯片与各功能模块
      ↓
去耦、定时、反馈或驱动等关键外围器件
      ↓
其余器件与丝印
```

有明确位置约束的器件可在确认后锁定，减少后续误移动。器件方向尽量利于连接、装配和识别，不能只追求排列整齐。

### 联系第 06 篇的 NE555 电路

如果以后把 NE555 振荡电路做成板子，应让供电去耦电容靠近电源与地引脚；定时电容和 2/6 脚的连接尽量短，R_A、R_B 围绕相关节点合理放置。需要较大电流的负载回路，也要有适当的回流路径，避免影响定时与信号节点。

这些是与前文相关的布局补充，并非本课程视频中某个具体板子的尺寸或摆放坐标。


## 12. 布线：按网络建立完整的导电路径

布线时，先确认焊盘属于哪个网络，再把需要连接的节点接起来。

| 内容 | 初步原则 |
|---|---|
| 关键连接 | 先完成供电、回流和对长度/噪声敏感的连接 |
| 线宽 | 根据电流、铜厚、允许温升和工艺选择，不只使用默认值 |
| 间距 | 同时满足制造和电气要求，不同电压或用途可能不同 |
| 过孔 | 换层时使用匹配网络的过孔，控制尺寸和数量 |
| 转角 | 基础板常用 45° 转角以便布线、观察和加工 |
| 回流 | 电流流出后还要能返回电源，不能只看信号或正电源线 |

### 换层与交叉

- 同层不同网络的铜线不能直接交叉接触。
- 不同铜层可以投影交叉，但不因此自动连通。
- 需要电气换层时，通常通过同网络的金属化过孔或合适的通孔焊盘连接。

不要为了让飞线消失，就把导线或焊盘的网络随意改成相同；连接需求应来自正确的原理图。

### 线宽不是通用承流保证

例如 `10 mil=0.254 mm`，这个数值只表示几何宽度。它能承受多少电流，还与铜厚、内外层、长度、温升、散热和连接处的细颈有关。

电源线、负载线和细小信号线可以使用不同规则；选择好规则后，再检查焊盘入口、过孔和铺铜连接处是否成为局部瓶颈。

对于本篇简单低压电路，先人工理解每条网络和回路。自动布线可以辅助，但结果仍需检查连接、路径与装配空间。


## 13. 铺铜与接地：铜区也必须有正确网络

铺铜是在某一铜层的区域内生成铜面，常用于地或电源连接。一个看起来很大的铜区，只有实际连到正确节点，才实现预期电气功能。

### 常见 GND 铺铜步骤

1. 指定需要铺铜的铜层与区域。
2. 将其网络设置为 GND。
3. 设置间距、板边距离和焊盘连接方式。
4. 生成或重建铜区。
5. 检查地焊盘、地过孔和铜区是否实际连通。
6. 检查孤岛、细颈、断裂和未完成网络。

| 设置或现象 | 含义 |
|---|---|
| 直连焊盘 | 焊盘直接接入铜面，电气与散热条件要核对 |
| 热焊盘/发散连接 | 用若干连接线接入铜面，便于控制焊接热量；连接尺寸仍需匹配电流 |
| 孤岛/死铜 | 没有按预期连接到目标网络的铜块 |
| 地过孔 | 在不同铜层间提供相应地连接 |
| 重建铺铜 | 根据当前走线、器件和规则重新计算铜区 |

顶层与底层的地铜不会因为同名就自动穿过绝缘材料相连，需要真实的层间连接。

### 修改后一定重新检查

标准版官方文档说明，移动器件、修改走线或铺铜属性后，应重建铺铜；标准版常用快捷键为 `Shift+B`，其他版本寻找相应重建功能。

改完 PCB 后使用旧的铜区显示，可能掩盖实际间距或连接问题。重建后重新运行网络和规则检查，再导出最终制造文件。

简单板子优先保持地连接连续、回流路径清楚。随意切割地铜或保留孤岛，不能仅凭画面效果证明能改善噪声。


## 14. 丝印、极性与装配细节

丝印是焊接、接线和维修时的说明书，应该帮助使用者快速找到正确方向。

| 建议标注 | 用途 |
|---|---|
| 元件编号 R1、C1、U1 等 | 与原理图和 BOM 对照 |
| 接口电压及信号名 | 避免接错电源或端口 |
| LED、二极管、电解电容极性 | 指示安装方向 |
| IC 的 1 脚标记或缺口方向 | 辅助识别芯片朝向 |
| 板名与版本号 | 区分不同设计版本 |
| 必要的测试点名称 | 方便测量供电或关键节点 |

### 丝印检查

- 不要压在需要焊接的裸露焊盘上。
- 避免被器件本体完全遮挡，或位于制造时会裁掉的位置。
- 字高、线宽和板边间距满足实际制造要求。
- 接口标签与真实引脚顺序一致。
- 底层文字和器件应按软件的视图与层规则确认方向，导出时不要自行凭感觉镜像文件。

### 外观预览的作用

2D 照片预览与 3D 视图可以帮助发现反向器件、干涉、超出板边和丝印遮挡等问题。但模型正确、画面漂亮，不单独证明封装尺寸、电路功能或规则都正确。

手工焊接还要考虑烙铁进入空间、连接器壳体和大器件遮挡；板子越小，不一定越容易做或越方便维修。


## 15. 原理图检查与 PCB 规则检查：检查不同层面的错误

原理图电气规则检查常称 ERC，PCB 设计规则检查称 DRC。软件提供的具体项目取决于版本、元件电气属性与规则配置；标准版还应查看设计管理器和封装管理器。

| 检查层面 | 重点 | 不能单独证明什么 |
|---|---|---|
| 原理图/电气检查 | 连接遗漏、重复编号、可能的电气类型冲突等 | 器件参数、计算和电路功能一定正确 |
| 封装与映射检查 | 封装是否存在、符号脚号与焊盘是否匹配 | 与采购实物的尺寸和脚序完全一致 |
| PCB DRC | 线宽、间距、孔和板边等是否满足所设规则 | 规则本身适合目标电流、耐压和厂商 |
| 网络/未布线检查 | 必须相连的网络是否完成 | 已连接的电路逻辑就是正确的 |
| 人工及实物复核 | 极性、器件尺寸、机械空间与接口方向 | 替代通电验证和制造检验 |

### DRC 要先设规则，再检查

按制造商当前能力和电路需求设置线宽、间距、孔径、过孔外径和板边距离。只使用软件默认值，可能与实际板厂或负载要求不一致。

标准版文档给出了“设计规则”设置和“检查 DRC”功能；检查结果应逐项定位、理解和处理。

```text
原理图/封装无未解决错误
             ↓
PCB 更新完成、铺铜已重建
             ↓
网络/未布线检查通过
             ↓
按合适规则运行 DRC，处理错误
             ↓
人工核对尺寸、极性、接口与版本
```

“DRC 为零”只是通过所配置的检查，不是功能验收。不能通过减小间距规则、改错网络或隐藏提示来替代修复真正的问题。


## 16. 生产资料：Gerber、钻孔、BOM 与贴片坐标

完成最终检查后，从对应 PCB 版本生成制造资料。嘉立创 EDA 标准版的 Gerber 导出通常得到 ZIP 包，其中包含制造图形和钻孔文件。

| 资料 | 表达什么 | 主要用途 |
|---|---|---|
| 铜层 Gerber | 走线、焊盘、铺铜等铜图形 | 裸板制造 |
| 阻焊 Gerber | 相应表面的阻焊开窗 | 裸板制造 |
| 丝印 Gerber | 表面文字和符号 | 裸板制造 |
| 板框文件 | 外形、相关切割轮廓 | 裸板制造 |
| 钻孔文件 | 孔的位置、尺寸及类型等 | 裸板制造 |
| 锡膏层文件 | 钢网开口图形 | 需要相应贴片装配时使用 |
| BOM | 器件编号、数量、型号与物料信息 | 元件采购和装配 |
| 坐标文件/CPL | 器件位置、旋转和安装面等 | 自动贴片与装配核对 |

### 常见文件后缀示例

标准版官方文档中，顶/底铜层常见 `.GTL/.GBL`，阻焊常见 `.GTS/.GBS`，丝印常见 `.GTO/.GBO`，板框示例使用 `.GKO`，钻孔使用 `.DRL`。

文件名与格式会随工具和输出设置变化。应在查看器中确认每份文件的实际层类型，而不是只凭后缀断定资料完整。

### 导出后的检查

在 Gerber 查看器中重新检查板形、铜层、焊盘、孔、阻焊、丝印和比例；确保它们来自最终设计版本。

Gerber ZIP 中的制造文件不能替代可编辑的原理图、PCB 与库文件。保留源工程和版本记录，后续改板才方便追踪。

裸板加工与装配所需资料不同：如果只做裸 PCB，不表示厂家会自动采购并焊好元件。需要贴片时，BOM、坐标和制造资料必须一致。


## 17. PCB 下单流程：先核对资料，再选择工艺参数

本节对应课程视频 58，整理学习用的下单步骤。平台界面、价格、活动和交付条件以实际页面为准。

```text
完成最终设计检查
        ↓
导出并预览最终 Gerber/钻孔文件
        ↓
选择裸板制造，或明确需要装配服务
        ↓
核对板形与工艺参数
        ↓
按平台流程上传/导入资料并检查解析结果
        ↓
处理工程审核问题，核对报价和交付条件
        ↓
确认订单后进入相应制造流程
```

| 参数或项目 | 核对目标 |
|---|---|
| 层数 | 与导出的铜层及设计一致 |
| 板形、尺寸、数量 | 与需要的版本和机械结构一致 |
| 基材与板厚 | 满足使用、安装和加工要求 |
| 铜厚 | 与电流、热设计和报价相匹配 |
| 阻焊/丝印颜色 | 与标识可读性和生产选择相符 |
| 表面处理 | 符合焊接、保存和成本需求 |
| 工艺限制 | 线宽、间距、孔、板边等满足当前能力 |
| 是否需要贴片 | 明确交付裸板还是相应装配服务 |

### 两种常见资料入口

- 从 EDA 的制造功能关联下单：仍要检查所提交的资料与当前版本一致。
- 导出 Gerber ZIP 后进入制造平台：上传后检查解析出的外形、层和孔。

如果选择 SMT，还需检查 BOM 匹配的完整物料、可用库存、封装、数量、安装面、旋转和极性。官方文档说明元件商城采购与贴片服务的库存、价格可能独立，不能用商城有货推断贴片一定有料。

下单前把最终源工程、制造包和版本号一起归档。对于学习打样，先把电路、封装和文件核对清楚，再看报价与活动条件。


## 18. 练习：把一个 5 V LED 电路做成小板

本节是自行设置的低压练习，**不是原视频电路参数的转录**。目标是练习原理图、封装、布局布线与检查的完整关系。

### 电路与计算

```text
J1 的 +5 V ── R1（330 Ω）── LED1 正极
                               │
                            LED1 负极
                               │
J1 的 GND ─────────────────────┘
```

假设 LED 在目标电流附近的正向压降约 `2.0 V`：

```text
I_LED ≈ (5.0 - 2.0) / 330 ≈ 9.1 mA
P_R ≈ I²R ≈ 0.027 W
```

可以初选功率足够的电阻，例如 `0.25 W` 轴向电阻，再根据实物尺寸选封装。LED 的实际压降与允许电流要查型号资料。

| 编号 | 练习用途 | 封装核对重点 |
|---|---|---|
| J1 | 两针直流电源接口 | 脚距、方向及 +5 V/GND 标注 |
| R1 | 330 Ω 限流电阻 | 实际本体和引脚跨度 |
| LED1 | 发光二极管 | 实物尺寸、脚距和 A/K 极性映射 |

### 原理图阶段

1. 放置三种器件并设置编号、参数和相符的封装。
2. 为电源、LED 前端和地建立清晰网络，可命名为 `VCC_5V`、`N_LED`、`GND`。
3. 核对路径是电源经电阻、LED 回到地，没有把 LED 或电阻短接。
4. 检查库中的 LED 脚号与 A/K 名称；不能把某个库的编号约定推广给所有 LED。

### PCB 阶段

1. 选择满足器件和接口空间的板形，在正确板框层画闭合外形。
2. 把 J1 放在便于接线的板边位置，R1、LED1 按回路摆放。
3. 完成三条相关网络的实际连接；若铺 GND，检查其真实连通。
4. 标出接口电压、地、LED 极性和版本号，检查装配空间。
5. 重建铺铜，完成网络和 DRC 检查，再导出、预览制造包。

练习完成后，记录所选实物型号、板尺寸、规则、检查结果和最终文件版本，便于复核与下一次修改。


## 19. 常见问题与到板后的验证

| 问题 | 排查方向 |
|---|---|
| 原理图转 PCB 失败 | 重复编号、封装缺失、绑定失效、脚号与焊盘号不匹配 |
| 走完线仍有未连接提示 | 端点是否真正接到焊盘？是否换错层、网络或过孔？ |
| 铺铜后地还没接通 | 网络设置、孤岛、分割、层间连接和重建状态 |
| 软件图形正常，元件却装不上 | 采购型号、脚距、本体尺寸、孔径、视图方向与封装匹配 |
| 修改原理图后板子仍异常 | PCB 是否导入变更？旧线、铜区和输出文件是否重新核对？ |
| 厂商识别不到板形 | 板框层、闭合性、重复轮廓与导出内容 |
| LED 不亮或方向错误 | 电源极性、焊点、LED A/K、限流电阻和实际连接 |
| 板子版本混淆 | 源工程、丝印、BOM 和 Gerber 是否统一版本 |

### 制造完成不等于功能验证完成

1. 检查板子与文件是否对应，查看焊盘、孔、板形和文字。
2. 焊接前核对物料，按正确方向装配并检查焊点。
3. 断电检查是否有明显短路、错焊或遗漏。
4. 使用合适的低压供电条件，初次通电时控制电流并观察异常。
5. 测关键电压、电流或波形，把结果与预计行为比较。
6. 记录故障、修改原因和下一版变化，不只在 PCB 上临时补线后忘记更新原理图。

万用表蜂鸣和阻值要结合电路判断，带电容的电路可能出现充电时的短暂读数变化；这不单独证明持续短路。

LED 练习的预计电流约 9.1 mA，实际值受供电、LED 压降和电阻误差影响。练习记录中应分开填写预计值、实测值与检查结果。

| 知识点 | 关键结论 |
|---|---|

| 飞线 | 提示连接需求，隐藏它不等于完成布线 |
| 布局 | 先机械约束与关键器件，再优化连接路径 |
| 走线与过孔 | 网络正确、线宽与间距合适，换层有真实连接 |
| 铺铜 | 必须指定网络并检查连通，修改后重建 |

| 原理图检查/ERC | 检查部分电气问题，结合人工功能复核 |
| DRC | 按已配置规则检查几何与制造相关问题 |
| Gerber 与钻孔 | 为制造提供图形和孔数据 |
| BOM 与坐标 | 为采购、贴片与装配核对提供信息 |
| 版本同步 | 原理图、PCB、BOM、制造包应来自一致版本 |

### 交付前的五问

1. 电路原理与连接对吗？
2. 实物、封装和脚号匹配吗？
3. 结构空间与实际导电路径合适吗？
4. 规则、网络和最终导出文件检查过吗？
5. 生产参数、装配资料与版本一致吗？


## 21. 自测题

1. 原理图和 PCB 分别表达什么？双层铜板是否必须两面都装元件？
2. 铜层、阻焊层、丝印层分别有什么作用？
3. 符号、封装和采购器件有什么区别？八脚器件能否随意使用任意八焊盘封装？
4. 导线在画面中接近引脚，能否据此确认网络连接正确？
5. 飞线是什么？把飞线隐藏之后，电气连接是否完成？
6. 1 mil 等于多少毫米？100 mil 又是多少？
7. 顶、底两面铺铜都设置为 GND，是否自动完成两层地连接？
8. 修改原理图后，为什么要更新 PCB 并复核原来的走线？
9. DRC 没有错误，是否能保证电路工作、封装匹配和装配成功？
10. 只下裸 PCB 订单，是否默认连元件一起焊好？SMT 通常还需什么资料？
11. 5 V 供电，LED 压降约 2 V，串联 330 Ω 时，电流和电阻功率约是多少？
12. 只保存旧 Gerber，能否代替保存可编辑工程和新版本制造包？

### 参考答案

1. 原理图表达电气连接与功能，PCB 表达物理布置和铜连接；铜层数与安装面数不同，双层板可以只在一面装元件。
2. 铜层导电，阻焊覆盖保护并在焊接处开窗，丝印用于文字与符号标识。
3. 符号表示电气端子，封装表示焊盘和尺寸，采购器件是具体实物；引脚数量相同不保证尺寸、映射和方向一致。
4. 不能。应确认端点吸附，并用网络高亮或管理器核对实际连接。
5. 它提示需要实现的网络连接；隐藏只改变显示。
6. 1 mil=0.0254 mm，100 mil=2.54 mm。
7. 不会自动穿过绝缘基材连接，需要相应金属化孔或其他真实层间连接。
8. 导入后旧走线可能保留，网络或器件可能变化；还要重建铺铜并重新检查和导出。
9. 不能。DRC 只检查已配置规则覆盖的项目，还需电路、实物、机械及功能验证。
10. 裸板与装配是不同服务；SMT 通常还需相应 BOM、坐标和一致的制造资料，并核对物料与极性。
11. I≈3/330≈9.1 mA，P_R≈I²R≈0.027 W。
12. 不能。旧 Gerber 表达旧版本的制造图形，也不是完整可编辑工程。
