## 2. 电路分析的四个基本量

在本课程的基础电路中，最常用的量是：

| 量 | 符号 | 单位 | 作用 |
|---|---|---|---|
| 电压 | `U` | `V` | 两点之间的电势差 |
| 电流 | `I` | `A` | 流过支路的电荷运动强度 |
| 电阻 | `R` | `Ω` | 对电流的阻碍程度 |
| 功率 | `P` | `W` | 单位时间转换或消耗的能量 |

它们之间的常用关系为：

$$
U=IR
$$

$$
P=UI=I^2R=\frac{U^2}{R}
$$

做题和调试时先统一单位，再判断是求电流、压降、阻值还是功率。一个数值本身没有意义，必须同时知道测量点、方向、时间范围和电路状态。


## 3. 欧姆定律和焦耳定律

### 3.1 欧姆定律

对于一定工作范围内的电阻性元件：

$$
U=IR
$$

它可以用来计算电流、压降和等效电阻，也可以作为测量结果的合理性检查。例如电阻两端电压增加而阻值不变，电流应按比例增加。

### 3.2 焦耳定律

电流通过电阻时会产生热量，电功率可写为：

$$
P=UI=I^2R=\frac{U^2}{R}
$$

在一段时间 `t` 内消耗的电能为：

$$
E=Pt
$$

若 `P` 用瓦特、`t` 用秒，`E` 的单位是焦耳；若 `P` 用千瓦、`t` 用小时，常用单位是千瓦时（`kWh`）。

**边界：**这些公式描述的是电路量之间的关系，不代表电阻永远保持固定温度和阻值。大功率工作时要考虑温升、散热和额定功率。


## 4. 基尔霍夫定律：从节点和回路检查电路

当电路不再是一个简单的串联或并联时，可以用基尔霍夫定律列方程。

### 4.1 基尔霍夫电流定律（KCL）

对任意节点，流入节点的电流之和等于流出节点的电流之和：

$$
\sum I_{in}=\sum I_{out}
$$

它来自电荷守恒。若一个节点有一条总电流流入、两条支路流出，则：

$$
I_{总}=I_1+I_2
$$

这也是并联电路“总电流等于支路电流之和”的一般表达。

### 4.2 基尔霍夫电压定律（KVL）

沿任意闭合回路走一圈，所有电压升和电压降的代数和为零：

$$
\sum U=0
$$

对一个电源和两个串联电阻，电源电压等于两个电阻压降之和：

$$
U_s-U_1-U_2=0
$$

列式前要先规定电流方向和电压正负方向；方向假设错了并不一定表示电路错，解出的负号会告诉你实际方向与假设相反。


## 5. 戴维宁定理和诺顿定理

讲义把戴维宁定理、诺顿定理列为 Circuit 仿真和电路分析中常见的基础定理。它们的用途是：把一个复杂的线性网络等效成更容易分析的形式。

### 5.1 戴维宁等效

从两个端口看，一个线性电路可以等效为：

```text
理想电压源 Uth + 串联电阻 Rth
```

其中：

- `Uth` 是端口开路时的电压；
- `Rth` 是把独立电压源置零（短路）、独立电流源置零（开路）后，从端口看进去的等效电阻。

接入负载 `RL` 后，负载电流可以写成：

$$
I_L=\frac{U_{th}}{R_{th}+R_L}
$$

### 5.2 诺顿等效

同一个线性网络也可以等效为：

```text
理想电流源 In + 并联电阻 Rn
```

两种等效之间的关系为：

$$
R_N=R_{Th},\qquad U_{Th}=I_N R_N
$$

初学时不必急着用定理处理所有电路。先确认电路满足线性分析条件，并明确端口、独立源置零规则和负载位置。


## 6. mAh 与 mWh：容量和能量不是一回事

电池标签常见 `mAh`（毫安时）和 `mWh`（毫瓦时）。两者描述的是不同的物理量：

| 单位 | 中文 | 主要表示 | 换算关系 |
|---|---|---|---|
| `mAh` | 毫安时 | 电荷容量，电流随时间的累计量 | `Q=I×t` |
| `mWh` | 毫瓦时 | 能量容量，考虑了电压 | `E=U×I×t` |

把电流用安培、时间用小时，得到安时；把电压乘以安时，得到瓦时：

$$
Wh=V\times Ah
$$

因此，同样是 `3000 mAh` 的电池，若标称电压不同，能提供的能量也不同。只比较 mAh 不能直接判断两块不同电压电池谁的能量更大。

### 6.1 示例

一节标称 `1.5 V`、容量 `800 mAh = 0.8 Ah` 的电池，理想能量约为：

$$
E=1.5V\times0.8Ah=1.2Wh
$$

若要比较不同电池、估算设备工作时间，应同时考虑电压、容量、负载功率、转换效率和电池可用放电范围。


## 7. 用功率估算续航时间

如果设备在稳定工作状态下消耗功率 `P_load`，电池可用能量为 `E_bat`，理想工作时间可粗略估算为：

$$
t\approx\frac{E_{bat}}{P_{load}}
$$

若电池通过升压、降压或稳压模块供电，还要考虑转换效率 `η`：

$$
t\approx\frac{E_{bat}\times\eta}{P_{load}}
$$

这个估算不会自动解决以下问题：

- 电池电压会随放电下降；
- 标称容量通常在特定放电条件下测得；
- 负载可能有启动电流和动态变化；
- 保护板可能在剩余能量耗尽前断开；
- 低温、老化和电池内阻会减少可用能量。

所以 mAh/mWh 适合用来比较和估算，不能把标签上的数字直接当作保证续航时间。


## 8. 开关电路：用电路状态控制输出

开关电路的核心是让电路在不同状态之间切换。最简单的开关只有两种状态：

- **导通**：电流可以通过，负载得到电压或电流；
- **断开**：电流路径被切断，负载停止工作。

数字电路也可以把开关状态抽象成高、低电平。继电器、三极管、MOS 管和逻辑门都能实现“用一个控制信号改变另一条电路状态”，只是速度、功率、隔离和驱动方式不同。

### 8.1 机械开关的检查

| 检查项 | 问题 |
|---|---|
| 常开/常闭 | 未按下或未通电时触点处于什么状态？ |
| 额定电压电流 | 触点能否承受负载的稳态和启动电流？ |
| 触点寿命 | 频繁通断是否会加速磨损？ |
| 绝缘与间距 | 高低压之间是否有足够隔离？ |
| 抖动 | 按键触点变化是否需要消抖？ |

开关不只是“把线接上或断开”，还要保证触点不会因电流过大而粘连、发热或拉弧。


## 9. 继电器：用小电流控制另一组电路

继电器是一种用电信号控制触点状态的器件。它通常包含：

1. **线圈**：通电后产生磁场；
2. **衔铁/机械结构**：受到磁力后移动；
3. **触点**：随机械结构切换另一条电路。

继电器的核心价值是：控制侧和触点侧可以分开，能够用低功率控制信号切换较高电压、较大电流或不同电源的负载，并提供一定的电气隔离。

### 9.1 触点名称

常见单刀双掷继电器的端子包括：

| 端子 | 含义 |
|---|---|
| `COM` | 公共端 |
| `NO` | Normally Open，常开触点，线圈未吸合时与 COM 断开 |
| `NC` | Normally Closed，常闭触点，线圈未吸合时与 COM 接通 |

线圈断电时，`COM-NC` 通常导通；线圈通电吸合后，触点切换为 `COM-NO` 导通。具体触点组合以继电器外壳标识和数据手册为准。


## 10. 继电器的关键参数和接线

选继电器至少要分开看两侧参数：

### 10.1 线圈侧

- 线圈额定电压：如 `5 V`、`12 V`；
- 线圈电阻或额定电流：决定驱动电源和驱动器件的负担；
- 吸合电压与释放电压：决定控制信号是否可靠；
- 吸合、释放时间：影响响应速度。

### 10.2 触点侧

- 触点额定电压和电流；
- 负载类型：阻性、感性、灯丝、电机等；
- 触点形式：常开、常闭、转换；
- 电气间隙、爬电距离和隔离要求。

### 10.3 单片机驱动继电器

单片机 GPIO 通常不能直接承担继电器线圈的电流，更常见的结构是：

```text
GPIO → 限流/基极或栅极驱动 → 三极管或 MOS 管 → 继电器线圈
                                              │
                                      线圈两端续流二极管
```

线圈是感性负载，断电时会产生反向感应电压。直流线圈常用续流二极管吸收尖峰；交流线圈的吸收方式不同，不能直接照搬直流接法。


## 11. 继电器仿真

继电器仿真要分别观察控制侧和触点侧：

1. 放置继电器线圈和控制电源；
2. 给线圈串入合适的电阻或驱动器件；
3. 在触点侧放置独立的电源和负载；
4. 初始状态下确认 `COM-NC` 或 `COM-NO` 的连接；
5. 接通线圈，观察衔铁/触点状态改变；
6. 断开线圈，确认触点回到释放状态；
7. 分别测量线圈电流和负载回路电流。

仿真中要明确触点侧是否与线圈侧电气隔离。某些简化模型只展示逻辑状态，不会完整模拟线圈反电动势、触点拉弧、机械抖动和温升。

### 11.1 常见仿真错误

- 把 `COM`、`NO`、`NC` 认反；
- 线圈额定电压与驱动电源不匹配；
- 控制侧和负载侧误接在同一条回路；
- 忘记给感性线圈提供反向电压吸收路径；
- 只看“触点动了”，没有检查负载电流是否超额定值。


## 12. 继电器和三极管/MOS 管的区别

课程后续会专门学习三极管和 MOS 管，这里先做边界区分：

| 对比项 | 继电器 | 三极管/MOS 管 |
|---|---|---|
| 控制方式 | 线圈磁力驱动机械触点 | 半导体器件控制载流 |
| 隔离 | 线圈侧和触点侧可实现电气隔离 | 普通低边/高边开关本身不提供隔离 |
| 速度 | 较慢，有机械动作 | 通常更快 |
| 静态功耗 | 线圈吸合时持续耗电 | 合理驱动时静态功耗可较低 |
| 负载能力 | 触点可切换较高电压/交流负载 | 受耐压、电流、散热和驱动条件限制 |
| 噪声与寿命 | 有机械声音和触点磨损 | 无机械触点，需关注发热和击穿 |

选择方案时看任务：是否需要隔离、切换交流、承受浪涌、快速开关、低功耗或静音，而不是只看“哪个能导通”。


## 13. 一页复习表

| 知识点 | 关键结论 |
|---|---|
| 欧姆定律 | `U=IR`，用于电压、电流和电阻之间的换算 |
| 功率 | `P=UI=I²R=U²/R`，检查发热和额定功率 |
| KCL | 节点流入电流之和等于流出电流之和 |
| KVL | 闭合回路电压升降代数和为零 |
| 戴维宁 | 复杂线性网络等效为电压源串联电阻 |
| 诺顿 | 复杂线性网络等效为电流源并联电阻 |
| mAh | 电荷容量，不能单独代表能量 |
| mWh | 能量容量，考虑电压因素 |
| 继电器线圈 | 通电产生磁场，驱动机械结构 |
| `COM/NO/NC` | 公共端、常开、常闭，按线圈状态切换 |
| 继电器驱动 | 线圈侧要考虑额定电流和断电反向电压 |

### 本篇核心公式

```text
U = I × R
P = U × I = I²R = U²/R
E = P × t
Wh = V × Ah
ΣIin = ΣIout
ΣU = 0
```


## 14. 自测题

1. 一个节点流入 `120 mA`，其中一条支路流出 `70 mA`，另一条支路应流出多少？
2. `5 V` 电源接 `100 Ω` 电阻，电阻功率是多少？
3. 戴维宁等效电路由哪些部分组成？
4. `1.5 V、800 mAh` 电池的理想能量约是多少？为什么不能只比较 mAh 判断两块不同电压电池的能量？
5. 继电器的 `COM`、`NO`、`NC` 分别是什么？线圈断电和通电时连接关系怎样变化？
6. 为什么单片机 GPIO 通常不能直接驱动继电器线圈？直流线圈两端常加什么器件？
7. 仿真继电器时，为什么要分别检查线圈侧和触点侧？

### 参考答案

1. `120-70=50 mA`。
2. `I=5/100=0.05 A`，`P=UI=0.25 W`。
3. 一个等效电压源 `Uth` 与串联电阻 `Rth`。
4. `E=1.5×0.8=1.2 Wh`；能量还与电压有关，同样 mAh 在不同电压下对应的 Wh 不同。
5. `COM` 是公共端，`NO` 是常开，`NC` 是常闭；通常线圈断电时 COM-NC 导通，通电吸合后切换为 COM-NO，具体以型号资料为准。
6. 线圈电流可能超过 GPIO 能力，且断电会产生反向感应电压；常用三极管/MOS 管驱动，直流线圈并联续流二极管。
7. 继电器用控制信号驱动线圈，再由触点切换另一条回路；两侧的电压、电流、隔离和故障风险不同。
