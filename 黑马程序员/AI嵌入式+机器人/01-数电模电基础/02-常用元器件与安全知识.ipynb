# 02-常用元器件与安全知识


## 1. 本篇学习范围

本篇对应课程视频 **11—20**：元器件和 PCB、电阻器、电容器、电和磁的关系、电感、LRC 振荡电路、保险丝和熔断器、安全警示、接插件、蜂鸣器。

学习目标是认识电路中最常见的被动元件和连接件，知道它们解决什么问题、怎样看参数、有哪些接线和安全边界。

| 类别 | 本篇涉及的对象 | 主要作用 |
|---|---|---|
| 被动元件 | 电阻、电容、电感 | 限流、储能、滤波、延缓变化、形成振荡 |
| 保护器件 | 保险丝、熔断器 | 过流时切断电路，保护线路和设备 |
| 连接件 | 排针排母、板对板、香蕉头、鳄鱼夹、开关 | 建立可靠、可维护的电气连接 |
| 输出器件 | 蜂鸣器 | 把电信号转换为声音 |

> 元件能否正常工作，不只取决于“名字对不对”，还要检查封装、标称值、额定电压、电流、功率、极性和安装方向。


## 2. 从元器件到 PCB

**元器件**是构成电子电路的基本器件。课程把电阻、电容、电感、二极管和晶体管等放在一起介绍，先建立“元件有什么功能”，再学习“怎样把元件连接起来”。

**PCB（Printed Circuit Board，印制电路板）**用绝缘基板、铜箔和机械结构承载元件并连接信号。初学者先关注三件事：

1. 元件的符号、引脚和极性；
2. PCB 上焊盘、走线、过孔和丝印分别表示什么；
3. 电路图中的连接关系如何对应到实物安装。

电路图解决“电气上怎样连接”，PCB 还要解决“元件放在哪里、线怎样走、能否制造和装配”。PCB 的完整绘制和生产流程放到第 8 篇，本篇只把元件识别与安全作为基础。


## 3. 电阻器：限制电流和分配电压

电阻器的基本作用是提供一定的电阻，限制电流、产生电压降，并把一部分电能转化为热能。它的阻值用 `Ω` 表示，功率用 `W` 表示。

### 3.1 选电阻时先看三个参数

讲义用“封装、标称、额定功率”概括电阻数据手册中最重要的内容：

| 参数 | 要回答的问题 | 影响 |
|---|---|---|
| 封装 | 元件是什么外形、尺寸多大、怎样安装？ | 能否放入 PCB、能承受多大功率 |
| 标称阻值 | 电阻值是多少？ | 决定电流和分压关系 |
| 额定功率 | 能长期承受多大功耗？ | 温升、可靠性和是否烧毁 |

电阻还有**精度/误差**参数。例如 `1%` 电阻的实际阻值允许在标称值附近有相应误差；精度越高，允许范围越小，价格和应用要求也可能更高。

额定功率不是“实际一定会消耗的功率”，而是元件在规定条件下允许承受的上限。设计时先计算实际功耗，再选择有余量的额定功率。


## 4. 常见电阻类型

课程配套讲义按功能和制造工艺列举了多种电阻。实际选型时先看它要完成的功能，再看阻值、功率、精度、温度系数和封装。

| 类型 | 特点或典型用途 |
|---|---|
| 普通电阻 | 最常用的固定阻值元件，用于限流、分压、偏置 |
| 精密电阻 | 阻值精度和稳定性较好，用于测量、基准和精密电路 |
| 光敏电阻 | 光照变化会引起阻值变化，用于光照检测 |
| 金属膜电阻 | 稳定性和精度通常较好，常见于一般电子电路 |
| 电位器 | 可连续调节阻值或分压，用于音量、亮度、阈值调整 |
| 旋钮可调电阻 | 通过调节机构改变阻值，常用于设备校准 |
| 热敏电阻 | 阻值随温度变化，用于温度检测或温度补偿 |
| 压敏电阻 | 电压变化会影响阻值，常用于过压保护 |

电位器和可调电阻的机械位置只是“当前状态”，并不代表接法已经正确。使用前要确认固定端、滑动端，以及它在电路中是作分压器还是作可变电阻。


## 5. 电阻值、功率与实际选型

电阻值可通过标识、万用表或数据手册确认。不能只凭外观猜测阻值：不同封装、不同颜色环和贴片标识的规则可能不同。

功率计算仍使用上一篇的关系：

$$
P = UI = I^2R = \frac{U^2}{R}
$$

选型时按下面顺序检查：

1. 根据电路功能算出理论阻值；
2. 选择可获得的标准阻值，并重新计算实际电流；
3. 计算电阻实际功耗；
4. 比较额定功率、允许误差和工作温度；
5. 检查封装尺寸和安装方式。

例如，若电阻两端实际电压为 `6.4 V`，希望电流约为 `20 mA`：

$$
R = \frac{6.4V}{0.02A} = 320Ω
$$

这个计算只得到理论阻值。真正选用时还要依据负载数据手册、电源波动和电阻功率余量确认最终型号。


## 6. 电容器：储存电荷、阻止突变和滤波

电容器由两个电极和中间的绝缘介质构成。它能够储存电荷和能量，电容值用法拉（`F`）表示。常用换算为：

```text
1 F = 1000 mF
1 mF = 1000 μF
1 μF = 1000 nF
1 nF = 1000 pF
```

### 6.1 电容器的常见作用

- **储能**：充电后可以在一定时间内释放能量。
- **滤波**：平滑电源或信号中的波动。
- **旁路/去耦**：在芯片附近提供瞬态电流，减少电源噪声。
- **隔直或耦合**：在特定电路中隔开直流分量、传递变化信号。
- **延时和振荡**：与电阻、电感或其他器件配合形成时间常数和振荡网络。

“电容隔直”是理想化的电路分析结论。真实电容有漏电、等效串联电阻、耐压和频率范围等限制，不能把所有电容都当成完全相同的理想元件。


## 7. 电容器类型与极性

配套讲义按介质和结构列出了玻璃、钽、薄膜、陶瓷、云母、空气、可调和金属化等电容。入门时重点先记住“介质不同，容量、耐压、频率特性、稳定性和体积也可能不同”。

| 类型 | 适合关注的特点 |
|---|---|
| 陶瓷电容 | 体积小、响应快，常用于高频旁路和去耦 |
| 铝电解电容 | 容量较大，常用于电源滤波；多数具有极性 |
| 钽电容 | 体积和稳定性有优势；极性和耐压必须核对 |
| 薄膜电容 | 稳定性和损耗表现较好，适合信号和交流场景 |
| 可调电容 | 容量可调，用于调谐和校准 |
| 超级电容 | 容量很大，适合短时储能和缓慢供电场景 |

### 7.1 有极性电容

铝电解、钽电容等有极性元件，正负端接反可能导致漏电、发热、失效甚至破裂。装配前看外壳上的负号、条纹、长短引脚或数据手册，不要只凭“看起来像”。

电容还要检查**额定电压**。额定电压必须高于电路中可能出现的最高电压，并考虑电源波动和瞬态。


## 8. 超级电容的放电估算

超级电容的放电时间可以用一个简化关系估算：

$$
t = \frac{C \times V}{I}
$$

其中：

- `t`：放电时间，单位为秒；
- `C`：电容量，单位为法拉；
- `V`：可用的电压变化范围，单位为伏特；
- `I`：放电电流，单位为安培。

讲义示例：`C = 1.5 F`、`V = 2.8 V`、`I = 20 mA = 0.02 A`：

$$
t = \frac{1.5 \times 2.8}{0.02} \approx 210s
$$

这是理想估算。实际可用时间还会受到漏电流、内阻、负载电流变化、最低工作电压和电容容量误差的影响。若负载需要稳定电压，不能只用这个公式判断系统是否能正常工作。


## 9. 电和磁：电感的工作基础

电流通过线圈时，线圈周围会产生磁场；磁场发生变化时，又可能在线圈中产生感应电动势。电和磁可以相互转换，电机、发电机、继电器和电感器都利用了这种关系。

**电感器**通常由线圈构成，用于储存磁场能量。它的电感量用亨利（`H`）表示，常用单位还有 `mH`、`μH`。

电感的两个核心特点是：

- 通过磁场储存能量；
- 阻碍电流的突然变化。

因此，电感两端的电流不会像理想开关那样瞬间改变。在电源、滤波、变换器和电机驱动中，电感常用来平滑电流或暂存能量。

> 直观理解：电容更倾向于阻碍电压突变，电感更倾向于阻碍电流突变。两者都能储能，但储能形式不同。


## 10. 电感、电阻和 LRC 振荡

真实电感有线圈电阻、磁芯损耗和漏磁，真实电容也有损耗。因此，电感和电容构成的振荡通常会逐渐衰减；要维持稳定振荡，需要额外的补偿或能量输入。

在 LRC 振荡电路中：

1. 电容先储存电场能量；
2. 放电时电流增大，电感储存磁场能量；
3. 电流变化使电容反向充电；
4. 能量在电场和磁场之间交换，形成周期性变化；
5. 电阻和元件损耗使振荡幅度逐渐减小。

理想 LC 电路的固有频率可写成：

$$
f_0 = \frac{1}{2\pi\sqrt{LC}}
$$

这个公式用于理解趋势：增大 `L` 或 `C`，振荡频率会降低；减小 `L` 或 `C`，振荡频率会升高。实际电路还要考虑阻尼、寄生参数和元件工作频率范围。


## 11. 保险丝与熔断器：让故障先断开

保险丝和熔断器用于防止过流、短路等故障继续损坏电路。电流超过允许值时，熔丝或熔断元件发热并断开回路，使后级设备得到保护。

| 器件 | 典型特点 |
|---|---|
| 保险丝 | 熔断后通常需要更换，结构简单、成本低 |
| 熔断器 | 通常指具有保护功能的熔断元件或组件；具体能否复位要看型号 |
| 自恢复保险丝 | 过流后电阻升高，故障排除并冷却后可能恢复；不能当作普通保险丝理解 |

选择保护器件时至少看：

- 额定电流：正常工作时不应误熔断；
- 额定电压：必须覆盖电路工作电压；
- 熔断速度：快断、慢断适合的负载不同；
- 分断能力：发生短路时能否安全切断故障电流；
- 封装、安装方式和环境温度。

保护器件要串联在需要保护的回路中，并放在故障电流能够经过的位置。用导线替代保险丝会失去保护作用，可能造成线路过热和起火。


## 12. 安全警示：特别注意电池与储能元件

讲义用锂电池案例强调了短路和机械损伤的危险。锂电池能够提供较大的电流，正负极直接连接、外壳破损、针刺挤压、过充过放或使用不匹配的充电器，都可能导致发热、冒烟、起火甚至爆炸。

实际操作应遵守：

- 不让电池正负极被金属工具、导线或零件直接短接；
- 使用与电池类型、节数和充电协议匹配的充电设备；
- 电池鼓包、破损、漏液、异常发热或受撞击后，不再继续使用；
- 不拆解、针刺、挤压或焊接未经允许的电芯；
- 电容器断电后仍可能带电，维修前先按规范放电并确认电压；
- 保护电路不能替代正确的结构、防短路和充电管理设计。

> 安全警示是元器件知识的一部分。看懂参数之后，还要判断故障状态下能量会怎样释放。


## 13. 接插件：建立可维护的连接

接插件是用于连接电路板、导线、模块和测试设备的连接器。讲义中展示了排针排母、板对板连接器、香蕉头、鳄鱼夹以及各种开关和按键。

| 类型 | 常见用途 | 使用时关注 |
|---|---|---|
| 排针/排母 | 模块与开发板、排线连接 | 针脚顺序、间距、方向和防呆 |
| 板对板连接器 | 两块 PCB 之间连接 | 配对型号、插拔次数、定位结构 |
| 香蕉头 | 电源和实验设备连接 | 额定电流、绝缘、防误插 |
| 鳄鱼夹 | 临时测试和实验连接 | 接触可靠性、极性、短路风险 |
| 按键、拨动开关 | 人机操作或信号切换 | 常开/常闭、额定电压电流、触点寿命 |

### 13.1 接插件选型清单

1. 电气参数：额定电压、额定电流、接触电阻；
2. 机械参数：针数、间距、插拔方向、锁扣和插拔寿命；
3. 结构参数：是否防呆、能否避免反插、是否有极性标记；
4. 环境参数：温度、振动、潮湿、腐蚀和防护等级；
5. 装配参数：焊接、压接、螺钉固定或快速插拔。

接插件不是“只要能插上就行”。松动、反插、错位和超额定使用都会造成间歇性故障或发热。


## 14. 蜂鸣器：把电信号变成声音

蜂鸣器是把电信号转换成声音的小型器件。声音来自物体振动；振动频率不同，听到的音调也不同。人耳通常能听到约 `20 Hz—20 kHz` 的声音，常见蜂鸣器工作频率大约在 `2—4 kHz` 范围。

### 14.1 有源与无源蜂鸣器

| 类型 | 工作方式 | 控制特点 |
|---|---|---|
| 有源蜂鸣器 | 内部带振荡电路，接通合适电源即可发声 | 主要控制通断，音调通常较固定 |
| 无源蜂鸣器 | 内部没有独立振荡源，需要外部周期信号 | 可通过改变频率产生不同音调 |

选型和接线前要确认：工作电压、额定电流、极性、是否需要外部驱动、声压和频率范围。单片机引脚驱动能力不足时，应使用三极管或其他驱动级，并在后续器件学习中进一步展开。

蜂鸣器不发声时，依次检查：电源电压、极性、接线是否闭合、器件类型、驱动波形/频率，以及驱动电流是否超过控制器能力。


## 15. 器件识别与装配检查流程

拿到一个陌生元件时，可以按这个顺序检查：

1. **看标识**：型号、阻值、容量、极性、额定电压和额定功率；
2. **看外形**：封装、引脚数量、缺口、条纹、长短脚和定位标记；
3. **查数据手册**：确认引脚定义、绝对最大额定值和推荐工作条件；
4. **看电路位置**：它在电源、信号、保护还是输出部分；
5. **断电测量**：必要时用万用表检查阻值、通断和明显短路；
6. **装配复核**：确认极性、方向、焊盘和连接器配对；
7. **限能上电**：优先使用限流电源，先观察电流和发热，再逐步测试功能。

### 15.1 最容易出错的地方

- 电解电容正负极装反；
- 电阻阻值正确但功率不够；
- 接插件针脚顺序看反或反插；
- 保险丝额定电流过大，故障时起不到保护作用；
- 蜂鸣器有源/无源类型判断错误；
- 电感和电容的额定电压、电流或频率范围没有核对。


## 16. 一页复习表

| 元件 | 核心作用 | 首要参数 | 典型风险 |
|---|---|---|---|
| 电阻 | 限流、分压、耗能 | 阻值、精度、额定功率、封装 | 功率过大导致发热或烧毁 |
| 电容 | 储能、滤波、去耦、延时 | 容量、耐压、极性、频率特性 | 极性反接、耐压不足、残余电压 |
| 电感 | 储存磁场能量、平滑电流 | 电感量、额定电流、直流电阻、频率 | 电流突变、磁芯饱和、发热 |
| 保险丝 | 过流时断开回路 | 额定电流、电压、速度、分断能力 | 用导线替代或额定值选错 |
| 接插件 | 建立可插拔连接 | 针数、间距、电流、方向、寿命 | 松动、反插、接触不良 |
| 蜂鸣器 | 电信号转声音 | 有源/无源、电压、频率、电流 | 驱动方式错误、超出引脚能力 |

### 关键关系

```text
电容：电场储能，倾向于阻碍电压突变
电感：磁场储能，倾向于阻碍电流突变
LRC：电场能量与磁场能量交换，实际会因损耗衰减
保险丝：故障时牺牲自己，切断过流路径
```

5. 保险丝为什么必须串联？为什么不能用导线替代？
6. 排针排母连接模块时，除了针脚数量还应检查什么？
7. 有源蜂鸣器和无源蜂鸣器在驱动方式上有什么区别？

### 参考答案

1. 还要确认精度、额定功率、封装、工作温度和实际电路中的电压电流。
2. `t = C×V/I = 1.5×2.8/0.02 ≈ 210 s`；真实时间还会受漏电、内阻和最低工作电压影响。
3. 有极性电容的内部结构决定了反向电压可能造成漏电、发热和失效；应核对外壳极性标记、引脚和数据手册。
4. 电容倾向于阻碍电压突变，电感倾向于阻碍电流突变。
5. 串联才能让故障电流经过保护元件并在超限时断开；导线不能按故障电流熔断，会失去保护并增加发热风险。
6. 针脚顺序、间距、方向、防呆结构、额定电流、电压、插拔寿命和安装高度。
7. 有源蜂鸣器内部有振荡电路，通常控制通断即可；无源蜂鸣器需要外部周期信号，频率可以影响音调。
