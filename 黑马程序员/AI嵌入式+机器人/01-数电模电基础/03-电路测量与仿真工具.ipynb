# 03-电路测量与仿真工具


## 1. 本篇学习范围

本篇对应课程视频 **21—30**：电阻的测量、万用表测电压、电压表量程、万用表测电流、Circuit 软件入门、欧姆定律和电阻串并联、电容器仿真、电感器仿真、分压电路、电位器仿真。

这一篇把前两篇的概念落实为两种能力：

- **实测**：知道万用表应该怎样接、怎样选档位、怎样避免损坏仪表和电路；
- **仿真**：在 CircuitJS 中搭建简单电路，通过电流、电压和波形观察公式是否符合预期。

```text
断电检查 → 预估量级 → 选功能/量程 → 正确接线 → 读数记录 → 解释结果
                                                           ↓
                                              用仿真复现和改变一个条件
```


## 2. 万用表能测什么

万用表是常用的电子测量仪器，可以测量电压、电流、电阻，也可能提供通断、二极管、电容、频率等功能。不同型号的旋钮、插孔和量程标识会有差异，操作前应以当前仪表说明书为准。

### 2.1 三个基本测量量

| 测量 | 电路中的含义 | 接入方式 | 典型档位 |
|---|---|---|---|
| 电阻 `Ω` | 元件对电流的阻碍程度 | 断电后跨接元件两端 | `Ω`、`kΩ`、`MΩ` |
| 电压 `V` | 两点之间的电势差 | 并联在被测两点 | 直流 `DCV` 或交流 `ACV` |
| 电流 `A/mA/μA` | 流过某支路的电流 | 串联到被测支路 | `A`、`mA`、`μA` |

先判断“我是在测两点之间的差值，还是要让电流经过仪表”，再决定并联还是串联。不要因为插孔位置相邻，就把不同测量方式混用。


## 3. 用万用表测电阻

配套讲义给出的测量顺序是：预估电阻值、调整合适的欧姆档位、把表笔接到电阻两端、读取显示数值。

### 3.1 标准步骤

1. 关闭电路电源，确认被测元件不带电。
2. 若电路中有电容，先按规范放电。
3. 预估阻值范围，选择欧姆档；不确定时先用较大量程。
4. 黑表笔接 `COM`，红表笔接电阻测量插孔。
5. 两支表笔分别接触电阻两端，保持接触稳定。
6. 等待读数稳定，并记录单位和档位。

### 3.2 读数异常时检查

- 显示 `OL` 或超量程：阻值可能超过当前量程，也可能是断路或接触不良；
- 读数接近 `0 Ω`：可能是短路，也可能是低阻值元件；
- 数值不断变化：电路仍带电、元件与其他支路并联、接触不稳或电容正在充放电；
- 在线路板上测到的阻值不等于元件标称值：并联支路会改变等效电阻。

如果需要确认某个独立电阻，最好拆下一端或断开相关并联路径后再测量。


## 4. 用万用表测直流和交流电压

电压测量是对两个点进行比较，所以表笔要并联在被测对象两端。配套讲义强调，测量前要预估电压、选择合适的直流/交流档位、把表笔接到被测电压两端，再读取数值。

### 4.1 标准步骤

1. 判断被测信号是直流还是交流。
2. 预估可能的最大电压，先选择较大量程。
3. 黑表笔接 `COM`，红表笔接 `V/Ω` 插孔。
4. 红、黑表笔并联到被测两点。
5. 观察读数、单位和正负号，再逐步换到合适量程。

直流电压测量时，红表笔接电势较高点、黑表笔接参考点，读数通常为正；反接时可能显示负号。交流档位不能替代直流档位，直流档也不能用来正确解释任意交流波形。

### 4.2 量程怎么选

- 量程过小：可能显示超量程，甚至损坏仪表；
- 量程过大：通常更安全，但分辨率可能较低；
- 不确定时：先从高量程开始，再逐级降低；
- 换档前：保持表笔接触稳定，遇到高压或未知电压先断电并查清楚。


## 5. 用万用表测电流

电流测量与电压测量不同：必须让被测电流经过万用表，因此要把电路断开一个节点，把仪表串入断点。

### 5.1 标准步骤

1. 断电并预估被测电流大小。
2. 选择直流或交流电流档，先使用较大量程。
3. 黑表笔接 `COM`，红表笔插到正确的 `mA` 或 `A/20A` 插孔。
4. 断开被测支路，把万用表串联到断点。
5. 检查接线和极性后上电，读取数值。
6. 测量结束立即断电，并把红表笔插回 `V/Ω` 插孔。

### 5.2 最危险的误操作

不要把电流档直接并到电源两端。电流档的内部电阻很小，这样做可能近似形成短路，烧断仪表保险丝、损坏电源或产生电弧。

> 测量电流前确认“串联”；测量电压前确认“并联”。这两个动作顺序比记住旋钮图标更重要。

某些万用表的 `mA` 插孔有独立保险丝，`A/20A` 插孔的允许测量时间和精度也可能不同。超过插孔额定值会产生危险，不能把 `20A` 当成“永远安全的档位”。


## 6. 测量前后的安全检查

### 6.1 上电前

- 电源电压和极性是否正确；
- 表笔是否插在与当前功能匹配的插孔；
- 档位是 `Ω`、`V` 还是 `A`，直流还是交流；
- 被测电路是否需要断开、放电或拆下一端；
- 电流是否可能超过仪表、表笔、接插件或保险丝的额定值。

### 6.2 测量后

- 先断电，再拆除串联的电流表；
- 把红表笔移回 `V/Ω` 插孔；
- 让旋钮回到安全的电压档或关闭仪表；
- 记录读数、条件和量程，避免只记一个没有上下文的数字。

涉及市电、高压电容、电池组或未知电源时，不用“试一下”代替风险判断。停止操作、断电并查清楚额定值和保护措施。


## 7. CircuitJS：用仿真观察电路行为

配套讲义介绍的 **CircuitJS** 是一个用 JavaScript 编写的电路仿真器。它可以从元件库拖入电源、电阻、电容、电感、开关和逻辑器件，用导线连接电路，再观察电流、电压、波形和示波器显示。

CircuitJS 适合做三件事：

1. 在接实物前检查连接关系和量级；
2. 改变一个参数，观察电流、电压或波形怎样变化；
3. 把欧姆定律、串并联、分压和储能元件的行为可视化。

仿真模型有理想化假设，元件参数、初始条件和显示结果未必等同于真实器件。仿真通过只能说明模型下的结果符合预期，实际电路仍要检查额定值、接线、寄生参数和安全边界。

### 7.1 基本操作顺序

1. 放置电源和需要观察的元件；
2. 用导线连接成闭合回路；
3. 设置电压、阻值、电容值或电感值；
4. 运行仿真，观察电流、电压和波形；
5. 只改变一个参数，再比较前后结果；
6. 将观察结果与公式和预期进行核对。


## 8. 电阻串联与并联

### 8.1 串联

串联电路只有一条电流路径，各电阻中的电流相同，总电阻为各电阻之和：

$$
R_{总} = R_1 + R_2 + \cdots + R_n
$$

总电压分配到各个电阻上：

$$
U_{总} = U_1 + U_2 + \cdots + U_n
$$

配套讲义中的仿真示例使用 `12 V` 电源和 `3 Ω、4 Ω、5 Ω` 三个串联电阻：

$$
R_{总}=3+4+5=12Ω
$$

$$
I=\frac{12V}{12Ω}=1A
$$

各电阻上的电压分别为 `3 V`、`4 V`、`5 V`，总和为 `12 V`。

### 8.2 并联

并联支路两端电压相同，总电流等于各支路电流之和：

$$
I_{总} = I_1 + I_2 + \cdots + I_n
$$

总电阻满足：

$$
\frac{1}{R_{总}} = \frac{1}{R_1}+\frac{1}{R_2}+\cdots+\frac{1}{R_n}
$$

两个电阻并联时也可写为：

$$
R_{总}=\frac{R_1R_2}{R_1+R_2}
$$

并联后的总电阻小于其中最小的单个电阻，这是检查计算结果是否合理的快速方法。


## 9. 用仿真验证串并联

在 CircuitJS 中搭建串联或并联电路时，先用公式算出预期值，再观察仿真中的电流和电压：

| 电路 | 先算什么 | 仿真中检查什么 |
|---|---|---|
| 串联 | `R总`、总电流、各段压降 | 各元件电流是否相同，压降之和是否等于电源电压 |
| 并联 | 等效电阻、总电流、各支路电流 | 各支路电压是否相同，支路电流之和是否等于总电流 |

推荐记录一张小表：

```text
参数      理论值        仿真值        差异原因
电源电压  ...           ...           模型设置/读数精度
总电阻    ...           ...           元件连接/单位
总电流    ...           ...           量程/显示精度
```

如果理论值和仿真值差异很大，先检查单位、接线、参考地、元件是否真正连接，以及仿真中是否启用了非理想模型。不要一开始就修改公式。


## 10. 电容器仿真：观察充放电

电容器在仿真中可以帮助观察“电压不能瞬间改变”的趋势。给电容串联电阻并接入电源时，电容电压会逐渐上升；断开电源并提供放电路径时，电容电压会逐渐下降。

常用时间常数为：

$$
τ = RC
$$

`τ` 越大，充放电变化越慢；`R` 或 `C` 增大都会使变化过程拉长。理想 RC 充电时，经过约 `1τ`、`3τ`、`5τ`，分别可以理解为接近最终值的不同程度；真实电路还会受负载、漏电和电源内阻影响。

### 10.1 仿真观察步骤

1. 放置直流电源、电阻和电容；
2. 把电阻串在电源与电容之间；
3. 观察电容两端电压和电阻电流；
4. 改变 `R` 或 `C`，比较上升/下降速度；
5. 使用开关或放电电阻观察放电过程。

仿真中看到电流随时间减小，是因为电容电压逐渐接近电源电压，电阻两端电压变小；这不是“电容把电流吃掉了”，而是电路状态正在改变。


## 11. 电感器仿真：观察电流变化

电感器更明显的特征是阻碍电流突变。给电感接入电源时，电流会逐渐建立；断开电源时，电感会努力维持原来的电流，可能在开关两端产生较高反向电压。

实际的继电器线圈、电机绕组和电感负载通常需要续流路径或吸收电路，例如二极管、RC 吸收或专用驱动器。具体方案要结合负载类型、开关速度和电压等级确定。

### 11.1 仿真观察步骤

1. 放置电源、开关、电阻和电感；
2. 闭合开关，观察电感电流从小到大建立；
3. 断开开关，观察电流和电压的瞬态变化；
4. 增大电感值，比较电流变化速度；
5. 增加电阻或吸收支路，观察尖峰如何变化。

电感仿真也可以帮助理解 LRC 电路：电容的电场能量与电感的磁场能量互相交换，电阻则带来阻尼和能量损耗。


## 12. 分压电路

两个电阻串联在输入电压上，可以从中间节点取得较低电压。若 `R1` 接输入端、`R2` 接地，输出取在 `R2` 两端，则理想空载分压公式为：

$$
U_{out}=U_{in}\times\frac{R_2}{R_1+R_2}
$$

### 12.1 例题

若 `Uin = 12 V`，`R1 = 2 kΩ`，`R2 = 1 kΩ`：

$$
U_{out}=12\times\frac{1}{2+1}=4V
$$

分压电阻中的电流为：

$$
I=\frac{12V}{2kΩ+1kΩ}=4mA
$$

### 12.2 分压的边界

上式假设输出端没有负载电流。如果输出端接入负载 `R_L`，下方电阻实际变为 `R2` 与 `R_L` 的并联，输出电压会下降：

$$
R_{下}=R_2\parallel R_L
$$

再用 `R下` 代入分压公式。负载阻值不能与分压电阻处于同一量级而不加考虑，否则计算出来的空载电压没有实际意义。


## 13. 电位器与可变电阻

电位器通常有三个端子：两端是电阻体的固定端，中间端是滑动端。滑动端移动时，两个部分的阻值同时变化，但它们的和（忽略误差）接近电位器总阻值。

### 13.1 三种常见接法

| 接法 | 用途 | 说明 |
|---|---|---|
| 三端作分压器 | 调节输出电压 | 两固定端接输入和地，滑动端取输出 |
| 滑动端与一固定端 | 可变电阻 | 形成一个随旋钮变化的两端电阻 |
| 固定端与滑动端并接 | 提高断触时的稳定性 | 常用于把电位器当作可变电阻使用 |

电位器仿真时，先确认旋钮位置和三个端子的关系，再观察输出电压如何变化。若把输入、地和输出端接错，旋钮变化可能与预期相反，甚至造成短路。

电位器的额定功率、总阻值、调节方式和机械寿命都需要根据用途选择；它不适合直接承受超过额定功率的负载。


## 14. 仿真排错方法

电路仿真“没有结果”时，按从基础到复杂的顺序排查：

1. **电源是否存在**：电压值、极性和开关状态是否正确；
2. **回路是否闭合**：导线端点是否真正吸附到元件引脚；
3. **参考点是否清楚**：需要地或参考节点的模型是否已经连接；
4. **元件数值是否合理**：`Ω`、`kΩ`、`mA`、`μF` 等单位是否混用；
5. **显示量是否正确**：看的是电压、电流还是波形，探针位置是否正确；
6. **初始条件是否影响结果**：电容初始电压、电感初始电流和开关状态可能改变瞬态；
7. **模型边界是否被忽略**：理想器件结果不能直接代表实物的发热、漏电和寄生参数。

每次只改一个条件，并保存修改前后的参数。这样才能知道差异来自哪个因素。


## 15. 一页复习表

| 场景 | 正确做法 | 最容易错在哪里 |
|---|---|---|
| 测电阻 | 断电、放电、跨接元件两端 | 在线测量被并联支路影响 |
| 测电压 | 选 DC/AC 档，表笔并联 | 把电流档或错误插孔接入电路 |
| 测电流 | 断开支路，仪表串联 | 把电流档直接并到电源两端 |
| 串联电阻 | `R总=ΣR`，各处电流相同 | 把电压相同误用到串联电阻 |
| 并联电阻 | 各支路电压相同，电流相加 | 算出的总阻值大于最小支路阻值 |
| 分压 | `Uout=Uin×R2/(R1+R2)` | 忽略输出负载导致的加载效应 |
| 电容仿真 | 观察 `RC` 时间常数和充放电 | 认为电容电压可以瞬间改变 |
| 电感仿真 | 观察电流建立和断开尖峰 | 忽略续流和吸收路径 |
| 电位器 | 确认三端关系和滑动端 | 接错电源/地/输出端 |

### 本篇核心公式

```text
R串 = R1 + R2 + ...
1/R并 = 1/R1 + 1/R2 + ...
Uout = Uin × R2/(R1 + R2)
τ = R × C
f0 = 1/(2π√(LC))   # 理想 LC 的理解公式
```


## 16. 自测题

1. 测量一个未知电阻前，为什么必须先断电？在线路板上测得的阻值为什么可能偏小？
2. 万用表测电压和测电流，分别应怎样接入？
3. 测电流时为什么必须先断开支路？测量结束后为什么要把红表笔移回 `V/Ω` 插孔？
4. `3 Ω、4 Ω、5 Ω` 串联接在 `12 V` 电源上，总电流是多少？各电阻压降是多少？
5. `2 kΩ` 与 `1 kΩ` 并联后的等效电阻是多少？
6. `12 V` 通过 `2 kΩ` 和 `1 kΩ` 分压，空载输出是多少？若输出接了负载，应如何修正？
7. 增大 RC 电路中的电阻，电容充电速度会怎样变化？增大电感，电流建立速度会怎样变化？

### 参考答案

1. 带电测电阻会使仪表读数失真，甚至损坏仪表；线路板上的并联支路会与被测电阻形成更小的等效阻值。
2. 电压表并联，电流表串联。
3. 需要让被测电流经过仪表；测完不移回表笔，下一次误用电压档可能造成危险短路。
4. `R总=12 Ω`，`I=1 A`；三只电阻压降分别为 `3 V`、`4 V`、`5 V`。
5. `R并=(2×1)/(2+1) kΩ=2/3 kΩ≈667 Ω`。
6. `Uout=12×1/(2+1)=4 V`；接入负载后先算 `R2` 与 `RL` 的并联等效值，再代回分压公式。
7. 两者都会让变化过程变慢：电阻增大使 `τ=RC` 变大，电感增大使电流建立更慢。
