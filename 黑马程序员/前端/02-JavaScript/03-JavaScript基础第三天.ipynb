## 1. for 循环

### 1.1 基本语法与执行顺序

`for` 把循环三要素写在同一行：初始化、条件和变化量，适合循环次数已知的场景。

```javascript
for (let i = 1; i <= 3; i++) {
  document.write('月薪过万<br>')
}
```

执行顺序是：

1. `let i = 1`：初始化，只执行一次。
2. 判断 `i <= 3`，为真才进入循环体。
3. 执行循环体。
4. 执行 `i++`，再回到条件判断。
5. 条件为假，跳出循环。

本例 `i` 为 `1`、`2`、`3` 时各执行一次，变成 `4` 后结束，页面共显示三行。改成 `i <= 10` 就能完成课堂“输出十句”的练习。三个部分之间是两个英文分号，变化量后面不要多写分号。

课堂原代码（`04-code/01-for循环的基本使用.html`）：

```javascript
    // 利用for循环输出三句话 月薪过万
    for (let i = 1; i <= 3; i++) {
      document.write('月薪过万<br>')
    }
    // 课堂练习： 输出10句月薪过万，并说出for循环的执行过程
  
```

**补充：**这里 `let i` 的作用范围在 `for` 中，不要在循环外依赖它。也不要写 `for (...);` 后再写大括号，那个多余分号会把循环体变为空语句。

### 1.3 continue、break 与无限循环

| 写法 | 作用 |
| --- | --- |
| `continue` | 结束本次循环，继续下一次 |
| `break` | 退出最近一层循环 |
| `for (;;)` | 条件省略，构造无限循环 |

下面两段分别输出 `1、2、4、5` 和 `1、2`：

```javascript
for (let i = 1; i <= 5; i++) {
  if (i === 3) {
    continue
  }
  console.log(i)
}

for (let i = 1; i <= 5; i++) {
  if (i === 3) {
    break
  }
  console.log(i)
}
```

`for` 遇到 `continue` 仍会执行头部的变化量 `i++`，再判断条件。这与 `while` 的变化量通常写在循环体里不同：`while` 跳过后面的 `i++` 可能造成死循环。

## 2. 循环嵌套

### 2.1 外层一次，内层一轮

一个循环里面再写一个循环就是循环嵌套。外层每执行一次，内层重新开始并完成一整轮；不是外层和内层各执行一次就结束。



课堂案例：每天记五个单词，连续三天。外层 `i` 控制天数，内层 `j` 控制当天的单词数量。

课堂原代码（`04-code/04-循环的嵌套.html`）：

```javascript
    // 外层循环打印 第 n 天
    for (let i = 1; i <= 3; i++) {
      document.write(`第${i}天<br>`)
      // 里层循环打印 第几个单词
      for (let j = 1; j <= 5; j++) {
        document.write(`记住了第${j}个单词<br>`)
      }
    }
  
```

`j = 1` 每次进入新的一天都会重新初始化，因此共执行 `3 × 5 = 15` 次“记住单词”的输出，另外还有三次“第几天”的输出。内外层变量分别用 `i` 和 `j`，变化量也要对应各自变量。

### 2.2 打印矩形：行数与列数

外层控制行数，内层控制每行的符号数量；换行写在内层循环外面、外层循环里面。

```javascript
for (let i = 1; i <= 5; i++) {
  for (let j = 1; j <= 5; j++) {
    document.write('☆')
  }
  document.write('<br>')
}
```

## 3. 数组的基本使用

### 3.1 为什么需要数组

数组（Array）按顺序保存一组数据。例如保存全班姓名，用一个数组比声明很多零散变量更便于遍历和管理。数组是对象类型的一种，数组内保存的每一项叫**元素**。

```javascript
let names = ['小明', '小刚', '小红', '小丽', '小米']
let scores = [78, 84, 70, 62, 75]
let mixed = [true, 1, false, 'hello']
let empty = []
```

元素之间用英文逗号隔开；数组能保存任意类型的值，也允许混合类型。本课进行求和或比较的案例使用数字数组，姓名案例使用字符串数组。

### 3.2 声明与访问

常用字面量 `[]` 声明数组；构造函数 `new Array(...)` 在本课只作了解。

```javascript
let arr = [1, 2, 'pink', true]
let another = new Array(1, 2, 3, 4)
console.log(arr[0])  // 1
console.log(arr[2])  // 'pink'
```

### 3.3 索引、长度与遍历

数组的编号叫**索引**或**下标**，从 `0` 开始。读取元素写 `数组名[索引]`，例如第三个元素是 `names[2]`。



`length` 属性表示数组长度。本课连续存放的数组中，它等于元素个数；最后一个元素的索引是 `length - 1`，不是 `length`。

```javascript
let names = ['小明', '小刚', '小红', '小丽', '小米']
console.log(names.length)                 // 5
console.log(names[0])                     // 小明
console.log(names[names.length - 1])      // 小米
console.log(names[5])                     // undefined

for (let i = 0; i < names.length; i++) {
  console.log(names[i])
}
```

`length` 是属性，不写成 `length()`。循环里不要把 `i <= arr.length` 当成标准边界，否则会多访问一项 `undefined`。

**补充：**JavaScript 允许稀疏数组，`length` 不一定等于实际存在的元素个数。例如直接给索引 `5` 赋值可能留下空槽；本日案例都按连续索引处理。

### 3.4 数组求和与平均值

### 3.5 数组最大值和最小值

采用“假设法”：先假设第一项最大、也最小，然后从第二项开始比较，遇到更大的就更新最大值，遇到更小的就更新最小值。

```javascript
let arr = [2, 6, 1, 77, 52, 25, 7]
let max = arr[0]
let min = arr[0]
for (let i = 1; i < arr.length; i++) {
  if (arr[i] > max) {
    max = arr[i]
  }
  if (arr[i] < min) {
    min = arr[i]
  }
}
console.log(max, min)  // 77，1
```

课堂原代码（`04-code/10-数组最大值.html`）：

```javascript
    let arr = [2, 6, 1, 7, 400, 55, 88, 100]
    // max里面要存的是最大值
    let max = arr[0]
    // min 要存放的是最小值
    let min = arr[0]
    // 遍历数组
    for (let i = 1; i < arr.length; i++) {
      // 如果max 比 数组元素里面的值小，我们就需要把这元素赋值给 max
      // if (max < arr[i]) max = arr[i]
      max < arr[i] ? max = arr[i] : max
      // 如果min 比 数组元素大， 我们就需要把数组元素给min
      // if (min > arr[i]) {
      //   min = arr[i]
      // }
      min > arr[i] ? min = arr[i] : min
    }
    // 输出 max
    console.log(`最大值是: ${max}`)
    console.log(`最小值是: ${min}`)
  
```

**资料差异：**讲义使用上面的七项数组，课堂文件改为 `[2, 6, 1, 7, 400, 55, 88, 100]`，因此课堂结果是最大值 `400`、最小值 `1`。原代码用三元表达式完成更新，整理示例用 `if` 展示同一过程。

初始值使用 `arr[0]`，不要随意设为 `0`：全为负数的数组，其最大值也可能小于 `0`。该写法也要求数组非空。

**XMind 拓展：记录极值位置。**更新极值时同时记录索引，例如 `maxIndex = i`。若只在严格更大时更新，重复最大值会保留第一次出现的位置；索引和数据值要分别保存。

### 4.2 新增：push 与 unshift

| 方法 | 添加位置 | 返回值 |
| --- | --- | --- |
| `arr.push(元素1, 元素2, ...)` | 数组末尾 | 添加后的新长度 |
| `arr.unshift(元素1, 元素2, ...)` | 数组开头 | 添加后的新长度 |

两种方法都直接修改原数组，也都能一次添加多项：

## 6. 冒泡排序与 sort

### 6.1 冒泡排序的思路

排序就是把数据按升序或降序排列。冒泡排序反复比较相邻两个元素，顺序错误时交换。

升序时，若 `arr[j] > arr[j + 1]` 就交换，第一趟会把最大项送到末尾；第二趟不用再比较已经确定的末项。长度为 `n` 的数组最多需要 `n - 1` 趟，第 `i` 趟只比较 `n - i - 1` 次（这里 `i` 从 `0` 开始）。



**图示说明：**原图把内层次数叫“交换次数”，更准确是比较次数的上限；只有前项大于后项才真的交换。

按课堂注释示例整理的可运行写法：

```javascript
let arr = [2, 4, 3, 5, 1]
for (let i = 0; i < arr.length - 1; i++) {
  for (let j = 0; j < arr.length - i - 1; j++) {
    if (arr[j] > arr[j + 1]) {
      let temp = arr[j]
      arr[j] = arr[j + 1]
      arr[j + 1] = temp
    }
  }
}
console.log(arr)  // [1, 2, 3, 4, 5]
```

交换要借助临时变量 `temp`，不能先把前一个值覆盖后再拿它去覆盖后一个值。内层条件保证 `j + 1` 仍在未排序部分的有效范围内。

### 6.2 数字数组使用 sort

课程给出两种写法，先记住比较方向，函数和参数会在后续课程展开：

原数组 `[1, 2, 3, 4, 5]`，循环从 `i = 1` 开始，所以漏掉索引 `0` 的数字 `1`。改成从 `0` 开始：

```javascript
let arr = [1, 2, 3, 4, 5]
for (let i = 0; i < arr.length; i++) {
  console.log(arr[i])
}
```

修改后完整输出 `1、2、3、4、5`。

### 8.3 排错题 3：内层循环写错变化量

原内层写 `for (let j = 1; j <= 5; i++)`，一直改变 `i`，而 `j` 始终为 `1`，因此内层条件始终为真。变化量应对应内层变量：

```javascript
for (let i = 1; i <= 5; i++) {
  for (let j = 1; j <= 5; j++) {
    console.log('这是双重for循环')
  }
}
```

修改后输出 `25` 次。排查嵌套循环时分别确认内外层的初始化、条件、变化量是否使用对应变量。

### 8.4 常见错误速查

| 现象 | 检查方向 |
| --- | --- |
| 数组遍历多出 undefined | 是否写成 `i <= arr.length` |
| 第一项没打印 | 是否从 `i = 1` 开始 |
| 求和得到 NaN | `sum` 是否初始化，是否累加到 `undefined` |
| 报 item is not defined | 是否误用了未声明变量，而非 `arr[i]` |
| 新数组变成数字 | 是否把 `push()` 返回的新长度赋给了数组变量 |
| 内层循环不结束 | 是否遗漏或写错 `j++` |
| 修改数组方法没效果 | 是否只引用方法名，遗漏 `()` |
| 方法不存在 | 是否拼错 `unshift`、`document.write` |
| 同一作用域重复声明报错 | 是否把两种参考方案一起复制，重复写了 `let max` 等 |

原 Markdown 还把 `switch` 和 `if` 的效率作了概括。复习它们的选择重点应是：`switch` 适合确定值匹配，注意全等与 `break`；`if` 适合区间和复杂条件。不能简单断言某一种在所有场景都更快，性能取决于实际代码和执行环境。
