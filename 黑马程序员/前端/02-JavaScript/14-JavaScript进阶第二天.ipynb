## 一、创建对象与构造函数

对象可以用对象字面量、`new Object()` 或构造函数创建。对象字面量最直接；`new Object()` 先得到空对象再追加属性；构造函数适合批量生成结构相同的实例。

构造函数按约定首字母大写，使用 `new` 调用。`new` 会创建新对象，把构造函数中的 `this` 指向它，并返回该实例。构造函数中的属性是实例属性；直接挂在构造函数上的属性或方法是静态成员，分别通过实例和构造函数访问。

```javascript
function Person(name, age) {
  this.name = name
  this.age = age
}
const p1 = new Person('张三', 18)
```

```html
<!DOCTYPE html>
<html lang="zh-CN">

<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
</head>

<body>
  <script>
    // 1. 构造函数的作用：创建对象
    // 2. 构造函数的两个约定
    // 2.1 构造函数名首字母要大写
    // 2.2 构造函数只能由 new 关键字来调用
    function Pig(name, age) {
      this.name = name
      this.age = age
    }
    // 使用new关键字调用函数的行为称为实例化
    // 构造函数内部无需写return，返回值为新创建的对象，return返回值无效
    const Gorge = new Pig('乔治', 3)
    const Peppa = new Pig('佩奇', 2)
    console.log(Gorge)
    console.log(Peppa)
    Peppa.name = '小猪佩奇'  // 实例属性(实例对象.属性)
    Peppa.sayName = function () { // 实例方法(实例对象.方法)
      console.log(this.name)
    }
    Peppa.sayName()

    Pig.eyes = 2 // 静态属性(构造函数.属性)
    Pig.sayEyes = function () { // 静态方法(构造函数.方法)
      console.log(this.eyes)
    }
    Pig.sayEyes()

  </script>
</body>

</html>
```

## 三、Object 静态方法

`Object.keys(obj)` 返回对象的可枚举属性名数组；`Object.values(obj)` 返回属性值数组；`Object.assign(target, ...sources)` 将源对象属性复制到目标对象并返回目标对象。

`Object.assign` 常用于追加属性或浅拷贝。浅拷贝只复制第一层，嵌套对象和数组仍然共享引用。

## 四、reduce 累计处理

`reduce` 依次处理数组，并把上一步结果传给下一步。回调常写成 `(pre, cur) => ...`。提供初始值时，第一次的 `pre` 是该值；不提供初始值时，`pre` 是第一个数组元素，空数组会报错。

```javascript
const nums = [1, 2, 3, 4]
const total = nums.reduce((pre, cur) => pre + cur, 0)
```

金额或工资累加应显式提供 `0`，让空数组也有明确结果。课堂员工工资案例合计为 `37000`。

课堂示例：

```html
<!DOCTYPE html>
<html lang="zh-CN">

<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
</head>

<body>
  <script>
    // 1. 数组的实例方法
    // 1.1 reduce() arr.reduce(function(上一次值, 当前值){}, 初始值)
    const arr = [1, 2, 3, 4, 5]
    // 1.1.1 没有初始值
    const sum = arr.reduce((pre, cur) => pre + cur)
    console.log(sum) // 15
    // 1.1.2 有初始值
    const sum2 = arr.reduce((pre, cur) => pre + cur, 10)
    console.log(sum2) // 25

    // 计算薪资案例
    const arr1 = [{
      name: '张三',
      salary: 10000
    }, {
      name: '李四',
      salary: 12000
    }, {
      name: '王五',
      salary: 15000
    }]
    // 计算所有员工的薪资总和
    // 第二个参数 0 表示初始值，没有初始值时，默认取数组的第一个元素，因此不能省略
    const totalSalary = arr1.reduce((pre, cur) => pre + cur.salary, 0)
    console.log(totalSalary) // 37000
  </script>
</body>

</html>
```

## 五、数组常用方法

| 方法 | 作用 | 返回值 |
| --- | --- | --- |
| `forEach` | 遍历并执行操作 | `undefined` |
| `map` | 对每个元素映射 | 新数组 |
| `filter` | 保留满足条件的元素 | 新数组 |
| `find` | 查找第一个满足条件的元素 | 元素或 `undefined` |
| `findIndex` | 查找第一个满足条件的下标 | 下标或 `-1` |
| `every` / `some` | 全部/至少一个满足条件 | 布尔值 |
| `join` | 连接数组元素 | 字符串 |
| `concat` | 合并数组 | 新数组 |
| `Array.from` | 伪数组或可迭代对象转数组 | 新数组 |

`forEach` 适合遍历操作，不能用它生成新数组；需要数组结果时使用 `map` 或 `filter`。这些方法通常不改变原数组，但新数组里的对象仍可能与原数组共享引用。

课堂示例：

```html
<!DOCTYPE html>
<html lang="zh-CN">

<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
</head>

<body>
  <script>
    // 1. 数组的实例方法
    // 1.1 reduce() arr.reduce(function(上一次值, 当前值){}, 初始值) 
    // 对数组中的每个元素进行累加，返回一个新数组
    const arr = [1, 2, 3, 4, 5]
    // 1.1.1 没有初始值
    const sum = arr.reduce((pre, cur) => pre + cur)
    console.log(sum) // 15
    // 1.1.2 有初始值
    const sum2 = arr.reduce((pre, cur) => pre + cur, 10)
    console.log(sum2) // 25

    // 计算薪资案例
    const arr1 = [{
      name: '张三',
      salary: 10000
    }, {
      name: '李四',
      salary: 12000
    }, {
      name: '王五',
      salary: 15000
    }]

    // 计算所有员工的薪资总和
    const totalSalary = arr1.reduce((pre, cur) => pre + cur.salary, 0)
    console.log(totalSalary) // 37000

    // 1.2 forEach() arr.forEach(function(当前值, 当前索引){}) 
    // 对数组中的每个元素进行操作，没有返回值
    arr.forEach((item, index) => {
      console.log(item, index)
    })

    // 1.3 map() arr.map(function(当前值, 当前索引){}) 
    // 对数组中的每个元素进行操作，返回一个新数组
    const arr2 = arr.map((item, index) => {
      return item * 2
    })
    console.log(arr2) // [2, 4, 6, 8, 10]

    // 1.4 filter() arr.filter(function(当前值, 当前索引){}) 
    // 过滤数组中的元素，返回符合条件的元素组成的新数组
    const arr3 = arr.filter((item, index) => {
      return item > 3
    })
    console.log(arr3) // [4, 5]

    // 1.5 join() arr.join('分隔符') 
    // 把数组中的所有元素转换为字符串，用分隔符隔开，返回一个新字符串
    const str = arr.join('-')
    console.log(str) // 1-2-3-4-5

    // 1.6 find() arr.find(function(当前值, 当前索引){}) 
    // 查找数组中符合条件的第一个元素，返回该元素
    const obj = arr1.find(item => item.name === '李四')
    console.log(obj) // {name: '李四', salary: 12000}

    // 1.7 every() arr.every(function(当前值, 当前索引){}) 
    // 检查数组中的所有元素是否都符合条件，返回一个布尔值
    const bool = arr.every(item => item > 0)
    console.log(bool) // true

    // 1.8 some() arr.some(function(当前值, 当前索引){}) 
    // 检查数组中是否有符合条件的元素，返回一个布尔值
    const bool2 = arr.some(item => item > 3)
    console.log(bool2) // true

    // 1.9 concat() arr.concat(数组1, 数组2, ...) 
    // 合并多个数组，返回一个新数组
    const arr4 = arr.concat(arr2, arr3)
    console.log(arr4) // [1, 2, 3, 4, 5, 2, 4, 6, 8, 10, 4, 5]

    // 1.13 findIndex() arr.findIndex(function(当前值, 当前索引){}) 
    // 查找数组中符合条件的第一个元素的索引，返回该索引
    const index = arr.findIndex(item => item > 3)
    console.log(index) // 3

    // 2. 数组的静态方法
    // 2.1 from() Array.from(类数组对象或可迭代对象) 
    // 把类数组对象或可迭代对象(伪数组)转换为数组
    const str2 = 'hello'
    const arr5 = Array.from(str2)
    console.log(arr5) // ['h', 'e', 'l', 'l', 'o']
    console.log(arr5.shift()) // 'h'
  </script>
</body>

</html>
```

## 六、字符串常用方法

常用属性和方法包括：`length`、`split`、`substring`、`slice`、`startsWith`、`endsWith`、`includes`、`indexOf`、`toUpperCase`、`toLowerCase`。字符串方法通常返回新字符串，不修改原字符串。

`split(',')` 可将逗号分隔的字符串拆成数组；`slice` 支持负数下标，`substring` 会把负数按 0 处理，二者规则不要混淆。

课堂示例：

```html
<!DOCTYPE html>
<html lang="zh-CN">

<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
</head>

<body>
  <script>
    // 实例属性 length
    // 字符串的长度，返回字符串中字符的个数
    const str0 = 'hello,world'
    console.log(str0.length) // 11

    // 1. 实例方法
    // 1.1 split() 字符串.split(分隔符)
    // 把字符串转换为数组，以指定的分隔符为界，返回新数组
    const str = 'hello,world'
    const arr = str.split(',')
    console.log(arr) // ['hello', 'world']

    // 1.2 substring() 字符串.substring(起始索引, 结束索引)
    // 提取字符串中指定索引之间的字符，不包含结束索引处的字符(左闭右开)，返回新字符串
    const str2 = 'hello,world'
    const str3 = str2.substring(0, 5)
    console.log(str3) // 'hello'

    // 1.3 startsWith() 字符串.startsWith(子字符串)
    // 判断字符串是否以指定的子字符串开头，返回布尔值
    const str4 = 'hello,world'
    console.log(str4.startsWith('hello')) // true
    console.log(str4.startsWith('world')) // false
    // endsWith() 判断字符串是否以指定的子字符串结尾

    // 1.4 includes() 字符串.includes(子字符串)
    // 判断字符串是否包含指定的子字符串，返回布尔值
    const str5 = 'hello,world'
    console.log(str5.includes('llo')) // true

    // 1.5 indexOf() 字符串.indexOf(子字符串)
    // 返回子字符串在字符串中第一次出现的索引，如果未找到则返回-1
    console.log(str5.indexOf('llo')) // 2
    console.log(str5.indexOf('abc')) // -1

    // 1.6 toUpperCase() 字符串.toUpperCase()
    // 把字符串转换为大写字母，返回新字符串
    const str6 = 'Hello World'
    console.log(str6.toUpperCase()) // 'HELLO WORLD'
    // toLowerCase() 把字符串转换为小写字母，返回新字符串
    console.log(str6.toLowerCase()) // 'hello world'

    // 1.7 slice() 字符串.slice(起始索引, 结束索引)
    // 提取字符串中指定索引之间的字符，不包含结束索引处的字符(左闭右开)，返回新字符串
    // 数组也有slice()方法
    const str7 = 'hello,world'
    const str8 = str7.slice(0, 5)
    console.log(str8) // 'hello'
  </script>
</body>

</html>
```

## 七、赠品字符串案例

商品赠品字符串可以先 `split`，再 `map` 生成标签，最后 `join('')` 拼回 HTML：

```javascript
const gift = '50g茶叶,清洗球'
const html = gift.split(',')
  .map(item => `<span>【赠品】${item}</span>`)
  .join('')
```

## 八、Number 常用方法

`toFixed(2)` 将数字格式化为两位小数，但返回值是字符串，适合展示，不适合直接继续计算。`toString()` 可以把数字转换为字符串。

课堂示例：

```html
<!DOCTYPE html>
<html lang="zh-CN">

<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Document</title>
</head>

<body>
  <script>
    const num = 3.1415926
    // 1. toFixed() Number.toFixed()
    // 保留小数位
    console.log(num.toFixed(2)) // 3.14

    // 2. 转换为字符串
    // console.log(String(num))
    console.log(num.toString()) // '3.1415926'
  </script>
</body>

</html>
```

## 十、易错点与练习

1. 构造函数要用 `new` 调用；实例成员和静态成员的访问对象不同。
2. `Object.assign` 是浅拷贝，嵌套对象仍共享引用。
3. `reduce` 处理可能为空的数组时显式提供初始值。
4. `forEach` 返回 `undefined`，`map/filter` 返回新数组。
5. `find` 找不到返回 `undefined`，`findIndex` 找不到返回 `-1`。
6. `toFixed` 返回字符串；金额计算尽量使用整数分或统一精度方案。
