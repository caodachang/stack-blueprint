# 04｜Git 版本控制系统与黑马头条部署实战

这份笔记对应《AJAX 入门到项目实战》的第 04 部分，依据配套 Markdown 笔记、课件、示例代码和作业资料整理，视频内容不作为笔记来源。主线是“本地版本管理 → 分支协作 → 远程仓库 → 黑马头条静态站点发布”。

- 能使用分支隔离功能开发、合并分支并处理冲突。
- 能配置远程仓库，完成 `clone / pull / push` 多人协作。
- 能用 VS Code 管理 Git，并把 Webpack 构建产物发布为可访问的静态站点。

> 课程示例使用 `master` 作为主分支；新建平台仓库可能默认使用 `main`。命令中的分支名必须与本地和远程实际名称一致。

## 示例项目：`git_study` 的提交与协作流程

资料中的 `03_code/git_study` 是一个前端页面项目，包含登录页、发布页、内容页、公共请求工具和静态资源。可以用它练习完整流程：

```bash
cd 03_code/git_study
git init
git status -s
git add page lib utils
git commit -m "feat: 初始化黑马头条页面"

# 新需求在独立分支完成
git switch -c content
git add page/content
git commit -m "feat: 完成内容列表页面"

git switch master
git switch -c login-bug
# 修复登录校验后
git add page/login utils
git commit -m "fix: 修复登录表单校验"

git switch master
git merge login-bug
git branch -d login-bug
```

> `git switch` 是较新的分支命令；课程使用的 `git checkout` 仍然可用。创建并切换分支也可以写成 `git checkout -b content`。提交前先运行 `git diff` 和 `git status -s`，确认没有把构建产物、账号信息或无关文件提交进去。

## 复习清单与安全实践

### 本地仓库闭环

```bash
git init
git status -s
git add .
git commit -m "feat: 完成某项功能"
git log --oneline --decorate --graph --all
```

### 撤销与回退

| 需求 | 命令 | 影响 |
| --- | --- | --- |
| 查看差异 | `git diff` | 只读，比较工作区与暂存区 |
| 取消暂存 | `git restore --staged <file>` | 文件回到未暂存状态 |
| 丢弃工作区改动 | `git restore <file>` | 覆盖未提交改动，先确认内容可丢弃 |
| 回退并保留工作区 | `git reset --mixed <commit>` | 默认模式，暂存区被重置 |
| 回退并保留暂存区 | `git reset --soft <commit>` | 适合重新组织提交 |
| 强制回退 | `git reset --hard <commit>` | 会覆盖暂存区和工作区，谨慎使用 |
| 找回历史指针 | `git reflog --oneline` | 查看回退、切换等本地记录 |

### 分支协作

1. 从最新主分支创建功能分支。
2. 一个功能拆成清晰的小提交，提交说明写清动词和目的。
3. 合并前先拉取主分支，减少长期分叉。
4. 发生冲突时逐个文件阅读冲突标记，确认最终代码后再 `git add` 并提交。
5. 合并完成并确认已推送后，再删除本地功能分支。

### `.gitignore` 建议

```gitignore
node_modules/
dist/
.DS_Store
*.log
.env
.env.*
```

不要提交密码、私钥、个人 Token 和本地配置；如果敏感内容已经进入历史，单纯新增 `.gitignore` 不会自动从历史删除，需要按团队安全流程处理。

### 远程协作速查

```bash
git remote -v
git remote add origin <仓库地址>
git push -u origin master       # 或 main
git pull --rebase origin master
git clone <仓库地址>
```

`git pull` 可以理解为 `git fetch` 加 `git merge`；多人同时修改同一文件时，先拉取再开发，提交前检查远程分支和当前分支，避免把别人的提交覆盖掉。
