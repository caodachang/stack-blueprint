## 1. 列表标签

### 1.1 列表的应用场景

新闻、排行榜、账单等内容彼此有关联，适合按行整齐展示。列表的选择取决于内容之间的关系：是否有顺序，是否是一个主题及其说明。

| 类型 | 整体标签 | 内部标签 | 常见场景 | 默认显示 |
| --- | --- | --- | --- | --- |
| 无序列表 | `ul` | `li` | 新闻列表、没有先后顺序的条目 | 每项前有圆点 |
| 有序列表 | `ol` | `li` | 排行榜、有先后顺序的条目 | 每项前有序号 |
| 自定义列表 | `dl` | `dt`、`dd` | 底部导航中的主题与说明 | `dd` 有缩进 |

### 1.2 无序列表：`ul` 和 `li`

`ul` 表示整个无序列表，`li` 表示其中一项。课堂文件 `01-列表-无序.html`：

```html
<ul>
    <li>榴莲</li>
    <li>香蕉</li>
    <li>苹果</li>
</ul>
```

打开后，三种水果各占一个列表项，前面有默认圆点。



> 课程要求的嵌套关系是 `ul > li`：`ul` 中的条目用 `li` 包裹。标题、段落、图片、链接等内容可以继续放在 `li` 内，不直接并列放到 `ul` 下面。

### 1.3 有序列表：`ol` 和 `li`

`ol` 表示整个有序列表，`li` 仍然表示每一项。课堂文件 `02-列表-有序.html`：

```html
<ol>
    <li>张三:100</li>
    <li>李四: 80</li>
</ol>
```

浏览器会自动显示序号，不需要在每个 `li` 的文字里手写 `1.`、`2.`。嵌套关系是 `ol > li`，`li` 内可以继续放需要的内容。



**原资料勘误：**讲义第 12 页小结误把 `ol` 和 `li` 解释成“无序列表”；本节应为“有序列表”，与第 11 页及配套代码一致。

### 1.4 自定义列表：`dl`、`dt`、`dd`

自定义列表把主题和针对主题的说明组织在一起。`dl` 包裹整体，`dt` 表示主题，`dd` 表示主题对应的内容。课堂文件 `03-列表-自定义.html`：

```html
<dl>
    <dt>帮助中心</dt>
    <dd>账户管理</dd>
    <dd>购物指南</dd>
</dl>
```



这里“帮助中心”是主题，“账户管理”和“购物指南”是说明，`dt` 与 `dd` 在 `dl` 内并列。按照课程写法，`dl` 内放 `dt`、`dd`，具体内容再写到它们里面。浏览器默认会让 `dd` 向右缩进。

原课的使用经验是：无序列表最常用，有序列表用于有顺序的内容，自定义列表常见于网页底部导航。


## 2. 表格标签

### 2.1 用行和列组织数据

成绩单等数据需要同时表达“某一行是谁”和“某一列是什么”，用表格的单元格结构来展示。

| 标签 | 含义 | 嵌套位置 |
| --- | --- | --- |
| `table` | 整张表格 | 最外层 |
| `tr` | 一行 | `table` 或表格结构标签内 |
| `td` | 普通单元格 | `tr` 内 |

基本关系是 **`table > tr > td`**。一行写一个 `tr`，这一行有几个单元格，就写几个 `td`。不要用换行标签代替表格的行。



**原资料勘误：**第 20 页对 `td` 的解释沿用了列表的描述；这里应理解为“表格中的普通单元格”。

### 2.2 表格的边框、宽度和高度

课程先用 HTML 属性观察表格效果。这三个属性写在 `table` 的开始标签中，作用于整张表格。

| 属性 | 课程中的属性值 | 作用 |
| --- | --- | --- |
| `border` | 数字，例如 `1` | 显示表格边框 |
| `width` | 数字，例如 `600` | 设置表格整体宽度 |
| `height` | 数字，例如 `400` | 设置表格整体高度 |

课堂文件 `04-表格-基本使用.html`：

```html
<!-- table 包含 tr, tr包含td -->
<table border="1" width="600" height="400">
    <tr>
        <td>姓名</td>
        <td>成绩</td>
        <td>评语</td>
    </tr>
    <tr>
        <td>小哥哥</td>
        <td>100分</td>
        <td>孩砸, 真棒啊</td>
    </tr>
    <tr>
        <td>郭德纲</td>
        <td>150分</td>
        <td>没有郭德纲相声早黄了, 有了郭德纲, 相声更黄了</td>
    </tr>
</table>
```

打开后应看到三行三列的表格。第一行在这一阶段仍使用 `td`，下一步再改成表头单元格。

> 原课提醒：实际开发中的样式效果推荐使用 CSS 设置。这里保留 `border`、`width`、`height`，用于理解讲义与完成当前练习。


### 2.3 表格标题和表头单元格

普通文字无法直接说明它是“整张表的大标题”还是“一列的标题”，`caption` 与 `th` 为它们赋予不同含义。

| 标签 | 用途 | 写在哪里 | 默认效果 |
| --- | --- | --- | --- |
| `caption` | 表格整体的大标题 | `table` 内，课程写在最前面 | 在表格上方居中显示 |
| `th` | 表头单元格，课程用于第一行的列标题 | `tr` 内，用于替换对应的 `td` | 文字加粗并居中 |

课堂文件 `05-表格-表格标题和表头.html`：

```html
<table border="1">
    <caption><strong>学生成绩单</strong></caption>
    <tr>
        <!-- <td></td> -->
        <th>姓名</th>
        <th>成绩</th>
        <th>评语</th>
    </tr>
    <tr>
        <td>张三</td>
        <td>100分</td>
        <td>真棒, 第一名</td>
    </tr>
    <tr>
        <td>李四</td>
        <td>99分</td>
        <td>真棒, 第二名</td>
    </tr>
    <tr>
        <td>总结</td>
        <td>郎才女貌</td>
        <td>真棒, 相亲成功</td>
    </tr>
</table>
```

浏览器会将“学生成绩单”放在表格上方，将“姓名、成绩、评语”加粗并居中。`caption` 不放在 `tr` 中，`th` 也不直接放在 `table` 中。


### 2.4 表格结构标签（了解）

表格还可以按用途分为头部、主体、底部，让结构更清晰。它们负责分组，内部仍用 `tr` 表示行，再由 `td` 或 `th` 表示单元格。

| 标签 | 含义 | 常见内容 |
| --- | --- | --- |
| `thead` | 表格头部 | 列标题 |
| `tbody` | 表格主体 | 数据行 |
| `tfoot` | 表格底部 | 总结、统计等 |

关系变成 `table > thead/tbody/tfoot > tr > td/th`。结构标签可以省略，但写出它们能更清楚地说明每组行的作用。



课堂文件 `06-表格-结构.html`：

```html
<table border="1">
    <caption><strong>学生成绩单</strong></caption>
    <thead>
        <tr>
            <!-- <td></td> -->
            <th>姓名</th>
            <th>成绩</th>
            <th>评语</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td>张三</td>
            <td>100分</td>
            <td>真棒, 第一名</td>
        </tr>
        <tr>
            <td>李四</td>
            <td>99分</td>
            <td>真棒, 第二名</td>
        </tr>    
    </tbody>
    <tfoot>
        <tr>
            <td>总结</td>
            <td>郎才女貌</td>
            <td>真棒, 相亲成功</td>
        </tr>
    </tfoot>
</table>
```

添加结构标签后，表格仍显示相同的行列内容；这一步主要改变内容分组与语义。


### 2.5 合并单元格

相邻单元格需要展示同一份内容时，可以水平或垂直合并。

| 属性 | 方向 | 属性值 |
| --- | --- | --- |
| `rowspan` | 跨行，垂直合并 | 合并的行数，包括保留单元格所在行 |
| `colspan` | 跨列，水平合并 | 合并的列数，包括保留单元格所在列 |



合并时按以下顺序操作：

1. 确定要合并哪几个相邻单元格。
2. 按 **左上原则** 保留一个：上下合并保留最上面的，左右合并保留最左面的，删除其他被合并的单元格标签。
3. 在保留的 `td` 或 `th` 上设置 `rowspan="行数"` 或 `colspan="列数"`。

课堂文件 `07-表格-合并单元格.html`：

```html
<table border="1">
    <caption><strong>学生成绩单</strong></caption>
    <thead>
        <tr>
            <!-- <td></td> -->
            <th>姓名</th>
            <th>成绩</th>
            <th>评语</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td>张三</td>
            <td rowspan="2">100分</td>
            <td>真棒, 第一名</td>
        </tr>
        <tr>
            <td>李四</td>
            <td>真棒, 第二名</td>
        </tr>    
    </tbody>
    <tfoot>
        <tr>
            <td>总结</td>
            <td colspan="2">郎才女貌</td>
        </tr>
    </tfoot>
</table>
```

在主体中，两行“成绩”由第一行的 `rowspan="2"` 占据，第二行删除对应成绩单元格。因此第二行只写姓名和评语两个 `td`，但视觉上仍对应三列。

在底部，“郎才女貌”用 `colspan="2"` 占据原来的两列，因此这一行只保留“总结”和合并后的单元格。

> 合并属性写在保留的单元格上。只加属性而没有删除被合并的标签，会多出单元格，导致表格错位。原课要求在同一结构分组中合并，不能跨 `thead`、`tbody`、`tfoot` 合并。


## 3. 表单标签

### 3.1 `input` 的不同形态

登录页、注册页需要收集用户填写的信息。输入框、选择框、按钮等可操作的元素统称为表单控件。

`input` 是单标签，可以通过 `type` 属性显示不同控件。

| `type` 值 | 显示形态 | 用途 |
| --- | --- | --- |
| `text` | 文本框 | 输入单行文本 |
| `password` | 密码框 | 输入密码，字符以遮蔽形式显示 |
| `radio` | 单选框 | 多选一 |
| `checkbox` | 复选框 | 多选多 |
| `file` | 文件选择框 | 选择之后要上传的文件 |
| `submit` | 提交按钮 | 提交表单 |
| `reset` | 重置按钮 | 恢复表单默认值 |
| `button` | 普通按钮 | 默认没有提交或重置功能，可配合 JavaScript 添加行为 |



课堂文件 `08-表单-input基本使用.html`：

```html
<!-- 写什么就显示什么 -->
文本框: <input type="text">
<br>
<br>
<!-- 书写的内容都会变成点点显示 -->
密码框: <input type="password">

<br>
<br>
单选框: <input type="radio">

<br>
<br>

多选框: <input type="checkbox">

<br>
<br>

上传文件: <input type="file">
```

这段代码先观察控件的外观。单选分组、默认选中、提交与重置的条件会在下面逐步添加。


### 3.2 文本框：占位符、默认值与字段名

`type="text"` 用于输入单行文本。`placeholder` 是输入前的提示文字；内容为空时显示，输入后会消失。

课堂文件 `09-表单-占位符.html` 同时展示文本框与密码框的提示：

```html
<input type="text" placeholder="请输入用户名">
<input type="password" placeholder="请输入密码">
```

讲义第 43 页与配套 XMind 进一步区分 `value` 和 `name`：

| 属性 | 含义 | 例子 |
| --- | --- | --- |
| `placeholder` | 提示应填写什么 | `placeholder="请输入昵称"` |
| `value` | 在代码中指定控件的默认值；提交时使用控件的当前值 | 默认显示“帅哥” |
| `name` | 提交时的字段名，用来说明数据含义 | `name="nickname"` |

讲义写法：

```html
昵称: <input type="text" value="帅哥" name="nickname">
```

这段代码打开后，输入框中真正有“帅哥”这个默认值。提交的数据可理解成 `nickname=帅哥`：`nickname` 是字段名，“帅哥”是字段值。

`placeholder` 与 `value` 的区别是：前者只是提示，后者是实际的默认内容。用户修改输入后，提交的是修改后的当前值。

**补充：**讲义用“内容”和“内容含义”解释前后端的数据关系。标准表单提交中，没有 `name` 的控件通常不会作为字段发送；仅写 `value` 并不意味着它会独立发送给后端。

### 3.3 密码框

把 `type` 设置为 `password`，即可让输入内容以遮蔽形式显示。它也可以使用 `placeholder`、`name` 与默认值。

> `type` 的值要准确写成 `password`，不要拼错或在值中多加空格。浏览器不认识该类型时，会按默认的 `text` 文本框处理。

检查效果时，在文本框和密码框中输入相同内容：文本框显示原字符，密码框显示遮蔽字符。


### 3.4 单选框：用 `name` 分组

`type="radio"` 表示单选框。仅写出多个圆形控件还不够，需要让同一组选项使用相同的 `name`，才能实现一组中同时只选一个。

课堂文件 `10-表单-单选框.html`：

```html
性别: <input type="radio" name="sex">男
      <input type="radio" name="sex" checked>女

<input type="checkbox" checked>
```

两个性别选项都使用 `name="sex"`，所以点击“男”对应的圆形控件时，“女”会取消选择。`checked` 让“女”默认选中。

同一组只给一个单选框设置 `checked`。如果页面还有“婚姻状况”这一组，使用另一字段名，如 `name="marry"`，使两组互不影响。

配套 XMind 还指出：单选框的 `value` 用来表示该选项对应的数据，`name` 表示字段含义，并承担分组作用。

### 3.5 复选框：允许选中多项

`type="checkbox"` 表示复选框，各选项可以独立选中、取消。它也用 `checked` 表示默认选中，但可以同时为多项设置。

依据讲义中的“爱好”效果整理：

```html
爱好:
<input type="checkbox" checked>敲代码
<input type="checkbox" checked>熬夜
<input type="checkbox">掉头发
```

打开后前两项默认选中。点击第三项时，前两项仍可以保持选中。配套总结中的 `name`、`value` 分别表示提交的字段名与选项数据，复选框不会因为 `name` 相同而变成互斥选择。

**补充：**`checked` 是布尔属性，写出它就表示默认选中。需要取消默认选中时删除该属性，不用 `checked="false"`。


### 3.6 文件选择

`type="file"` 显示文件选择控件。点击后由浏览器打开文件选择窗口。默认选择一个文件，添加 `multiple` 可以选择多个文件，这是配套总结中标出的 HTML5 属性。

课堂文件 `12-表单-上传文件.html`：

```html
<input type="file" multiple>
```

在文件选择窗口中选择多个文件，确认后应能看到控件显示已选文件的信息。

**补充：**这一节演示的是“选择文件”。选择完成并不代表已经上传成功，实际上传还需要提交与接收文件的功能。


### 3.7 `input` 按钮与 `form`

| 写法 | 功能 | 需要注意 |
| --- | --- | --- |
| `<input type="submit">` | 提交表单 | 使用 `value` 可以设置按钮文字 |
| `<input type="reset">` | 恢复默认值 | 恢复到初始状态，不一定清空 |
| `<input type="button">` | 普通按钮 | 当前没有内置提交或重置行为 |

要按本课写法实现提交和重置，将相关控件与按钮一起包裹在 `form` 中。课堂文件 `13-表单-按钮input.html`：

```html
<form action="">
    用户名: <input type="text">
    <br>
    <br>
    密码: <input type="password">
    <br>
    <br>
    <!-- 按钮 -->
    <input type="submit" value="免费注册">
    <input type="reset">
    <input type="button" value="普通按钮">
    <!-- 属性  xx="xxx" -->
</form>
```

先修改用户名和密码，再点击重置。这个示例没有设置输入框默认值，所以会恢复为空。如果控件原先有 `value`，或选项原先有 `checked`、`selected`，重置会恢复这些默认状态。

`value="免费注册"` 设置了提交按钮显示的文字。普通按钮默认只显示按钮外观。

**补充：**这里的 `action=""` 是原课堂代码，没有配置实际注册接口，输入框也未设置提交字段名。这段代码用于观察控件与按钮，不会完成真实的账户注册。


### 3.8 `button` 按钮标签

`button` 是双标签，按钮文字放在开始标签与结束标签之间，还可以包裹图片等内容。它的三种 `type` 值与 `input` 按钮一致。

| `type` | 行为 |
| --- | --- |
| `submit` | 提交 |
| `reset` | 重置 |
| `button` | 普通按钮 |

课堂文件 `14-表单-按钮button.html`：

```html
<button>我是按钮</button>
<button type="submit">提交按钮</button>
<button type="reset">重置按钮</button>
<button type="button">普通按钮, 没有任何功能</button>
```

原课指出，谷歌浏览器中不写 `type` 时默认为提交按钮。**补充：**提交或重置要有关联的表单；上面的课堂代码没有包裹 `form`，主要演示按钮写法。放进表单后，若只希望它是普通按钮，明确写 `type="button"`。


### 3.9 `select` 下拉菜单

下拉菜单提供多个候选项，`select` 表示整个菜单，内部的每个 `option` 表示一项。`selected` 写在 `option` 上，指定默认选中项。

课堂文件 `15-表单-下拉菜单.html`：

```html
<select>
    <option>北京</option>
    <option>上海</option>
    <option>广州</option>
    <option selected>深圳</option>
</select>
```

打开后默认显示“深圳”，展开菜单可以换成其他城市。`selected` 和单选、复选的 `checked` 用在不同标签上，不要混用。

### 3.10 `textarea` 文本域

单行文本框不适合个人介绍等大段文字，`textarea` 提供多行输入区域。它是双标签。

课堂文件 `16-表单-文本域.html`：

```html
<textarea cols="60" rows="30"></textarea>
```

| 属性 | 含义 |
| --- | --- |
| `cols` | 可见宽度，以字符列数衡量 |
| `rows` | 可见行数 |

在区域内输入文字并按回车，应能继续输入下一行。讲义中的浏览器还可以拖拽右下角改变大小；实际开发时，尺寸等样式效果推荐使用 CSS 设置。

**补充：**文本域的默认内容放在双标签之间，例如 `<textarea>默认介绍</textarea>`，不使用 `input` 的 `value` 写法。


### 3.11 `label`：让文字与控件关联

小圆点、小方框的点击区域较小。用 `label` 将文字与控件关联后，点击文字也可以操作对应控件。

方法一，用 `for` 对应控件的 `id`：

1. 给控件设置 `id`。
2. 用 `label` 包裹对应文字。
3. 让 `label` 的 `for` 值与该控件的 `id` 值一致。

方法二，把文字和控件一起包裹在 `label` 中，不写 `for`。

课堂文件 `17-表单-label.html` 在同一组单选框中演示两种方式：

```html
性别: 
<input type="radio" name="sex" id="nan"> <label for="nan">男</label>
<label><input type="radio" name="sex"> 女</label>
```

保存后点击“男”或“女”这两个文字，应能切换对应选项。`name="sex"` 保证单选分组，`id="nan"` 与 `for="nan"` 负责文字关联，两者用途不同。

**补充：**同一页面中的 `id` 应唯一；多个控件可以有相同的 `name`，用来表达同一字段或单选组。


## 4. 语义化标签

### 4.1 没有语义的布局标签：`div` 和 `span`

它们提供通用容器，用于包裹内容，本身不说明内容是头部、导航还是文章。下面讨论的是浏览器默认显示，后续 CSS 可以改变效果。

| 标签 | 默认显示特点 | 配套总结中的尺寸特点 |
| --- | --- | --- |
| `div` | 独占一行 | 宽度通常占满所在容器，高度由内容撑开 |
| `span` | 一行可以显示多个 | 宽度随内容变化，行内高度随字体与行高表现 |

课堂文件 `18-没有语义的标签.html`：

```html
普通文字
<div>这是div标签</div>
<div>这是div标签</div>

<span>这是span标签</span>
<span>这是span标签</span>
```

两个 `div` 默认分别显示在两行；两个 `span` 在空间足够时显示在同一行。配套 XMind 将 `span` 的尺寸概括为“由内容撑开”，这里按其默认行内显示来理解。

### 4.2 有语义的布局标签（了解）

HTML5 增加了含义更明确的布局标签。原课将它们的默认显示特点与 `div` 对比：默认独占一行，同时说明内容的作用。

| 标签 | 原课对应含义 |
| --- | --- |
| `header` | 网页头部 |
| `nav` | 网页导航 |
| `footer` | 网页底部 |
| `aside` | 网页侧边栏 |
| `section` | 网页区块 |
| `article` | 网页文章 |



课堂文件 `19-有语义的标签-手机端.html`：

```html
<header>网页头部</header>
<nav>网页导航</nav>
<footer>网页底部</footer>
<aside>侧边栏</aside>
<section>网页区块</section>
<article>文章</article>
```

**补充：**标签表达结构与含义，图中的颜色、宽高、左右排列等需要样式设置。仅写这些标签时，浏览器不会自动排成效果图中的侧栏布局；这些标签也不限于手机端使用。


## 5. 字符实体

### 5.1 HTML 的空格合并现象

在普通文本中连续写多个空格、换行和缩进，浏览器通常会把它们合并成一个空格显示。因此，在代码中敲很多空格并不能直接得到等量的视觉间隔，代码换行也不等于页面换行。

### 5.2 使用字符实体表示特殊符号

字符实体的命名写法是 `&实体名称;`，以 `&` 开头，以英文分号结束。课程重点是空格、小于号和大于号，讲义表中还列出了其他常见字符。

| 显示字符 | 含义 | 实体写法 |
| --- | --- | --- |
| 空格 | 不换行空格，课程用于保留空格效果 | `&nbsp;` |
| `<` | 小于号 | `&lt;` |
| `>` | 大于号 | `&gt;` |
| `&` | 和号 | `&amp;` |
| `"` | 双引号 | `&quot;` |
| `'` | 撇号、单引号 | `&apos;` |
| `¢` | 分（cent） | `&cent;` |
| `£` | 英镑（pound） | `&pound;` |
| `¥` | 元（yen） | `&yen;` |
| `€` | 欧元（euro） | `&euro;` |
| `§` | 小节符号 | `&sect;` |
| `©` | 版权（copyright） | `&copy;` |

课堂文件 `20-字符实体.html`：

```html
<!-- 网页不认识多个空格, 只认识一个 -->
这是HTML文档, 现在要学&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;习字符实体.
```

这段代码中的多个 `&nbsp;` 会保留多个空格的效果。需要在页面显示尖括号时，用实体避免它被当作标签语法的一部分。当前知识点的练习示例：

```html
&lt;p&gt;这是段落标签&lt;/p&gt;
```

浏览器应将其显示为文字 `<p>这是段落标签</p>`，而不是解析成一个段落元素。

**原资料兼容性说明：**第 73 页对 `&apos;` 标注“IE 不支持”，这是讲义保留的旧浏览器提示，现代浏览器支持该实体。

其余文本框按原资料分别设置所在地区提示“奋斗者之家”、学历提示“小学”、月薪提示“1000~100000”；手机号与昵称使用普通文本框。自我介绍是 `cols="50"`、`rows="10"` 的 `textarea`，其后接“我承诺”三条列表与提交、重置按钮。图片文件放在 HTML 旁的 `images` 目录。

验证时重点检查：点击性别文字或图片可以切换，婚姻状况组不影响性别组，喜好多项可同时选中，默认条款复选框重置后仍为选中。原资料只演示静态控件，出生年月没有日期联动，参考代码中的空 `option value=""` 也未表达实际的年月日提交值。
