## 1. HTML 初体验

### HTML 是什么

HTML 是 **HyperText Markup Language，超文本标记语言**。课程用“超链接”帮助理解“超文本”：网页中的内容可以通过链接关联到其他页面；它也能组织图片、音视频等内容。

“标记”也叫“标签”，是在尖括号中写入标签名，用来告诉浏览器这段内容是什么。例如，`strong` 表示重要的文字，浏览器通常将它显示为加粗。

### 标签语法

讲义写法：

```html
<strong>需要加粗的文字</strong>
```



`<strong>` 是开始标签，`</strong>` 是结束标签，中间放内容。结束标签比开始标签多一个 `/`。

| 类型 | 写法 | 本日示例 |
| --- | --- | --- |
| 双标签 | `<标签名>内容</标签名>` | `strong`、`p`、`a` |
| 单标签 | `<标签名>`，没有结束标签 | `br`、`hr`、`img` |

在 `demo.html` 中输入上面的示例，保存并打开后，应看到加粗的文字。标签本身不会以尖括号的形式显示在页面上。

本节自查：HTML 文件的扩展名是 `.html`；标签名放在 `<>` 中；双标签的内容放在开始标签与结束标签之间。


## 2. HTML 基本骨架

网页中的信息有不同用途：有些是标题、编码等供浏览器使用的信息，有些是正文、图片等要显示给用户的内容。HTML 骨架为它们规定了位置。

| 标签 | 作用 | 所在位置 |
| --- | --- | --- |
| `html` | 包裹整个网页 | 最外层 |
| `head` | 网页头部，存放标题、样式等信息 | `html` 内 |
| `title` | 网页标题，通常显示在浏览器标签页上 | `head` 内 |
| `body` | 网页主体，存放正文、图片等内容 | `html` 内，与 `head` 并列 |

讲义的简化骨架：

```html
<html>
    <head>
        <title>网页标题</title>
    </head>
    <body>
        网页主体内容
    </body>
</html>
```



### 在 VS Code 中生成骨架

在 `.html` 文件中输入英文 `!`，按 `Enter` 或 `Tab` 展开。讲义截图中的骨架如下：

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta http-equiv="X-UA-Compatible" content="IE=edge">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>

</body>
</html>
```

本日先记住 `html → head / body` 和 `head → title` 的位置关系。将 `title` 改成“我的第一个网页”，在 `body` 中写入“你好，HTML”，保存并刷新：标签页标题应改变，页面正文应显示这句话。

**补充：**`<!DOCTYPE html>` 是文档类型声明；`charset="UTF-8"` 指定字符编码。中文练习页面可将 `lang="en"` 改成 `lang="zh-CN"`。新版 VS Code 生成的骨架可能不包含 `X-UA-Compatible` 这一行，它属于旧版 IE 的兼容设置，不影响本日学习。


## 3. 标签的关系

理解标签关系可以确定书写位置，也能让代码更整齐。本日主要区分父子关系与兄弟关系。

### 父子关系：嵌套

```html
<html>
    <head></head>
</html>
```

`head` 写在 `html` 内部，`html` 是父级，`head` 是子级。**子级换行并缩进**，可用 `Tab`。

### 兄弟关系：并列

```html
<html>
    <head></head>
    <body></body>
</html>
```

`head` 与 `body` 有同一个父级，彼此是兄弟关系。**兄弟标签换行后对齐**，开始标签与对应的结束标签也应对齐。

讲义自查：`<html><body></body></html>` 中，`html` 和 `body` 是父子关系；骨架中的 `head` 与 `body` 是兄弟关系。


## 4. 注释

注释是对代码的解释和说明，用来帮助理解、查找和后续修改代码。讲义要求在关键代码旁加注释。

HTML 注释写法：

```html
<!-- 这里是注释 -->
```

练习示例：

```html
<body>
    <!-- 页面主要标题 -->
    <h1>我的学习记录</h1>
</body>
```

在 VS Code 中，将光标放到代码行或选中代码，按 `Ctrl + /` 可添加或取消注释。浏览器中应只看到“我的学习记录”，不会显示注释内容。


## 5. 标题标签

标题标签用于新闻标题、文章标题、网页区域名称、产品名称等，标签名是 `h1` 到 `h6`，全部是双标签。

```html
<h1>一级标题</h1>
<h2>二级标题</h2>
<h3>三级标题</h3>
<h4>四级标题</h4>
<h5>五级标题</h5>
<h6>六级标题</h6>
```

默认显示特点：文字加粗，字号从 `h1` 到 `h6` 逐渐减小，每个标题独占一行。把示例放入 `body`，保存并刷新，就能比较六级标题的效果。

讲义的使用经验是：一个网页中 `h1` 只用一次，放新闻标题或网页 Logo；`h2` 到 `h6` 没有使用次数限制。

> **补充说明：**“`h1` 只用一次”是本课程的组织建议，浏览器不会禁止多个 `h1`。本日练习按一个主要标题安排，再用较低级别标题组织章节；默认字号也可以在后续 CSS 学习中修改。


## 6. 段落标签

段落标签 `p` 用于新闻段落、文章段落、产品描述信息等，是双标签。

```html
<p>这是第一段内容。</p>
<p>这是第二段内容。</p>
```

默认显示特点是：每个段落独占一行，段落之间存在间隙。保存并刷新后，应看到两段分开的文字，而不是挤成一个连续的文本块。

`h1` 等标题标签表达“这是标题”，`p` 表达“这是正文段落”。根据内容的作用选择标签。


## 7. 换行与水平线标签

### 换行：`br`

在普通 HTML 正文中，编辑器里的 `Enter` 换行不会直接变成网页中的强制换行。需要换行时使用单标签 `<br>`。

```html
<body>
    第一行内容
    <br>
    第二行内容
</body>
```

保存并刷新后，应看到两行文字。去掉 `<br>` 后，两行源码通常会显示在同一行中。



### 水平线：`hr`

`<hr>` 是单标签，默认显示一条水平线。讲义在个人简介案例中用它分隔标题和正文。

```html
<h1>尤雨溪</h1>
<hr>
<p>这里是个人简介。</p>
```

`p` 组织段落，`br` 在当前文字中换行，`hr` 表示内容之间的分隔。三者产生的效果不同。


## 8. 文本格式化标签

文本格式化标签为文字添加加粗、倾斜、下划线或删除线，帮助突出内容。它们都是双标签，默认不会像标题或段落一样各自占一行。

| 默认效果 | 讲义重点标签 | 另一组标签 |
| --- | --- | --- |
| 加粗 | `strong` | `b` |
| 倾斜 | `em` | `i` |
| 下划线 | `ins` | `u` |
| 删除线 | `del` | `s` |

```html
<p>
    <strong>加粗文字</strong>
    <em>倾斜文字</em>
    <ins>下划线文字</ins>
    <del>删除线文字</del>
</p>
```

讲义把 `strong`、`em`、`ins`、`del` 概括为自带强调含义的标签，作为本节重点。

**补充说明：**它们的具体语义不同：`strong` 表示重要性，`em` 表示强调，`ins` 表示插入的内容，`del` 表示删除的内容。例如，用 `<del>` 表达旧内容被删去，用 `<ins>` 表达新内容被加入，比单纯记住线条效果更准确。

讲义自查：`strong` 默认加粗，`em` 默认倾斜；两个格式化标签通常在一行内显示，空间不足时仍会自然换行。


## 9. 图像标签

### 基本使用

图像标签 `<img>` 用于在网页中插入图片，是单标签。讲义写法：

```html
<img src="图片的 URL">
```

本日使用 `src` 指定图片的位置和名称。讲义中的两张图片示例：

```html
<img src="./cat.jpg">
<img src="./dog.jpg">
```



练习时，需要把真正的 `cat.jpg`、`dog.jpg` 文件放在 HTML 文件同一文件夹中。输入 `./` 后，VS Code 可以提示当前文件夹里的文件。保存并打开页面后，应看到对应图片。

### 常见属性

| 属性 | 作用 | 讲义说明 |
| --- | --- | --- |
| `src` | 图片地址 | 本日写法中必须指定，用于查找图片 |
| `alt` | 替换文本 | 图片无法显示时显示的文字 |
| `title` | 提示文本 | 鼠标悬停在图片上时显示 |
| `width` | 图片宽度 | 值写数字，不写单位 |
| `height` | 图片高度 | 值写数字，不写单位 |

讲义写法：

```html
<img src="./cat.jpg" alt="替换文本" title="提示文本">
```

属性的通用写法是 `属性名="属性值"`。属性写在开始标签的尖括号里、标签名后面，标签名和属性之间用空格隔开；不同属性的先后顺序不影响含义。



练习示例：

```html
<img src="./cat.jpg" alt="一只猫" title="猫咪照片" width="200">
```

保存并刷新后，图片宽度应变成 200 像素；悬停时可看到“猫咪照片”。将 `src` 暂时改为不存在的文件，可观察替换文本，再恢复正确路径。

**补充：**`alt` 还用于给辅助阅读工具提供图片含义。尺寸可先只设宽或高，让另一边按原比例缩放；同时设置不匹配的宽高会使图片变形。Notebook 中的配图是讲义效果图，并不会自动成为练习项目里的 `cat.jpg` 文件。


## 10. 路径

路径是查找文件时，从起点到终点经过的路线。写图片、音频、视频或页面地址时，都要先判断起点在哪里。

### 相对路径：从当前文件出发

讲义用下面的目录说明当前 HTML 文件与图片之间的位置关系。



| 写法 | 含义 |
| --- | --- |
| `./` | 当前文件所在文件夹 |
| `../` | 当前文件所在文件夹的上一级 |
| `文件夹名/` | 进入该文件夹 |
| `/` | 路径各级之间的分隔符 |

在 HTML 文件中，当前文件指当前的 `.html` 文件。练习示例：

```text
day01/
├── cat.jpg
└── 案例/
    └── 11-相对路径/
        ├── 相对路径.html
        ├── 1.jpg
        └── images/
            └── dog.jpg
```

在 `相对路径.html` 中写：

```html
<!-- 同级：当前文件夹里的图片 -->
<img src="./1.jpg" alt="同级图片">

<!-- 下级：进入 images 文件夹 -->
<img src="./images/dog.jpg" alt="下级图片">

<!-- 上两级：先回到 案例，再回到 day01 -->
<img src="../../cat.jpg" alt="上级图片">
```

保存并刷新后，三张图片都应显示。文件夹层级变了，路径也需要重新计算，不能只复制原来的字符串。

### 绝对路径与在线地址

讲义区分：Windows 的本地绝对路径从盘符出发，Mac 的本地绝对路径从根目录 `/` 出发。讲义展示的本地写法是：

```html
<img src="C:\images\mao.jpg">
```

Windows 常见文件路径使用 `\`，课程建议网页中统一使用 `/`。讲义还给出完整的在线地址：

```html
<img src="https://www.itheima.com/images/logo.png">
```

完整在线地址可用于引用外部资源或制作友情链接。

> **补充说明：**本地盘符路径依赖自己的电脑，发布网页后，访问者无法读取这个路径。项目内资源通常用相对路径，外部资源用完整 URL。以 `/images/a.jpg` 开头的 Web 路径是从站点根目录查找，不能把开头的 `/` 直接理解成“进入当前文件夹”。在线图片也可能因地址失效或访问限制而无法显示。


## 11. 超链接

超链接标签 `a` 用于点击后跳转到其他页面，是双标签。`href` 指定跳转地址，标签中间的内容是用户可点击的文字。

讲义写法：

```html
<a href="https://www.baidu.com">跳转到百度</a>
```

默认在当前页面所在窗口中打开目标地址。要在新窗口或新标签页打开，可以加 `target="_blank"`：

```html
<a href="https://www.baidu.com" target="_blank">跳转到百度</a>
```

保存并打开后，点击第一种写法会在当前标签页跳转；点击第二种通常会打开新标签页，具体由浏览器设置决定。

### 开发初期的空链接

讲义在尚未确定目标地址时使用 `#`：

```html
<a href="#">暂未确定的链接</a>
```

> **勘误：**讲义说 `#` 会“在当前页面刷新一次”。更准确地说，它是当前页面的空片段链接，通常会改变地址中的片段并滚动到页面顶部，不会重新加载整个页面。它适合作为本日练习的占位地址，正式页面应补上实际目标。

本节自查：`a` 用来跳转，`href` 决定目标，`target="_blank"` 决定在新窗口或标签页打开。


## 12. 音频标签

音频标签 `audio` 是双标签。讲义基本写法：

```html
<audio src="音频的 URL"></audio>
```

| 属性 | 作用 | 讲义说明 |
| --- | --- | --- |
| `src` | 音频地址 | 本日直接指定地址；常见格式为 MP3、Ogg、Wav |
| `controls` | 显示音频控制面板 | 可让用户播放、暂停等 |
| `loop` | 循环播放 | 播放完后再次播放 |
| `autoplay` | 自动播放 | 浏览器一般会限制自动播放 |

练习时，将真实音频文件 `music.mp3` 放在 HTML 文件同一文件夹中：

```html
<audio src="./music.mp3" controls loop></audio>
```

保存并刷新后，应看到音频控制面板。点击播放可听到音频，播放结束后重复播放。未添加 `controls` 时，通常不会出现可见的播放面板。

### 属性简写

讲义拓展：属性名和属性值相同时，可以简写成一个单词。例如：

```html
<audio src="./music.mp3" controls="controls"></audio>
<audio src="./music.mp3" controls></audio>
```

两种写法含义相同。**补充：**`controls`、`loop`、`autoplay` 属于布尔属性，出现就表示启用；不要用 `autoplay="false"` 来关闭，关闭时删除该属性。这个规则只适用于相应的布尔属性，并不适用于所有 HTML 属性。


## 13. 视频标签

视频标签 `video` 是双标签。讲义基本写法：

```html
<video src="视频的 URL"></video>
```

| 属性 | 作用 | 讲义说明 |
| --- | --- | --- |
| `src` | 视频地址 | 本日直接指定地址；常见格式为 MP4、WebM、Ogg |
| `controls` | 显示视频控制面板 | 可播放、暂停、调整音量等 |
| `loop` | 循环播放 | 播放结束后再次播放 |
| `muted` | 静音播放 | 关闭视频声音 |
| `autoplay` | 自动播放 | 讲义说明浏览器支持静音状态下自动播放 |

练习时，将真实视频文件 `video.mp4` 放在 HTML 文件同一文件夹中：

```html
<video src="./video.mp4" controls></video>
```

保存并刷新后，应看到视频区域和控制面板，点击播放可观看。需要尝试静音自动循环播放时，可以写：

```html
<video src="./video.mp4" controls autoplay muted loop></video>
```

**补充：**`muted` 可满足常见自动播放策略，但自动播放仍由浏览器策略决定，不保证在所有环境成功。音视频文件能否播放还取决于内部编码是否受支持，扩展名相同不代表一定能播放。练习中先验证手动播放，再观察自动播放。


## 综合案例一：个人简介

讲义的目标页面介绍尤雨溪，将标题、水平线、段落、链接和图片组合起来。



### 制作思路

> 从上到下，先整体再局部，逐步分析制作。

按照讲义给出的流程：**分析内容 → 写代码 → 保存 → 刷新浏览器看效果**。

1. 整体分成个人姓名、简介和照片，再分成学习经历、主要成就、社会任职等部分。
2. 姓名用 `h1`，下方用 `hr` 分隔。分区名称用较低级标题，说明文字用 `p`。
3. 简介中的 Vue.js、HTML5、JavaScript 用 `a`，照片用 `img`。
4. 先完成文字结构，再逐步补图片和链接，每次保存后观察效果。
