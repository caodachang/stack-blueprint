## 1. 基础认知

### 1.1 CSS 的作用

CSS 是 **Cascading Style Sheets（层叠样式表）**，用于给 HTML 标签设置样式。前两天用 HTML 表达页面内容和结构，本节用 CSS 改变它们的颜色、大小、字体和对齐方式。

| 技术 | 负责的部分 | 本课中的例子 |
| --- | --- | --- |
| HTML | 结构 | 用 `p` 表示一段文字 |
| CSS | 表现 | 将这段文字设为红色、30px |
| JavaScript | 行为 | 页面交互和动态效果，后续学习 |

### 1.2 CSS 语法与初体验

一条规则由选择器和大括号内的声明组成：选择器找到标签，声明描述要改的样式。每条声明写成 `属性名: 属性值;`。

```css
p {
    color: red;
}
```

`p` 是选择器，`color` 是属性名，`red` 是属性值。标点使用英文输入状态下的 `{ } : ;`，每条声明写完加分号；CSS 注释写成 `/* 注释内容 */`。



课堂文件 `01-体验css.html` 在 `<head>` 的 `<title>` 下方放置 `<style>`：

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta http-equiv="X-UA-Compatible" content="IE=edge">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
    <style>
        /* css注释 */
        /* 这里写的都是css */
        /* 选择器 {css属性} */
        /* 选择器: 查找标签 */
        p {
            /* 文字颜色变红色 */
            color: red;
            /* 字变大 px:像素 */
            font-size: 30px;
            /* 背景颜色 */
            background-color: green;
            /* width  height; */
            width: 400px;
            height: 400px;
        }
    </style>
</head>
<body>
    <p>这是一个p标签</p>
</body>
</html>
```

打开后，段落文字变成红色，字号为 `30px`，段落区域为 `400px × 400px`，背景为绿色。

| 属性 | 作用 | 课堂取值 |
| --- | --- | --- |
| `color` | 文字颜色 | `red` |
| `font-size` | 字体大小 | `30px` |
| `background-color` | 背景颜色 | `green` |
| `width` | 宽度 | `400px` |
| `height` | 高度 | `400px` |


### 1.3 CSS 的三种引入方式

| 方式 | 写在哪里 | 作用范围 | 课程中的使用场景 |
| --- | --- | --- | --- |
| 内嵌式 | HTML 的 `style` 标签中 | 当前页面 | 小案例、课堂演示 |
| 外联式（外链式） | 单独的 `.css` 文件，通过 `link` 引入 | 引入该文件的页面 | 项目中，多页复用样式 |
| 行内式 | 标签的 `style` 属性中 | 当前标签 | 本阶段少用，后续配合 JS 学习 |

内嵌式已经在上一个示例中使用。课程约定把 `<style>` 放在 `<head>` 内。

外联式需要两个文件。课堂文件 `02-css引入方式.html` 与 `my.css` 放在同一目录，HTML 用 `href="./my.css"` 指向样式表，`rel="stylesheet"` 表示与当前页面的关系是样式表。

`my.css` 只写 CSS，不添加 `<style>` 标签：

```css
/* 选择器 {} */
p {
    color: red;
}
```

`02-css引入方式.html` 同时演示外联式和行内式：

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta http-equiv="X-UA-Compatible" content="IE=edge">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
    <!-- 关系: 样式表 -->
    <link rel="stylesheet" href="./my.css">
</head>
<body>
    <!-- css到底能写在哪里 -->
    <p>这是p标签</p>
    <div style="color: green; font-size: 30px;">这是div标签</div>
    <div>这个div是什么颜色</div>
</body>
</html>
```

保存两个文件后打开 HTML：`p` 因外部样式变为红色；第一个 `div` 因自身的 `style` 变为绿色、30px；第二个 `div` 使用默认文字样式。只有引入 `my.css` 的页面才会应用其中的规则。


## 2. 基础选择器

选择器先选中页面里的标签，再为它们设置样式。四种基础写法如下，注意 CSS 中的点号或井号不写进 HTML 属性值。

| 选择器 | 写法 | 查找依据 | 范围 |
| --- | --- | --- | --- |
| 标签选择器 | `p { ... }` | 标签名 | 页面中的所有 `p` |
| 类选择器 | `.red { ... }` | `class` 中的类名 | 所有带 `red` 类的元素 |
| ID 选择器 | `#blue { ... }` | `id` 的值 | 合法页面中具有该唯一 ID 的元素 |
| 通配符选择器 | `* { ... }` | 任意标签 | 页面中的所有元素 |

### 2.1 标签选择器

标签选择器直接使用标签名。它选中一类标签，不能只凭这个写法区分其中某一个；不论嵌套多深，对应标签都能被找到。

课堂文件 `03-选择器-标签.html`：

CSS（放在 `<style>` 内）：

```css
/* 选择器 {} */
/* 标签选择器 就是 以标签名命名的选择器 */
p {
    color: red;
}

/* 标签选择器 选中所有的这个标签都生效css */
```

HTML（放在 `<body>` 内）：

```html
<p>pppppppp</p>
<p>这个p是什么颜色呢</p>
<p>2222</p>
```

三段 `p` 都显示红色。


### 2.2 类选择器

先在 CSS 中用 `.类名` 定义规则，再在 HTML 的 `class` 属性中使用类名。类名可以重复，一条类规则可以被不同标签复用；同一标签也可以同时使用多个类，类名之间用空格分隔。

课堂文件 `04-选择器-类.html`：

CSS（放在 `<style>` 内）：

```css
.red {
    color: red;
}

.size {
    font-size: 66px;
}
```

HTML（放在 `<body>` 内）：

```html
<!-- 类: 定义 和 使用才能生效 -->
<p>111</p>
<!-- 一个标签可以使用多个类名 , 需要空格隔开即可 -->
<p class="red size">222</p>
<div class="red">这个标签文字也要变红</div>
```

第一段 `p` 没有使用类，保持默认样式；第二段同时使用 `red` 和 `size`，文字红色、字号 66px；`div` 只使用 `red`，文字红色、字号保持默认。

> `class="red size"` 表示两个类，CSS 分别写 `.red`、`.size`；HTML 中不写 `class=".red"`。只定义类而没有在标签上使用，也不会产生对应效果。

课程建议类名由字母、数字、下划线、中划线组成，并且不要以数字或中划线开头，例如 `news-title`、`color1`。把重复的样式抽成公共类，可以减少重复代码。

**补充：**这里沿用课程的命名约定；CSS 语法允许某些以中划线开头的名称，课程规则是初学时采用的保守写法。


### 2.3 ID 选择器，以及类与 ID 的区别

HTML 用 `id="blue"` 标识元素，CSS 用 `#blue` 选择它。一个元素只有一个 ID 值，同一个文档中的 ID 必须唯一；不能像 `class` 那样用空格填写多个 ID。

课堂文件 `05-选择器-id.html` 的样式如下：

```css
/* 定义id选择器 */
#blue {
    color: skyblue;
}
```

**原资料修正：**该文件同时给 `div` 和 `p` 设置了 `id="blue"`，与讲义第 19 页“ID 不可重复”的要求矛盾。下面保留原选择器和文字，仅移除第二个标签重复的 ID：

```html
<div id="blue">这个div文字是蓝色的</div>
<p>111</p>
```

修正后只有 `div` 被 `#blue` 选中，显示天蓝色。若想让多个标签共用相同样式，应使用类。

| 比较项 | `class` | `id` |
| --- | --- | --- |
| 可否重复 | 多个元素可以用同一个类 | 文档内唯一 |
| 一个元素的取值 | 可以有多个类，空格分隔 | 一个 ID |
| 选择器前缀 | `.` | `#` |
| 课程中的开发建议 | 最常用，用于设置和复用样式 | 常配合 JS 使用，本课一般不用来组织样式 |


### 2.4 通配符选择器

`*` 可以选中页面中的所有标签。课堂文件 `06-选择器-通配符.html`：

CSS（放在 `<style>` 内）：

```css
* {
    color: red;
}
```

HTML（放在 `<body>` 内）：

```html
<div>div</div>
<p>pppp</p>
<h1>h1</h1>
<span>span</span>
<p>pppp</p>
<h2>h2</h2>
```

这些标签的文字都会变为红色。课程中一般使用标签和类选择器；通配符可能用于小页面中统一清除标签的默认 `margin`、`padding`，相关属性后续学习，本节先理解其范围。


## 3. 字体和文本样式

### 3.1 字体大小：`font-size`

本课使用 `数字 + px` 设置字号。`px` 表示像素，写 `font-size: 30;` 不符合本例所需的长度写法，应写 `30px`。

课堂文件 `07-字号.html`：

CSS（放在 `<style>` 内）：

```css
p {
    font-size: 30px;
}
```

HTML（放在 `<body>` 内）：

```html
<!-- 默认字号是16 -->
<p>段落文字</p>
```

文字显示为 30px。课程以 Chrome 常见默认字号 16px 为参照。

**补充：**浏览器设置或已有样式可能改变默认字号，16px 不是所有文字都固定采用的大小。

### 3.2 字体粗细：`font-weight`

| 效果 | 关键字 | 常用数字 |
| --- | --- | --- |
| 正常 | `normal` | `400` |
| 加粗 | `bold` | `700` |

课程还介绍 `100`～`900` 的整百数。字体不一定提供全部粗细，因此换一个数字后可能看不到变化；本阶段重点掌握 400、700。

课堂文件 `08-文字粗细.html`：

CSS（放在 `<style>` 内）：

```css
div {
    /* 加粗 */
    font-weight: 700;
}

h1 {
    /* 不加粗 */
    font-weight: 400;
}
```

HTML（放在 `<body>` 内）：

```html
<div>这是div</div>
<h1>一级标题</h1>
```

`div` 被加粗；`h1` 虽然默认加粗，但这里用 `400` 改为正常粗细。标题标签的语义仍然是一级标题。

### 3.3 字体是否倾斜：`font-style`

`normal` 表示正常，`italic` 表示倾斜。课堂文件 `09-倾斜.html`：

CSS（放在 `<style>` 内）：

```css
div {
    /* 倾斜 */
    font-style: italic;
}

em {
    /* 正常的, 不倾斜 */
    font-style: normal;
}
```

HTML（放在 `<body>` 内）：

```html
<div>div文字</div>
<em>em</em>
```

`div` 的文字倾斜；`em` 原本常以斜体呈现，这里改为正常显示，强调语义仍然保留。


### 3.4 常见字体系列

字体系列是一类字体的统称，具体字体名称则指某一种字体。

| 系列 | 课程介绍的特点 | 常见用途 | 例子 |
| --- | --- | --- | --- |
| `sans-serif` 无衬线 | 笔画粗细较均匀，首尾没有笔锋装饰 | 网页 | 黑体、Arial |
| `serif` 衬线 | 笔画粗细有变化，首尾有笔锋装饰 | 报刊、书籍 | 宋体、Times New Roman |
| `monospace` 等宽 | 字符采用固定宽度，便于纵向对齐 | 程序代码 | Consolas、Fira Code |

### 3.5 指定字体：`font-family`

取值可以是一组具体字体，最后用通用字体系列兜底：

```css
font-family: "Microsoft YaHei", 黑体, sans-serif;
```

浏览器按从左到右的顺序尝试字体：本机没有前一个字体时再使用后一个；具体字体都不可用时，使用操作系统为通用系列提供的字体。课程举例 Windows 常见微软雅黑、macOS 常见苹方，实际还取决于安装情况。

包含多个单词的字体名称推荐加引号，例如 `"Times New Roman"`；最后的 `sans-serif`、`serif`、`monospace` 不加引号。优先使用常见系统字体，并提供后备字体。

课堂文件 `10-字体.html`：

CSS（放在 `<style>` 内）：

```css
div {
    /* font-family: 宋体; */
    /* 如果用户电脑没有安装微软雅黑, 就按黑体显示文字 */
    /* 如果电脑没有安装黑体, 就按任意一种非衬线字体系列显示 */
    font-family: 微软雅黑, 黑体, sans-serif;
}
```

HTML（放在 `<body>` 内）：

```html
<div>
    这是一个div标签
</div>
```

按本机可用字体显示这段文字；没有微软雅黑时尝试黑体，再尝试无衬线系列。


### 3.6 样式的层叠

同一条规则重复声明同一个属性时，后面的有效声明覆盖前面的声明。课堂文件 `11-拓展-层叠性.html`：

CSS（放在 `<style>` 内）：

```css
p {
    /* 层叠性: 后面的覆盖前面的属性 */
    /* 覆盖 */
    color: red;
    color: blue;
}
```

HTML（放在 `<body>` 内）：

```html
<p>ppppp</p>
```

这段文字最终显示蓝色。这说明 CSS 可以逐层叠加、覆盖样式。

**补充：**本例的声明处于同一规则、同一优先级下。不能据此推断所有规则冲突都只看谁写在后面；本节先掌握这里展示的覆盖现象。

### 3.7 字体属性连写：`font`

`font` 是复合属性，一次设置多项字体样式。课程使用的顺序是：

```css
font: italic 700 66px 宋体;
```

对应 `font-style → font-weight → font-size → font-family`，即“倾斜、加粗、字号、字体”。在本课的连写形式中，前两项可以省略，字号和字体必须保留：

```css
font: 100px 微软雅黑;
```

**省略不表示保留之前的设置。**连写会将未写出的相关属性重置，例如省略倾斜和粗细时，恢复正常样式。若需要混用连写与单独属性，要么把取值写进 `font`，要么把单独的覆盖声明放在 `font` 后面。

课堂文件 `12-font属性.html`：

CSS（放在 `<style>` 内）：

```css
p {
    /* font-size: ;
    font-style: ;
    font-weight: ;
    font-family: ; */
    /* font: style   weight   size    字体; */

    /* font: italic 700 66px 宋体;
    font-style: normal; */
    font: 100px 微软雅黑;

    /* 一个属性冒号后面书写多个值的写法 -- 复合属性 */
}
```

HTML（放在 `<body>` 内）：

```html
<p>这是p标签</p>
```

当前生效的是 100px 微软雅黑，注释中的倾斜、加粗示例不生效。练习另一组写法时，先注释掉最后的 `font: 100px 微软雅黑;`，再取消上面两条声明的注释：先连写倾斜、加粗、66px、宋体，再用后面的 `font-style: normal;` 改回不倾斜。否则最后的 100px 连写仍会覆盖前面的字体设置。


### 3.8 文本首行缩进：`text-indent`

`text-indent` 控制文本第一行的缩进，后面的行不会跟着缩进。可以用 `px`，也可以用 `em`。

`1em` 等于当前元素的 `font-size`；中文正文常用 `2em` 实现约两个汉字宽度的首行缩进。当前字号为 40px 时，`2em` 就是 80px，修改字号后缩进会相应变化。

课堂文件 `13-文本缩进.html`：

CSS（放在 `<style>` 内）：

```css
p {
    /* text-indent: 50px; */
    /* 首行缩进2个字的大小 */
    /* 默认字号: 16px ; 32 */
    /* text-indent: 40px;
    font-size: 20px; */

    /* em: 一个字的大小 */
    text-indent: 2em;
    font-size: 40px;
}
```

HTML（放在 `<body>` 内）：

```html
<p>2019年，事件视界望远镜团队让世界首次看到了黑洞的样子。不过，研究人员公布的这张发光环形物体的图像并不是传统的图片，而是经过计算获得的。利用位于美国、墨西哥、智利、西班牙和南极地区的射电望远镜所得到的数据，研究人员进行了数学转换，最终合成了这张标志性的图片。研究团队还发布了实现这一壮举所用的编程代码，并撰文记录这一发现，其他研究者也可以在此基础上进一步加以分析。</p>
```

首行向右缩进 80px，段落其余行保持正常起点。与固定 `40px` 缩进相比，`2em` 更适合随字号变化的正文。

课程 XMind 提醒，本课把该属性设置在 `p` 这样的块级文本容器上；直接设置在普通行内 `span` 上，不能得到同样的段落首行缩进效果。


### 3.9 文本水平对齐：`text-align`

| 值 | 效果 |
| --- | --- |
| `left` | 左对齐 |
| `center` | 居中对齐 |
| `right` | 右对齐 |

给**承载内容的容器**设置 `text-align`。例如标题文字位于 `h1` 内，就给 `h1` 设置；图片放在 `div` 内，就给这个 `div` 设置。

课程中的 `text-align: center;` 可以让文字，以及默认显示方式下的 `span`、`a`、`input`、`img` 等行内内容在容器内水平居中。

课堂文件 `14-水平对齐方式.html`：

CSS（放在 `<style>` 内）：

```css
h1 {
    /* text-align: left; */
    /* text-align: right; */
    text-align: center;
}

body {
    text-align: right;
}
```

HTML（放在 `<body>` 内）：

```html
<h1>新闻标题</h1>

<img src="./images/1.jpg" alt="">
```

使用课堂目录中的 `images/1.jpg`。标题因自身的 `text-align: center;` 居中；图片直接放在 `body` 中，因 `body` 的 `text-align: right;` 靠右。若要让图片居中，把它所在容器的取值改为 `center`。

> `text-align` 调整的是容器内的内容，不会让这个容器本身移动到页面中央。盒子本身的居中方法在后面的拓展中学习。


### 3.10 文本修饰：`text-decoration`

| 值 | 效果 | 课程中的使用提示 |
| --- | --- | --- |
| `underline` | 下划线 | 常见 |
| `line-through` | 删除线 | 较少使用 |
| `overline` | 上划线 | 很少使用 |
| `none` | 没有装饰线 | 常用于清除链接默认下划线 |

课堂文件 `15-文本修饰.html`：

CSS（放在 `<style>` 内）：

### 3.11 行高：`line-height`

行高控制一行文字所占的高度，影响多行文本的疏密。讲义用“上间距 + 文本高度 + 下间距”说明一行的组成：



| 写法 | 含义 | 例子 |
| --- | --- | --- |
| 数字 + `px` | 固定行高 | `line-height: 50px;` |
| 不带单位的数字 | 当前字号的倍数 | `line-height: 1.5;` |

行高也可以放进 `font`，用斜杠与字号分隔，顺序为 `style weight size/line-height family`。课堂文件 `16-行高.html`：

CSS（放在 `<style>` 内）：

```css
p {
    /* line-height: 50px; */
    /* 自己字号的1.5倍 */
    /* line-height: 1.5; */

    /* 66px 宋体 倾斜 加粗 行高是2倍 */
    font: italic 700 66px/2 宋体;
}
```

HTML（放在 `<body>` 内）：

```html
<p>2019年，事件视界望远镜团队让世界首次看到了黑洞的样子。不过，研究人员公布的这张发光环形物体的图像并不是传统的图片，而是经过计算获得的。利用位于美国、墨西哥、智利、西班牙和南极地区的射电望远镜所得到的数据，研究人员进行了数学转换，最终合成了这张标志性的图片。研究团队还发布了实现这一壮举所用的编程代码，并撰文记录这一发现，其他研究者也可以在此基础上进一步加以分析</p>
```

当前字号为 66px，`/2` 表示两倍行高，即 132px；字体为宋体、倾斜、加粗。注释中还演示了独立设置固定行高与 1.5 倍行高。

与其他字体属性一样，`font` 会重置行高。需要保留特定行高时，把它写在连写的斜杠后，或在 `font` 后再单独写 `line-height`。

### 3.12 让单行文字垂直居中

本课的方法是将行高设为文字容器的高度。课堂文件 `17-内容垂直居中.html`：

CSS（放在 `<style>` 内）：

```css
div {
    width: 552px;
    height: 400px;
    background-color: pink;
    /* background-color: green; */
    text-align: center;
    /* 文字是单行的 */
    /* 垂直居中技巧: 设置行高属性值 = 自身高度属性值 */
    line-height: 400px;
    font-size: 61px;
}
```

HTML（放在 `<body>` 内）：

```html
<div>文字</div>
```

粉色区域宽 552px、高 400px；`text-align: center;` 让文字水平居中，`line-height: 400px;` 让这行文字在高度方向居中。该技巧用于单行文字，不能直接套到多行段落。

课程还介绍精准布局时使用 `line-height: 1;` 减少文字上下的额外行间距。

**补充：**`1` 表示行高等于字号，不是行高为 1px，也不会自动清除 `p` 等标签的默认外边距；实际字形仍取决于字体。

## 5. 拓展：颜色的常见取值

课程在调试工具之后补充颜色写法。同一组颜色值既能用于 `color` 设置文字颜色，也能用于 `background-color` 设置背景颜色。

| 表示方式 | 含义 | 例子 |
| --- | --- | --- |
| 关键字 | 预定义的颜色名 | `red`、`green`、`blue`、`yellow`、`orange`、`skyblue`、`pink` |
| `rgb()` | 红、绿、蓝三个通道 | `rgb(255, 0, 0)` |
| `rgba()` | RGB 加透明度 | `rgba(0, 0, 0, 0.5)` |
| 十六进制 | `#` 后依次写 RGB 通道 | `#ff0000`、`#f00` |

### 5.1 RGB 和 RGBA

本课的 RGB 写法中每个通道范围为 0～255：

| 值 | 颜色 |
| --- | --- |
| `rgb(255, 0, 0)` | 红色 |
| `rgb(0, 255, 0)` | 绿色 |
| `rgb(0, 0, 255)` | 蓝色 |
| `rgb(0, 0, 0)` | 黑色 |
| `rgb(255, 255, 255)` | 白色 |

RGBA 多出的 `a` 表示透明度，范围为 0～1：0 完全透明，1 完全不透明，0.5 表示半透明。`rgba(0, 0, 0, 0.5)` 可以简写为 `rgba(0, 0, 0, .5)`。

### 5.2 十六进制颜色

`#rrggbb` 每两位一组，依次表示红、绿、蓝。每位可用 `0`～`9`、`a`～`f`，例如 `ff` 相当于通道值 255，`00` 相当于 0。



只有三组各自的两位都相同时，才能缩写为三位：`#ffaabb → #fab`、`#ffffff → #fff`、`#000000 → #000`。三组不必彼此相等，但每一组内部必须是重复字符。

`#ffaabc` 中最后一组是 `bc`，不能按这种方法写成 `#fabc`。

**补充：**四位十六进制本身在现代 CSS 中有其他含义，但它不是这里六位颜色的正确缩写，本课仍按六位、三位形式练习。

课程强调：案例先使用效果图标注的颜色，实际按设计稿或测量工具取得颜色后直接填写，不必凭感觉重新选择。


## 6. 拓展：让标签本身水平居中

前面的 `text-align: center;` 让容器内部的内容居中。如果要让 `div`、`p`、标题这样的块级盒子本身居中，课程使用固定宽度配合 `margin: 0 auto;`。

课堂文件 `18-拓展-标签水平居中.html`：

CSS（放在 `<style>` 内）：

```css
div {
    width: 300px;
    height: 300px;
    background-color: pink;
    margin: 0 auto;
}
```

HTML（放在 `<body>` 内）：

```html
<div></div>
```

打开后，300px 宽的粉色盒子位于父容器水平方向的中央。这里 `0` 设置上下外边距，`auto` 让左右外边距自动分配，形成居中效果。

| 目标 | 写法 | 设置给谁 |
| --- | --- | --- |
| 文字、`span`、`a`、`input`、`img` 等内容水平居中 | `text-align: center;` | 承载它们的容器 |
| 本课中的固定宽度块级盒子水平居中 | `margin: 0 auto;` | 盒子本身，并设置宽度 |

> 未设置宽度时，普通块级盒子通常占满父容器的可用宽度，左右没有明显的剩余空间，便看不到盒子移动到中央的效果。盒子与盒子内部的内容可以分别设置居中。
