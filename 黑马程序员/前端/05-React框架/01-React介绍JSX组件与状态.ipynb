## 二、React 是什么

React 由 Meta（原 Facebook）开发，是一个用于构建 Web 和原生交互界面的 JavaScript 库。它把用户界面拆成一个个组件，组件拥有自己的结构、逻辑和样式，并可以相互嵌套、重复使用。

### React 的主要优势

1. **组件化开发**：把复杂页面拆成独立、可复用、易维护的组件。
2. **较好的更新性能**：根据状态变化计算需要更新的界面，减少无意义的 DOM 操作。
3. **生态丰富**：路由、状态管理、请求、测试和工程化工具选择较多。
4. **跨平台能力**：同一套 React 思想可以扩展到 React Native 等原生平台。

React 的核心心智模型是：**状态决定 UI**。当状态变化时，组件重新执行并得到新的界面描述。

## 四、JSX：在 JavaScript 中写 UI

JSX 是 JavaScript 和 XML/HTML 的语法扩展，用来声明 React 组件的 UI：

```jsx
const message = 'this is message'

function App() {
  return (
    <div>
      <h1>this is title</h1>
      <p>{message}</p>
    </div>
  )
}
```

JSX 的优势是把 HTML 的声明式结构和 JavaScript 的编程能力放在一起。JSX 不是浏览器原生语法，构建工具会先把它转换成浏览器可以执行的 JavaScript。

### JSX 的基本规则

- 一个组件返回多个元素时，需要用一个根元素或空 Fragment `<>...</>` 包裹。
- 标签必须闭合，例如 `<img />`。
- HTML 属性使用驼峰写法，如 `className`、`tabIndex`、`onClick`。
- `class` 在 JSX 中要写成 `className`，`for` 要写成 `htmlFor`。
- 注释写在 `{/* 注释内容 */}` 中。

## 五、JSX 中的 JavaScript 表达式

在 JSX 中使用一对大括号 `{}` 插入 JavaScript **表达式**：

```jsx
const message = 'this is message'

function getAge() {
  return 18
}

function App() {
  const user = { name: 'jack' }
  return (
    <div>
      {'this is str'}
      <p>{message}</p>
      <p>{user.name}</p>
      <p>{getAge()}</p>
      <p>{1 + 2}</p>
    </div>
  )
}
```

可以使用变量、属性访问、算术运算、函数调用、三元表达式和逻辑运算。`if`、`switch`、变量声明等是语句，不能直接放进 `{}`；复杂判断应提前计算，或封装成函数后在 JSX 中调用。

```jsx
const type = 1

function getArticleJSX() {
  if (type === 0) return <div>无图模式</div>
  if (type === 1) return <div>单图模式</div>
  if (type === 3) return <div>三图模式</div>
  return null
}
```

## 六、列表渲染：`map` 与 `key`

React 没有 `v-for`，通常使用 JavaScript 原生 `map` 把数组转换成 JSX：

```jsx
const list = [
  { id: 1001, name: 'Vue' },
  { id: 1002, name: 'React' },
  { id: 1003, name: 'Angular' },
]

function App() {
  return (
    <ul>
      {list.map(item => (
        <li key={item.id}>{item.name}</li>
      ))}
    </ul>
  )
}
```

`key` 用于帮助 React 识别列表中的每一项，应该使用稳定、唯一的字符串或数字，优先使用后端数据中的 id。不要把数组下标作为长期列表的 key，尤其是列表会插入、删除或排序时，否则可能出现 DOM 复用和输入状态错位。

## 七、条件渲染

### 1. 逻辑与 `&&`

条件为真时渲染右侧内容：

```jsx
{hasMessage && <span>有一条新消息</span>}
```

### 2. 三元表达式

在两个 UI 结果之间选择：

```jsx
{loading ? <span>加载中...</span> : <span>加载完成</span>}
```

### 3. 复杂条件

当判断分支较多时，把逻辑放进函数或提前计算，返回对应 JSX：

```jsx
function renderStatus(status) {
  if (status === 'loading') return <Loading />
  if (status === 'error') return <ErrorTip />
  if (status === 'success') return <Result />
  return null
}

function App() {
  return <section>{renderStatus('success')}</section>
}
```

条件渲染的关键是让每个分支最终得到 JSX、字符串、数字、数组或 `null`。不需要展示内容时返回 `null`。

## 八、React 事件绑定

React 事件属性使用 `on` 加事件名，并遵循驼峰命名，例如 `onClick`、`onChange`、`onMouseEnter`。值应传入函数引用：

```jsx
function App() {
  const clickHandler = () => {
    console.log('按钮被点击了')
  }

  return <button onClick={clickHandler}>click me</button>
}
```

### 事件对象

回调函数的第一个参数就是事件对象：

```jsx
function App() {
  const handleInput = event => {
    console.log(event.target.value)
  }

  return <input onChange={handleInput} />
}
```

### 自定义参数

不能直接写 `onClick={clickHandler('jack')}`，否则组件渲染时就会立即调用。应使用箭头函数包一层：

```jsx
const clickHandler = name => {
  console.log(name)
}

<button onClick={() => clickHandler('jack')}>click me</button>
```

同时传递自定义参数和事件对象时，手动接收并传递事件：

```jsx
const clickHandler = (name, event) => {
  console.log(name, event)
}

<button onClick={event => clickHandler('jack', event)}>click me</button>
```

## 九、函数组件

在 React 中，函数组件通常是首字母大写的 JavaScript 函数，函数内部返回 UI：

```jsx
function Button() {
  return <button>click me</button>
}

function App() {
  return (
    <div>
      <Button />
      <Button></Button>
    </div>
  )
}
```

组件使用时像 HTML 标签一样书写。首字母大写非常重要：小写标签会被当成原生 HTML 标签，大写名称才会被 React 识别为自定义组件。组件可以嵌套，也可以在不同页面中重复使用，后续还会通过 props 接收外部数据。

## 十、`useState`：让组件拥有状态

`useState` 是 React Hook，用于给函数组件添加状态变量。状态更新后，React 会重新执行组件函数并更新界面：

```jsx
import { useState } from 'react'

function App() {
  const [count, setCount] = useState(0)

  return (
    <button onClick={() => setCount(count + 1)}>
      点击了 {count} 次
    </button>
  )
}
```

`useState(0)` 返回一个数组：第一项是状态值，第二项是更新状态的函数。通常使用数组解构命名为 `[state, setState]`。

### 状态修改规则

React 状态应当被视为只读数据：**替换它，不要直接修改它**。直接修改对象或数组可能不会触发正确的更新，也会让数据流难以追踪。

```jsx
const [user, setUser] = useState({ name: 'jack', age: 18 })

function updateName() {
  setUser({ ...user, name: 'rose' })
}

const [list, setList] = useState([])
setList([...list, newItem])
setList(list.filter(item => item.id !== id))
```

更新依赖旧值时可以使用函数形式：

```jsx
setCount(prevCount => prevCount + 1)
```

## 十一、React 组件样式

### 1. 行内样式

JSX 中的 `style` 接收 JavaScript 对象，CSS 属性名使用驼峰写法：

```jsx
<div style={{ color: 'red', fontSize: 20 }}>红色文字</div>
```

外层 `{}` 表示 JSX 表达式，内层 `{}` 表示样式对象。

### 2. CSS 类名

```css
/* index.css */
.foo {
  color: red;
  font-size: 20px;
}
```

```jsx
import './index.css'

function App() {
  return <span className="foo">this is span</span>
}
```

React 中使用 `className` 而不是 HTML 的 `class`。项目中通常把组件结构和样式文件拆开，保持 JSX 清晰。

## 十二、综合案例：B 站评论列表

案例需求：

1. 根据数组渲染评论列表。
2. 只有当前用户发表的评论显示删除按钮。
3. 渲染“最热/最新”导航并高亮当前 Tab。
4. 点击 Tab 后按点赞数或时间排序。
5. 评论删除后同步更新评论数量。

建议先拆分数据和交互，再实现界面：

```jsx
const defaultList = [
  {
    rpid: 3,
    user: { uid: '13258165', uname: '周杰伦', avatar: '' },
    content: '哎哟，不错哦',
    ctime: '10-18 08:15',
    like: 88,
  },
  {
    rpid: 2,
    user: { uid: '36080105', uname: '许嵩', avatar: '' },
    content: '我寻你千百度 日出到迟暮',
    ctime: '11-13 11:29',
    like: 88,
  },
]

const user = { uid: '13258165', uname: '当前用户' }
const tabs = [
  { type: 'hot', text: '最热' },
  { type: 'time', text: '最新' },
]
```

组件中用 `useState` 管理当前 Tab 和评论列表：

```jsx
import { useState } from 'react'

function App() {
  const [activeTab, setActiveTab] = useState('hot')
  const [list, setList] = useState(defaultList)

  const onDelete = rpid => {
    setList(prevList => prevList.filter(item => item.rpid !== rpid))
  }

  const onToggle = type => {
    setActiveTab(type)
    const newList = [...list].sort((a, b) => {
      if (type === 'time') return b.ctime.localeCompare(a.ctime)
      return b.like - a.like
    })
    setList(newList)
  }

  return (
    <div className="app">
      <nav>
        <span>评论 {list.length}</span>
        {tabs.map(tab => (
          <button
            key={tab.type}
            className={tab.type === activeTab ? 'active' : ''}
            onClick={() => onToggle(tab.type)}
          >
            {tab.text}
          </button>
        ))}
      </nav>

      <section>
        {list.map(item => (
          <article key={item.rpid}>
            <h4>{item.user.uname}</h4>
            <p>{item.content}</p>
            <small>{item.ctime} · 点赞数：{item.like}</small>
            {user.uid === item.user.uid && (
              <button onClick={() => onDelete(item.rpid)}>删除</button>
            )}
          </article>
        ))}
      </section>
    </div>
  )
}
```

真实项目中可以继续把导航、评论输入框、评论项拆成独立组件。拆分后每个组件的职责更单一，也更容易复用和测试。

## 十三、案例中的关键知识点

- `list.map(...)` 负责把评论数据转换为评论项 JSX。
- `key={item.rpid}` 为每条评论提供稳定身份。
- `activeTab` 决定当前 Tab 的 `className`，从而实现高亮。
- 删除时使用 `filter` 生成新数组，并调用 `setList` 触发更新。
- 排序前使用 `[...list]` 复制数组，避免直接修改 state。
- 条件表达式控制删除按钮是否出现。
- 评论数量直接读取 `list.length`，因此删除后会自动同步。

这体现了 React 的单向数据流：数据保存在状态中，UI 根据状态渲染，用户操作通过事件处理函数更新状态。

## 十四、常见错误与排查顺序

1. **JSX 报语法错误**：检查标签是否闭合、多个根元素是否使用 Fragment 包裹。
2. **属性不生效**：检查是否使用 `className`、`htmlFor` 和驼峰事件名。
3. **点击时页面加载或函数提前执行**：事件属性传函数引用，带参数时使用箭头函数包裹。
4. **列表警告缺少 key**：给最外层被 `map` 返回的元素添加稳定唯一的 `key`。
5. **状态修改后界面不更新**：不要直接 `push`、`splice` 或修改对象属性后等待 React 猜测，使用新数组或新对象调用 setter。
6. **条件渲染结果不对**：确认 `&&` 左侧不是意外的数字 0；复杂逻辑先写成函数或变量。
7. **排序改变了原始状态**：排序前复制数组，如 `const next = [...list].sort(...)`。

排查顺序建议是：先看浏览器控制台，再确认事件是否触发，然后检查 state 是否通过 setter 更新，最后检查 JSX 条件和列表 key。

## 十五、本日总结

- React 是以组件为核心、以状态驱动 UI 的前端库。
- JSX 是 JavaScript 的语法扩展，使用 `{}` 插入表达式。
- 列表用 `map` 渲染，条件用 `&&`、三元表达式或函数判断。
- 事件使用 `onClick` 等驼峰属性，传参时用箭头函数包裹。
- 函数组件首字母大写，组件可以嵌套和复用。
- `useState` 为组件提供响应式状态，状态更新必须通过 setter，并优先创建新对象或新数组。
- `className` 和 `style` 是 React 中最常见的基础样式写法。
