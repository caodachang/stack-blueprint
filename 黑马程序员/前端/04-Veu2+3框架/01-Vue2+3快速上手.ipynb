## 二、Vue 是什么

Vue 是一套用于构建用户界面的渐进式 JavaScript 框架。它把界面拆成可复用的组件，并通过数据驱动视图更新。

### 1. UI 与数据驱动

传统做法需要手动查询 DOM、修改文本、拼接 HTML；Vue 中只需要维护数据，数据变化后 Vue 会把变化同步到页面。

### 2. 渐进式

可以只引入 Vue 做简单的视图渲染，也可以继续使用组件、路由、状态管理和工程化工具，按项目复杂度逐步增加能力。

### 3. 框架与库

库通常解决某一类问题，调用时由业务代码主导；框架会规定一套组织代码的方式，并在合适的时机调用业务代码。Vue 既能像库一样按需使用，也提供完整的前端应用开发能力。

## 三、创建第一个 Vue 应用

Vue 2 CDN 方式的基本流程可以概括为 4 步：

1. 准备一个挂载容器。
2. 引入 Vue。
3. 创建 `new Vue()` 实例。
4. 通过 `el` 指定容器，通过 `data` 提供页面数据。

```html
<div id="app">
  <h1>{{ message }}</h1>
</div>

<script src="https://cdn.jsdelivr.net/npm/vue@2/dist/vue.js"></script>
<script>
  const app = new Vue({
    el: '#app',
    data: {
      message: 'Hello Vue'
    }
  })
</script>
```

`el` 决定 Vue 管理哪一段 DOM；`data` 中的属性可以直接在模板中使用。Vue 2 中常见的是 `new Vue({ ... })`，Vue 3 中通常使用 `createApp(App).mount('#app')`，但“数据驱动视图”的核心思想相同。

## 四、插值表达式 `{{ }}`

双大括号用于把数据或表达式的结果插入文本节点：

```html
<p>{{ username }}</p>
<p>{{ price + 100 }}</p>
<p>{{ isVip ? '会员价' : '普通价' }}</p>
<p>{{ user.name }}</p>
<p>{{ formatPrice(price) }}</p>
```

插值中只能写表达式，不能写完整语句：

- 可以写变量、属性访问、算术运算、三元表达式和函数调用。
- 不能写 `if`、`for`、变量声明等语句。
- 模板中使用的变量必须先在 `data`、`computed`、`methods` 等选项中声明。
- 插值只负责文本内容，不能用于 HTML 属性；属性绑定要使用 `v-bind`。

例如：

```html
<!-- 错误：不会把 src 当作 Vue 表达式解析 -->
<img src="imgUrl">

<!-- 正确 -->
<img :src="imgUrl">
```

## 五、响应式与 Vue Devtools

Vue 会把 `data` 中的属性代理到 Vue 实例上，因此模板中可以直接使用 `message`，在方法中也可以通过 `this.message` 访问。

当响应式数据发生变化时，Vue 会重新计算相关视图并更新 DOM：

```js
const app = new Vue({
  el: '#app',
  data: { count: 0 },
  methods: {
    add () {
      this.count++
    }
  }
})
```

开发时可以安装 Vue Devtools，在浏览器中查看组件树、响应式数据和事件，定位“数据变了但页面不对”的问题。

## 六、内容渲染指令：`v-text` 与 `v-html`

### `v-text`

把数据作为纯文本写入元素，效果类似 `textContent`：

```html
<p v-text="message"></p>
```

### `v-html`

把数据作为 HTML 解析后写入元素，效果类似 `innerHTML`：

```html
<div v-html="htmlContent"></div>
```

`v-html` 适合显示确实需要解析的可信 HTML；如果内容来自用户输入或不可信接口，可能产生 XSS 风险，默认应优先使用插值或 `v-text`。

## 七、条件渲染：`v-show` 与 `v-if`

```html
<div v-show="visible">频繁切换的内容</div>
<div v-if="isLogin">已登录</div>
<div v-else-if="isLoading">加载中</div>
<div v-else>请先登录</div>
```

- `v-show` 通过切换 `display` 控制显示，元素始终存在，适合频繁切换。
- `v-if` 真正创建或销毁元素，初始条件为假时不会渲染，适合条件变化不频繁或内容较重的场景。
- `v-else-if`、`v-else` 必须紧跟在 `v-if` 或 `v-else-if` 元素后面。

选择依据是“切换成本”和“初始渲染成本”：频繁显示隐藏用 `v-show`，按条件决定是否创建用 `v-if`。

## 八、事件处理：`v-on`、`methods` 与 `$event`

`v-on` 用于监听 DOM 事件，简写为 `@`：

```html
<button v-on:click="count++">点我</button>
<button @click="add(2)">加 2</button>
<button @click="handleClick">执行方法</button>
```

复杂逻辑放在 `methods` 中：

```js
new Vue({
  el: '#app',
  data: { count: 0 },
  methods: {
    add (step) {
      this.count += step
    },
    handleInput (event) {
      console.log(event.target.value)
    }
  }
})
```

方法中的 `this` 指向当前 Vue 实例。需要原生事件对象时，可以显式传入 `$event`：

```html
<button @click="handleClick($event, '保存')">保存</button>
```

常见事件修饰符有 `@submit.prevent`、`@keyup.enter`、`@click.stop`，用于减少重复的事件处理代码。

## 九、属性绑定：`v-bind` 与图片轮播

`v-bind` 用于把表达式结果绑定到 HTML 属性，简写为 `:`：

```html
<img :src="imageUrl" :alt="title">
<a :href="link">打开链接</a>
<button :disabled="isDisabled">提交</button>
```

### 图片轮播核心思路

1. 用数组保存图片地址。
2. 用 `index` 表示当前图片。
3. `:src="images[index]"` 显示当前图片。
4. 上一张、下一张按钮修改 `index`。
5. 到达边界时隐藏或禁用按钮。

```html
<div id="app">
  <img :src="images[index]" alt="轮播图">
  <button v-show="index > 0" @click="index--">上一张</button>
  <button v-show="index < images.length - 1" @click="index++">下一张</button>
</div>
<script>
new Vue({
  el: '#app',
  data: {
    index: 0,
    images: ['./images/1.jpg', './images/2.jpg', './images/3.jpg']
  }
})
</script>
```

## 十、列表渲染：`v-for` 与 `key`

`v-for` 可以遍历数组、对象和数字：

```html
<!-- 数组 -->
<li v-for="(item, index) in list" :key="item.id">
  {{ index + 1 }} - {{ item.name }}
</li>

<!-- 对象 -->
<span v-for="(value, key) in user" :key="key">
  {{ key }}：{{ value }}
</span>

<!-- 数字 -->
<span v-for="n in 5" :key="n">{{ n }}</span>
```

`key` 用来帮助 Vue 识别每一条列表项，应该使用稳定且唯一的字符串或数字，优先使用后端返回的 `id`。当列表会增删或排序时，不建议把数组下标作为 `key`，否则可能造成复用错误和输入框状态错位。

## 十一、列表案例：书架删除

```html
<div id="app">
  <ul>
    <li v-for="book in books" :key="book.id">
      <span>{{ book.name }}</span>
      <button @click="removeBook(book.id)">删除</button>
    </li>
  </ul>
</div>
<script>
new Vue({
  el: '#app',
  data: {
    books: [
      { id: 1, name: 'JavaScript 高级程序设计' },
      { id: 2, name: 'Vue.js 实战' },
      { id: 3, name: '深入理解计算机系统' }
    ]
  },
  methods: {
    removeBook (id) {
      this.books = this.books.filter(book => book.id !== id)
    }
  }
})
</script>
```

删除操作本质是根据 id 生成一个新数组，再交给 Vue 更新视图。不要直接通过 DOM 删除节点，否则数据和页面容易失去同步。

## 十二、双向绑定：`v-model`

`v-model` 适合表单控件，让“数据 → 表单”和“表单 → 数据”保持同步。

```html
<input v-model="username" placeholder="请输入用户名">
<textarea v-model="remark"></textarea>

<input type="checkbox" v-model="remember"> 记住我

<label><input type="radio" value="male" v-model="gender">男</label>
<label><input type="radio" value="female" v-model="gender">女</label>

<select v-model="city">
  <option value="beijing">北京</option>
  <option value="shanghai">上海</option>
</select>
```

登录表单通常把输入值放到 `data` 中，点击登录时读取 `this.username` 和 `this.password`。重置表单时直接恢复对应数据即可，页面会自动同步。常用修饰符包括：`v-model.trim` 去掉首尾空格，`v-model.number` 转为数字，`v-model.lazy` 改为在 change 时同步。

## 十三、综合案例：小黑记事本

功能要求：

- 展示事项列表。
- 输入内容并新增事项。
- 删除单条事项。
- 显示剩余数量。
- 一键清空全部事项。

```html
<div id="app">
  <input v-model.trim="todoName" @keyup.enter="addTodo" placeholder="请输入任务">
  <button @click="addTodo">添加</button>

  <ul>
    <li v-for="(todo, index) in todos" :key="todo.id">
      <span>{{ index + 1 }}. {{ todo.name }}</span>
      <button @click="removeTodo(todo.id)">删除</button>
    </li>
  </ul>

  <footer v-show="todos.length">
    <span>还有 {{ todos.length }} 条任务</span>
    <button @click="clearTodos">清空</button>
  </footer>
</div>
<script>
new Vue({
  el: '#app',
  data: {
    todoName: '',
    nextId: 4,
    todos: [
      { id: 1, name: '学习 Vue 基础指令' },
      { id: 2, name: '完成一个小案例' },
      { id: 3, name: '整理学习笔记' }
    ]
  },
  methods: {
    addTodo () {
      if (!this.todoName) return
      this.todos.push({ id: this.nextId++, name: this.todoName })
      this.todoName = ''
    },
    removeTodo (id) {
      this.todos = this.todos.filter(todo => todo.id !== id)
    },
    clearTodos () {
      this.todos = []
    }
  }
})
</script>
```

这个案例串联了 `v-model`、`@keyup.enter`、`v-for`、`:key`、`v-show`、事件方法和响应式数组，是当天内容的综合练习。

## 十四、常见问题与排查顺序

1. **页面显示空白**：检查 Vue 脚本是否成功引入，挂载元素 id 是否与 `el` 一致。
2. **模板变量找不到**：检查属性是否写在 `data`、`computed` 或 `methods` 中，拼写和大小写是否正确。
3. **属性没有生效**：HTML 属性要用 `:属性="表达式"`，不要把插值写进属性值。
4. **事件没有触发**：检查事件名、方法名和方法是否放在 `methods` 中。
5. **列表更新异常**：给每一项提供稳定唯一的 `key`，不要随意使用下标。
6. **HTML 被当成文字显示**：需要解析可信 HTML 时使用 `v-html`，普通文本使用插值或 `v-text`。
7. **数组删除后页面不变**：优先通过响应式数据修改数组，例如 `filter`、`push`、`splice`，不要只操作 DOM。

排查时先看控制台报错，再确认数据是否真的发生变化，最后检查模板指令和绑定关系。

## 十五、本日总结

- Vue 用数据驱动视图，响应式数据变化会触发页面更新。
- `{{ }}` 适合文本插值，只能写表达式。
- `v-text` 输出纯文本，`v-html` 解析 HTML，需要注意安全。
- `v-show` 适合频繁切换，`v-if` 适合条件性创建和销毁。
- `v-on`/`@` 处理事件，复杂逻辑放在 `methods`，用 `this` 访问数据。
- `v-bind`/`:` 绑定属性，`v-for` 渲染列表并配合稳定的 `key`。
- `v-model` 建立表单控件与数据之间的双向同步。

下一步是在组件化的基础上学习生命周期、工程化和组件通信，把这些指令组合成可维护的 Vue 应用。
