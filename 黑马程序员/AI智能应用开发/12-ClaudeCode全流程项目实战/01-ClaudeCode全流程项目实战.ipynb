## 第一章 · VibeCoding 初识

### 1. VibeCoding 的含义与边界

讲义将 Vibe Coding（氛围编程）介绍为 Andrej Karpathy 在 2025 年 2 月提出的一种 AI 辅助编程风格：开发者用自然语言表达想要的效果，让 AI 生成和修改代码，再运行程序检查行为，继续用对话修正。

基本过程是：描述需求 → AI 生成代码 → 运行验证 → 描述问题或新期望 → AI 继续修改。判断结果主要依靠运行效果，对代码内部细节的参与较少。

| 对比维度 | 传统编码 | 讲义中的 VibeCoding |
| --- | --- | --- |
| 代码产出 | 开发者逐行编写 | 用自然语言表达需求，由 AI 生成 |
| 对代码的态度 | 理解每一行及其作用 | 更多依靠 AI 输出与运行效果 |
| 调试方式 | 阅读代码、断点调试 | 描述症状，让 AI 修正 |
| 迭代方式 | 直接修改实现 | 通过对话逐步调整 |

> 原课的重要边界：即使每一行代码都由 LLM 生成，只要开发者全部审查、测试并理解，也不属于这里强调的狭义 VibeCoding。区别在于参与和审查程度。

后续企业项目会引入设计、规范和验收。不能把最初快速试错的方式直接当作复杂项目的全部研发流程。

## 3. Git 与 Node.js 前置环境

### 3.1 Git

Git 负责版本管理，也是本课 Windows 环境中 Claude Code 的依赖之一。安装包可从课程资料取得，官网为 [Git](https://git-scm.com/)。按安装向导完成安装，讲义建议开发工具放在没有中文和空格的路径，例如 `D:\develop\git`。

安装完成后，在新打开的命令行中验证：

```bat
git --version
```

能够输出版本号，说明当前终端能找到 Git。后续 Claude Code 启动时若提示 Git 相关错误，先回到这一步检查安装和环境变量。

### 3.2 Node.js

本课通过 npm 安装 Claude Code，所以先安装 Node.js。官网为 [Node.js](https://nodejs.org/zh-cn)。安装包架构要与电脑匹配：普通 Intel / AMD 的 64 位 Windows 选 `x64`，ARM 设备选 `arm64`。讲义也介绍了旧 `x86` 架构，但不作为普通新电脑的默认选择。

运行 MSI 安装程序，选择安装路径，按向导完成。随后在新终端验证：

```bat
node -v
```

若 `node` 或 `npm` 无法识别，检查安装目录是否进入 `PATH`，以及终端是否在安装后重新打开。

### 3.3 npm 镜像与全局工具目录

讲义使用以下命令配置 npm 下载源与全局安装路径：

```bat
npm config set registry https://registry.npmmirror.com
npm config set prefix "D:\develop\NodeJS"
```

第二条中的路径是课程示例，应替换为自己的目录。它改变后续全局工具的安装位置；若安装后找不到 `claude`，还要检查这个目录是否已加入 `PATH`。这两项是本机工具配置，不是笔记仓库的项目配置。


## 4. 大模型准备与 Claude Code 接入

### 4.1 为什么还要准备模型

Claude Code 是执行编程任务的工具，底层还需要能够理解和生成内容的大模型。讲义对比本地部署和云端 API：本地部署需要硬件并受模型规模影响，云端 API 不需自行部署，但要按供应商规则付费。

本课使用 DeepSeek 云端服务。课程操作顺序为：到 [DeepSeek 官网](https://www.deepseek.com/) 注册账号 → 进入开放平台准备调用额度 → 创建 API Key → 复制并妥善保存。创建后的完整 Key 通常只显示一次，后续配置需要用到它。

### 4.2 安装和验证

讲义在 Windows 管理员 CMD 中执行 npm 全局安装：

```bat
npm install -g @anthropic-ai/claude-code
claude -v
```

看到版本号后，再启动：

```bat
claude
```

安装成功与模型连通是两个检查点。能够显示版本号，尚不代表已经连通模型服务。

### 4.3 课程中的首次启动处理

讲义首次启动遇到如下连接错误：

```text
Unable to connect to Anthropic services
Failed to connect to api.anthropic.com: ERR_BAD_REQUEST
```

课程接下来使用第三方兼容接口，并在当前 Windows 用户目录的 `.claude.json` 中增加以下字段，表示已完成新手引导：

```json
{
  "hasCompletedOnboarding": true
}
```

这是需要合并到现有 JSON 对象中的字段，**不要用上面这个最小示例覆盖已有文件**。用户目录例如 `C:\Users\你的用户名`，不是项目根目录。原课示例中的首次启动时间、迁移记录和 `userID` 属于某次安装状态，无需照抄。

此步骤记录原课处理方式；跳过引导不会修复所有网络、凭证或模型兼容问题。


## 5. DeepSeek 配置与 CC Switch

### 5.1 用户级模型配置

在当前用户目录的 `.claude` 文件夹中找到或创建 `settings.json`，例如 `C:\Users\你的用户名\.claude\settings.json`。课程使用如下配置，Key 改为自己的值：

```json
{
  "env": {
    "ANTHROPIC_BASE_URL": "https://api.deepseek.com/anthropic",
    "ANTHROPIC_AUTH_TOKEN": "<替换为自己的 DeepSeek API Key>",
    "API_TIMEOUT_MS": "3000000",
    "CLAUDE_CODE_DISABLE_NONESSENTIAL_TRAFFIC": "1",
    "ANTHROPIC_MODEL": "deepseek-flash",
    "ANTHROPIC_DEFAULT_SONNET_MODEL": "deepseek-flash",
    "ANTHROPIC_DEFAULT_OPUS_MODEL": "deepseek-flash",
    "ANTHROPIC_DEFAULT_HAIKU_MODEL": "deepseek-flash"
  }
}
```

| 配置项 | 作用 |
| --- | --- |
| `ANTHROPIC_BASE_URL` | 指向供应商的 Anthropic 兼容接口 |
| `ANTHROPIC_AUTH_TOKEN` | 调用该供应商的凭证 |
| `API_TIMEOUT_MS` | 超时时间，单位毫秒；示例为 3,000,000 毫秒，即 50 分钟 |
| `CLAUDE_CODE_DISABLE_NONESSENTIAL_TRAFFIC` | 课程设置为 `1`，禁用非必要流量 |
| `ANTHROPIC_MODEL` | 默认模型 |
| 三个 `ANTHROPIC_DEFAULT_*_MODEL` | 分别将 Sonnet、Opus、Haiku 模型档位映射到供应商模型 |

原课将三个档位都映射为 `deepseek-flash`，所以切换档位仍指向同一模型。模型名按讲义记录；实际接入时需与账号可调用的模型名一致，不要把其他供应商的模型名直接填进来。

重新运行 `claude`，核对工作目录，在信任提示中确认当前目录，检查界面显示的模型并发送简单请求。能收到模型回复，才说明接入完成。API Key 属于个人凭证，不应放入课程示例代码或提交进仓库。

### 5.2 CC Switch 管理供应商

课程使用 [CC Switch 发布页](https://github.com/farion1231/cc-switch/releases)提供的桌面应用管理不同 AI 工具的供应商配置。操作顺序为：

1. 安装并打开 CC Switch，选择要配置的 AI 工具与模型供应商。
2. 填入该供应商的 API Key、访问地址和模型名。
3. 保存后，在供应商列表点击“启动”，使 Claude Code 使用对应配置。
4. 需要其他模型时，先准备对应供应商账号与 Key，再按相同过程添加和切换。

手动编辑配置和 CC Switch 解决的是同一个接入问题；切换之后要重新核对实际生效的模型。


## 6. 初体验：水果消消乐

先用一个范围很小的纯前端游戏体验“描述—生成—运行—修正”。创建 `VibeCoding演示` 目录，在这个目录中启动 `claude`，输入课程提示词：

```text
请帮我开发一个水果消消乐（三消）网页游戏，要求如下：
1. 纯前端实现，所有代码放在一个 HTML 文件里（内嵌 CSS 和 JavaScript），双击就能在浏览器打开玩，不依赖任何外部库和网络；
2. 10×10 棋盘，用 6 种水果（emoji：🍎🍊🍇🍓🍉🍒）表示，开局随机填充；
3. 当棋盘上没有任何可消除的交换时，提示“没有可用步数”并自动重新洗牌；
4. 界面：顶部计分板显示当前得分和剩余步数（初始 30 步），消除时水果有简单的放大消失动画和得分飘字，整体配色鲜艳活泼；
有不清楚的地方，可以直接和我确认。
```

生成后，双击目录中的 HTML 文件，在浏览器检查棋盘大小、初始步数、计分、消除动画与无步可走时的洗牌行为。结果与要求不符时，具体描述现象并继续让 Claude 修改，再重新打开页面验证。

这里的练习产物是一个 HTML 文件，不涉及数据库、接口和部署。它帮助建立反馈循环，后续项目才逐步增加设计与工程约束。


## 7. 权限模式与常用命令

### 7.1 课程中的四种模式

权限模式决定 AI 执行操作时怎样请求确认。讲义给出的使用建议如下，名称与可切换范围以课程界面为准：

| 模式 | 原课建议的使用场景 |
| --- | --- |
| Manual Mode | 不确定操作后果，希望逐步确认 |
| Accept Edits | 日常编码，希望减少编辑确认 |
| Plan Mode | 先分析和规划，再决定是否执行 |
| Auto Mode | 较长任务，已明确目标并信任执行方向；原课不建议用于敏感环境 |

课程通过 `Shift + Tab` 切换权限模式。先识别当前模式，再给任务，不能只看 AI 的文字回复来判断它是否会修改文件。

### 7.2 斜杠命令

以下命令在 Claude Code 交互界面中输入，不是在系统终端直接执行。要带 `/`，否则会按普通提示词处理。

| 命令 / 按键 | 用途 | 课程中的使用场景 |
| --- | --- | --- |
| `/help` | 查看帮助 | 忘记命令时 |
| `/model` | 查看或切换模型 | 检查当前模型；本课三个档位映射相同 |
| `/clear` | 清空当前对话 | 开始全新任务 |
| `/btw` | 临时插问，按讲义描述不计入主任务上下文 | 工作中询问旁支问题 |
| `/context` | 查看上下文各部分占比 | 诊断上下文占用 |
| `/compact` | 压缩对话上下文 | 长对话减少上下文压力 |
| `/init` | 生成项目 `CLAUDE.md` | 进入新项目建立项目说明 |
| `/rewind` | 回退到历史节点 | 撤回不合适的对话或操作节点 |
| 双击 `Esc` | 课程用于停止或回退 | 执行中先停止；空闲时进入回退 |
| `/simplify` | 从质量、效率和复用角度审查并修改代码 | 开发后整理优化 |
| `/resume` | 选择历史会话继续 | 异常终止后恢复工作 |

这些命令的效果不同：`/clear` 清空上下文，`/compact` 保留压缩后的任务信息，`/resume` 重新选择历史会话。


## 第二章 · 实战准备：项目定位与业务架构

### 1. 为什么开发《AI-智多星》

平台面向 IT 求职学员，把学习路线、题库练习与 AI 模拟面试的数据连接起来，解决原课提出的三个问题：刷题与学习路径割裂、背答案缺少真实面试对话、错题和薄弱点缺少持续追踪。

业务内容沿“学科 → 分类 → 题目 / 课程 / 路线”组织。学员先按路线学习或刷题，再通过 AI 面试检验掌握情况；AI 参与出题、追问、评分和报告生成。

### 1.1 两个用户入口

| 入口 | 面向的人 | 主要业务 |
| --- | --- | --- |
| 学员前台 `portal` | 互联网注册学员 | 学习路线、课程入口、题库练习、错题本、AI 面试、面试记录与报告、个人中心 |
| 管理后台 `admin` | 平台运营和管理人员 | 学科、课程库、题库、知识库、学习路线、学员、运营数据、用户角色和操作日志 |



图中还列出 DeepSeek、MinerU、阿里云短信、MinIO 和 B 站视频平台等外部能力。这里只记录项目已有的业务全景，不据此提前实现尚未进入本章的功能。


## 2. 技术架构与服务边界

项目采用前后端分离、双服务端协作。原文所说“四个端”，结合技术架构图理解为两个前端应用与两个服务端；前面的概览表只列了学员前台和管理后台两个用户入口。

| 部分 | 课程技术栈 | 主要职责 |
| --- | --- | --- |
| 学员前台 | Vue 3、Vite、TypeScript、Pinia、Vue Router、Axios、Web Speech API | 学习、练习、面试的展示和交互 |
| 管理后台 | Vue 3、Element Plus、ECharts、md-editor-v3、Sass | 平台业务数据维护与统计 |
| Java 应用服务端 | Java 17、Spring Boot 3.2、Spring MVC、MyBatis-Plus、Redis、JWT、RBAC + AOP 等 | 业务逻辑、权限、业务数据读写 |
| Python 智能体服务端 | Python、FastAPI、Uvicorn、LangGraph、LangChain、Pydantic、httpx、Loguru | RAG、Agent 编排、出题追问、评分、报告与结构化输出 |



### 2.1 通信与数据边界

前端负责交互展示，Java 服务管理业务，Python 服务提供 AI 能力，通过 REST 接口和 SSE 流式推送协同。SSE 是服务端持续向客户端发送事件的机制，用于逐步展示生成过程等场景。

架构图明确指出：Python 智能体服务仅内网可达、不直接连接业务库；业务数据的唯一写入口位于应用服务端。不要看到 AI 服务要使用业务数据，就擅自让它绕过业务接口直接写 MySQL。

### 2.2 各层使用的组件

| 层次 | 课程组件与用途 |
| --- | --- |
| 接入层 | Nginx：静态资源、反向代理、HTTPS、跨域、限流、前端 history 路由回退、SSE 长连接 |
| 数据层 | MySQL 8.0：业务持久化；Redis 7.x：缓存、令牌、会话；Milvus：知识库语义检索；MinIO：文档与附件 |
| 工程与运维 | Knife4j：接口文档；Git：版本管理；Maven / Vite：构建；Docker：容器；Logback / Loguru：日志；TraceId：链路标识 |

原课概览表把 Python 技术重复写为 `LangGraph, LangGraph`，这里以随后技术架构段明确给出的 `LangChain + LangGraph + FastAPI` 为准。

## 5. VibeCoding 与 SDD

SDD 在本课指规格驱动 / 规则驱动的工程化开发：先形成需求、设计、数据库与 API 规范，再让 AI 依照确认后的文档生成代码，并逐条验收。

| 维度 | VibeCoding | SDD |
| --- | --- | --- |
| 起点 | 先表达想法，迅速生成运行结果 | 先明确规范和验收要求 |
| 迭代 | 对话、运行、反复调整 | 对照文档实现与验收，同步更新设计 |
| 优势 | 启动快、试错便宜、适合探索 | 一致性、协作、追溯和维护更有依据 |
| 成本与风险 | 上下文易断裂、结构与质量不稳定 | 前期设计投入、人工校准、文档维护 |
| 适用场景 | 学习、原型、POC、一次性工具 | 企业产品、长期维护、多人协作 |



### 5.1 在项目中混合使用

1. 需求探索、原型验证：用 VibeCoding 快速把想法变成可体验的界面。
2. 规范设计、方案确认：用 SDD 明确业务流程、技术架构和接口契约。
3. 代码生成、快速迭代：在确认的规范内让 AI 实现并及时运行。
4. 测试验收、质量把控：按规范检查功能与代码质量，修复问题。
5. 上线与持续迭代：结合反馈调整实现，持续维护文档和规则。



讲义将文档称作 AI 开发的契约，强调先确认规则、再实现、再验证。文档仍需由人校正；AI 写出文档或测试代码，并不自动代表设计正确或测试充分。


## 第三章 · 全流程实战：开发流程

传统研发和 AI 辅助研发都需要需求、设计、实现、测试与部署。课程对比的重点是 AI 参与深度，以及需求、原型和代码之间的反馈速度。

| 阶段 | 原课强调的变化 | 必须保留的判断 |
| --- | --- | --- |
| 需求分析 | 从前期集中分析转向持续提问、探索、收敛 | 用户、范围、关键业务规则要由人确认 |
| 原型设计 | 从绘制设计稿扩展到直接生成可交互网页 | 原型不等于已具备后端、数据库、权限和完整校验 |
| 项目设计 | AI 辅助模块、架构、数据库、接口的方案评估 | 判断模块边界、关系约束、接口一致性与扩展需求 |
| 编码实现 | AI 生成功能代码，人运行并反馈 | 复杂任务要拆分、明确约束、审查与测试 |
| 测试验收 | 原型和编码阶段都持续验证 | 最后仍需集成测试、验收和必要的安全检查 |
| 部署上线 | AI 辅助配置、脚本、Dockerfile 和流水线 | 环境隔离、迁移备份、权限、监控与回滚仍需设计 |

例如“新增课程”，涉及前端表单、接口、参数校验、持久化和异常处理，不能仅以页面能打开作为完成标准。课程以这个例子说明：任务表达越具体、验证标准越清晰，AI 执行越容易保持一致。


## 2. 需求探索：先把问题问清楚

需求探索要把模糊想法变成可确认的目标、用户、功能和业务规则。直接让 AI “开发一个学习管理系统”，容易遗漏学习记录怎样保存、成绩怎样计算、不同角色可以做什么等问题，导致后期返工。

### 2.1 要问清楚的内容

| 方面 | 要解决的问题 | 原课举例 |
| --- | --- | --- |
| 项目目标 | 为什么做，解决什么问题 | 统一管理课程学习、练习成绩和进度 |
| 用户对象 | 谁使用，各自完成什么任务 | 学员学习，教师维护内容，管理员配置系统 |
| 核心业务 | 操作怎样流转，有哪些分支 | 发布课程 → 学习 → 练习 → 查看成绩 |
| 未知与风险 | 哪些规则未决定或有歧义 | 能否重复学习、提交后能否修改、教师的数据范围 |

探索时优先关注第一版范围、业务流程、权限与状态等规则、待确认事项。不要为了补全清单而任意扩张功能；AI 的建议也不能自动变成已确认需求。

### 2.2 阶段产物：需求基线

需求基线至少包括目标用户与项目目标、角色、核心功能、核心流程、关键规则、本期范围与暂不实现内容、待确认问题、初步验收目标。可以随原型验证逐步细化，不要求第一轮就生成一份巨大的完整规格书。

完成本阶段后，应该能够说明项目服务谁、先做什么、关键操作怎样进行、哪些规则仍需确认，而不是仅得到一串功能名称。


## 2.3 在 CLAUDE.md 中写入需求阶段规则

在新建的项目工作目录中准备 `CLAUDE.md`。下面是原课需求阶段规则的整理，约束的是课程项目开发过程：

```markdown
# 1. 需求阶段规则

适用范围：需求梳理、需求探索、产出需求文档阶段。

1. 有不确定的需求或问题，与我交互确认，等我回答后再继续追问。
2. 一轮一轮推进，直到关键需求清晰、没有阻塞性问题。
3. 每轮结束输出简短的“当前需求快照”：已确认结论、待确认清单、本轮新增结论。
4. 问题必须具体、可回答，尽量给出候选选项和倾向建议。
5. 每个问题简要说明“为什么问”，即影响哪些设计决策。
6. 优先提出最影响系统架构与开发落地的问题。
7. 禁止一次性抛出所有问题。
8. 禁止提出“你想要什么”这类空泛问题。
```

这里让 AI 以产品经理的方式推进对话，同时保留人工决定关键业务规则的环节。每轮快照用于防止多轮探索后丢失已确认信息。


## 2.4 需求探索演示提示词

以下保留原课的目标、功能范围与任务边界。项目名“云枢智学”沿用本段示例。

```text
你现在是一名资深产品经理，擅长企业级软件项目的需求分析与需求挖掘。

【项目背景】
我准备开发一个“学员在线学习、刷题与AI面试”的系统（云枢智学），
主要面向两类用户：互联网上的学员、机构运营人员。

1. 学员注册登录后，可以看到不同方向的学习路线，例如“AI全栈开发”、
   “Agent开发”、“Java开发”；进入路线后看到各阶段课程，
   课程关联 B 站视频，点击跳转到 B 站观看。
2. 学员选择学科下的课程阶段进行刷题，题型支持判断题、单选题、
   多选题、简答题，提交后给出判定结果。
3. 学员可以进行 AI 面试，检测技能掌握情况，结束后给出面试报告。

【初步设想的功能清单】
1. 学员前台（C端）：注册登录、首页展示、课程学习路线（关联B站课程）、
   题库练习、AI面试、学习记录、面试记录、错题本、个人信息管理。
2. 管理后台（B端）：学科管理、B站课程库管理、刷题题库管理、
   面试题库管理、学习路线管理、学员管理、系统用户管理、角色管理、
   操作日志、数据看板统计等。

以上都只是初步想法，大量业务细节尚未确定。

【本轮任务边界】
- 不要直接编写完整需求文档。
- 不要开始写代码，也不要讨论技术选型。
- 不要臆造我没说过的需求，不要替我做决定。
- 核心职责是“把需求问清楚”。

请先完成以下4件事：
1. 分析目标用户、诉求、典型场景与核心业务目标。
2. 梳理已明确的需求，以及模糊、缺失或矛盾的部分。
3. 按业务模块列出需要进一步确认的问题。
4. 优先提出最影响系统架构与开发落地的问题。

现在，请开始第一轮。
```

任务边界与功能清单同样重要。该轮先探索需求，不提前做技术选型或实现代码；后续经人工整理和调整的需求文档，才成为原型设计的输入。


## 3. 产品原型：把需求变成可体验的界面

原型在正式开发前表达页面结构、布局、交互和业务流程。它可以是草图、交互设计稿，也可以是可运行网页。课程侧重让 AI 直接生成 HTML、CSS、JavaScript 原型。

### 3.1 为什么需要原型

需求文字比较抽象。把登录、学习首页、课程详情、学习记录等页面做出来后，产品、业务、前后端和测试可以围绕实际界面确认入口、布局、跳转和操作顺序，在改动成本较低时发现遗漏。

原型还能为后续工作提供输入：前端识别组件与交互，后端识别数据与操作，数据库设计识别实体，测试整理使用场景。但权限、数据一致性、异常处理等技术设计仍需单独完成。

| 维度 | 传统原型 | AI 原型 |
| --- | --- | --- |
| 设计方式 | Axure、Figma 等工具 | 自然语言描述后生成页面 |
| 形态 | 设计稿或交互原型 | 可运行的交互网页 |
| 修改 | 调整稿件和交互配置 | 对话修改实现 |
| 验证 | 查看和模拟操作 | 浏览器直接体验 |
| 衔接开发 | 开发者再实现 | 原型代码作为参考或起点，仍需整理和工程化 |

### 3.2 操作顺序与提示词

先明确初步需求，再生成原型，随后在浏览器体验搜索、筛选、弹窗、页面跳转等行为，发现问题继续修改。三个步骤可以循环。

在项目根目录向 Claude 提交：

```text
请根据 docs 目录下，初步梳理出来并调整优化后的
《云枢智学-需求文档.md》，帮我基于 HTML、CSS、JS 生成这个项目
的可交互的页面原型（只考虑交互设计，不需要考虑服务端逻辑），
生成页面原型放在 prototype 目录下。
```

完成后应能在 `prototype/` 找到页面并实际操作。原型通过体验后，再进入总体设计，不直接把原型当成生产系统。


## 4. 项目设计：五个阶段与交付物

设计阶段把需求与原型转化为能够指导 AI 编码的方案，采用“先明确约束、确认方案，再实施”的思路。按下面的顺序推进，但允许后面的设计暴露问题后回到前面调整。

| 阶段 | 重点 | 交付物 |
| --- | --- | --- |
| 1. 需求与功能模块梳理 | 功能清单、职责、依赖、流程、歧义 | `docs/design/01.功能模块梳理.md` |
| 2. 技术选型与整体架构 | 技术依据、分层、模块依赖、目录 | `docs/design/02.整体架构设计.md` |
| 3. 数据库设计 | 实体、字段、约束、逻辑关系、索引、删除策略 | `docs/design/03.数据库设计.md`、`scripts/sql/schema.sql` |
| 4. API 接口设计 | 路径、方法、入参、出参、错误、权限 | `docs/design/04.API接口设计.md` |
| 5. 整体评审与定稿 | 完整性、一致性、约束、修正与待确认项 | `docs/design/05.设计方案评审结果.md` |

各阶段结束都汇报结论和待确认问题，按课程提示词等待审核后进入下一步。这个控制点避免尚未确认的业务规则被后续设计固化。


## 4.1 需求与功能模块梳理

本阶段先整理“系统要做什么、谁负责什么”，不进行技术选型、表结构和接口设计。

```text
请开始云枢智学项目设计的第一阶段：需求与功能模块梳理。
请先阅读需求文档 docs/云枢智学-需求文档.md 和 prototype 目录下的页面原型。

1. 梳理项目的主要功能。
2. 划分功能模块，明确各模块职责。
3. 梳理模块之间的关系和核心业务流程。
4. 整理尚不明确、需要进一步确认的问题。

将设计结果整理到 docs/design/01.功能模块梳理.md。

注意：
- 以现有需求和原型为依据，不要擅自增加功能。
- 暂时不进行技术选型、数据库设计、API接口设计或代码开发。
- 不明确的问题先询问，不自行决定。
- 完成后汇报主要设计结论和待确认问题，等待审核。
```

验收时对照需求和原型，检查核心功能是否都有模块承接、职责是否重复、跨模块流程是否说清楚。


## 4.2 技术选型与整体架构

本阶段确定技术与代码组织方式，必须承接上一阶段的功能模块。课程已明确的技术栈应继续遵循，不能由 AI 随意换成另一套框架。

```text
请阅读 docs/design/01.功能模块梳理.md，结合现有需求文档、页面原型及资料，
完成云枢智学项目的技术选型与整体架构设计。

1. 技术选型：确定前后端、后端框架、数据库、缓存等方案，说明依据，遵循既有技术约束。
2. 整体架构：明确整体组成、技术分层、各层职责和协作方式。
3. 模块架构：明确各模块在系统中的位置、职责、依赖。
4. 项目目录：规划目录结构，说明主要目录的用途。
5. 关键决策：记录重要技术决策、待确认问题和风险。

保存至 docs/design/02.整体架构设计.md；流程图用 Mermaid 格式。

注意：
- 以第一阶段确认的模块为依据。
- 不展开具体表结构和接口定义，不写业务代码。
- 未确定的方案提出有依据的建议，并与用户交互确认。
- 汇报技术选型、架构、目录和待确认问题，审核后再进入下一阶段。
```

检查重点是模块能否映射到架构、调用边界是否合理、目录是否支撑后续开发，而不是架构图是否复杂。


## 4.3 数据库设计

明确系统要保存的数据、实体关系与约束，再根据业务查询场景规划索引。表之间有逻辑关系，不代表一定要创建物理外键；本课评审规则明确禁止创建物理外键。

```text
请阅读：
- docs/design/01.功能模块梳理.md
- docs/design/02.整体架构设计.md

结合项目需求，完成云枢智学的数据库设计：
1. 梳理核心实体及关系，设计表结构。
2. 明确字段、数据类型、主键、约束、索引。
3. 梳理表的逻辑关联与数据删除策略。
4. 检查是否满足业务需求和数据库规范。

设计文档：docs/design/03.数据库设计.md
建表脚本：scripts/sql/schema.sql

注意：
- 必须阅读并遵循 docs/rules/database-rules.md。
- 业务规则不明确时直接交互确认，不自行决定。
- 汇报主要设计和待确认问题，审核后再进入下一阶段。
```

完成后应同时得到可评审的表结构说明与 SQL 脚本，两者必须一致。本阶段形成脚本，不代表已经获准在数据库执行。


## 4.4 API 接口设计

接口定义把功能、业务数据和调用方连接起来。每个接口要说清用途、方法、路径、输入、输出、校验、业务异常与权限。

```text
请阅读：
- docs/design/01.功能模块梳理.md
- docs/design/02.整体架构设计.md
- docs/design/03.数据库设计.md

完成云枢智学 API 接口设计：
1. 梳理各模块接口清单、用途和所属模块。
2. 明确请求路径、方法、参数、响应数据和业务规则。
3. 统一响应结构、参数校验、异常处理和权限校验。
4. 检查接口与功能需求、系统架构、数据库设计的一致性。
5. 有疑问的接口设计直接交互确认，不擅自假设关键规则。

保存至 docs/design/04.API接口设计.md。

注意：
- 遵循 CLAUDE.md 中的 API 规则，并读取其引用的 docs/rules/api-rules.md。
- 字段与数据库保持一致，不随意新增或修改数据库字段。
- 本阶段只设计接口，不编写业务代码、不实现接口。
- 完成完整性与一致性检查，汇总问题并等待审核。
```

例如接口要返回的字段，应能在已确认的数据模型或明确的计算规则中找到依据；不能为了让页面方便展示而悄悄改表。


## 4.5 整体评审与定稿

整体评审用于检查设计之间是否一致，避免各文档分别看起来完整，却无法协同实现。

```text
请全面阅读并评审：
- docs/design/01.功能模块梳理.md
- docs/design/02.整体架构设计.md
- docs/design/03.数据库设计.md
- docs/design/04.API接口设计.md
- docs/rules/database-rules.md
- docs/rules/api-rules.md

结合需求文档、原型和已有规范，完成：
1. 功能完整性：核心需求是否都有模块、数据和API支撑。
2. 架构合理性：技术选型、职责、服务边界和调用关系是否合理。
3. 数据库与接口一致性：参数与响应是否匹配表结构，是否缺失或冗余。
4. 规范与约束检查：技术栈、数据库、API等约束是否遵守，特别是禁止创建物理外键。
5. 问题修正：可明确判断的问题直接修正文档与SQL，保证同步；
   涉及关键业务规则或需用户决策的问题先列出，等待确认。

要求：
- 功能、架构、数据库与API相互一致。
- docs/design/03.数据库设计.md 与 scripts/sql/schema.sql 一致。
- 汇总发现的问题、已修改内容、待确认事项。
- 保存到 docs/design/05.设计方案评审结果.md。

边界：只评审和完善设计，不写业务代码、不实现API、不执行SQL。
不得擅自更换技术栈，不得为消除文档差异而修改已确认的需求。
完成后汇报，等待审核确认。
```

设计定稿后，需求和关键决策才成为后续功能规格与实现的依据。后续设计变化，也要同步检查受影响的文档。

进入设计阶段，在前面需求规则后追加：

## 6. Docker 环境准备

项目依赖的数据库、缓存等服务适合用容器统一准备。课程先安装 Docker，再让 Claude 编写 Docker Compose 文件准备项目依赖。

### 6.1 方式一：AI 辅助安装

原课提示词为：

```text
请帮我直接在当前系统中安装一下 docker。
```

讲义演示的过程包括系统授权、安装 WSL、下载 Docker Desktop、完成安装并打开。这里记录这一做法，不在当前电脑执行。

### 6.2 方式二：手动安装

1. 在控制面板的“程序 → 启用或关闭 Windows 功能”中，启用“适用于 Linux 的 Windows 子系统”和“虚拟机平台”，然后重启。
2. 安装课程提供的 `docker环境准备/wsl.3.0.4.0.x64.msi`。
3. 在命令行执行 `wsl --version`，看到版本号后继续。
4. 运行 `Docker Desktop Installer.exe`，按安装向导完成。
5. 打开 Docker Desktop，按原课完成首次启动界面并进入主页面。

```bat
wsl --version
```

能够看到 WSL 版本和 Docker Desktop 主界面，是本段的检查点。WSL、虚拟机功能和 Docker Desktop 有先后关系，不能只下载 Docker 安装包就认为环境已经完成。

### 6.3 Docker 数据位置与镜像仓库

若系统盘空间不足，可在 Docker Desktop 设置中调整镜像数据的存放位置。原课还在 Docker Engine 配置中设置镜像加速地址：

```json
{
  "builder": {
    "gc": {
      "defaultKeepStorage": "20GB",
      "enabled": true
    }
  },
  "experimental": false,
  "features": {
    "buildkit": true
  },
  "registry-mirrors": [
    "https://mirrors.tuna.tsinghua.edu.cn",
    "https://docker.m.daocloud.io",
    "https://docker.1ms.run",
    "https://mirror.baidubce.com",
    "https://hub-mirror.c.163.com",
    "https://dockerproxy.net"
  ]
}
```

上面保留原课配置，**镜像地址不等于持续可用性承诺**，也不能把 PyPI / npm 镜像与 Docker Registry 镜像混为一谈。复现时根据实际可用的 Docker Registry 服务配置，并保留本机已有必要设置。

## 8. Java 后端开发规范

原课在 `CLAUDE.md` 中继续追加下面的开发约定。技术栈为 Java、Spring Boot 3、MyBatis-Plus、MySQL、Redis、MinIO。

```markdown
# 5. Java后端项目开发规范

1. 开发语言：Java。
2. 后端技术栈：Spring Boot 3 + MyBatis-Plus + MySQL + Redis + MinIO。
3. 类名使用 UpperCamelCase，例如 QuestionService。
4. 方法、变量使用 lowerCamelCase，例如 getQuestionDetail。
5. 常量使用大写下划线，例如 DEFAULT_PAGE_SIZE。
6. 命名表达业务意图，避免含义模糊的缩写。
7. 请求参数进行必要的格式、范围和业务校验。
8. 优先用 MyBatis-Plus；复杂查询按需使用自定义 SQL。
9. 避免循环中无必要的大量单条查询，防止 N+1 查询。
10. 使用统一异常处理，不在各 Controller 重复捕获异常。
11. 业务异常携带清晰、可识别的信息或错误码。
12. 不向前端返回 SQL 详情、堆栈、内部路径、密钥等敏感信息。
13. 注释重点说明业务意图、复杂逻辑和关键设计原因，避免冗余。
```

N+1 查询是先查一批数据，再为每条记录分别追加查询，查询次数随结果数量增长。该规范要求在 AI 生成功能代码时同时考虑校验、数据库访问和异常契约，避免只完成“接口能返回”。


## 9. Vue 3 前端开发规范

```markdown
# 6. Vue3前端项目开发规范

1. 技术栈为 Vue 3 生态，管理端使用 Element Plus。
2. 优先采用 Composition API 和 <script setup>。
3. 组件、方法、变量命名清晰表达业务意图，避免模糊缩写。
4. 页面结构和交互符合已确认的原型与业务设计，不擅改核心功能和布局。
5. 风格统一、布局清晰、层次分明，兼顾常用屏幕尺寸。
6. 表单有必要的参数校验、错误提示、提交状态和操作反馈。
7. 接口调用统一封装和管理，遵守 API 规范，避免页面重复请求逻辑。
8. 路由守卫负责前端导航控制，真正的认证和接口权限由后端执行。
9. 合理处理加载中、请求失败、数据为空等状态。
10. 前端不暴露 API Key、服务端凭证等敏感信息。
11. 注释说明业务意图、复杂逻辑和关键设计原因，避免冗余。
```

前端校验用于改善交互，后端校验负责保证业务约束；隐藏按钮和路由守卫不能代替服务端权限校验。功能页面还应能表达加载、成功、失败和空数据等实际状态。
