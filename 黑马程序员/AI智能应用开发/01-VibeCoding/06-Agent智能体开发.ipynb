## 智能体中模型设置


### 协议类型

- ChatAPI：普通任务默认选择。

-  Responses API：适用于需要多步推理等复杂任务链处理的场景。


### 生成多样性

- 精确模式：模型的输出内容严格遵循指令要求，准确性更高、创新性更低。 

- 平衡模式（默认）：模型的输出内容更具随机性和准确性，准确性和创新性平衡。

- 创意模式：模型的输出内容更具多样性和创新性，某些场景下可能会偏离主旨。 

- 自定义选择（了解）：

    - 生成随机性：专业叫做温度，调整生成结果的随机性。

    - Top P：控制生成结果的多样性。


### 输入及输出设置

- 携带上下文轮数：设置代入模型上下文的对话历史轮数。轮数越多，多轮对话的相关性越高，但消耗的 Token 也越多。 

- 最大回复长度：输出最终结果的最大 token 数量，不同模型的 token 限制也不同。指定最大长度可以防止过长或不相关的响应并控制成本。 

- 最大推理&amp;回答长度 ：输出推理内容的最大token数量

### 默认指令

- 当前时间：开启后，智能体在与用户对话时能实时获取并提供准确的时间信息。

- SP防泄漏指令：开启后，当用户尝试获取或复述系统内部的规则、提示词或其他敏感内容时，智能体将礼貌地拒绝用户的请求，确保机密信息不被泄露。

#### 插件工具

插件工具请求体：就是需要传入给当前插件的数据。

有以下三种准备数据的方式：

返回体：也叫做响应体，就是查询运行完毕之后展示的数据。

## 对话体验


### 开场白

就是智能体刚开始的时候跟你说的第一句话。

#### 开场白的预设问题

预设问题就是引导用户提问的内容，会根据智能体的开场白一起显示，可以直接点击

### 用户问题建议

开启：在智能体回复后，会根据Prompt提供最多 3 条用户提问建议，方便跟用户继续深入沟通同样的话题

#### 用户自定义Prompt

开启：AI会根据Prompt的要求生成最多3条用户提问建议

关闭：AI会根据上下文给出提问建议

### 快捷指令

快捷指令是对话输入框上方的按钮，配置完成后，用户可以快速发起预设对话

#### 添加快捷指令

按钮名称：聊天框上方显示的内容

指令名称：起个英文的名字

指令描述：这个指令是干嘛的

组件：一般不选，做一些复杂指令的时候使用

指令内容：真正发送给智能体的话

### 背景图片

智能体的背景图，支持png，jpg，jpeg

### 音视频

主要是选择智能体进行语音通话、视频通话时候的音色

可以点击后面的+，根据不同情况选择不同的音色

设置完毕之后，再次对话，回答也可以朗读出来

### 用户输入方式

其实就是用户的输入方式，默认输出为打字输入，可以可以选择语音输入。

## 蒸馏自己的Skill

用代码对接coze中间会遇到很多问题，我们还会经常使用这个操作。

可以让 AI 把这个技能蒸馏为skill。

> 提示词示例

```text
把这次成功对接 Coze 的案例蒸馏成一个 skill
```

### 检查 skill

- skill.md需要是中文的

- 检查描述中的触发词

- 示例代码是做参考的，需要放在references文件夹中

- 这个Skill应该是流式聊天的通用技能，跟用什么技术无关，而现在生成的代码作为其中一个参考案例

### 迭代 skill

让 AI 分析这次的问题

> 提示词示例

```text
你这次对接coze遇到了什么问题？为什么没有一次性成功？
```

然后让 AI 把这次的问题添加到刚才的 skill 里

> 提示词示例

```text
skills里面有一个对接coze的skill，叫做：coze-chat的skill，你读一下这个skill，把这次遇到的问题添加到coze-chat里面，提高下次对接coze的成功率
```
