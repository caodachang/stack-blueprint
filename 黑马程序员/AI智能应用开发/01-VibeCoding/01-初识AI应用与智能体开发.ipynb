# 初识AI应用与智能体开发

## 1 AI应用开发的介绍


### 1.1 什么是AI应用？

AI应用：带有AI功能的智能软件。

> 利用目前AI大模型的能力，去开发智能软件（传统模块 + 智能模块）

其他称呼：智能软件/AI全栈/AI应用/大模型应用/Agent开发/智能体开发… 

### 1.2 生活中的AI应用

目前生活中的AI应用可以分为两种：

- 传统项目的智能化升级

- 全新的智能软件

> 补充一句：AI虚拟试衣功能目前在国外非常流行，短时间之内在国内基本无法流行起来，不是因为国内技术不行，而是因为国内快速且非常便宜的物流市场，还有用户的消费习惯。

其他业务场景：

音视频软件：即梦

外卖软件：智能点餐系统

物流软件：智能仓储管理系统

医疗软件：智能问诊辅助系统 / AI私人医生

小说软件：AI续写 / AI小说推荐

### 推理大模型：

传统大语言模型基础上，强化了推理、逻辑分析和决策能力的模型。

- 上下文理解能力

- 复杂关系推理能力

- 链式思维：在回答问题之前，推理大模型会生成一系列中间推理步骤，逐步推导出最终答案
