# 提示词工程


## 提示词与提示词工程

提示词（Prompt）是提交给模型、用于说明任务和提供信息的输入。

提示词工程（Prompt Engineering）是设计提示词、观察结果、测试修改、保存有效版本的过程。

## 上下文：模型回答时能看到哪些信息

**上下文（Context）是本次回答时模型能够使用的输入信息。**除了刚发出的消息，还可能有应用设定的规则、被带入的历史对话、上传文件中提取的文字和工具返回结果。

## 第一部分：一条提示词，分成四块写

身份 → 指令 → 例子 → 资料。

### 四块各自解决什么问题

| 部分 | 要回答的问题 | 缺失时常见的结果 |
| --- | --- | --- |
| 身份 | 用什么角度、面向什么读者表达 | 解释难度或语气不合适 |
| 指令 | 具体做什么，遵循什么要求 | 回答了相近的问题，却没交付需要的东西 |
| 例子 | 合格结果具体长什么样 | 标签、字段和表达风格不稳定 |
| 资料 | 回答依据是什么 | 使用通用知识，忽略任务中的事实 |

## 第一块：身份（可选）

**1. 写什么：**

这个 AI 该以什么身份、什么语气跟你说话。

**2. 什么时候写：**

当你希望它有特定口吻或立场时。

**3. 举例：**

❌ 你是一位世界顶级的、经验极其丰富的、享誉全国的语文特级教师……

✅ 用高中语文老师的口吻帮我批改作文。

**4. 要点：**

一句话就够。堆砌"顶级"、"资深"、"享誉全国"没有任何用处——那是形容词，不是标准。

## 第二块：指令（必写）

**1. 写什么：**

要它做什么、不要它做什么。

**2. 怎么写：**

分两列写：

【要做的】

- 列出 5 条建议，每条不超过 30 字

- 每条后面说明理由

- 用中文回答

【不要做的】

- 不要写总结段

- 不要用感叹号

- 不要问我"还需要什么吗"

**3. 举例：**

❌ 写得专业一点，别太啰嗦。

✅ 用词正式，不出现"我觉得"、"大概"这类词；全文不超过 400 字。

**4. 要点:**

具体、可验收的标准，比空泛的形容词管用。

"不要写鸡汤结尾"比"写得真诚一点"有效十倍。不是因为它是否定句，而是因为它具体、能验收。

OpenAI 原话：

对目标的描述要非常具体，给出可衡量的成功标准。

有明确约束就写死："预算低于 500 元""字数在 300 到 500 之间"。

## 第三块：例子（格式特殊时写）

**1. 写什么：**

给它一个或者两个输出格式的例子。

OpenAI 叫 few-shot learning（少样本学习）：

给例子时，尽量覆盖多样的输入情况。

**2. 什么时候用:**

| 情况 | 要不要给例子 |
| --- | --- |
| 常见的问答、总结、翻译 | 不用，先直接写，不行再改 |
| 输出格式很特别（非普通文本） | 要，给 1~3 个 |
| 分类、打标签类任务 | 要，每种情况给一个 |

## 第四块：资料（有材料就写）

你要它依据的材料——文章、数据、笔记。

## 如何删掉"AI 味"

去除大模型生成内容的AI味:

```text
不要使用这些套话:"值得注意的是""深入探讨""赋能""助力""总而言之""简而言之""综上所述"。
不要用"这不是关于X,而是关于Y""这是X,不是Y"这类对比句式。
不要在结尾写总结段或升华。
不要声明"本次不涉及…""以下内容保持不变""我将分三类说明"。
不要自造复合词,不要用"在某种程度上"这类模糊限定词,不要用"由此可见""接下来我们看"这类过渡语。
直接说你要做什么,把关系讲清楚。
```
