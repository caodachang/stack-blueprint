# 工作流

## 工作流的组成

- 节点：流程里面的每一个步骤也叫做节点。

- 开始节点：工作流里面的第一个节点，只负责接收数据，不做任何处理。

- 结束节点：工作流里面的最后一个节点，只负责输出数据，不做任何处理。

- 输入数据：当前节点接收到的数据，该数据是由前面的节点传递而来的。

- 输出数据：当前节点的计算结果，会把该结果输出交给后面的节点。

## 创建一个工作流

进入资源库 --- 点击右上角的"+资源" --- 点击工作流准备开始创建

- 工作流名称：给工作流起个名字

> 注：工作流的名字只能是字母，数字和下划线，并且以字母开头，长度不能超过30

- 工作流描述：需要告诉大模型什么时候调用工作流。 在一个智能体中添加多个工作流时， 大模型可以通过工作流描述，自动去调用工作流。在后续的项目中，我们将使用这种方式实现一个智能体根据业务场景不同调用多个工作流。

> 注：描述非常重要，如果乱写，大模型无法调用。

工作流名称：Travel_Plan_Assistant

工作流描述：当用户询问旅游规划、制定旅行攻略、查询目的地路线或安排行程时使用此工作流。它能够根据用户的出行天数、预算和兴趣偏好，自动整合景点、交通与住宿信息7，生成详细的每日行程安排表及出行建议。

- 其他工作流

我们添加两个插件：

#### 设置节点输入数据

只用设置输入（不需要视觉理解）

- 变量名input：表示当前大模型节点的输入数据

- 变量值：str表示字符串文本，该数据为开始节点传递过来的数据，选择开始节点的input变量

#### 设置模型提示词

给当前大模型节点添加提示词，提示词分为两类：

- 系统提示词：就是大模型的人设与回复逻辑

- 用户提示词：用户在本轮对话中的输入，用于给模型下达最新的指令或问题。

>系统提示词示例

```text
你是一个旅游规划助手，能够根据用户输入的内容规划行程，并根据互联网搜索结果给出可执行的方案。需要注意

1. 你的回复必须言简意赅，不要有太多的长篇大论
2. 请从每天的行程安排、天气和应对措施、主要景点、特色食物、民族特色等多个维度按天安排行程
3. 最后整理一个表格，按照吃、住、行等多个维度，计算2-4个人的人均支出
```

用户提示词:此时写为：{{input}}，表示引用当前节点input变量中记录的数据。

> 注：只要打一个{，剩余内容会自动关联出来

选择输出格式为JSON

### 调试工作流

- 点击试运行，并给开始节点的所有输入变量赋值后，再点击右侧下方的试运行按钮，即可完成调试。

输入数据如下：

> 帮我制定一个2026年10月1日到2026年10月7日，武汉7天旅游攻略

- 运行过程中，每个节点的执行结果都可以展开看到，方便我们进行调试。 也会展示整体的执行时间，以及token的消耗数。

点击试运行左边的调试按钮可以看到更详细的信息，点击步骤后面的圆环按钮可以看到这个步骤的运行结果，可以让 AI 帮助分析这一大串 JSON

## 其他常用节点

### 选择器节点

该节点是一个 if-else 节点，用于设计工作流内的分支流程。

### 意图识别节点

意图识别节点对用户意图进行归类，给每一个意图设定自己的分支

> 注：每一个分支上节点的输入数据都要进行修改
