# 10 · 独立实战


## 项目：

学生就业帮助系统


## 功能模块：


### 登录

利用用户名、密码进行登录


### 注册

用户名，密码，身份证号码，手机号码，邮箱


### 忘记密码

校验身份证号码，如果正确就修改


### 信息弹框

如果用户信息不全，每次登录在屏幕中央都有弹框提醒：请进入用户信息页面录入数据


### 用户信息

登录成功之后调整，点击用户头像，进入用户信息界面

展示用户注册的所有信息，用户名不能修改，其余都可以修改，提交修改后保存到数据库


### 退出登录

登录成功之后调整，点击用户头像，可以退出登录，回到登录页面


### 简历评估

查看历史信息

根据上传的简历进行评估

简历评估对接coze

展示简历评估结果

简历评估结果持久化保存


### 录音分析

查看历史信息

根据上传的录音进行分析

录音分析对接coze

展示录音分析结果

录音分析结果持久化保存


### 面试题生成

查看历史信息

根据上传的简历生成面试题

面试题生成对接coze

展示生成之后的结果

面试题持久结果持久化保存


### 模拟面试（直接对接Coze智能体，不需要搭建工作流）

根据生成的面试题，和简历进行模拟面试

聊天记录不需要保存

右上角有清空聊天记录的功能

在coze当中要做什么事情：

1. 创建智能体

2. 系统提示词

3. 关联插件（联网搜索）

4. 关联知识库（AI面试题、修改之后的简历）
