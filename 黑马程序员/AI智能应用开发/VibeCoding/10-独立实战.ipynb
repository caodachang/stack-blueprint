# 10 · 独立实战


来源：[飞书讲义《第10章：独立实战》](https://heuqqdmbyk.feishu.cn/docx/FCp6dHsKioAJcdxHy63cQu0cnLd) · 整理日期：2026-09-25。

本章是“学生就业帮助系统”的独立实战需求，承接第 09 章的三个工作流。以下完整保留原讲义的功能要求及本地运行说明；原文没有配图、代码或详细搭建步骤，仅在必要处加入简短补充与勘误。


## 项目：

学生就业帮助系统


## 功能模块：


### 登录

利用用户名、密码进行登录


### 注册

用户名，密码，身份证号码，手机号码，邮箱


### 忘记密码

校验身份证号码，如果正确就修改


> **补充：**身份证号码不是秘密凭据，不能仅凭号码匹配就允许修改密码。实际应用应通过已验证的手机、邮箱等完成身份验证，再使用有时限、只能使用一次的重置凭据修改密码。


### 信息弹框

如果用户信息不全，每次登录在屏幕中央都有弹框提醒：请进入用户信息页面录入数据


### 用户信息

登录成功之后调整，点击用户头像，进入用户信息界面

展示用户注册的所有信息，用户名不能修改，其余都可以修改，提交修改后保存到数据库


> **补充：**“展示所有信息”不包括显示密码原文或密码哈希；数据库应保存安全的密码哈希。手机号、邮箱、身份证号等按需脱敏展示，敏感信息变更需重新验证身份。用户名不可修改等规则应由后端同时检查。


### 退出登录

登录成功之后调整，点击用户头像，可以退出登录，回到登录页面


> **补充：**退出登录除了返回登录页，还要清理本地登录状态，并让服务端会话或可撤销的登录凭据失效，避免仍能用旧会话读取历史记录。


### 简历评估

查看历史信息

根据上传的简历进行评估

简历评估对接coze

展示简历评估结果

简历评估结果持久化保存


### 录音分析

查看历史信息

根据上传的录音进行分析

录音分析对接coze

展示录音分析结果

录音分析结果持久化保存


### 面试题生成

查看历史信息

根据上传的简历生成面试题

面试题生成对接coze

展示生成之后的结果

面试题持久结果持久化保存


> **补充：**简历评估、录音分析、面试题生成可分别对接第 09 章对应工作流。三类历史记录都应关联当前登录用户，后端检查记录归属；刷新或重新登录后应能读回保存结果，不能只保存在页面内存中。


### 模拟面试（直接对接Coze智能体，不需要搭建工作流）

根据生成的面试题，和简历进行模拟面试

聊天记录不需要保存

右上角有清空聊天记录的功能

在coze当中要做什么事情：

1. 创建智能体

2. 系统提示词

3. 关联插件（联网搜索）

4. 关联知识库（AI面试题、修改之后的简历）


> **补充：**“聊天记录不需要保存”可理解为本项目不做聊天历史持久化，不代表 Coze 侧一定不保留会话或日志。清空后如果要求模型也忘记之前的对话，应按接口能力重置会话或新建会话。用户简历应按用户隔离，不能混入所有人共享的知识库。

注意：

1. 要把开始节点的入参input告诉

服务器：

其实就是一台电脑，一旦作为服务器之后，别人就可以通过网址，直接访问到你电脑上的软件了。

此时服务器就要开启内网穿透。

内网穿透一旦开启之后：

别人可以访问我的电脑

我们现在做的项目：我的电脑要访问coze的服务器

提示词：

我的电脑现在不需要作为服务器，我是把（简历，录音）上传到coze的服务器当中解析我的（简历，录音），此时不需要开启内网穿透，我只要自己的项目在本机上能运行就可以了。


> **补充：**原文“要把开始节点的入参input告诉”未写完整。对接时应把工作流的输入名称、类型和必填项明确提供给编写调用代码的 AI 或开发者，并按实际接口要求上传文件、传递文件标识或 URL。Coze 访问凭据应放在后端配置中。

> **勘误：**电脑作为服务器不必然需要内网穿透。仅供本机使用时可以监听本地地址；让外部访问才需要合适的网络入口。当前项目主动请求 Coze 并上传文件，通常无需内网穿透；若传给 Coze 的只是本机路径或 localhost 链接，远程服务无法直接读取，应先按平台支持的方式上传。
