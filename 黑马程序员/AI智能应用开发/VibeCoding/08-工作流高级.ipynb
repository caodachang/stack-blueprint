## 知识库节点

### 新建知识库

在资源库新建知识库

- 火山知识库： 适用于企业用户，以及大规模的查询场景。

- 扣子知识库： 适用于轻量检索，低代码，轻量级应用的场景

文档解析策略分两种：

- 精准解析：支持从文档中提取图片元素、扫描件（OCR）、表格元素；支持设置过滤策略，以文档页的粒度过滤掉当前文档中不需要导入的内容。精准解析需要耗时更长的时间。

- 快速解析：不支持从文档中提取图像、表格等元素，适用于纯文本。

### 添加知识库节点

回到工作流，添加节点选择知识库检索

配置三个节点

## 数据库

### 数据库和知识库的区别

- 知识库：一般只做查询处理，存放大模型不知道的数据。（专业领域的数据、公司内部的数据）

- 数据库：可以增删改查。真实的开发中，用来保存业务数据的。（程序在运行的时候产生的数据，注册）

### Coze中的数据库

在Coze当中，我们可以在智能体中直接使用数据库，也可以在工作流中使用数据库，方式都是一样的。

#### 创建数据库

在资源库创建一个数据库

- 选择单用户模式

- 选择隔离渠道

- 导入数据，确定格式正确

#### 工作流中使用数据库

- 添加意图识别节点，判断增删改查

- 添加大模型节点处理用户输入

- 选择对应的数据库节点

- 添加大模型节点处理输出

- 连接输出节点

## 循环节点

循环节点有两部分

- 上面的节点：叫做循环控制节点，负责控制循环的次数，接收数据，输出数据

- 下面的节点：叫做循环体节点，负责设置每一次循环中的具体业务

循环体中可以添加其他节点
