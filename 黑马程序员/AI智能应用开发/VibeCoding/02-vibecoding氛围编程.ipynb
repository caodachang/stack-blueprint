## 4. claudecode的启动方式

> 聊天记录在 .claude\projects 下

方式一：claude：常规启动，新建一个对话

方式二：claude --continue / claude -c：继续上一次对话启动

方式三：claude --resume/ claude -r：选择某一次对话

方式四：claude --dangerously-skip-permissions：绕过安全检查启动，可组合 -c -r

方式五：claude -n / claude –-name：新建会话，给这次会话起别名。例如：claude -n "用户登录接口开发"

## 5. claudecode的权限控制

claudecode因为权限问题，很多操作需要我们自己点击yes，下面有三种方式可以让claudecode拥有更多的权限，让我们少点yes

方式一：启动时直接绕过安全检查

方式二：全自动/半自动方式

方式三：配置文件精细控制


### 跳过检测方式

启动方式：claude --dangerously-skip-permissions


### 全自动/半自动方式

为了在危险性和自动之间寻求一个平衡。

claudecode中，又引入了多种操作模式，可以利用快捷键进行切换

快捷键：Shift + Tab （ Shift + 回车：输入换行 ）

| 可切换的模式 | 说明 |
| --- | --- |
| plan mode on | 计划模式，第五章详细学习 |
| manual mode on（纯手动） | 所有操作需确认  |
| accept edits on（半自动） | 自动接受文件编辑 |
| auto mode on（全自动） | 自动执行低风险命令和文件编辑 |

工作原理：

claudecode 里面内置了AI 分类器，会进行安全判断。

- 低风险操作（如读文件、编辑已知代码、部分命令）：分类器判定安全，自动放行。

- 高风险操作（如 rm -rf、修改系统文件、执行未经验证的网络请求、部分命令）：分类器会判定危险，直接拒绝执行，会出现大段红色报错，此时可以继续等待 claudecode 尝试其他方法，或回退到 "manual mode on" 手动模式（手动可以yes）。

### 配置文件方式

这是最精确的控制权限的操作，我们可以把禁止的和允许的写到配置文档当中。

- 允许的操作allow：不需要点击yes，直接放行

- 禁止的操作deny：直接停止

配置方式如下：

- 打开Claude Code 的全局配置文件，位于 ~/.claude/settings.json

```jsonc
{
  "env": {
    "ANTHROPIC_BASE_URL": "大模型的的BASE_URL",
    "ANTHROPIC_AUTH_TOKEN": "自己的APIKEey",
    "API_TIMEOUT_MS": "3000000",
    "CLAUDE_CODE_DISABLE_NONESSENTIAL_TRAFFIC": "1",
    "ANTHROPIC_MODEL": "大模型的名字",
    "ANTHROPIC_DEFAULT_SONNET_MODEL": "大模型的名字",
    "ANTHROPIC_DEFAULT_OPUS_MODEL": "大模型的名字",
    "ANTHROPIC_DEFAULT_HAIKU_MODEL": "大模型的名字"
  }, // 注意这里要加个逗号，跟模型配置加以区分
  "permissions": {  // 给claudecode添加的权限
   "allow": [ // 允许的权限
     "Read",  // 读取文件中的内容      
     "Write", // 写入文件      
     "Bash(npm *)", // npm命令  
     "Bash(git *)", // git命令   
     "Bash(node *)" // node命令  
   ],
   "ask": [...], // 可以删除，未配置默认ask
   "deny": [ // 禁止的权限
     "Bash(rm -rf *)"  // 全局删除
   ]
  }
}
```

配置完毕之后，重启claudecode，可以再次询问：

```text
你现在被允许直接操作的权限有哪些？禁止操作的权限有哪些？
```

## 6. claudecode的常见命令


### 基础命令：

| 命令 | 作用 | 使用场景 |
| --- | --- | --- |
| /help | 显示帮助信息 | 忘记命令时查看 |
| /model | 查看/切换当前模型（高/中/低档） | 我们设置的模型都是一样的，暂时没用 |
| /clear | 完全清空当前对话 | 开始全新任务，创建新的聊天记录文件 |
| /btw | 该命令下的对话不会记入上下文，适合临时提问 | 主任务进行中想问个无关问题 |


### 常见命令：

| 命令 | 作用 | 使用场景 |
| --- | --- | --- |
| /context | 详细查看上下文中每个模块占比 | 防止上下文过多导致模型幻觉 |
| /compact | 压缩当前对话的上下文 | 对话太长，AI开始“遗忘”早期内容时 |
| /init | 自动生成项目的 CLAUDE.md | 进入新项目后的第一件事 |
| /rewind | 回退到某个节点 | 说错几句话，不污染上下文 |
| 双击esc | 停止/回退 | 如果claudecode正在执行，停止当前操作<br><br>如果claudecode没有执行，等同于/rewind |
| /simplify | 从"代码质量"、"运行效率"、"可复用性"三个角度去审核并修改项目代码 | 开发完毕之后，项目优化时使用 |
| /resume | 对出当前对话，选择其他对话 | 异常终止 |


## 7. claude.md文件

claudecode里面有一个执行规则，在执行的时候必须要遵守里面的规则，他就是claude.md文件。

### 文件内容

我们会把开发时的注意点，AI要遵守的规则，都写在claude.md中。

比如：

```text
核心行为准则
1. 所有回答都需要使用简体中文。
2. 上下文接近80%左右的时候，提醒我。
3. 遇到配置文件settings.json中deny绝对禁止的操作，立即停止并提醒我，不要尝试其他方式执行。
4. 如果要修改settings.json的内容，修改之前务必告诉我，允许之后才能修改。
5. 每次回答要精准，不绕弯子，拿不准的提前问我，不编造。

工作流程
1. 涉及 3 个以上文件改动时，先输出中文实施方案，等我确认后再编码。
2. 生成代码时，文件名、变量名、函数名等命名要简洁明了，见名知意。
3. 遇到报错时，先用通俗的中文解释错误含义，再给修复方案。
```

每次claudecode跟大模型交互的时候，都会把claude.md中的内容拼接到提示词中。


### 文件位置

他可以放在任意的目录下，但是一般我们只会在三个层级进行配置：

全局配置、项目根目录和模块子目录

- 用户全局配置：

    位于：C:\Users\你的用户名\\.claude\CLAUDE.md：他是全局的配置，自己开发的所有项目都需要遵守

- 项目根目录：

    位于：项目文件夹\CLAUDE.md：他是本项目的配置，本项目中需要遵守的规则，团队共享。

- 模块子目录：

    位于：项目文件夹\模块文件夹\CLAUDE.md：他是单独模块的配置，当前模块需要遵守的规则。


## 8. Memory记忆


### Memory是自动维护的

地址在：C:\Users\用户名\\.claude\projects\项目名\memory当中

记忆内容：

- user：用户的身份、用户的偏好

- feedback：agent该如何配合你，纠错经验

- project：项目相关的内容（进度、阶段结论、重要决策等）

- reference：外部资源位置（xxx文档在xxx位置）

### 记忆结构

```text
memory/
├─ MEMORY.md          简短索引：有哪些记忆、何时需要读
├─ user.md            用户偏好
├─ project.md         项目背景与重要决定
└─ troubleshooting.md 已验证的排错经验
```

子文件名不固定

## 9. Skill 结构化的提示词

路径：~/.claude/skills/skill文件夹名

Skill 就是一个有.md文件的文件夹。.md文件包含了以下内容：

- name：skill的名字，如果手动调用就是通过名字调用的。

- description：skill的描述，这个描述非常重要，claudecode就是根据描述自动调用当前的Skill，里面包含了很多触发词。

- 内容：详细的操作手册

- 条款 --- 有没有都行

### Skill的执行原理

- 把所有skill的name和description加载到上下文

- 由大模型判断，完成当前的任务需要哪个skill

- 读取skill全文

- 按照skill的提示进行操作

#### Skill文件夹的组成

完整的skill里面可以包含了以下内容：

- Skill.md：主要的核心流程

- references：参考文档

- scripts：利用代码使用一些工具

- assets：资源/图片

# ChatGPT 桌面版

## 权限控制

**沙箱**是工具执行时的技术边界，例如允许写入哪些目录、能否访问网络。

**审批**是在操作超出既有权限时请求授权。

### 项目规则文件 `AGENTS.md`

建议在项目根目录创建 `AGENTS.md`，内容可以这样写：

```markdown
# 项目开发规则

## 技术栈
- 前端：Vite + React
- 后端：Python + FastAPI
- 包管理：前端 npm，后端 uv

## 工作流程
1. 先阅读相关文件，再提出最小修改方案。
2. 修改后运行与改动相关的检查。
3. 报告文件路径、测试命令和结果。

## 代码约定
- 使用简体中文说明，但代码标识符使用英文。
- 不把密钥写入代码、日志或提交。
- 不随意引入新依赖。

## 验收
- API 返回结构与前端调用保持一致。
- 错误路径有清晰的用户提示。
- 不能通过测试的改动不能标记为完成。
```

`AGENTS.md` 给编码助手提供的项目指导。

`README.md` 是项目说明文档

## 用文件保存项目记忆

推荐在项目中维护：

```text
docs/
├── project.md          项目目标、用户和当前阶段
├── decisions.md        已确认的技术决策及原因
├── progress.md         已完成、失败和下一步
└── troubleshooting.md  已验证的错误原因与解决办法
```

开始新任务时可以这样要求：

```text
先阅读 README.md、AGENTS.md 和 docs/ 下与当前任务相关的文件。
把它们作为项目事实来源；如果文件之间冲突，列出冲突并暂停决定。
完成任务后更新 docs/progress.md，只记录已验证的结果，不把猜测写成事实。
```
