## 1. 企业筛选简历的过程

**课件第2、5～9页：** 简历“石沉大海”“已读不回”可能与市场、岗位供需和简历本身有关。不能仅凭没有回复，就判断自己的某个技术点必然不合格。

课件给出的大致链路是：HR筛选、部门负责人筛选、预约面试。

| 阶段 | 课件列出的关注项 | 简历准备时可以做什么 |
|---|---|---|
| HR初筛 | 学历、院校、经验、年龄、跳槽频率等 | 把教育、工作时间、求职状态写清楚，避免信息互相冲突 |
| 部门负责人筛选 | 当前项目技术栈、业务领域是否匹配 | 让岗位相关的技能和项目职责容易找到 |
| 额外匹配项 | 高可用与高并发、公有云开发、团队管理经验 | 仅列真实经历，准备解释自己负责的范围与结果 |

课件用Boss直聘和智联招聘的招聘方后台说明：招聘者能按关键词、经验、学历、城市等字段检索简历。下面是**课程中的历史截图**，用于理解检索机制。



**补充建议：** 在线简历和附件简历保持一致；技能名称使用正确、常见的拼写，例如Spring Boot、MyBatis、Redis。不要为了出现搜索词，把没有实践过的技术堆进去。

课件描述了部分招聘筛选现象，不能据此认为所有企业采用相同规则，也不能把筛选条件当作对个人能力的完整评价。真实经历与岗位匹配程度应分别分析。


## 2. 岗位要求与简历匹配

**课件第8～10、24页：** 部门负责人还会看业务领域，例如电商、银行、物流。岗位要求中的Java基础、框架、数据库、业务建模、微服务经验，应该能在简历中找到对应事实。

**补充方法：** 选择一个目标岗位，把招聘要求拆成四列，不必为每个要求写一大段说明。

| 岗位要求 | 自己的真实证据 | 简历中的位置 | 面试前待补内容 |
|---|---|---|---|
| SQL优化 | 某次查询的执行计划、修改与对比记录 | 技能条目及项目职责 | 为什么选该索引、其他写入成本 |
| 消息队列 | 某模块的发送、消费、失败重试代码 | 项目模块描述 | 重复消费、积压、确认时机 |
| 权限认证 | 登录与角色授权实现 | 技能及权限模块 | 认证和授权区别、越权验证 |
| 业务建模 | 实际订单状态与表关系 | 项目介绍 | 非法状态、重复请求如何处理 |

这张表是待自己填写的练习，不表示已经完成过上述工作。可以分别标记“只了解”“练习实现过”“在真实工作中负责过”，据此确定表述强度。

如果招聘要求强调某种行业经验，而自己只有课程项目经历，就说明项目是学习实践，再用业务链路展示理解。没有必要编造公司、工龄、上线规模来补齐要求。

每次调整简历可优先改变相关内容的顺序、删减无关条目、补充清楚的职责。工作时间、学历、实际贡献等事实应保持一致。


## 3. 简历整体结构

**课件第11～12、15页：** 完整简历可包含基本信息、教育背景、求职意向、工作经历、职业技能、项目经历、个人优势和个人荣誉。课件强调，在真实前提下优化表达，重点体现难点和亮点。

| 部分 | 应写清楚的内容 | 检查重点 |
|---|---|---|
| 基本信息 | 姓名、可靠的联系方式等必要信息 | 联系方式准确，避免加入与招聘无关的敏感信息 |
| 教育背景 | 院校、专业、学历、起止时间 | 学历与学习形式据实填写 |
| 求职意向 | 目标岗位、目标城市、到岗状态 | 区分现居地、求职城市与到岗时间 |
| 工作经历 | 公司、岗位、日期、主要职责 | 公司与项目时间关系可解释 |
| 职业技能 | 与岗位相关且可以展开的能力 | 表述与实践深度相符 |
| 项目经历 | 背景、技术、职责、难点、结果 | 明确团队成果和个人贡献 |
| 个人优势 | 能支持岗位工作的事实 | 避免只有“认真”“学习能力强”等空话 |
| 荣誉或成果 | 真实证书、比赛、文章、开源贡献 | 名称、时间、链接与贡献可核对 |

**配套批注建议：** 技术突出时可把技能放在项目之前；项目更能证明能力时可把项目提前。最近、最相关的项目通常先写。

**整理建议：** 应届生没有正式工作经历时，可以如实写实习、课程实践、个人项目或开源贡献，结构不必强行凑齐八部分。无需复制范例中的年龄、性别、籍贯等所有字段。

阅读者应能快速回答：你应聘什么岗位？相关能力是什么？你做过哪些具体工作？这些事实有什么证据？排版和措辞服务于这些问题。


## 4. 职业技能的写法

**课件第10、13、15页的核心准则：写在简历上的必须能聊。** 技能应放在容易看到的位置，并且引导面试官询问自己真正理解的部分。

课件公式为“职业技能＝必要技术＋其他技术”；总结页将后者写成“第三方技术”。结合所列JVM、源码、设计能力等内容，应理解为基础之外的扩展能力，并不限于第三方库。

课件以Spring Boot、SSM、Redis、数据库作为示例基础，再根据经验深度加入微服务、ES、MQ、源码、高并发、JVM、技术选型和设计能力。课件举例：1～2年扩展2～3项，3～5年扩展3～4项，5年以上扩展5项以上。**这是课程的准备参考，不是招聘统一门槛；年限不能直接证明掌握程度。**

### 将笼统表述具体化

| 笼统写法 | 课件举出的更具体写法 | 需要准备的证据或追问 |
|---|---|---|
| 熟悉MySQL，有SQL优化经验 | 理解事务、锁、索引，能根据EXPLAIN分析SQL | 哪条SQL、为什么慢、如何比较优化效果 |
| 熟悉Java并发 | 说明线程池、AQS、CountDownLatch、Future等具体范围 | 在哪里使用、线程如何协作、失败如何结束 |
| 熟练使用Redis | 说明数据结构、缓存问题、持久化或高可用中的实际范围 | 穿透/击穿/雪崩、失效、一致性如何处理 |

课件Redis示例使用了“精通”“自主搭建过高可用集群”等强表述。配套批注则建议慎用“精通”。**整理建议：先按实际事实写，较强词语需要较强证据。**

例如，完成过某练习后才可写：

> 在个人学习项目中使用Redis缓存课程查询结果，实现更新后缓存失效，并通过请求记录检查缓存命中情况。正在补充分布式部署和故障切换实践。

技能条目可按“技术＋实际应用范围＋能解释的机制”组织。避免连续列十几个框架名称，却无法说明一个请求怎样执行。只阅读过源码时，不把它表述成生产调优经历。


## 5. 项目经历的写法

**课件第14～15页：** 项目数量以真实经历为准，较久且不相关的项目可以简写。描述应体现业务深度或技术深度，能说明模块设计、开发、迭代和结果。

**配套批注给出的结构：** 项目定位、架构和组成模块、技术选型、项目职责、难点或亮点、业绩。简介建议控制在几行，将展开空间留给个人职责。

### 可填写的项目卡片

~~~text
项目名称与性质：课程练习 / 个人项目 / 实习项目 / 工作项目
项目周期：真实的开始、结束或迭代时间
业务定位：谁使用？解决什么问题？
项目组成：主要用户端、管理端和模块
技术栈：实际版本与关键依赖
自己的职责：负责哪些模块、设计或代码？
关键难点：触发条件、影响、约束是什么？
处理方案：关键流程、为什么选择这个方案？
结果与证据：日志、用例、测试条件、指标或功能验收
边界与改进：尚未解决的问题、失败情况和代价
~~~

“参与”“负责”“主导”需要区分：参与表示承担部分工作；负责应能说清模块范围；主导设计还需要解释需求澄清、方案比较、决策和落地。0到1是首次建设，1到2是既有能力的迭代，两者都可以展示价值。

课件建议尽可能展示QPS、数据量等指标。**补充建议：指标同时记录环境、数据集、并发、测试时长、错误率和延迟。** 没有测量时可以写功能完成范围或验证方法，不填写推测出来的性能提升。

一条职责尽量同时包含业务动作和技术选择，例如“实现订单超时关闭，处理支付与关闭并发时的状态校验”。只有“使用Redis、RabbitMQ”通常不足以解释贡献。


## 6. 黑马头条项目示例与追问

**课件第8、14、23页：** 使用黑马头条展示项目描述，涉及文章发布与审核、延时发布、实时排行榜、冷热数据迁移。该示例用于说明如何把简历条目引到业务和技术问题。

| 简历中的主题 | 可准备的业务问题 | 可准备的技术问题（补充） |
|---|---|---|
| 文章发布审核 | 审核哪些内容？哪些状态可以发布？审核失败怎么办？ | 自动审核与人工审核怎样衔接？重复回调如何处理？ |
| 延时发布 | 用户能否修改或取消？到时间但审核未通过怎么办？ | 实际使用Redis、MQ还是任务调度？延迟、重试与幂等如何保证？ |
| 实时排行榜 | 排行榜依据什么行为？如何去重和计算热度？ | 统计窗口、乱序、状态更新、持久化与查询如何连接？ |
| 冷热数据迁移 | 什么条件算冷数据？历史文章怎样查询？ | 迁移校验、并发修改、重试、回滚和路由怎样处理？ |

课件包含“迁移超过1000万冷数据”等范例数字，以及HBase等选型。**这些是课程示例中的陈述，不是用户已经完成的工作，也不是本篇实际测量的结果。** 配套PDF和课件的技术列表并不完全相同，应按自己实践过的版本回答。

课件中的`kafka-starem`应规范写为 **Kafka Streams**，`Hbase`规范写为 **HBase**。整理简历时要核对技术名称，避免拼写错误。

可以将真实的学习经历写成：

> 基于黑马头条课程学习文章发布流程，梳理文章状态、自动审核与延时发布的关系。独立实践及验证过的部分为____，待补充的部分为____。

填写完成后，挑一个模块沿着输入、状态、数据变化、失败处理讲清楚。不要直接复制范例里的公司背景、项目日期和“主导设计”。


## 7. 配套参考简历的结构

**资料PDF共3页：** 第1页安排基本信息、教育、个人技能和工作经历；第2～3页安排项目经历与荣誉。项目按较近经历在前的方式组织，每个项目包含日期、技术列表、业务简介和负责模块。

这里提炼表达结构，不复制范例的个人信息、工作年限、任职时间和荣誉。

| 范例项目 | 范例呈现的业务内容 | 可以学习的描述方法 |
|---|---|---|
| 帝可得 | 智能售货机、设备管理、工单、订单 | 从业务背景落到工单创建、库存和出货等责任范围 |

范例中的技术列表较长，不能仅据此判断所有技术都由简历作者个人负责。自己的简历应把“项目整体采用的技术”与“本人负责模块用到的技术”区分清楚。

配套简历和话术文档对零售项目使用了不同名称，本篇保留各自来源名称，未核验它们的版本或业务关系。


## 8. 练手项目的来源与学习步骤

**课件第16～20页：** 来源包括Gitee、GitHub、黑马程序员B站项目课程、其他项目课程和相关技术文章。可以搜索感兴趣的技术或业务点，例如Spring Boot、权限。



以上为课件历史页面。Star数量可以作为发现项目的线索；项目是否适合练习，还需要检查代码、文档、依赖、维护情况和运行要求。

课件给出的学习过程：

1. 找到感兴趣的技术点或业务点。
2. 在本地运行项目。
3. 用断点跟踪关键代码，梳理业务。
4. 尝试自己独立完成相关功能。
5. 选择一两个点深入挖掘，参考其他项目和资料。

**补充建议：** 每一步留下一份小产出：运行说明、请求链路、表关系、独立实现、问题记录。重新实现某个模块时说明参考来源，区分开源原有功能与自己的改动。

课件推荐研究能复用到多个业务的通用模块：**权限认证、文件服务、日志、搜索、秒杀、支付、缓存、SaaS、订单、优惠券、CMS和报表。** 并不要求应届生一次实现全部模块。

选择模块时先确认：自己是否理解业务？能否完成最小功能？是否有值得深入的问题？环境与时间是否允许？项目规模大并不自动带来更好的面试回答。


## 9. 权限认证模块如何深入

**课件第19页采用三个层次：功能实现、常见问题、系统设计。** 用这三个层次可以判断自己对模块掌握到哪里。

### 9.1 功能实现

课件列出用户名密码登录、二维码登录、手机短信登录，以及用户、角色、权限管理与分配。技术支撑包括RBAC模型、Spring Security或Apache Shiro。

**补充解释：** 认证回答“你是谁”，授权回答“你能做什么”。RBAC通过用户、角色、权限关系表达访问范围。

~~~text
用户 —— 用户角色关系 —— 角色 —— 角色权限关系 —— 权限
请求进入 → 确认身份 → 加载或校验权限 → 执行业务或拒绝访问
~~~

前端隐藏按钮只能影响展示，服务端还要校验接口权限和具体数据的访问权限。

### 9.2 常见问题

课件提示token刷新、加密解密和XSS防护。可以进一步准备：token过期、刷新并发、退出后失效、角色变更何时生效、重复发送短信、验证码重放、二维码状态与超时。

**修正说明：** 密码一般保存经过适当密码散列算法处理的结果，如bcrypt或Argon2，并通过比较验证。不要设计成能解密还原用户密码。JWT签名验证也应与加密解密分开理解。

XSS防护需要结合输出编码、安全的渲染方式及适当策略，不能认为引入某个认证框架就自动解决。身份认证、授权、XSS、CSRF分别有自己的威胁和防护条件。

### 9.3 系统设计

课件关注可扩展性、高可用性、通用性。准备回答角色是否能扩展、权限变更如何传播、多实例如何共享会话、失效如何处理、审计如何记录。SaaS场景还需要解释租户隔离。

**待实践自测：** 普通用户访问管理员接口是否被拒绝？能否通过更换资源ID读取他人数据？过期、撤销、刷新后的token各有什么行为？这些是练习题，本篇未执行相关实验。


## 10. 通用模块的练习问题

**本表是根据课件模块清单补充的练习题。** 每个模块从“能做功能”继续走到“能解释失败和设计”。

| 模块 | 最小功能 | 常见问题 | 设计追问 |
|---|---|---|---|
| 文件服务 | 上传、下载、文件元数据 | 重复上传、中断、类型和大小校验 | 大文件、存储权限、分片合并、访问控制 |
| 日志 | 请求日志与错误日志 | 缺失、敏感内容、定位不到请求 | 关联ID、采集、检索、保存期限 |
| 搜索 | 索引、查询、分页 | 更新延迟、重复、索引与库不一致 | 重建、增量同步、相关性与深分页 |
| 秒杀 | 资格校验、扣库存、订单 | 超卖、重复请求、热点 | 原子性、限流、排队、失败补偿 |
| 支付 | 发起支付、回调、查单 | 重复回调、超时、伪造回调 | 验签、幂等、对账、状态一致性 |
| 缓存 | 查询缓存与失效 | 穿透、击穿、雪崩、旧数据 | TTL、热点、容量与一致性策略 |
| SaaS | 租户及租户资源 | 跨租户越权、配置差异 | 隔离方式、扩展、运维和成本 |
| 订单 | 创建与状态变化 | 重试、取消与支付竞态 | 状态机、幂等、事务和补偿 |
| 优惠券 | 领取、核销、退券 | 超发、重复核销、计算歧义 | 规则表达、锁定释放、退款拆分 |
| CMS | 内容编辑、审核、发布 | 重复发布、历史版本、权限 | 审核流、发布版本、检索和静态化 |
| 报表 | 统计、展示、导出 | 口径不一致、大数据导出慢 | 时间窗口、分页、异步与可追溯性 |

不用每行都准备。挑一个自己已经有基础的模块，补齐功能说明、异常用例、设计取舍和验证记录。权限认证可直接使用上一节的三层问题。

回答时明确实现边界，例如“只做了单实例演示”“只验证了模拟支付回调”“尚未做压力测试”。继续实践后再更新笔记和简历，不能把待做事项写成已取得的结果。


## 11. 面试轮次与考察重点

**课件第21～22页：** 不同企业的面试轮次不同，可能只有一轮技术面，也可能有多轮技术面和HR面。课件列出资深开发或技术经理、业务部门经理、HR等角色。

| 角色 | 可以重点准备的内容 |
|---|---|
| 开发人员或技术负责人 | 基础、原理、代码、项目实现、排错与技术取舍 |
| 业务或部门负责人 | 业务理解、负责范围、交付结果、沟通与思考方式 |
| HR | 求职动机、经历一致性、协作、岗位条件和到岗安排 |

课件对角色的技术能力、出场轮次和薪资决定权做了经验性概括。**修正说明：实际组织安排会不同，不能假定某个职务技术较弱，也不能假定某一人拥有最终薪资决定权。** 按问题内容认真回答即可。

面试前向招聘联系人确认形式、预计时长、轮次、是否有笔试或现场编码。远程面试还应检查声音、网络和共享屏幕需要展示的内容。

课件第23页的过程主线是：自我介绍、项目介绍、具体模块、业务深度、技术实现。项目细节很容易成为后续追问入口，需要提前准备并在面试后复盘。


## 12. 自我介绍

**课件第23页：** 采用总分结构，先给整体经历，再分阶段说明。配套话术参考3～5分钟，并强调内容与简历一致、逻辑清楚、能概括且聚焦。

**补充建议：** 根据面试官要求控制长度，也准备一个较短版本。没有三段工作经历就不必凑三段。姓名、岗位、相关背景、一个代表项目或优势、当前求职方向通常已经足够。

### 应届生或转行学习者模板

> 您好，我叫____，应聘Java开发岗位。我的背景是____。我主要学习了____，在____课程/个人项目中，已经独立完成____，并通过____验证。这个过程中，我遇到过____，通过____解决。希望在____方向继续积累实践。

### 有真实工作经历的模板

> 您好，我叫____，应聘____岗位。我的主要经历可以分为____几个阶段：____。最近在____项目中负责____，重点处理了____，结果通过____确认。我比较熟悉____，希望在与这些能力相关的岗位继续发展。

模板里的每个空都应填写自己的事实。不要使用配套范例中的姓名、公司、工龄、项目日期或“高级工程师”等身份。

自我介绍结束后，可以自然停下，让面试官选择追问方向。用自己能够解释的具体模块作为入口，比一次列出所有学过的框架更容易展开有效交流。


## 13. 项目介绍与技术追问

**课件第23页：** 从项目介绍进入具体模块，再展开业务深度和技术实现。例如写了“主导文章发布审核、延时发布”，就需要回答审核什么、用什么技术、延时发布如何实现。

**补充讲述顺序：** 先说明服务对象和业务目标，再讲整体模块和本人职责，最后选择一个功能展开。先帮助对方建立背景，再讲细节。

~~~text
业务目标：____
输入、输出和状态：____
我负责的模块与改动：____
关键实现链路：____
当时的约束和方案比较：____
遇到的问题、定位依据：____
结果、验证条件及局限：____
~~~

对每个重点模块至少准备下列问题：

- 正常请求经过哪些接口、服务、表和组件？
- 哪一步可能失败？如何重试？重复请求有什么结果？
- 为什么用这个方案？有没有更简单的方案？
- 事务边界在哪里？并发更新如何处理？
- 如何发现问题？日志、指标或测试能说明什么？
- 哪些是自己负责，哪些来自框架、课程或同事？

**不知道时的表达建议：** 说明已经掌握的范围，再提出需要确认的条件。例如“我实现过单实例的过期检查，多实例下的撤销机制还没有实践；我会先确认状态存储和传播方式，再验证一致性”。

资料中的项目话术可以帮助发现追问，但不能证明方案已经正确，也不能替代自己的代码和验证记录。技术细节以实践版本为准。


## 14. 人事面试与薪资沟通

**配套“面试参考话术”列出五类问题：** 为什么离职、如何谈薪资、如何看待加班、工作中最有意义的事、为什么转行。下面提炼准备方法，不复制它的个人经历范例。

| 问题 | 准备重点（整理建议） |
|---|---|
| 为什么离职 | 真实原因、目前状态、接下来想承担的工作；不虚构家庭事件或公司变动 |
| 如何谈薪资 | 岗位职责、市场信息、本人能力和总体待遇；明确计算口径 |
| 如何看待加班 | 如实说明能配合的安排，交流工作节奏、频率、补偿或调休 |
| 最有意义的事 | 一个真实任务，说明困难、本人行动、结果和学到什么 |
| 为什么转行 | 真实动机、已付出的学习与实践、对岗位的理解 |

### 薪资比较的口径

原资料建议预期涨幅30%，还用月薪、福利、公积金相加比较公司。**修正说明：30%不是固定规律，不能直接用相加后的数字当作可支配收入。**

比较时分别核对：税前固定月薪、承诺的发薪月数、奖金是否保证、试用期待遇、福利、社保公积金缴纳基数与比例、工作安排。公积金和现金薪资的性质不同，个人缴纳和单位缴纳也不能重复计入。

~~~text
固定税前年现金 = 固定税前月薪 × 明确承诺的发薪月数
奖金、补贴、福利、公积金：分别列出条件和口径
到手收入：还要结合税费与个人缴纳部分计算
~~~

本篇不替用户设定具体工资，也不把不确定年终奖或口头加薪承诺当作已经保证的收入。

**Offer沟通：** 配套资料强调取得书面Offer。可以确认岗位、薪资口径、试用期、地点、到岗时间和附带条件，再进行比较。谈话中的意向和承诺应尽量通过正式书面信息核对。

对加班、家庭和离职原因的回答以事实和自己的边界为准；无需照搬“单身、可以全身心投入”等话术。用具体协作经历支持表达，比固定套话更有说服力。


## 15. 面试准备与复盘

**课件第23～25页：** 提前准备，面试后复盘，找出不足并针对性补强。课程用较高目标的准备要求鼓励学习，不能把薪资例子或口号理解成录用保证。

### 面试前

1. 逐条检查简历，确认每个技能和职责都能解释。
2. 练习自我介绍和一个代表模块的展开回答。
3. 根据岗位要求复习薄弱项，准备可展示的项目材料。
4. 检查面试形式、时间、地点或远程设备。
5. 准备自己关心的职责、团队、技术和待遇问题。

### 面试后

**补充复盘卡片：**

~~~text
岗位及面试轮次：
问题原文或准确概括：
面试官给出的条件：
当时的回答：
卡点：知识不懂 / 条件没问清 / 项目证据不足 / 表达不清
核对后的结论及来源：
需要补做的最小实验：
下一次闭卷回答：
~~~

复盘先记下真实问答，再区分事实和猜测。对方没有反馈时，不把自己猜测的淘汰原因记成确定结论。也不要仅凭一次面试就得出“这类企业必考某题”的结论。

每次选择少量具体问题补强，例如把“不了解线程池”改成“解释不清线程数、队列与拒绝策略的关系”。能够闭卷讲清并用例子验证，才说明补强有进展。

## 17. 配套材料中的表达修正

配套批注有些建议与课件“真实前提”的要求不一致，或带有旧版本背景。本篇采用下面的处理方式。

| 原资料中的建议或表述 | 本篇采用的解释 |
|---|---|
| 通过虚岁、现居城市、时间调整迎合招聘 | 年龄、地址、学历形式、工作与项目日期据实表达；可单独注明求职地点和迁居意愿 |
| 按工龄固定配公司数、项目数和周期 | 数量与时间由真实经历决定，主要展示相关且能深入说明的项目 |
| 复制范例技能、公司职责或业绩 | 只参考结构和提问方向，用自己的实践与贡献填写 |
| 列出许多“加分技术” | 依据岗位与实际掌握程度筛选，不把所有词语写成技能 |
| 课程示例中的“精通”、高并发和千万级数据 | 保留为范例背景，自己的强表述与数字需要证据 |
| 范例项目性能改善、使用率提高等结果 | 不能移植为个人业绩，需记录测量或统计依据 |
| 邮箱类型、个人字段或话术作为统一规则 | 联系方式可靠、内容适合岗位即可，不把历史偏好视为通行标准 |

批注中“不能相互copy”“不要夸大”“拼写要正确”“将技术与功能结合描述”的建议，和课件原则相符，可以保留。

部分材料出现Struts2、Hibernate、EasyUI、Ribbon、Hystrix等技术名称。它们说明示例的技术背景，是否写入简历取决于岗位和实际经历，不能把这份旧列表当作当前必须掌握的全部组件。

参考荣誉只学习“名称和时间清楚”的写法，自己的证书名称按官方名称和证书填写，不沿用范例中的称谓。


## 18. 配套技术题的复习入口

“面试参考话术”还收录15类技术题。它们是准备篇的补充题，不在本篇替代后续专题。下表完整列出题目，并指出需要核对的地方。

| 配套题目 | 复习时应补齐的边界或修正 | 后续入口 |
|---|---|---|
| ThreadLocal原理 | 值存在线程的ThreadLocalMap中；线程隔离不自动消除共享对象的数据竞争；线程池中用完及时remove | 08并发 |
| 同步锁、死锁、乐观锁、悲观锁 | 死锁包含循环等待等条件；阻塞不一定是死锁；CAS和锁方案都有竞争成本 | 08并发 |
| 反射 | Class对象获取方式各有用途；访问控制和模块边界仍会限制反射，不能任意访问所有方法 | Java基础补充 |
| HashMap、Hashtable、ConcurrentHashMap | 线程安全、null规则、版本与锁粒度分别解释，不保证性能“提升N倍” | 07集合 |
| 线程池分类 | 同时解释队列容量、最大线程数、拒绝策略与关闭过程，无界配置需要结合任务负载分析 | 08并发 |
| JDK8堆内存结构 | 元空间位于本地内存，与堆中的对象区分；分代与分配行为还要结合收集器 | 09JVM |
| HashMap底层原理 | 树化不仅看链表长度，还看容量条件；查找和扩容要按具体JDK实现分析 | 07集合 |
| MySQL存储引擎 | InnoDB事务、锁、索引与MyISAM比较要标注版本，不套用所有历史文件格式 | 03数据库 |
| 避免索引失效 | 结合索引定义、选择率、谓词与执行计划，避免把范围、OR或函数使用说成绝对规律 | 03数据库 |
| SQL调优 | 子查询不一定更慢；排序、GROUP BY与索引效果须结合版本和计划 | 03数据库 |
| Spring事务传播 | REQUIRED等7种传播行为，重点是已有事务时的行为、挂起与新事务，NESTED还依赖事务管理器支持 | 04框架 |
| RocketMQ消息不丢失 | 发送确认、持久化/复制、消费成功与重试逐段分析；消费确认方式与客户端版本有关 | 06消息中间件 |
| Redis持久化 | RDB/AOF、配置、恢复和可能丢失的时间窗口；不能等同于零丢失保证 | 02Redis |
| Spring Cloud组件 | 资料列Eureka、Ribbon、Hystrix、Feign、Gateway、Config，属于历史组合，需按项目版本核对 | 05微服务 |
| Spring Boot等常用注解 | @SpringBootApplication的组合、映射与Bean职责；@MapperScan属于MyBatis集成，不是Boot本身提供的通用注解 | 04框架 |

**额外核对点：** 原话术把Class.forName称为“最安全/性能最好”，没有可普遍成立的依据。它可能触发类初始化，使用时要结合目的与环境，不能机械套用性能结论。

MySQL的GROUP BY不应无条件被描述为自动排序，排序要求应显式表达。索引是否使用及使用到哪一部分，用当前环境的EXPLAIN和实际数据验证。

这里的修正是整理者补充，未运行上述源码、SQL或消息系统实验。后续笔记依据对应专题的实际课件继续展开。


## 19. 配套项目话术的技术修正

配套文档还给出资讯、自动零售、电商项目的介绍与追问。可以学习“业务选择如何引出技术方案”，但下列陈述需要补齐或修正。

### 19.1 数据规模与分库分表

原资料把1000 TPS与“每月3000万数据”直接关联。**修正：** 如果持续每秒写入1000条且运行30天，记录数为：

~~~text
1000 × 60 × 60 × 24 × 30 = 2,592,000,000条
~~~

峰值TPS、平均TPS、每请求写入条数、活跃时长和保留期限分别影响规模。不能把峰值持续乘满整月，也不能在没有负载假设时断言“单表5000万”“必须20张表”。需要结合行宽、索引、磁盘、查询模式和维护成本评估。

资料以用户行为、关注表和粉丝表说明按查询主体设计数据布局。可以追问两份关系如何同步、幂等、修复以及分页。异步双写MySQL和MongoDB也不会自动保证一致性，需要明确失败与重放机制。

### 19.2 零售工单与消息分发

资料讲到区域、智能派单、补货和维修工单，用Redis ZSet表示工作量，用任务调度分片处理批量任务，用MQTT/EMQ传递设备状态。准备时继续问：并发派单是否分配到同一人？每日初始化与任务执行是否冲突？重复设备报文会不会产生重复工单？

资料写支付成功后发送出货指令，设备上报结果，订单和设备服务各自处理。**修正：** MQTT共享订阅的同一共享组通常会把一条消息分配给组内一个订阅者。如果两个服务都必须收到结果，应使用分别的订阅组或其他能满足广播需求的方案，不能默认同一共享组中的两个服务都会收到。

该文档没有提供本篇可验证的完整代码与配置，以上作为设计问题保留，不能据此宣称已经实现可靠出货。

### 19.3 令牌桶与身份识别

原资料对令牌桶写了“处理结束后删除”“到最低限额时不删除”。**修正：** 经典令牌桶按速率补充令牌，桶满不再增加，请求到达时消耗令牌。令牌不足时按设计拒绝或等待。请求完成后归还额度更接近并发许可控制，不能与按速率限流混为一谈。

原资料把JWT身份识别描述为“公钥解密成功”。**修正：** 对非对称签名JWT应验证签名，并检查有效期、签发方、接收方及其他约束；公钥验签与解密不同。有有效身份也仍需要授权校验。

token放在URL中可能进入日志或历史记录，应结合协议选择合适的传递方式。网关向下游传递身份时，还要防止外部请求伪造身份头，并明确服务之间的信任边界。


## 20. 闭卷自测

1. **HR和部门负责人分别看什么？** 前者常先看招聘字段，后者看技术和业务匹配；具体流程因公司而异。
2. **没有面试回复说明什么？** 可能与市场、岗位、匹配和简历有关，不能从单个现象确定原因。
3. **技能条目的底线是什么？** 真实，并且能够解释实际范围、机制和使用场景。
4. **课件“必要技术＋其他技术”如何使用？** 根据岗位和能力组织基础与扩展，年限和技术数量只是参考。
5. **“熟悉MySQL”如何写得具体？** 写出实际使用和能解释的事务、索引、锁或执行计划分析，提供真实证据。
6. **项目介绍要包含什么？** 业务定位、组成、技术、个人职责、难点、处理和结果。
7. **项目指标为什么需要条件？** 相同QPS在不同硬件、数据、并发和错误率下含义不同。
8. **有课程笔记就能写工作经历吗？** 不能。课程学习、个人实践和正式工作应据实区分。
9. **练手项目从哪里找？** 开源平台、项目课程和相关资料，以自己能运行、理解和深入实践为判断依据。
10. **怎样判断模块已经吃透？** 说明功能、常见问题、系统设计，并通过自己的实现和验证支持。
11. **认证与授权有什么区别？** 前者确认身份，后者检查可执行操作与可访问数据。
12. **自我介绍为什么用总分结构？** 先提供背景，再展开相关经历，便于理解和选择追问。
13. **项目追问怎么准备？** 沿输入、状态、数据链路、失败、取舍和证据展开。
14. **薪资必须涨30%吗？** 不是，依据岗位、能力、市场和整体条件协商。
15. **如何复盘？** 记录真实问题和回答，识别卡点，核对来源，补实验，再闭卷回答。
16. **JWT公钥能验证什么？** 在非对称签名场景验证签名，还需检查声明和授权，不能只说解密成功就有权限。
17. **令牌桶的令牌何时消耗？** 请求获取准入时消耗，按速率补充，与任务完成后归还许可不同。
18. **同一MQTT共享订阅组是否保证每个服务都收消息？** 不保证，组内一般分担消费，需要按广播或分摊需求设计订阅。
