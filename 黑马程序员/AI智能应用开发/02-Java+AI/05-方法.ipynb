
## <strong>方法的注意事项</strong>

### <strong>方法的注意事项</strong>

- 方法不能嵌套定义

    <strong>示例代码：</strong>

    ````java
    public class MethodDemo {
        public static void main(String[] args) {

        }

        public static void methodOne() {
            public static void methodTwo() {
                // 这里会引发编译错误!!!
            }
        }
    }
    ````

- void表示无返回值，可以省略return，也可以单独的书写return，后面不加数据

    <strong>示例代码：</strong>

    ````java
    public class MethodDemo {
        public static void main(String[] args) {

        }
        public static void methodTwo() {
            //return 100; 编译错误，因为没有具体返回值类型
            return;        
            //System.out.println(100); return语句后面不能跟数据或代码
        }
    }
    ````

> **勘误：**示例中的无条件 `return;` 已结束方法，在同一代码块后面直接加打印语句会造成不可达代码。但如果 `return` 位于某个 `if` 分支中，分支后仍可写其他语句，供没有执行该 `return` 的路径继续运行。

### <strong>方法的通用格式</strong>

- 格式：

````java
public static 返回值类型 方法名(参数) {
   方法体;
   return 数据 ;
}
````

- 解释：

| <strong>public static</strong> | 修饰符，目前先记住这个格式 |
| --- | --- |
| <strong>返回值类型</strong> | 方法操作完毕之后返回的数据的数据类型，如果方法操作完毕，没有数据返回，这里写void，而且方法体中一般不写return |
| <strong>方法名</strong> | 调用方法时候使用的标识 |
| <strong>参数</strong> | 由数据类型和变量名组成，多个参数之间用逗号隔开 |
| <strong>方法体</strong> | 完成功能的代码块 |
| <strong>return</strong> | 如果方法操作完毕，有数据返回，用于把数据返回给调用者 |

- 定义方法时，要做到两个明确

    - <strong>明确返回值类型：</strong>主要是明确方法操作完毕之后是否有数据返回，如果没有，写void；如果有，写对应的数据类型
    - <strong>明确参数</strong>：主要是明确参数的类型和数量

> <strong>调用方法时注意：</strong>
> 
> - void类型的方法，直接调用即可
> - 非void类型的方法，推荐用变量接收调用


## <strong>方法重载</strong>

### <strong>方法重载</strong>

- 概念：方法重载指同一个类中定义的多个方法之间的关系，满足下列条件的多个方法相互构成重载

    - 多个方法在同一个类中
    - 多个方法具有相同的方法名
    - 多个方法的参数不相同，类型不同或者数量不同

> **补充：**参数类型的顺序不同也能构成重载，例如 `fn(int a, double b)` 与 `fn(double a, int b)`；只改变参数名不能构成重载。

> <strong>注意：</strong>
> 
> - 重载仅对应方法的定义，与方法的调用无关，调用方式参照标准格式
> - 重载仅针对同一个类中方法的名称与参数进行识别，与返回值无关，换句话说不能通过返回值来判定两个方法是否相互构成重载

- 正确范例：

````java
public class MethodDemo {
    public static void fn(int a) {
            //方法体
    }
    public static int fn(double a) {
            //方法体
    }
}
public class MethodDemo {
    public static float fn(int a) {
            //方法体
    }
    public static int fn(int a , int b) {
            //方法体
    }
}
````

> **勘误：**“正确范例”只是在演示合法的重载关系，两个同名 `MethodDemo` 是独立示例。里面返回类型为 `int`、`float` 的方法体只有注释，直接编译会报缺少返回语句；运行前应补上返回值，例如 `return 0;` 或 `return 0.0f;`。

- 错误范例：

````java
public class MethodDemo {
    public static void fn(int a) {
            //方法体
    }
    public static int fn(int a) {         
    /*错误原因：重载与返回值无关*/
            //方法体
    }
}
public class MethodDemo01 {
    public static void fn(int a) {
        //方法体
    }
} 
public class MethodDemo02 {
    public static int fn(double a) { 
    /*错误原因：这是两个类的两个fn方法*/
        //方法体
    }
}
````

### 练习1：去除重复元素

<strong>需求：</strong>

获取10个1-100之间的随机数并存入到数组当中,要求保证数据是唯一的

<strong>代码示例：</strong>

````java
public class MethodDemo2 {
    public static void main(String[] args) {
        // 获取10个1-100之间的随机数并存入到数组当中,要求保证数据是唯一的

        // 1.创建数组
        int[] arr = new int[10];
        // 2.生成随机数
        Random r = new Random();
        for (int i = 0; i < arr.length; ) {
            int num = r.nextInt(100) + 1;
            // 对num进行判断, 存在---不存  不存在---存入
            if(!contains(num,arr)){
                arr[i] = num;
                // 只有生成一个满足要求的随机数,索引才会自增
                i++;
            }
        }
        // 3.遍历
        for (int i = 0; i < arr.length; i++) {
            System.out.print(arr[i] + " ");
        }
        System.out.println();

    }

    // 定义一个方法，判断num在数组arr中是否存在
    public static boolean contains(int num, int[] arr){
        for (int i = 0; i < arr.length; i++) {
            if(arr[i] == num){
                // 如果遇到一个符合条件的，直接返回true
                // 此时方法直接结束
                return true;
            }
        }
        // 如果循环结束了，数组里面所有的元素都判断完毕了，还没有找到一样，直接返回false，表示num在arr当中是不存在的。
        return false;
    }
}
````

### 练习4：定义方法的小诀窍

<strong>需求：</strong>

给定两个长方形，判断哪个长方形的面积更大？

如何定义方法？

<strong> 小诀窍:</strong>

1. 观察在大段的代码当中, 反复使用的独立功能是什么?   反复使用 + 独立功能   求长方形的面积
2. 这个独立功能,需要什么才能完成? --- 形参  长 宽
3. 方法的调用处,是否需要这个独立功能的结果继续做其他事情?  必须要把面积返回

<strong>代码示例：</strong>

````java
public class MethodDemo5 {
    public static void main(String[] args) {
        double len1 = 10.1;
        double width1 = 5.1;

        double len2 = 9.1;
        double width2 = 6.1;

        double area1 = getArea(len1,width1);
        double area2 = getArea(len2,width2);

        if(area1 > area2){
            System.out.println("第一个长方形面积更大");
        }else if(area2 > area1){
            System.out.println("第二个长方形面积更大");
        }else{
            System.out.println("两个长方形面积相等");
        }
    }
    public static double getArea(double len, double width){
        return len * width;
    }
}
````

````java
public class Test1{
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        int[] scores = new int[5];

        // 要求1：录入5个有效分数
        for (int i = 0; i < 5; ) {
            System.out.print("请输入第" + (i + 1) + "位评委的分数(0-100): ");
            int score = scanner.nextInt();
            if (checkScore(score)) {
                scores[i] = score;
                i++;
            } else {
                System.out.println("分数无效！请输入0-100之间的整数。");
            }
        }

        // 要求2：获取最大值和最小值
        int max = getMaxScore(scores);
        int min = getMinScore(scores);

        // 要求3：计算总分
        int total = getSum(scores);

        // 要求4：计算最终平均分
        double finalScore = (total - max - min) / 3.0;

        // 输出结果
        System.out.println("最高分: " + max);
        System.out.println("最低分: " + min);
        System.out.println("总分: " + total);
        System.out.println("最终得分: " + finalScore);
    }


    // 验证评分的打分是否在合理范围之内
    public static boolean checkScore(int score) {
        return score >= 0 && score <= 100;
    }

    // 获取最大值方法
    public static int getMaxScore(int[] scores) {
        int max = scores[0];
        for (int i = 1; i < scores.length; i++) {
            if (scores[i] > max) {
                max = scores[i];
            }
        }
        return max;
    }

    // 获取最小值方法
    public static int getMinScore(int[] scores) {
        int min = scores[0];
        for (int i = 1; i < scores.length; i++) {
            if (scores[i] < min) {
                min = scores[i];
            }
        }
        return min;
    }

    // 求和
    public static int getSum(int[] scores){
        int sum = 0;
        for (int i = 0; i < scores.length; i++) {
            sum = sum + scores[i];
        }
        return sum;
    }


}
````

### 练习2：计算班级分数

<strong>需求：</strong>

班主任需要统计10名学生的数学成绩（0-100分），

计算及格率，平均分，并找出最高分。

要求1：键盘录入10名学生的成绩，存入数组。超出范围，提示“成绩无效，请重新输入”。

要求2：定义方法，求及格人数，根据及格人数，求及格率。

要求3：定义方法求总分，根据总分求平均分

要求4：定义方法求最大值。

<strong>代码示例：</strong>

````java
public class Test2{
    public static void main(String[] args) {
        int[] grades = new int[10];
        Scanner scanner = new Scanner(System.in);

        // 输入10个有效成绩
        for (int i = 0; i < grades.length; ) {
            System.out.print("请输入第" + (i + 1) + "个学生的成绩(0-100)：");
            int grade = scanner.nextInt();
            if (checkScore(grade)) {
                grades[i] = grade;
                i++;
            } else {
                System.out.println("成绩无效，请重新输入！");
            }
        }

        // 计算及格率
        double passRate = calculatePassRate(grades);

        // 计算平均分
        double averageScore = calculateAverageScore(grades);

        // 找出最高分
        int maxScore = findMaxScore(grades);

        // 输出结果
        System.out.println("===== 成绩分析结果 =====");
        System.out.println("及格率: " + (passRate * 100) + "%");
        System.out.println("平均分:" + averageScore);
        System.out.println("最高分: " + maxScore);
    }

    // 验证评分的打分是否在合理范围之内
    public static boolean checkScore(int score) {
        return score >= 0 && score <= 100;
    }

    // 计算及格率
    public static double calculatePassRate(int[] grades) {
        int passCount = countPassStudents(grades);
        return passCount / (double) grades.length;
    }

    // 计算及格人数
    public static int countPassStudents(int[] grades) {
        int count = 0;
        for (int i = 0; i < grades.length; i++) {
            if (grades[i] >= 60) {
                count++;
            }
        }
        return count;
    }

    // 计算平均分
    public static double calculateAverageScore(int[] grades) {
        int total = calculateTotalScore(grades);
        return total / (double) grades.length;
    }

    // 计算总分
    public static int calculateTotalScore(int[] grades) {
        int total = 0;
        for (int i = 0; i < grades.length; i++) {
            total += grades[i];
        }
        return total;
    }

    // 找出最高分
    public static int findMaxScore(int[] grades) {
        int max = grades[0];
        for (int i = 1; i < grades.length; i++) {
            if (grades[i] > max) {
                max = grades[i];
            }
        }
        return max;
    }

}
````

### 练习3：计算快递邮费

<strong>需求：</strong>

某快递公司的运费规则如下（首重1kg，超出部分按kg计算，不足1kg按1kg算）：

> **补充：**本题代码按“超出首重的重量”分档：总重减去 `1 kg` 后向上取整，其中前 `5 kg` 超重按 `2 元/kg`，再往后的部分按 `1.5 元/kg`。因此总重 `1 kg`、`1.1 kg`、`6 kg`、`6.1 kg` 分别计费 `10`、`12`、`20`、`21.5` 元。

首重1kg：10元；

超出1-5kg：每kg加2元；

超出5kg以上：每kg加1.5元。 

键盘录入小数，表示用户快递的重量，计算最终的结果

要求1：快递重量必须大于0，否则重新输入

要求2：不同价位的计算，单独定义一个方法

<strong>代码说明：</strong>

- 代码示例1：是最基本最简单的代码实现，但是代码中计算超重部分重复代码了，现在重复代码业务较为简单可以忽略，但是如果重复代码比较多，再这么写就不合适了，可以参考代码示例2。
- 代码示例2：解决了代码重复问题，并且对代码进行了优化。

<strong>代码示例1：</strong>

````java
public class ExpressCostCalculator {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        double weight;

        // 输入重量并验证
        while (true) {
            System.out.print("请输入快递重量(kg): ");
            weight = scanner.nextDouble();
            if (weight <= 0) {
                System.out.println("重量必须大于0，请重新输入！");
            } else {
                break;
            }
        }

        // 计算运费
        double cost = calculateCost(weight);
        System.out.println("快递重量:" + weight);
        System.out.println("运费:" + cost);
    }

    // 计算总费用
    public static double calculateCost(double weight) {
        double baseCost = 10.0; // 首重1kg费用

        if (weight <= 1.0) {// 首重1kg以内
            return baseCost;

        } else if (weight > 1.0 && weight <= 6.0) {// 总重1-6kg（超出1-5kg）
            // 1. 计算超出部分的重量
            double overWeight = weight - 1.0;
            // 2. 计算超出部分的价格
            int fullKg = ceil(overWeight);
            return baseCost + fullKg * 2.0;

        } else {// 总重超过6kg（超出5kg以上）
            // 1. 计算前6kg的价格
            double cost = baseCost + 5 * 2.0;

            // 2. 计算超出6kg的部分（向上取整）
            double overWeight = weight - 1.0;
            int fullKg = ceil(overWeight);

            // 3. 剩余部分按1.5元计算
            if (fullKg > 5) {
                cost += (fullKg - 5) * 1.5;
            }
            return cost;
        }
    }

    // 小数向上取整
    // 1.1 -> 2
    // 1.8 -> 2
    public static int ceil(double weight) {
        // 备注：此时无需考虑负数，因为在main方法中已经判断过，以后本方法可以用Math里面的ceil替代
        // 如果是整数，直接返回
        if(weight % 1 == 0){
            return (int)weight;
        }

        // 如果是小数，则向上取整
        return (int) (weight + 1);
    }
}
````

<strong>代码示例2：</strong>

````java
public class ExpressCostCalculator {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        double weight;

        // 输入重量并验证
        while (true) {
            System.out.print("请输入快递重量(kg): ");
            weight = scanner.nextDouble();
            if (weight <= 0) {
                System.out.println("重量必须大于0，请重新输入！");
            } else {
                break;
            }
        }

        // 计算运费
        double cost = calculateTotalCost(weight);
        System.out.println("快递重量: " + weight);
        System.out.println("运费:" + cost);
    }

    // 计算总费用
    public static double calculateTotalCost(double weight) {
        double baseCost = 10.0; // 首重1kg费用

        if (weight <= 1.0) {
            return baseCost;
        }

        // 计算超出部分重量（向上取整）
        double overWeight = weight - 1.0;
        int ceilKg = (int) Math.ceil(overWeight);

        // 计算超出部分费用
        double overCost = calculateOverCost(ceilKg);

        return baseCost + overCost;
    }

    // 计算超出部分费用
    // 下面代码采取了fail fast原则，也是实际开发首选。
    // 即：每段代码负责独立功能，互不影响，遇到问题或者得到结果立刻结束。
    //    google、Oracle鼓励此风格
    public static double calculateOverCost(int kg) {
        // 无超重
        if (kg <= 0) {
            return 0.0;
        }

        // 超重 1kg ~ 5kg
        if (kg <= 5) {
            return kg * 2.0;
        }

        // 超重5kg（前5kg按2元/kg，剩余部分按1.5元/kg）
        return 5 * 2.0 + (kg - 5) * 1.5;
    }
}
````

> **勘误：**这里更准确的说法是“提前返回”：满足某种情况就直接返回结果，减少分支嵌套。通常所说的 fail fast（快速失败）是尽早发现并报告错误，和正常得到结果后返回不是同一个概念。
