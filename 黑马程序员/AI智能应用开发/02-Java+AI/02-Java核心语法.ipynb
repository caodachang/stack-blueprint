## <strong>字面量</strong>

作用：告诉程序员，数据在程序中的书写格式。

| <strong>字面量类型</strong> | <strong>说明</strong> | <strong>程序中的写法</strong> |
| --- | --- | --- |
| 整数 | 不带小数的数字 | 666，-88 |
| 小数 | 带小数的数字 | 13.14，-5.21 |
| 字符 | 必须使用单引号，有且仅能一个字符 | ‘A’，‘0’， ‘我’ |
| 字符串 | 必须使用双引号，内容可有可无 | “HelloWorld”，“黑马程序员” |
| 布尔值 | 布尔值，表示真假，只有两个值：true，false | true 、false |
| 空值 | 一个特殊的值，空值 | 值是：null |

````java
public class Demo {
public static void main(String[] args) {
        System.out.println(10); // 输出一个整数
        System.out.println(5.5); // 输出一个小数
        System.out.println('a'); // 输出一个字符
        System.out.println(true); // 输出boolean值true
        System.out.println("欢迎来到黑马程序员"); // 输出字符串
    }
}
````

### <strong>区分技巧</strong>

1. 不带小数点的数字都是整数类型的字面量。
2. 只要带了小数点，那么就是小数类型的字面量。
3. 只要用双引号引起来的，不管里面的内容是什么，不管里面有没有内容，都是字符串类型的字面量。
4. 字符类型的字面量必须用单引号引起来，不管内容是什么，但是个数有且只能有一个。
5. 字符类型的字面量只有两个值，true、false。

> **勘误：**`true` 和 `false` 是布尔字面量，不是字符字面量。字符字面量写成单引号包围的一个字符，例如 `'A'`；`null` 可用于引用类型，不能直接赋给 `int` 等基本类型。
6. 空类型的字面量只有一个值，null。

> 备注：所有的标点符号必须是英文状态下的如果


## <strong>变量</strong>

### <strong>什么是变量？</strong>

变量就在程序中临时存储数据的容器。但是这个容器中只能存一个值。

### <strong>变量的定义格式</strong>

数据类型 变量名 = 数据值；

#### <strong>格式详解</strong>

<strong>数据类型：</strong>限定了变量当中能存储什么类型的数据。

- 如果要存10，那么数据类型就需要写整数类型。
- 如果要存10.0，那么数据类型就需要写小数类型。

<strong>变量名：</strong>其实就是这个容器的名字。

当以后想要使用变量里面的数据时，直接使用变量名就可以了。

<strong>数据值：</strong>真正存储在容器中的数据。

<strong>分号：</strong>表示语句的结束，就跟以前写作文时候的句号是一样的。

#### <strong>常用的数据类型</strong>

整数：int

小数：（浮点数）double

其他数据类型稍后讲解

<strong>举例：</strong>

````java
public class VariableDemo{
public static void main(String[] args){
    //定义一个整数类型的变量
    //数据类型 变量名 = 数据值;
    int a = 16;
    System.out.println(a);//16
    //定义一个小数类型的变量
    double b = 10.1;
    System.out.println(b);//10.1
  }
}
````

- 变量在使用之前必须要赋值。

### <strong>练习2 - 游戏人物</strong>

<strong>需求</strong>：

我方：叉子           对方：长手

攻击：220            攻击：210

防御：85             防御：80

血量：1012.5         血量：1223.3

技能加成:  1.2       技能加成: 1.3

技能造成伤害的公式：攻击力 &#42; 技能加成 – 对方防御力

普攻造成伤害的公式：攻击力 – 对方防御力

计算：

&emsp;我方第一次进行普通攻击，造成多少伤害，对方还剩余多少血量？

&emsp;我方第二次进行技能攻击，造成多少伤害，对方还剩余多少血量？

 规则：经常发生改变的数据，用变量记录

<strong>代码解析</strong>：

````java
package com.itheima.variable;

public class VariableDemo2 {
    public static void main(String[] args) {
        /*
            我方：叉子           对方：长手
            攻击：220            攻击：210
            防御：85             防御：80
            血量：1012.5         血量：1223.3
            技能加成:  1.2       技能加成: 1.3

            技能造成伤害的公式：攻击力 * 技能加成 – 对方防御力
            普攻造成伤害的公式：攻击力 – 对方防御力

            计算：
                我方第一次进行普通攻击，造成多少伤害，对方还剩余多少血量？
                我方第二次进行技能攻击，造成多少伤害，对方还剩余多少血量？

             规则：经常发生改变的数据，用变量记录
        */


        // 1. 定义变量记录我方的攻击力
        int attack1 = 220;

        // 2. 定义变量记录我方的防御力
        int defense1 = 85;

        // 3. 定义变量记录我方的血量
        double blood1 = 1012.5;

        // 4. 定义变量记录我方技能加成
        double skill1 = 1.2;

        // 5. 定义变量记录对方的攻击力
        int attack2 = 210;

        // 6. 定义变量记录对方的防御力
        int defense2 = 80;

        // 7. 定义变量记录对方的血量
        double blood2 = 1223.3;

        // 8. 定义变量记录对方的技能加成
        double skill2 = 1.3;

        // 9.  我方第一次进行普通攻击，造成多少伤害，对方还剩余多少血量？
        // 普攻造成伤害的公式：攻击力 – 对方防御力
        double damage1 = attack1 - defense2;
        blood2 = blood2 - damage1;
        System.out.println(blood2);// 1083.3

        // 10. 我方第二次进行技能攻击，造成多少伤害，对方还剩余多少血量？
        // 技能造成伤害的公式：攻击力 * 技能加成 – 对方防御力
        double damage2 = attack1 * skill1 - defense2;
        blood2 = blood2 - damage2;
        System.out.println(blood2);// 899.3

    }
}
````

适用于类名

- 如果是一个单词，那么首字母大写。比如：Demo、Test。
- 如果是多个单词，那么每一个单词首字母都需要大写。比如：HelloWorld

1. 尽量不要用拼音。但是一些国际通用的拼音可视为英文单词。

3. 错误：jiage、dazhe


## <strong>算术运算符</strong>

### 基本介绍

<strong>分类</strong>：+  -  &#42;  /  %

<strong>运算特点：</strong>

&#43; - &#42; ：跟小学数学中一模一样没有任何区别

/：

1. 整数相除结果只能得到整数，如果结果想要是小数，必须要有小数参数
2. 小数直接参与运算，得到的结果有可能是不精确的

<strong>案例：</strong>

````java
public class OperatorDemo1 {
    public static void main(String[] args) {
        /*
            算数运算符：+ - * / %

            整数计算、小数计算

        */
        // 1. 整数计算
        // 细节：整数相除结果还是整数，就是商
        //       其他运算跟数学中是一模一样的
        int a = 10;
        int b = 3;

        System.out.println(a + b);// 13
        System.out.println(a - b);// 7
        System.out.println(a * b);// 30
        System.out.println(a / b);// 3
        System.out.println(a % b);// 1

        System.out.println("-------------------------");

        // 2.小数计算
        // 细节：
        //     小数直接参与计算，结果有可能不精确的
        double c = 1.1;
        double d = 1.01;

        System.out.println(c + d);// 13.0
        System.out.println(c - d);// 7.0
        System.out.println(c * d);// 30.0
        System.out.println(c / d);// 3.3333333333333335
        System.out.println(c % d);// 1.0



    }
}
````

> **勘误：**这一段小数运算代码后的若干输出注释沿用了上面整数示例的结果，不能照抄为运行结果。例如 `1.1 + 1.01` 约为 `2.11`，而不是 `13.0`；以实际运行输出为准。

%：取模、取余。

他做的也是除法运算，只不过获取的是余数而已。

<strong>案例：</strong>

````java
System.out.println(10 % 2);//0
System.out.println(10 % 3);//1
````

<strong>应用场景：</strong>

````java
//可以利用取模来判断一个数是奇数还是偶数
System.out.println(15 % 2);//1  奇数
````

### 练习1 - 数值拆分

<strong>需求：</strong>键盘录入一个三位数，将其拆分为个位、十位、百位后，打印在控制台

<strong>代码示例：</strong>

````java
//1.键盘录入一个三位数
//导包 --- 创建对象 --- 接收数据
Scanner sc = new Scanner(System.in);
System.out.println("请输入一个三位数");
int number = sc.nextInt();//123

//2.获取这个三位数的个位、十位、百位并打印出来
//公式：
//针对于任意的一个数而言
//个位： 数字 % 10
int ones = number % 10;
//十位： 数字 / 10 % 10
int tens = number / 10 % 10;
//百位： 数字 / 100 % 10
int hundreds = number / 100  % 10;

//输出结果
System.out.println(ones);
System.out.println(tens);
System.out.println(hundreds);
````

> <strong>公式：</strong>
> 
> 获取任意一个数上每一位数。
>

## <strong>隐式转换</strong>

> <strong>概念：</strong>也叫自动类型提升。就是把一个取值范围小的数据或者变量，赋值给另一个取值范围大的变量。此时不需要我们额外写代码单独实现，是程序自动帮我们完成的。

<strong>简单记忆：</strong>就是小的给大的，可以直接给。

<strong>触发时机：</strong>

- 不同类型的数据进行计算，默认采取隐式转换，Java自动转换，无需我们写代码

<strong>提升规则：</strong>

- <strong>规则一：</strong>如有byte short类型的数据，先提升为int类型
- <strong>规则二：</strong>把取值范围小的提升为取值范围大的，再进行运算

> 备注1：
> 
> 提升的时候是先执行规则一，再执行规则二，如果计算的数据没有byte和short类型，则跳过规则一，直接执行规则二。
> 
> 备注二：
> 
> <strong>取值范围从小到大的关系：</strong>byte short int long float double

> **补充：**这串顺序用于记忆数值转换方向，但“自动转换”不保证所有数值都精确不变。例如 `long` 转 `float` 可能丢失整数精度；混合运算也按表达式的实际结合顺序逐步提升类型。

````java
// &  //两边都是真，结果才是真。
System.out.println(true & true);//true
System.out.println(false & false);//false
System.out.println(true & false);//false
System.out.println(false & true);//false

System.out.println("===================================");

// | 或  //两边都是假，结果才是假，如果有一个为真，那么结果就是真。
System.out.println(true | true);//true
System.out.println(false | false);//false
System.out.println(true | false);//true
System.out.println(false | true);//true
````

### <strong>短路逻辑运算符</strong>

<strong>分类：</strong>  &amp;&amp;   &#124;&#124;

<strong>&amp;&amp;：</strong>运算结果跟&amp;是一模一样的，只不过具有短路效果。

<strong>&#124;&#124;：</strong>运算结果跟&#124;是一模一样的。只不过具有短路效果。

<strong>逻辑核心：</strong>

- 当左边不能确定整个表达式的结果，右边才会执行。
- 当左边能确定整个表达式的结果，那么右边就不会执行了。从而提高了代码的运行效率。

<strong>举例：</strong>

- <strong>用户登录案例</strong>

<strong>用户名正确  &amp; 密码正确</strong>

如果使用一个&amp;，不管用户名是否正确都会去验证密码。

<strong>思考：</strong>

如果用户名输入正确了，那么我们再判断密码是否正确，是符合业务逻辑的。但是如果用户名输入错误了，那么现在还有必要去比较密码吗？没有不要了。

如果使用一个&amp;，那么左边和右边不管什么情况下，都会执行。

<strong>用户名正确  &amp;&amp;  密码正确</strong>

如果用户名输入正确了，那么才会验证密码是否输入正确。如果用户名输入错误了，那么就不会再去验证密码是否正确，最终的结果直接为false。从而提高了程序运行的效率。

> <strong>总结：</strong>
> 
> - &amp;&amp; 和 &amp; 、&#124;&#124;和&#124;的运行结果都是一模一样的。
> - 但是短路逻辑运算符可以提高程序的运行效率。

> <strong>建议：</strong>最为常用： &amp;&amp;   &#124;&#124;   ！

### 练习3 - 回文数

<strong>需求</strong>：

键盘录入一个四位整数，判断这个数字是否为回文数。

<strong>代码解析</strong>：

````java
public class OperatorDemo12 {
    public static void main(String[] args) {
        // 键盘录入一个四位整数，判断这个数字是否为回文数
        // 1. 键盘录入一个四位整数
        Scanner sc = new Scanner(System.in);
        System.out.println("请输入一个四位整数：");
        int number = sc.nextInt();
        // 2. 数字拆分
        int ge = number % 10;
        int shi = number / 10 % 10;
        int bai = number / 100 % 10;
        int qian = number / 1000 % 10;
        System.out.println(ge);
        System.out.println(shi);
        System.out.println(bai);
        System.out.println(qian);
        // 3. 判断是否为回文数
        //   qian == ge  &&  bai == shi
        System.out.println(qian == ge && bai == shi);
    }
}
````

### 练习4 - 7的有缘数

<strong>需求</strong>：

寻找7的有缘数，定义一个两位整数，只要该数字包含7或者是7的倍数，就是7的有缘数

<strong>代码解析</strong>：

````java
public class OperatorDemo13 {
    public static void main(String[] args) {
        // 寻找7的有缘数，定义一个两位整数，只要该数字包含7或者是7的倍数，就是7的有缘数
        // 1. 定义一个两位整数
        int number = 22;

        // 包含7: 个位或者十位是7即可
        // 7的倍数：number % 7 == 0
        // 2. 获取个位、十位
        int ge = number % 10;
        int shi = number / 10;

        // 3. 判断当前数字是否是7的有缘数
        // ge == 7  ||  shi == 7  || number % 7 == 0
        System.out.println(ge == 7 || shi == 7  || number % 7 == 0);
    }
}
````


## <strong>三元运算符</strong>

又叫做：三元表达式或者问号冒号表达式。

<strong>格式：</strong>关系表达式 ？ 表达式1 ：表达式2 ；

<strong>计算规则：</strong>

- 计算关系表达式的值。
- 如果关系表达式的值为真，那么执行表达式1。
- 如果关系表达式的值为假，那么执行表达式2。

> <strong>注意：</strong>三元运算符的最终结果一定要被使用，要么赋值给一个变量，要么直接打印出来。

<strong>案例：</strong>

````java
public class OperatorDemo12 {
    public static void main(String[] args) {
        //需求：求两个数的较大值
        int a = 10;
        int b = 20;

        //格式：关系表达式 ？ 表达式1 ： 表达式2 ；
        //注意点：
        //三元运算符的最终结果一定要被使用。
        //要么赋值给一个变量，要么直接输出。
       int max =  a > b ? a : b ;
        System.out.println(max);


        System.out.println(a > b ? a : b);
    }
}
````
