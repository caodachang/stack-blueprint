### 环境变量

#### 配置环境变量的作用

如果我想要在CMD的任意目录下，都可以启动某一个软件，那么就可以把这个软件的路径配置到环境变量中的PATH里面。

在启动软件的时候，操作系统会先在当前路径下找，如果在当前路径没有找到，再到环境变量的路径中去找。如果都找不到就提示无法启动。

#### 把任意软件配置环境变量的步骤

1. 右键我的电脑，选择属性
2. 点击左侧的高级系统设置
3. 选择高级，再点击下面的环境变量
4. 找系统变量里面的PATH
5. 把软件的完整路径，配置到PATH当中就可以了
6. （可做可不做）就是把自己配置的路径，移动到最上面

（如果无法出现第二步界面，可以打开我的电脑之后右键点击空白处）



第二步：点击高级系统设置。

#### 把JDK配置到环境变量当中

开发Java程序，需要使用JDK提供的各种开发工具，而这些工具在JDK的安装目录的bin目录下，如果不配置环境变量，那么这些命令只可以在bin目录下使用，而我们想要在任意目录下都能使用，所以就要配置环境变量。

解释一下java -version出现之后信息的意思（以下内容可以不看）

````java
java version "17.0.12" 2024-07-16 LTS  // 本次使用java的大版本为17，小版本为12，表示把17的版本，更新12次的小版本，更新时间为：2024年7月16日，是LTS正式版
Java(TM) SE Runtime Environment (build 17.0.12+8-LTS-286)//Java标准版运行环境的构建号为17.0.12+8-LTS-286。286是一个补丁号。
Java HotSpot(TM) 64-Bit Server VM (build 17.0.12+8-LTS-286, mixed mode, sharing)//这表示Java使用的是HotSpot虚拟机，这是一个64位的服务器版虚拟机。构建号与运行时环境相同。mixed mode表示虚拟机运行在混合模式下，即同时使用解释执行和即时编译（JIT）执行Java代码。sharing表示虚拟机使用了类数据共享（CDS）功能，这可以提高启动速度和减少内存占用。
````

> 以上三行信息，我们一般只看第一行，知道自己安装的是哪个版本即可，版本只看前面的大版本

## IDEA 介绍

### IDEA的基本使用

<strong>IntelliJ IDEA</strong> 是一款强大的 Java 集成开发环境（IDE），被广泛认为是开发 Java 程序的最佳工具之一。

在开始本文档之前，请确保你已正确安装 <strong>JDK</strong> 和 <strong>IDEA</strong>

### 项目层级结构

在 IDEA 中，Java 项目按以下层级进行组织：

1. <strong>Project（项目）</strong>：整体工程，比如微信、QQ 等应用。
2. <strong>Module（模块）</strong>：项目的功能子模块，如聊天、通讯录等。
3. <strong>Package（包）</strong>：模块内的业务分类，例如聊天模块中的朋友圈、视频号等。
4. <strong>Class（类）</strong>：具体的代码单元，实际开发工作主要在类中完成。

> 注：模块暂时用不到，我们先不创建，等后面章节用到了模块，我们再进行使用
> 
> <strong>目前使用的层级</strong>：<strong>项目（Project）</strong>→ <strong>包（Package）</strong>→ <strong>类（Class）</strong>。

## 解释HelloWorld代码的含义

HelloWorld代码作为我们第一个写的Java代码，具有很重要的意义。以后所有的代码结构、运行方式都会按照本案例进行。

### <strong>注释</strong>

注释是对代码的解释和说明文字。

Java中的注释分为三种：

- 单行注释：

````java
// 这是单行注释文字
````

- 多行注释：

````java
/*
这是多行注释文字
这是多行注释文字
这是多行注释文字
*/
````

> 注意：多行注释不能嵌套使用。

- 文档注释（暂时用不到）：

````java
/**
这是多行注释文字
这是多行注释文字
这是多行注释文字
*/
````

<strong>使用的技巧</strong>

1. 如果我们要对代码进行解释，那么就可以使用注释。
2. 当注释的内容比较少，一行就写完了，可以用单行注释。
3. 如果注释的内容比较多，需要写在多行，那么可以使用多行注释。

> <strong>注意：</strong>注释的内容不会参与编译和运行的，仅仅是对代码的解释说明而已。所以，不管在注释当中写什么内容，都不会影响代码运行的结果。

### <strong>关键字</strong>

#### 什么是关键字？

被Java赋予了特定含义的英文单词。

当我们在代码中写了关键字之后，程序在运行的时候，就知道要做什么事情了。

> <strong>注意</strong>：关键字很多，不用刻意去记。

> **嵌入表格说明：**原讲义在此处有一张飞书嵌入表格，但公开页面未显示其单元格，因此无法逐字转写。可回到[原讲义](https://heuqqdmbyk.feishu.cn/docx/Zx5tdqkrpoVXFuxTPBSc5cEhnuJ)查看。下面只补充本章示例中实际出现的词：

| 词 | 在本章中的作用 |
| --- | --- |
| `class` | 定义类 |
| `public` | 声明公开可见 |
| `static` | 让 main 方法属于类本身 |
| `void` | 方法不返回值 |
| `package` | 声明所属包 |

> `String` 是类名；`main`、`args`、`println` 是标识符，不是 Java 关键字。

#### <strong>第一个关键字 class</strong>

表示定义一个类。创建一个类。

类：Java 项目最基本的组成单元，一个完整的 Java 项目有可能会有成千上万个类来组成的。

class 后面跟随的就是这个类的名字，简称：类名。

在类名后面会有一对大括号，表示这个类的内容。

<strong>举例：</strong>

````java
public class HelloWorld{

}
````

<strong>解释：</strong>

class表示定义类。

类名：HelloWorld

HelloWorld 后面的大括号表示这个类的范围。

### HelloWorld代码解析

> **补充：**代码第一行声明了 `package com.itheima.test;`，它必须与前面创建的包一致；若把文件直接放在默认包里，应去掉这一行。源文件名应为 `HelloWorld.java`，与 `public class HelloWorld` 保持一致。

了解了注释和关键字后，让我们结合注释逐行解析刚刚写的完整 Java 程序：

````java
package com.itheima.test; // IDEA 自动生成，package 是一个关键字，表示当前类所属的包，便于组织代码。

// class：定义一个类，后面跟随类名
// HelloWorld：类的名字，通常与文件名一致
// {}：类的范围，所有代码都需要写在这对大括号中
public class HelloWorld {

    // 程序的主入口（main 方法）
    // 程序运行时会从这个入口点开始执行代码
    // 快速生成方式：psvm
    public static void main(String[] args) {

        // 输出语句
        // 作用：将小括号中的内容打印到控制台
        // 快速生成方式：sout
        System.out.println("Hello World");
    }
}
````
