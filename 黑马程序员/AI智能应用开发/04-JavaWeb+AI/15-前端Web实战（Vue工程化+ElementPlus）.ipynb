## 复习与排错（补充）

| 现象 | 检查方法 | 处理思路 |
| --- | --- | --- |
| `node` 不是内部或外部命令 | `where node`、检查 PATH | 重新打开终端，让环境变量生效；必要时重新安装 |
| `npm` 命令不可用 | `npm -v` | 确认 Node.js 安装目录已加入 PATH |
| 全局安装后找不到命令 | `npm config get prefix`、检查 PATH | 把 prefix 对应的可执行目录加入 PATH |
| 下载依赖很慢或失败 | `npm config get registry` | 检查镜像地址、网络和证书；必要时恢复官方 registry |
| 配置改错 | `npm config list` | 使用 `npm config delete <key>` 删除配置，或重新设置正确值 |

### 推荐验收命令

```powershell
node -v
npm -v
npm config get prefix
npm config get registry
npm doctor
```

`npm config set` 会写入 npm 配置文件，配置对后续终端和项目生效。切换 Node.js 版本时，注意全局包目录和 PATH 可能需要重新检查。镜像适合加速依赖下载，发布包和生产构建仍应固定 lockfile，并在 CI 中使用可复现的依赖源。
