# 01-Dify入门：Agent概念、平台介绍与本地部署

本篇对应课程第 1 章《Dify 入门》，围绕两个问题展开：什么是能够完成任务的 AI Agent，以及如何在 Windows 电脑上安装并运行 Dify。内容依据课程《第一章节：Dify 入门》课件整理；课件中把安装过程展开为 Docker Desktop、Dify 仓库、环境配置、启动和首次登录五步，具体命令以课程附件文档为准。




## 1. 先看课程要解决的问题

课程从企业应用中的四类痛点切入：客服会反复回答相同问题，内容创作速度跟不上需求，数据分析需要等待专业人员，个性化服务又会带来较高的人力成本。

这些问题有共同特点：任务可以描述清楚，但执行过程包含资料查询、判断、调用工具和多步操作。单纯的聊天模型能生成文字，却不一定能把任务完整做完，因此课程先引出 AI Agent，再介绍用于搭建 Agent 应用的 Dify。




## 2. 大模型本身为什么不够用

课件把大模型的局限归纳为四点：

| 局限 | 具体表现 | 对应用的影响 |
| --- | --- | --- |
| 知识有截止时间 | 知识停留在训练数据的时间范围内 | 无法直接回答最新信息 |
| 默认不能联网 | 不会自动访问实时网页或业务系统 | 难以查询外部数据 |
| 专业深度有限 | 对企业内部领域知识掌握不足 | 专业回答可能不准确 |
| 不能直接执行 | 只能输出文本，不能自动完成业务操作 | 无法独立推进完整任务 |

因此，应用开发需要为模型补充知识、工具和执行流程。Agent 正是把这些能力组织起来的一种应用形态。




## 3. 什么是 AI Agent

课程给出的直观说法是：**AI Agent 是一个能够干活的 AI 管家**。它接收目标后，会结合环境信息进行判断，选择合适的工具，并通过多步行动完成任务。

一个 Agent 至少包含下面三类能力：

- **自主性**：根据任务和当前结果独立做出下一步决策。
- **工具使用**：调用搜索、数据库、天气服务或企业接口等外部能力。
- **任务规划**：把复杂目标拆成多个步骤，按顺序执行并根据结果继续推进。

这三个能力共同决定了 Agent 能否从“一问一答”扩展到“提出目标后完成工作”。




## 4. Agent 与普通大模型的区别

可以从任务执行方式理解两者的差异：

| 对比维度 | 传统大模型 | AI Agent |
| --- | --- | --- |
| 定位 | 被动响应问题 | 主动执行任务 |
| 信息获取 | 主要依赖训练数据 | 可以实时搜索或读取外部数据 |
| 能力边界 | 以对话和文本生成为主 | 可以调用工具 |
| 任务处理 | 通常给出单轮回答 | 可以进行多步规划 |
| 应用场景 | 咨询、问答 | 需要实际执行的业务 |

课程用一句话概括两者的关系：**Agent = 大模型 + 工具**。这里的“工具”不只是一段函数，也可以是知识库、搜索服务、数据库或企业 API。大模型负责理解和决策，工具负责提供数据或执行动作。




## 5. Agent 能带来什么变化

课件用三个应用案例说明 Agent 的落地价值：

| 案例 | 课件给出的结果 | 说明 |
| --- | --- | --- |
| 中国电信客服 Agent 集群 | 服务 145 万次 | Agent 可以承担大量重复客服服务 |
| 华大基因专业遗传咨询 Agent | 效率提升 60% | Agent 可以辅助专业咨询流程 |
| 平安保险智能理赔 Agent | 用户满意度 4.8/5.0 | Agent 可以参与理赔等业务办理 |

这些数据用于说明应用方向。真正部署时仍需要结合业务数据、权限边界和人工审核机制评估效果。


## 6. 本章小结：先区分两个概念

1. **什么是 AI Agent？**

   能感知环境、自主决策、使用工具并完成任务的智能实体。

2. **Agent 和大模型有什么区别？**

   大模型主要负责理解和生成内容，Agent 在此基础上增加工具调用、任务规划和执行能力。

理解这两个概念后，下一步就是选择一个平台，把模型、工具、知识库和流程编排成可运行的应用。


## 7. 什么是 Dify

Dify 是一个开源的大语言模型（LLM）应用开发平台，用来简化生成式 AI 应用的创建和部署。课程官网地址为 [https://dify.ai/zh](https://dify.ai/zh)。

课件强调了 Dify 的三个特点：

- **低代码或无代码**：通过可视化界面编排业务逻辑，像拖拽积木一样组合应用能力。
- **功能完整**：支持接入多种主流模型，覆盖聊天、知识库、工作流和 Agent 等常见场景。
- **开源并支持本地部署**：可以把平台部署在自己的电脑或服务器上，便于学习和管理数据。




## 8. Dify 可以构建哪些应用

课程把 Dify 的核心应用能力分成四类：

| 能力 | 作用 | 适合的任务 |
| --- | --- | --- |
| 聊天助手 | 管理上下文，支持多轮对话 | 客服、咨询、问答 |
| 工作流 | 在可视化画布中编排业务步骤 | 信息抽取、审核、自动化处理 |
| 知识库（RAG） | 接入私有文档，让回答基于自有知识 | 企业制度、产品资料、游戏攻略 |
| Agent 智能体 | 自主调用工具并拆解复杂任务 | 查询、分析、需要行动的业务助手 |

这四类能力可以组合使用。例如，一个客服 Agent 可以同时连接企业知识库，并通过工作流执行工单处理。




## 9. Dify 的本地安装路线

课程把安装和首次运行拆成五步：

1. **安装 Docker Desktop**：Windows 和 macOS 的安装方式需要区分，先准备好 Docker 运行环境。
2. **克隆 Dify 仓库**：把 Dify 项目文件下载到本地，并进入课程指定的 Docker 部署目录。
3. **配置 Docker 环境**：复制环境变量示例文件，生成本地配置文件。
4. **启动 Dify**：使用 Docker Compose 启动平台所需的服务。
5. **初始化登录**：首次打开页面后注册账号并设置登录密码。




## 10. Docker 部署中的关键文件和命令

课件明确展示了环境配置和启动命令：

```bash
# 在 Dify 仓库的 Docker 部署目录中执行
cp .env.example .env
docker compose up -d
```

两条命令的作用不同：

- `cp .env.example .env`：从示例配置生成实际使用的环境变量文件。后续需要修改配置时，应编辑 `.env`。
- `docker compose up -d`：按 Compose 配置启动 Dify 依赖的服务，并让命令在后台运行。

课程课件只展示了这两个关键步骤；仓库克隆地址、目录切换和 Windows Docker Desktop 的安装细节，按照随课附件文档执行。执行命令前要确认当前目录确实包含 `.env.example` 和 Compose 配置文件。


## 11. 启动后的检查顺序

完成启动后，可以按这个顺序检查：

1. Docker Desktop 已启动，相关容器处于运行状态。
2. 浏览器能够打开 Dify 页面。
3. 首次访问页面时可以完成注册和登录。
4. 登录后能看到聊天助手、工作流、知识库或 Agent 等应用入口。

如果页面打不开，先回到 Docker Desktop 检查服务状态，再核对 `.env` 文件和 Compose 配置所在目录。不要在错误目录反复执行启动命令，也不要把 `.env` 中的密钥和密码提交到公开仓库。
