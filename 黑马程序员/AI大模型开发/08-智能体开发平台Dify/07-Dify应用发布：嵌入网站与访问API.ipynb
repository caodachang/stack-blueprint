# 07-Dify应用发布：嵌入网站与访问 API

本篇对应课程第 7 章《Agent 发布》，介绍如何把已经完成的 Agent 应用交付给用户：一种方式是嵌入网站，另一种方式是通过 API 访问。课程以 NLP2SQL 数据分析助手为例，演示发布入口、测试运行和 API Key 使用。




## 1. 从应用搭建到用户使用

前面的章节完成了提示词、知识库、工具和工作流。本章解决最后一步：如何让其他用户触达已经配置好的 Agent。

Dify 课件展示两种发布方式：

| 方式 | 使用场景 | 用户入口 |
| --- | --- | --- |
| 嵌入网站 | 把聊天能力放进已有网站页面 | 网站中的聊天窗口 |
| 访问 API | 由前端、后端或其他程序调用 | 程序通过接口发送请求 |

两种方式使用的是同一个 Agent 应用，但接入位置不同。


## 2. 嵌入网站：打开发布入口

课件以 NLP2SQL 数据分析助手为例，操作起点是：

1. 打开目标 Agent 应用。
2. 选择“发布”。
3. 进入应用发布页面。

发布页面会展示当前应用可用的交付方式。先确认选择的是准备交付的 Agent，避免在错误的应用中生成入口或配置。




## 3. 选择网站嵌入方式

在发布方式中选择第二种“嵌入网站中”。Dify 会为当前 Agent 提供网站嵌入相关的配置或代码入口。

课件演示的重点是发布路径：从 Agent 进入发布页面，再选择网站嵌入方式。具体展示的代码或配置应以当前 Dify 页面生成的内容为准，不要手动猜测或改写嵌入参数。




## 4. 测试网站中的聊天窗口

完成嵌入配置后运行测试，页面右下角会显示聊天窗口。可以通过测试输入确认：

- 聊天窗口能否正常打开。
- 用户消息是否能发送到 Agent。
- Agent 是否返回预期回答。
- NLP2SQL 助手是否能按照之前编排的流程处理问题。




## 5. 网站嵌入的发布检查

网站嵌入属于面向最终用户的交付入口，发布前应检查：

1. Agent 使用的模型、知识库和工具配置正确。
2. 测试问题能够得到稳定结果。
3. 页面上的聊天窗口位置和打开方式符合预期。
4. 不把 API Key、内部地址或其他敏感配置直接暴露给访客。

课件只展示了嵌入方式和测试窗口，没有展开具体网页框架或前端代码，因此这里保留 Dify 生成的嵌入配置作为实际来源。


## 6. 通过 API 访问 Agent

另一种发布方式是 API。操作起点与网站嵌入相同：

1. 打开 NLP2SQL 数据分析助手。
2. 选择“发布”。
3. 在发布方式中进入 API 访问配置。



API 方式适合由自己的前端、后端或自动化程序调用 Agent，把模型能力接入现有业务系统。


## 7. 获取 API Key

在 API 发布页面获取当前 Agent 的 API Key。API Key 用来证明调用者有权访问该应用。

使用时要注意：

- 只在受控的服务端配置 API Key。
- 不要把真实 Key 写进公开代码仓库、网页源码或截图。
- 发现 Key 泄露时，应按平台提供的方式及时更换或撤销。
- 不同 Agent 使用各自的访问凭证，调用前确认 Key 对应的应用。




## 8. 使用 Postman 测试 API

课件使用 Postman 进行接口测试。基本过程是：

1. 在 Dify 发布页面查看当前 Agent 的 API 访问信息。
2. 在 Postman 中创建对应请求。
3. 按页面说明填写请求地址、认证信息和请求内容。
4. 发送请求，观察返回结果。

具体请求方法、接口地址、请求头和 JSON 字段以当前 Dify 页面提供的 API 文档为准。课件页没有完整展示这些字段，因此本篇不虚构一套固定请求示例。




## 9. API 测试要检查什么

测试运行时，重点确认：

- API Key 能够通过认证。
- 请求内容能被 Agent 正确理解。
- 返回结果与网页聊天窗口中的结果一致或符合应用设计。
- NLP2SQL 流程中的 SQL 生成、数据库查询和结果展示仍然可用。
- 错误请求能够返回可理解的错误信息。




## 10. 网站嵌入与 API 的选择

| 需求 | 推荐方式 | 原因 |
| --- | --- | --- |
| 希望用户直接在网页聊天 | 嵌入网站 | 用户通过页面中的聊天窗口使用 |
| 已有业务系统需要调用 Agent | API | 程序可以把 Agent 接入现有流程 |
| 想快速验证应用效果 | 先嵌入或使用页面测试 | 不需要先开发完整调用程序 |
| 需要自定义前端体验 | API | 由业务系统控制界面和交互 |

无论选择哪种方式，都要先在 Dify 中完成 Agent 的模型、提示词、知识库和工具配置，再进行发布测试。


## 11. Dify-Agent 课程主线回顾

课件最后把整个课程归纳为五个环节：

1. **提示词**：学习提示词工程，控制模型行为。
2. **RAG**：构建和应用知识库，补充模型知识。
3. **工具制作**：通过自定义插件扩展外部能力。
4. **工作流**：编排节点，组织复杂业务流程。
5. **Bot 应用**：发布和运营，让用户真正使用 Agent。




## 12. 发布前的最小检查清单

在把 Agent 交给其他用户之前，可以完成一次最小检查：

1. 用几组真实问题测试提示词和模型回答。
2. 检查知识库检索结果是否相关。
3. 检查插件和工作流节点是否能正常执行。
4. 检查没有把内部资料和敏感凭证暴露在前端。
5. 分别测试网站嵌入入口或 API 入口。
6. 记录当前发布的 Agent 版本和可用范围。

这样可以把“应用能运行”与“应用适合交付”区分开，减少发布后才发现配置错误的情况。


## 13. 本章小结

Dify Agent 可以通过两种方式触达用户：嵌入网站，或通过 API 访问。网站嵌入适合直接提供聊天窗口，API 适合接入已有系统。发布流程的共同起点是打开 Agent 的发布入口，选择对应方式，完成测试，再交付给用户。
