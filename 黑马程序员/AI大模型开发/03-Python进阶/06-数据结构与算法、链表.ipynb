本篇对应《Python 进阶》第 6 天《算法详解_链表》讲义，按“算法与复杂度 → 数据结构与顺序表 → 单链表”的顺序整理。重点是理解数据如何组织、操作效率如何衡量，以及如何用 Python 面向对象的方式实现单链表。

## 一、算法与时间、空间复杂度

### 1. 数据结构与算法的关系

- **数据结构**是存储、组织数据的方式，描述数据元素之间的关系。
- **算法**是为解决实际问题设计的方法和步骤。
- 数据结构是算法处理问题的载体，算法决定如何对数据进行运算；二者结合后形成程序。
- 常见数据运算包括插入、删除、修改、查找和排序。

同一批学员信息用列表保存时，按姓名顺序查找通常需要遍历，复杂度约为 `O(n)`；用姓名作为键保存到字典时，平均查找可达到 `O(1)`。这说明数据结构的选择会直接影响算法效率。

### 2. 算法的五大特性

1. **输入**：可以有零个或多个输入。
2. **输出**：至少产生一个结果。
3. **有穷性**：执行有限步骤后能够结束。
4. **确定性**：每一步都有明确含义，同样输入得到确定的执行过程。
5. **可行性**：每一步都能通过已经实现的基本操作完成。

算法独立于具体编程语言，同一个算法可以用 Python、Java 等不同语言实现。

### 3. 时间复杂度

程序运行时间可以抽象为：`T = 操作步骤数量 × 单个操作步骤的执行时间`。

实际分析时，关注问题规模 `n` 增大后步骤数量如何增长，用大 `O` 记法描述渐进复杂度。默认讨论最坏时间复杂度，除非特别说明。

- 基本操作：`O(1)`。
- 顺序结构：复杂度相加，保留增长最快的项。
- 嵌套循环：按各层循环的实际执行次数分析；两层都执行 `n` 次的嵌套循环为 `O(n²)`。
- 分支结构：取各分支复杂度的最大值。
- 忽略低次项、常数项和常数系数。

例如，三层都遍历 `n` 次的穷举过程为 `O(n³)`；若通过整理或预处理减少一层循环，可优化到 `O(n²)`。还要区分最优、最坏和平均复杂度：最优是最快情况，最坏是最慢情况，平均是所有输入情况的期望。

### 4. 常见复杂度与空间复杂度

常见时间复杂度从低到高为：`O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(n³)`。

空间复杂度衡量算法额外占用的存储空间，例如固定数量的临时变量为 `O(1)`，随 `n` 增长的一维辅助数组为 `O(n)`，二维表通常为 `O(n²)`。时间和空间之间常常需要权衡：用额外空间保存索引，可能换取更快的查找。

## 二、数据结构与顺序表

### 1. 内存中的存储结构

讲义用教学模型说明：内存以字节为基本存储单位，每个字节都有自己的地址；一个地址对应 `8 bit` 的存储空间。数据类型占用的字节数不同，例如讲义以整型占 4 个字节、单个字符占 1 个字节进行示意。

这里的字节数是理解地址和连续存储的模型。Python 中整数、字符串和列表元素是对象，实际占用还包括对象管理信息，不能直接把示意图中的字节数当成 Python 对象的精确内存大小。

### 2. 数据结构的分类

按照数据元素之间的关系，数据结构分为：

| 类型 | 关系特点 | 典型结构 |
| --- | --- | --- |
| 线性结构 | 结点最多只有一个直接前驱和一个直接后继 | 栈、队列、线性表 |
| 非线性结构 | 一个结点可能有多个直接前驱或后继 | 树、图 |

线性结构还要考虑实际存储方式：

- **顺序表**：元素连续存放在一块连续存储区中，顺序由物理存储顺序表示。
- **链表**：元素存放在彼此链接的存储块中，结点在内存中可以不连续。

### 3. 顺序表的存储方式

顺序表有一体式和分离式两种组织方式：

- **一体式结构**：信息区和数据区连续存储，扩容时通常需要整体搬迁。
- **分离式结构**：信息区保存数据区的地址、容量和当前元素个数，扩容时可以更换数据区并更新地址。

顺序表完整信息包括数据区和信息区。由于顺序表可通过首地址加下标偏移直接定位元素，按下标访问的复杂度为 `O(1)`。当元素大小不固定时，实际实现通常保存元素对象的地址，而不是假设所有元素占用相同字节数。

### 4. 顺序表扩充与增删复杂度

扩充数据区常见两种策略：

| 策略 | 优点 | 代价 |
| --- | --- | --- |
| 每次增加固定数量 | 节省空间 | 扩充频繁，操作次数多 |
| 每次容量加倍 | 减少扩充次数，推荐 | 可能暂时占用更多空间 |

常见操作复杂度如下：

- 尾端插入：`O(1)`（不考虑扩容搬迁时）。
- 不保持顺序的指定位置插入：把该位置原有元素移到尾端空位，再把新元素放入指定位置，为 `O(1)`（不考虑扩容），但会破坏顺序。
- 保持顺序的指定位置插入：后续元素整体后移，为 `O(n)`。
- 尾端删除：`O(1)`。
- 不保持顺序的指定位置删除：用末元素填补空位，约为 `O(1)`，但会破坏顺序。
- 保持顺序的指定位置删除：后续元素整体前移，为 `O(n)`。

## 三、单链表的结构

### 1. 为什么需要链表

顺序表要求一段连续的存储空间。内存充足时容易找到连续空间；内存不足或空间碎片较多时，即使总空闲空间足够，也可能无法完成扩充。

链表把元素分散到多个存储块中，每个结点保存数据和下一个结点的地址，不要求结点连续。最后一个结点的链接域指向 `None`。

![单链表结点与链接结构](06-img/06-链表结点.jpg)

单链表结点包含两个域：`item` 保存数据，`next` 保存下一个结点的位置。`head` 指向头结点，从 `head` 出发沿 `next` 可以访问后续结点。

### 2. 结点对象与链表对象

从面向对象角度把职责分开：`SingleNode` 表示一个结点，`SingleLinkList` 管理头指针以及链表操作。访问结点时，`node.item` 读取数据，`node.next` 取得下一个结点。

![结点结构与对象实现](06-img/07-链表结点代码.jpg)

```python
class SingleNode(object):
    # 链表结点实现
    def __init__(self, item):
        self.item = item
        self.next = None


class SingleLinkList(object):
    # 单链表的实现
    def __init__(self, node=None):
        self.head = node
```

后文的 `is_empty`、`length`、`travel`、`add`、`append`、`insert`、`remove`、`search` 都是 `SingleLinkList` 的实例方法，放入该类时应与 `__init__` 保持同级缩进。

## 四、判空、长度与遍历

### 1. 判断链表是否为空

只需检查头指针：`head is None` 表示没有任何结点。

```python
def is_empty(self):
    return self.head is None
```

### 2. 测量链表长度

用游标 `cur` 从头结点开始，沿 `next` 移动；每访问一个真实结点就把计数加一，直到 `cur` 变为 `None`。空链表长度为 `0`，时间复杂度为 `O(n)`，额外空间复杂度为 `O(1)`。

![链表长度测量](06-img/13-链表长度.jpg)

```python
def length(self):
    cur = self.head
    count = 0
    while cur is not None:
        cur = cur.next
        count += 1
    return count
```

### 3. 遍历链表

每次处理 `cur.item` 后都要执行 `cur = cur.next`，否则会停留在同一个结点。

```python
def travel(self):
    cur = self.head
    while cur is not None:
        print(cur.item)
        cur = cur.next
```

## 五、增加结点

### 1. 头部插入

头插只需要改变两个引用，顺序不能颠倒：先让新结点指向原来的 `head`，再让 `head` 指向新结点。如果先覆盖 `head`，就会丢失原链表入口。时间复杂度为 `O(1)`。

![头部添加结点](06-img/18-头部添加.jpg)

```python
def add(self, item):
    node = SingleNode(item)
    node.next = self.head
    self.head = node
```

### 2. 尾部插入

空链表直接把新结点设为 `head`；非空链表从头结点出发，找到满足 `cur.next is None` 的尾结点，再令 `cur.next = node`。没有尾指针时需要遍历，时间复杂度为 `O(n)`。

![尾部添加结点](06-img/19-尾部添加.jpg)

```python
def append(self, item):
    node = SingleNode(item)
    if self.is_empty():
        self.head = node
    else:
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = node
```

易错点是循环条件：尾插应停在最后一个真实结点，不能让 `cur` 先变成 `None`。

### 3. 指定位置插入

课程按从 `0` 开始的位置处理边界：`pos <= 0` 按头部插入，`pos >= length()` 按尾部插入，其他情况找到插入位置的前一个结点。先执行 `node.next = cur.next`，再执行 `cur.next = node`，才能保持链表不断链。

![指定位置添加结点](06-img/24-指定位置添加.jpg)

```python
def insert(self, pos, item):
    if pos <= 0:
        self.add(item)
    elif pos >= self.length():
        self.append(item)
    else:
        cur = self.head
        count = 0
        node = SingleNode(item)
        while count < pos - 1:
            cur = cur.next
            count += 1
        node.next = cur.next
        cur.next = node
```

## 六、删除与查找结点

### 1. 删除指定元素

删除需要两个游标：`cur` 指向当前结点，`pre` 指向前一个结点。删除头结点时移动 `head`；删除其他结点时让 `pre.next` 跳过 `cur`。删除完成后立即返回，因此有重复元素时只删除第一个匹配结点；遍历结束仍未找到目标，链表保持不变。

![删除结点](06-img/28-删除节点.jpg)

```python
def remove(self, item):
    cur = self.head
    pre = None
    while cur is not None:
        if cur.item == item:
            if cur == self.head:
                self.head = cur.next
            else:
                pre.next = cur.next
            return
        pre = cur
        cur = cur.next
```

![删除结点时的当前游标与前驱游标](06-img/29-删除节点辅助指针.jpg)

### 2. 查找元素

从 `head` 开始逐个比较，找到后返回 `True`，走到 `None` 仍未找到则返回 `False`。查找时间复杂度为 `O(n)`，额外空间复杂度为 `O(1)`。

```python
def search(self, item):
    cur = self.head
    while cur is not None:
        if cur.item == item:
            return True
        cur = cur.next
    return False
```

## 七、顺序表与链表的比较

![顺序表和链表的复杂度比较](06-img/33-顺序表链表比较.jpg)

| 操作 | 链表 | 顺序表 |
| --- | ---: | ---: |
| 访问元素 | `O(n)` | `O(1)` |
| 头部插入/删除 | `O(1)` | `O(n)` |
| 尾部插入/删除 | `O(n)`（本课单链表从头开始定位尾部） | `O(1)`（插入不发生扩容时） |
| 中间插入/删除 | `O(n)` | `O(n)` |

上表按本课实现分析，并包含定位结点的成本。若已知待操作结点的前驱，链表中间插入或删除只需修改链接，为 `O(1)`；按元素值删除仍需先查找，整体为 `O(n)`。顺序表单次扩容可能搬移全部元素，为 `O(n)`。

链表牺牲了顺序表按下标随机访问的优势，并且每个结点增加链接域，会带来额外空间开销；换来的好处是结点不要求连续存储，插入和删除时不需要整体搬移大量元素。

## 八、课程小结与易错点

### 核心结论

1. 数据结构负责组织数据，算法负责处理数据，二者共同影响程序效率。
2. 顺序表适合按下标随机访问；链表适合不要求连续空间、且需要频繁修改链接关系的场景。
3. 单链表结点由 `item` 和 `next` 组成，`head` 是访问整条链的入口。
4. 链表的长度、遍历、查找都要沿 `next` 逐结点移动，通常是 `O(n)`。
5. 头部插入的两个指针更新顺序、尾部遍历的循环条件、中间插入的两次链接更新，是最容易出错的地方。

### 自查清单

- 空链表时，`head` 是否为 `None`？
- 头插前是否先保存原来的链表入口（`node.next = self.head`）？
- 尾插是否停在最后一个结点，而不是走到 `None`？
- 删除头结点和删除中间结点是否分别处理？
- 指定位置插入时，是否先连接新结点的后继，再连接前驱？
- 遍历、长度和查找循环中是否每轮都执行 `cur = cur.next`？
