## 一、什么是面向对象

### 1. 编程思想

编程思想是人们利用计算机解决实际问题时采用的思维方式，与具体使用哪一种编程语言关系不大。课程先对比两种常见思想：

- **面向过程**：先列出解决问题所需的步骤，再按步骤逐一写代码。思考重点是“先做什么、再做什么”。
- **面向对象**：把现实中的实体抽象为对象，围绕对象组织数据和操作。思考重点从“步骤”转向“操作哪个对象”。

Python 同时支持面向过程和面向对象。面向对象的宗旨是尽可能模拟现实世界：现实中的事物都可以看作对象，对象具有描述自身的属性，也具有能够完成动作的方法。这样，程序中的概念更接近业务中的实体。

例如，“让冰箱工作”可以按面向过程拆成打开冰箱门、放入物品、关闭冰箱门；按面向对象思考，则是找到冰箱对象，再调用它的开门、放入和关门等方法。

### 2. 面向对象的三大特性

| 特性 | 含义 | 课程强调的作用 |
| --- | --- | --- |
| 封装 | 把属性和方法封装在一起，只对外公开使用所需的接口，并隐藏实现细节 | 简化编程、增强安全性 |
| 继承 | 子类继承父类的属性和方法，使子类对象具有父类的特征和行为 | 代码复用；继承关系满足 `is-a` |
| 多态 | 不同类的对象对同一消息作出不同响应 | 解耦合、便于扩展 |

这里的“封装”并不只是把代码放到一个文件，而是把描述对象的数据与操作这些数据的函数结合成一个有机整体；使用者通过规定的访问方式使用对象，不必了解内部细节。继承和多态将在后续面向对象高级内容中继续展开，本篇先掌握它们的基本含义。

## 二、类和对象

### 1. 两个基本概念

- **类（class）**：对现实事物的抽象描述，是创建对象的模板。它规定对象应当具有什么数据和行为。
- **对象（object）**：类的具体实例，是现实事物在程序中的具体体现。

可以把汽车图纸看作类：图纸规定生产出的汽车要有“跑起来”的行为；按照图纸生产出的某一辆具体汽车，就是对象。类是抽象的、看不见摸不着的模板，对象是具体的实体。一个类可以实例化出多个对象。

### 2. 定义类和定义方法

定义类的基本语法是：

```text
class 类名:
    # 定义方法
    方法列表...
```

下面定义一个具有“跑起来”行为的 `Car` 类：

```python
class Car:
    def run(self):
        print("能跑起来了...")
```

类中的函数称为方法。方法的第一个参数通常写成 `self`，它表示调用该方法的对象本身，后文会专门解释。

### 3. 创建对象并调用方法

对象实例化的语法是 `对象名 = 类名()`，调用对象方法的语法是 `对象名.方法名()`：

```python
class Car:
    def run(self):
        print("能跑起来了...")

# 创建对象
car = Car()

# 调用对象的行为
car.run()
```

`Car` 是类，`car` 是由它创建的一个具体对象。调用 `car.run()` 时，Python 会把 `car` 这个实例传给方法的 `self` 参数。

### 4. `self` 关键字

`self` 是 Python 方法中用于指向**对象实例本身**的参数。它不是固定必须叫这个名字，但按约定使用 `self`，这样代码最容易阅读。

同一个类可以创建多个对象，`self` 让方法知道“这一次到底是哪个对象在调用”：

```python
class Car:
    def run(self):
        print(f"self 的值: {self}")
        print("能跑起来了...")

car = Car()
print(f"输出 car 对象名: {car}")
car.run()

my_car = Car()
print(f"输出 my_car 对象名: {my_car}")
my_car.run()
```

两次调用中打印的实例标识不同，说明 `self` 分别指向 `car` 和 `my_car`。在类的内部调用同一个对象的其他方法，也要通过 `self`：

```python
class Car:
    def run(self):
        print("能跑起来了...")

    def work(self):
        # 在类内部调用方法
        self.run()

car = Car()
car.work()
```

课程练习要求定义一个手机类，包含开机、关机和拍照三个方法：

```python
class Phone(object):
    def open(self):
        print("手机能正常开机啦~~~")

    def close(self):
        print("关机, 睡觉啦.")

    def take_photo(self):
        print("我超自恋, 就爱自拍, 哈哈哈.")

phone = Phone()
phone.open()
phone.take_photo()
phone.close()
```

## 三、添加和获取对象属性

### 1. 属性是什么

属性表示对象的固有特征。在 Python 中通常使用变量保存属性，例如人的姓名、年龄、身高和体重。属性属于对象，每个对象可以拥有自己的属性值。

### 2. 在类外部添加和获取属性

类外部可以用 `对象名.属性 = 属性值` 添加或修改属性，用 `对象名.属性` 读取属性：

```python
class Car:
    def run(self):
        print("能跑起来了...")

car = Car()

# 添加对象属性
car.color = "红色"
car.number = 4

# 获取对象属性
print(f"车的颜色: {car.color}")
print(f"轮胎数: {car.number}")
```

这里的 `color` 和 `number` 是添加到 `car` 对象上的属性。另一个由 `Car()` 创建的对象不会自动拥有这两个属性，除非也为它设置。

### 3. 在类内部获取对象属性

在类的方法内部，通过 `self.属性名` 读取当前对象的属性：

```python
class Car:
    def run(self):
        print("能跑起来了...")

    def show(self):
        print(f"车的颜色为: {self.color}")
        print(f"车的轮胎个数为: {self.number}")

car = Car()
car.color = "红色"
car.number = 4

# 属性已经设置后，再调用方法读取
car.show()
```

本节的两个固定写法是：

- 类外部设置属性：`对象名.属性 = 属性值`；
- 类内部获取属性：`self.属性名`。

调用 `show()` 的位置要注意：必须先给 `car.color` 和 `car.number` 赋值，否则方法读取尚未存在的属性会报错。

## 四、魔法方法

### 1. 概念

Python 中有一类特殊方法，总是被双下划线包围，例如 `__init__()`、`__str__()` 和 `__del__()`。它们会在特定情况下由 Python 解释器自动调用，因此称为“魔法方法”，通常不需要开发者手动调用。基本形式是：

```python
def __魔法方法名__(self):
    ...
```

### 2. `__init__()`：创建对象时初始化属性

现实中的人出生时就有姓名、年龄等属性。对应到程序中，可以在创建对象时自动设置初始属性。实例化对象时，Python 会自动触发 `__init__()`。

**无参数初始化**：属性值直接写在类内部，外部不需要传入参数。

```python
class Car:
    def __init__(self):
        self.color = "Black"
        self.number = 3

    def show(self):
        print(f"车的颜色为: {self.color}")
        print(f"车的轮胎个数为: {self.number}")

car = Car()
print(f"类外面访问属性: {car.color}")
car.show()
```

**有参数初始化**：需要由外部决定初始值时，把参数写在 `self` 后面，并在实例化时传入：

```python
class Car:
    def __init__(self, color, number):
        self.color = color
        self.number = number

car = Car("Red", 4)
print(car.number)
```

`Car("Red", 4)` 的过程可以理解为：先创建一个 `Car` 实例，再自动调用 `__init__()`，把实例作为 `self`、`"Red"` 作为 `color`、`4` 作为 `number`，最后把参数保存到对象属性中。

### 3. `__str__()`：打印对象时返回可读信息

如果没有实现 `__str__()`，直接 `print(对象)` 通常会看到对象的默认标识信息。定义 `__str__()` 后，`print()` 会自动调用它，并输出其返回的字符串。

```python
class Car:
    def __init__(self, color, number):
        self.color = color
        self.number = number

    def __str__(self):
        return "车的颜色:" + self.color + " 轮胎个数:" + str(self.number)

car = Car("Red", 4)
print(car)
```

注意：`__str__()` 必须返回字符串；`number` 是整数时，需要先用 `str(self.number)` 转换，才能与其他字符串拼接。

### 4. `__del__()`：对象删除时自动调用

调用 `del 对象` 删除对象，或程序执行结束、对象即将被回收时，Python 解释器会自动调用 `__del__()`。它常用于演示对象被销毁的时机。

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

    def __del__(self):
        print("del 自动调用了 __del__() 方法")

car = Car("BMW")
print(car.brand)
del car

# car 已被删除，继续访问会报错
# print(car.brand)
```

本节三个魔法方法的作用可以这样记：`__init__()` 在创建时初始化，`__str__()` 在打印对象时提供字符串，`__del__()` 在删除对象时执行清理相关逻辑。

## 五、综合案例：减肥

### 1. 需求分析

小明当前体重是 `100 kg`：每跑步一次减少 `0.5 kg`，每大吃大喝一次增加 `2 kg`。需求中只有一个主要事物“学生”，所以可以设计 `Student` 类：

- 属性：当前体重 `current_weight`；
- 方法：`run()` 跑步，`eat()` 大吃大喝；
- 初始化：在 `__init__()` 中把体重设为 `100`。

```python
class Student:
    def __init__(self):
        self.current_weight = 100

    def run(self):
        self.current_weight = self.current_weight - 0.5
        print(f"跑步一次, 减少0.5kg, 当前体重为 {self.current_weight}kg.")

    def eat(self):
        self.current_weight += 2
        print(f"大吃大喝一顿, 增加2kg, 当前体重为: {self.current_weight}kg.")

student = Student()
student.run()
student.eat()
```

这里的关键是：方法通过 `self.current_weight` 读取并更新同一个对象的状态，因此连续调用会保留前一次操作的结果。

## 六、综合案例：烤地瓜

### 1. 需求拆分

“烤地瓜”涉及一个主要事物，因此设计 `SweetPotato` 类。对象需要保存：

1. 地瓜被烤过的总时间 `cook_time`；
2. 当前生熟状态 `cook_state`；
3. 用户添加的调料列表 `condiments`。

对象需要提供三个行为：

- `cook(time)`：累计烤制时间，并依据总时间更新状态；
- `add_condiments(condiment)`：把调料追加到列表；
- `__str__()`：直接打印总时长、状态和调料。

### 2. 课程代码

```python
class SweetPotato:
    def __init__(self):
        self.cook_time = 0       # 烤过的时间
        self.cook_state = '生的'  # 烤的状态
        self.condiments = []      # 调料列表

    def cook(self, time):
        # 先累计总烤制时间
        self.cook_time += time

        # 再根据总时间判断状态
        if 0 <= self.cook_time < 3:
            self.cook_state = '生的'
        elif 3 <= self.cook_time < 5:
            self.cook_state = '半生不熟'
        elif 5 <= self.cook_time < 8:
            self.cook_state = '熟了'
        elif self.cook_time >= 8:
            self.cook_state = '烤糊了'

    def add_condiments(self, condiment):
        self.condiments.append(condiment)

    def __str__(self):
        return (f'这个地瓜被烤过的时间是 {self.cook_time}, '
                f'状态是 {self.cook_state}, 调料有 {self.condiments}')

digua1 = SweetPotato()
print(digua1)
digua1.cook(2)
digua1.add_condiments('辣椒面儿')
print(digua1)
digua1.cook(2)
digua1.add_condiments('酱油')
print(digua1)
```

### 3. 代码阅读要点

- `cook()` 累计的是总时间，所以第二次烘烤要在第一次结果上继续判断；
- 调料使用列表保存，允许多次添加并在 `__str__()` 中整体展示；
- `print(digua1)` 会自动触发 `__str__()`，因此可以直接查看对象状态。

> 资料中的状态表写的是“0–3 分钟生的、3–7 分钟半生不熟、7–12 分钟熟了、超过 12 分钟烤焦”，而示例代码实际使用的是 `0–3`、`3–5`、`5–8`、`>=8`。上面保留了课件代码的判断边界；如果按状态表实现，应同步把 `cook()` 中的条件改成表格区间。
