## 1. 人工智能、机器学习与深度学习

### 人工智能（AI）

人工智能（Artificial Intelligence）研究能够表现出智能行为的计算系统。课程用“仿智”帮助理解：让计算机模拟人类的部分思考、感知或行动能力。

课程从四个方向说明 AI 的期望：像人一样思考、理性地思考、像人一样行动、合理地行动。

### 机器学习（ML）

机器学习（Machine Learning）让计算机从数据中学习规律，不需要程序员把每一种情况都写成明确的判断规则。

以识别汽车为例，人通过经验归纳汽车的特征，再判断一张新图片是不是汽车。机器则通过历史数据训练出模型，再把新的输入交给模型预测。

这里的“自动学习”仍然需要人选择任务、数据和学习算法，关键区别是具体规律由数据学习得到。



### 深度学习（DL）与三者关系

深度学习（Deep Learning）采用多层神经网络学习数据的表示。课程借助“大脑仿生”的比喻，说明神经元与多层网络的联系。

机器学习是实现人工智能的一种途径，深度学习是机器学习中的一类方法。概念范围可以理解为 `DL ⊂ ML ⊂ AI`，不能把三者当作完全相同的概念。

> 整理说明：“大脑仿生”是入门比喻，人工神经网络并不等同于对真实大脑的完整复刻。





## 2. 基于规则与基于模型的学习

### 基于规则的方式

程序员根据经验发现规律，再把规律写成 `if-else`，用这些规则对输入作判断。课程以邮件分类说明这一过程：人先规定判断规则，再据此把邮件分为正常或垃圾邮件。

这要求规则能够明确写出来。图像、语音识别和自然语言处理往往有大量复杂变化，手工规则难以覆盖所有情况。

课件中的大象有真实动物、雕塑和绘画等多种形态。它们外观差异很大，却都需要被识别为大象，这说明“把全部规则写清楚”十分困难。

### 基于模型的方式

先提供历史数据，再由学习算法得到模型。训练结束后，把新输入交给模型，模型输出预测结果。

- **训练阶段**：历史数据用于学习参数。
- **应用阶段**：已经学好的模型用于新数据预测。





### 房价预测：模型与参数

课件用房屋面积和价格建立线性关系，示例数据如下。价格单位为万元。

| 面积（平方米） | 价格（万元） |
| --- | --- |
| 123 | 250 |
| 150 | 320 |
| 87 | 160 |
| 102 | 220 |

将样本画成散点后，可以用一条直线近似它们的关系。模型写成：

$$y=ax+b$$

`x` 表示面积，`y` 表示价格。直线关系是**模型**，斜率 `a` 和截距 `b` 是需要通过训练求出的**参数**。参数确定后，输入新面积即可得到预测价格。

课程用“尽量通过这些点、不能通过的点分布在直线两侧”帮助理解拟合过程，本节先建立直观认识，具体求解在后续线性回归章节展开。



## 3. 应用领域和发展史

### 机器学习的应用领域

| 领域或任务 | 课程中的例子 |
| --- | --- |
| 计算机视觉（CV） | 图像识别、人像、动物与交通工具识别 |
| 自然语言处理（NLP） | 机器翻译、摘要生成、理解交流内容 |
| 用户分析 | 社交网络、影评、商品评论分析 |

这些任务都需要将数据中的信息转化为能够支持判断或预测的规律。

### 数据、算法、算力

数据、算法、算力相互作用，是 AI 发展的三要素。

- **数据**提供学习材料，质量、数量和覆盖范围影响模型能学到什么。
- **算法**规定如何从数据中学习规律。
- **算力**支撑训练与预测所需的计算。

课件对硬件作了入门比较：CPU 负责调度和通用任务，GPU 更适合大量矩阵运算，TPU 是针对神经网络计算设计的处理器。

> 整理说明：课件用“CPU 适合 I/O 密集型、GPU 适合计算密集型”作简化对比。CPU 同样可以处理计算任务，实际选择还取决于任务是否适合并行及数据传输成本。



## 4. 样本、特征和标签

课程以“同学就业薪资表”解释数据的基本组成。

| 同学编号 | 培训学科 | 作业考试 | 学历 | 工作经验 | 工作地点 | 就业薪资 |
| --- | --- | --- | --- | --- | --- | --- |
| 1 | Java | 90 | 本科 | 1 | 北京 | 14k |
| 2 | Java | 80 | 本科 | 1 | 武汉 | 10k |
| 3 | AI | 90 | 本科 | 0 | 北京 | 15k |
| 4 | AI | 92 | 研究生 | 2 | 上海 | 25k |
| 5 | 测试 | 95 | 本科 | 0 | 上海 | 11k |
| 6 | 测试 | 80 | 专科 | 0 | 武汉 | 7k |
| n | AI | 91 | 本科 | 1 | 上海 | 待预测 |

- **样本（sample）**：一行记录是一条样本，多条样本组成数据集（dataset）。
- **特征（feature）**：用于预测的属性信息。此例有培训学科、作业考试、学历、工作经验、工作地点五个特征。
- **标签/目标值（label/target）**：模型需要预测的结果，此例是就业薪资。

同学编号用于标识记录，课程没有把它计入这五个预测特征。

> 从表格结构看，特征对应列。从模型任务看，特征应是对当前预测有用的信息。标签是哪一列，要由任务目标决定。



### 数据集划分

训练集（training set）用于学习模型，测试集（testing set）用于评估模型。课程给出的常用比例是 `8:2` 或 `7:3`，这是常见选择，并不是固定要求。

特征与标签拆分后，用四个变量表示：

| 变量 | 含义 |
| --- | --- |
| `x_train` | 训练集特征 |
| `y_train` | 训练集标签 |
| `x_test` | 测试集特征 |
| `y_test` | 测试集标签 |

训练时使用 `x_train` 和 `y_train`。测试时将 `x_test` 交给模型，再把预测结果与 `y_test` 比较。

薪资未知的新样本用于实际预测，与已经有真实标签、用来评价结果的测试样本作用不同。

> 补充：训练集与测试集应保持独立，不能用测试集标签参与训练，否则测试结果无法反映对新数据的预测能力。



## 5. 机器学习算法分类

### 有监督学习

训练数据包含输入特征和目标值，也就是样本具有标签。课程强调数据标注的作用：学习时让模型知道“这条输入对应什么结果”。

记第 `i` 条样本为 `(x_i, y_i)`，训练数据可写成：

$$D=\{(x_1,y_1),(x_2,y_2),\ldots,(x_n,y_n)\}$$

学习目标是根据这些对应关系得到模型，并预测新样本的目标值。

#### 分类与回归

| 问题 | 目标值 | 课程示例 |
| --- | --- | --- |
| 分类（classification） | 离散类别 | 好瓜/坏瓜，薪资“低/中/高” |
| 回归（regression） | 连续数值 | 根据面积、位置、楼层、朝向预测房价 |

分类可以是二分类或多分类。判断任务类型时要看**目标值的含义**。预测具体薪资属于回归，预测薪资档位属于分类。



### 无监督学习

训练数据没有目标标签，可以表示为：

$$D=\{x_1,x_2,\ldots,x_n\}$$

课程以聚类（clustering）说明无监督学习：根据样本间的相似性将样本分组，发现数据内部结构和相互关系。

聚类得到的组别由数据结构决定，训练前没有给定每个样本应属于哪个类别。它与有监督分类的区别，在于训练数据是否提供目标标签。



### 半监督学习

半监督学习同时利用少量有标签数据与未标记数据，以降低标注成本。课程给出的直观流程是：

1. 专家先标注少量数据，训练出初步模型。
2. 用模型处理未标记的数据。
3. 对照领域专家的分类结果，继续改进模型。

图中还呈现了有标记、未标记数据共同参与学习的方式。本节重点是理解“部分有标签、部分无标签”的数据条件。

> 整理说明：课程这里也借助了询问专家的例子。主动学习强调选择哪些样本让专家标注，半监督学习强调利用有标签和无标签数据，两种思路可以结合。



### 强化学习

强化学习（Reinforcement Learning）让智能体不断与环境交互，通过试错学习行动策略，目标是获得更多**累计奖励**。

| 要素 | 含义 | 小孩学走路的例子 |
| --- | --- | --- |
| 智能体（Agent） | 选择行动的主体 | 小孩 |
| 环境及状态 | 所处环境和当前情况 | 地面及当前站立、行走状态 |
| 行动（Action） | 智能体作出的选择 | 走一步 |
| 奖励（Reward） | 环境对行动的反馈 | 走好几步后获得巧克力 |

小孩每走一步，状态发生变化，完成子任务得到奖励，没有完成则得不到奖励。这个例子帮助理解行动与反馈的循环。

课程列举的应用有 AlphaGo、各类游戏、对抗比赛和无人驾驶。



### 四种方式的比较

| 学习方式 | 训练条件 | 学习目标 |
| --- | --- | --- |
| 有监督学习 | 有特征，也有目标标签 | 预测新样本的类别或数值 |
| 无监督学习 | 有特征，无目标标签 | 发现内部结构，如聚类 |
| 半监督学习 | 部分样本有标签，部分无标签 | 利用两部分数据学习，降低标注成本 |
| 强化学习 | 与动态环境交互并获得奖励 | 学到能够提高累计奖励的策略 |

“有监督、无监督、半监督、强化”描述学习方式。“分类、回归、聚类”描述具体任务，要结合数据条件和目标理解。

## 6. 机器学习建模流程

课程按照以下顺序建立一个机器学习任务：

1. **获取数据**：收集与任务相关的经验数据、图像或文本等。
2. **数据基本处理**：处理缺失值、异常值等问题。
3. **特征工程**：提取特征、预处理特征，必要时选择或降维，形成模型能够使用的表示。
4. **模型训练**：根据任务选择合适算法。课件列举线性回归、逻辑回归、决策树、GBDT。
5. **模型评估**：根据回归、分类或聚类任务选择相应指标。
6. **在线服务与预测**：评估满足需求后，把模型用于新的输入。

效果不理想时，要回到数据、特征或训练步骤继续改进。流程不是只运行一次的直线。

> 课件强调：数据基本处理和特征工程，通常是整个建模过程中最耗时、最耗精力的工作。



### 有监督学习的训练与预测

训练时，先把原始输入转换成特征，再把特征与标签交给学习算法，得到模型。

预测时，对新输入执行相同的特征提取方式，再交给已经训练好的模型，得到预测标签或数值。

两阶段的输入作用不同：标签指导训练，实际预测时需要由模型给出未知的结果。



## 7. 特征工程

特征工程（Feature Engineering）利用专业背景知识和处理技巧，把原始数据转换成更适合当前任务的特征。

课程强调：特征设计可能困难、耗时，需要领域知识。数据和特征影响模型能够达到的效果，选算法只是建模工作的一部分。

### 特征提取

特征提取（feature extraction）从原始数据中抽取与任务相关的信息，形成特征向量。

课程用鸢尾花解释这个过程：不同类别的花有不同形态，可以测量花萼长度、花萼宽度、花瓣长度、花瓣宽度，组织成四个特征。每朵花是一条样本，花的类别是标签。

这些测量值把外观信息转成模型能够处理的数值。对于文本和图片，也需要先形成可用于学习的特征表示。





### 特征预处理

特征预处理（feature preprocessing）处理特征的尺度、单位等问题。课程的图例中，不同特征原本分别落在 `60—90`、`2—4`、`10—15`、`40—46` 等范围，转换后落在可比较的数值范围内。

某些算法会受到量纲影响，数值大的特征可能影响更强。标准化、归一化等方法用于处理这种尺度差异。

> 整理说明：缩放处理的是数值尺度，并不意味着每个特征对结果应具有完全相同的重要性。具体预处理方法在后续课程中展开。



### 特征降维

特征降维降低特征表示的维度。课程用三维地球与二维地图帮助理解“降低维度”的直观含义。

降维会改变数据的表示，也可能丢失部分信息，因此要尽量保留任务所需的主要信息。新维度未必与原始字段一一对应。

### 特征选择

特征选择从已有特征中挑选重要子集。图例把 `f1` 至 `f7` 选择为 `f1`、`f2`、`f3`，选中的字段仍保持原有含义。

**降维与选择的区别**：降维可以变换原有表示，特征选择保留已有字段的子集。两者都可能减少输入特征数量。





### 特征组合

特征组合（feature crosses）通过加法、乘法等运算构造新特征。教师笔记给出的例子是：

- `A × B`：将两个特征值相乘。
- `A × B × C × D × E`：组合多个特征。
- `A × A`：对单个特征求平方。

组合的目的，是让输入表达任务需要的关系。构造哪些组合，需要结合领域知识和模型效果判断。



## 8. 拟合、欠拟合与过拟合

拟合（fitting）描述模型对样本分布的模拟情况。判断拟合问题时，要同时观察训练数据和未参与训练的数据。

| 状态 | 训练集表现 | 测试集表现 | 课程列举的原因 |
| --- | --- | --- | --- |
| 欠拟合（under-fitting） | 差 | 差 | 模型过于简单，学习到的信息不足 |
| 过拟合（over-fitting） | 好 | 差 | 模型太复杂、数据不纯、训练数据太少 |

课件的直观图中，简单直线没能表达样本趋势，适度复杂的曲线较好地模拟趋势，过于曲折的曲线则过分追随训练样本细节。



### 天鹅识别案例

课程用两个例子说明拟合问题。

- **欠拟合**：只学到“有翅膀、嘴巴长”等粗略特征，把其他鸟类也识别成天鹅，说明模型没有学到足够的区分信息。
- **过拟合**：训练图片全是白天鹅，模型把“白色”也当成判断依据，遇到黑天鹅便无法正确识别。

第二个例子还提醒我们检查训练数据的覆盖范围：模型学到的规律受它看到的样本限制。



### 泛化能力与模型复杂度

泛化（Generalization）是模型对新数据的预测能力。训练集成绩好，只说明模型适应了训练数据，还要通过独立数据判断它能否泛化。

教师笔记的示意图中，训练效果持续提高，但测试效果可能在达到合适位置后下降。这帮助理解为什么要同时观察两部分数据的表现。

课程给出的奥卡姆剃刀原则是：当两个模型具有相同的泛化误差时，倾向于选择更简单的模型。

课件练习列举的改进方向包括：

- 欠拟合时，增加有用特征或提高模型表达能力。
- 过拟合时，考虑正则化、异常值处理、特征降维等方法。

> 整理说明：这些是根据原因选择的改进方向，并非任何任务都能用同一种方法解决。过拟合的关键是对训练数据的过度适应，不应简单理解为“特征数量多”。
