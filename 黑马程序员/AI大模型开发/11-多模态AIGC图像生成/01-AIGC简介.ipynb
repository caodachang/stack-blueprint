# 多模态 AIGC 图像生成 01：AIGC 简介

本篇对应讲义《01-AIGC.pdf》，按原资料整理三个问题：

1. AIGC 是什么，以及从技术和生成内容两个角度如何分类；
2. AIGC 在影视、电商、娱乐、教育和工业等场景怎样落地；
3. AIGC 产品如何分为基础层、中间层和应用层。

讲义中的宣传页已省略。

## 一、AIGC 是什么

AIGC 是 **AI-Generated Content** 的缩写，中文可理解为“人工智能内容生成”。它指利用人工智能模型学习数据中的模式和抽象概念，再生成新的内容。这里的“内容”不局限于文字，也可以是图像、音频、视频，或多种模态的组合。

因此，AIGC 的核心关注点是“由 AI 生成内容”，而不是只把 AI 当作检索或分类工具。

## 二、按技术路径划分

讲义把 AIGC 相关技术分为三类：

| 类型 | 主要含义 | 讲义中的例子 |
| --- | --- | --- |
| 内容孪生 | 把数字内容映射到另一个模态，或对内容做智能增强、智能转译 | 图像超分、语音转字幕、文字转语音 |
| 内容编辑 | 理解内容及其属性，并按控制条件修改已有内容 | 场景剪辑、虚拟试衣、人声分离 |
| 内容生成 | 从海量数据中学习抽象概念，生成全新的内容 | AI 绘画、AI 写作、视频生成、多模态生成 |

三者的区别在于处理对象和结果：内容孪生强调跨模态映射，内容编辑强调对已有内容进行可控修改，内容生成则强调产出新的内容。

## 三、按生成内容划分

从生成结果看，AIGC 可以分为以下五类：

### 1. 文本生成

基于自然语言处理的文本生成包括非交互式和交互式两种形式，常见任务有摘要、标题生成、聊天机器人和图生文。讲义列举的代表性产品包括 ChatGPT、文心一言。

### 2. 图像生成

图像生成包括编辑修改和自主生成，应用覆盖超分、修复、人脸替换、文生图等方向。讲义列举 Midjourney、DALL·E、文心一格作为代表性产品。

### 3. 音频生成

音频生成可用于语音克隆和场景化语音生成等 C 端产品，讲义列举 DeepMusic、WaveNet。

### 4. 视频生成

视频生成包括视频编辑和自主生成，涉及超分、修复、剪辑等任务，讲义列举 Deepfake、VideoGPT。

### 5. 多模态生成

多模态生成把不同模态组合起来进行转换，例如文本生成图像、文本生成音频、文本生成视频、图像生成文本和图像生成视频。讲义列举 DALL·E、Midjourney、Stable Diffusion。

## 四、AIGC 的行业应用

### 1. 影视行业

AIGC 在影视行业覆盖创作、拍摄和后期三个阶段：

- **前期创作**：学习大量剧本，按预定风格生成剧本，为创作者提供更多灵感和选择；创作者再筛选、加工和完善内容，以缩短创作周期。
- **中期拍摄**：合成虚拟场景，用于无法实拍或实拍成本过高的环境。拍摄前收集场景素材并建模，演员在绿棚中拍摄，再通过人员识别和抠图把人物融入虚拟场景。
- **后期制作**：进行画质增强、人脸替换、人声替换和自动剪辑，节省预告片、片段集锦等制作时间和成本。

### 2. 电商行业

电商场景主要体现在三方面：

- **商品展示**：生成商品 3D 模型，用于商品展示和虚拟试用，让消费者更直观了解外形、尺寸和效果，提高购买决策的准确性。
- **主播打造**：打造虚拟主播，进行自动化直播，提升效率并降低成本。
- **交易场景**：构建虚拟商城，并用智能聊天机器人赋能线上线下秀场，为消费者提供新的购物场景。

### 3. 娱乐行业

娱乐行业中的图像内容生成可以用于人脸美妆、图像融合、黑白上色、风格转换和属性变换。

在社交互动方面，虚拟主播、虚拟网红、聊天机器人和互动游戏等形式不断丰富社交体验，也为社交互动市场提供新的产品形态。

### 4. 教育、工业等行业

- **教育**：生成教学素材，例如用数字人生成历史人物对话，提升课堂互动；也可以用 ChatGPT 生成创意教学方案，拓宽授课思路。
- **工业**：AIGC 与 CAD 等工业设计软件结合，通过文字提示生成特定机构模型，作为工业设计参考。例如输入“设计一款卫星太阳能电池板可伸缩折翼机构”，生成相应的 3D 设计机构。

讲义对这些场景的共同判断是：AIGC 将提升内容创作者、设计师、工程师和教育工作者的效率与质量，并推动企业数字化、智能化进程。

## 五、AIGC 产品的三层形态

讲义从产品和商业形态出发，把 AIGC 产品分成基础层、中间层和应用层：

| 层级 | 定位 | 能力与产品形态 |
| --- | --- | --- |
| 基础层（模型服务） | 由少数头部企业或研发机构主导的基础设施 | 基于预训练大模型提供能力，常见形态是通过 API 收取调用费，或对基础设施封装后提供专业软件平台 |
| 中间层（2B） | 面向行业和企业的场景化能力 | 通常没有从零开发大模型的能力，而是基于开源大模型进行改进、抽取或二次开发，形成场景化、垂直化、定制化的模型或工具 |
| 应用层（2C） | 面向终端用户的场景化工具或软件 | 基于基础层和中间层开发，关注具体用户需求，通过网页、小程序、群聊、App 等载体提供功能 |

## 六、本篇要点

1. AIGC 的全称是 AI-Generated Content，生成对象包括文本、图像、音频、视频和多模态内容。
2. 按技术路径可以区分内容孪生、内容编辑和内容生成。
3. AIGC 已经覆盖影视、电商、娱乐、教育和工业等行业，落地方式取决于行业的内容生产和交互流程。
4. 产品形态可以按基础层、中间层、应用层理解：基础层提供模型能力，中间层做行业化改造，应用层直接服务终端用户。
