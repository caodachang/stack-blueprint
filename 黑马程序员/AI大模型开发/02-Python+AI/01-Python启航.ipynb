## 环境准备：解释器与开发工具

计算机最终执行的是二进制指令。Python 代码需要由 **Python 解释器** 翻译并执行；解释器负责“让代码跑起来”，开发工具负责编辑、分析和调试代码，两者是不同的东西。

手写文件再切换命令行可以完成开发，但课程指出这种方式缺少提示、效率低且容易出错。IDE（集成开发环境）把代码编写、分析、调试等功能整合到一个软件中，首章选用 PyCharm。

### 安装 Python

1. 打开 [Python 官网](https://www.python.org/)，下载 Windows 安装包。公开安装讲义建议使用 `3.13.x`，因为后续课程依赖可能尚未为最新版本准备好；课件里的 `3.13.9`、`3.14.0` 截图只是操作示例，不要求照抄截图版本。
2. 安装界面勾选 `Add python.exe to PATH`，再选择 `Customize installation`。
3. 特性选择页按课程说明直接点 `Next`。
4. 自定义安装路径时，课程建议开发软件放在不含中文、空格较少的目录，例如 `D:\develop\python`；这是便于课程环境统一的习惯。
5. 点击 `Install`，结束后关闭安装程序。



### 验证解释器是否可用

安装后按 `Win + R`，输入 `cmd` 并回车打开命令提示符，然后输入：

```bash
python
```

如果进入 Python 交互提示符并显示版本号，说明命令已经找到解释器。退出交互模式可输入 `exit()`。还可以用下面的命令核对版本、命令来源和实际解释器路径：

```bash
python --version
where.exe python
python -c "import sys; print(sys.executable)"
```

安装后如果刚修改过 PATH，要关闭旧 CMD，再打开一个新的终端验证。

### 常见问题：PATH 找不到 Python

在 CMD 中输入 `python` 时，系统先检查当前目录；找不到时，再按 PATH 环境变量列出的目录搜索可执行文件。因此会出现两类现象：

- 提示“`python` 不是内部或外部命令”，通常是 Python 所在目录没有加入 PATH。
- 运行 `python` 没有进入解释器，或跳转 Microsoft Store，通常是 PATH 顺序、应用执行别名或安装选项没有配置好。

课程给出的处理思路是：在 Windows 的“系统→高级系统设置→环境变量”中编辑 `Path`，将 Python 安装目录和 `Scripts` 目录加入列表。例如：

```text
D:\develop\python\
D:\develop\python\Scripts\
```

用户变量只对当前登录用户有效，系统变量对所有用户有效。修改后重新打开 CMD，再用 `where.exe python` 和 `python --version` 检查。



## 第一个 Python 程序：交互模式与脚本

进入交互提示符后，可以立即输入一行代码，适合快速试验：

```python
print("Hello World")
print("Hello Python")
```

如果要一次执行多行代码，把内容保存为 `hello.py`，再在保存文件所在目录运行：

```bash
python hello.py
```

资料中的 `hello.py` 原例是三行打印，输出一个边框和标题：

```python
print("##############################")
print("####### Hello Python  #########")
print("##############################")
```

这里的 `.py` 是 Python 源文件后缀。代码中的括号、引号等符号使用英文半角；字符串内容可以写中文。

## PyCharm：创建第一个项目

PyCharm 是 JetBrains 的 Python IDE。课程给出的下载地址是 [JetBrains PyCharm](https://www.jetbrains.com/pycharm/download/?section=windows)，公开安装讲义的基本流程是双击安装包、选择安装路径、按需要创建桌面快捷方式，然后完成安装。首次打开后可以按个人习惯选择主题和字体。

创建项目时重点确认三项：

1. **项目位置**：选择项目的磁盘路径并确定项目名。
2. **解释器类型**：首章示例使用“项目 `venv`”，让 PyCharm 在项目目录中创建 `.venv`。
3. **Python 版本**：选择已经安装并通过命令行验证的解释器。



### 在项目中创建 Python 文件

在项目窗口中右键项目目录，选择 `New → Python File`，新建 `01.快速入门.py`。



文件中输入：

```python
# 输出Hello World
print("Hello World")
print("Hello Python")
```

运行文件后，运行窗口应看到两行文本。`#` 开头的是注释，只给开发人员阅读，Python 解释器会忽略它；在 PyCharm 中可以使用 `Ctrl + /` 切换当前行注释。

## 项目目录与基础配置

课程示例的项目目录包含：



- `.idea`：PyCharm 保存的项目配置。
