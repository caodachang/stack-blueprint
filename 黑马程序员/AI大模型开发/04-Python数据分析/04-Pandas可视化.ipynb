## 一、为什么要做数据可视化

Matplotlib 是 Python 中常用的 2D/3D 数据可视化库，支持以渐进、交互式的方式绘制图表。

可视化的作用：

- 用图形直观呈现数据和变化趋势；
- 帮助发现异常点、分布形态和变量之间的关系；
- 让分析结论更容易被比较和沟通。

选图时先看数据的业务问题：看趋势通常用折线图，比较类别用柱状图，查看分布用直方图，观察占比用饼图，研究两个数值变量的关系用散点图。

## 二、Matplotlib 绘图流程

### 2.1 导入 `pyplot`

```python
import matplotlib.pyplot as plt
```

`matplotlib.pyplot` 提供了一组接近 MATLAB 风格的绘图函数。常见绘图过程可以固定为三步：

1. `plt.figure()` 创建画布；
2. `plt.plot()` 或其他绘图函数绘制数据；
3. `plt.show()` 显示图像。

### 2.2 绘制第一个折线图

折线图适合表达随时间或有序变量变化的趋势。`figsize` 设置画布尺寸，`dpi` 设置清晰度。

```python
import matplotlib.pyplot as plt

# 一周的温度
x = [1, 2, 3, 4, 5, 6, 7]
y = [17, 17, 18, 15, 11, 11, 13]

plt.figure(figsize=(10, 6), dpi=100)
plt.plot(x, y)
plt.show()
```

如果要保存图片，应在 `show()` 之前调用 `savefig()`，因为显示图像后当前 figure 资源可能被释放。

## 三、完善折线图

课程以某城市 11 点到 12 点每分钟的温度变化为例，演示坐标刻度、中文字体、网格、标题和保存图片。

### 3.1 准备连续数据

```python
import random

x = range(60)
y_shanghai = [random.uniform(15, 18) for _ in x]

plt.figure(figsize=(20, 8), dpi=80)
plt.plot(x, y_shanghai)
```

`random.uniform(a, b)` 生成 `[a, b]` 区间内的随机浮点数，适合制作演示数据。正式分析应使用真实观测值。

### 3.2 自定义 x、y 轴刻度

`xticks` 和 `yticks` 可以把默认数字替换成更有业务含义的标签。刻度太密时可使用切片减少显示数量。

```python
x_ticks_label = [f"11点{i}分" for i in x]
y_ticks = range(40)

plt.xticks(x[::5], x_ticks_label[::5])
plt.yticks(y_ticks[::5])
```

刻度值和刻度标签的数量需要一一对应，否则会出现警告或显示错位。

### 3.3 解决中文和负号显示

如果系统没有配置中文字体，标题、图例和坐标轴标签可能显示为方框。可以在脚本中动态设置字体：

```python
from pylab import mpl

mpl.rcParams["font.sans-serif"] = ["SimHei"]
mpl.rcParams["axes.unicode_minus"] = False
```

也可以安装支持中文的字体并修改 Matplotlib 配置文件。字体名称应以本机实际安装情况为准。

### 3.4 网格、标签、标题和保存

```python
plt.grid(True, linestyle="--", alpha=0.5)
plt.xlabel("时间")
plt.ylabel("温度")
plt.title("11点到12点的温度变化", fontsize=20)

# 在显示前保存
plt.savefig("./temperature.png")
plt.show()
```

网格线用于帮助读数；标题和坐标轴名称要说明数据的对象、单位和时间范围。

### 3.5 一张图绘制多条折线

在同一个坐标系中多次调用 `plot` 即可比较多组趋势，使用颜色、线型和图例区分它们。

```python
y_beijing = [random.uniform(1, 3) for _ in x]

plt.plot(x, y_shanghai, label="上海")
plt.plot(
    x, y_beijing,
    color="r",
    linestyle="--",
    label="北京",
)
plt.legend(loc="best")
```

常见颜色字符包括 `r`、`g`、`b`、`k`；常见线型包括 `-` 实线、`--` 虚线、`-.` 点划线和 `:` 点虚线。也可以直接使用完整颜色名称。

## 四、多坐标系与面向对象绘图

### 4.1 `plt.subplots`

当两组数据量纲差异很大，或希望在同一画布中分别展示，可以使用多个坐标系。`plt.subplots` 返回画布对象 `fig` 和坐标系对象 `axes`。

```python
fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(20, 8),
    dpi=100,
)

axes[0].plot(x, y_shanghai, label="上海")
axes[1].plot(x, y_beijing, color="r", label="北京")

axes[0].set_title("上海温度")
axes[1].set_title("北京温度")
axes[0].set_xlabel("时间")
axes[1].set_xlabel("时间")
axes[0].set_ylabel("温度")
axes[1].set_ylabel("温度")
axes[0].legend()
axes[1].legend()

plt.tight_layout()
plt.show()
```

`plt.xlabel()` 是面向过程的写法；`axes.set_xlabel()` 是面向对象的写法。在多个子图中推荐使用 `axes`，因为每个坐标系的设置更清晰。

### 4.2 面向对象 API 对照

| 面向过程 | 坐标系对象方法 |
| --- | --- |
| `plt.plot` | `ax.plot` |
| `plt.xticks` | `ax.set_xticks` / `ax.set_xticklabels` |
| `plt.yticks` | `ax.set_yticks` |
| `plt.xlabel` | `ax.set_xlabel` |
| `plt.ylabel` | `ax.set_ylabel` |
| `plt.title` | `ax.set_title` |
| `plt.grid` | `ax.grid` |
| `plt.legend` | `ax.legend` |

当图表结构变复杂时，明确保存 `fig` 和 `ax`，有助于控制每个子图。

### 4.3 双坐标轴

## 五、常见图形及适用场景

| 图形 | API | 主要用途 |
| --- | --- | --- |
| 折线图 | `plt.plot` | 连续数据的变化趋势 |
| 柱状图 | `plt.bar` | 离散类别之间的数值比较 |
| 直方图 | `plt.hist` | 连续数值的分布情况 |
| 饼图 | `plt.pie` | 分类数据的占比 |
| 散点图 | `plt.scatter` | 两个数值变量的关系、聚集和离群点 |

图形选择要服从问题本身，避免为了“好看”而使用不适合数据结构的图。

### 5.1 柱状图

柱状图用柱子的高度比较离散类别。`width` 控制柱宽，`align` 控制对齐方式。

```python
categories = ["A", "B", "C", "D"]
values = [3, 7, 5, 4]

plt.figure(figsize=(8, 5))
plt.bar(categories, values, color="steelblue", width=0.6)
plt.title("类别数量比较")
plt.xlabel("类别")
plt.ylabel("数量")
plt.show()
```

### 5.2 直方图

直方图先把连续数值划分为若干区间，再统计每个区间中的样本数量。`bins` 影响分组数量和观察到的分布形状。

```python
import numpy as np

data = np.random.randn(500)

plt.figure(figsize=(10, 6))
plt.hist(
    data,
    bins=30,
    color="steelblue",
    alpha=0.7,
    rwidth=0.85,
)
plt.title("数据分布")
plt.xlabel("数值")
plt.ylabel("频数")
plt.grid(True, alpha=0.3)
plt.show()
```

柱状图强调类别对比；直方图强调连续变量的分布，两者不要混用。

### 5.3 饼图

饼图适合类别数量较少且总和有明确意义的占比场景。

```python
sizes = [25, 35, 25, 15]
labels = ["A", "B", "C", "D"]

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
)
plt.title("类别占比")
plt.show()
```

类别太多或各部分数值接近时，饼图不利于精确比较，可以改用排序后的柱状图。

### 5.4 散点图

散点图用一组 x、y 坐标点展示两个变量的关系，可以观察相关趋势、聚类和离群点。

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.scatter(x, y, color="red")
plt.title("两个变量的关系")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 六、数学函数与可视化

`plot` 不只用于业务折线图，也可以绘制数学函数。`np.linspace` 能在指定区间生成等间隔的 x 值。

```python
import numpy as np

x = np.linspace(-10, 10, 1000)
y = np.sin(x)

plt.figure(figsize=(12, 5))
plt.plot(x, y)
plt.grid(True)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("正弦函数")
plt.show()
```

## 七、链家数据集可视化案例思路

课程前 3 个视频使用链家数据集练习多个需求。具体需求的共同方法是：

1. 读取并检查数据字段；
2. 处理缺失值、异常值和不合理记录；
3. 根据需求选择城市、区域、户型、面积、总价等字段；
4. 用 `groupby` 或 `value_counts` 汇总；
5. 用合适图形展示比较结果；
6. 添加标题、单位、图例和必要的刻度标签。

示例流程：

```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取链家数据；文件名以本地资料为准
house = pd.read_csv("./data/lianjia.csv")

# 查看结构和缺失情况
print(house.head())
print(house.info())
print(house.isna().sum())

# 按区域统计房源数量
area_count = house["区域"].value_counts().head(10)

# 绘制前十个区域的房源数量
area_count.sort_values().plot(kind="barh")
plt.title("各区域房源数量（前十）")
plt.xlabel("房源数量")
plt.ylabel("区域")
plt.tight_layout()
plt.show()
```

字段名称和数据文件路径应以实际资料为准，先用 `house.columns` 确认列名，再编写筛选和分组表达式。

### 7.1 用 Pandas 快速绘图

Pandas 的 Series 和 DataFrame 提供了 `.plot()` 接口，本质上仍然调用 Matplotlib。它适合快速探索，复杂图表再切换到 `plt` 或 `axes` API。

```python
# Series 快速绘图
house["总价"].plot(kind="hist", bins=30)

# 分组后绘制柱状图
price_by_area = house.groupby("区域")["总价"].mean()
price_by_area.sort_values(ascending=False).head(10).plot(
    kind="bar",
    figsize=(10, 5),
)
plt.ylabel("平均总价")
plt.title("平均总价最高的区域")
plt.tight_layout()
plt.show()
```

### 7.2 可视化前的数据检查

- 数值列先确认单位，例如面积是平方米还是平方英尺；
- 房价、面积等字段要处理无法转换的字符串；
- 长尾分布可以先查看分位数，必要时限制展示范围；
- 分类字段过多时只展示前若干类别，并说明筛选规则；
- 图表标题、坐标轴和图例必须能独立说明数据含义。

```python
# 将可能包含字符串的字段转换为数值，无法转换的值变成 NaN
house["面积"] = pd.to_numeric(house["面积"], errors="coerce")
house["总价"] = pd.to_numeric(house["总价"], errors="coerce")

# 删除关键字段缺失的数据
house = house.dropna(subset=["区域", "面积", "总价"])
```

## 八、绘图质量与常见问题

### `plt.show()` 后保存图片为空

把 `plt.savefig()` 放在 `plt.show()` 之前，或显式保存 figure 对象。

### 中文显示为方框

设置支持中文的字体，并将 `axes.unicode_minus` 设置为 `False`，避免负号显示异常。

### 刻度标签重叠

减少刻度数量、旋转标签或增大画布；分类太多时只展示 Top N。

```python
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
```

### 图表信息不完整

补充标题、坐标轴名称、单位、图例和数据时间范围。图表应该让读者不依赖代码也能理解。

### 多图布局拥挤

调整 `figsize`，并使用 `plt.tight_layout()`；面向对象绘图时也可以使用 `fig.subplots_adjust()` 精细控制边距。
