### 3. 通过命令管理环境

在 Anaconda Prompt 或已经配置 `conda` 的终端中执行。`虚拟环境名字` 与 `python版本` 是占位文字，需要按实际环境替换。

```bash
# 创建指定 Python 版本的环境
conda create -n 虚拟环境名字 python=python版本

# 进入环境
conda activate 虚拟环境名字

# 退出当前环境
conda deactivate

# 删除指定环境及其中的包
conda remove -n 虚拟环境名字 --all
```

## 七、NumPy 简介

NumPy（Numerical Python）重在数值计算。它提供 N 维数组对象 `ndarray`，适合存储和处理大量数值数据，许多科学计算工具把它作为基础。

课程强调的能力包括：

- 高性能科学计算与数据分析。
- 用 `ndarray` 表示多维数组，完成矢量运算。
- 用数组操作表达计算，减少自己编写 Python 循环。
- 提供读写磁盘数据和操作内存映射文件的工具。

NumPy 的底层大量使用 C 等编译语言实现，数组运算通常比逐个操作 Python 嵌套列表更高效。广播也是它的核心能力之一，本篇课件主要演示数组创建、内置函数与乘法。

课程统一使用别名 `np`：

```python
import numpy as np
```

后续示例都沿用这个导入。`ndarray` 的下标从 `0` 开始，本篇数值数组中的元素由同一个 `dtype` 描述。

## 八、ndarray 的属性

拿到数组后，先确认它有几维、每个维度多大、共有多少元素，以及元素的数据类型。课程介绍以下属性：

| 属性 | 含义 | 读数组时要看什么 |
| --- | --- | --- |
| `ndim` | 维度数，也称轴的数量 | 一维、二维还是更多维 |
| `shape` | 各个维度大小组成的元组 | 二维数组的行数和列数 |
| `size` | 元素总个数 | 各维大小的乘积 |
| `dtype` | 元素的数据类型 | 如 `int32`、`float64` |
| `itemsize` | 每个元素占用的字节数 | 注意单位是字节 |

例如，二维数组有 `n` 行、`m` 列，`shape` 就是 `(n, m)`；`shape` 元组有两个成员，所以 `ndim` 为 `2`。讲义中的 `(2, 3)` 是两行三列的具体例子。

课件用下面的数组检查属性：

```python
import numpy as np

a = np.arange(15).reshape(3, 5)
print("数组的维度:", a.shape)
print("数组轴的个数:", a.ndim)
print("数组元素类型:", a.dtype)
print("数组中每个元素的字节大小:", a.itemsize)
print("数组元素的总个数:", a.size)
print("类型查询:", type(a))

# 创建一个数组
b = np.array([6, 7, 8])
print("数组b:", b)
print("数组b类型:", type(b))
```

`np.arange(15)` 创建 15 个元素，`reshape(3, 5)` 把它们组织为 3 行 5 列，元素数量保持不变。`reshape` 的目标形状必须容纳相同数量的元素。

课件第一条打印文字把 `shape` 称为“维度”，实际打印的是**形状**；维度数要看 `ndim`。还要区分 `type(a)` 与 `a.dtype`：前者说明变量是数组对象，后者说明数组元素是什么类型。默认整数位数可能随 NumPy 版本和平台变化，应读取实际 `dtype`。

## 九、创建 ndarray

### 1. 从列表创建：array

`np.array()` 把现有数据转换为 NumPy 数组。下面是课件的一维整数数组和浮点数组示例：

```python
import numpy as np

a = np.array([2, 3, 4])
print("数组a元素类型:", a)
print("数组a类型:", a.dtype)

b = np.array([1.2, 3.5, 5.1])
print("数组b元素类型:", b.dtype)
```

示例中 `a` 的元素是整数，`b` 的元素是浮点数。第一条 `print` 实际打印 `a` 的内容，虽然课件中的提示文字写了“元素类型”，识别类型仍应看后面的 `dtype`。

### 2. 按步长创建：arange

`np.arange()` 类似 Python 的 `range()`，创建一维数组。常用参数顺序是起点、终点、步长，还可以指定 `dtype`。

```python
np_arange = np.arange(10, 20, 5, dtype=int)
print("arange创建np_arange:", np_arange)
print("arange创建np_arange的元素类型:", np_arange.dtype)
print("arange创建np_arange的类型:", type(np_arange))
```

这个例子从 `10` 开始，每次增加 `5`，到 `20` 前停止。**终点不包含在范围内**，因此这里表示的两个数是 `10` 和 `15`。

### 3. 创建随机数矩阵

课程通过随机数生成 3 行 4 列的数组，用来说明形状、取值范围和分布。

```python
import numpy as np

# 生成 3 行 4 列的随机浮点数，rand 的取值范围为 [0.0, 1.0)
arr = np.random.rand(3, 4)
print(arr)
print(type(arr))

# 生成 3 行 4 列的随机整数，取值范围为 [-1, 5)
arr = np.random.randint(-1, 5, size=(3, 4))
print(arr)
print(type(arr))

# 生成 3 行 4 列的随机浮点数，来自指定区间的均匀分布
arr = np.random.uniform(-1, 5, size=(3, 4))
print(arr)
print(type(arr))
```

| 调用 | 数据与范围 | 形状参数 |
| --- | --- | --- |
| `np.random.rand(3, 4)` | `[0, 1)` 内的均匀分布浮点数 | 多个位置参数 |
| `np.random.randint(-1, 5, size=(3, 4))` | `-1` 至 `4` 的整数 | `size` 元组 |
| `np.random.uniform(-1, 5, size=(3, 4))` | 指定区间内的均匀分布浮点数 | `size` 元组 |

随机数示例的每次结果可以不同，学习时主要检查数组形状、类型和取值范围，不把某次随机结果当成固定答案。

### 4. 指定与转换数据类型

创建数组时用 `dtype` 指定类型，创建后用 `astype()` 转换类型。类型名通常包含类型和位数，例如 `float64`、`int32`。

课件先创建全为零的浮点数组，再把它转换成整数数组：

```python
# 初始化 3 行 4 列数组，数据类型为 float64
zeros_float_arr = np.zeros((3, 4), dtype=np.float64)
print(zeros_float_arr)
print(zeros_float_arr.dtype)

# astype 转换数据类型，把已有数组的元素类型转换为 int32
zeros_int_arr = zeros_float_arr.astype(np.int32)
print(zeros_int_arr)
print(zeros_int_arr.dtype)
```

`np.zeros()` 的第一个参数是形状元组。此处转换前后都为 3 行 4 列，变化的是元素类型。

`astype()` 默认返回转换后的数组，所以课件用另一个变量接收结果。浮点数转整数时可能丢失小数部分，不能把类型转换当作四舍五入。

### 5. 等比数列：logspace

`np.logspace(start, stop, num)` 创建在对数尺度上等间隔的数列。默认底数是 `10`，`start` 和 `stop` 表示的是**指数**，对应区间端点为 `10**start` 和 `10**stop`。

课件先让开始指数和结束指数都为 `0`：

```python
a = np.logspace(0, 0, 10)
a
```

两端都对应 `10**0`，所以 10 个元素都为 `1`。再把结束指数改成 `9`：

```python
a = np.logspace(0, 9, 10)
a
```

这 10 个点的指数依次为 `0` 到 `9`，对应 `10**0`、`10**1`，一直到 `10**9`。

通过 `base` 改变底数：

```python
a = np.logspace(0, 9, 10, base=2)
a
```

这里对应 `2**0` 到 `2**9`。底数、指数区间和元素个数共同决定数列，不能把 `start`、`stop` 直接理解为数组的数值端点。

### 6. 等差数列：linspace

`np.linspace(start, stop, num)` 按指定的元素个数创建等差数列。课程首先演示从 `1` 到 `10` 取 10 个点：

```python
a = np.linspace(1, 10, 10)
a
a.dtype
```

默认包含终止值。本例的 10 个点对应 `1` 到 `10`，默认元素类型为浮点型。

`endpoint` 控制是否包含终止值，默认是 `True`：

```python
a = np.linspace(1, 10, 10, endpoint=False)
a
```

`endpoint=False` 仍创建 10 个点，但不包含 `10`，本例相邻点的间隔变为 `0.9`。

| 创建方式 | 第三个常用参数 | 终点规则 |
| --- | --- | --- |
| `arange(start, stop, step)` | 步长 `step` | 不包含终点 |
| `linspace(start, stop, num)` | 元素个数 `num` | 由 `endpoint` 控制，默认包含 |
| `logspace(start, stop, num)` | 元素个数 `num` | 起点和终点表示指数，默认包含指数区间端点 |

## 十、NumPy 内置函数：基本函数

这些函数可以处理单个数，也可以对数组中的元素进行运算。对数组调用时，通常直接写一次函数调用就能处理全部元素。

| 函数 | 课程中的用途 |
| --- | --- |
| `np.ceil()` | 向上取整 |
| `np.floor()` | 向下取整 |
| `np.rint()` | 取最接近的整数 |
| `np.isnan()` | 判断元素是否为 `NaN`（Not a Number） |
| `np.multiply()` | 对应元素相乘 |
| `np.divide()` | 对应元素相除 |
| `np.abs()` | 计算元素的绝对值 |
| `np.where(condition, x, y)` | 条件成立选 `x`，否则选 `y` |

课件使用 `randn()` 生成标准正态分布的随机数，再逐项演示函数：

```python
# randn 返回标准正态分布的随机数，形状为 2 行 3 列
arr = np.random.randn(2, 3)
print(arr)
print(np.ceil(arr))
print(np.floor(arr))
print(np.rint(arr))
print(np.isnan(arr))
print(np.multiply(arr, arr))
print(np.divide(arr, arr))
print(np.where(arr > 0, 1, -1))
```

最后一个表达式为每个元素作选择：大于 `0` 时选 `1`，其余情况选 `-1`。`np.isnan()` 则返回对应位置的布尔判断结果。

补充两个直接影响示例理解的细节：

- 讲义把 `np.rint()` 简称为“四舍五入”。它在恰好位于两个整数中间时取偶数，例如 `0.5` 对应 `0`，`1.5` 对应 `2`。
- `np.divide(arr, arr)` 是逐元素相除。非零数除以自身为 `1`，如果数据中有 `0`，对应位置会出现无效运算，因此不能把它当作任意数组都生成全 `1` 的方法。

## 十一、NumPy 内置函数：统计函数

### 1. 统计全部元素

| 函数 | 含义 |
| --- | --- |
| `np.mean()` | 平均值 |
| `np.sum()` | 和 |
| `np.max()`、`np.min()` | 最大值、最小值 |
| `np.std()`、`np.var()` | 标准差、方差 |
| `np.argmax()`、`np.argmin()` | 最大值、最小值所在的索引 |
| `np.cumsum()` | 累加和 |
| `np.cumprod()` | 累乘积 |

最大值与最大值的索引是两件事。`max()` 回答“数值是多少”，`argmax()` 回答“它在哪个位置”。默认不指定轴时，多维数组的 `argmax()`、`argmin()` 返回展平后的索引。

累加和、累乘积会保留沿途的累计结果。每个位置都包含截至当前位置的元素；不指定 `axis` 时，`cumsum()` 和 `cumprod()` 先把数组按默认顺序视为一维数据，再累计。

### 2. 用 axis 指定统计方向

多维数组不指定 `axis` 时，这些统计函数默认处理全部元素。对于二维数组，课件重点区分以下三种写法：

| 调用 | 聚合范围 | 对 3 行 4 列数组的结果形状 |
| --- | --- | --- |
| `np.sum(arr)` | 全部元素 | 一个标量 |
| `np.sum(arr, axis=0)` | 沿行方向合并，得到每列的和 | 长度为 `4` 的数组 |
| `np.sum(arr, axis=1)` | 沿列方向合并，得到每行的和 | 长度为 `3` 的数组 |

课程示例：

```python
arr = np.arange(12).reshape(3, 4)
print(arr)

print(np.cumsum(arr))  # 返回一维数组，保留截至当前位置的累加和
print(np.sum(arr))     # 全部元素的和
print(np.sum(arr, axis=0))  # 按列统计和
print(np.sum(arr, axis=1))  # 按行统计和
```

`axis` 指定的是要沿哪个轴聚合。二维数组的 `axis=0` 消去行这个维度，每列得到一个结果；`axis=1` 消去列这个维度，每行得到一个结果。课程所说的“0 按列，1 按行”适用于这里的二维数组。

## 十二、NumPy 内置函数：去重与排序

### 1. 去重：unique

`np.unique()` 查找唯一值并返回排序后的结果。课程把它类比为 Python 的 `set`，但 NumPy 返回的是数组，而且默认结果有序。

```python
arr = np.array([[1, 2, 1], [2, 3, 4]])
print(arr)
print(np.unique(arr))
```

这个二维数组中有重复的 `1` 和 `2`。默认调用按全部元素去重，因此得到的是唯一值的一维数组，而不是保持原有两行的布局。

### 2. 排序：sort

课件对比 `np.sort(arr)` 与 `arr.sort()`，关键在于是否修改原数组。

```python
arr = np.array([1, 2, 34, 5])
print("原数组arr:", arr)

# np.sort 返回排序后的副本
sortarr1 = np.sort(arr)
print("numpy.sort()函数排序后的数组:", sortarr1)

# ndarray 直接调用 sort，在原数组上进行修改
arr.sort()
print("数组.sort()方法排序:", arr)
```

| 写法 | 原数组 | 返回值 |
| --- | --- | --- |
| `np.sort(arr)` | 保留原来的排列 | 排序后的新数组 |
| `arr.sort()` | 在原数组上排序 | `None` |

需要保留原数据时接收 `np.sort()` 的结果；调用 `arr.sort()` 后应查看 `arr`，不要用它的返回值覆盖数组变量。

## 十三、NumPy 运算：逐元素计算

NumPy 数组的基本算术运算按对应元素进行。课程先演示两个一维数组相减：

```python
import numpy as np

a = np.array([20, 30, 40, 50])
b = np.arange(4)
c = a - b
print("数组a:", a)
print("数组b:", b)
print("数组运算a-b:", c)
```

`b` 包含 `0`、`1`、`2`、`3`，计算时分别与 `a` 中同一位置的元素配对。普通的 `a - b` 创建结果数组，由 `c` 接收。

课程接着演示二维数组对应元素相乘：

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 2, 3], [4, 5, 6]])
a * b
np.multiply(a, b)
```

`a * b` 与 `np.multiply(a, b)` 在这里都是逐元素相乘。两个数组形状都是 `(2, 3)`，结果也保持 `(2, 3)`，每个位置分别计算 `1*1`、`2*2` 等。

> 二维 `ndarray` 使用 `*` 时计算的是对应元素的乘积。判断运算含义要看操作符和函数，不能仅凭数组是二维就认定它做矩阵乘法。

## 十四、NumPy 运算：矩阵乘法

### 1. 行与列对应相乘再求和

矩阵乘法中，结果的一个元素来自左矩阵的一行与右矩阵的一列：先对应相乘，再把乘积相加。

左矩阵形状为 `(m, n)`，右矩阵形状为 `(n, p)` 时，可以相乘，结果形状为 `(m, p)`。**左矩阵列数必须等于右矩阵行数。**

课件用 2 行 3 列的矩阵 A 与 3 行 2 列的矩阵 B 展开计算：

<img src="01-img/matrix-product.jpg" alt="课件中 2×3 与 3×2 矩阵相乘的公式展开" width="850">

例如结果的第一行第一列，使用 A 的第一行和 B 的第一列，计算三个乘积之和。其余位置按同样规则得到，最终构成 2 行 2 列的矩阵。

### 2. 使用 dot 完成矩阵乘法

课件示例使用二维数组，分别调用数组方法 `x.dot(y)` 和 NumPy 函数 `np.dot(x, y)`：

```python
import numpy as np

x = np.array([[1, 2, 3], [4, 5, 6]])
y = np.array([[6, 23], [-1, 7], [8, 9]])

print(x)
print(y)

print(x.dot(y))
print(np.dot(x, y))
```

`x.shape` 为 `(2, 3)`，`y.shape` 为 `(3, 2)`，内侧大小都是 `3`，所以可以进行矩阵乘法，结果形状为 `(2, 2)`。两种 `dot` 写法在这个二维示例中表示同一项计算。

| 运算 | 本篇示例的计算规则 | 形状关系 |
| --- | --- | --- |
| `a * b`、`np.multiply(a, b)` | 对应位置的元素相乘 | 示例中两边同为 `(2, 3)`，结果为 `(2, 3)` |
| `x.dot(y)`、`np.dot(x, y)` | 左边的一行与右边的一列相乘求和 | `(2, 3)` 与 `(3, 2)` 相乘，结果为 `(2, 2)` |

这里讲的是二维数组的矩阵乘法。检查表达式时，先读清 `shape`，再确认使用的是逐元素运算还是 `dot`。
