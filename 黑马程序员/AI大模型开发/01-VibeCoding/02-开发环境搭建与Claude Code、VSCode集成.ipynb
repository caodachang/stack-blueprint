# VibeCoding开发环境：部署与Claude Code、VSCode集成


## 1. 开发环境的组成与安装顺序

AI生成代码之后，仍需要本地工具保存文件、安装依赖、运行程序和检查结果。开发环境就是完成这些动作的一组软件及配置。

本篇对应视频 **02—06、08**，并结合课程配套讲义整理。Windows采用PowerShell，macOS采用系统终端中的zsh或bash。

| 组件 | 作用 | 本篇怎样确认可用 |
|---|---|---|
| 终端与Shell | 输入、解释和执行命令 | 能切换到指定项目目录 |
| Node.js与npm | 运行JavaScript工具、管理JavaScript依赖 | 输出版本，执行简单JavaScript |
| Python与pip | 运行Python程序、管理Python依赖 | 确认解释器和pip属于同一环境 |
| Git | 记录代码版本；Windows安装包还提供Git Bash | 输出版本 |
| Claude Code | 读取项目、修改文件、调用开发工具 | 能完成一个小任务 |
| 模型服务 | 为编程Agent提供模型能力 | 配置正确，实际请求成功 |
| VSCode与扩展 | 查看文件、终端、改动和Agent交互界面 | 能打开项目并使用官方扩展 |

```text
选系统和Shell → 安装运行时与Git → 安装Claude Code
                                  ↓
                       配置模型服务并验证CLI
                                  ↓
                       安装VSCode和扩展 → 小项目联调
```

**CLI**是命令行界面。先验证CLI，再验证编辑器扩展，能把“安装问题”“模型连接问题”和“编辑器问题”分开定位。

课程安装包有固定版本，重新搭建时应结合项目要求选择受支持的版本。新建Node.js环境优先选择官网标注的LTS（长期支持）版本。


## 2. 选择Windows、macOS与实际运行环境

命令运行在哪个系统，就需要在哪个系统安装对应工具。Windows上的Python安装完成，并不代表WSL里的Python也已经可用。

| 环境 | 常用Shell | 本篇路线 | 特点 |
|---|---|---|---|
| Windows原生 | PowerShell | 主路线 | 直接安装Windows版工具 |
| macOS | zsh或bash | 对应补充 | 使用macOS安装包和Shell命令 |
| WSL | Linux中的bash等 | 本篇不展开安装 | 在Windows上运行Linux环境，工具和依赖需按Linux管理 |

**WSL**是Windows Subsystem for Linux。**Shell**是解释命令的程序；**终端**是承载Shell的窗口。例如Windows Terminal可以承载PowerShell、CMD或WSL。

课程选择Windows Terminal中的PowerShell。Windows 10可从Microsoft Store获取Windows Terminal；没有这个应用时，系统PowerShell也能执行本篇的PowerShell命令。

| 操作 | Windows PowerShell | macOS zsh/bash |
|---|---|---|
| 查看当前目录 | `Get-Location` | `pwd` |
| 查看目录内容 | `Get-ChildItem` | `ls` |
| 查看命令位置 | `Get-Command python` | `command -v python3` |
| 设置当前进程环境变量 | `$env:NAME="value"` | `export NAME="value"` |

不要把PowerShell的环境变量语法直接粘到CMD或macOS终端里。VSCode的集成终端也需要选择相应的Shell配置。


## 3. 当前目录、路径与PATH

终端中的命令从一个“当前目录”开始执行。相对路径从这里计算；PATH则是系统查找可执行命令时使用的目录列表。两者解决不同问题。

| 概念 | 解决什么问题 | 示例 |
|---|---|---|
| 当前目录 | 此刻在哪个文件夹操作 | `C:\dev\vibe-env-check` |
| 绝对路径 | 明确指向一个位置 | `C:\dev\vibe-env-check\index.html` |
| 相对路径 | 从当前目录寻找文件 | `.\index.html` |
| PATH | 输入命令名时去哪里找程序 | 输入`python`后定位到某个`python.exe` |

Windows示例：

```powershell
New-Item -ItemType Directory -Path 'C:\dev\vibe-env-check'
Set-Location -LiteralPath 'C:\dev\vibe-env-check'
Get-Location
Get-ChildItem

Get-Command python -All
Get-Command node -All
```

macOS示例：

```bash
mkdir -p ~/dev/vibe-env-check
cd ~/dev/vibe-env-check
pwd
command -v python3
command -v node
```

**正例：**先进入练习目录，再运行`claude`，让任务与项目位置一致。

**反例：**终端仍在上一个项目中，却要求AI“修改当前项目”，导致修改范围与预期不一致。

> 安装器修改PATH之后，已经打开的终端通常仍持有旧环境。关闭并重新打开终端再验证；VSCode已经运行时，也可能需要重启它及其终端。

**场景一：命令找不到。**检查安装是否完成、安装目录是否在PATH、终端是否重启。

**场景二：版本不对。**检查命令解析到了哪个位置，确认是否有旧版本或其他环境排在前面。


## 4. Node.js与npm的安装和验证

浏览器能运行网页里的JavaScript，Node.js则让JavaScript在浏览器之外运行。前端构建工具、部分开发服务和npm命令都需要它。

**npm**是Node.js生态的包管理工具，通常随官方Node.js安装包一起安装。**包**是可安装的代码或工具；**依赖**是项目运行或构建需要的包。

| 平台 | 课程讲义示例 | 新建环境的选择 |
|---|---|---|
| Windows | `node-v24.16.0-x64.msi` | 官网LTS、匹配CPU架构的Windows安装包 |
| macOS | `node-v24.16.0.pkg` | 官网提供的对应macOS安装包 |

**CPU架构**例如x64或ARM64，决定安装包是否适配处理器。Mac的Apple Silicon与Intel机型也需要选择适配的构建。

安装后重新打开终端。Windows PowerShell：

```powershell
node --version
npm.cmd --version
node -e "console.log('Node.js OK')"
```

macOS：

```bash
node --version
npm --version
node -e "console.log('Node.js OK')"
```

Windows示例使用`npm.cmd`，明确调用npm的命令包装器，避免PowerShell优先调用`npm.ps1`时遇到脚本策略限制。两种入口通常对应同一套npm安装。

日后运行项目时，`npm install`用于安装项目依赖，`npm run dev`用于运行项目定义的`dev`脚本。后者能否使用，要查看`package.json`中的`scripts`，不能假设每个项目都有它。

参考：[Node.js官方下载](https://nodejs.org/en/download)。


## 5. Python、解释器与pip

Python代码由**解释器**执行；`python`、`python3`通常是启动解释器的命令。**pip**是安装Python包的工具。

课程讲义使用Python 3.14.6安装包。实际项目如果规定了Python版本范围，应优先满足项目约束，例如`pyproject.toml`中的`requires-python`。

### Windows

Python官网当前提供Python Install Manager；较早的完整安装器也常见。Install Manager是管理Python运行时的工具，安装方式和旧安装器的界面不同。

课程采用完整安装器，强调勾选 **Add python.exe to PATH**，并给出自定义目录示例。这是课程示例，Python并不要求必须安装在D盘，也不是所有中文路径都不能使用。

### macOS

可以使用python.org提供的`.pkg`安装包。官方安装器还提供`Install Certificates.command`完成该Python环境的证书配置。不要修改或删除Apple管理的系统Python。

安装后检查：

| 检查 | Windows | macOS |
|---|---|---|
| 版本 | `python --version` | `python3 --version` |
| 解释器实际位置 | `python -c "import sys; print(sys.executable)"` | `python3 -c "import sys; print(sys.executable)"` |
| 对应pip | `python -m pip --version` | `python3 -m pip --version` |

Windows如果安装了可用的Python启动器，也可用`py`启动；具体行为取决于安装方式。

**优先使用`python -m pip`。**`-m`表示用这一个解释器运行指定模块，可以减少“pip装到了另一个Python里”的问题。Mac保持使用`python3`即可，不需要为了跟视频命令一样而强行创建`python`别名。

参考：[Windows官方说明](https://docs.python.org/3/using/windows.html)、[传统安装器说明](https://docs.python.org/3.13/using/windows.html)、[macOS说明](https://docs.python.org/3.13/using/mac.html)。


## 6. 用虚拟环境隔离Python项目

两个项目可能依赖同一个库的不同版本。**虚拟环境**为一个项目提供独立的Python包目录，避免把所有依赖混在系统环境里。

```text
vibe-env-check/
├── .venv/          项目Python环境，不放业务代码
├── index.html      后面的练习页面
└── requirements.txt  有第三方依赖时记录依赖
```

在项目目录创建环境。Windows：

```powershell
python -m venv .venv
.\.venv\Scripts\python.exe --version
.\.venv\Scripts\python.exe -m pip --version
.\.venv\Scripts\python.exe -c "import sys; print(sys.executable)"
```

macOS：

```bash
python3 -m venv .venv
./.venv/bin/python --version
./.venv/bin/python -m pip --version
./.venv/bin/python -c "import sys; print(sys.executable)"
```

以上直接调用环境内的解释器，不需要先执行激活脚本。后续安装依赖也使用同一解释器，例如Windows的`.\.venv\Scripts\python.exe -m pip install -r requirements.txt`。

激活只是让当前Shell优先查找该环境中的命令：

| 平台 | 激活 | 退出激活 |
|---|---|---|
| Windows PowerShell | `.\.venv\Scripts\Activate.ps1` | `deactivate` |
| macOS zsh/bash | `source .venv/bin/activate` | `deactivate` |

> 激活脚本被拦截时，可以直接调用环境内的Python。创建了虚拟环境，也要确认VSCode和Agent实际使用的是这个解释器。

**场景一：装了库仍提示找不到。**核对执行程序的解释器与安装库的解释器是否相同。

**场景二：换电脑或移动项目。**根据依赖记录重新创建虚拟环境。虚拟环境中的路径可能绑定原位置，不适合直接复制或提交到Git。

如果已有项目由uv管理，继续遵循该项目的`uv sync`、`uv run`流程，不要另外随意创建一套依赖环境。本篇示例使用Python自带的venv。

参考：[venv官方文档](https://docs.python.org/3/library/venv.html)。


## 7. Git与Git Bash

AI可能一次修改多个文件。Git用于记录这些修改，让你能比较版本、保存进度和协作。Git安装完成不会自动替每个目录建立版本记录，初始化与提交留到后续工程化笔记。

### Windows

课程提供Git for Windows安装包。新建环境也可从[Git官方安装页面](https://git-scm.com/install/windows)下载，或使用官方页面列出的命令：

```powershell
winget install --id Git.Git -e --source winget
git --version
```

**WinGet**是Windows包管理工具；若系统没有`winget`，使用官网安装包即可。

Git for Windows还提供**Git Bash**，它是带有Bash及相关工具的环境。`git.exe`与`bash.exe`承担不同任务。DeepSeek接入指南要求Windows先安装Git for Windows；当Claude Code明确提示缺少Bash时，需要检查Git Bash安装及工具识别情况。

### macOS

课程使用Apple的命令行开发工具：

```bash
xcode-select --install
git --version
```

若已经有可用Git，可以先验证，不必重复安装。已有Homebrew的用户也可选择`brew install git`，选一种路线即可。

首次提交前需要配置作者名称和邮箱，例如：

```bash
git config --global user.name "你的名称"
git config --global user.email "your-name@example.com"
```

`--global`表示对当前用户生效，示例值需要替换。作者信息是提交记录的一部分，与登录Gitee或GitHub的认证是不同配置。

参考：[Git macOS安装页面](https://git-scm.com/install/mac)。


## 8. 安装Claude Code：课程方式与当前推荐方式

Claude Code是编程Agent工具。安装它解决“本地能否运行工具”，接入模型解决“工具能否得到模型响应”，需要分别验证。

**课程讲义使用npm安装；核对时Anthropic官方仓库已将npm安装标为弃用，推荐原生安装、WinGet或Homebrew。**本篇保留课程方式用于识别旧教程，实际新装优先采用官方推荐方式。

| 平台 | 推荐路线之一 | 另一种官方路线 |
|---|---|---|
| Windows | WinGet | 官方PowerShell安装脚本 |
| macOS | 官方Shell安装脚本 | 已有Homebrew时使用cask |

Windows选择其中一种：

```powershell
# 路线A：WinGet
winget install Anthropic.ClaudeCode

# 路线B：官方脚本；选择B时不必再执行A
irm https://claude.ai/install.ps1 | iex
```

macOS选择其中一种：

```bash
# 路线A：官方脚本
curl -fsSL https://claude.ai/install.sh | bash

# 路线B：已有Homebrew时可选
brew install --cask claude-code
```

`irm`是PowerShell的`Invoke-RestMethod`缩写；`iex`执行取得的脚本文本；macOS的管道把下载脚本交给Bash执行。这些命令会下载并运行官方安装脚本，应核对域名和来源。

安装后重新打开终端，执行：

```bash
claude --version
claude --help
```

旧npm方式如下，**不作为本篇的新装默认路线**：

```bash
npm install -g @anthropic-ai/claude-code
```

`-g`表示全局安装命令行工具。npm路线需要Node.js；课程里安装Node.js还用于后续JavaScript项目，不能把所有运行时都简单视为Claude Code本体的依赖。

本篇不要求先切换npm镜像，也不把课程中的`sudo npm install`当作macOS通用步骤。

参考：[Anthropic官方仓库README](https://github.com/anthropics/claude-code#readme)。


## 9. 选择模型服务：工具、地址、凭证和模型名

Claude Code是工具名称，实际连接哪个模型取决于配置。课程选择**DeepSeek**作为模型提供商；也可以按Claude Code官方引导使用有相应访问权限的官方账户。

| 项目 | 含义 | 课程DeepSeek路线 |
|---|---|---|
| 工具 | 负责项目操作的客户端 | Claude Code |
| 服务地址 | 请求发送到哪里 | `https://api.deepseek.com/anthropic` |
| 凭证 | 服务如何确认调用者 | DeepSeek API Key |
| 模型名 | 服务实际接收的模型标识 | 使用DeepSeek当前接入指南中的标识 |

**API**是软件之间调用功能的接口；**API Key**是调用服务的凭证。`/anthropic`是DeepSeek提供的Anthropic兼容接口地址，用于适配Claude Code所使用的请求协议。

```text
项目文件与工具 ←→ Claude Code ←→ 配置的模型API
                                  ├─ 官方账户路线
                                  └─ DeepSeek兼容接口路线
```

按课程路线，需要在[DeepSeek开放平台](https://platform.deepseek.com/)管理API Key及用量。可用额度、计费和模型范围以平台账户实际状态为准。

**不要把DeepSeek Key当成Anthropic Key使用。**服务地址与凭证必须属于同一配置路线。已有官方登录或其他提供商配置时，先弄清当前生效配置再调整，避免多套认证混用。


## 10. 在当前终端配置DeepSeek

**环境变量**是进程可以读取的配置。终端中设置后，由这个终端启动的Claude Code可以继承它；关闭终端后，这种会话配置通常不会继续保留。

以下模型标识与映射按 **2026-10-09核对的DeepSeek官方接入指南**整理。模型服务会更新，日后遇到不存在的模型时应重新核对该指南，不要仅靠旧视频中的名字。

Windows PowerShell：

```powershell
$env:ANTHROPIC_BASE_URL = "https://api.deepseek.com/anthropic"
$env:ANTHROPIC_AUTH_TOKEN = "<你的DeepSeek API Key>"
$env:ANTHROPIC_MODEL = "deepseek-flash[1m]"
$env:ANTHROPIC_DEFAULT_OPUS_MODEL = "deepseek-flash[1m]"
$env:ANTHROPIC_DEFAULT_SONNET_MODEL = "deepseek-flash[1m]"
$env:ANTHROPIC_DEFAULT_HAIKU_MODEL = "deepseek-flash"
$env:CLAUDE_CODE_SUBAGENT_MODEL = "deepseek-flash"
$env:CLAUDE_CODE_EFFORT_LEVEL = "max"
```

macOS：

```bash
export ANTHROPIC_BASE_URL="https://api.deepseek.com/anthropic"
export ANTHROPIC_AUTH_TOKEN="<你的DeepSeek API Key>"
export ANTHROPIC_MODEL="deepseek-flash[1m]"
export ANTHROPIC_DEFAULT_OPUS_MODEL="deepseek-flash[1m]"
export ANTHROPIC_DEFAULT_SONNET_MODEL="deepseek-flash[1m]"
export ANTHROPIC_DEFAULT_HAIKU_MODEL="deepseek-flash"
export CLAUDE_CODE_SUBAGENT_MODEL="deepseek-flash"
export CLAUDE_CODE_EFFORT_LEVEL="max"
```

| 配置 | 作用 |
|---|---|
| `ANTHROPIC_BASE_URL` | 指定兼容接口地址 |
| `ANTHROPIC_AUTH_TOKEN` | 提供服务认证凭证 |
| `ANTHROPIC_MODEL` | 指定主要模型 |
| `ANTHROPIC_DEFAULT_*_MODEL` | 将Claude Code中的模型档位映射到提供商模型 |
| `CLAUDE_CODE_SUBAGENT_MODEL` | 指定子Agent模型，后续课程会展开 |
| `CLAUDE_CODE_EFFORT_LEVEL` | 接入指南给出的推理努力配置 |

这里的档位映射不代表DeepSeek变成了Claude模型。各键的取值和兼容性需要随工具及提供商更新。

只在本地替换API Key占位符，不把真实Key写进笔记、Git仓库或共享截图。先用这类会话配置验证CLI，再考虑持久化。

参考：[DeepSeek接入Agent工具](https://api-docs.deepseek.com/zh-cn/guides/coding_agents)。


## 11. 持久化配置：settings.json与进程环境

CLI在一个终端中成功，不代表已经打开的VSCode自动取得了同样的环境变量。**进程**是正在运行的程序实例；它通常在启动时继承父进程的环境。

课程采用Claude Code的用户级配置文件，使之后启动的工具能够读取设置：

| 平台 | 用户级文件 |
|---|---|
| Windows | `C:\Users\你的用户名\.claude\settings.json` |
| macOS | `~/.claude/settings.json` |

`~`表示当前用户的主目录。这是工具配置文件，不是项目代码目录里的Python虚拟环境。

下面是将上一节环境变量写入`env`字段的示例。**已有配置时合并字段，不要覆盖其他设置。**

```json
{
  "env": {
    "ANTHROPIC_BASE_URL": "https://api.deepseek.com/anthropic",
    "ANTHROPIC_AUTH_TOKEN": "<你的DeepSeek API Key>",
    "ANTHROPIC_MODEL": "deepseek-flash[1m]",
    "ANTHROPIC_DEFAULT_OPUS_MODEL": "deepseek-flash[1m]",
    "ANTHROPIC_DEFAULT_SONNET_MODEL": "deepseek-flash[1m]",
    "ANTHROPIC_DEFAULT_HAIKU_MODEL": "deepseek-flash",
    "CLAUDE_CODE_SUBAGENT_MODEL": "deepseek-flash",
    "CLAUDE_CODE_EFFORT_LEVEL": "max"
  }
}
```

**JSON**是结构化配置格式：键和字符串使用双引号，末尾不多加逗号，标准JSON不写注释。该文件如果保存真实Key，就含有凭证，不应复制进项目仓库。

修改后关闭旧Claude Code会话，再启动验证；VSCode扩展会话也需要重新启动。多种配置同时存在时，应检查有效配置与实际请求，不能只看某个文件已写入。

**场景一：CLI能连，扩展不能连。**核对是否只在CLI所在终端临时设置了变量，扩展是否读取了用户级配置。

**场景二：换目录后服务改变。**检查项目是否存在额外设置，以及当前启动环境是否相同。课程演示的设置读取方式还需要在实际扩展版本中联调确认。


## 12. 首次启动与分层验证

先在自己创建的练习目录启动Claude Code。Windows：

```powershell
Set-Location -LiteralPath 'C:\dev\vibe-env-check'
claude
```

macOS：

## 15. 小项目联调：验证环境确实贯通

版本号检查完成后，用一个无第三方依赖的小网页验证“模型响应 → 文件生成 → 本地运行 → 页面交互”。这个练习不代替后续贪吃蛇实战。

在CLI或VSCode的Claude面板提出：

```text
在当前练习目录创建index.html，使用原生HTML、CSS、JavaScript。
页面显示“环境联调”，有一个计数按钮，初始计数为0，
每点击一次增加1。不要引入第三方依赖，不要修改目录外的文件。
完成后说明文件位置、打开方式、已完成检查和待人工检查项。
```

确认文件生成到预期目录后，使用本项目的Python虚拟环境启动本地HTTP服务。

Windows：

```powershell
.\.venv\Scripts\python.exe -m http.server 8000 --bind 127.0.0.1
```

macOS：

```bash
./.venv/bin/python -m http.server 8000 --bind 127.0.0.1
```

**HTTP服务**把当前目录的文件通过网页地址提供给浏览器。`127.0.0.1`是本机回环地址，`8000`是本例监听的端口。浏览器访问：

```text
http://127.0.0.1:8000/
```

逐项检查：

| 操作 | 预期 |
|---|---|
| 打开页面 | 出现“环境联调”和初始计数0 |
| 点击一次按钮 | 计数变为1 |
| 连续点击三次 | 计数继续正确增加 |
| 刷新页面 | 因未实现持久化，计数回到0 |
| 在VSCode查看文件 | 内容与Agent交付说明一致 |

服务会持续占用当前终端，其他操作另开终端即可；在服务终端按`Ctrl+C`停止。本例服务只用于本地练习。

本篇仅提供操作与验收模板，没有实际安装软件、连接模型或生成这个网页，因此不把这些检查标为“已通过”。


## 16. 常见故障的定位顺序

先确认是“命令不可用”“配置没有生效”“模型请求失败”还是“项目代码出错”，再处理对应层次。

| 现象 | 优先检查 | 处理方向 |
|---|---|---|
| `node`、`python`或`git`找不到 | 安装完成情况、PATH、旧终端 | 查看命令位置，重开终端 |
| 版本与预期不同 | 多个安装位置及PATH顺序 | 核对实际解释器或可执行文件 |
| `npm.ps1`不能运行 | PowerShell选择了脚本入口 | 先使用`npm.cmd`，确认npm本体可用 |
| `Activate.ps1`被拦截 | 激活脚本策略 | 直接调用`.venv`中的Python |
| Claude Code提示缺少Bash | Git for Windows及Git Bash | 核对安装组件和提示中的路径要求 |
| `claude --version`成功但无法对话 | 服务地址、认证、额度、网络 | 阅读具体请求错误，分开检查 |
| HTTP 401 | 认证凭证及提供商是否匹配 | 检查Key与服务地址 |
| HTTP 403 | 响应正文中的访问或额度原因 | 根据正文处理，不能统一判为Key错误 |
| 提示模型不存在 | 模型标识和映射 | 重新核对提供商接入文档 |
| CLI成功但VSCode失败 | 进程环境、用户配置、扩展会话 | 核对有效配置，重启相应会话 |
| Python提示缺少模块 | 安装依赖和运行程序用的解释器 | 使用同一解释器的`-m pip` |
| 浏览器无法访问本地页面 | 服务是否运行、地址和端口 | 查看服务终端日志与当前目录 |
| 8000端口被占用 | 已有监听服务 | 换用8001并同步修改浏览器地址 |

Windows的定位命令：

```powershell
Get-Location
Get-Command node -All
Get-Command python -All
Get-Command git -All
Get-Command claude -All
python -c "import sys; print(sys.executable)"
python -m pip --version
```

这些命令用于确定“在哪儿、用了什么工具”。如果怀疑认证问题，查看配置时不要把真实Key输出到共享日志中。

不要仅凭“安装失败”反复重装所有软件。记录失败的命令、完整错误和实际环境，才能缩小问题范围。
