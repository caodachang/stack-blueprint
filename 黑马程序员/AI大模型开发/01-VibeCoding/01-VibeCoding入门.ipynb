# VibeCoding入门：概念、初体验与开发流程


## 1. VibeCoding是什么

写程序时，我们先有一个目标，例如“做一个能开始、暂停和重置的番茄钟”，再把目标转换成界面、数据和程序逻辑。VibeCoding把自然语言作为主要入口，让AI完成大量代码实现，再根据运行结果继续调整。

课程课件把VibeCoding（氛围编程）概括为：**用自然语言描述需求，由AI生成可运行代码的开发方式。**

| 课件中的特点 | 开发时的含义 | 番茄钟示例 |
|---|---|---|
| 自然语言驱动 | 先说明想实现的功能和约束 | “25分钟专注，5分钟休息” |
| 即时生成与反馈 | 生成代码后，尽快运行并观察结果 | 点击开始，观察倒计时是否变化 |
| 快速试错迭代 | 根据具体问题分步修改 | 发现暂停后时间仍减少，要求修复 |
| 聚焦意图而非实现 | 人主要表达目标，AI承担代码实现 | 先确定按钮应该怎么工作 |

“生成可运行代码”是目标，实际输出仍需要运行验证。AI也可能生成语法错误、遗漏功能或误解需求。

> 本笔记采用课程的广义用法。实践中应保留对代码、测试和交付结果的判断；下面的开发流程是在课程入门概念上补充的工程实践。


## 2. AI辅助编程的三种常见形态

同样是“让AI帮我编程”，AI能接触到的项目资料、能执行的动作可能不同，因此不能只看它是否会回答问题。

| 形态 | AI通常能做什么 | 人主要做什么 | 适合的任务 |
|---|---|---|---|
| 聊天问答 | 解释代码、生成片段、根据贴出的报错分析 | 提供资料，复制代码，自己运行 | 学概念、理解错误、写小片段 |
| 编辑器辅助 | 根据当前文件和部分项目上下文补全、解释或修改代码 | 选择修改范围，检查建议 | 局部重构、修改函数 |
| 编程Agent | 在授权范围内读文件、改文件、调用终端并查看结果 | 定义任务与边界，验证交付 | 连续完成项目任务 |

这里的**编辑器**是写代码的工具；**终端**是输入命令、运行程序的入口；**Agent**是围绕目标调用工具并根据结果继续行动的系统。

课程主要使用Claude Code，并在后续介绍VSCode集成和Qoder。具体能否执行某个动作，取决于工具功能、当前配置和权限。


## 3. 大模型、工具与编程Agent

大模型能够生成“应该怎么改”的回答，但模型本身不会自动改变你电脑里的文件。要完成项目任务，还需要一个系统把模型的决定转换成实际操作，再把结果送回模型。

| 概念 | 含义 | 在番茄钟任务中的作用 |
|---|---|---|
| 大模型（LLM） | 根据输入内容生成文本或工具调用请求的模型 | 理解需求，提出实现方式和代码 |
| 上下文 | 本次任务中提供给模型的信息，例如需求、文件和报错 | 让模型知道现有代码和要修复的问题 |
| 工具 | 执行具体动作的能力，例如读文件、写文件、运行命令 | 保存HTML文件，执行检查 |
| 编程Agent | 将模型、上下文和工具组织起来，持续完成开发任务的系统 | 根据运行结果决定下一步修改 |

```text
人给出目标与约束
        ↓
Agent组织上下文 → 大模型决定下一步 → 工具执行动作
        ↑                                  ↓
        └────────── 文件、日志、测试结果 ────┘
        ↓
人根据验收标准确认结果
```

**使用场景一：**程序报错，把完整错误信息和相关文件提供给Agent，让它定位原因、修改代码，再运行同一条命令验证。

**使用场景二：**程序能启动，但暂停按钮无效。提供复现步骤和预期行为，让Agent修复逻辑，而不只是解释代码。

> “AI说修改完成”是一条反馈。修改是否有效，要看实际文件和运行结果。


## 4. AI参与开发与应用调用AI

“使用AI写程序”和“程序本身有AI功能”是两个独立维度。课程中的贪吃蛇、记账APP可以由AI辅助开发，但运行这些应用不一定需要大模型。

| 场景 | 开发过程中使用AI | 应用运行时调用大模型 |
|---|---|---|
| 用Claude Code开发本地番茄钟 | 是 | 不需要 |
| 用AI开发普通记账软件 | 可以 | 不一定需要 |
| 手工开发一个AI聊天助手 | 不一定 | 需要 |
| 用编程Agent开发RAG知识库问答系统 | 是 | 通常需要 |

**RAG（检索增强生成）**是在回答前先检索相关资料，再把资料提供给大模型的做法。它属于应用能力，课程后半部分会用它做项目。

学习时先明确当前问题属于哪一层：是“怎样更好地让AI帮我开发”，还是“怎样在软件里接入大模型”。番茄钟初体验重点是前者。


## 5. 人与AI怎样分工

当AI承担更多代码实现，人仍需要决定功能是否合理、结果是否正确。入门阶段最重要的能力，是把目标说清楚，并根据证据判断完成情况。

| 开发环节 | 人的责任 | AI可承担的工作 |
|---|---|---|
| 定义需求 | 明确用户、功能、约束与验收标准 | 提出遗漏项，整理需求 |
| 确定方案 | 判断复杂度和技术选择是否合适 | 给出方案及取舍 |
| 实现功能 | 限定修改范围，提供业务规则 | 编写或修改代码 |
| 运行与调试 | 提供真实环境和复现现象 | 分析报错，修复后复测 |
| 验收与交付 | 确认功能、数据、质量和使用方式 | 汇总改动、检查结果和运行说明 |

**新手的用法：**要求AI解释入口文件、运行方式、修改涉及的文件及验证步骤；遇到术语先问清它在当前项目里起什么作用。

**有经验开发者的用法：**补充接口契约、代码规范、测试要求和项目约束，让AI按已有工程规则完成任务。

课件总结分别面向零基础新手和有经验开发者，强调自然语言沟通；开发任务越复杂，越需要专业约束和验证。


## 6. 把一句想法写成可验收的需求

“帮我做个好看的番茄钟”缺少功能边界。AI需要自行猜测按钮状态、计时规则和页面风格，生成结果就可能偏离预期。

一个可执行的需求至少包含五项：

| 要素 | 要回答的问题 | 示例 |
|---|---|---|
| 目标 | 帮谁完成什么任务 | 帮个人安排专注与休息时间 |
| 功能 | 用户能进行哪些操作 | 开始、暂停、继续、重置 |
| 约束 | 使用什么技术，避免哪些复杂度 | 本地运行，原生HTML/CSS/JavaScript |
| 验收 | 怎样证明行为正确 | 暂停后剩余时间不变，继续后接着倒计时 |
| 交付 | 最后需要拿到什么 | 源代码、打开方式、检查结果 |

**模糊表达：**

> 做个番茄钟，功能齐全，漂亮一点。

**具体表达：**

> 做一个本地网页番茄钟。默认专注25分钟、休息5分钟，支持开始、暂停、继续和重置。计时显示为“分:秒”，显示当前阶段。先不做登录、数据库和云同步。交付可直接打开的文件，并逐项说明按钮和阶段切换的检查结果。

提示词越长不一定越有效。优先补足规则和验收条件；与当前任务无关的背景可以省略。


## 7. 初体验：开发一个番茄钟

课件的初体验需求是：

> 我需要一个番茄钟，25分钟专注，5分钟休息，有开始、暂停、重置按钮。

下面将这句话扩展成一个完整任务。**扩展规则、提示词和验收项是本笔记补充，并非课程逐字记录。**环境安装留到第二篇笔记，本节假设已经有可用的AI编程工具。

### 7.1 先限制项目规模

采用原生HTML、CSS和JavaScript，在浏览器本地运行。**HTML**描述页面结构，**CSS**控制外观，**JavaScript**处理按钮交互和计时逻辑。第一版不接入后端，不添加账户系统。

### 7.2 给AI一个明确任务

```text
在当前空目录开发一个本地网页番茄钟，使用原生HTML、CSS、JavaScript。

需求：
1. 专注阶段默认25分钟，休息阶段默认5分钟。
2. 显示当前阶段和剩余时间，时间格式为“分:秒”。
3. 支持开始、暂停、继续和重置。
4. 点击开始后计时；重复点击不能创建多个倒计时。
5. 暂停后时间不再减少；继续后从暂停处接着计时。
6. 重置恢复到专注25分钟，并处于未开始状态。
7. 专注结束后切换到休息5分钟，等待用户点击开始；
   休息结束后切换到专注25分钟，同样等待开始。
8. 不做登录、数据库、云同步，也不引入第三方依赖。

请先简要说明文件结构和实现思路，再实现。
完成后说明如何打开页面、改动了哪些文件、
做了哪些检查，以及哪些行为仍需要人工验证。
```

这里把“阶段结束后是否自动开始”写清楚，是为了避免双方对同一句需求有不同理解。

### 7.3 打开项目并观察结果

实际文件名以AI交付为准。一种可能的结构是：

```text
pomodoro/
├── index.html   页面结构与脚本入口
├── style.css    布局、颜色与按钮样式
└── app.js       计时与状态切换
```

按交付说明打开页面，先检查是否出现25:00和专注阶段，再操作开始、暂停、继续、重置。按钮存在只说明界面元素已经生成，还需要观察状态变化。

### 7.4 用复现步骤反馈问题

```text
问题：点击暂停后，倒计时仍然减少。
复现：打开页面 → 点击开始 → 等待3秒 → 点击暂停 → 再等待3秒。
预期：第二次等待期间，显示的剩余时间不变。
实际：剩余时间继续减少。
请定位并修复原因，保留其他功能，再检查开始、暂停、继续、重置。
```

反馈应包含**操作、预期、实际结果**。如果有报错，再附上完整错误信息。


## 8. 从需求到交付的开发闭环

一次代码生成只能覆盖开发过程中的一部分。可复用的流程是：先确定目标，完成一个小功能，再用实际结果决定下一步。

```text
明确需求 → 确定方案 → 实现一个小版本 → 运行验证
                                          │
                              有问题：给出证据并修复
                                          │
                              通过：保存版本，再加功能
```

| 步骤 | 具体动作 | 产出 |
|---|---|---|
| 明确需求 | 写下功能、约束和验收条件 | 可核对的需求清单 |
| 确定方案 | 选运行方式，明确文件和功能边界 | 简短实现计划 |
| 最小实现 | 优先完成能展示、能操作的基础版本 | 可运行代码 |
| 运行验证 | 检查真实交互、报错和边界情况 | 成功证据或问题记录 |
| 修复迭代 | 一次处理一个可复现问题，修复后复测 | 行为符合预期的版本 |
| 保存交付 | 保存源代码、运行说明和检查结果 | 可继续开发的项目 |

**Git**是记录代码版本的工具。后续课程会讲初始化、提交、分支和远端同步；当前先建立“每完成一小步，就保存一个可恢复版本”的习惯。

小版本可以这样安排：第一步显示时间，第二步实现计时，第三步加入暂停和重置，第四步完成阶段切换，最后再调整界面。


## 9. 上下文与反馈为什么影响结果

AI需要根据上下文判断当前项目是什么、哪些代码已经存在、你希望改成什么。缺少信息时，它可能猜错文件、技术选型或业务规则。

| 情况 | 应提供的上下文 | 可用的表达 |
|---|---|---|
| 新建项目 | 目标、运行环境、技术约束 | “只需本地网页，不需要后端” |
| 修改已有功能 | 相关文件、现有行为、目标变化 | “保留当前页面结构，只调整计时逻辑” |
| 修复错误 | 复现步骤、预期、实际、完整报错 | “暂停后仍减少，步骤如下……” |
| 继续旧任务 | 已完成内容、待办和已知问题 | “开始和重置已通过，继续检查阶段切换” |

**上下文容量有限**，模型不能无限保留所有对话和文件细节。**Token**是模型处理文本的基本单位，通常与字、词或片段有关，不能简单按“一字一个Token”换算。

入门时可以维护一份简短项目说明，记录目标、运行方式、关键规则和待办。它帮助人和AI重新对齐任务。课程后续会进一步讲上下文压缩和memory记忆管理。


## 10. 怎样判断项目真的完成

验收条件应能观察、能复现。对于番茄钟，可以把“做完了”拆成下面的检查。

| 检查项 | 操作 | 预期结果 |
|---|---|---|
| 初始状态 | 首次打开页面 | 专注阶段，25:00，未开始 |
| 开始 | 点击开始并等待 | 时间逐步减少 |
| 重复开始 | 计时中多次点击开始 | 倒计时速度保持正常 |
| 暂停 | 暂停后继续等待 | 剩余时间保持不变 |
| 继续 | 暂停后点击继续 | 从剩余时间继续减少 |
| 重置 | 运行中或暂停时点击重置 | 回到专注25:00，计时停止 |
| 专注结束 | 将专注倒计时推进到结束 | 切换休息5:00，等待开始 |
| 休息结束 | 将休息倒计时推进到结束 | 切换专注25:00，等待开始 |

验证结束行为时，可在测试期间临时缩短时长，检查后恢复默认值；也可让AI编写测试来控制时间，避免每轮都等待25分钟。**不要把未操作、未检查的项目写成“已通过”。**

暂停、继续和阶段切换涉及**状态**，即程序当前处于哪种情况。可以先把状态理解成：

```text
未开始 ──开始──→ 运行中 ──暂停──→ 已暂停
                     ↑               │
                     └────继续───────┘

任意状态 ──重置──→ 专注阶段未开始
运行中 ──计时结束──→ 下一阶段未开始
```

除了功能，还要确认：按说明能否重新打开项目、依赖是否齐全、出错时是否有明确提示。交付应包含运行方法和验证记录。


## 11. 常见误区与能力边界

| 误区 | 容易出现的问题 | 更有效的做法 |
|---|---|---|
| 一次让AI完成所有功能 | 改动过多，难定位失败原因 | 分成小版本，每步验证 |
| 只说“有Bug，修一下” | 缺少定位信息，修改可能偏离问题 | 给出操作、预期、实际和报错 |
| 只看页面截图 | 计时和状态切换错误无法发现 | 实际操作并检查边界情况 |
| 看见“检查通过”就接受 | 检查范围可能只覆盖语法或启动 | 核对检查内容与验收项 |
| 每次修复都允许大改 | 引入额外变化，已有功能可能失效 | 明确修改范围，复测相关功能 |
| 代码生成后从不保存版本 | 出错后难以恢复到可用状态 | 完成一步就保存可恢复版本 |

AI可能误解业务规则、使用不存在或已变化的接口，也可能为了消除报错而修改错误的地方。涉及真实用户数据、权限或发布时，需要结合项目风险增加检查。

本篇只建立入门判断：**需求清楚、范围可控、反馈具体、结果可验证、版本可恢复。**测试、代码审查、Skills、Subagent和Hook会在后续笔记中展开。
