## 一、认识 AI 与 AIGC

来源：《AIGC》1-4 页。

**AI（Artificial Intelligence，人工智能）** 是研究和实现模拟、延伸、扩展人类智能的理论、方法、技术与系统。识别、预测、规划和生成都可以是 AI 的任务。

**AIGC（Artificial Intelligence Generated Content，人工智能生成内容）** 强调由 AI 生成文本、图像、音频、视频等内容，是 AI 的一种应用方向。

| 概念 | 关注点 | 内容创作中的例子 |
| --- | --- | --- |
| AI | 完成智能任务 | 识别人声、分析图像、推荐内容 |
| AIGC | 根据输入生成内容 | 写旁白、生成海报、合成配音、生成视频片段 |
| 多模态 | 处理文字、图像、声音等多种信息形式 | 上传参考图并用文字描述修改要求 |

**原稿表述补充**：多模态让创作链路更丰富，但 AIGC 不要求每个模型同时处理所有模态。只生成文本的模型也能参与 AIGC 工作。

本课中，文本通常是起点：先确定主题、人物、画面与节奏，再把描述交给图像或视频工具。

## 二、AI文本处理：提示词怎样写

来源：《AI文本处理大师》1-3 页。

原课核心：**好工具 + 好方法 = 好结果**；好的输出需要多次尝试和优化。

| 原则 | 要明确什么 | 常用补充内容 |
| --- | --- | --- |
| 明确性 | 要完成的任务与目标 | 写什么、给谁看、用于哪里 |
| 具体性 | 内容边界与交付要求 | 背景、结构、字数、语气、输出格式 |
| 创意性 | 区别于普通内容的亮点 | 特殊角色、独特场景、比喻、反常规设定 |

可以先按“任务 → 受众 → 背景 → 结构 → 风格 → 篇幅 → 输出格式”列要求，再组装成自然语言。这个顺序是整理补充，不是每次必须写满的固定模板。

比如“写一篇中秋文案”缺少使用场景；补充“给亲子观众看的 60 秒微缩景观视频，四场景，输出画面、旁白和绘画提示词表格”，就更容易交给后续制作环节。


### 2.1 原课三组提示词

来源：《AI文本处理大师》2 页，按原文整理标点。

**明确性：规定角色感与文本形式。**

```text
模仿初中生，写一篇关于“武汉樱花”非常绚烂美丽的诗，七言绝句。
```

七言绝句的基本外形是四句、每句七字；生成后仍要核对字数，不能只看标题就认定符合要求。

**具体性：规定主题、受众、结构和篇幅。**

```text
针对AI大模型对各行各业的影响，写一篇关于《AI引发社区工作的革命性变革》的公众号文章。
针对人群是中老年人，需要通俗易懂。
可以从“AI能做什么、AI带来的社区工作变革、工作人员如何用AI”的结构写，3000字左右。
```

**创意性：先给出独特设定，再限定叙述效果。**

```text
写一篇1500字左右的公众号推文。
标题：猫的奇妙之旅。
内容：一只能空间穿越的猫，在樱花盛开季节环游武汉的故事，描述它在武汉7天的冒险经历。
要求：结构合理，风趣幽默，还有一些人生启发。
```

优化时指出具体问题，例如“第三天和第四天的地点重复，请更换第四天的地点，并保留前后的时间顺序”。这种追问比“再写好一点”更便于修改。


### 2.2 用 AI 处理已有文档

来源：《AI文本处理大师》3-5 页。

原课流程：**原始文档导入 → 输入处理要求 → AI改写 → 人工校验与导出**。

| 处理方式 | 目标 | 交付时重点检查 |
| --- | --- | --- |
| 编写与重组 | 从材料生成可阅读文章 | 原文事实是否保留，背景和建议是否分开 |
| 改写与扩写 | 调整受众、表达和内容展开 | 新增内容有无依据，重复段落是否删去 |
| 修订与翻译 | 提升专业度，再转换语言 | 术语、数字、专有名词与语义是否一致 |
| 解读与策划 | 从政策或背景提出项目方案 | 原文要求、推断和建议是否区分清楚 |

先确认工具真的读到了附件。链接读取失败时，提供需要处理的原文；单纯粘贴一个链接并不保证模型已看到正文。

讲义中的党建材料、三篇养生文章、文章改写翻译材料、政策解读材料与三个 PDF 是课堂输入附件。本篇整理讲义中展示的处理方法与提示词，没有把这些附件名称当作已经读取过的正文。


### 2.3 原课文档处理模板：编写与扩写

来源：《AI文本处理大师》3-4 页。

**政务推文：材料重组。**

```text
我是【湖南省衡山县融媒体部门】，需要针对这个政策，结合【衡山县特色】，从实际情况出发，生成一篇【适合发公众号的推文】：
• 删除冗余重复信息
• 按“背景—分析—建议”结构重组
有专业度，有深度，1500字左右。
```

理解模板时抓住“身份与场景、已有材料、本地背景、文章结构、篇幅”这五部分。地方特色和政策细节应来自输入材料；材料不足时先列待补信息。

**营销文章：多篇材料改写并扩写。**

```text
针对上面三篇文章，重新生成一篇关于45岁左右中年人养生的文章，风格可以更具营销性与煽动性，直戳痛点，并给出可行的解决方案。
2000字左右。
```

这段保留原课措辞，用于理解受众与语气控制。实际整理知识文章时，应把“更有吸引力”落实为标题、结构和可理解的表达；原文未给出的事实或效果不能因为要求扩写而自动补成确定结论。


### 2.4 原课文档处理模板：修订、翻译与策划

来源：《AI文本处理大师》4 页。

**科普修订，随后翻译。**

```text
帮我把这篇文章，改写为专业性比较强的科普文章，知识准确、结构严谨、深入浅出。
1200字左右。
```

```text
把改写后的文章，翻译成英语，并用英国人喜欢的风格编写。
```

“英国人喜欢的风格”比较宽泛。整理补充：可以进一步说明为“使用英式英语拼写，面向普通读者，表达自然，保持原文术语和事实不变”。先核对中文版，再翻译，减少错误在两种语言中传播。

**政策解读，随后制定工作计划。**

```text
我是一家武汉的文创公司，解读一下上面这个链接中的政策，找到政策导向。
制定出【相约春天赏樱花】的文创产品策略。
```

```text
针对这个政策与策略，设计【樱花主题徽章】，帮我制定一份工作计划。
涉及到【人员、时间、成本】，【策划、设计、营销、售后】等，需要有专业度。
为了方便管理，也可以一些【报表】的形式。
```

把流程拆成“读材料 → 提炼导向 → 提出策略 → 变成计划表”。计划表至少能回答任务、负责人、起止时间、交付物、预算与检查点；这些字段是对原课计划要求的整理补充。


### 2.5 从 Markdown 到 XMind

来源：《AI文本处理大师》5 页。

原课流程：**生成 Markdown 文档 → 保存为 .md → 导入 XMind → 保存或导出思维导图**。

```text
我要做一个【电商视觉设计的课程大纲】，生成xmind思维导图，并且以markdown格式输出，方便粘贴。
```

```text
读取【PDF】中的内容，帮我生成xmind思维导图，并且以markdown格式输出，方便粘贴。
```

这里让 AI 输出的是 Markdown 大纲；再由 XMind 导入结构。只把文件扩展名改成 `.xmind` 不会生成有效的 XMind 文件。

补充练习：可导入的大纲结构示意，具体内容应来自所读材料。

```markdown
# 电商视觉设计课程
## 产品视觉基础
### 受众与品牌定位
### 构图与配色
## 商品图制作
### 产品主图
### 场景图
## 营销素材
### 节日海报
### 短视频配图
```

如果 AI 输出带代码围栏，保存文件时取围栏内的 Markdown 正文。确保文件不是隐藏了扩展名的 `课程大纲.md.txt`。不同版本的 XMind 导入菜单可能不同，以当前软件提供的 Markdown 导入功能为准。


### 2.6 职场文档的三项要求与 5W2H

来源：《AI文本处理大师》5-6 页。

| 要求 | 原课要点 |
| --- | --- |
| 格式规范 | 标题层级、段落缩进、字体字号统一 |
| 逻辑结构 | 总分总框架、信息优先级、过渡衔接 |
| 语言表达 | 精准无歧义、简洁去冗余、专业术语规范 |

**5W2H** 用七个问题补齐项目条件：

| 要素 | 含义 | 写项目需求时问自己 |
| --- | --- | --- |
| What | 做什么 | 要完成哪件事，交付什么 |
| Why | 为什么 | 目的是什么，解决什么问题 |
| Where | 在哪里 | 地点、平台或范围是什么 |
| When | 何时 | 开始、截止与关键节点是什么 |
| Who | 谁 | 谁负责，谁参与，谁是受众 |
| How | 怎么做 | 方法、步骤与协作方式是什么 |
| How much | 多少资源 | 时间、费用、数量或验收标准是什么 |

**原表修正**：讲义第一张表把 Where 的提示写成“谁，由谁来做”，把 Who 写成“何处，在哪里做”；这里按正确含义整理。

讲义案例是新品发布：目标为提升品牌知名度，选择会展中心，安排在 10 月中旬，由市场部统筹、设计与媒体团队协作，采用线上宣传和现场展示；示例预算为 50 万元（场地 20 万、宣传 15 万、人员 10 万、其他 5 万）。它用于学习把项目要求写具体，不是本篇要求实际举办活动或投入这些费用。




### 2.7 文本处理课堂作业

来源：《AI文本处理大师》6 页。

用 AI 完成自我介绍的两个版本，并整理为 PPT：

1. 简单版：300-450 字。
2. 进阶版：1000-1200 字。
3. 将自我介绍制作成 PPT。

整理补充：先列真实经历、介绍场景、受众与希望强调的能力，再让 AI 组织内容；两版使用同一份事实材料。PPT 按“我是谁 → 相关经历 → 能力与作品 → 目标”分配信息，不把长文章直接铺满一页。

交付检查：字数符合要求、经历没有被编造、结构清楚、幻灯片便于口头讲述。这里整理作业要求，未自动生成个人经历或替用户完成自我介绍。


## 三、AI图像创意：节日海报

来源：《AI图像创意大师》1-4 页。

节日海报把**节日元素、品牌信息与情感表达**组合在一张图中。原课展示端午、青年节、教师节与中秋等效果，主练习是即梦中的中秋海报。

原课操作顺序：进入即梦的图像生成 → 写提示词 → 选择模型 → 选择画面比例 → 生成多张结果 → 挑选并调整。

原课推荐竖版 `9:16`。它适合本课的竖屏示例；横幅、方形封面或印刷海报应按实际用途设置比例。

当画面不符合要求时，先判断是主题、构图、色彩、人物还是文字出了问题，再修改相应描述。需要长期保持统一的项目，选定能稳定产出所需效果的模型后记录模型与设置。

**整理补充**：海报上的日期、活动信息、品牌字样应单独核对；如果生成文字不准确，可以先生成背景与主体，再在设计或剪辑软件中排版文字。






### 3.1 AI绘画提示词的四部分

来源：《AI图像创意大师》4 页。

**内容类型 + 描述说明 + 风格描述 + 负面描述（可选）**。

| 部分 | 应描述什么 | 示例关键词 |
| --- | --- | --- |
| 内容类型 | 想得到哪类图像 | 摄影、插画、素描、雕塑 |
| 描述说明 | 谁在什么环境中做什么 | 一家人围坐古风庭院，共享月饼 |
| 风格描述 | 光线、构图、画面质感与艺术风格 | 暖光、侧光、水墨、极简、赛博朋克 |
| 负面描述 | 不希望出现的内容或缺陷 | 多余手指、错误肢体、水印、模糊背景文字 |

原课常用正面词：`超高清、8K、大师级杰作、最佳质量、细腻`。
负面词覆盖：低质量、模糊、像素化、噪点、水印、签名、多余或缺失肢体、结构错误等。

**理解边界**：这些是描述词，不是质量保证。写“8K”不等于导出的文件真的有 8K 分辨率；实际像素尺寸由工具的输出设置决定。负面提示词也要服从创作目标：若追求胶片颗粒或动态模糊，就不要同时要求完全无颗粒、无模糊。

不要为凑关键词堆进彼此冲突的要求。先让主体和构图成立，再增加光线与风格细节。


### 3.2 中秋海报提示词练习

下面是**根据原课四部分结构补写的练习示例**，不是原稿逐字提示词。

```text
竖版国风插画海报，9:16。
一家人在古风庭院的圆桌旁共赏明月，桌上有月饼、茶盏和桂花，人物神态自然温暖。
远景为圆月与屋檐，近景有少量桂花枝，主体位于画面中下部，顶部留出标题排版空间。
以墨青色和暖金色为主，灯笼暖光，层次清楚，节日团圆氛围。
避免多余肢体、人物面部变形和杂乱的文字。
```

检查时按顺序看：主题是否一眼可辨 → 主体是否突出 → 留白能否排标题 → 人物有无明显错误 → 配色是否统一。

例如，若桌上缺少月饼，补充其位置与外形；若所有人物挤在一起，调整人数、景别与构图；若标题位置被占满，强化顶部留白。每次尽量解决一个明确问题，保留较好的版本用于比较。


### 3.3 图片修复与高清放大

来源：《AI图像创意大师》5 页。

原课展示百度 AI 图片助手的图片处理入口，包含修复、去除画面中的不需要元素、清晰度增强等功能。

整理后的使用顺序：选择自己的练习素材 → 指定需要修复的区域或处理目标 → 查看修复结果 → 比较边缘与细节 → 保存所需尺寸。

放大可以提高输出尺寸，但细节可能经过重建；原本模糊的字或纹理不一定被准确还原。涉及商品标签、活动信息时，仍应核对具体内容。对外使用图片时，保留原素材授权要求，不把去掉标识理解为获得使用权。

原课“免费、不需要积分”的描述属于当时界面信息，实际功能和费用以当前工具为准。




### 3.4 人像写真：五部分描述与参考图

来源：《AI图像创意大师》6-12 页。

原课结构：**主谓宾 + 环境背景 + 镜头视角 + 表现风格 + 比例质量**。

| 部分 | 人像图需要明确的条件 |
| --- | --- |
| 主体与动作 | 人物、服装、表情、姿态与手持物 |
| 环境背景 | 室内/室外、场景元素、前后景和光线 |
| 镜头视角 | 俯视/平视/仰视、远景/中景/近景/特写 |
| 表现风格 | 写实、国风、油画、插画、胶片质感等 |
| 比例质量 | 横竖比例、细节方向和实际导出尺寸 |

原课用“人像写真”入口上传参考图，再生成单人或合照。参考图用于约束外貌，仍要检查输出是否保留关键特征，以及面部、手部、服装是否自然。

课堂效果包括竹林古风、红色婚纱、西域沙漠、舞蹈、武侠和自拍合照。人物身份、外貌特征与画面风格可以分别描述，再组合成提示词。使用自己的或已获授权的参考照片；合成画面与真实拍摄的身份、事件信息分开记录。






### 3.5 原课写真提示词与风格比较

来源：《AI图像创意大师》8-12 页。

**红色婚纱：重点是人物、服装细节与柔和光线。**

```text
画面主体为一位身着红色婚纱、露肩女子半身特写，妆容精美、面容姣好，眼神温柔幸福。
婚纱材质细腻，有精致刺绣或装饰，凸显国风之美。
背景简洁或带有淡淡国风元素且模糊，以突出主体。
色彩以红色为主，搭配金色或白色装饰增加华丽感。
柔和光线从侧面或正面照亮女子面容和婚纱，营造出温馨浪漫氛围。
整体充满国风美学的浪漫与庄重，展现女子美丽与优雅，画质高质量，画面清晰细节丰富。
```

**西域沙漠：重点是环境、人物姿态与逆光氛围。**

```text
西域公主，深红色长袍，红色色调，肤白如雪，长发飘飘随风吹动，仙气飘飘，绿色眼眸深邃。
坐在地上，全身，背景是沙漠、驼队，画面唯美，宛如仙境，画面清晰、锐化、丁达尔光线。
```

其余案例按视觉要素归纳：竹林古风侧重竹筏、竹叶前景和自然光；沙漠舞蹈侧重飘动服饰、身体动作和逆光；武侠侧重持剑、飞雪、灰暗色调、动态模糊和水墨晕染；随手自拍侧重非正式构图、轻微运动模糊与不均匀打光。

原课的泳池写真、合成明星合照属于效果展示。本篇提炼构图与光影方法，没有把合成图当作真实事件。人物案例之间不要默认套用同一份“万能负面词”：武侠图中的模糊与颗粒感本来就是风格要求。


## 四、AI视频编导：香水产品案例

来源：《AI视频编导大师》1-3 页。

原课流程：**即梦生成产品图 → 即梦图生视频 → 剪映合成**。

先让产品图成立，再生成运动。原课强调“图像效果是最终 AI 视频的关键”：主体位置、产品外形、画面光线和环境若已出错，直接生成视频通常会延续问题。

原课产品图提示词：

```text
一瓶“香奈儿可可小姐馥郁香水”放置在一片茂密的森林中，周围环绕着各种植物、茉莉花、玫瑰花。
整个画面以蓝色调为主，倒影、水波纹增加画面的纵深感和质感。
瓶子被聚焦在画面中央，周围的环境包括植物和水都被虚化处理。
自然光源，明亮通透，商业拍摄，产品摄影，极简主义，磨砂质感，8K超高清，杰作。
```

这里保留原课产品名称作为示例。练习时可替换成自己的产品与参考图；检查瓶型、瓶盖、标签、倒影及透视关系。模型生成的标签文字不自动等于真实产品信息。

“多次抽卡”是反复生成、筛选与调整的课堂说法。每次记录保留的版本和修改原因，避免只增加生成次数而没有目标。




### 4.1 图生视频：把重点放在动作与运镜

来源：《AI视频编导大师》2-3 页。

原课建议图生视频阶段“弱化提示词”，重点描述**怎么动、镜头怎么走**。参考图已经交代主体和环境，就不必再重写大量外观细节。

推荐示例是**希区柯克变焦（Dolly Zoom）**：摄影机前后移动，同时反向改变焦距，让主体在画面中的大小相对稳定、背景透视产生变化。它是明确的视觉目标，AI 生成效果仍要逐帧检查。

补充练习：

```text
使用参考图中的香水瓶与森林场景。
镜头缓慢推进，并产生希区柯克变焦效果；香水瓶保持画面中央，瓶身形状与标签稳定。
水面轻微波动，背景柔和变化，不增加新的主体。
```

先判断瓶子是否变形、标签是否闪烁、倒影是否合理，再判断运镜是否符合预期。动作太多容易互相干扰，可以先保留一个主要镜头运动。

原课文字素材为 `PERFUME` 和“一起解锁夏日美好”。整理补充：用剪辑软件单独排字，便于改内容、位置和动画；本页产品案例只需理解合成方法，不要求生成相同的商业成品。




## 五、AI微缩景观：从脚本到四个场景

来源：《AI微缩景观视频》1-3 页。

**微缩景观**是把真实场景按比例缩小，用模型表现城市、自然、人物活动等。在本课的 AI 图像中，通过“小人物 + 巨大食物/物件 + 微距细节”形成缩小世界的视觉效果。

原课介绍的用途包括直观展示、教育理解、艺术装饰和动态互动。视频的关键是让这个小世界中的人物行动有意义，而不是只显示几个静态摆件。

原课流程：

```text
DeepSeek：创意与脚本
    ↓
即梦：生成各场景图像
    ↓
可灵：把图像生成视频片段
    ↓
剪映：裁剪、配音、配乐、文字与导出
```

本课中秋主题是从“制作月饼”发展到“团圆庆祝”。先确定故事，再逐个做图，能够减少素材之间只有风格相似、却没有前后关系的问题。




### 5.1 脚本提示词的六个条件

来源：《AI微缩景观视频》3 页。

结构：**角色 + 任务 + 主题 + 核心约束 + 风格参考 + 负面提示**。

| 条件 | 本课中的作用 |
| --- | --- |
| 角色 | 告诉 AI 从编导、视觉叙事等角度组织内容 |
| 任务 | 指定写脚本、分镜表或绘画提示词 |
| 主题 | 保持故事围绕中秋与团圆 |
| 核心约束 | 时长、场景数、字数、人物、比例 |
| 风格参考 | 对称构图、古风、复古或暖色等视觉方向 |
| 负面提示 | 排除现代元素、悲伤结局或不需要的广告感 |

原课四场景提示词：

```text
我是一名AI视频创作人员，想做一个中秋节视频脚本，1:20的微缩景观，从做月饼到吃月饼的4个场景。
一群中国古风人物在做一个大大的月饼，中国古风。
以表格形式呈现，并且给出对应的中文AI绘画提示词，需要保证这4个场景的统一性与连贯性。
```

这里的 `1:20` 是微缩比例，**不是 1 分 20 秒的视频时长**。时长需要另行写清楚。

原课另一个示例：

```text
你是一位擅长电影感视觉叙事的导演。
请编写一个60秒的短视频脚本，主题是“中秋月圆人团圆”。
脚本需包含5个场景，总旁白字数控制在180字以内。
视觉风格参考韦斯·安德森的对称构图和复古色调。
避免出现悲伤的结局和商业广告感。
```

这是**另一套五场景需求**，不要把它的“五个场景”直接混到前一个四场景案例中。

### 5.3 人物、色彩与空间一致性

来源：《AI微缩景观视频》4、7 页。

| 一致性维度 | 原课固定规则 |
| --- | --- |
| 核心角色 | 白须老匠：青灰短打与木槌；束发妇人：赭色襦裙与铜尺；双髻孩童：红肚兜与坚果篮 |
| 主色 | 饼体金棕 `#d4a017`，场景黛青 `#2a5c6a` |
| 点缀色 | 工具朱红 `#c3272b`，夜光青白 `#e0f7fa` |
| 共同背景 | 左后侧六角粮仓贯穿场景1-4；右前侧桂花树从花苞发展到盛放 |
| 衔接设计 | 场景1→2：面粉飞溅转为测量尺刻度；3→4：热气衔接烟花；结尾：孔明灯聚成圆月 |

**一致性**：各镜头保留相同人物特征、风格与关键道具。
**连贯性**：按故事顺序推进，上一镜头的状态能解释下一镜头。
**创意性**：在统一规则之内，让动作、尺度和细节有亮点。

整理补充：建立一段固定的“共同设定”，每个场景都带上它；只替换该场景的动作、景别与环境状态。支持参考图或角色参考的工具，可使用相同参考素材，但仍要检查实际输出，不能只靠“保持一致”四个字。

原稿颜色代码可以作参考，不能保证模型生成像素严格匹配这些色值。必要时在后期统一调色。


### 5.4 脚本的五类常见问题

来源：《AI微缩景观视频》4-6 页。

| 问题 | 原课给出的改法 |
| --- | --- |
| 指令空泛，只有“写中秋脚本” | 补受众、平台、时长、风格、主题亮点，可借助5W2H |
| 完全接受初稿 | 人工删减、重组、润色，补充自然的情感变化 |
| 只有旁白，没有画面 | 每一幕要求画面描述、运镜建议与视觉元素 |
| 多种矛盾风格混杂 | 确定一个核心调性，在后续提示中持续保留 |
| 长度失控 | 写清总时长、场景数量和旁白字数 |

原课把旁白每分钟约180-200字作为估算参考。语速、停顿、人物对白和画面留白会影响成片，最终用实际配音长度核对，不能只按字数认定时长符合要求。

好的脚本不仅能读，还要能指导做图、生成片段和剪辑。每个镜头都应知道主体是什么、发生什么、怎么动以及持续多久。


### 5.5 分步骤迭代与专业描述词

来源：《AI微缩景观视频》5-6 页。

原课分层迭代：

1. 创意定位：以治愈系微缩景观编导的角色提出3个中秋创意。
2. 大纲细化：选择“月饼诞生记”，拆成4个场景。
3. 内容填充：扩展成60秒脚本，加入画面、旁白与音效建议。

提供参考时，说明要借鉴哪一层：叙事节奏、视觉配色、镜头语言或三幕式“遇到问题 → 解决问题 → 圆满结局”。只写作品名称，容易让模型抓错重点。

| 描述方向 | 原课术语与作用 |
| --- | --- |
| 运镜/拍摄 | Dolly shot（推轨）、Time-lapse（延时）、Slow motion（慢动作）、Handheld shaky cam（手持晃动）、Drone dive（无人机俯冲）、Macro shot（微距） |
| 光影 | Tyndall effect（丁达尔效应）、Cyberpunk neon（霓虹）、Warm tone（暖色）、Chiaroscuro（明暗对照） |
| 剪辑节奏 | Fast-paced editing（快节奏剪辑）、Jump cut（跳切）、Fade to black（渐隐黑场） |

这些描述并不都属于摄影机移动：微距是视角/拍摄尺度，慢动作与延时关系到时间表现，跳切和黑场通常由后期剪辑实现。

原课的负面提示示例：避免夸张辞藻、现代手机电脑、说教式结尾。让负面条件服务项目目标，不要机械复制一长串。

预留后续接口：旁白可以交给语音合成；关键画面描述用于文生图；需要后期叠加的特效在脚本中明确标注。


### 5.6 图像与视频生成的验收

来源：《AI微缩景观视频》6-8 页。

**图像阶段**：按场景复制提示词，生成后及时“缺哪补哪”；核对比例、风格、人物、道具与前后关系。先把四张关键图稳定下来，再生成运动。

**视频阶段**：同样需要多次生成与筛选。除了画面好看，还要符合可理解的事物规律：人物不应突然消失，月饼不应无原因变大变小，工具应与动作对应。

| 原课推荐工具 | 课堂评价 | 当前练习如何比较 |
| --- | --- | --- |
| 可灵 | 效果较好、消耗快、非会员等待较长 | 动作是否稳定，生成成本与排队是否可接受 |
| 即梦 | 从图到视频方便 | 与已生成的图像衔接是否方便 |
| VIDU | 功能较多，课堂观察等待较短 | 是否支持所需参考与输出方式 |

以上是原课阶段经验，不是当前固定排名。先用同一张图与一个简单动作测试，再选择适合项目的工具。

整理补充：长动作可拆成短片段，在剪辑中组合；当工具单次时长不够时，不要靠一次生成硬塞入整段故事。




## 六、剪映合成：让片段成为完整视频

来源：《AI微缩景观视频》8-12 页。

原课流程：**导入视频 → 裁剪与变速 → 添加音乐/旁白 → 文字与动画 → 导出**。

先把片段按脚本放到时间轴，再处理节奏。画面轨道、音乐轨道、旁白与文本的时间位置相互配合；开头、转场和结尾都要检查。

| 常用操作 | 原课快捷键/方法 |
| --- | --- |
| 选择与移动素材 | `A` 选择工具 |
| 切换分割工具 | `B` |
| 在播放头位置分割 | `Ctrl+B` |
| 加入音乐或素材 | 点“+”或拖到时间轴 |
| 初学者调整时长 | 常规变速 |

快捷键以当前剪映版本与焦点位置为准。先选中需要操作的轨道/片段，并确认播放头位置，再进行分割。

变慢可能出现卡顿，原课建议按需要启用智能补帧。补帧后的运动也要检查，避免人物边缘和物件形状出现异常。

原课提及“绿色版永久免费会员”下载，不作为本篇的安装步骤。软件使用官方版本；会员与免费功能按当前账号实际可用范围选择。






### 6.1 音乐、音效与旁白

来源：《AI微缩景观视频》10-11 页。

原课步骤：音乐库选曲 → 下载所需素材 → 点“+”或拖到轨道 → 对齐画面 → 调整音量 → 必要时淡入淡出。

喜欢的音乐可收藏。带钻石标记的素材属于讲义界面中的会员功能，选用前查看实际账号是否可用。

整理补充：旁白出现时，让背景音乐保持在能听清人声的音量；切换场景时对齐动作与节拍。烘烤、搬运、敲打等可以使用对应音效，但音效不应让画面节奏变得杂乱。

检查完整播放过程：有没有断音、突然变响、旁白被音乐盖住或声音延续到空白画面。淡入淡出用于让开头与结尾自然，不必每一段都套用相同长度。




### 6.2 文字、字幕与动画

来源：《AI微缩景观视频》11-12 页。

原课区分两种文本方式：

| 方式 | 优点与边界 |
| --- | --- |
| 文字模板 | 快速套用设计与动画；可改文字内容，样式调整范围受模板限制 |
| 新建文本 | 调整字体、大小、位置、颜色，以及入场/出场动画 |

课上提到的“模板不能调整样式”应按具体模板与当前版本理解，不作为所有模板的统一规则。

整理补充：标题说明主题，字幕帮助理解旁白，装饰文字营造氛围；三者不必同时铺满画面。先保证能读清，再考虑动态效果。字幕时间应与讲话对应，不能因套模板遮住主体或超出画面边缘。

像产品案例中的 `PERFUME`、中秋故事中的“月饼诞生记”，适合在合成阶段单独添加，便于控制排版与时序。




### 6.3 导出与成片检查

来源：《AI微缩景观视频》12 页。

原课要求：设置好**标题与导出位置**，导出分辨率不低于 **1080P**，等待导出完成。

整理补充：竖版9:16常见的1080级输出是1080×1920，横版16:9常见为1920×1080；以当前软件显示的具体宽高为准。项目、生成素材与导出视频的比例应配合，避免意外裁切。

视频导出后，重新打开最终文件从头到尾播放，检查画面比例、字幕、声音、总时长、黑场和结尾。文件名、保存位置与剪映工程文件也要明确，便于以后修改。

提示词里写“8K”，并不能代替这里的导出尺寸设置；把低分辨率素材放进高分辨率工程，也不自动得到同等细节。




## 七、原课作业与交付要求

来源：《AI微缩景观视频》12-13 页，结合《AI文本处理大师》6 页。

| 作业 | 课堂资料/主题 | 需要完成的内容 |
| --- | --- | --- |
| 文本作业 | 自我介绍 | 300-450字简单版、1000-1200字进阶版、PPT |
| 视频作业1 | 万物生花、漫步古镇 | 视频导入、剪辑、音频、文字、贴纸与导出 |
| 视频作业2 | 端午节微缩景观、悯农少儿诗词、武侠 | 视频、音乐音效、文字字幕、动画与旁白剪辑，标识设计和导出 |

原课附件名包括 `万物生花.zip`、`漫步古镇.zip`、`端午节-微缩景观.zip`、`悯农-少儿诗词.zip`、`武侠.zip`。讲义显示的是附件名称与任务要求；本篇没有把这些素材包写成已完成的作品。

整理补充的最小练习：用“月饼诞生记”完成一份四场景脚本、一份固定人物设定、四张同风格图与对应片段，合成为一条可完整播放的视频。保留脚本、提示词、原素材、工程与导出文件，便于发现问题后回到对应环节修改。


### 7.1 讲义中的服务价格与积分示例

来源：《AI图像创意大师》2、6 页，《AI视频编导大师》1 页。

以下只记录原课用于介绍服务层次的数字，不作为当前市场报价、实际收益或费用承诺。

| 类别 | 原课层次 | 原课数字与示例范围 |
| --- | --- | --- |
| 节日海报 | 基础 / 精品 | 50-200元 / 200-1000元以上 |
| 写真 | 经济型 | 200-600元；1套服装、1个简单场景，30-50张底片，精修3-8张 |
| 写真 | 标准型 | 800-1500元；2-3套服装、2个场景，60-100张底片，精修15-20张 |
| 写真 | 高端定制 | 2000-5000元以上；4套以上服装、多场景，100-200张底片，精修30-50张 |
| 视频视觉服务 | 基础 / 创意 / 影视 | 800-2000元 / 3000-5000元 / 1万元以上 |

写真表里的服装、底片与拍摄场景是原课对服务套餐的类比，不能直接等同于 AI 生成服务的必然交付。视频讲义以简单模板、定制动画/特效、高规格宣传区分层次。

即梦页面的原课示例是每天66积分，生图一次1-2积分，生视频一次10-20积分。工具政策与不同模型耗用会变化；估算项目成本时，应查当前实际积分，并把多次生成、失败重试与后期修改算进去。


## 八、复习与跟做检查

- [ ] 能区分 AI、AIGC 与多模态，不把“生成内容”理解成所有智能任务。
- [ ] 写提示词时交代任务、受众、背景、格式与主要约束。
- [ ] 能用追问定位问题，生成后人工核对事实、结构、字数与表达。
- [ ] 知道 Markdown 大纲需由 XMind 导入，不是直接改成 .xmind 后缀。
- [ ] 5W2H 中 Who 对应人员，Where 对应地点/平台。
- [ ] 能写出绘画提示词的主体、环境、构图、光线与风格。
- [ ] 负面提示词与想要的艺术效果不冲突，实际输出尺寸已核对。
- [ ] 产品图的外形、标签和倒影成立后，再生成动作与运镜。
- [ ] 理解微缩比例1:20与视频时长是两个不同条件。
- [ ] 四场景故事顺序清楚，人物、色彩、道具与空间有共同规则。
- [ ] 视频中没有明显的突然消失、变形或无原因缩放。
- [ ] 剪辑中的音乐、旁白、文字与镜头节奏相互配合。
- [ ] 最终视频按要求导出，并重新打开完整播放检查。
