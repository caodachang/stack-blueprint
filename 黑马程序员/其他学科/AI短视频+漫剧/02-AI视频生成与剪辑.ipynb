## 一、AI生成和剪辑分别解决什么问题

**整理补充，对应原课“视频认知”和“视频生成”的学习主题。**

AI 视频生成负责根据文字、图片等条件生成随时间变化的画面；剪辑负责挑选、排列和组合已有素材，建立内容、时间与声音之间的关系。

| 环节 | 输入 | 输出 | 重点检查 |
| --- | --- | --- | --- |
| 脚本 | 主题、受众、时长、风格 | 镜头表、旁白、提示词 | 每个镜头是否有明确目的 |
| 生图 | 画面描述、参考图 | 关键帧或场景图 | 构图、人物、风格和留白 |
| 生视频 | 文字、参考图、运动要求 | 单个动态片段 | 动作、运镜、形变与稳定性 |
| 初剪 | 筛选后的片段 | 能完整观看的内容顺序 | 内容是否成立、时长是否合理 |
| 精剪 | 初剪结果 | 节奏、声音、文字更完整的成片 | 衔接、可读性与视听统一 |

**生成完成不等于成片完成。** 一段画面可能很好看，却没有叙事用途；一个镜头前半段合格，后半段可能出现变形。剪辑时可以只保留其中有价值的部分。

例如：生成了五段各约 5 秒的风景，素材总长约 25 秒。若目标是 20 秒，就应根据内容挑选有效区间，而不是要求每段都完整放入。


### 1.1 常见生成方式

**整理补充。** 下表描述通用概念，不代表每个平台或每个版本都有全部入口。

| 方式 | 怎么输入 | 适合做什么 | 常见问题 |
| --- | --- | --- | --- |
| 文生视频 | 用文字描述画面与动作 | 试创意、试运镜、生成独立场景 | 人物与布局较难精确控制 |
| 图生视频 | 给一张图，再描述动态 | 保留已确定的构图和主体，让图动起来 | 大运动可能导致结构改变 |
| 首尾帧生成 | 提供起点图和终点图 | 尝试从 A 状态过渡到 B 状态 | 两图差异过大，过渡容易变形 |
| 多图参考 | 提供人物、场景或风格参考 | 提高跨镜头的一致性 | 参考冲突，或模型只保留部分特征 |
| 视频参考或视频编辑 | 用现有视频作为条件 | 参考动作、改变局部或风格 | 具体能力和权限因平台而异 |

本篇推荐练习顺序：先确定一张合格画面 → 生成一个简单运动 → 筛选有效片段 → 再做多镜头衔接。

**不要把图生视频理解为只给图片加平移。** 模型会生成新的中间画面，因此也可能改变手指、脸、屋顶、船和文字等细节。


### 1.2 先确定成片规格

**整理补充，下面是练习起点，不是原课的固定交付参数。**

| 项目 | 竖屏短视频练习 | 横屏城市景观练习 |
| --- | --- | --- |
| 画面比例 | `9:16` | `16:9` |
| 导出分辨率示例 | `1080 × 1920` | `1920 × 1080` |
| 目标时长示例 | 20-30 秒 | 20-30 秒 |
| 帧率 | 按项目与主要素材统一，常用起点为 25 或 30 fps | 同左 |
| 文字安排 | 给标题、字幕预留区域 | 注意画卷和城市名的位置 |

`9:16` 是宽高比，`1080 × 1920` 是像素数量，`30 fps` 是每秒帧数，20 秒是时长。它们解决不同问题。

生成提示词中的“8K”“电影级”通常是风格表达，**不会把低分辨率素材直接变成真正的 8K 视频**。画面尺寸、帧率和导出质量应在生成与剪辑界面中确认。

竖屏项目使用横屏素材时，先决定裁切、留边或重新生成。人物脸部、手部、重要地标和字幕都应留在安全区域内。

### 3.1 分镜表中每一列解决什么问题

**整理补充。** 分镜不是把文章平均分成几段，而是把内容转成能拍、能生成、能剪的镜头任务。

| 字段 | 应写什么 | 为什么要写 |
| --- | --- | --- |
| 镜号 | `S01`、`S02` 等稳定编号 | 关联图片、视频、旁白和修改记录 |
| 成片时长 | 例如 0-4 秒 | 控制整体节奏与总长度 |
| 画面内容 | 主体、环境、景别与构图 | 决定图像生成的内容 |
| 主体动作 | 谁以什么方式做什么 | 决定人物、船、植物等如何运动 |
| 运镜 | 推进、横移、环绕、拉升等 | 决定观众如何观看场景 |
| 旁白与字幕 | 本镜头承载的信息 | 避免画面和文字各讲各的 |
| 音乐与音效 | 气氛、动作声、环境声 | 建立空间与情绪 |
| 衔接要求 | 出画方向、落点、遮挡物 | 提高下一镜头接上的可能性 |

“一位老人很思念家乡”是情绪目标；“老人站在江边，目光从白鸟移向远山，手指轻捻胡须”才是可生成的画面动作。

一个短镜头先安排一个主要动作。把“起身、奔跑、拔剑、跳跃、落地、回头说话”同时塞进几秒钟，模型更容易遗漏动作或破坏人物结构。


### 3.2 可直接修改的脚本提示词

**整理补充模板。** 可以交给文本工具先生成表格，再人工删改。

```text
你是一位短视频分镜编导。
请为“宋代清明节微缩景观”设计一条20秒竖屏短视频。

受众：喜欢传统文化和国风视觉的观众。
基调：清雅、温润、节奏舒缓。
统一风格：宋代工笔水墨、微缩模型质感、轻烟雨。
画幅：9:16。

输出5个镜头，每镜头成片约4秒。
请用表格输出：镜号、成片时间、景别、画面内容、主体动作、
镜头运动、图片提示词、视频提示词、字幕、配乐与音效。

约束：
1. 场景为青瓦白墙、拱桥、河流、柳树组成的同一片江南空间。
2. 每个镜头只有一个主要动作，运镜以缓慢为主。
3. 人物服装、场景材质、光线和色调保持统一。
4. 不出现现代建筑、电线、车辆与明显现代服饰。
5. 不让AI画面直接承担可读的长篇文字，字幕留给剪辑阶段。
6. 镜头时长总和必须为20秒。
```

拿到结果后先检查时间是否相加等于 20 秒，再检查每个镜头能否独立生成。旁白过长时，先精简文字或延长镜头，不要靠极快语速硬塞进去。


### 3.3 清明节微缩景观完整练习脚本

**整理补充，沿用原课已确认的风格，以下五镜头不是原课表格的完整转录。**

| 镜号 | 成片时间 | 画面与动作 | 运镜 | 字幕与声音 |
| --- | --- | --- | --- | --- |
| S01 | 0-4 秒 | 青瓦白墙与小河建立空间，薄雾缓慢流动 | 缓慢推进 | “烟雨入江南”；古筝淡入、细雨声 |
| S02 | 4-8 秒 | 拱桥上人物缓步，桥下小舟轻行 | 缓慢横移 | “春水过小桥”；轻水声 |
| S03 | 8-12 秒 | 石桌上的青团、白菊与酒盏，柳影轻动 | 轻微环绕 | “一口青团，几分春意”；音乐留白 |
| S04 | 12-16 秒 | 河岸人物放飞纸鸢，衣袖随风微动 | 稍微拉远 | “把春风放得更远”；风声与鸟鸣 |
| S05 | 16-20 秒 | 回到城郊全景，河流连接桥与屋舍 | 缓慢拉升后停稳 | “清明”；音乐与环境声淡出 |

前两镜头交代地点与生活，第三镜头给观众停顿，第四镜头带出春日活动，第五镜头收束主题。这样五个漂亮画面之间才有观看顺序。

若生成结果没有足够好的放纸鸢动作，可以换成“人物站在河岸，手持纸鸢，衣袖轻动”。先保证主体结构，再逐步增加动作难度。


## 四、把脚本变成风格统一的图像

**原课画面**：《04-AI视频生成》开头可以看到在即梦中处理竖屏江南微缩景观。画面中的图像参考描述是“图片1内容不变，风格与颜色改为图片2”，并展示了一组桥、河道、白墙灰瓦和柳树的候选图。



这一步强调：确定主体内容后，还要统一风格与色彩。截图中的模型版本、生成尺寸是当时界面信息，不作为所有练习的固定设置。

**整理补充的操作顺序**：

1. 先用 S01 生成一张能代表整条视频的样张。
2. 检查构图、画幅、材质、光线和色调，选出参考图。
3. 后续镜头沿用同一段风格描述，并在支持的入口添加参考图。
4. 仅改变镜头内容、景别或动作准备状态。
5. 将合格图按镜号保存，再进入视频生成。

参考图可以提高控制程度，但不能保证人物、建筑和布局绝对不变，仍然要逐张检查。


### 4.1 图片提示词和视频提示词分开写

**整理补充。** 图片提示词主要说明“这一帧是什么”，视频提示词主要说明“它怎么随时间变化”。

| 图片提示词关注 | 视频提示词关注 |
| --- | --- |
| 主体、服饰、道具、空间关系 | 主体动作、动作先后与幅度 |
| 景别、构图、光线和色彩 | 运镜方向、速度与结束位置 |
| 材质、时代、风格 | 水、烟、风、树叶等环境运动 |
| 给文字预留区域 | 维持人物、建筑、材质等约束 |

**S01 图片提示词示例**：

```text
宋代江南城郊微缩景观，青瓦白墙沿小河错落排列，石拱桥连接两岸，
近处柳枝垂入画面，远处山峦被轻雾遮掩。略微俯视构图，前中后景清楚，
宋代工笔水墨与细腻微缩模型质感结合，青灰与浅绿色为主，柔和阴天光线，
宁静、清雅、温润。竖屏9:16，底部保留简洁空间供后期添加字幕。
不要文字、水印、现代车辆、电线和现代高层建筑。
```

**S01 图生视频提示词示例**：

```text
镜头缓慢向石拱桥推进，运动平稳，构图保持原图的空间关系。
河面出现细小水纹，柳枝轻微摇动，薄雾缓慢流动。
保持白墙灰瓦、桥体轮廓和水墨微缩风格稳定，无突然切镜。
```

图片已经给出建筑和光线时，视频提示词无需反复堆积所有静态形容词。先让“缓慢推进 + 水纹 + 柳枝轻动”成立，再增加复杂要求。


### 4.2 一致性检查：不只检查脸

**整理补充。** 连续镜头中以下内容最好保持一致：

| 类别 | 检查项 |
| --- | --- |
| 人物 | 脸型、发型、服装、年龄、身高关系、配饰 |
| 道具 | 船的颜色、纸鸢形状、桌上物件、手持物的位置 |
| 场景 | 桥和河的方向、房屋布局、地标的主要结构 |
| 视觉 | 材质、色调、景深、光线方向与明暗 |
| 动作 | 起止姿态、视线、行进方向、手与物的接触 |

镜头 A 中小舟向画面右侧行进，镜头 B 突然反向，观众可能理解为掉头。除非脚本需要这个变化，否则应优先选择运动方向相容的素材。

人物换脸、手指多出、物体穿透等明显问题，通常应回到生成阶段解决；轻微色差、边缘裁切与镜头长度，则可以在剪辑阶段调整。

一个实用判断：**看单帧发现的问题，先检查图像；只在播放时出现的问题，再检查动作与运镜。**


## 五、图生视频：描述运动与变化

**整理补充，对应原课“AI视频生成”主题。**

提示词可以按下面的顺序组织：

```text
主体动作 + 环境运动 + 镜头运动 + 速度/幅度 + 保持不变的要素
```

例如，桥下小舟这一镜头：

```text
一叶小舟从石拱桥下缓慢划过，船身平稳，划桨动作轻柔，水面留下细小涟漪。
镜头沿河道向右缓慢横移，近处柳枝轻摆，远山薄雾缓缓流动。
保持舟的数量、桥体结构、建筑布局和宋代水墨微缩质感稳定，避免快速旋转与突变。
```

注意“小舟运动”和“镜头横移”是两种运动。两者方向和速度叠加后会改变观感；主体已经移动很明显时，可以先使用固定镜头测试。

生成前核对：图像是否选对、画幅是否合适、模型是否支持当前输入、时长是否满足镜头需求、是否有声音生成选项、预计消耗是否可接受。

生成后至少观看开头、中段和结尾，再决定是否加入候选素材。一个漂亮封面不能证明整段视频都合格。


### 5.1 常见运镜与适用情况

**整理补充。** 同一个运镜词在不同模型中可能得到不同结果，方向和速度最好一起写。

| 运镜 | 观众看到的变化 | 适用场景 | 提示词写法示例 |
| --- | --- | --- | --- |
| 推进 | 越来越靠近主体，空间关系发生变化 | 进入场景、突出细节 | 镜头缓慢向桥推进，停在桥前 |
| 拉远 | 主体变小，更多环境出现 | 揭示空间、收束结尾 | 镜头缓慢后退，露出完整河岸 |
| 横移 | 摄影机向左或向右移动 | 展示街景、跟随船只 | 沿河道向右平稳横移 |
| 环绕 | 从主体不同侧面观看 | 产品、静物、地标展示 | 围绕青团石桌轻微环绕 |
| 拉升 | 视点升高，逐渐俯视 | 城市景观、结尾全景 | 镜头缓慢升高，俯视完整微缩场景 |
| FPV穿梭 | 第一人称快速经过空间 | 城市穿越、奇观展示 | 向前穿过建筑间隙，进入下一场景 |
| 希区柯克变焦 | 主体大小较稳定，背景透视感变化 | 紧张、惊讶或视觉强调 | 主体大小保持稳定，背景产生纵深变化 |

推进和变焦并不相同：推进改变摄影机位置，变焦改变视角。AI 输出有时会混用效果，应以实际画面是否符合目标为准。

“高速俯冲、环绕、旋转、拉升、变焦”同时出现时容易失控。对舒缓的清明景观，先采用单一、缓慢的运动更容易形成统一气氛。


### 5.2 时长、候选版本与重试

**整理补充。** 生成设置随工具版本变化，以下是制作方法，而非固定平台参数。

1. 先用较短片段验证画面与运动方向。
2. 每个镜头保留少量候选，并记录参考图、提示词、模型与生成时长。
3. 对比候选时先看主体稳定，再看运动和画面美感。
4. 每次修改一个主要因素，如动作幅度或运镜方向。
5. 确定可用片段后，再扩展到其余镜头。

| 问题 | 优先调整 |
| --- | --- |
| 人物或桥体持续变形 | 先换更清楚的参考图，减少大幅动作 |
| 画面几乎不动 | 把主要运动写具体，检查是否选了正确入口 |
| 镜头高速乱转 | 删除冲突运镜，明确平稳与缓慢 |
| 动作没有完成 | 简化动作，或增加生成时长后再裁剪 |
| 后半段质量下降 | 只取前半段有效区间，必要时重生成 |
| 风格和前镜头差异大 | 统一参考图与风格描述，再做有限调色 |

生成素材可以略长于成片镜头，为入点、出点和转场留出余量。不同模型对首尾帧、时长和分辨率的支持不同，不应把教程里的选项当作永远不变的规则。


## 六、延伸案例：一镜到底穿梭城市

**来源：课程配套云笔记《AI一镜到底穿梭城市》，页面标注 2025 年 12 月 28 日修改。** 本节作为视频生成控制的延伸阅读。

原文操作路径：

1. 在即梦 AI 中选择“无限画布”。
2. 选择“Agent模式”，勾选“灵感搜索”与“创意设计”。
3. 选择“4.5图片模型”，画面比例 `16:9`。
4. 让 AI 按统一格式编写六个城市地标的提示词并生成图像。
5. 切换为“视频生成”，将生成图片“添加到对话”。
6. 原文使用“视频3.0 Fast”模型，并输入连续穿梭运镜提示词。



这个示例封面包含广州塔、岭南建筑、烟花和节庆人物。画面中的城市字样有重复，提醒我们：即使整体视觉已经成立，也要单独检查生成文字，必要时由后期重新制作标题。

**原文边界**：Agent 模式可以帮助组织任务与复用上下文，但一致性仍要人工检查；不要把自动化流程理解为每次都能准确完成所有镜头。


### 6.1 城市景观提示词：先统一格式

**配套云笔记原提示词，标点和分行整理。**

```text
根据下面我给出的这段参考提示词，保持格式不变，
从南到北对6个城市地标建筑进行描述：

3D微缩景观，篆刻“广州”与“CHINA”文案，岭南建筑夜晚庆典场景，
关键元素为广州塔（银色螺旋造型高塔）、岭南骑楼建筑，
穿鲜艳服饰的人（于步道、广场上互动），及红色灯笼、暖光街灯；
场景是夜间节庆，构图采用略微俯视视角，广州塔居画面中心偏右，
周围建筑错落分布，前景人物动态丰富，背景烟花绽放；
色彩以深暗蓝夜空为基底，交织烟花的金红亮色、建筑的银/红/黄撞色、
灯笼的暖橙光，色调绚烂温馨；画面风格是写实微缩景观风，
细腻还原建筑纹理与人物神态，背景要素含缀星光的深夜空、绽放的烟花，
搭配微缩草地步道与广场，整体营造热闹喜庆的岭南庆典氛围。
```



**整理补充**：先检查输出城市与地标是否对应、南北顺序是否正确，再生成图片。保持格式不变的作用，是让每座城市都有相近的描述维度：地标、构图、环境、色彩和材质。

画面要求中若同时出现“广州塔中心偏右”和“进入左侧地标”的运镜，需要根据最终选图统一方向，不能机械复制。


### 6.2 视频提示词：把穿梭路线写清楚

**配套云笔记原提示词，分行整理。**

```text
FPV视角，电影级无人机镜头，从高空鸟瞰急速俯冲快速穿梭推进到某个建筑面前，
锁定此建筑进行快速环绕运镜，接着镜头继续向前穿梭，
从高空急速穿过建筑间的缝隙，拉近至下一个城市画面。
镜头向上拉升俯视整个城市微缩景观。
全程贯串高速飞行，与结合运镜抖动特效，突出环境细节。
```



原文提醒：提示词没有测试好，批量生成会造成较大浪费。页面中的“Abent”应理解为 **Agent**。

**整理补充的测试方法**：先选两座城市试一次“接近地标 → 穿过遮挡 → 进入下一城”的衔接。重点检查飞行方向、地标形状、速度和切换位置；通过后再增加城市数量。

“一镜到底”描述的是观看时连续穿梭的效果。AI 生成的连续感可能来自模型过渡，也可以用后期衔接改善，并不证明画面存在真实的、物理连续的拍摄路径。


### 6.3 城市画卷：把材质也纳入统一风格

**来源：课程配套云笔记《AI一镜到底城市画卷》。**

原文要求设计十个中国城市的提示词：超现实数字艺术、厚涂油画质感、微缩景观，每幅包括四种以上地标，空中有云雾等漂浮物，画布上有城市名字，并描述画布材质、笔触和景观细节。操作上先“图片生成”，原文选“图片4.5”，挑选合格画面后“添加到对话”，切换“视频生成”和“视频3.0 Fast”。



**校对说明**：页面正文有“微缩录观、微缩最观、画布材、烟节”等明显文字错误，本篇按上下文整理为“微缩景观、画布材质、细节”。视频提示词中的“螺旋馆冲、朦就、气流震领”等异常词不逐字当作有效术语。

**整理补充的清晰版本**：

```text
第一人称FPV视角，镜头向画面中的主要地标快速推进，
经过地标后沿屋檐方向转向，穿过朦胧烟雾，进入下一幅城市画卷。
最后镜头拉升，俯视完整微缩画卷。
保持厚涂油画笔触、画布材质与微缩景观比例统一。
只保留轻微的飞行震动，避免大幅旋转和突然跳切。
```

地标位于左侧还是右侧，应以选定的图片为准。城市名称需要准确时，优先用剪辑软件叠加文字，并校对字体、错字与重复字。


## 七、筛选和管理素材

**整理补充。** 把生成文件与剪辑选用文件分开，可以避免导入错误版本。

```text
清明微缩景观/
  01-脚本/镜头表.md
  02-参考图/风格参考.png
  03-关键帧/S01-v01.png
  04-生成视频/S01-v01.mp4
  05-选用视频/S01-selected.mp4
  06-声音/旁白.wav、雨声.wav、配乐.mp3
  07-剪辑工程/
  08-导出/清明微缩-v01.mp4
```

这里是命名示例，不要求与网盘原文件完全同名。保留原始文件，不要覆盖唯一一份合格素材。

| 镜号 | 版本 | 建议记录 |
| --- | --- | --- |
| S01 | v01 | 参考图、提示词、模型、生成时长、主要问题 |
| S01 | v02 | 修改了什么、是否改善、可用区间 |
| S02 | v01 | 与上一镜头的色调、方向、人物是否相容 |

可用片段的筛选顺序：主体完整 → 没有明显穿透变形 → 运动符合脚本 → 镜头衔接合理 → 细节美观。

原文件名中的随机编号不作为排序依据。按 `S01` 到 `S05` 导入，能减少后期对照脚本的成本。


## 八、初剪：先把内容顺序搭起来

**原课画面**：《05-AI视频初剪》约 00:27 附近展示剪映中的五段竖屏景观素材，素材区、画面属性、预览和主轨时间线均可见，五段画面顺次排列。



从这一布局可以理解初剪的目标：先把视频素材按脚本串成能看的结构。

**整理补充的初剪步骤**：

1. 新建工程，确定画幅与项目规格，使用能识别的项目名。
2. 导入选用素材和所需声音，先核对有没有混入错误版本。
3. 按镜号把视频放到主轨，播放检查镜头顺序。
4. 删除明显失败、重复、停顿或与主题无关的区间。
5. 根据分镜调整每段时长，形成目标时长附近的粗剪版本。
6. 暂时只放必要的旁白或节奏参考，先看内容是否成立。

不要因为素材生成了 5 秒，就认为成片必须使用完整 5 秒。原课示范脚本为 20 秒，而初剪画面的素材组合可以更长，**计划时长、生成时长和初剪时长是不同阶段的数据**。


### 8.1 裁剪、分割、排序与留余量

**整理补充。** 以下操作用动作名称描述，具体按钮与快捷键以当前剪映版本为准。

| 操作 | 用途 | 检查点 |
| --- | --- | --- |
| 拖动片段边缘 | 缩短开头或结尾 | 是否误删动作起点、字幕结束位置 |
| 在播放头位置分割 | 从中间去掉失败或冗余片段 | 分割点是否落在动作突变处 |
| 调整片段顺序 | 按脚本建立内容关系 | 景别、视线和运动方向是否衔接 |
| 缩放时间线 | 精确定位入点、出点 | 是否留有极短空隙或多余碎片 |
| 开关吸附等辅助功能 | 帮助对齐片段与声音 | 调整时有没有误碰其他片段 |

如果镜头需要转场，源素材两端通常应有可用余量。转场效果和时间线规则因软件而异，最终时长要看时间线显示，不能仅凭片段标注的秒数相加。

例如：S02 中船到第 4 秒才开始穿模，可保留前面稳定部分，再让下一镜头在动作或声音上接住。裁剪能减少问题暴露，但不能修复整个镜头都错误的结构。

初剪完成后关掉声音看一遍：没有旁白解释时，画面顺序是否仍然容易理解。再只听声音看一遍：有无突兀的音量或未完成的句子。


## 九、精剪：让镜头、声音和文字互相配合

**原课画面**：《06-AI视频精剪》开头展示了剪映转场面板，以及主轨上方的效果轨道；界面中有带 VIP 标识的效果项目。



**整理补充的精剪顺序**：

1. 先修节奏，让每个镜头在观众理解后及时进入下一镜头。
2. 再检查衔接，选择直接切换或适当转场。
3. 统一色调、明暗、裁切和主体位置。
4. 完成旁白、配乐、环境声与动作音效。
5. 添加标题和字幕，逐字校对并检查遮挡。
6. 最后决定是否需要特效，检查片头片尾与导出范围。

精剪不等于不断加效果。对于清雅水墨景观，强烈闪白、快速旋转和密集震动可能破坏主题气氛。

原课界面中出现的 VIP 效果不必机械照搬。采用可用的基础效果，也可以完成结构清楚、节奏合理的练习。


### 9.1 转场和关键帧怎么选

**整理补充。**

| 方法 | 适合情况 | 使用时注意 |
| --- | --- | --- |
| 直接切 | 动作、内容或景别本身能形成关系 | 检查动作是否突然中断 |
| 短叠化 | 舒缓情绪、景物变化、时间推移 | 两张画面重叠时是否杂乱 |
| 淡入淡出 | 开场、结束、明确段落变化 | 不要让所有镜头都频繁变黑 |
| 方向性转场 | 原镜头有明显移动方向 | 下一镜头方向应相容 |
| 遮挡衔接 | 柳枝、屋檐、烟雾暂时遮住画面 | 遮挡大小和位置要接近 |
| 关键帧 | 调整位置、缩放、透明度等 | 变化应平稳，起止状态应明确 |

关键帧可以用来给静态标题做轻微淡入，或让图片缓慢推近。但对已经具有明显运镜的 AI 视频再次大幅缩放，可能造成运动叠加与眩晕。

转场时长可先尝试短于镜头时长的一小部分，再根据实际观感调整。不要把某个秒数当成通用答案。

**检查方法**：把注意力放在两个镜头交界处，问“这个变化是内容需要，还是只是为了展示效果”。如果移除效果后表达更清楚，就可以直接切。


### 9.2 声音：旁白优先，环境声负责空间

**整理补充。**

| 声音层 | 作用 | 清明景观示例 |
| --- | --- | --- |
| 旁白或朗读 | 提供主要信息 | 诗句、主题文案 |
| 配乐 | 连接情绪与节奏 | 古筝或轻柔器乐 |
| 环境声 | 建立空间与氛围 | 雨声、风声、河水声 |
| 动作音效 | 让具体动作更可信 | 划桨入水、脚步声 |

推荐操作：先把旁白调到清楚可听，再降低配乐，让环境声只在需要的段落出现。没有旁白的纯景观视频，也应避免配乐和环境声互相抢占。

音乐和环境声在开头、结尾适当淡入淡出，避免突然开始或截断。镜头切换时可以让一段环境声略早进入或稍晚离开，帮助观众连接两个空间。

不要只看音量数值：戴耳机和用普通扬声器各听一遍，确认无爆音、明显底噪、忽大忽小和不必要的重复声音。存在响度规范的发布渠道，应按该渠道要求处理。

音乐、配音和音效能否公开使用，应以实际授权为准；课堂案例中的素材存在，不等于已确认适用于任何商业项目。


### 9.3 标题、字幕和画面文字

**整理补充。** AI 画面中的文字容易出现错字、缺笔、重复和前后变化，可靠的字幕应在剪辑阶段生成并校对。

1. 导入或识别旁白，先核对识别文本。
2. 按完整语意断句，再调整每条字幕的出现与消失时间。
3. 统一字体、字号、颜色与位置，检查明暗背景上的可读性。
4. 避开人物脸部、关键道具和地标。
5. 在接近手机实际观看大小时检查一次，避免只在放大的编辑器里觉得清楚。

清明练习可以用“烟雨入江南”等短句分镜出现，片尾留“清明”作为主题；不必同时放入长篇旁白、完整诗句、城市名和大标题。

标题素材如配套“楚风”PNG，导入后要检查透明背景、边缘、分辨率与画幅比例。艺术字是标题素材，不能替代所有正文和字幕。

少儿诗词类视频应特别校对诗题、作者、诗句与读音。自动字幕常把人名、古词和同音字识别错误。


### 9.4 画面统一与导出

**整理补充。**

先选一段作为色彩参考，逐段检查亮度、色温和饱和度。清明水墨景观不宜一个镜头青灰低饱和、下一个镜头突然变成强烈橙红夜景，除非脚本明确需要转换。

| 导出项 | 起点建议 | 必查问题 |
| --- | --- | --- |
| 画幅与尺寸 | 与工程一致；竖屏示例为 1080 × 1920 | 有没有误导出横屏或黑边 |
| 帧率 | 与工程和主要素材相容 | 是否无意义提高帧率或产生不顺畅运动 |
| 格式 | 常用 MP4，编码按播放端支持选择 | 目标设备是否能正常播放 |
| 质量 | 选择足以保留细节的质量设置 | 水墨细节、烟雾、渐变是否明显压缩 |
| 声音 | 包含所需音轨 | 是否漏旁白、重复声音或无声 |
| 导出范围 | 完整有效时间线 | 有没有多余黑尾、空白或碎片 |

**示例规格不等于原课固定标准**：20 秒、9:16、1080 × 1920、30 fps 可以作为一次练习的约定。若素材和项目以 25 fps 工作，保持相容通常比盲目提高数字更合理。

导出后用普通播放器打开成片，完整播放一次。编辑器预览正常，不能证明导出文件的字幕、音轨、范围和编码都正常。


## 十、原素材案例：杜甫诗句

**来源：第 02 模块“杜甫诗句”文件夹中五段 MP4 的文件名描述。** 原文件名末尾使用省略号，本节只整理可读的描述，不把缺失部分补成原文，也不据此声称完整观看了视频。

| 建议叙事顺序 | 原文件编号 | 文件名可读内容 |
| --- | --- | --- |
| 建立景物 | 9464 | 国风江南江畔，碧江、青山、红色山花与白鸟 |
| 人物反应 | 8418 | 杜甫捻须，目光追随白鸟，眉头微蹙，穿插白鸟和山花 |
| 情绪转折 | 1773 | 杜甫坐到江边石桌，笔墨纸砚齐全，望向远山，神情沉重 |
| 落笔表达 | 1418 | 蘸墨落笔写诗，穿插面部特写 |
| 文字收束 | 4625 | 黑屏淡入诗句和落款 |

这组描述体现了“景物 → 人物反应 → 情绪 → 表达 → 主题收束”的结构，而不是按随机编号从小到大排序。

**整理补充**：碧江、白鸟、青山、红花与思归情绪，可联系杜甫《绝句二首·其二》理解：

```text
江碧鸟逾白，山青花欲燃。
今春看又过，何日是归年。
```

诗题与完整诗句在这里是关联学习补充，文件名并未完整给出作品题名。做成片时仍要核对原素材、脚本和朗读版本。

“写全诗”的视频动作不代表生成画面中的字一定准确。可保留蘸墨与落笔动作，在后期单独添加校对过的诗句，并让朗读、字幕和画面情绪一致。


## 十一、完整复现练习

**整理补充练习，本篇没有替你实际生成或发布视频。**

目标：制作一条 20 秒“清明节微缩景观”短片，五个镜头，竖屏 `9:16`，有短句字幕、配乐和少量环境声。

| 阶段 | 需要完成的产物 | 通过条件 |
| --- | --- | --- |
| 策划 | 一张五镜头表 | 总时长20秒，画面与主题对应 |
| 生图 | 五张关键帧 | 风格、画幅、光线一致，主体清楚 |
| 生视频 | 每镜头至少一段候选 | 动作简单、主体稳定，有可用区间 |
| 筛选 | 镜号与选用区间记录 | 能说明为什么选这一版 |
| 初剪 | 一份结构完整的工程 | 无明显空隙，顺序清楚，接近目标时长 |
| 精剪 | 字幕、声音与适量转场 | 字幕可读、旁白清楚、视觉统一 |
| 导出 | 一个可播放的MP4 | 完整播放无黑尾、无漏音、无错字 |

第一轮先用“慢推进、慢横移、轻微环绕、拉远、拉升”各一个镜头。第二轮再任选一个镜头改成更复杂的运动，与第一轮比较失败率和观感。

完成后记录：哪张图最难保持稳定、哪个运镜最容易变形、哪个镜头删短后更好看、哪一处声音让衔接更自然。这个记录比只保存一次成功的提示词更有复用价值。


## 十二、常见问题排查

| 现象 | 先检查 | 处理方向 |
| --- | --- | --- |
| AI只生成漂亮风景，没有故事 | 是否有分镜目的和内容顺序 | 补充主体、动作、转折与结尾 |
| 五个镜头像五种画风 | 参考图、风格描述、模型版本 | 统一样张与提示词，再生成候选 |
| 画面中人物不断换脸 | 人物参考与大幅动作 | 固定人物设定，先减小动作 |
| 船穿桥、手穿物体 | 原图结构和动作关系 | 重选图或重生成，剪掉局部失败区间 |
| 视频动作和运镜相互冲突 | 是否同时写了太多运动 | 先保留一个主要动作与一种运镜 |
| 成片超过目标时长 | 素材是否全部完整使用 | 按信息价值剪短，检查转场占用 |
| 转场一加就显得乱 | 主题、方向、时长与效果强度 | 优先直接切或减少效果 |
| 音乐盖过旁白 | 各音轨的相对音量 | 先保证旁白，再调配乐与环境声 |
| 字幕出现错字 | 自动识别文本与古词人名 | 对照脚本逐字校对 |
| 导出后有黑边或黑尾 | 画幅、裁切、导出范围 | 回到工程修正，再重新导出 |
| 批量生成浪费积分 | 是否先验证样张和运镜 | 先测一镜头或两城市，再扩大数量 |

排查顺序：**原图是否合格 → 动态是否稳定 → 素材是否选对 → 时间线是否正确 → 声音文字是否完整 → 导出是否正常**。
