### 1.1 AI漫剧的重点：故事成立，画面连得上

**整理补充。** AI 漫剧常由生成图片、动态片段、旁白、对白、音效与剪辑共同构成。不同项目可以采用漫画、动画、写实或玄幻等视觉形式；不能只因“漫剧”这个名称就断定每个案例都是二维动画。

| 层次 | 解决的问题 | 合格表现 |
| --- | --- | --- |
| 故事 | 谁遇到什么问题，为什么采取行动 | 角色目标和因果关系清楚 |
| 画面 | 人物、环境、动作是什么 | 主体可辨，风格与身份稳定 |
| 镜头 | 观众何时看见哪些信息 | 远近、顺序与方向能帮助理解 |
| 声音 | 谁在说话，场景有什么声音 | 配音可懂，音效与动作对应 |
| 剪辑 | 将独立素材组成连续时间 | 衔接自然，重点不被打断 |

一段龙飞过城墙的画面可以是视觉素材；加入“城中人看见威胁 → 守城人作出选择 → 行动带来结果”，才建立更完整的叙事关系。

不是每个镜头都需要对白，也不是每个镜头都需要大动作。反应、停顿、细节和环境镜头都可以承担剧情信息。

## 二、AI漫剧剧本编写

**原课画面**：《01-AI漫剧剧本编写》预览片段展示豆包的新对话界面。可确认使用文本对话入口；预览画面未展示可完整读取的成稿，因此下面的模板与示例属于整理补充。



**整理补充：先确定六件事。**

1. **主角**：观众主要跟随谁。
2. **目标**：主角想得到、守住或改变什么。
3. **阻碍**：为什么不能直接达成目标。
4. **选择**：主角必须作出什么决定。
5. **结果**：决定带来什么变化。
6. **形式**：目标时长、画面比例、受众和视觉风格。

“古城、龙、雷电、电影感”是一组元素；“守城人必须在城门关闭前找回信物，否则城中人无法撤离”才开始形成行动目标与时间压力。

先限制故事规模：短片优先少角色、少场景、一个主要冲突。复杂世界观可以后续扩展，不必一次塞进一分钟。


### 2.1 一句话故事与人物动机

**整理补充。** 可以用下列结构检查构思：

```text
当【事件发生】，
【主角】为了【目标】，
必须在【限制】下完成【行动】，
却遇到【阻碍】，
最终通过【选择】带来【结果】。
```

自编练习《城门之前》：风暴将至，守城人为了让迟归的人进城，必须在城门关闭前确认一位陌生人的身份。他先把陌生人当成威胁，随后发现对方是在送回失落的信物，于是选择开门接应。

这个设定是练习，不是“大汉龙劫”的原课剧情。它适合少量角色和一个主要场景，能练习视线、反应、道具特写与动作衔接。

角色动机应通过可见行动表现。例如“犹豫”可以表现为手停在门闩上、看向远处、听见求救后回头；不必把所有心理活动都写成旁白。


### 2.2 可直接修改的剧本提示词

**整理补充模板。** 先要求故事成立，再要求输出镜头表。

```text
你是一名短片编剧，请为原创AI漫剧《城门之前》写一个约60秒的剧本。

受众：喜欢简洁奇幻故事的成年观众。
故事：风暴将至，守城人误把送回信物的陌生人当成威胁，
      在发现真相后决定开门接应。
角色：守城人、陌生人，两人均为成年角色。
场景：一处古城门及城门外的道路。
风格：写实奇幻，冷灰风暴背景，暖色灯火作为视觉对照。
约束：只保留一个主要冲突；不要新增复杂势力、回忆支线或第三位主角。
对白：短句，配合动作，保留自然停顿。

先给出：一句话故事、角色目标、阻碍、转折、结局。
再给出：场景剧本，包含可见动作、对白和声音提示。
最后检查：事件因果、角色动机、道具位置和总时长。
无法确定的内容请列为待确认项，不要假装已经完成视频制作。
```

得到初稿后重点改“不可见的心理描述”“突然出现的新设定”“太长的对白”和“无法在几秒内完成的连续动作”。

不要在一条提示中同时要求完整世界观、十位人物、打斗、反转、长对白和一分钟时长。先缩小故事，再逐步补充需要的细节。


### 2.3 剧本、分镜与生成提示词的区别

**整理补充。**

| 文档 | 面向谁 | 应写什么 |
| --- | --- | --- |
| 故事梗概 | 编剧与创作者 | 情节、人物动机、冲突和结果 |
| 场景剧本 | 导演与配音等环节 | 可见行动、对白、场景和声音 |
| 分镜表 | 生图、生视频与剪辑环节 | 景别、构图、时长、动作、运镜与衔接 |
| 生图提示词 | 图像工具 | 某一时刻的主体、环境、构图和风格 |
| 视频提示词 | 视频工具 | 从参考图开始发生的动作与镜头变化 |

“他意识到对方不是敌人”是故事信息；“镜头切到陌生人手中的信物，再切回守城人松开的手”是镜头任务。

生图不必描述接下来完整一分钟的剧情；生视频也不必重复图片中已经明确的全部细节。每个环节只提供完成当前任务所需的信息。


### 2.4 人物设定表：固定身份与可变状态

**整理补充。** 为主要人物建立稳定编号与参考图。

| 字段 | 练习示例 | 使用原则 |
| --- | --- | --- |
| 编号 | C01 守城人 | 文件和分镜都用同一编号 |
| 外貌 | 成年、短黑发、较清晰的眉眼特征 | 先确认参考图，不只依赖文字 |
| 服饰 | 深灰外衣、暗红肩巾、皮质腰带 | 跨镜头保持主要形状和颜色 |
| 道具 | 城门钥匙或信物 | 记录哪只手持有、何时转交 |
| 性格与动机 | 谨慎，负责保护城中人 | 决定对白和行动，不能只写外貌 |
| 可变状态 | 警惕 → 迟疑 → 决断 | 表情和动作可以变化，身份不变 |

参考图至少检查正面、侧面、全身与主要服饰。不同角度可以增加控制信息，但图片太多且相互矛盾也会干扰生成。

不要把“使用同一个随机种子”理解为一定保持同一人物。相同文字、种子或风格词都不是跨镜头身份一致的保证；参考图和人工检查仍然重要。


### 2.5 场景、道具与视觉基准

**整理补充。** 除了人物，环境也要稳定。

- **空间**：城门、道路、塔楼之间的关系；人物从哪边进入画面。
- **时间**：是风暴前、风暴中还是风暴后，不要无原因地晴雨切换。
- **光线**：主光来自灯火还是天空，人物和背景的照明是否一致。
- **道具**：信物的形状、材质、大小与持有者。
- **画面比例**：横屏或竖屏，在生成阶段就给人物动作与字幕留位置。

建立一张场景全景基准图，再生成局部镜头，比每镜重新描述一个不同城门更容易保持空间关系。

自编练习可使用“冷灰环境＋暖灯火”的方向，但这是练习选择，不是课程固定配色。风格要服务故事，不必为了“电影感”把所有画面压到看不清。


### 2.6 把 60 秒故事拆成八个镜头

**整理补充自编分镜，不是原课镜头表。**

| 镜号 | 时间 | 景别与画面 | 动作、声音与用途 |
| --- | --- | --- | --- |
| S01 | 0—6 秒 | 城门与风暴云层的远景 | 建立地点和威胁；风声与低强度音乐 |
| S02 | 6—12 秒 | 守城人中景，手扶门闩 | 他望向道路，表现警惕 |
| S03 | 12—18 秒 | 陌生人远中景，靠近城门 | 小幅向前移动，传来简短求助 |
| S04 | 18—26 秒 | 守城人近景 | 手停住，短句询问，建立选择 |
| S05 | 26—34 秒 | 信物特写 | 陌生人展示信物；揭示转折信息 |
| S06 | 34—42 秒 | 守城人反应近景 | 视线落向信物，表情放松 |
| S07 | 42—52 秒 | 城门与两人中景 | 开门接应，可拆为两段较简单动作 |
| S08 | 52—60 秒 | 门内暖光或城门收束镜头 | 故事结束，音乐收住，简短片名 |

时间合计 60 秒，是成片规划，不代表每个生成工具都能一次生成对应长度。较长段落可由多个有效片段组成，或根据素材调整时长。

每镜至少明确：主体、景别、动作、声音、前后连接点。图像生成阶段先做一个能代表该镜头的时刻，视频生成阶段再安排有限动作。


### 2.7 剧本审核：先修因果，再润色句子

**整理补充检查表。**

| 检查 | 应能回答的问题 |
| --- | --- |
| 目标 | 主角为什么要行动？ |
| 阻碍 | 什么让行动变得困难？ |
| 转折 | 新信息从哪个镜头被观众看见？ |
| 决策 | 主角的选择是否由前面信息推动？ |
| 结果 | 结尾是否回应开头的问题？ |
| 可制作性 | 哪个动作过于复杂，能否拆镜？ |
| 时长 | 对白是否真的能在指定时间内自然读完？ |

先把对白试读或生成试听，再决定镜头长度。文字工具估计的时间不能代替实际声音长度。

反转需要前面的信息支持；“突然出现一个新角色解释一切”往往增加制作负担，也削弱已有角色的行动。


## 三、AI图像与视频生成

**原课画面**：《02-AI图像与视频生成》约 `00:25` 展示可灵页面、参考图相关入口和图像候选；约 `00:55` 可见风暴、城墙与巨龙等写实玄幻画面。





能确认的是工具界面、候选图与主要视觉元素；画面未完整显示原课剧本，因此不能据此编造各镜头的人名、对白和剧情顺序。

**整理补充**：先确定合格图像，再生成动态片段。错误的人脸、道具、服饰或空间关系一旦进入视频生成，往往会在动态画面中持续出现。


### 3.1 生图提示词：固定项与镜头项分开

**整理补充模板。**

```text
【固定角色】C01守城人，以已确认人物参考图为身份依据，
成年、短黑发、深灰外衣、暗红肩巾，主要服饰与道具保持一致。
【固定场景】同一古城门，风暴将至，冷灰天空，门内有暖色灯火。
【本镜头】人物中景，站在门内侧，右手扶门闩，向画面右侧道路看去。
【构图】主体清楚，手与门闩可辨，给后续字幕留出空间。
【光线与风格】写实奇幻，灯火照亮局部脸部，服饰纹理自然。
【约束】不要增加第二把钥匙、额外手臂或无关现代物品。
```

生成时同时使用工具实际支持的参考图功能。文字里的“同一人物”只是意图，不等于工具已经收到该人物的图像。

固定项保持身份、场景和风格；镜头项改变景别、动作、视线与构图。每次修改记录原因，避免为修一个动作顺手改掉全部设定。


### 3.2 图像审核与角色一致性

**整理补充。** 用参考图逐项比较，而不是只判断“整体挺像”。

| 检查对象 | 需要看什么 |
| --- | --- |
| 人脸 | 发型、轮廓、眉眼、年龄感是否稳定 |
| 服饰 | 颜色、领口、肩巾、饰品和衣服层次 |
| 身体 | 手指、肢体连接、比例与动作可读性 |
| 道具 | 数量、形状、持有的手、大小与位置 |
| 场景 | 门的位置、建筑结构、道路方向与光源 |
| 风格 | 写实程度、色彩和材质是否属于同一作品 |

发现错误时，优先按工具能力局部修改或重生成。剪辑裁切可以隐藏少量边缘问题，但不能让变化的人脸自动恢复一致。

可以先做少量关键图验证设定，再扩展整个分镜。不要在人物参考还没确认时一次生成几十张图。


### 3.3 图生视频：一个镜头先做一个主要变化

**整理补充。** 视频提示词应说明“从这张图开始发生什么”。

```text
使用已确认的守城人中景图作为起点。
人物缓慢转头看向城门外，手保持扶住门闩，
肩巾随风轻微摆动。镜头保持稳定，仅有轻微推进。
脸部、服饰、城门结构与人物数量保持稳定。
不要突然出现新人物，不要让手与门闩融合。
```

这里的动作和数值都不是原课提示词。工具若不支持单独的负面提示入口，可在描述中简洁表达约束；约束是生成目标，不保证每次实现。

把“拔剑、奔跑、跃上城墙、打斗、落地、转身说话”塞进一个短片段，容易遗漏动作或产生形变。可以拆成准备、动作、结果三个镜头，再在剪辑中建立连续感。

主体现有运动较大时，镜头运动保持简单；镜头大幅绕行时，人物动作也应相对简单，降低同时变化的复杂度。


### 3.4 景别、动作与运镜分别控制什么

**整理补充。**

| 维度 | 例子 | 叙事作用 |
| --- | --- | --- |
| 景别 | 远景、中景、近景、特写 | 决定观众看到环境、动作还是细节 |
| 主体动作 | 转头、伸手、递信物、开门 | 推动事件与人物选择 |
| 运镜 | 推进、后退、横移、跟随 | 改变观看距离和角度 |
| 环境运动 | 风吹肩巾、烟雾、雨水 | 增加场景状态与氛围 |

特写适合让观众看清信物；远景适合建立城门与道路的位置。不是所有镜头都用近景，也不是所有场景都需要环绕运镜。

图生视频中的大角度变化会要求模型生成原图没有的侧面或背面，更容易改变身份和空间结构。需要强视角变化时，先准备合适的目标视角图或拆镜。


### 3.5 首尾状态、视线与道具连续性

**整理补充。** 生成每个镜头时，要提前想它怎样接到下一个。

| 连续性 | 示例与检查 |
| --- | --- |
| 视线 | 守城人看向画面右侧，后续对象镜头应让空间关系说得通 |
| 动作 | 上一镜开始举信物，下一镜特写接到同一动作阶段 |
| 道具 | 信物不能从右手无原因跳到左手，或突然改变大小 |
| 身体方向 | 人物进城方向与下一镜所在位置对应 |
| 光线与天气 | 同一段时间不要突然白天转夜晚、风暴转晴天 |

某个工具提供首尾帧功能时，可以尝试明确起止状态，但两图差异过大仍可能生成不合理的过渡。它不是人物、动作和空间完全连续的保证。

如果两段无法自然连接，先尝试插入有意义的反应或道具特写，再考虑改分镜或重新生成。转场特效不能自动补足缺失的因果。


### 3.6 视频筛选与有效区间

**整理补充。** 生成文件的总时长与可用时长不同。

| 检查 | 淘汰或修订原因 |
| --- | --- |
| 主体 | 脸变化、人物突然增加、道具变形 |
| 动作 | 运动与意图相反，手和物体融合，动作没完成 |
| 时间 | 某一时刻突然消失、闪变或跳跃 |
| 镜头 | 意外快速旋转、构图失控、主体出画 |
| 故事 | 画面好看但无法表达当前分镜信息 |

例如一个 5 秒素材，后 2 秒手部开始崩坏，可考虑只保留前 3 秒；如果关键剧情发生在后 2 秒，则应换素材或重新生成，不能因为前面好看就牺牲剧情信息。

对每段记录“可用区间、问题、分镜用途”。编号不能代替检查，同一提示词的不同结果也不能默认都合格。


### 3.7 管理镜头与生成版本

**整理补充目录示例。**

```text
城门之前/
  01-剧本与分镜/
  02-人物与场景参考/
  03-关键画面/
  04-生成视频/
  05-配音与音效/
  06-海报与片名/
  07-导出/
```

| 推荐名称 | 含义 |
| --- | --- |
| C01-守城人-正面-v02.png | 角色参考及版本 |
| S05-信物特写-img-v03.png | 第五镜的关键图 |
| S05-信物特写-video-v02.mp4 | 第五镜的视频版本 |
| S05-旁白-v01.wav | 对应镜头或语句的声音 |

记录表可包含：镜号、源图版本、提示词版本、候选文件、可用区间、审核状态。保留审核通过的版本，同时知道为什么淘汰其他版本。

生成工具输出的文件名可以保留在记录里，再复制一份按镜号命名用于剪辑。不要为了整理而丢失源文件与分镜之间的对应关系。


### 3.8 提示词优化：改具体问题，不只堆形容词

**原课画面**：《02-AI图像与视频生成》约 `01:20` 可见页面中的提示词优化面板，标题含“DeepSeek-R1 提炼版”，旁边仍显示城池与风暴画面。



画面能确认有优化结果展示；不能从这一帧确认所有隐藏设置、完整提示词或后续生成效果。

**整理补充**：优化提示词应针对具体问题。

| 问题 | 可尝试的修改 |
| --- | --- |
| 城门位置不一致 | 采用场景基准图，明确主建筑与道路关系 |
| 人物数量变化 | 简化群体内容，明确主要人物与数量 |
| 手部动作复杂 | 简化到一个主要动作，必要时拆为两个镜头 |
| 脸在大运镜中变化 | 减小视角变化，使用更合适的参考图 |
| 氛围和剧情无关 | 删除无用效果，强调当前镜头的信息用途 |

“电影级、8K、震撼、超真实”不能替代主体、构图、动作和空间关系。提示词优化工具也需要人工审核；生成更长的文字不一定更容易执行。


## 四、AI漫剧剪辑成片

**原课画面**：《03-AI漫剧剪辑成片》約 `00:55` 展示剪映素材区、预览与时间线。时间线已排列多个玄幻片段，预览中是持剑人物。



约 `01:20` 可见水墨或碎片扩散形态的动态画面，以及选中片段的属性区。



两张图能确认剪辑软件与素材排列方式；仅凭这些预览不能确认完整成片的配音、字幕、片头时长或最终导出参数。

**整理补充**：本节沿用第 03 篇的时间线操作，但重点变为故事连贯性。先按剧情建立初剪，再处理声音、字幕、颜色和包装。


### 4.1 初剪：先让观众知道发生了什么

**整理补充操作流程。**

1. 导入审核通过的素材，按分镜编号建立顺序。
2. 逐段保留有效内容，裁掉等待、形变或与剧情无关的部分。
3. 检查每段结束状态能否接到下一段开始状态。
4. 临时放入旁白或对白，验证镜头长度是否合适。
5. 从头连续播放，先解决因果、顺序和时长。

初剪阶段优先直接切换。转场、滤镜和字幕花字可以后加，但缺少一个关键反应镜头时，应先补镜头而不是继续堆效果。

有八个素材不等于必须全部用上。选一个镜头的理由应是它表达了剧情信息，或支持情绪、空间和节奏。


### 4.2 镜头衔接：动作、视线与声音

**整理补充示意图。**



| 衔接方式 | 《城门之前》的自编示例 | 检查 |
| --- | --- | --- |
| 视线连接 | 守城人看向道路 → 陌生人出现 | 观看方向与对象位置是否说得通 |
| 动作连接 | 伸出信物 → 信物特写 | 动作阶段、持有的手是否对应 |
| 反应连接 | 看见信物 → 守城人放下戒备 | 反应由新信息推动 |
| 声音提前 | 先听见求助，再看见来人 | 声音让观众预期下一镜内容 |
| 声音延续 | 对白继续，画面切到听者反应 | 听者反应没有破坏语义 |

声音先于下一镜出现，常称为 J-cut；上一镜声音延续到下一镜，常称为 L-cut。它们让声音与画面切点不必完全重合，使用时仍应确认说话者和空间关系。

片段中运动方向突然相反时，不要一律通过水平翻转修正。翻转也会改变持剑的手、文字、服饰和空间关系，应综合判断。


### 4.3 对白与配音：区分说话者，控制信息量

**整理补充。** 优先把每句对白写成易于自然说出的短句。

| 自编对白 | 对应画面 | 作用 |
| --- | --- | --- |
| 陌生人：“等等，我带回了这个。” | 来人接近城门，准备展示信物 | 提出新信息 |
| 守城人：“从哪里来的？” | 警惕的近景 | 表现谨慎与阻碍 |
| 陌生人：“他让我把它交给你。” | 信物特写 | 推动识别与转折 |
| 守城人：“进来。” | 守城人作出决定，门开始打开 | 行动回应信息 |

这些对白是自编练习，不是原课台词。短句仍需听声音是否符合人物状态，不能所有句子都用同一语调。

为每个主要角色保持相对稳定的音色、语速和表达方式。重新生成某句后，比较与前后声音的音量、节奏和音质是否一致。

没有可靠口型同步时，可在听者反应、道具特写或环境镜头上承接部分对白；不要为了掩盖口型问题而让观众始终不知道谁在说话。


### 4.4 音乐、环境声与动作音效

**整理补充示意图。**



| 声音层 | 作用 | 调整原则 |
| --- | --- | --- |
| 对白／旁白 | 承载人物与剧情信息 | 先保证可懂，不被其他声音盖住 |
| 环境声 | 风暴、城门空间、远处人声 | 音量稳定，避免突然有声或无声 |
| 动作音效 | 脚步、门闩、信物碰撞 | 对齐动作发生时刻，避免每一下都过重 |
| 音乐 | 紧张、缓和与收束 | 根据剧情变化，给对白留空间 |

先听对白，再逐层加入其他声音。有对白时可降低音乐，重点动作也可以保留短暂安静，让观众注意新的信息。

生成视频可能带有自动声音或异常音效。导入剪映后检查原声是否需要保留，防止它与后加配音、音乐重复播放。

不要把所有音轨都拉到很大。整体混合后仍需检查输出过载、明显破音和尾部突停，最后用手机扬声器试听。


### 4.5 字幕与角色对白校对

**整理补充。** 先保证文字准确，再考虑样式与动画。

- 检查人名、专有词、数字、同音字和标点。
- 每句字幕对应正确说话者与时间，不把一句话切得支离破碎。
- 字幕尽量不遮住人物脸、手部动作和信物。
- 角色切换明显时，可通过声音和镜头区分；需要额外说话者提示时再设计文字标识。
- 场景背景复杂时，使用适度描边或底板增强可读性。

AI 配音读对不代表自动字幕一定识别对；修改字幕也不会自动修正旧配音。声音与文字分别检查。

竖屏字幕须考虑发布平台的覆盖区域。在手机尺寸查看整段，确认快速镜头下仍能读完，不只看电脑上的静态大图。


### 4.6 精剪：把颜色与视觉包装统一起来

**整理补充。** 精剪建立在内容和声音已经成立的基础上。

| 项目 | 处理目标 |
| --- | --- |
| 时长 | 去掉重复等待，让重要转折有理解时间 |
| 颜色 | 同一场景的人脸、服饰、明暗与天气相对稳定 |
| 构图 | 重要信息处于画面内，横竖屏适配合理 |
| 转场 | 服务时间、空间或情绪变化，避免连续炫技 |
| 片名 | 简短可读，与作品视觉方向一致 |
| 结尾 | 回应故事问题，声音和画面自然结束 |

科幻或玄幻效果可以夸张，但基础动作与叙事关系仍需清楚。没有必要给每个镜头叠加闪光、抖动和巨大音效。

沿用第 03 篇的校色顺序：基础明暗与白平衡 → 镜头匹配 → 少量风格处理 → 连续复查。不要期待一个滤镜自动统一所有 AI 素材。


## 五、海报与片名设计

**原素材观察**：“大汉龙劫”海报使用大面积黑色背景、金色毛笔字与少量红色点缀；主标题占据纵向主体位置，信息层次集中。

**整理补充**：参考的是视觉组织方法，不直接把原海报标题改掉就当作自己的原创成果。

| 设计对象 | 应先确定 |
| --- | --- |
| 主标题 | 正确文字、长度、阅读方向 |
| 背景 | 是否需要人物或场景，是否与成片一致 |
| 色彩 | 主色、标题对比与少量点缀 |
| 信息层级 | 片名最重要，其他信息是否真的需要 |
| 平台规格 | 横竖比例、封面裁切、安全区域 |

文字生成容易出现缺字、错字和结构变形。需要准确片名时，可以先生成无字背景，再用剪映或设计软件添加正式文字。


### 5.1 自编海报提示词与排版流程

**整理补充模板。**

```text
为原创奇幻短片《城门之前》设计一张竖版海报背景。
场景是一座古城门，门外冷灰风暴，门内透出少量暖色灯火。
一位守城人的背影站在门侧，人物较小，突出选择与边界的主题。
构图简洁，上方为片名预留清晰空白，主体不被裁切。
写实奇幻，材质自然，冷暖对照克制。
只生成背景，不添加文字、字母、印章和品牌标志。
```

排版步骤：确认背景 → 添加准确片名 → 确定字号与对比 → 放入少量必要信息 → 缩小检查可读性 → 对照成片检查人物与风格。

片名与字幕可以使用不同层级，但尽量保持作品整体统一。书法或装饰字体适合短标题，未必适合快速阅读的长对白。

发布前确认实际使用的字体、音乐和素材适用范围。不要把工具里可生成或可预览当作所有用途都已获授权。


## 六、导出与完整复查

**整理补充，不是原课最终导出参数。**

| 项目 | 练习选择与检查 |
| --- | --- |
| 画面比例 | 按目标平台选 9:16 或 16:9，生成与剪辑尽量一致 |
| 分辨率 | 竖屏可用 1080×1920，横屏可用 1920×1080 |
| 帧率 | 与项目和主要素材匹配，常见起点为 25 或 30 fps |
| 输出文件 | MP4 常用于播放，编码按目标平台要求选择 |
| 命名 | 城门之前-竖屏-v01.mp4，明确版本与方向 |
| 保留资料 | 工程、原素材、参考图、提示词和分镜记录 |

导出到更高分辨率不等于恢复原素材细节；把低帧率素材导出为更高帧率也不等于获得真实的更多动作信息。

导出后播放实际文件，从头看完，再在手机上检查字幕、音量与画面方向。工程预览正常不能代替导出文件检查。


### 6.1 成片审核：故事、画面与声音一起看

**整理补充。**

| 维度 | 通过标准 |
| --- | --- |
| 故事 | 角色目标明确，转折有依据，结尾回应问题 |
| 身份 | 人物脸、发型、服饰与主要道具稳定 |
| 空间 | 视线、行动方向、城门与人物位置可理解 |
| 动态 | 无严重形变、闪变、突然消失与无意义运动 |
| 声音 | 说话者明确，台词可懂，音效贴合动作 |
| 字幕 | 文案正确、时间对应、手机上可读 |
| 包装 | 片名准确，封面与成片风格相符 |
| 文件 | 完整可播放，无意外黑屏、长尾和声音截断 |

至少做三种检查：只看画面是否能理解行动 → 只听声音是否能理解信息 → 画面与声音是否互相支持。

写修改记录时使用实际问题，例如“信物从右手跳到左手 → 换特写候选并调整切点 → 完整播放确认”。不要只写“已美化”“已优化”。


## 七、完整练习任务

**整理补充自编练习。** 依据前面的《城门之前》方案，完成一条约 60 秒的短片，或把故事进一步缩成 30 秒版本。

需要准备的内容：

1. 一句话故事、两个人物设定、一处场景基准。
2. 一张分镜表，写明画面、时长、声音和连接点。
3. 审核通过的关键图与视频有效片段。
4. 校对后的对白或旁白、音乐、必要音效与字幕。
5. 剪辑工程、成片与封面，以及一次修改记录。

可以先做三镜验证：守城人警惕 → 信物特写 → 守城人决定开门。三镜能连起来，再扩展环境和完整动作。这样能较早暴露身份、空间和剧情问题。

本篇只交付学习笔记和练习方案，没有代替读者生成角色、制作视频或导出这条练习短片。


## 八、常见问题速查

**整理补充。**

| 问题 | 优先处理 |
| --- | --- |
| 剧本很长却没有故事 | 先明确目标、阻碍、选择与结果，再删支线 |
| 对白塞不进镜头 | 试读，删冗余，合理增加有效画面 |
| 人物跨镜头像换了人 | 核对参考图与固定设定，减小视角变化，重做错误候选 |
| 场景像不同城市 | 用场景基准图，明确建筑与空间关系 |
| 动作复杂而崩坏 | 拆成短镜头，每镜只安排一个主要变化 |
| 道具突然换手或消失 | 检查起止状态和切点，换素材或补特写 |
| 镜头很漂亮却无法理解 | 回到分镜用途，补必要信息或删除无关镜头 |
| 音乐盖住对白 | 先调清人声，再压低音乐，完整试听 |
| 字幕正确但配音读错 | 分别修改字幕与重新生成／录制音频 |
| 海报片名错字 | 生成无字背景，手动添加准确标题 |
| 导出片尾过长 | 检查最右侧剩余音频、文本和叠加片段 |

问题发生在哪个环节，就优先在那个环节修。用更多转场、滤镜和音效遮掩问题，通常会增加后续检查难度。
