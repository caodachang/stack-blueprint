## 一、认识剪辑软件：从素材到成片

**原课画面**：《01-认识剪辑软件》约 `00:19` 展示剪映首页。能看到“开始创作”、草稿列表以及图文成片、创作脚本等入口。



这张图能确认的是首页与草稿入口，不能据此推断完整课程里每个功能的操作过程。

**整理补充：先分清三个对象。**

| 对象 | 保存什么 | 常见误解 |
| --- | --- | --- |
| 原始素材 | 视频、图片、音频、字体或图片标题等文件 | 把素材导入后就可以随意删除原文件 |
| 工程／草稿 | 素材引用、片段位置、裁剪、字幕、效果与参数 | 草稿就是能交付给别人看的 MP4 |
| 导出成片 | 按时间线合成后的播放文件 | 导出文件里还能直接改每个字幕和镜头 |

剪辑通常不会直接改写原始素材。草稿常依赖原素材的保存位置；移动、重命名或删除后，可能需要重新定位。工程打包或云同步是否包含素材，应按软件实际能力确认。

实用做法：先把素材集中保存，再开始剪辑；保留工程用于修改，导出视频用于观看和交付。


### 1.1 界面分区与最短操作路线

**整理补充示意图，不是课程截图。**



| 区域 | 主要任务 | 操作前先确认 |
| --- | --- | --- |
| 素材／功能区 | 导入文件，选择音频、文本、转场等 | 素材是否已经下载到本地 |
| 预览区 | 检查当前位置实际呈现的画面 | 播放头位置是否正确 |
| 属性区 | 改选中对象的位置、缩放、音量、颜色等 | 当前选中的是视频、文本还是音频 |
| 时间线 | 安排片段顺序、时长、轨道关系 | 修改是否会影响后面的片段 |

一次完整入门操作：新建草稿 → 导入一个视频 → 放入时间线 → 移动播放头 → 裁掉不需要的开头 → 添加标题 → 预览 → 导出。

**播放头和选中片段是两回事。** 播放头决定当前预览哪一帧；选中对象决定属性区修改谁。很多“参数改了没效果”的问题，来自选错对象或播放头没有经过它。


### 1.2 建立素材目录与成片规格

**整理补充，可用于自己的练习工程。**

```text
城市微缩短片/
  01-脚本/
  02-图片/
  03-视频/
  04-配音/
  05-音乐音效/
  06-标题与封面/
  07-导出/
```

给文件起能对应镜头的名字，例如 `S01-城市全景-v2.mp4`、`S02-桥边人物-v1.mp4`。原素材保留，试验修改另存版本。

开始前确定：目标平台、横竖屏、时长、画面比例与主要素材帧率。竖屏练习可用 `9:16`、`1080×1920`；横屏练习可用 `16:9`、`1920×1080`。这是练习起点，不是原课规定。

比例解决画布形状，分辨率解决像素数量，帧率解决每秒画面数量。将低清素材导出为 4K，只是增加输出像素，不能恢复不存在的细节。


## 二、轨道与视频分割

**整理补充，对应第 02 课时。** 时间线的横向表示时间，纵向组织不同层的素材。同一时刻可能有背景视频、叠加画面、字幕、人声、音乐和音效共同出现。



| 轨道内容 | 常见用途 | 检查重点 |
| --- | --- | --- |
| 主视频 | 建立主要镜头顺序 | 是否有空隙、错误顺序或无效片段 |
| 叠加视频／图片 | 画中画、贴图、局部合成 | 不透明画面是否遮住下面的主体 |
| 文本 | 标题、字幕、注释 | 出现与消失时间是否合适 |
| 人声 | 旁白、对白 | 是否对应画面与字幕 |
| 音乐 | 气氛与节奏 | 是否抢人声、是否突然中断 |
| 音效 | 动作、环境与空间感 | 是否贴合动作发生的时间 |

一般情况下，叠加层覆盖下面的画面；透明区域、蒙版、抠像或混合设置会改变结果。声音轨道则是同时混合播放，不是“上轨遮住下轨”。

主轨吸附、自动接合与联动移动会影响删除后的行为。操作前观察开关状态，操作后检查后面的声音和字幕是否仍然对齐。


### 2.1 裁剪、分割、移动：分别改变什么

**整理补充。**

| 操作 | 改变什么 | 适合解决的问题 |
| --- | --- | --- |
| 拖动片段边缘裁剪 | 改可见素材的入点或出点 | 去掉镜头开头等待、结尾变形 |
| 在播放头处分割 | 把一个时间线片段分为两段 | 去掉中间错误，分别调整两段 |
| 移动整个片段 | 改它在成片中的开始时间 | 排顺序、对齐旁白或节拍 |
| 删除片段 | 去掉选中的时间线对象 | 清理无用内容 |
| 缩放时间线显示 | 改界面显示比例 | 更准确地定位切点，不改变成片时长 |



例：原素材 10 秒，仅保留第 2—7 秒，保留长度是 `7−2=5 秒`。这是裁剪，不是把 10 秒加速成 5 秒。

若要删去中间一段，在这段内容的开始和结束分别分割，再删除中间片段。删除后可能留下空隙，也可能因吸附或自动接合而移动后续片段，应以当前软件行为为准。


### 2.2 分割操作步骤与对齐检查

**整理补充操作流程。**

1. 在时间线选中需要修改的视频片段。
2. 将播放头移到目标位置，放大时间线，必要时逐帧定位。
3. 使用“分割”功能；快捷键先查看软件的快捷键设置。
4. 删除或裁剪不需要的部分，再调整相邻片段的位置。
5. 连续播放切点前后，检查动作、声音、字幕和节奏。

剪切判断应看内容：一个动作完成、旁白换句、视线变化或节奏落点，都可能成为切点。不能只凭片段长度相同就认为节奏自然。

**分割后重点检查**：视频自带音频是否一起分割；音乐是否被误删；文本是否仍对应画面；后面是否多出黑屏；是否误把不同镜头移到叠加轨。

撤销能恢复最近操作，但不能代替保存。做大范围调整前先复制草稿或保存阶段版本。


## 三、其他基本功能与音乐添加

**整理补充，对应第 03 课时。** 先完成内容顺序，再使用画面功能解决具体问题。

| 功能 | 用途 | 常见问题 |
| --- | --- | --- |
| 缩放、位置、旋转 | 适配比例、调整构图、强调主体 | 放大过多会损失清晰度 |
| 画面裁切 | 只保留源画面的一部分 | 主体移动后可能被切出画面 |
| 变速 | 调整动作时长和节奏 | 人声与声音可能变化，需试听 |
| 静帧 | 将某一帧作为静态片段停留 | 停留太久会让观众误以为卡住 |
| 转场 | 处理两个镜头的连接 | 长转场可能削弱动作和信息 |
| 分离／提取音频 | 单独处理视频里的声音 | 原音频与分离音频重复播放 |

优先用直接切换建立镜头关系。只有在时间、空间、情绪变化需要提示时，再考虑淡化、叠化等转场。转场不一定会自动延长工程总时长，效果时长和素材重叠方式要看实际时间线。

竖屏中放横屏素材时，选“裁切填满”或“保留完整画面并设计背景”，先确保脸、动作和文字不会被切掉。


### 3.1 音乐导入、截取与淡入淡出

**整理补充操作流程。**

1. 导入本地音乐或选择软件音乐素材，先试听与主题是否合适。
2. 放入独立音轨，按内容确定开始位置。
3. 选取适合成片的音乐段落，而不是一律从音乐的第 0 秒开始。
4. 将音乐长度调整到需要的范围，检查结尾是否自然。
5. 给开始与结尾设置适当淡入淡出，再听整段。

**淡入淡出**改变音量；视频淡化改变画面透明度，两者需要分别检查。音乐短于成片时，优先换合适的段落或歌曲；需要拼接循环时，让接点落在相近节拍与乐句位置，并试听是否出现突跳。

按“旁白能听清 → 音乐支持情绪 → 音效强调动作”的顺序调整。不要因为音乐波形看起来小，就把音量直接拉满。

发布前确认所选音乐、字体、贴纸等是否允许目标用途。软件里能预览或导出，不等于适用于所有发布平台和商业场景。


### 3.2 声音层次与音量判断

**整理补充。** 声音波形帮助定位，最终判断要靠完整试听。

| 声音 | 调整目标 | 可采取的动作 |
| --- | --- | --- |
| 人声 | 内容清楚、句子间音量接近 | 调整分句音量，适度降噪，保留自然呼吸 |
| 背景音乐 | 支持情绪且不盖过对白 | 整体降低，有旁白时进一步压低 |
| 动作音效 | 提示动作、增加现场感 | 对齐动作，避免每个切点都加重音 |
| 原视频声音 | 保留需要的环境或对白 | 无用原声静音，避免与配音重复 |

音量界面可能使用百分比或 dB。它们是调节方式，不能只看数值比较不同素材的响度。在存在 dBFS 峰值表的工作流里，人声峰值约 `−12 至 −6 dBFS` 可作为试调起点，但不是固定交付标准，也不代表最终整体响度。

降低音量能减少输出过载，却不能修复录音时已经发生的削波失真。明显破音的录音应优先重录或换素材。

至少检查三次：单听人声 → 人声加音乐 → 所有声音一起播放。最后再用手机扬声器听一遍，确认弱音量下仍能理解内容。


## 四、人声配音

**整理补充，对应第 04 课时。** 配音为镜头提供信息与情绪。先确定文案，再确定声音，最后调整画面与节奏。

| 配音方式 | 适合情况 | 重点检查 |
| --- | --- | --- |
| 导入已有录音 | 已用录音设备或其他工具录制 | 音质、文件完整性、分句对应 |
| 软件内录音 | 一边看画面一边讲解 | 麦克风、输入电平、环境噪声 |
| 文字转语音 | 快速试读或生成旁白 | 读音、多音字、停顿、语气和使用权限 |

真人录音流程：准备文案 → 试录一小段 → 检查距离和爆破音 → 分段录制 → 导入或保留在音轨 → 裁掉明显无效空白 → 轻度处理 → 完整试听。

录音时尽量保持麦克风距离一致。降噪用来减轻稳定噪声，过强处理容易出现水声、金属声或吞掉轻声字；环境太吵时应先改善录音条件。


### 4.1 文字配音与语句校正

**整理补充操作流程。**

1. 将旁白拆成意义完整的短句，避免一大段没有标点的文字。
2. 使用文本朗读或文字配音入口，选择合适音色与语气。
3. 生成后逐句听，检查多音字、专名、数字和英文缩写。
4. 对读错部分调整措辞、标点或读音设置，再生成该句。
5. 将配音与镜头对齐，给重要信息保留理解时间。

示例文案：

```text
晨雾沿着河面散开。
小桥那边，城市慢慢醒来。
一艘小船，把春天带进了画面。
```

上述文案是自编练习，不是原课配音内容。标点可以提示停顿，但不同语音工具的解释不同，应听实际输出。

字幕和配音应分别校对：配音读对了，自动字幕仍可能识别错；修改文本后，旧音频通常也不会自动变成新说法，需要重新生成或录制。


### 4.2 用旁白带动镜头节奏

**整理补充。** 一句话太长时，先简化文案，不要只靠加速把它硬塞进镜头。

| 问题 | 优先处理 |
| --- | --- |
| 镜头结束了，旁白还没说完 | 调整文案、延长有效画面或增加合理补充镜头 |
| 配音说完后长时间没内容 | 缩短空白，或用画面动作与音乐承接 |
| 每句之间完全没有停顿 | 留自然停顿，让信息有理解空间 |
| 人声和音乐争抢注意力 | 下调音乐，必要时分段调整音乐音量 |
| 配音画面不同步 | 先核对语义对应，再微调音频位置 |

可以先建立旁白骨架，再围绕声音安排镜头；音乐驱动的风景短片也可以先定音乐段落，再安排画面。两种方法都要回到“观众是否看懂、是否听清”。


## 五、字幕设计

**整理补充，对应第 05 课时。** 字幕用于理解语音，标题用于提出主题，强调文字用于突出重点。三者不要争抢同一位置。

字幕基本流程：选定需识别的音频范围 → 自动识别或手动输入 → 逐句校对 → 调整时间 → 统一样式 → 手机尺寸预览。

先校正文案，再做花字动画。校对重点包括：人名、地名、专业词、数字、标点、多音字和 AI 配音中的非正常停顿。

| 对象 | 样式原则 | 时间原则 |
| --- | --- | --- |
| 普通字幕 | 清晰、稳定、与背景有对比 | 对应说话内容，结束后及时消失 |
| 主标题 | 主题明确，长度适中 | 出现时给足阅读时间 |
| 强调词 | 少量突出关键词 | 与重点语句或画面变化一致 |

自动识别结果只是初稿。AI 生成画面里的伪文字通常不能作为正式字幕，重要文字应在剪辑软件中重新添加。


### 5.1 字体、描边、断句与安全区域

**整理补充。** 字幕设计先解决“看得清”，再考虑装饰。

- 尽量使用清晰字体，统一字号、行距和主要颜色。
- 浅色字幕遇到明亮背景，可使用适度描边、阴影或半透明底板。
- 按意义断句，不把一个人名、地名或紧密词组拆到两行。
- 控制同屏文字量，按实际语速与阅读速度决定显示时长。
- 字幕避开人物脸部、手部动作和画面关键细节。

竖屏发布界面的按钮、昵称、评论入口可能覆盖画面。安全区域应参考目标平台并用实际设备检查，不能把“底部固定留某个百分比”当成所有平台通用标准。

例如“剪辑的第一步／是选出有效画面”比“剪辑的第／一步是选出有效画面”更便于理解。断句应配合语音，不能只追求两行一样长。


### 5.2 字幕时间与样式联动

**整理补充检查流程。**

1. 从一句话的实际起点确定字幕入点。
2. 从语句结束、下句开始和阅读需要共同确定出点。
3. 确认相邻字幕没有意外重叠，也没有过长停留。
4. 统一基础样式，再单独设计少量重点字。
5. 连续播放完整视频，检查静止截图看不出的闪跳与阅读问题。

部分软件提供批量样式或“应用到全部”功能。使用前确认修改作用于哪些字幕，使用后抽查首、中、尾三处，避免标题与普通字幕一起被改。

重新识别音频可能生成一组新字幕。应检查是否与旧字幕叠加，而不是仅看当前一句是否正常。


## 六、片头片尾设计与视频导出

**整理补充，对应第 06 课时。** 片头帮助观众迅速理解主题，片尾让内容有完整落点。

| 位置 | 设计内容 | 检查重点 |
| --- | --- | --- |
| 片头 | 主题、代表性画面、必要的标题 | 是否尽快进入内容，文字是否易读 |
| 正片 | 镜头顺序、旁白、字幕与节奏 | 是否兑现片头提出的内容 |
| 片尾 | 最后一个有效镜头、简短结束语或署名 | 是否自然结束，是否有多余黑屏 |

短视频不必每次都加入独立片头模板。可以在第一个有效镜头上叠加标题；片尾也可以在最后一个镜头自然收住。

导出前检查时间线最右侧：零散文本、隐藏叠加层或音频可能把成片拉长。不要仅凭主视频最后一个镜头的位置判断总时长。


### 6.1 导出参数分别决定什么

**整理补充参数示例，不是原课固定设置。**

| 参数 | 含义 | 常用练习思路 |
| --- | --- | --- |
| 画面比例 | 画布宽高关系 | 先确定 9:16 或 16:9，再检查构图 |
| 分辨率 | 输出画面的像素数量 | 竖屏 1080×1920，横屏 1920×1080 |
| 帧率 | 每秒输出多少帧 | 与项目和主要素材匹配，常见为 25 或 30 fps |
| 码率／质量 | 压缩后的数据量与质量取舍 | 先用推荐或高质量，检查细节与文件大小 |
| 编码 | 用什么方式压缩视频 | H.264 常用于兼容性较好的播放文件 |
| 封装格式 | 如何组织视频与音频数据 | MP4 常见；封装格式不等于编码 |

把 25 fps 素材导出为 60 fps，不会自动获得真实拍摄的 60 fps 动作细节。普通重复帧和 AI 插帧是不同处理，插帧也需要检查运动伪影。

大致文件体积可估算为 `总码率 × 时长 ÷ 8`。例如总码率约 8 Mbps、时长 30 秒，体积约 30 MB，不计封装等少量开销；实际结果取决于编码模式与内容。

普通 MP4 成片通常不保留透明背景。若要交付透明叠加素材，先确认软件支持的透明视频格式与对方播放器要求。


### 6.2 导出不是最后一次检查

**整理补充交付流程。**

导出到明确目录，采用容易区分版本的名称，例如 `城市微缩-竖屏-1080p-v01.mp4`。不要把草稿文件名与最终视频文件名混淆。

导出后打开实际文件，从头播放到尾，检查：

- 开头是否出现错误帧、长时间黑屏或未完成标题。
- 中段是否有声音突变、字幕错字、切点闪帧与比例问题。
- 结尾是否截断旁白、音乐突停或多出长空白。
- 时长、分辨率、画面方向是否符合预期。
- 在手机上，文字是否可读、人声是否清楚。

本篇提供的是操作笔记和练习方案，没有代替读者实际完成剪辑或导出成片。工程预览正常也不能保证导出结果完全正常，以导出文件复查为准。


## 七、色彩调节的作用

**整理补充，对应第 07 课时。** 调色先让画面可信、镜头之间统一，再建立需要的氛围。

| 任务 | 解决的问题 | 判断依据 |
| --- | --- | --- |
| 基础校正 | 太暗、太亮、偏色、对比异常 | 主体是否可辨，细节是否保留 |
| 镜头匹配 | 相邻镜头的明暗、白平衡不一致 | 连续播放时是否突然变色 |
| 风格塑造 | 温暖、清冷、柔和、紧张等氛围 | 是否服务主题与叙事 |
| 注意力引导 | 背景抢眼、主体不突出 | 视线是否落在重要内容上 |

AI 素材特别要注意：同一人物跨镜头肤色变了、同一场景时冷时暖、某些区域异常发光。生成阶段尽量保持风格，剪辑阶段再校正镜头差异。

调色不能修复手指畸形、建筑结构跳变或文字乱码。遇到明显生成错误，优先筛掉片段或重新生成。

参考顺序：选择代表性镜头 → 调好基础明暗和白平衡 → 匹配其他镜头 → 添加适度风格 → 连续播放 → 对比调整前后。


### 7.1 单镜头好看与整片统一

**整理补充。** 一个镜头单独看很有气氛，接到另一个镜头后却可能显得突兀。

例如：第一镜头是偏蓝的清晨河流，第二镜头是偏橙的同一座桥。如果两者表示同一时间地点，应先检查白平衡和亮度；若脚本明确跨越时间，也可以保留差异，但应通过叙事让变化成立。

选择一个表现正常的参考镜头，比较其他镜头的：主体亮度、阴影深浅、白色是否偏色、肤色与主要环境色。不要只复制一组参数后就认为全部镜头统一了。

统一风格不等于所有画面都同样明亮。夜景应保留夜景感，雾景也不必拉出和晴天一样强的对比。

复制调色设置后，至少对明亮、阴暗、有人物这三类镜头分别检查。不同源素材对同一参数的反应可能很不一样。


## 八、色彩调节基础

**整理补充，对应第 08 课时。** 先看画面问题，再选择对应控制项。

| 控制项 | 主要影响 | 调整时注意 |
| --- | --- | --- |
| 曝光／亮度 | 整体明暗，具体响应因软件而异 | 提亮后高光可能失去细节 |
| 对比度 | 明亮与暗部的差异 | 太高容易死黑、死白；太低会发灰 |
| 高光 | 较亮的部分 | 不能从已完全过曝的数据中恢复真实细节 |
| 阴影 | 较暗的部分 | 大幅提亮会显露噪声或生成瑕疵 |
| 白色色阶／黑色色阶 | 明暗端点，若软件提供 | 检查端点是否被压成大片纯白或纯黑 |
| 色温 | 蓝黄方向的整体偏色 | 不把所有温暖画面都纠正成冷白 |
| 色调 | 常用于绿与洋红方向偏色 | 小幅调整，特别检查肤色 |
| 饱和度 | 色彩浓度 | 太高会让皮肤、天空和植被不自然 |

“更亮”不一定更清楚。如果高光失去纹理、暗部变得发灰或主体仍与背景混在一起，需要调整的是局部明暗关系，而不是继续拉亮度。

软件参数刻度不同，不能将某个教程的 `+20` 直接视为跨软件通用的颜色结果。


### 8.1 一套容易复查的基础校色流程

**整理补充操作流程。**

1. 选中片段，先关闭或重置已有滤镜、风格效果，确定问题来自哪里。
2. 调整白平衡，检查应接近中性的物体和人物肤色。
3. 调整整体明暗，让主体能看清，保留该场景应有的氛围。
4. 微调高光、阴影与对比度，观察亮暗细节。
5. 小幅调整饱和度，再按需要处理具体颜色。
6. 对比调整前后，播放整个镜头，确认移动主体没有变得异常。

不要一次把每个滑块都改很多。每次改一个主要问题，能更容易判断哪个控制项真正有效。

出现“画面越调越脏”时，回到中性状态，依次排查：是否多个滤镜叠加、是否过度提阴影、是否混用了通道曲线和色轮、是否错误选择了调整对象。


## 九、HSL 与色轮调节

**整理补充，对应第 09 课时。** HSL 按颜色范围调整，色轮按阴影、中间调、高光等亮度区域调整，两者选取对象的方式不同。

| HSL 参数 | 意义 | 例子 |
| --- | --- | --- |
| H：Hue，色相 | 颜色朝相邻颜色偏移 | 让某一绿色范围偏黄绿或蓝绿 |
| S：Saturation，饱和度 | 颜色浓度增强或减弱 | 降低过艳的蓝色天空 |
| L：Lightness／Luminance，明亮度 | 所选颜色范围更亮或更暗，名称随软件变化 | 压暗蓝色范围，突出云层对比 |

操作顺序：找到画面中的问题颜色 → 选对应色段 → 小幅改一个参数 → 观察全部受影响区域 → 连续播放。

例如想让植被不那么鲜艳，可以适度降低绿色饱和度，但绿色衣服或人物周围的绿光也可能一起变化。HSL 选择的是颜色范围，不是“树”这个物体。

皮肤常涉及红、橙等范围。调整橙色时也可能改变木材、灯光或衣服，不能认为橙色只代表肤色。


### 9.1 色轮：按明暗区域加入颜色倾向

**整理补充。** 若所用版本提供阴影、中间调和高光色轮，可理解为给不同亮度区域加入颜色倾向；各区域通常有重叠，不是完全独立的三块。

| 区域 | 常见观察对象 | 调整注意 |
| --- | --- | --- |
| 阴影 | 暗处、背光部分 | 加色太多会让暗部变脏 |
| 中间调 | 主体、大部分皮肤与物体 | 对整体观感影响很大，优先保持主体自然 |
| 高光 | 云、灯光反射、明亮表面 | 偏色过重容易出现不自然的白色 |

通常，色轮控制点的方向决定颜色倾向，离中心的距离决定强度；亮度控制若提供，则单独影响该区域的明暗。具体控制项以当前界面为准。

“阴影略冷、高光略暖”可以作为自编练习，但不必对所有视频使用同一种方案。清雅诗词、节日热闹和科幻冷峻的风格目标不同。

HSL 与色轮都能影响颜色，叠加时要防止互相抵消或过度染色。先完成基础校正，再做少量风格处理。


## 十、曲线调色

**整理补充，对应第 10 课时。** 曲线描述“输入亮度如何映射为输出亮度”。



在常见亮度曲线中：横轴从左到右是输入从暗到亮，纵轴从下到上是输出从暗到亮。对角直线表示保持原来的映射；某一位置上提，通常让对应亮度范围变亮；下压则变暗。

| 操作 | 常见结果 | 注意 |
| --- | --- | --- |
| 下压较暗区域，上提较亮区域 | 形成适度 S 形，提高对比 | 过强会丢失亮暗细节 |
| 抬高左下端点 | 最暗部分变亮，形成抬黑效果 | 过多会让画面发灰 |
| 下压右上端点 | 限制最亮部分的输出 | 不是恢复已经丢失的高光细节 |
| 小幅上提中间位置 | 提亮中间调 | 周围区域也会受到平滑曲线影响 |

曲线上一个控制点不只影响一个精确亮度值，它会通过插值影响附近范围。少量控制点更容易判断和复查。


### 10.1 RGB 通道与曲线排错

**整理补充。** 软件若提供单独的红、绿、蓝曲线，就可以在不同亮度范围调整颜色通道。

| 通道变化 | 常见颜色倾向 |
| --- | --- |
| 增加红通道 | 更偏红 |
| 减少红通道 | 相对偏青 |
| 增加绿通道 | 更偏绿 |
| 减少绿通道 | 相对偏洋红 |
| 增加蓝通道 | 更偏蓝 |
| 减少蓝通道 | 相对偏黄 |

这些是相对变化方向，最终观感取决于原图、亮度范围和其他调整。软件中的总曲线与亮度曲线也可能有不同实现，不要假定所有“RGB”按钮都只影响明暗。

排错时先确认当前选中的是总曲线还是颜色通道。画面突然偏色，常见原因是改了单通道；变得发灰，可能是黑端点抬高；暗部没有层次，可能是曲线过度下压或多个对比效果叠加。

用中性画面试调，比较调整前后，再回到完整镜头。不要只用一帧决定人物肤色和全片风格。


## 十一、综合案例：完成一条 30 秒城市微缩短片

**整理补充自编练习，对应“综合案例演示”的学习主题。不是对未完整观看的原课案例进行复述。** 可沿用第 02 篇已准备的风景或微缩景观素材，也可以使用自己的合适素材。

练习目标：竖屏 `9:16`、约 30 秒，包含多个有效镜头、旁白、音乐、校对后的字幕、简短标题、统一颜色。后面再给静态图片增加关键帧，并用蒙版尝试局部叠加。

| 成片时间 | 自编镜头任务 | 文字或声音 |
| --- | --- | --- |
| 0—3 秒 | 城市代表性全景，直接进入主题 | 标题“微缩城市的一天”，音乐缓入 |
| 3—9 秒 | 河流、桥或街道建立环境 | 旁白介绍场景，保留自然停顿 |
| 9—16 秒 | 人物或小船等一个明确动作 | 旁白与动作对应，可加少量环境声 |
| 16—24 秒 | 建筑、植物或生活细节 | 镜头节奏略作变化，避免重复信息 |
| 24—30 秒 | 回到全景或有结束感的画面 | 简短收束，音乐自然结束 |

时间表是规划工具，不要求每个镜头强行拉到精确长度。若旁白与素材不匹配，先删减文案、替换镜头或调整时间。


### 11.1 从初剪到精剪的操作顺序

**整理补充练习步骤。**

1. **筛素材**：去掉主体错误、画面变形严重、文字乱码明显的片段。
2. **初剪**：把有效镜头排到主轨，先讲清内容，裁掉无用头尾。
3. **配音**：分句录制或生成，核对读音，并与画面意义对齐。
4. **音乐**：选合适乐段，调整入点、结尾和旁白期间的音量。
5. **字幕**：识别或输入，逐句校对，统一位置与基础样式。
6. **片头片尾**：标题不拖慢开头，最后一个镜头和声音能自然结束。
7. **颜色**：先校正一个参考镜头，再匹配其他镜头。
8. **动画与合成**：按需要添加关键帧、蒙版，检查是否引入新问题。
9. **导出**：按目标规格输出，播放实际文件并在手机上复查。

检查顺序很有用：先只看画面能否理解 → 只听声音能否理解 → 画面与声音一起看是否互相支持。

当画面已经有明显运镜时，不必再强行叠加大幅缩放。视觉运动过多，会掩盖主体动作并让字幕难以阅读。


### 11.2 练习交付与自检标准

**整理补充。** 练习应保存工程与导出视频，便于回看和修改。

| 检查项 | 通过标准 |
| --- | --- |
| 内容 | 能说出主题，每个镜头有用途，无重复堆砌 |
| 时间线 | 无意外空隙、漏删片段和超长尾巴 |
| 声音 | 人声可懂，音乐不抢，结尾无突停与明显破音 |
| 文字 | 无关键错字，时间对应，手机尺寸能读 |
| 颜色 | 相邻镜头无不合理突变，人物与环境不异常 |
| 动画 | 主体不被切出，运动速度符合内容 |
| 合成 | 蒙版边缘合适，无意外遮挡与穿帮 |
| 文件 | 导出文件完整可播放，规格与目标一致 |

记录一次修改：发现的问题、调整的对象、采取的动作、复查结果。例如“桥边字幕遮住人物 → 上移字幕并缩短文字 → 手机上完整播放，人物和字幕均清楚”。

复查结果要基于实际成片，不能只写“按了导出按钮”或“参数设置正确”。


## 十二、关键帧使用

**整理补充，对应第 12 课时。** 关键帧在指定时间保存某个参数值，软件根据相邻关键帧计算中间变化。



| 参数 | 关键帧可完成的变化 |
| --- | --- |
| 位置 | 画面、图片或文本从一个位置移到另一个位置 |
| 缩放 | 缓慢推进、拉远或局部强调 |
| 旋转 | 角度随时间改变 |
| 不透明度 | 淡入淡出，若该参数支持关键帧 |
| 音量 | 某句旁白期间降低音乐，若该参数支持关键帧 |

不是所有参数都支持关键帧，应查看参数旁是否提供关键帧标记。只放一个关键帧，不能定义两个不同值之间的动画。

关键帧没有生成新的真实场景运动。对静态图做缩放，只改变取景；它不能让图片中的人真实转身或让建筑露出原图没有的背面。


### 12.1 给静态图做一次缓慢推进

**整理补充操作示例，数值为练习值。**

1. 在时间线放入一张分辨率足够的图片，调整显示时长为约 5 秒。
2. 选中图片，将播放头放到片段开头。
3. 在缩放参数处添加关键帧，初始值设为 `100%`；同时确认构图合适。
4. 将播放头移到接近片段末尾，缩放改为 `110%`，确认生成了第二个关键帧。
5. 预览整段，检查重要主体是否始终在画面里。
6. 如要向特定主体靠近，同时调整位置参数，并检查对应关键帧。

线性变化示例：`缩放(t)=100%+10%×t/5`，其中 `0≤t≤5 秒`。在第 2.5 秒，线性结果为 `105%`；采用缓入缓出后，中间各时刻的值不一定按这个公式变化。

只有修改参数、没有确认关键帧时，可能整段都变成相同缩放。播放头没移动就反复修改，只是在同一个时间点改变值，也不会形成所需动画。


### 12.2 关键帧速度、缓动与排错

**整理补充。** 关键帧时间间距影响变化速度：相同位移在 1 秒完成比在 5 秒完成更快。

| 情况 | 原因与检查 |
| --- | --- |
| 动画完全不动 | 是否存在两个时间不同、数值不同的关键帧 |
| 动画中途突然跳一下 | 是否混入了多余关键帧，或相邻值差异过大 |
| 修改后整段都变了 | 是否真正启用了该参数的关键帧 |
| 人物被放大到画面外 | 同时检查位置、缩放与片段完整运动 |
| 多张图运动方向混乱 | 先确定镜头意图，再统一运动方向 |
| 片段裁短后动画异常 | 查看剩余关键帧位置，重新确认起止值 |

缓入缓出让运动启动和停止更柔和；线性变化让参数以稳定速率变化。软件若提供曲线或缓动预设，按运动目的选择，并播放检查。

关键帧与转场分别解决参数动画和镜头衔接，不能混为一谈。先做简单的两帧动画，再增加更多时间点。


## 十三、蒙版使用

**整理补充，对应第 13 课时。** 蒙版控制一个图层的哪些区域显示、哪些区域隐藏。它不会从原文件里永久删除像素。



| 设置 | 作用 | 常见用途 |
| --- | --- | --- |
| 形状 | 决定显示区域的轮廓 | 圆形头像、矩形窗口、分屏 |
| 位置与大小 | 决定蒙版覆盖哪里 | 对准主体或合成区域 |
| 旋转 | 改变边界方向 | 斜向分屏、倾斜窗口 |
| 羽化 | 让边缘逐渐过渡 | 柔和融合，减轻生硬边界 |
| 反转 | 交换显示与隐藏区域 | 保留外部、隐藏内部 |

具体形状与可动画的参数以软件版本为准。羽化太大可能让主体边缘半透明，太小则可能显得硬；没有适用于所有分辨率的固定数值。

蒙版通常按形状控制范围；抠像常依据颜色、主体识别或其他方法提取对象，两者用途有关但实现不同。移动人物需要跟随轮廓时，固定圆形蒙版通常不够。


### 13.1 做一个圆形叠加窗口

**整理补充操作步骤。**

1. 将背景视频放在主轨。
2. 将另一段视频或图片放在叠加轨，时间与背景重叠。
3. 选中上层素材，打开蒙版，选择圆形或相近形状。
4. 调整大小和位置，让需要展示的主体处于显示区域内。
5. 按画面需要微调羽化，检查边缘是否自然。
6. 播放整个重叠区间，确认主体运动后仍在窗口里。

如果要移动整个窗口，调整图层位置与缩放；如果要改变上层素材内部的显示范围，调整蒙版位置与大小。二者的效果不同，应观察预览确认当前改的是谁。

结合关键帧时，可以让窗口移动或逐步揭示画面，但前提是对应属性支持动画。先验证两帧变化，再做复杂运动。


### 13.2 蒙版、抠像与合成排错

**整理补充。**

| 问题 | 检查方法 |
| --- | --- |
| 上层整张图遮住背景 | 是否选对上层，蒙版是否启用，是否覆盖整个画面 |
| 加蒙版后什么都看不见 | 蒙版位置、大小、反转状态和片段重叠时间 |
| 显示区域和预期相反 | 检查“反转”，先用简单形状验证 |
| 羽化边缘像雾一样 | 降低羽化，检查上层透明度 |
| 主体一移动就被裁掉 | 调整范围或跟随位置，必要时换合适抠像方法 |
| 分屏中间出现空白 | 两层是否同时间存在，蒙版边界与图层位置是否匹配 |
| 遮挡关系不对 | 查看轨道层次与不透明区域，调整上层内容 |

“一个人物走到另一个物体后面”涉及前后遮挡关系。往往需要把遮挡物作为前景层，再用蒙版或抠像组织层次；单纯降低人物透明度不会产生正确遮挡。

合成完成后检查亮度、颜色、尺寸和运动方向。边缘柔和只能解决部分边界问题，不能自动让两个素材的光照和透视一致。


## 十四、常见问题速查

**整理补充。** 先确定出错对象，再检查时间、轨道和参数。

| 现象 | 优先检查 |
| --- | --- |
| 预览黑屏 | 播放头所在时间是否有可见素材，是否被隐藏或透明度为零 |
| 导入后素材丢失 | 原文件是否被移动、改名、删除，是否是未下载的云文件 |
| 删除视频后字幕错位 | 主轨吸附、联动或自动接合是否移动了视频 |
| 人声重复或有回声 | 原声、提取音频、配音是否同时播放 |
| 识别字幕不完整 | 音频范围是否选对，是否静音，识别后是否有重复字幕组 |
| 画面突然偏色 | 滤镜叠加、RGB 单通道、HSL 或色轮是否改错对象 |
| 导出时长超出预期 | 最右侧是否还有文本、叠加片段或音频 |
| 导出比预览模糊 | 源素材分辨率、裁切放大程度、导出分辨率与质量 |
| 播放卡顿但导出正常 | 预览负荷与效果复杂度；必要时按软件能力降低预览质量或用代理 |
| 文件打不开或导出失败 | 路径、空间、格式、编码、素材完整性与报错信息 |

排错时一次改一个问题，并复查原来正常的部分。不要通过反复添加滤镜、字幕或音频来掩盖根本问题。


## 十五、复习题与参考答案

**整理补充，用来检验是否理解。**

| 问题 | 参考答案 |
| --- | --- |
| 工程文件为什么不等于成片？ | 工程记录编辑结构和素材引用，成片是合成后的播放文件 |
| 缩放时间线是否会加快视频？ | 不会，只改变编辑界面的时间显示比例 |
| 裁剪和变速有什么区别？ | 裁剪选择源素材区间；变速改变素材播放速度 |
| HSL 能否只修改画面中的某棵树？ | 不能直接按物体选择，它可能影响其他相同颜色区域 |
| 色轮与 HSL 的区别是什么？ | 色轮常按亮度区域加入颜色倾向，HSL 按颜色范围调整 |
| 曲线抬高黑端点会怎样？ | 暗端变亮，可能形成抬黑效果；过多会发灰 |
| 一张图只有一个关键帧为什么不动？ | 没有定义不同时间的两个值供插值变化 |
| 关键帧缩放能否生成真实转身？ | 不能，只改变画面取景或属性 |
| 蒙版羽化解决什么问题？ | 软化显示与隐藏区域的边缘，不自动修正光照或透视 |
| 导出到更高分辨率是否恢复真实细节？ | 不会，输出像素增加不等于源素材细节增加 |

练习后应能独立完成：整理素材、排镜头、删无效片段、配音、加音乐、校对字幕、基础调色、两帧动画、简单蒙版合成，以及导出后的完整复查。
