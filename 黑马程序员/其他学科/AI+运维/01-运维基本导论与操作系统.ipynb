## 一、IT 运维的基本概念

### 1.1 运维到底在维护什么

**原课依据：第 3–4 页。**

IT 运维是通过专业技术手段，让企业的 IT 系统和网络持续、安全、稳定地运行，保障业务连续性。它涉及计算机网络、应用系统、硬件环境和服务流程的综合管理。

理解这句话时，抓住三个词：

| 目标 | 具体含义 | 整理补充例子 |
| --- | --- | --- |
| 持续运行 | 用户需要时，服务能正常提供 | 网站、接口、数据库不能频繁中断 |
| 安全运行 | 控制访问、保护系统和数据 | 管理权限、更新补丁、限制不必要的网络入口 |
| 稳定运行 | 请求增加或资源变化时，服务仍可用 | 发现 CPU、内存、磁盘或响应时间异常并处理 |

“修电脑、修网线”可能是工作的一部分，但本课的重点是 **保障服务器和应用所承载的业务**。判断运维工作是否有效，要看用户的业务能否继续正常进行。


### 1.2 六个常见运维方向

**原课依据：第 5–7 页。**



| 方向 | 主要关注的问题 | 课件列举的技术或工作 |
| --- | --- | --- |
| 系统运维 | 基础设施状态、服务器与网络稳定性 | Zabbix、Prometheus、负载均衡、冗余设计、故障响应 |
| 数据库运维 | 数据可靠性与访问性能 | 备份恢复、主从复制、慢查询分析、索引优化、MySQL Performance Schema |
| 自动化运维 | 重复操作的效率、一致性与标准化 | Ansible、配置管理、自动巡检、CI/CD |
| 容器运维 | 容器生命周期、调度与服务管理 | Kubernetes、服务发现、自动扩缩容、Istio |
| 云计算运维 | 云上资源弹性、性能与成本 | AWS／阿里云、自动扩缩容、Serverless、资源标签、账单分析 |
| 信创运维 | 国产软硬件适配与迁移 | 鲲鹏 CPU、统信 OS、达梦数据库、兼容性测试 |

**整理补充：** 这些方向会交叉。一次应用上线可能同时涉及 Linux、数据库、网络、安全组和监控。初学阶段先掌握共同基础，不必在第一篇就深入每种工具。


### 1.3 运维的四项核心工作

**原课依据：第 6–7 页。** 课件把核心工作概括为 **软件部署、监控、告警、故障快速恢复**。

| 工作 | 要解决的问题 | 整理补充：应留下的结果 |
| --- | --- | --- |
| 软件部署 | 如何让应用在目标环境运行 | 版本、配置、启动方式、访问入口和上线记录 |
| 监控 | 如何知道系统现在是否正常 | 指标、日志、健康状态与变化趋势 |
| 告警 | 如何及时发现需要处理的异常 | 告警内容、影响范围、负责人和通知方式 |
| 故障恢复 | 如何恢复业务并验证效果 | 处理记录、恢复证据与后续改进事项 |

**整理补充例子：** 应用上线后磁盘被日志写满，监控发现空间不足，告警通知负责人，负责人定位原因、处理日志并验证应用恢复，再完善日志轮转和容量预警。这四项工作在一次事件中形成连续过程。


## 二、软件开发流程与三大环境

### 2.1 从产品设计到运维

**原课依据：第 8–9、12 页。**

```text
产品设计 → 项目研发 → 项目测试 → 项目上线 → 运维
                        ↓ 不通过
                    返回研发修改
```



| 阶段 | 要完成的工作 |
| --- | --- |
| 产品设计 | 明确需求、功能与交互 |
| 项目研发 | 编写前后端程序；课件举例前端 HTML、CSS、JavaScript，后端 Java、Python、Go |
| 项目测试 | 检查功能、性能等，发现问题后反馈修改 |
| 项目上线 | 把经过验证的应用部署到正式提供服务的环境 |
| 运维 | 管理上线后的运行状态，持续维护系统 |

**整理补充：** 实际项目中运维可以更早参与容量、部署架构、日志与监控设计。课件流程表达的是软件从需求到运行的主要阶段，并不表示运维必须等到最后一步才参与。


### 2.2 开发、测试、生产环境

**原课依据：第 10–12 页。**

环境是相对独立的软件、硬件资源，以及对应的配置和数据。应用在这些资源中安装、运行和接受测试。



| 环境 | 主要使用者 | 主要用途 | 关注点 |
| --- | --- | --- | --- |
| 开发环境（dev） | 开发人员 | 编写、调试代码，快速验证想法 | 开发效率、调试方便 |
| 测试环境（test） | 测试工程师、QA 团队 | 功能、自动化、性能测试，模拟真实场景 | 发现问题、验证质量 |
| 生产环境（prod） | 最终用户 | 正式对外提供业务服务 | 可用性、稳定性、安全性 |

课件强调：运维人员尤其要关注 **生产环境**。

**整理补充：** 独立环境未必对应三台物理服务器，也可以通过虚拟机、云实例、容器和网络隔离实现。不同环境要明确区分数据库、配置、账号和访问入口，避免测试操作影响真实用户。


### 2.3 同一应用，为什么需要不同环境

**整理补充案例。** 假设同一个订单应用部署在三个环境中：

| 项目 | 开发环境 | 测试环境 | 生产环境 |
| --- | --- | --- | --- |
| 配置标识 | dev | test | prod |
| 数据 | 本地或模拟数据 | 专门准备的测试数据 | 真实业务数据 |
| 日志 | 可以包含较多调试信息 | 用于复现和验证问题 | 按业务需要记录，避免暴露敏感内容 |
| 发布操作 | 开发人员频繁修改 | 以验证版本为单位 | 按受控流程发布并保留回退方案 |

“测试通过”并不自动保证生产正常。发布前还要核对生产配置、依赖版本、数据连接、端口、资源和权限；发布后从实际用户路径验证服务。

**整理补充上线检查顺序：** 确认版本 → 核对配置 → 验证依赖 → 部署启动 → 检查日志与健康状态 → 验证业务 → 观察监控。出现问题时依据准备好的方案回退。


## 三、工单系统

### 3.1 工单是可跟踪的工作记录

**原课依据：第 13–16 页。**

工单系统（Ticketing System）用于 **记录、分配和处理工作请求、问题报告、变更申请等事项**，是运维与 IT 支持团队日常工作的工具。

**整理补充：** 同一个问题从发现到处理，可能跨越不同班次和团队。工单把问题、负责人、过程和结果保留下来，避免只靠口头沟通。

| 工单类型 | 整理补充例子 | 需要说明的内容 |
| --- | --- | --- |
| 问题报告 | 应用无法访问、接口频繁超时 | 时间、现象、影响、环境和证据 |
| 工作请求 | 申请账号或部署测试环境 | 需求、范围、负责人和完成条件 |
| 变更申请 | 修改配置或发布新版本 | 原因、步骤、时间窗口、验证与回退方式 |


### 3.2 工单的四步流程

**原课依据：第 14、16 页。**

```text
提交工单 → 自动分配 → 处理反馈 → 验证关闭
```



1. **提交工单**：用户填写问题描述、类别、优先级等信息。
2. **自动分配**：系统按规则把工单交给相关人员。
3. **处理反馈**：处理人记录诊断和解决方案，并与用户沟通。
4. **验证关闭**：确认问题已经解决，再把工单关闭。

**整理补充：** “执行了一个操作”和“问题已经解决”是不同状态。关闭前应检查原问题是否消失、业务是否恢复；验证失败时继续处理。实际系统也可能采用人工分配、转派和升级机制。


### 3.3 工单怎样连接运维、测试、产品和开发

**原课依据：第 15 页的生产运维流程图。**



原图表达的主要判断关系：

| 初步判断 | 处理方向 |
| --- | --- |
| 用户操作问题 | 说明正确操作流程，并确认问题解决 |
| 数据问题 | 排查数据并按流程修复 |
| 环境问题 | 运维修复环境、验证服务 |
| 代码缺陷 | 由测试复现、开发修复，再测试与发布 |
| 需求缺陷或新需求 | 转为需求或变更，进入产品评审流程 |

**整理补充：** 分类需要证据，不能只因“页面打不开”就直接认定为运维问题。错误可能来自网络、服务、代码、数据或操作方式，工单应记录判断依据和交接信息。


### 3.4 一份工单应该怎样写

**整理补充模板，编号、时间和现象均为练习示例。**

```text
标题：测试环境订单页面访问超时
工单编号：LAB-001
环境：test
发现时间：2026-10-12 10:20（记录时注明时区）
影响范围：测试人员无法打开订单列表；生产环境尚未发现同类问题
现象：访问测试入口后约 30 秒超时
证据：失败时间、请求编号、脱敏截图或日志片段
近期变化：10:00 发布测试版本 v0.2
初步排查：确认网络入口和应用状态，继续检查应用日志
负责人：负责该测试服务的人员
处理过程：按时间记录检查结果、修改和验证
完成条件：订单列表正常加载，相关测试通过
后续改进：根据根因补充监控、文档或发布检查
```

优先级应结合影响范围和紧急程度确定。不要把“着急”作为唯一依据，也不要在工单中粘贴密码或私钥。


## 四、网络通信的三要素

### 4.1 协议、IP、端口分别解决什么问题

**原课依据：第 17–30 页。** 网络通信是在不同设备之间，通过网络传输和交换数据。

| 要素 | 解决的问题 | 入门记忆方式 |
| --- | --- | --- |
| 协议（Protocol） | 通信遵循什么规则和格式 | 按什么规则交流 |
| IP 地址 | 数据应送到哪个网络地址 | 找到通信的目标 |
| 端口（Port） | 找到目标后，交给哪项服务 | 找到服务入口 |

**整理补充例子：** 访问 `https://example.com` 时，域名需要解析出地址，浏览器根据 HTTPS 的规则与目标服务通信；未显式写端口时通常使用 443。远程管理 Linux 则常用 SSH 和 TCP 22。

协议、地址和端口都正确，仍需网络路由可达、访问规则允许、对应服务正在监听，连接才能成功。


### 4.2 常见协议与它们的关系

**原课依据：第 19–20、30 页；下表的分层解释为整理补充。**

| 名称 | 主要作用 | 需要记住的区别 |
| --- | --- | --- |
| IP | 网络寻址与数据包转发 | IPv4、IPv6 是 IP 的两个版本 |
| TCP | 面向连接的传输，提供有序、可靠的字节流 | 常用于需要可靠传输的应用 |
| UDP | 以数据报传输，本身不保证送达和顺序 | 应用可按自身需求处理重传等机制 |
| HTTP | Web 请求与响应 | 负责网页、接口等应用层交互 |
| HTTPS | 使用 TLS 保护 HTTP 通信 | 通常使用 TCP 443；HTTP/3 使用 QUIC／UDP |
| FTP | 文件传输 | 控制连接通常使用 TCP 21，数据连接另有端口 |
| SSH | 加密远程登录和命令操作 | 本模块连接 Linux 云服务器时使用 |

**整理补充：** `TCP/IP` 常指一组网络协议构成的协议体系；HTTP、TCP、IP 处于不同层次，不是互斥的选择。一次 HTTPS 请求通常会同时用到应用层、传输层和网络层的协议。


### 4.3 IP 地址：IPv4 与 IPv6

**原课依据：第 21–25、30 页。**



| 对比项 | IPv4 | IPv6 |
| --- | --- | --- |
| 地址长度 | 32 位 | 128 位 |
| 完整格式 | 4 组，每组 8 位 | 8 组，每组 16 位 |
| 常见写法 | 点分十进制 | 冒号分隔的十六进制 |
| 课件示例 | `192.168.88.102` | `2001:0db8:85a3:0000:0000:8a2e:0370:7334` |

**整理补充：** IPv4 的每组十进制数为 `0–255`；IPv6 可以按规则省略前导零，并用一次 `::` 压缩连续全零组。

课件用“设备地址”帮助理解 IP。更准确地说，IP 是网络接口的地址，设备可能拥有多个地址；地址的唯一性要结合所在网络范围理解，不能把 IP 当作永久的设备身份证。


### 4.4 公网 IP 与私网 IP

**原课依据：第 65–66 页；地址范围与网络条件为整理补充。**

| 类型 | 主要用途 | 从自己的电脑直接连接时 |
| --- | --- | --- |
| 公网 IP | 公共网络中的地址，可在具备路由和授权条件时访问 | 本课 MobaXterm 示例使用云实例的公网 IP |
| 私网 IP | 在 VPC、局域网等内部网络通信 | 电脑需要有通往该网络的路径 |

常见 IPv4 私有地址范围：

- `10.0.0.0/8`：10.0.0.0–10.255.255.255。
- `172.16.0.0/12`：172.16.0.0–172.31.255.255。
- `192.168.0.0/16`：192.168.0.0–192.168.255.255。

**整理补充纠正：** 课件中的“不能使用私网 IP”指电脑没有进入实例内网的连接场景。有 VPN、专线、堡垒机或其他可用路由时，也可以访问私网地址；私网 IP 也不是只有某一家云厂商才可以使用。

有公网 IP 并不等于所有人都能访问：还要看实例状态、路由、端口监听、安全组和系统防火墙。重新创建实例或调整地址后，应重新查询当前 IP。


### 4.5 端口范围与分类

**原课依据：第 26–30 页。** 端口帮助区分服务入口。TCP 与 UDP 的端口号范围都是 **0–65535**。



| 范围 | 分类 | 入门理解 |
| --- | --- | --- |
| 0–1023 | 知名／系统端口；课件称知名（保留）端口 | 常见基础服务使用，如 HTTP 80、HTTPS 443 |
| 1024–49151 | 注册端口 | 许多应用和数据库采用这一范围内的端口 |
| 49152–65535 | 动态／私有端口 | 常用于临时连接和客户端通信 |

**整理补充：** 端口 0 是特殊保留值，不能把整个范围都理解成可直接部署服务的普通端口。客户端临时端口的实际分配范围也取决于操作系统配置，不一定严格等于最后一段。

同一个数字在 TCP 与 UDP 中属于不同的端口空间。端口号是常见约定，不会自动决定服务类型；服务可以改用其他端口。


### 4.6 常用端口速查

**整理补充。** 原课举例了 HTTP 80、HTTPS 443、FTP 21，云控制台截图还出现 SSH 22 与 RDP 3389。

| 服务 | 常用默认端口 | 说明 |
| --- | --- | --- |
| SSH | TCP 22 | Linux 远程登录 |
| HTTP | TCP 80 | Web 与接口服务 |
| HTTPS | TCP 443；HTTP/3 常用 UDP 443 | 加密 Web 通信 |
| FTP 控制连接 | TCP 21 | 数据连接另有端口要求 |
| RDP | TCP／UDP 3389 | 常用于 Windows 远程桌面 |
| MySQL | TCP 3306 | 数据库；通常仅允许所需内部网络访问 |
| Redis | TCP 6379 | 数据服务；通常仅允许所需内部网络访问 |

学习端口的目的，是知道连接对象和排查位置。**把端口列在表中不代表要全部开放。** 本模块的 Linux SSH 练习只需按连接方式配置必要入口。


### 4.7 将三要素应用到一次连接

**整理补充示例。** 下列地址为文档示例地址，不对应真实练习服务器。

```text
远程登录：SSH → 203.0.113.10 → TCP 22
访问网页：HTTP → 203.0.113.10 → TCP 80
访问接口：http://203.0.113.10:8080 → 使用 TCP 8080
```

| 现象 | 可以先区分什么 |
| --- | --- |
| 连不上服务器 | 地址是否正确、路由是否可达、端口是否允许、服务是否监听 |
| SSH 能登录，网页打不开 | SSH 与 Web 是不同服务，分别检查 Web 的监听、规则与应用状态 |
| 浏览器能打开域名，IP 访问结果不同 | 虚拟主机、Host、TLS 证书等配置可能有关 |
| 网络可达，但提示认证失败 | 连接已进入认证阶段，应核对账号、密码或密钥及登录策略 |

不能只靠 `ping` 判断应用是否正常：ICMP 可能被限制；能收到 ping 响应，也不证明 Web 或 SSH 服务可用。


## 五、常见操作系统与 Linux

### 5.1 操作系统在服务器中做什么

**原课依据：第 31–33 页；职责解释为整理补充。**

操作系统负责管理 CPU、内存、存储、进程、网络和用户权限，为应用提供运行基础。

| 常见系统 | 典型使用场景 |
| --- | --- |
| Windows／Windows Server | 个人电脑及 Windows 生态的服务器应用 |
| macOS | 苹果电脑的桌面与开发环境 |
| Unix／Linux | 多种服务器、基础设施与开发环境；Unix 和 Linux 并非同一个系统 |
| Android | 手机及移动设备，底层使用 Linux 内核 |
| 鸿蒙 | 课件图片及总结中列举的系统，具体形态取决于版本和设备 |

本课程后续以 **Linux 云服务器** 作为练习环境。电脑上使用 Windows，并不影响通过网络管理远端 Linux。


### 5.2 为什么本课程选择 Linux

**原课依据：第 32–33 页。** 课件强调 Linux 的性能、稳定性、安全性和远程管理便利性。

**整理补充：更准确地理解这些优点。**

| Linux 常见优势 | 理解方式 |
| --- | --- |
| 丰富的服务器软件生态 | Web、数据库、容器与自动化工具广泛支持 Linux |
| 方便远程与自动化管理 | SSH、命令行、脚本和配置管理适合重复执行 |
| 多用户、多任务与权限机制 | 不同用户和程序可以在受控权限下工作 |
| 可选择较精简的系统环境 | 可以按业务需求安装组件并配置资源 |
| 开源与发行版生态 | 可以学习实现方式，选择不同社区或商业支持 |

课件的 Windows／Linux 对比是一种入门简化：**Windows Server 也支持多用户、远程管理和长期运行；Linux 不会因为开源就天然安全或永不故障。** 实际性能、稳定性与安全性取决于版本、配置、补丁、权限和维护质量。

选系统应考虑应用兼容性、团队能力、支持周期和成本。这里选择 Linux，是为了学习课程所采用的服务器运维方法。


### 5.3 Linux 内核与 Linux 发行版

**原课依据：第 35–36 页。**

Linux 的创建者是 **Linus Torvalds（林纳斯·托瓦兹）**。课件介绍其起源于 1991 年，并写有“1991 年 10 月首次发布”；记忆重点是人物与年份。

| 概念 | 含义 |
| --- | --- |
| Linux 内核 | 操作系统核心，承担进程、内存、设备、文件系统与网络等管理职责 |
| Linux 发行版 | 以 Linux 内核为基础，加上系统工具、软件包、安装与管理机制等形成的完整系统 |

**整理补充：** 课件中的“狭义 Linux”指内核，不应把“由 Linux 编写”理解为 Linux 是编写代码的人。发行版也不只是对内核改名，软件仓库、默认配置、包管理方式和支持政策都可能不同。

```text
Linux 发行版 ≈ Linux 内核 + 系统工具 + 软件包与管理机制 + 默认配置
```

学习时既要识别“这是 Linux”，也要明确 **哪种发行版、哪个版本、什么架构**，再查对应操作方式。


### 5.4 常见 Linux 发行版

**原课依据：第 36–39 页；家族和包管理说明为整理补充。**



| 发行版或体系 | 入门认识 |
| --- | --- |
| CentOS／CentOS Stream | 课件列举的红帽生态发行版；本课使用 CentOS Stream 9 |
| RHEL（Red Hat Enterprise Linux） | 红帽企业 Linux，生产选型时关注其订阅与支持政策 |
| Ubuntu | 基于 Debian 生态的常用发行版，注意具体版本与 LTS 支持周期 |
| 麒麟 | 课件列举的国产发行版，操作需结合产品与版本 |

**整理补充：** 红帽生态常使用 RPM 与 DNF；Debian／Ubuntu 生态常使用 DEB 与 APT。网上一条 Ubuntu 命令未必能直接用于 CentOS Stream，应先确认系统和软件包名称。

本篇先建立发行版概念，包管理、目录结构与常用命令将在后续模块继续学习。


### 5.5 认识 CentOS Stream 9

**原课依据：第 37–39 页。** 课程采用 **CentOS Stream 9 64 位**，强调更新较快、与 RHEL 生态接近、社区支持等特点。

**整理补充核对：** [CentOS 官方说明](https://www.centos.org/centos-stream/)将 CentOS Stream 定位为持续交付的发行版，位于 **Fedora 与 RHEL 之间**，跟踪略领先于 RHEL 的开发更新。

```text
Fedora → CentOS Stream → RHEL
             本课程使用 Stream 9
```

这是便于理解的生态关系，不表示三者每个版本都能一一直接互换。

| 容易混淆的说法 | 正确理解 |
| --- | --- |
| CentOS Stream 等于过去的 CentOS Linux | 两者在发布与 RHEL 的关系上不同 |
| Stream 9 会自动变成 Stream 10 | Stream 9 在该主要版本内持续更新，主要版本迁移需另行处理 |
| 与 RHEL 有关系就等于享有相同企业支持 | CentOS Stream 的生命周期和社区支持要单独核对，不等同于 RHEL 商业订阅 |
| 更新较快就可以无检查用于任何生产系统 | 仍需评估兼容性、支持周期与变更验证 |

复现课程时优先使用同一版本；生产选型则按业务要求评估，不能只因课堂使用就直接决定。


## 六、云服务器的基本介绍

### 6.1 云服务器是什么

**原课依据：第 40–43 页。**

云服务器是云平台提供的计算资源。课件以虚拟化服务器为例：把物理服务器的 CPU、内存、存储等资源划分成独立的虚拟运行环境，用户通过网络按需使用。



| 本地电脑 | 云服务器 |
| --- | --- |
| 直接操作自己身边的机器 | 通过控制台或远程连接管理云上实例 |
| 自行维护硬件与系统 | 厂商维护底层基础设施，用户管理所用系统与应用 |
| 资源通常随本机配置确定 | 可按平台规则选择或调整实例规格 |

**整理补充：** ECS 是本课阿里云示例使用的产品名称；不同厂商的名称并不完全相同。云上还存在裸金属、容器等形态，本模块重点是 Linux 云实例。

使用云服务不会免除应用部署、权限、数据备份和系统维护等工作。


### 6.2 常见云厂商

**原课依据：第 42–43 页。**

| 市场 | 课件列举 |
| --- | --- |
| 国内 | 阿里云、腾讯云、华为云、UCloud、百度云、金山云、京东云 |
| 国外 | AWS、Microsoft Azure、Google Cloud（GCP） |

本课选择 **阿里云** 展示申领、管理和连接流程。不同平台的按钮、产品名和计费规则可能不同，但都需要理解计算、存储、网络、身份与费用。

**整理补充：** 课件第 42 页出现市场份额数字，但没有在该页标明统计时间和来源，因此不把这些比例作为当前市场结论。学习本模块只需识别常见厂商和管理入口。


### 6.3 云实例有哪些组成部分

**整理补充，结合第 47–54 页的配置界面理解。**

| 组件或配置 | 作用 | 常见混淆 |
| --- | --- | --- |
| vCPU 与内存 | 提供运算能力与运行空间 | vCPU 数不等于独占相同数量的物理核心 |
| 镜像 | 决定初始操作系统及可能预装的软件 | 镜像不是实例的全部资源 |
| 系统盘／数据盘 | 存放系统、程序和业务数据 | 磁盘释放规则会影响数据是否保留 |
| VPC／交换机 | 提供实例所属的虚拟网络与子网 | 同在云上不表示自动互通 |
| 公网 IP 与带宽 | 提供所需外部网络连接条件 | 带宽上限与实际流量费用是不同概念 |
| 安全组 | 控制云侧网络访问规则 | 安全组开放端口不代表应用已启动 |
| 用户、密码或密钥 | 认证系统登录身份 | 云账号与 Linux 登录账号不同 |
| 计费与生命周期 | 决定资源费用、到期和释放方式 | 关机、停止、释放的效果不同 |

后面配置 ECS 时，可以用这张表逐项确认“我选择的是什么、为什么需要它”。

| --- | --- |

**原课依据：第 50 页。**

| 需要 | 对应配置 | 整理补充建议 |
| --- | --- | --- |
| 电脑直接 SSH 连接实例 | 可达地址、TCP 22 与相关入站规则 | 将来源限制到需要的管理地址或网络 |

## 八、云服务器的基本使用

### 8.1 启动、停止、重启和释放

**原课依据：第 58 页的控制台实例操作。** 课件演示在实例列表使用“停止”，或在更多菜单的实例状态中选择启动、重启、停止等操作。

**整理补充：区分四种操作。**

| 操作 | 对运行状态的影响 | 需要注意 |
| --- | --- | --- |
| 启动 | 让已停止实例进入运行状态 | 仍需确认应用是否随系统启动 |
| 停止 | 关闭实例的运行 | 保留资源及收费效果因配置和平台规则不同 |
| 重启 | 重新启动实例系统 | 会中断当前连接，应用需重新恢复 |
| 释放 | 删除相应实例资源 | 可能删除随实例释放的磁盘及其中数据 |

课件作业要求尝试重启。练习前确认操作的是自己的练习实例，检查数据保存情况；重启后等待状态恢复，再重新连接并验证应用。

重启不是通用故障修复方式。真实业务中应按维护流程操作，并确认影响、原因与恢复结果。


### 8.2 使用 Workbench 连接

**原课依据：第 59–61 页。**



课堂操作路径：

1. 进入 ECS 实例列表，确认目标实例运行中。
2. 点击该实例的 **远程连接**。
3. 进入 **Workbench** 的登录界面。
4. 核对目标实例、地址、认证方式和用户名。
5. 输入自己配置的认证材料并连接。
6. 看到远程终端提示符后，确认已进入目标服务器。

**整理补充：** Workbench 是浏览器中的远程管理入口；登录云控制台的身份和进入 Linux 的身份仍然不同。能看到实例列表，不代表已登录实例内部系统。


### 8.3 Workbench 中为什么能选择私网地址

**原课依据：第 60 页。** 课堂截图在 Workbench 中选择了私网 IP，并以 root、密码认证登录。



这与后面 MobaXterm 填公网 IP 的示例并不矛盾：**两个入口所处的网络路径不同**。

| 连接场景 | 地址如何选择 |
| --- | --- |
| 课件的 Workbench 入口 | 按平台提供的可达连接通道选择；截图使用私网地址 |
| 电脑没有进入 VPC 的直接 SSH | 通常使用目标实例可达的公网地址 |
| 电脑已通过 VPN、专线等进入内部网络 | 在路由和授权允许时可以使用私网地址 |

**整理补充：** 平台界面和连接通道可能变化，不能只根据“这是网页里的终端”就假定私网必然可达。要结合实际实例与入口说明选择。

图片里的实例、IP 和账户属于授课示例，应换成自己的实例信息。


### 8.4 使用 MobaXterm 建立 SSH 会话

**原课依据：第 62–66 页。** 原课使用 Windows 上的 **MobaXterm（MX）**。



课堂操作步骤：

1. 安装并打开 MobaXterm。
2. 选择 **Session／会话**，创建新会话。
3. 选择 **SSH**。
4. 在 **Remote host／远程主机** 中填写自己实例当前可达的 IP。
5. 在 **Specify username／指定用户名** 中填写实际 Linux 用户。
6. 填写端口；课件采用默认 **22**。
7. 确认后进入连接与认证流程。

| 字段 | 本课示例 | 自己使用时 |
| --- | --- | --- |
| 协议 | SSH | 确认使用 SSH |
| 远程主机 | 课件实例的公网 IP | 填自己的当前地址，不复制课件 IP |
| 用户名 | root | 根据镜像与账号策略确认 |
| 端口 | 22 | 若已调整 SSH 端口，客户端与规则也需匹配 |

**整理补充：** 需要安装工具时使用可信官方来源，并注意版本与许可。工具截图能帮助定位字段，具体界面以本机版本为准。


### 8.5 首次连接时的主机身份提示

**原课依据：第 67 页。** 课件展示首次连接提示，并标注 **Accept**。



**整理补充：** SSH 的主机密钥用于识别连接的服务器。首次连接时，客户端还没有保存该主机的身份信息，所以会提示确认。

推荐理解顺序：

1. 确认目标地址和实例是自己要连接的对象。
2. 能核对指纹时，通过云控制台等可信渠道核对主机密钥指纹。
3. 确认身份后接受，客户端保存相关记录。
4. 以后出现“主机身份发生变化”的警告，先查重建实例、重装系统、IP 变化等原因，再处理记录。

不把“每次点接受”当作固定操作，也不为消除警告而直接关闭主机身份检查。


### 8.6 密码输入与登录成功的判断

**原课依据：第 68–69 页。** 原课提醒在终端中输入密码时，通常 **不会显示字符**，输入后按回车即可。

**整理补充示例，主机名为虚构名称。**

```text
服务器要求输入密码
    ↓
输入自己的密码（终端通常不显示字符或星号）
    ↓
回车
    ↓
出现欢迎信息及 shell 提示符，例如：
[root@lab-server ~]#
```

看懂提示符：

| 部分 | 含义 |
| --- | --- |
| root | 当前用户示例 |
| lab-server | 主机名示例 |
| ~ | 当前用户的家目录 |
| # | 常见 root 提示符形式；普通用户常见 $，但提示符可以自定义 |

出现欢迎文字不能独自证明登录到正确主机，应再核对用户、主机和系统。关闭终端窗口通常只结束当前会话，不等于关闭云服务器或停止资源计费。


### 8.7 登录后怎样确认目标系统

**整理补充：以下均为查看信息的练习命令，不是课件第 1 份的逐字命令。** 在自己的练习服务器终端中手动运行，可记录系统与身份；本篇笔记没有代你执行远程操作。

```bash
whoami
hostname
cat /etc/os-release
uname -r
pwd
```

| 命令 | 查看什么 |
| --- | --- |
| whoami | 当前登录用户 |
| hostname | 主机名 |
| cat /etc/os-release | 操作系统发行版及版本信息 |
| uname -r | 内核版本 |
| pwd | 当前所在目录 |

建议把实际结果填入实验记录：用户、主机名、发行版、版本、连接方式和验证时间。不要把本地 Windows 终端输出当作 Linux 服务器输出。

离开 shell 可使用 `exit`。它用于退出当前会话，资源生命周期应在云控制台另外确认。


### 8.8 远程连接失败时的排查顺序

**整理补充，围绕本模块的地址、端口、实例和认证概念。**

先查看错误停在哪一步，再按证据排查：

| 现象 | 优先检查 |
| --- | --- |
| 控制台找不到实例 | 云账号、地域、资源是否已创建或被释放 |
| 连接超时 | 实例是否运行、目标地址、网络路径、安全组与防火墙、是否使用了不可达的私网 IP |
| Connection refused／连接被拒绝 | 目标端口是否监听、端口是否填错；也可能是网络策略主动拒绝 |
| Permission denied／认证失败 | 用户名、密码或密钥、认证方式、用户权限与 SSH 策略 |
| 主机身份变化警告 | 地址对应的实例是否变化、系统是否重装、指纹是否匹配 |
| 登录后出现意外系统或用户 | 核对地址、实例、会话配置与当前身份 |

```text
实例与地址 → 网络可达性 → 端口与访问规则 → SSH 服务 → 用户认证 → 登录后核对
```

排查时一次改变一个条件，记录结果；不要同时重启、换密码、修改规则，导致无法判断原问题。

### 9.2 三个概念练习

**整理补充自测，先尝试回答再看分析。**

**练习一：** 应用在开发环境正常，生产环境启动后无法连接数据库。应该先看什么？

分析：先区分环境，核对生产配置中的数据库地址、端口、账号、网络路径和权限，再看日志；不要直接把生产数据拿到开发环境试。

**练习二：** 电脑上 MobaXterm 填写实例私网 IP 后超时，而 Workbench 可以连接。一定是 SSH 服务坏了吗？

分析：不一定。两个入口的网络路径可能不同；先确认电脑是否能进入 VPC、地址是否可达，再检查规则和服务。

**练习三：** SSH 登录成功，但网页打不开。能否直接认为云服务器正常、问题一定在代码？

分析：SSH 成功说明该连接路径和认证可用；网页还需要 Web 服务监听、对应端口规则、应用配置等条件，需继续按证据排查。

| 退出终端就停止云费用 | 会话结束与资源停止、释放、计费是不同概念 |

### 10.4 参考答案

1. 通过技术手段保障企业 IT 系统和网络持续、安全、稳定运行，保障业务连续性。
2. 系统、数据库、自动化、容器、云计算、信创；部署、监控、告警、故障快速恢复。
3. 产品设计 → 项目研发 → 项目测试 → 项目上线 → 运维；测试发现问题会返回修改。
4. 开发环境供开发调试，测试环境供质量验证，生产环境对真实用户提供服务；运维尤其关注生产。
5. 记录、分配与处理请求、问题和变更；提交工单 → 自动分配 → 处理反馈 → 验证关闭。
6. 协议规定通信规则，IP 帮助定位网络目标，端口标识服务入口。
7. IPv4 为 32 位、点分十进制；IPv6 为 128 位、冒号分隔十六进制。端口 0–65535，分为 0–1023、1024–49151、49152–65535 三段，端口 0 有特殊含义。
8. 内核提供核心管理功能，发行版还包含系统工具、软件包和管理机制；本课采用 CentOS Stream 9 64 位。
9. 云平台按需提供计算资源；配置计费、地域、规格、镜像、存储、网络、登录方式后核对实例，通过 Workbench 或 SSH 客户端连接并验证。
10. 可能存在地址错误、实例未运行、路由不可达、系统防火墙限制、SSH 未监听或认证不匹配等原因。
11. 两者网络路径不同；电脑没有进入 VPC 时通常不能直接到达私网地址，具备 VPN、专线等路径时也可使用私网。
12. 云账号管理云资源；Linux 用户是系统身份；root 是超级用户；sudo 按授权临时以其他身份执行命令。
13. 首次确认并保存服务器身份；以后变化时先核对实例和指纹，再处理，不直接忽略警告。
14. 停止改变运行状态，重启重新启动系统，释放可能删除资源及数据，退出终端只结束会话；实际收费与数据保留需看平台规则。
