# 01 AI做人形猫咪账号

## 资料定位与课程目标

本篇对应网盘课件 **《01-AI做人形猫咪账号》**，课件共 14 页。课程用一个“治愈猫咪账号”案例，演示如何把对标账号拆成提示词、图片、图文发布和视频四个环节。

课件开场的宣传语是“AI 做治愈猫咪账号，新手也能轻松月赚 3 万”。这句话属于课程宣传口径，笔记只记录课件展示的方法，不把收益数字当成结果承诺。实际效果还会受到内容质量、平台规则、账号运营和工具额度等因素影响。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)（提取码：1234）。

## 一、整体流程：先做图文，再扩展到视频

课件把“AI 做治愈猫咪账号”拆成四步：

1. **AI 生成提示词**：把对标猫咪形象交给 AI 分析，再生成标题、文案和图片提示词。
2. **AI 生成猫咪图**：把提示词输入即梦 AI；需要保持角色特征时，导入参考图。
3. **下载并发布猫咪图文**：先把图片变高清、下载原图，再到小红书后台发布，补充文案和话题。
4. **把猫咪图片转成视频**：使用 AI 图生视频，再用剪映导入素材、添加背景音乐，输出成片。

课件强调这一套流程全程使用国产工具：豆包 AI 负责提示词、标题和文案，即梦 AI 负责生图与图生视频，最终在小红书发布内容。

## 二、AI 生成提示词：先分析对标猫咪形象

课件先让 AI 分析一张对标图片，关注两类信息：

- **视觉形象**：猫咪的品种或外观、站立姿态、服装、配色、场景和画面氛围。
- **内容表达**：图片配套文案的主题、语气、标题写法和互动方式。

示例图中的猫咪穿着白色时尚服装，使用双脚站立，像人类模特一样展示穿搭。课件文案以“小猫咪时装周开启”为主题，并用“酷酷喵、帅气喵、可爱喵、呆呆喵”等称呼增加轻松感。这里的分析结果会继续服务于后面的提示词生成。

使用工具：豆包 AI（[doubao.com/chat](https://www.doubao.com/chat/)）。

### 2.1 用豆包一次生成图片提示词、标题和文案

课件给出的 Prompt 可以直接作为账号内容生产模板。它要求模型先理解账号定位，再把每组内容拆成图片提示词、标题和文案三部分，并用表格输出 10 组内容。

~~~text
身份：现在你是一名专业的小红书运营

技能：你很擅长运营猫咪穿搭分享的账号，这个账号的特色在于猫咪模特会像人类一样站立并展示穿搭，注意，猫咪在分享穿搭时会使用双脚站立，这是它们像人类的关键特点。

要求：现在需要你为这个账号创作内容，包括 3 部分：图片、标题、文案。对于图片，我需要你提供相应的 AI 生图提示词，对于标题和文案，也请你将它们用表格列给我。

输出：现在请你帮我列出 10 组相应内容
~~~

这段 Prompt 的关键不在于某个固定句式，而在于把**身份、账号特色、输出字段和数量**交代清楚。生成结果后，先检查猫咪是否保持双脚站立、穿搭展示和治愈风格，再选择适合发布的组。

## 三、AI 生成猫咪图：使用即梦 AI

拿到生图提示词后，课件把它输入即梦 AI 生成猫咪图片。

操作顺序：

1. 打开即梦 AI 的 AI 创作页面：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)。
2. 将豆包生成的图片提示词输入生图区域。
3. 提交生成任务，等待图片结果。
4. 检查画面是否符合账号设定，重点看猫咪是否站立、服装是否完整、画面是否有治愈感。
5. 选出适合后续发布的图片。

课件展示的是工具操作流程，具体按钮名称和免费额度可能随网站版本调整。

### 3.1 导入参考图，保持角色特征

课件的下一步是在即梦 AI 中**导入参考图**，让模型参考已有角色的外观特征。这样做的目的，是让同一个账号里的猫咪形象更容易保持一致。

可以按下面的顺序检查：

- 先准备一张已经确认效果较好的猫咪图；
- 在即梦 AI 的参考图入口导入图片；
- 再输入新的穿搭、动作或场景提示词；
- 生成后对照参考图，检查脸部、毛色、体型和整体风格是否偏离。

参考图用于维持角色特征，新的文字提示词用于改变本次内容的穿搭或场景；两者需要一起检查。

## 四、下载高清图片并发布猫咪图文

### 4.1 变高清并下载原图

课件把发布前的图片处理分成两步：

1. 使用 AI 工具把图片变高清；
2. 下载原图，保留发布所需的清晰度。

课件还列出了百度图片工具作为去水印入口：[image.baidu.com](https://image.baidu.com)。使用第三方图片处理或去水印工具时，应确认图片来源和使用权限，避免处理或发布没有授权的素材。

### 4.2 在小红书后台发布作品

图片准备好后，进入小红书后台发布作品：

1. 选择处理好的猫咪图片；
2. 填写豆包生成的标题和文案；
3. 添加与内容相关的 #话题；
4. 预览排版，确认图片、文字和话题对应；
5. 发布作品。

课件只展示发布入口和“添加文案、#话题”的要点，没有规定固定的标题长度或话题数量。发布时以小红书当前页面规则为准。

## 五、从图文扩展到视频

课件在图文发布后提出一个问题：**图片感觉不够生动，能不能生成视频发布？**

这一步说明账号内容可以从静态图文扩展到动态视频。先用一张猫咪图做图生视频，再把生成的视频剪成可发布的短片。

### 5.1 使用 AI 把猫咪图片转成视频

课件展示的图生视频流程包括：

1. 选择一张猫咪图片；
2. 在 AI 转视频工具中描述画面；
3. 描述希望出现的动作；
4. 点击“立即生成”，等待视频结果；
5. 检查生成的视频是否保持猫咪形象，动作是否自然。

提示词至少要让工具知道“画面主体是谁”和“需要发生什么动作”。例如，先说明猫咪穿着什么、站在什么场景，再补充走路、转身或展示服装等动作。课件没有给出固定动作 Prompt，所以动作描述应以当次图片和账号风格为准。

### 5.2 用剪映剪成片

AI 生成视频后，课件使用剪映做最后处理：

1. 导入 AI 生成的视频素材；
2. 添加背景音乐；
3. 检查画面和音乐的衔接；
4. 生成视频成片。

剪映负责把 AI 生成的片段整理成可发布的成片。课件只列出导入素材、添加背景音乐和生成视频三个动作，没有要求复杂特效；先保证主体、动作和声音完整。

## 六、课程小结：一条可复用的猫咪账号生产链

课件最后把方法归纳为三类工具和四个产出环节：

- 用 **豆包 AI** 生成 AI 生图提示词、文案和标题；
- 用 **即梦 AI** 结合提示词生成高清猫咪图片；
- 在 **小红书** 发布图文，也可以把图片转成视频发布；
- 内容链路是：**生成提示词 → 生成图片 → 图文发布 → 图生视频 → 剪辑发布**。

后续复用这套流程时，先固定猫咪角色特征，再批量生成穿搭和场景内容；每次发布前检查图片授权、平台规则和工具当前的使用限制。
