## 二、整体流程：先规划分镜，再逐个制作素材

第 3、14 页将制作流程归纳为四步：

| 步骤 | 使用工具 | 操作与产出 |
| --- | --- | --- |
| 1. 生成治愈 Vlog 脚本 | DeepSeek／豆包 | 确定生活主题，输出分镜及每个镜头的 AI 绘画提示词 |
| 2. 生成 Vlog 配图 | 即梦 AI | 将提示词与参考图结合，生成各场景的人物插画 |
| 3. 生成 Vlog 视频 | 即梦 AI | 把选定图片作为参考首帧，描述动作和环境变化，生成短视频 |
| 4. 剪辑完整成片 | 剪映 | 按顺序组合镜头，调整比例与画面变化，添加音乐、环境声和动作音效 |

本课有两类用途不同的提示词：**生图提示词描述某一刻的画面，视频提示词描述这张图接下来怎样动**。前者确定人物、构图与画风，后者确定动作、环境和光影变化。

原课先生成 8 个分镜，再重点演示办公与乘车配图及视频。最终示例还出现进门、卸妆和厨房画面，课件没有逐一展示所有镜头的完整生成操作。


## 三、用 DeepSeek／豆包生成分镜脚本与生图提示词

### 3.1 原课完整对话提示词

在 DeepSeek 或豆包中输入以下原课指令。提示词中的 `Al` 按课件文字保留，指向的仍是 AI 绘画提示词：

~~~text
#身份：你是一个治愈系生活赛道的vlog博主，你非常擅长制作治愈系的爆款vlog视频。
#技能：你擅长输出vlog分镜头脚本以及对应的Al绘画提示词，且我需要用到这些AI绘画提示词，在其他AI大模型中生成图片，所以这些Vlog分镜画面逻辑要通顺、剧情要合理，AI提示词也需要保持风格一致，剧情连贯。
~~~

## 七、添加背景音乐、环境声与动作音效

### 7.1 三类声音分别承担什么作用

第 13 页要求“添加环境音效、动作音效”。从时间线可以分清三类声音：

| 声音层 | 原课可见素材或片段名 | 对应作用 |
| --- | --- | --- |
| 背景音乐 | `蜜桃物语` | 贯穿多个镜头，提供整条短片的听觉氛围 |
| 环境音 | `城市街道汽车喇叭声`、`嘈杂街道人潮车流声` | 交代办公楼外、街道或乘车等环境 |
| 动作音效 | `打印文件`、`开门声`、`涂抹洗手液音效`、`切菜声1`等 | 与打印、进门、清洁、切菜等动作相对应 |

音乐可以跨越多个片段，动作音效通常只覆盖相关动作，环境声则根据场景安排。将它们放在不同音轨，便于单独移动和调整。

课件时间线还显示了一个名称以“使用香…”开头的音效片段，后半部分被遮挡，这里不补全名称。素材名只是当时工程中的标签，课件没有提供音效下载地址或完整素材包。






### 7.2 让声音与画面中的动作对齐

结合原课时间线，复做时可按下面顺序检查：

1. 先按分镜安排好画面，确定每个动作发生的位置。
2. 添加背景音乐，试听它与整条短片的节奏是否协调。
3. 在相应场景添加环境声，切换场景时检查声音衔接。
4. 将开门、清洁、切菜等音效移动到对应动作处。
5. 从头播放，检查各层音量与起止时间，避免多个声音同时过响。

原课没有给出固定音量、淡入淡出时长或精确音效切点，上面是依照示例整理的操作检查建议。前面两段即梦视频不含音频，因此声音需要在剪辑阶段另行制作。

第 13 页最后写“去除水印，导出视频即可”，没有展示去水印或导出设置。第 14 页成品仍能看到即梦标识，不能据此认定已完成去水印。导出时使用平台允许的方式，保留要求展示的 AI 标识，并按发布要求选择参数。

## 九、课程小结

原课第 14 页将本篇方法归纳为四步：

1. 使用 **DeepSeek／豆包** 生成治愈 Vlog 分镜脚本和生图提示词。
2. 使用 **即梦 AI** 结合参考图生成各场景配图。
3. 使用 **即梦图片生视频**，给分镜图片添加动作和环境变化。
4. 使用 **剪映** 组合画面，安排背景音乐、环境声与动作音效，导出成片。

本课的重点是把一个日常主题拆成连贯镜头，并在不同场景中维持人物与画风。先确定完整分镜，再逐个生成配图与视频，最后用声音和剪辑节奏把它们组织成一段生活记录。
