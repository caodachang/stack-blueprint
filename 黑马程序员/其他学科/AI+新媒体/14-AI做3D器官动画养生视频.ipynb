# 14 AI做3D器官动画养生视频


## 资料定位与课程目标

本篇对应课件 **《14-AI做3D器官动画养生视频》**，共 15 页。第 2 页展示案例，第 3 页给出五步流程，第 4—11 页依次讲解文案、分镜、生图和图生视频，第 12—13 页讲剪映合成，第 14 页小结，第 15 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇以本地 PPTX 的正文、操作截图和 4 段内嵌视频为准，其中 2 段为案例录屏，2 段为生成的动画示例。

课程用拟人化器官讲述饮食主题：让器官有表情、拿食物或饮品，再生成短动画，最后给每个镜头配上对应的旁白和字幕。“3D”在这里指卡通视觉风格，课件没有教授三维建模或医学解剖建模。

养生文案是课程中的 AI 创作示例。“器官最爱某食物”“定时排毒”“穴位唤醒自愈力”等说法未在课件中提供医学依据，下面按资料保留，用来学习内容制作，实际使用前应核对、修改相关健康表述。


## 一、先看器官拟人动画的表达方式

第 2 页两段案例的账号均显示为“聚友食养馆”。画面以单个器官角色为主体，配一个对应食物，字幕采用“我是××，喜欢××”的句式；不同器官依次出场，形成清单式节奏。

| 案例观察 | 录屏可见的表现 | 制作上可以学习的地方 |
| --- | --- | --- |
| 心脏与红薯 | 卡通心脏长出手脚和五官，旁边是红薯；画面有明显表情变化 | 让器官、食物和一句短台词同时出现，降低理解成本 |
| 血管与西兰花 | 拟人血管举着西兰花，置于厨房，底部有白字黑描边字幕 | 食物成为角色手里的道具，厨房提供饮食场景 |

## 二、五步制作流程

第 3 页的流程按素材依赖展开：先定台词，再确定每个镜头画什么、怎么动，最后合成。

| 阶段 | 工具 | 任务 | 主要产出 |
| --- | --- | --- | --- |
| 1. 文案 | DeepSeek | 参考 3 篇样例，再写 3 篇器官养生文案 | 选定并核对的旁白 |
| 2. 分镜 | DeepSeek | 根据第 1 条新文案，生成 13 个绘画提示词 | 12 个器官镜头＋1 个综合镜头 |
| 3. 图片 | 即梦 AI | 逐条输入提示词，选择适合的 3D 卡通图 | 每个镜头的静态图片 |
| 4. 视频 | 即梦 AI | 用图片加动作提示词生成短视频 | 可供剪接的器官动画素材 |
| 5. 成片 | 剪映 | 拼接画面，添加口播稿、朗读音频、字幕、提示与 BGM | 导出的视频 |

同一段台词要在文案、分镜、图片、视频和字幕中保持一致。例如第一句是“我是心脏，最爱燕麦”，它对应心脏角色、燕麦牛奶道具和心脏的动画片段。

课件总览标有“免费生成”，但没有说明账号额度或收费条件，操作时以工具实际界面为准。

## 四、把文案转成 13 个绘画分镜

### 4.1 继续对话的完整提示词

第 6 页在同一段对话中，要求根据“上述第 1 条”生成分镜：

```text
写的不错，请选择上述你生成的第1条文案，根据文案内容帮我生成即梦AI出图13个分镜提示词，提示词需要包含以下内容，以下【】中的内容请根据第1条实际文案做相应调换：

1.图片风格：每张图片风格为3D卡通，形象简洁，色彩鲜艳真实，轻松日常的氛围，适合用于宣传营养饮食的知识
2.捕述主体：拟人化的【xx器官】形象，带有表情
3.捕述动作：【xx器官】正在用【吸管需一杯牛奶】
4.描述场最：背景是【厨房场景】，有【餐具桌椅】等元素
```

原文中的“捕述”“场最”“吸管需一杯牛奶”存在错字。复做时可分别按“描述”“场景”“用吸管喝一杯牛奶”理解；方括号内只是示例，应替换成每句台词实际对应的器官、食物和动作。

| 约束 | 需要交代的内容 | 本课示例 |
| --- | --- | --- |
| 风格 | 视觉风格、色彩、氛围 | 3D 卡通、轻松日常、色彩鲜艳 |
| 主体 | 什么角色、什么表情 | 拟人心脏，开心眯眼笑 |
| 动作 | 角色与食物怎样互动 | 双手捧杯、用吸管喝饮品 |
| 场景 | 背景与辅助道具 | 厨房、橱柜、食材、餐具 |

第 1 篇新文案有 **12 个器官或身体部位条目**，分镜结果在此基础上增加“综合展示”，所以总数是 **13 镜**。这里的“器官”是课件的统称，清单中也包含头发、皮肤、血管等不同类别的身体部位。

## 六、用图片生成器官动画

### 6.1 从选中的图片进入“生成视频”

小结页的图片详情截图展示了 **“生成视频”** 入口。操作顺序是先选定图片，再从详情进入视频生成，将该图作为画面参考。



图片提示词负责角色造型、道具和背景；动作提示词继续交代“说话、喝饮品、身体扭动”等变化。图片与动作中的器官、食物要一致，避免在图生视频时换了角色身份。

## 七、剪映中拼接画面、添加口播与朗读

### 7.1 让音频、画面、文案逐句对应

第 12 页给出三个任务：

1. **导入画面，拼接画面**：按分镜顺序排列生成的视频，检查每个片段对应的器官和食物。
2. **添加口播稿，朗读音频**：将选定文案加入工程，生成朗读。下一页左侧“文本”区可见“添加口播稿”入口，但课件没有展开具体音色与语速设置。
3. **音频＋画面＋文案三者对齐**：每句声音、字幕与对应的动画同时出现，再按句子长度调整镜头。



截图中的心脏镜头字幕是“我是心脏，最爱燕麦”，采用白字黑描边。左侧打开的是画面特效库，只能确认当时浏览了特效，课件没有指定必须添加哪一个效果。



这张时间线仍有未对齐之处：在心脏、肝脏、肺部之后又重复了前三种素材，第四句字幕已经是“我是肠道，最爱香蕉”，下方却仍为心脏画面和心脏朗读片段。应按器官逐句替换、校对，不能直接把示例时间线视为已完成的成片。

## 八、复做时的对应关系与易错点

| 容易出错的地方 | 本课材料中的例子 | 复做时的处理 |
| --- | --- | --- |
| 分镜数量不清楚 | 12 个条目，却要求 13 个分镜 | 12 个单独角色镜头，加 1 个综合展示镜头 |
| 文字名称沿用错误 | 肝脏图生视频提示词写成“脾脏” | 对照文案、图片与动作提示词，统一器官名称 |
| 有素材就反复铺到后面 | 肠道台词下方仍是心脏素材 | 为每句台词替换正确画面与朗读片段 |
| 图生视频的“说话”等于已经配好音 | 即梦页面只有图片和动作提示词 | 朗读在剪映添加，再和字幕、镜头逐句对齐 |
| 5 秒素材等于最终镜头必须保留 5 秒 | 示例每段动画约 5 秒，但台词长短不同 | 按朗读长度截取可用动作，再检查衔接 |
| 提示层与字幕时长遗漏 | 提示轨道约 15 秒，工程约 33 秒 | 检查提示与字幕覆盖完整输出范围 |

复做整条视频时，可以先按“01心脏、02肝脏……”保存对应图片、视频与台词，合成时按同一顺序使用。清单式内容最需要检查的是每句说什么、画面是什么，而后才是动画与特效。


## 九、课程小结

第 14 页回顾的制作顺序是：**DeepSeek 写器官养生文案 → 生成 13 个分镜绘画提示词 → 即梦生图 → 即梦图生视频 → 剪映朗读、合成并导出**。

本课可复用的做法，是把“一个角色＋一个食物＋一句话”做成重复的镜头结构，再通过拟人表情和简单动作形成连续视频。先保证文案、角色、食物与声音逐条对应，再统一画风和字幕，才能把分散生成的素材接成完整内容。
