# 20 AI新闻日报稳定涨粉搞流量


## 课程目标与制作流程

本篇对应课件 **《20-AI新闻日报稳定涨粉搞流量》**，共 13 页。封面主题是“AI做行业新闻日报，稳定涨粉搞流量”，实操围绕一个固定领域，把网页新闻整理为有日期、有序号的日报，再转换成长图用于社交平台。

| 阶段 | 课件中的做法 | 产出 |
| --- | --- | --- |
| 搜集新闻 | 选择信息量丰富的垂类新闻网页 | 待处理的网址 |
| 提取新闻 | 用 Gobble Bot 提取网页内容 | `all-in-1-file.txt` |
| AI精写 | 在 Kimi 中上传文本，并按提示词整理 | Markdown格式的新闻日报 |
| 转换长图 | 用 ReadPo 海报编辑器排版 | 可保存、发布的日报长图 |

“垂类”指围绕同一领域持续选题，例如足球、篮球或AI行业。账号可以保持相对固定的选题范围、日报结构和视觉风格，读者更容易快速浏览与识别。

本课案例日期为 **2025年4月25日和4月27日**，属于历史演示。笔记保留其流程与界面记录，实际制作时需要替换为本次日期和对应新闻。课件展示了文章阅读与互动数字，没有提供连续涨粉数据，因此“稳定涨粉”是课程定位，不能据此保证运营结果。


## 一、先看案例：把多条新闻组织成一篇日报

第 2 页用三张手机文章截图展示成品形态，标注“网络原图”。页面中能看到“热点追踪”“今日热议 / HOT NEWS”等栏目，以及 **新闻序号、醒目的标题、分段正文和结尾互动引导**。



日报的阅读顺序通常是先看日期与主题，再浏览各条标题，最后选择感兴趣的内容细读。编号帮助读者定位，小标题区分事件，段落间留白让手机阅读更轻松。

案例结尾有邀请评论、关注后续内容等表达。图中可见“阅读2.5万”等单篇数据，这些数字用于认识案例页面，不能直接推出整个账号的长期表现。后面的实操重点是做出结构清楚、事实可核对的日报。

### 3.2 在结果区下载合并文本

第 5—6 页右侧出现 **“Gobbling results - 19028 chars - Download all-in-1-file”**，下方有标题、图片地址和新闻链接的预览。点击 **“Download all-in-1-file”**，下载得到 `all-in-1-file.txt`。

## 四、检查TXT：保留新闻，去掉重复和杂项

第 6 页打开的文件名为 **`all-in-1-file.txt`**。文本里能看到Markdown形式的链接、图片地址、重复出现的新闻标题，以及 `* 1`、`* 2` 等分页内容，后面还有“热门推荐”。



从截图可以确认，提取结果仍是待整理的网页材料。部分条目主要提供标题和链接，不能据此认定每篇新闻的完整正文都已下载。

| 检查对象 | 处理方法 |
| --- | --- |
| 相同链接或重复标题 | 去重，避免日报把同一条新闻反复收录 |
| 同一事件的多个相关报道 | 合并核心信息，保留各自来源；有分歧时继续核对 |
| 图片地址、导航、分页、推荐区 | 与新闻内容分开，去掉无关文本 |
| 只有标题、没有正文 | 打开来源链接补充正文，或标注“待补充正文” |
| 日期与领域 | 确认是否符合本期日报的时间范围和主题 |

“AI改写”用于整理已有信息。只有标题时，不能为了凑出“正文详细介绍”而补写原文未提供的人物发言、过程细节或评价。日报篇幅应由有效材料决定。


## 五、让Kimi设计新闻日报提示词

### 5.1 先描述想要的日报格式

第 7 页使用 [Kimi](https://kimi.moonshot.cn/)。课件给出的完整请求是：

```text
帮我设计一个新闻内容提取的提示词，我将网页内容提供给你，你就可以从中提取出标题和内容，并以markdown格式输出，最后形成一篇新闻日报。

要求新闻日报有日期、有新闻序号排列、每个新闻有小标题、摘要、正文详细介绍，让人一目了然。
```

这一步是在让AI先写出“如何处理新闻”的规则，之后才把规则与网页文本一起交给新会话。请求已明确日期、序号、小标题、摘要和正文详细介绍五类结构。

### 5.2 课件中可完整读出的结构化提示词部分

截图中的回复采用角色、背景、能力、目标、约束、输出格式和工作流程的结构。下面转录 **从Role到Workflow的可见部分**，保留原文的 `Constrains` 拼写：

```text
- Role: 新闻内容提取与编辑专家
- Background: 用户需要从网页中提取新闻标题和内容，并以Markdown格式整理成具有日期、新闻序号排列、小标题、摘要和正文详细介绍的新闻日报，以便快速获取和整理新闻信息，提高工作效率。
- Profile: 你是一位经验丰富的新闻编辑，擅长从海量信息中精准提取关键内容，并将其整理成结构清晰、内容完整的新闻报道。你对新闻的排版和格式化有着深刻的理解，能够确保新闻日报的可读性和逻辑性。
- Skills: 你具备强大的文本分析能力、信息筛选能力以及Markdown格式化技巧，能够快速识别新闻标题和正文内容，并将其整理成规范的新闻日报格式。你还能根据新闻内容添加适当的序号和小标题，提取摘要，使新闻日报一目了然。
- Goals: 从提供的网页内容中提取新闻标题和正文内容，以Markdown格式输出，生成一篇具有日期、新闻序号排列、小标题、摘要和正文详细介绍的新闻日报。
- Constrains: 提取的新闻内容应准确无误，标题和正文内容应完整且不包含无关信息，Markdown格式应规范，新闻日报应具有良好的可读性和逻辑性。
- OutputFormat: Markdown格式
- Workflow:
  1. 从提供的网页内容中识别并提取新闻标题。
  2. 从网页内容中提取新闻正文内容，去除无关信息和广告。
  3. 为每条新闻添加序号和小标题。
  4. 从正文内容中提取摘要，确保摘要简洁明了，涵盖新闻核心要点。
  5. 将提取的标题、小标题、摘要和正文内容整理成Markdown格式，生成新闻日报。
  6. 在新闻日报的开头添加日期，确保整体结构清晰、一目了然。
```



`Workflow` 后面还有 `Examples`，但截图只露出示例开头，没有展示完整内容；因此这里不补写缺失的示例。下一页输入框另能读到 `Initialization` 结尾，见下一节。

这份规则解决了输出结构问题。实际处理时还应明确本次日期、素材范围和正文不足时的处理方法，避免模型自行猜测。


## 六、新建会话：上传TXT与提示词，生成日报

### 6.1 把材料与处理规则一起交给Kimi

第 8 页的操作是 **“新建会话 上传文本+提示词”**。附件卡片显示 `all-in-1-file.txt`，大小为 **`23.89 KB`**，输入区粘贴前面生成的提示词。



操作时确认附件已经上传，再提交提示词。截图下方显示“已联网”和“长思考(k1.5)”，这是录课时的界面状态；整理附件的关键仍是让模型明确处理哪份文件、依据哪些材料。

### 6.2 处理提示词中的开场白

输入框末尾可读到以下初始化要求：

```text
- Initialization: 在第一次对话中，请直接输出以下：您好！作为新闻内容提取与编辑专家，我将根据您提供的网页内容，提取新闻标题和正文内容，并以Markdown格式输出，生成一篇具有日期、新闻序号排列、小标题、摘要和正文详细介绍的新闻日报。请提供网页内容，我将为您处理。
```

这段会要求模型在第一次回复中问候并索要网页内容。附件已上传时，可删除这一段，或明确补充：**“附件已经提供，请直接读取并生成日报。”** 否则可能只收到一段开场白。

### 6.3 检查生成结果的结构与内容

结果截图的标题是 **“2025年4月27日 新闻日报”**，依次出现“国王杯决赛巴萨夺冠”“吕迪格赛场奇葩行为”“国家德比裁判争议”等编号条目，每条带有“摘要”和“正文详细介绍”。



先检查日期和序号，再把每条内容与来源对应，核对人名、时间、数字、事件结果与引述。课件中的生成段落是演示输出，不能仅凭排版完整就认定事实准确。确认后保留Markdown源文本，下一步用于长图排版。


## 七、实操补充：让输出更容易核对

以下提示词是根据本课流程补充的实用版本，**不是课件原文**。上传TXT后，可填写日期、领域和时间范围再使用：

```text
请读取我已上传的 all-in-1-file.txt，直接整理新闻日报。
日报日期：【填写本次日报日期】
选题领域：【填写领域，例如足球或AI行业】
收录时间范围：【填写开始与结束时间】

1. 仅使用附件中能够确认的信息，保留新闻来源链接；不要自行补写人物发言、事件背景、数字或结论。
2. 删除导航、广告、分页、热门推荐和图片链接等杂项；相同链接去重，同一事件的多条信息合并整理。
3. 逐条检查发布时间与领域。时间不明确、超出范围或不属于本领域的条目，列入“待核对”，不强行放入正式日报。
4. 如果原文只有标题和链接，没有新闻正文，列入“待补充正文”并保留链接，不生成详细介绍。正文不足时说明缺少什么。
5. 正式条目按“编号小标题 → 摘要 → 正文详细介绍 → 来源”输出。摘要简明概括事实，正文只整理原文支持的内容，不重复堆砌摘要。
6. 使用Markdown：一级标题写日期与日报名称，二级标题写编号与新闻标题，三级标题写摘要和正文详细介绍。条目序号连续。
7. 附件已经提供，请直接处理，不再输出索要网页内容的开场白。若无法读取附件，明确说明，不生成新闻。
```

这份补充把三件事写清楚：处理已经上传的附件；没有正文时先补材料；每条新闻保留来源。若确实要另行联网补充，应明确列出实际查阅到的链接，并逐条核对，不把模型生成的说明当作已取得的原文。

整理结束后，把“待核对”和“待补充正文”的项目留在编辑清单，先解决缺失项，再决定是否进入发布版。


## 八、用ReadPo把Markdown日报排成长图

### 8.1 打开免费海报编辑器

课件使用 [ReadPo](https://readpo.com/zh)，进入顶部 **“免费海报编辑器”**。页面左侧编辑Markdown，右侧实时预览海报。把审核后的日报源文本粘贴到左侧，再观察右侧标题、段落和留白。



如果复制内容时带上了包围整篇的代码围栏（开头和结尾的三个反引号），粘贴时去掉外层围栏，保留正文中的标题符号与空行，避免整篇显示成代码块。

### 8.2 用标题层级控制日报结构

第 9 页左侧实际使用 `#` 表示日报标题，`##` 表示编号新闻标题，`###` 表示“摘要”和“正文详细介绍”。符号后面留一个空格，正文段落之间用空行分隔。



下面是方便替换的结构模板，**占位文字与“来源”行是笔记补充，不是课件逐字内容**：

```markdown
# 【年】年【月】月【日】日 【领域】新闻日报

## 1. 【新闻小标题】

### 摘要
【用简短文字概括已核实的核心事实】

### 正文详细介绍
【根据已取得的新闻正文整理，保留必要的时间、主体、事件与结果】

来源：【媒体或网站名称】（【原文链接】）

## 2. 【下一条新闻小标题】

### 摘要
【摘要】

### 正文详细介绍
【正文】

来源：【媒体或网站名称】（【原文链接】）
```

粘贴后检查右侧是否正确区分总标题、新闻标题和栏目。出现标题变普通文字、内容挤在一起等情况时，先回到左侧检查井号后的空格与段落空行，再调整外观。

### 10.3 发布前检查

1. 核对日期、新闻所属领域与收录范围，确认没有混入过期或无关条目。
2. 检查标题序号是否连续，同一事件是否重复，摘要与正文是否大量重复。
3. 回看原文，核对姓名、机构、时间、数字和引用，并保留可追溯的来源。
4. 在手机上查看文字大小、背景对比与图的首尾，确认没有缺段、截字或多余空白。
5. 按目标平台要求完成发布，并记录后续阅读、收藏、评论和关注变化，用实际反馈调整选题与篇幅。


## 十一、课程小结

第 12 页把本课归纳为四步：**搜集新闻 → 提取新闻 → AI精写 → 转换长图**。第 13 页没有新增教学内容。

先围绕账号领域选择信息丰富的新闻网页，用Gobble Bot提取并下载TXT。打开文件检查标题、正文、重复项与杂项，再把材料和处理规则交给Kimi，输出有日期、编号和清楚层级的Markdown日报。

确认事实和来源后，在ReadPo“免费海报编辑器”粘贴Markdown，设置背景、排版主题、文字大小和边距。检查手机阅读效果，保存完整长图，再用于平台发布。

日常复用时，保留提示词、Markdown结构和视觉风格，更新每期日期与新闻材料。稳定的制作流程有助于持续产出，内容是否准确、是否符合读者需求，仍需逐期检查。
