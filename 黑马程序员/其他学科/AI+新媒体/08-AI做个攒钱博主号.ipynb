## 二、整体流程：配图、选题、文案与起号

第 3、12 页把本课分为四部分：

| 环节 | 工具或方法 | 本课的具体产出 |
| --- | --- | --- |
| 1. 生成并设计配图 | 豆包 AI＋图怪兽 | 参考图生成记账本图片，再将图片、背景和文字排成封面 |
| 2. 扩写选题 | DeepSeek／豆包 | 围绕 4 个参考方向，各扩写 5 个，共 20 个选题 |
| 3. 生成文案 | DeepSeek／豆包 | 参考同类内容，为确定主题生成主标题、副标题和正文 |
| 4. 持续制作与优化 | 观察同类账号，循环制作发布 | 寻找低粉账号中的受欢迎内容，调整封面、标题与文案 |

**生图与排版承担不同任务**：豆包负责产生记账本等视觉素材，图怪兽负责把素材与可编辑的标题组合成封面。DeepSeek 的选题结果用于挑选写作方向，文案结果再把其中一个方向展开。

原课先示范视觉形式，再生成选题。实际制作某一篇时，最终主题、封面文字和正文内容需要相互对应。课程没有展示批量内页排版、图片导出和平台发布的完整操作。


## 三、用豆包参考图片生成记账本配图

### 3.1 先观察参考封面的组成

### 3.2 上传参考图，描述要生成的主体与文字

第 4 页单独列出的豆包提示词为：

~~~text
帮我生成一张一样的图片，是一个记账的笔记本，笔记本封面上写着2个字：暴富
~~~

同一页的豆包对话截图实际输入更长，结合参考图写为：

~~~text
帮我生成一张类似的图片，上面是一个记账的本子，封面写着暴富2个字，上下也写着财源广进，八方来财的成语。
~~~

两者都要求生成记账本，并把封面大字改为“暴富”；实际对话还要求上下添加“财源广进、八方来财”。复做时先放入参考图，再描述主体、文字及需要保留的外观，让 AI 有明确的画面依据。

~~~text
8种方法： 
1、能给目标用户带来巨大的价值，对他们有用； 
2、针对目标人群的需求痛点，能解决他们的痛点； 
3、能引起人们的好奇，制造悬念；
~~~

## 七、图文账号如何从 0 到 1 起号

第 11 页把起号思路归纳为四步：

| 顺序 | 原课表述 | 对应的内容工作 |
| --- | --- | --- |
| 1 | 挖掘 1–5 万低粉爆款 | 观察同类小体量账号中受欢迎的选题和表现形式 |
| 2 | AI 复刻类似爆款 | 参考主题、标题和结构，用 AI 制作自己的配图与文案 |
| 3 | 优化封面标题文案 | 调整封面的可读性、标题表达和正文内容 |
| 4 | 循环制作发布 | 持续重复制作与发布，再改进内容 |

课件把“1–5 万”与“低粉爆款”写在一起，没有进一步说明筛选范围、样本数量或互动门槛。这是寻找参考账号的方向，不能把它理解为每篇内容必须达到的点赞量或固定起号标准。

本课的“复刻”可以拆开理解：参考封面如何分配文字和图片，观察标题如何交代人群与问题，学习正文如何分段，再替换为自己确认过的主题和信息。制作时要处理好内容之间的关系，避免只换标题数字，正文却沿用另一篇。

课件没有展开发布按钮、更新频率、复盘指标和变现渠道，也没有验证“暴力起号”“百万＋”等宣传表述的实际效果。这里保留四步流程，不补出固定涨粉或收益承诺。


## 八、课程小结

本课把攒钱图文的制作拆成四项工作：

1. **豆包生成素材，图怪兽完成设计**：先参考记账本图片生图，再把主体、标题和编号排成 `3:4` 封面。
2. **DeepSeek 扩写选题**：明确 18–40 岁的目标人群，以 4 个参考方向生成 20 个候选题目。
3. **DeepSeek 生成文案**：收集同类参考，确定主题，要求输出主标题、副标题和正文。
4. **持续制作并优化**：寻找合适的同类内容，调整封面、标题和文案，循环制作发布。

复做一篇时，先选定主题，准备能够支持内容的信息，再让图片、标题与正文围绕同一个问题展开。课程示范了从参考到封面和文案的过程，成品发布前仍需检查 AI 图片文字、事实依据与文案承诺。
