# 10 AI做老照片转视频


## 资料定位与课程目标

本篇对应课件 **《10-AI做老照片转视频》**，共 11 页，首页标题为“AI 将老照片生成视频，开启副业之旅”。课程先展示应用与需求，再介绍用一键抠图修复、上色，用可灵 AI 把图片转成视频，最后用剪映组合照片、视频和音乐，制作可发布的成片。第 10 页为小结，第 11 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇按本地课件整理，读取操作截图，抽看 5 段内嵌视频，并保留原课人物动态图。

本课给出了工具入口、案例和剪辑工程，没有给出图生视频的完整提示词、模型、生成参数或详细接单流程。以下按课件可见内容记录，必要的操作整理与观察会在对应步骤说明。


## 一、什么是老照片转视频，有哪些实际应用

第 2 页先看效果与需求。“老照片转视频”在本课中包含两层处理：先让模糊、黑白的照片变得更清晰并带上颜色，再让照片中的人物产生表情或动作，形成一小段动态画面。

开场录屏展示多张人物肖像的动态效果，并切换到评论区。后面的家庭照片案例则展示夫妻微笑拥抱、长辈抱孩子时低头看孩子等变化。课程借这些画面说明：既可以做人物肖像展示，也可以把家庭合影制作成有情感表达的纪念短片。

同页两张需求截图展示了不同侧面：

| 原课材料 | 可见内容 | 在课程中的作用 |
| --- | --- | --- |
| 文章内的聊天与转账截图 | “用 AI 工具有成本的，49 元可以做一张，不同图片做出来效果会略微不同”，随后出现“先来一张试试”和 `¥49.00` 转账卡片 | 举例说明有人愿意为照片处理付费，也提示不同原图的效果会不同 |

## 二、基本流程：修复上色、图生视频、剪辑成片

第 3–5 页逐步补全制作链路，第 10 页再次归纳为三步：

| 顺序 | 工具与入口 | 输入 | 本阶段产出 |
| --- | --- | --- | --- |
| 1. 修复照片、提升清晰度并上色 | [一键抠图](https://www.yijiankoutu.com/)的照片修复功能 | 老照片、黑白照片 | 可用于后续生成的清晰彩图 |
| 2. 将照片转换成视频 | [可灵 AI](https://klingai.kuaishou.com/)的 AI 视频入口 | 处理后的照片 | 人物有表情或动作的短视频素材 |
| 3. 剪辑并发布 | 剪映 | 原照片、生成的视频、背景音乐 | 经过编排和调整的成品视频 |

这三步分别解决照片质量、人物动态和视频表达的问题。原照片用于展示“处理前”，生成视频用于展示“处理后”，剪辑把两种状态按顺序连接起来，让观众看出变化。

课件先演示“黑白图、清晰彩图、AI 转视频”的关系，再讲剪映处理。复做时也按这个顺序准备素材，避免把未经检查的照片直接作为最终输入。


## 三、用一键抠图修复老照片、提升清晰度并上色

### 3.1 从照片修复入口上传图片

第 3 页给出网址 `www.yijiankoutu.com/`，截图用箭头指向顶部 **照片修复**，左侧选中 **老照片修复**。页面标题是“在线老照片修复、老照片翻新”，说明文字写可以修复清晰度并自动上色。

按截图可整理出以下操作顺序：

1. 打开一键抠图网站，进入“照片修复”。
2. 在“老照片修复”页点击 **添加图片**，或者将图片拖到上传区域。
3. 对照原图检查处理后的清晰度与颜色，将满意的图片留作下一步的输入。

截图左侧还列出“黑白照片上色”“模糊人脸修复”，它们是可见的其他入口。原课没有逐项演示这几个功能，也没有展示上传后的处理面板、付费说明或保存按钮，不能把它们写成已经演示过的一整套参数设置。

### 3.2 对照原图与清晰彩图

同页用一张女性黑白肖像和一张彩图做对照。彩图中能看到棕色背景、肤色、唇色等，左侧还有“上官云珠”的红色竖排文字。

对照图帮助理解本阶段的目标，但课件没有展示这张结果的完整生成过程，也没有说明竖排文字在哪一步加入。复做时检查眼睛、嘴部、发际线和衣服边缘，确认处理后仍符合自己要保留的人物特征。

> 补充：AI 上色是依据图像推测颜色，不能把生成的衣服颜色、肤色和背景颜色当作原照片拍摄时的真实色彩记录。

## 七、使用剪映组合图片、视频与背景音乐

### 7.1 导入素材，建立处理前后的对照

第 8 页的 Step1 是“使用剪映，生成视频”，列出两项操作：**导入图片视频素材，添加背景音乐，生成发布视频**。

截图左侧的媒体区可见 `output.mp4`、`素材.png` 和 `夫妻照.mp4`。其中 `素材.png` 的缩略图为黑白夫妻合影，两个视频缩略图为彩色画面；截图没有说明两个视频文件之间的完整关系，不能只凭文件名认定它们如何生成。

可以按截图的组织方式复做：

1. 将原照片和下载的视频导入剪映的媒体区。
2. 在时间线上先放原照片，再接生成的视频，让观众先看到原始状态，再看到人物动起来。
3. 根据画面衔接调整照片停留时间和视频片段长度，添加背景音乐。
4. 从头播放，核对照片与视频是否属于同一组人物，画面变化和音乐是否协调。

### 7.2 从时间线观察转场、变速和音乐

原课截图中的具体设置如下：

| 项目 | 截图可见内容 | 用途 |
| --- | --- | --- |
| 片段顺序 | `素材.png` 在前，`夫妻照.mp4` 在后 | 形成原照片与动态结果的对照 |
| 转场 | 右侧名称为 **叠化**，时间线连接处有转场块 | 让黑白照片向彩色视频逐渐过渡 |
| 视频变速 | 视频片段标为 **变速 0.9X** | 将这一段素材略微放慢 |

“叠化”可以理解为前一幅画面逐渐淡出，同时后一幅画面逐渐出现。原课没有给出转场的具体秒数，也没有展示音乐音量、淡入淡出或全部片段的时长设置，因此上表是这一张工程截图的记录，不是每条作品必须照抄的参数。

- 原照片与生成视频的衔接是否自然，人物主体是否被裁掉。
- 是否存在明显的人脸、手部或衣服变化异常，片段首尾是否有多余画面。
- 音乐是否覆盖需要展示的片段，切换是否突兀。

以上是对“调整细节”的补充检查。成片约 15.37 秒、第 6 页单段案例约 5.1 秒，反映的是**素材与剪辑成片的不同阶段**，不能据此给所有作品规定统一时长。

## 九、课程小结

第 10 页的问题是“将老照片变清晰并转成视频的步骤是？”，原课归纳为：

1. 使用 **一键抠图**，将老照片修复、变清晰并上色。
2. 使用 **可灵 AI**，将图片转为视频，并下载视频。
3. 使用 **剪映**，将图片和视频素材剪辑成片，再发布。

这一页再次展示女性肖像的黑白图、清晰彩图、动态效果和剪映成片示例，回顾同一条制作链路。单段 AI 视频负责呈现动作，剪映负责安排原图、动态片段、转场与音乐。

本课的实践重点是先把照片处理好，再检查人物动态，最后通过处理前后的对照呈现效果。商业部分则提醒：有作品制作能力之后，还要解决客源问题，课件中的报价与需求截图只作为应用示例。
