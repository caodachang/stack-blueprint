# 02 AI做民国美女故事号


## 资料定位与课程目标

本篇对应网盘课件 **《02-AI做民国美女故事号》**，共 14 页。课程从民国人物故事视频案例入手，依次讲解故事脚本、人物配图、剪映图文自动成片，以及替换素材和细节优化。第 14 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。

课件开场用“快速涨粉带货变现”说明账号方向，本篇的具体实操集中在视频制作，没有展开商品选择、挂商品链接或带货运营步骤。涨粉、收益和“免费工具”的表述按课件理解，实际结果与工具额度以使用时的情况为准。


## 一、先看案例：人物故事怎样做成短视频

课件先展示两条民国女性人物故事视频。案例画面以黑白或复古人物形象为主，配有解说和醒目的黄色字幕。开场字幕分别使用“她是民国最传奇的女人”“她是民国最痴情的豪门千金”这类人物标签，先引出主角，再展开人物经历。

后续操作围绕这种形式展开：故事文案提供叙事，人物图片提供画面，剪映把文案组织成带解说、字幕和配图的视频，再通过人工调整改善成片。

下面保留案例视频在课件中的封面帧，便于对照人物画面和字幕形式。






## 二、整体流程：先自动成片，再替换素材优化

课件把制作过程分为四步：

| 步骤 | 使用工具 | 具体任务与产出 |
| --- | --- | --- |
| 1. 生成民国故事脚本 | 豆包 AI | 根据人物、受众和叙事要求，生成短视频文案 |
| 2. 生成民国故事配图 | 即梦 AI | 生成复古人物图片，变高清并下载 |
| 3. 自动剪辑成片 | 剪映的图文成片功能 | 输入已有文案，生成包含画面、解说和字幕的初版视频 |
| 4. 调整素材优化成片 | 剪映编辑界面 | 换成自己生成的图片，处理画中画、抠图、关键帧、转场、光影、音乐和字幕，最后导出 |

**自动成片后仍要替换、检查素材。** 剪映先把文案组织成视频结构，随后再把不合适的自动配图换成前一步准备的人物图片。


## 三、AI 生成民国故事脚本

### 3.1 用 ISRO 模型组织提示词

课件先介绍与 AI 对话的 `ISRO` 模型，把一条指令拆成四部分：

| 部分 | 课件英文 | 需要交代的内容 | 本课中的对应要求 |
| --- | --- | --- | --- |
| I：身份 | Identity | 给 AI 一个身份 | 人物传记赛道的民间故事博主 |
| S：技能 | Skill | 告诉 AI 它具备什么技能 | 擅长人物传记故事、短视频文案和有张力的开场白 |
| R：要求 | Require | 说明结果、数量及具体约束 | 面向喜欢近代史的成年人，简洁有力，增加悬念，剧情逻辑通顺 |
| O：输出 | Output | 说明如何展示结果 | 输出指定人物的短视频文案，约 1000 字 |

这个框架帮助 AI 同时理解“以什么身份写”“为谁写”和“最终交付什么”。课件在输出部分还补充了字数、开头和情节要求，实际使用时要把四部分合起来阅读。


### 3.2 在豆包中生成约 1000 字的人物故事

打开豆包 AI：[doubao.com/chat](https://www.doubao.com/chat/)，输入课件给出的完整提示词：

~~~text
#身份：你是一个人物传记赛道的民间故事博主，你经常制作人物传记猎奇类的爆款短视频。
#技能：你最擅长输出人物传记类故事，以及撰写相应的短视频文案，且擅长用简短富有张力的开场白吸引读者。
#要求：这些故事开头猎奇，剧情跌宕起伏且逻辑通顺，适合目标读者是“喜欢近代史的成年人”，你会用简洁有力、增加悬念、强化冲突感的方式来介绍人物，避免陈词滥调，并且善于用倒叙/悬疑/对话等方式讲述故事。
#输出：现在请你帮我输出【宋美龄】的人物传记短视频文案，字数在1000字左右，要求开头有悬念能留人，中间越跌宕起伏，越令人唏嘘的越好。
~~~

这段提示词用倒叙、悬疑、对话等方式组织人物经历，并要求开头能留住读者。豆包返回的是可继续编辑的故事文案，课件截图中的标题为《宋美龄：权力背后的传奇女性》。后续把确认后的正文送入剪映。

> 补充：人物传记涉及真实人物。AI 为了满足“悬念”“跌宕起伏”可能生成没有依据的场景、对话或心理描写，进入配音前应核实人物关系、时间和事件，不把生成文案直接当成史料。




## 四、AI 生成民国故事配图

### 4.1 在即梦中输入提示词并导入参考图

打开即梦 AI：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)，进入“图片生成”。课件截图中的生图提示词是：

~~~text
林徽因，民国美女，黑白，文静，半身照，超高清分辨率，肤质清晰
~~~

提示词确定人物、时代风格、黑白色调和半身构图，画面适合后续竖屏人物解说视频。截图还导入了一张参考图片，并显示“参考风格”。

按原课页面操作：

1. 在“图片生成”输入上述提示词。
2. 使用“导入参考图”加入参考图片，让生成结果参考所需风格。
3. 选择生图模型和图片比例。课件截图为 `图片 2.0 Pro`、精细度 `5`，比例选中 `9:16`。
4. 提交生成，查看人物图片结果。

这些是课件当时的界面设置，模型名称和可用额度以当前页面为准。

> 原课示例差异：上一页的脚本提示词使用“宋美龄”，这一页及后续剪辑使用“林徽因”。笔记按原资料保留两组示例。实际制作同一条视频时，需要统一脚本、配图和解说的人物，不能直接混用。




### 4.2 查看并选择生成结果

课件展示了同一组提示词生成的多张人物图片：主体为穿着民国服饰的女性，采用黑白或偏复古的半身构图。先从结果中挑出适合视频的图片，再进入高清处理和下载。

生成图用于故事配图。即使提示词写了真实人物的名字，也不能据此把生成结果标为该人物的真实历史照片。




### 4.3 变高清并下载原图

课件下一步强调“AI 变高清、下载原图”。操作时先选中需要的图片，使用图片工具栏中的 `HD / 超清` 功能，处理后下载原图，供剪映替换素材使用。

课件同时列出“百度去水印”入口：[image.baidu.com](https://image.baidu.com)。本页没有展开具体操作；如需处理图片，应先确认素材的使用权限。




## 五、剪映 AI 图文自动成片

### 5.1 从“图文成片”进入已有文案模式

前两步准备好脚本和配图后，打开剪映。课件演示的是电脑端界面：

1. 在首页选择“图文成片”。
2. 在新窗口左上角找到“已有文案？”下的“自由编辑文案”。
3. 进入正文编辑界面，使用已经生成的故事脚本。

先选择已有文案入口，才能继续使用豆包写好的正文。页面中其他题材的“智能写文案”选项不属于这里的操作步骤。






### 5.2 粘贴文案、选择解说并生成视频

在“自由编辑文案”页面粘贴故事正文。课件此处演示的是约 1000 字的林徽因故事，右下角显示 `997/20000`，解说音色显示“沉稳解说”。

操作顺序：

1. 将确认后的故事正文粘贴到编辑区。
2. 检查段落是否完整，去掉不需要朗读的提示词、说明文字。
3. 在右下角选择解说音色，原课示例为“沉稳解说”。
4. 点击“生成视频”，等待剪映自动成片。

`997/20000` 是截图中的文案计数显示，本课的脚本要求仍是约 1000 字，不需要写满页面额度。




### 5.3 查看自动生成的视频与时间线

生成完成后进入剪辑界面。课件截图中已经出现竖屏预览，时间线上有分段字幕、图片素材、解说音频和背景音乐。

先播放初版，确认正文已经组织成视频，再进入下一步的素材替换。此时的画面包含剪映自动匹配的素材，和自己在即梦生成的人物图片可能不一致，需要继续调整。




## 六、调整素材，优化成片

课件先回看案例，再把优化分成两组：一组处理人物素材和运动效果，另一组处理转场、光影、背景音乐和字幕。

### 6.1 替换自动匹配的素材

把前面下载的人物图片导入剪映，在时间线中找到需要调整的画面片段，将素材换成自己的 AI 配图。替换后播放对应段落，确认图片与当前讲述的人物、情节相符，画面长度能承接解说。

原课小结明确要求“将成片的素材，替换自己刚才用 AI 生成的”。因此，完成“生成视频”只得到初版，素材替换是本课后续实操的一部分。

### 6.2 插入画中画，使用抠图和关键帧

课件列出三项连续操作：

1. **插入画中画**：在主画面上叠加人物图片，时间线上会多出一层画面素材。
2. **对画中画使用抠图效果**：去掉叠加图片不需要的背景，保留人物，使其与底层画面组合。
3. **给画中画添加关键帧效果**：在不同时间点记录画面状态，让静态素材产生位置或大小的变化。

“画中画”是叠加在主画面上的另一层画面，“关键帧”用来记录某个时间点的属性状态。课件截图可以看到多层图片轨道以及素材上的变化标记，体现了在自动成片基础上继续编辑的过程。

本页没有给出固定的抠图参数、关键帧数值和运动时长，跟做时按当前素材预览调整。




### 6.3 添加转场和画面光影

接下来处理画面之间的衔接：

- **画面转场**：在相邻片段之间添加切换效果，让图片衔接更自然。
- **画面光影**：给画面叠加光影效果，增加静态人物图的变化。

原课时间线截图中可见片段连接处的转场标记，以及上方的“发光线条光影”效果片段。它们都是局部画面处理，不需要重新生成故事文案。


### 6.4 添加背景音乐，调整字幕并导出

课件最后列出“添加背景音乐、调整字幕导出”。自动成片已经生成部分音频和字幕，最终要结合当前画面继续检查、调整：

1. 添加或检查背景音乐，试听它与故事解说的搭配。
2. 对照解说调整字幕，检查人物姓名、文字与画面的对应关系。原课预览使用醒目的黄色字幕。
3. 从头播放调整后的片段，确认替换图片、画中画、转场和光影的位置合适。
4. 使用剪映的导出功能生成最终视频。

原课时间线中的背景音乐名为 `Last Reunion`，这里只作为截图内容记录。课件没有规定固定音量、字幕字号、导出码率或分辨率，不把这些参数补写成原课要求。




## 七、课程小结

原课把民国故事账号的视频制作归纳为四步：

1. 使用 **豆包 AI** 生成民国故事脚本。
2. 使用 **即梦 AI** 生成民国故事配图。
3. 使用 **剪映 AI 图文成片**，把文案生成初版视频。
4. 将自动成片中的素材替换为自己生成的图片，优化后导出。

这篇的主要产出是一条人物故事视频。脚本决定讲什么，配图决定人物画面，图文成片完成初步组织，最后的素材替换和细节调整让画面与故事对应起来。
