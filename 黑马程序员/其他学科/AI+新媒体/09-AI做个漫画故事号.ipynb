## 资料定位与课程目标

本篇对应课件 **《09-AI做个漫画故事号》**，共 16 页。课程以民间故事类短视频为例，演示用 DeepSeek 生成文案和分镜提示词、用即梦生成古风背景图，再用剪映“图文成片”生成配音与视频，替换素材并添加人物、动画和字幕。第 14 页补充故事内容与带货产品的衔接，第 15 页为小结，第 16 页没有新增教学内容。

课件开场的“有门槛、有深度、能变现”和流程图中的“免费生成”属于课程表述。资料没有提供收益验证，工具当前的模型、额度与收费需要按实际使用时的说明确认。


## 一、先看案例：背景、人物、旁白与字幕分层组合

第 2 页展示两条民间故事视频，第 13 页再次展示同样的两条案例。抽看的画面可以看到：

| 案例 | 可见内容 | 制作上可以观察的结构 |
| --- | --- | --- |
| 范蠡相关故事 | 古代室内、城门、庭院等背景，叠加古装人物与金元宝，顶部有大字字幕 | 用背景交代地点，人物与道具承载故事动作，字幕随讲述推进 |
| 人生智慧类故事 | 老先生、猎户等人物，庭院、山崖、村落等场景，出现“听懂这几个小故事”“成为利益共同体”等文字 | 将多个故事片段组合成一个主题，切换人物和背景表现不同情节 |

这些案例的“漫画”主要表现为**场景图＋人物素材＋动画＋解说字幕**。本课没有演示先生成整段人物动画，而是先制作背景，再在剪辑阶段叠加人物与变化。

下面保留案例抽帧，供观察人物与背景的组合方式。案例中的点赞、商品销量等界面数字不能据此推算自己制作后的效果，本篇也未逐句转录案例音频。






## 二、整体流程：先做故事，再做背景和剪辑

第 3、15 页将本课分为四个环节：

| 顺序 | 工具 | 操作与产出 |
| --- | --- | --- |
| 1. 生成故事与分镜 | DeepSeek／豆包 | 参考文案写新故事，再将选定故事拆成分镜与生图提示词 |
| 2. 生成故事配图 | DeepSeek＋即梦 AI | 将分镜提示词放入即梦，生成与文案场景对应的古风背景 |
| 3. 自动生成视频初稿 | 剪映“图文成片” | 输入已有文案，选择解说音色，生成配音、字幕与画面初稿 |
| 4. 替换素材并优化 | 剪映 | 用准备的图片替换画面，叠加人物和贴纸，调整动画、音乐、字幕后导出 |

这里需要区分两类素材：**故事文案用于朗读和字幕，绘画提示词用于生成场景图**。把生图提示词直接放进朗读区，会把镜头说明也读出来。

课件第 6 页要求背景“只需古代的场景，不需要人物”，第 11 页再添加人物素材。这样背景与人物可以分层调整，后续分别安排人物动画和背景运动。


## 三、用 DeepSeek 生成民间故事文案

### 3.1 原课完整故事提示词

第 4 页的指令先定义民间故事博主的身份，再要求 AI 分析 5 篇参考文案，仿写 3 篇。完整原文如下：

~~~text
#身份：我是一个民间故事赛道的短视频博主，我需要制作猎奇民间故事的爆款短视频。
#需求：下面我会给你5条爆款的民间故事文案，这些文案都十分吸引读者，我需要你分析这5篇文案的风格及特征，然后模仿写出3篇同样风格的文案，并且将现有5篇文章的核心观点和风格表达，运用在新生成的3篇文案中。
#要求：
有新主题：参考5篇原文后，请你给我写3个新的，吸引人的文章标题
文章结构：需要你识别5篇原文的核心论点和支撑论据，并且在模仿文案中，也严格按照这个文章结构来撰写
风格模仿：严格模仿这5篇参考文案的风格、语气，有观点，有故事，有转折
剧情起伏：撰写的新故事文案，剧情需要起伏，有逻辑、有寓意、有深度，且能够吸引自媒体用户
字数时长：生成的3篇新文案，每一篇时长在3-5分钟，每一篇文案字数约在1000-1200字左右。
价值观正确：故事不能迷信、不能讲风水玄学，要符合主流的核心价值观。
5篇参考文案，在文件中。
~~~

这条提示词同时约束了主题、结构、语气、转折和篇幅：每篇约 **1000–1200 字**，目标时长 **3–5 分钟**，并明确要求“不迷信、不讲风水玄学”。时长是创作目标，最终仍取决于朗读速度、停顿和剪辑，不能把字数直接当成成片时长。

提示词最后写“5 篇参考文案，在文件中”，但所给目录没有找到这份文件，课件内部也没有嵌入对应文档。复做这一步需要自行准备要分析的参考文案；仅粘贴这段指令，AI 并没有读到课件所说的 5 篇原文。


### 3.2 原课生成结果二：出嫁的女儿与娘家灯火

第 5 页展示的是“新文案 2”和“新文案 3”。第一篇故事全文没有出现在本页，不能将后面两个故事重新编号为第一篇。以下保留第二篇可见全文，标点按阅读整理：

~~~text
新文案2：为什么出嫁的女儿不能看娘家的灯火？一场深夜的诡异遭遇揭晓答案

“嫁出去的女儿，除夕夜不能看娘家的灯！”这规矩听起来不近人情，但背后的因果却让人脊背发凉。

故事一：宋朝时，富商刘家嫁女，女儿玉娘思念父母，除夕夜偷偷跑回娘家。路过村口时，她见自家宅院灯火通明，窗上映出父母身影，心中一暖。谁知推门而入，屋内竟空无一人，灯火瞬间熄灭，黑暗中传来凄厉哭声：“你看了不该看的灯，带走了刘家的运数！”次日刘家突发大火，百年基业毁于一旦，玉娘从此疯癫，逢人便喊：“灯是假的！人也是假的！”

故事二：民国时期，寡妇秀娥因婆家苛待，除夕夜逃回娘家。弟弟见她可怜，破例让她进门。半夜秀娥起身喝水，见父母房中亮着灯，便凑近窗缝偷看——只见父母竟在烧纸人，纸人脸上赫然画着她的生辰八字！母亲泣道：“嫁出去的女儿是外人，看了娘家的灯，就得替全家挡灾……”秀娥连夜逃回婆家，三日后暴毙，娘家人却安然度过饥荒。

故事三：九十年代，山村姑娘小芳远嫁他乡，因思念父母，除夕夜偷偷用望远镜眺望娘家。镜头中，娘家灯火璀璨，她却突然看见另一个“自己”正坐在饭桌前吃饭！当晚小芳高烧昏迷，梦里有个白发老妪冷笑：“活人看灯，死人替身。你这一眼，换了你的阳寿！”半年后小芳病逝，娘家人却意外中了彩票，从此富甲一方。

寓意：古时认为，女儿出嫁便是“外人”，除夕灯火象征家族气运，若女儿窥视，便会吸走娘家福分，自身也要背负因果。如今虽提倡破除迷信，但许多老人仍严守此规，只因亲情背后牵扯着难以言说的利益与牺牲。所谓禁忌，不过是人性在命运面前的挣扎与妥协。
~~~

从结构看，这一篇用疑问标题和一条禁忌引出悬念，随后分成三个年代不同的小故事，最后用“寓意”收束。这能帮助观察“标题、引入、故事、结尾”的组织方式。

需要同时看到生成偏差：正文将家族气运、替身、阳寿等设定写成情节因果，末尾的说法也没有得到史料支持。它没有满足前面“不迷信”的要求，只能作为原课 AI 生成故事记录，不能作为历史或民俗事实引用。




### 3.3 原课生成结果三：门前不栽桑，屋后不种柳

同页的另一张截图展示第三篇全文，末尾另有未完全展开的风格分析。可见故事原文如下，标点按阅读整理：

~~~text
新文案3：“门前不栽桑，屋后不种柳”真的是迷信吗？百年老宅的诅咒之谜

“前不栽桑，后不栽柳，院中不种鬼拍手！”这句俗语绝非空穴来风，一座百年凶宅的往事，揭开了树木与风水之间的生死关联。

故事一：明朝嘉靖年间，县令李茂才为显风雅，在宅前种了一排桑树。不料三年内，长子坠马、次子溺亡，夫人也突发疯癫症，整日对着桑树念叨：“丧门星，吊死鬼……”李茂才请道士做法，道士怒斥：“桑谐音‘丧’，引阴魂聚宅，速速砍树！”当夜李家仆役砍树，斧头竟溅出黑血，树根下挖出七具无棺尸骨。次年李茂才因贪腐被斩，宅子沦为鬼宅。

故事二：清末，山西盐商赵老爷为防贼，在屋后种了一圈柳树。柳枝繁茂后，家中怪事频发：账房先生莫名吊死在柳树下；小妾半夜见窗外人影晃动，开窗却只有柳条摇曳。风水先生叹道：“柳树招孤魂，枝条如缰绳，这是要锁活人魂魄啊！”赵老爷不信邪，结果不出半年，库房失火，万贯家财化为灰烬，他也在柳林中自缢身亡。

故事三：上世纪六十年代，村长王大力带头破除迷信，偏偏在院中种了棵杨树（即“鬼拍手”）。夏夜风起，树叶哗哗作响，如千百人拍手。某夜，王大力听见院中有人低语，提灯查看却见树影中站着数十个面色惨白的“人”，齐声冷笑：“树是我们坟头的碑，你吵得我们睡不着……”次日王大力暴毙，村民砍树时发现，树下竟是一处乱葬岗。

寓意：桑柳杨树自古被视为“阴木”，其名、形、声皆暗合灾祸，古人避之如蛇蝎。看似迷信的规矩，实则是通过自然征兆规避风险。如今住楼房者虽无此虑，但若回乡修宅，仍需谨记——风水不是玄学，而是人与自然博弈千年的生存智慧。
~~~

截图中的风格分析明确写到“标题：以疑问句制造悬念，结合民俗禁忌，激发猎奇心理”；结构部分可见“开篇抛出核心论点，用 2–3 个故事佐证，每个故事包含诡异细节、反转结局及因果逻辑”，后半句被输入框遮挡。

这一篇同样包含鬼魂、诅咒和风水因果，与第 4 页的创作要求不一致。结尾把相关说法解释成“生存智慧”，也不能替代事实依据。学习时可以观察叙事结构；复做时还需要审查内容是否满足自己已经写明的要求。




## 四、把选定故事拆成分镜和生图提示词

### 4.1 原课完整分镜提示词

第 6 页继续在同一对话中，要求 AI 根据**第一个民间故事**生成分镜：

~~~text
文案创作得不错，请将第一个民间故事文案，帮我生成如下内容：
1、生成36个分镜头脚本
2、每个分镜头的AI绘画提示词
3、分镜头只需古代的场景，不需要人物

分镜头用表格+序号表示，每个分镜头的提示词需要包含：
固定的绘画风格（古风水墨画风格）、主体和主体描述、镜头语言、景别、环境描述、光影氛围等；
需确保我可以在即梦AI使用这些提示词生成完整图片；
每个分镜提示词字数50字左右，中文。
~~~

课件旁边写“3 分钟视频约＝36 个分镜头”。按 180 秒除以 36 个镜头计算，平均约 **5 秒一个镜头**，这是安排素材数量时的估算。实际镜头应跟随句子和情节调整，课件没有要求每个片段固定 5 秒。

这一轮输出有两个用途：分镜描述用于核对“当前要表现什么”，绘画提示词用于告诉即梦“画面里要有什么”。提示词要求固定古风水墨画风格，并描述主体、镜头、景别、环境与光影。

这里的“第一个民间故事”对应课件未展示全文的第一篇，与上面转录的第二、第三篇不同。复做时要让 AI 明确正在拆分哪一篇故事；若另开对话，先提供自己选定的完整正文，再发送分镜要求。

## 五、用即梦生成与文案对应的古风背景

### 5.1 输入场景提示词并设置图片参数

第 7 页强调“画面需要匹配文案描述场景”。打开即梦：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)，进入“图片生成”。原课输入框中的实际提示词为：

~~~text
古风水墨画风格，主体：半开木门（门环锈蚀）；镜头：中景平视；环境：门缝透烛光，台阶青苔；光影：冷雨斜飘，门内暖黄光晕。
~~~

| 设置 | 原课截图可见值 |
| --- | --- |
| 生图模型 | `图片 2.0 Pro` |
| 精细度 | `7` |
| 图片比例 | `3:4` |
| 尺寸栏 | `768×1024` |
| 参考图 | 可见“导入参考图”入口，未显示已经上传参考图 |

输入区还有一个 `Deepseek-R1` 按钮，但截图明确标出的生图模型是“图片 2.0 Pro”，两者不能混作同一个模型设置。

### 5.2 检查候选图，选择并下载背景

第 7 页返回四张古代院门候选图，门的开合、屋檐、灯笼和暖光位置不同，整体都表现雨中建筑与石阶。下方还有一条带“超清”标记的结果。

选图时先对照当前文案：场景是不是需要院门，门缝的光是否符合叙事，画面中是否出现不需要的人物或现代物品。这里使用的是场景背景，后面还要留出叠加人物的位置。

课件展示了“超清”结果，但没有逐步演示高清处理与下载按钮。第 8 页已经进入下载后的图片处理，不能据此补写固定下载分辨率或免费额度。

原课虽然写“古风水墨画风格”，生成图实际保留较丰富的建筑细节和颜色，并非纯黑白水墨。需要通过图片结果判断风格，不能只看提示词中的名称。




### 5.3 原课下载后的图片处理

第 8 页写“下载后去水印”，给出百度图片入口：[image.baidu.com](https://image.baidu.com/)。左图展示“选择编辑方式”，其中选中 **AI 去水印**；右图显示上传的雨中古街图片，界面提示“正在消除…16%”。

这页记录的是处理入口与运行状态，没有展示完成后的结果或保存操作。右图处理对象是另一张古街背景，也不是前页展示的半开木门候选图，不能把两张图当作同一个任务前后的对比。

> 补充：只处理自己有权编辑的素材，并按平台允许的导出方式使用图片。平台要求保留的 AI 标识应继续保留。画面清理与生成内容标识需要分别确认，不能因课件写了“去水印”就认定所有标识都可以移除。






## 六、用剪映“图文成片”生成视频初稿

### 6.1 进入图文成片，选择已有文案

第 9 页在剪映首页用红框标出 **图文成片**。进入后，左上角“已有文案？”下面有 **自由编辑文案**，原课用箭头指向这个入口。

本课前面已经用 DeepSeek 写好故事，因此接下来选择“自由编辑文案”，将确定的故事放入编辑区。页面其他位置还有情感关系、励志鸡汤、美食教程等智能写文案分类，本课没有沿这些分类重新生成故事。

可以按下面的顺序操作：打开图文成片，进入自由编辑文案，粘贴准备朗读的正文，再检查是否带入了提示词、分镜表和不应朗读的说明。






### 6.2 选择解说音色，生成视频

第 10 页展示自由编辑文案界面，右下角选中 **沉稳解说**，旁边是 **生成视频**。可见的输入计数为 `997/20000`，它反映当前示例的输入状态，不代表课程要求每篇都写 997 字。

操作重点是：先确认朗读内容，再选择和试听音色，最后生成视频。生成后还要检查断句、漏字和画面是否匹配。

**本页截图粘贴的是林徽因相关人物故事，开头为“故事要从一个迷雾重重的夜晚说起”**，与前面“北宋富商施恩”的古代民间故事不同。截图可以用来说明按钮和生成步骤，但不是本课前面文案已经完整输入的证明。实际复做时，应替换为自己最终选定的故事。




### 6.3 自动成片后，检查字幕、配音与画面

第 10 页右侧展示生成后的剪映工程，时间线已有多层内容：

| 图层或轨道 | 原课可见内容 | 后续需要检查什么 |
| --- | --- | --- |
| 字幕 | 上方连续的文字片段，与句子对应 | 是否有错字、断句不顺或不该朗读的说明 |
| 图片 | 黑白人物、书法、建筑等画面 | 是否与当前故事和当前一句相符 |
| 解说配音 | 按句子分开的声音片段与波形 | 台词是否读全，语速是否合适 |
| 背景音乐 | 下方长音轨，片段名 `Last Reunion` | 与解说是否协调，音量是否影响听清正文 |

这一页仍然是人物故事示例，预览为 `9:16`。自动配图不一定符合自己准备的画风和场景，后面需要用即梦生成的背景替换，并重新检查图文对应关系。

因此，“图文成片”在这里提供的是可继续编辑的视频初稿。不能把自动生成视为已经完成古风漫画故事的全部制作。




## 七、替换背景，添加人物、贴纸与动画

### 7.1 将自动配图替换为准备好的古风场景

第 11 页回到古代民间故事工程。素材区能看到雨中古街和半开院门，预览画面上方写“施恩不图报，图报莫施恩”，下方叠加古装老者与挑担人物；预览比例为 **`3:4`**。

这一页的细节优化列出：

1. 调整画面，添加贴纸动效。
2. 添加人物，增加人物动画。
3. 给背景图片设置关键帧动画。

复做时先将自己的背景图片导入剪映，再按文案场景替换时间线上的画面，调整每张背景的持续区间与构图，保留人物活动和字幕的位置。背景从古街切换到院门时，也要检查解说是否已经讲到相应地点。

原课没有逐个展示“替换”的按钮或拖拽方式，也没有提供所有人物素材的来源。截图中后面仍能看到其他画风的图片，所以不能将本页视为全部 36 个镜头已经统一完成。




### 7.2 让人物与背景分别产生变化

从第 11 页时间线可以看到，人物和贴纸位于背景图片上方，人物还以“复合片段”的形式出现。可以把各层理解为：

| 元素 | 在本课中的作用 |
| --- | --- |
| 背景图 | 交代街巷、宅院等地点与氛围 |
| 人物素材 | 表现老者、挑担人等角色，配合叙事出现或离开 |
| 贴纸动效 | 为局部动作或强调内容增加变化 |
| 背景关键帧 | 让背景的位置、缩放等属性随时间变化，形成简单运镜 |

关键帧可以理解为先在一个时间点记录画面状态，再在另一个时间点改变状态，让中间产生连续变化。本课只提出这个处理方向，没有展示具体关键帧数量、缩放数值、运动曲线和固定动画名称。

> 补充检查：按当前句子的内容安排人物出场时间，避免同一人物在不相关场景里一直停留。背景运动时观察是否露出边缘、人物是否被裁掉，字幕也需要避开关键动作。




## 八、调整背景音乐、字幕并导出

第 12 页继续写“细节优化”，列出“调整背景音乐（片片相思赋予谁）”“去除相关水印（AI 平台水印）”“更换背景音乐、调整字幕导出”。

这里需要对照文字与截图：文字推荐的曲名是 **`片片相思赋予谁`**，但截图下方长音轨仍标为 **`Last Reunion`**。不能据此认定推荐音乐已经替换完成。选择音乐后，需要从头播放，确认它不会盖过解说。

这一页截图再次使用林徽因相关人物故事，预览为 `9:16`，右侧打开字幕列表，画面中能看到黄色文字配深色描边。它展示的是调整字幕与声音的方向，不能把该人物画面记作前面的古代漫画故事最终成片。

本课的比例也要按环节分别记录：即梦场景图和第 11 页古代故事工程是 `3:4`，第 10、12 页人物故事工程是 `9:16`。复做同一条视频时，先确定最终画幅，再检查图片裁切、人物位置和字幕，不能直接混用各截图的设置。

课件没有给出字体字号、字幕位置数值、音乐音量、导出分辨率、帧率和码率，也没有展示水印处理的最终结果。收尾时核对字幕与配音、图文对应和音量，再按素材及发布要求导出，保留平台要求展示的 AI 标识。




## 九、对照案例，理解故事内容如何衔接带货

第 13 页“看案例剪辑”再次播放第 2 页的两条对标案例，使用同样的内嵌视频。它没有另外提供一条明确对应本课全部制作步骤的完整成片，因此不能把这些案例直接写成老师刚刚做好的成果。

第 14 页接着问“怎么出单带货？”，明确给出两点：

1. **短视频中有销售、介绍产品的内容。**
2. **带货产品与短视频强相关。**

从这一页两段案例录屏的抽帧中，可以观察到：

| 案例 | 内容与产品的连接 |
| --- | --- |
| 人生感悟故事 | 前半段用人物与故事讲观点，后面切到实体书内页，字幕写“有意义的小故事”，界面有“视频同款”商品卡 |
| 范蠡相关故事 | 先讲范蠡和处世认知，后面出现“是你读过的书”“认知”等字幕，界面商品卡写“范蠡传” |

这一页强调的是内容衔接：先让故事主题与书的内容建立联系，再在视频中实际介绍商品，给观众理解“为什么是这件商品”的依据。商品卡只是承接入口，课程并没有展示挂卡后必然成交的验证。

课件没有展开开通带货权限、选品佣金、商品上架、挂载链接、下单流程或转化数据。下面保留书页与商品卡、故事人物与相关商品卡的画面，用来对照原课这两条原则。






## 十、课程小结

第 15 页把本课方法归纳为四步：

1. 使用 **DeepSeek** 生成故事文案，再生成分镜与 AI 绘画提示词。
2. 使用 **即梦 AI** 生成与故事场景相符的背景图片。
3. 使用 **剪映图文成片**，将已有文案转换为可编辑的视频初稿。
4. **替换素材并细化画面**，补充人物、贴纸、动画、音乐与字幕，完成导出。

本课可以复用的制作方法是：先明确故事，再把场景背景和人物分开准备，利用自动成片建立解说与字幕，最后逐段调整画面。带货部分则要求视频里实际介绍商品，并让商品与故事主题相符。

复做时仍需核对 AI 故事是否满足原先的创作要求，补齐自己的参考资料与完整分镜，并统一同一条视频的画幅、人物、背景和声音。
