# 15 AI做古人跳养生操视频


## 资料定位与课程目标

本篇对应课件 **《15-AI做古人跳养生操视频》**，共 12 页。第 2 页展示案例，第 3 页给出四步流程，第 4 页获取参考视频，第 5—7 页生成人物并处理图片，第 8—9 页演示动作模仿，第 10 页剪辑，第 11 页小结，第 12 页没有新增教学内容。

笔记以本地 PPTX 的正文、操作截图和 **4 段内嵌视频**为准：2 段案例录屏、1 段真人健身操参考视频、1 段林黛玉动作模仿结果。

这节课用真人视频提供动作，用古风人物图片提供形象与场景，让静态人物按参考视频的节奏抬手、下蹲、摆动，再添加标题和音乐。总览把这一步写成“AI换脸”，实际操作使用的是即梦 **“数字人”里的“动作模仿”**，输出包含整个角色的动作变化。


## 一、先看古人健身视频的表现形式

第 2 页给出两种“消气操”案例，主要看人物、背景、动作与标题怎样组合。

| 案例 | 录屏中的画面 | 可观察的做法 |
| --- | --- | --- |
| 单人古风消气操 | 账号显示“人民网”，灰袍古人在水墨山水、松树和石栏前跳操，底部有“每天一遍 快乐无限”文字 | 用一个全身角色占据画面主体，通过连续动作形成节奏 |
| 多人古风消气操 | 账号显示“养生课堂”，三位古人在院落中跳操，上方是红色《消气操》标题 | 主角居中、配角分列两侧，标题与字幕交代主题 |





两段内嵌录屏分别约 **15.03 秒、25.23 秒**，这是课件截取的录屏长度。案例二后半段还打开了评论区，因此不能把整段录屏长度当成原作品时长。课件后面的实操演示使用一个林黛玉角色，没有展开三人同框的制作方法。

案例中的点赞数和课程标题中的“涨粉”属于当时的展示信息，课件没有给出复做后的效果保证。


## 二、四步制作流程与素材关系

第 3 页的顺序是先准备动作，再准备人物，然后生成视频并剪辑。

| 阶段 | 课件工具 | 当前要完成的事 | 下一步使用的素材 |
| --- | --- | --- | --- |
| 1. 获取参考视频 | 短视频平台、轻抖小程序 | 找到简单易学的真人健身操，复制链接并下载 | 本地动作参考视频 |
| 2. 生成人物 | 即梦 AI、百度 AI 图片助手 | 生成古代人物全身照，做超清与图片处理后下载 | 清楚的全身人物图 |
| 3. 动作模仿 | 即梦 AI | 上传人物图和参考视频，让新角色模仿动作 | 古人跳操视频 |
| 4. 剪辑成片 | 剪映 | 导入视频，添加字幕，处理水印，更换背景音乐并导出 | 完整视频 |

人物图片和参考视频在第三步汇合：**图片决定角色长相与画面场景，参考视频提供动作顺序与节奏**。因此前两步都要准备好本地文件，才能继续生成。

总览页标注了“AI免费生成”“无水印下载”，但没有说明账号额度与收费条件，操作时以工具实际界面为准。


## 三、获取一段简单的健身操参考视频

### 3.1 复制链接，再用轻抖下载

第 4 页给出的工具是 **轻抖小程序**。两张手机截图分别显示原视频已经复制链接，以及轻抖首页的“去水印”入口。

1. 在短视频平台找到准备参考的健身操，复制视频链接。
2. 打开轻抖小程序，进入“去水印”功能，按界面提示使用复制的链接解析、下载。
3. 把下载的视频保存为后面能上传的本地文件，先播放确认动作和时长。





课件没有展开链接粘贴框、解析结果和保存弹窗，因此上面记录的是操作衔接，没有额外的下载参数。参考素材与配乐应使用本人或获准使用的内容。

### 3.2 本课选用的参考动作

课件内嵌的是一位穿白衣的人在圆形砖门前跳操，画面顶部写着“见笑开心养生操”“每天一遍 烦恼再见”。视频约 **19.04 秒、720 × 1280、30 帧/秒**，人物全身可见，包含双臂展开、下蹲扶腰、左右摆臂和小步跳动等变化。

这类素材把动作集中在一个清楚的全身角色上，便于后面检查动作是否迁移成功。选视频时先看手脚是否完整、动作是否连续，再确认它适合自己的角色构图。


## 四、用即梦生成古代人物全身照

### 4.1 完整人物提示词

第 5 页正文给出：

```text
林黛玉全身照，正面看向镜头，古风水墨画风格。
```

这条提示词包含人物身份“林黛玉”、构图“全身照”、朝向“正面看向镜头”和“古风水墨画”风格。后续要迁移全身动作，所以选图时要检查头部、身体和裙摆是否完整，手臂是否容易辨认。



截图中的实际输入比正文略短：

```text
林黛玉全身照，看向镜头，古风水墨画风格
```

它没有正文里的“正面”二字。四张结果的构图也不同，前两张更接近半身或局部人物，不能只因输入了“全身照”就认定每张都符合要求。第 6 页选中放大的是 **第 4 张青绿色衣服、白色裙摆、古建筑背景的人物**，后面的动作模仿沿用这张图。


### 4.2 生图模型、比例与尺寸

第 5 页的参数截图显示：

| 设置 | 课件中的选择 |
| --- | --- |
| 生图模型 | 图片 2.0 Pro，`by Seedream 2.0` |
| 图片比例 | `9:16` |
| 图片尺寸 | `576 × 1024` |



输入框下方还显示 `Deepseek-R1` 按钮，实际“生图模型”栏选择的是“图片 2.0 Pro”。这些是课件当时使用的界面和设置，人物图采用竖幅，后面的动作模仿截图也提示竖屏输出。


### 4.3 超清处理与下载

第 6 页的任务是“AI变超清、下载”。先打开选中的人物图，在编辑工具里选择 **“HD 超清”**，再下载处理后的图片。





详情页右侧还显示“生成视频”“局部重绘”“消除笔”等入口，本节只用到超清与下载。课件没有给出超清后的具体像素尺寸或倍率，不能把前面的 `576 × 1024` 直接写成处理后的输出尺寸。

下载前再次看清脸、双手和衣服边缘。超清处理用于改善图片细节，人物是否为完整全身仍要靠选图与构图检查。


## 五、按课件处理人物图片水印

第 7 页给出的工具地址是 [百度图片](https://image.baidu.com/)，截图已经进入 **“百度AI图片助手”** 的编辑工作区，并载入刚才的人物图。编辑方式的局部截图用红框标出“AI去水印”。





按本页任务，衔接顺序是打开图片编辑器、载入下载的人物图、选择“AI去水印”处理，再保存用于动作模仿的图片。课件没有展示从百度图片首页进入编辑器的具体按钮，也没有展开去水印的选区、参数和处理完成提示。

整张工作区截图中，“变清晰”仍显示为选中状态，红框局部图才指出“AI去水印”入口，二者是不同的功能。下一页人物预览右下角仍可见“即梦AI”标记，所以材料不足以证明上传前的图片已经完成去水印。复做时应检查自己保存的实际文件，而不是仅凭步骤标题判断处理成功。


## 六、即梦动作模仿：人物图与动作视频一起输入

### 6.1 进入数字人，上传两个素材

第 8 页显示的路径是 **“数字人” → “动作模仿”**。界面上方另有“对口型”标签，本课选中的是“动作模仿”。

1. 在“人物”区域上传处理后的人物全身图，预览确认是青绿色衣服的林黛玉形象。
2. 在“选择动作”区域选择 **“上传本地文件”**。
3. 上传前面下载的真人健身操视频，预览确认是白衣人物与圆形砖门，播放器显示约 `00:19`。
4. 检查两个素材是否对应，再提交动作模仿生成。课件截图没有截到具体生成按钮的位置。



界面也提供“预设模版”，本课使用的是上传本地视频。底部提示 **“仅支持输出竖屏效果”**，但没有展示视频模型、输出分辨率或其他展开参数。

这一步的动作来自参考视频，不需要把“抬手、下蹲”等动作重新逐条写成提示词。本课也没有要求输入旁白音频或做新台词对口型。


### 6.2 查看生成结果与动作对应关系

同页的结果记录标注“参考人物”“动作模版”“动作模仿”。预览中的林黛玉已经开始摆动身体，背景仍是人物图里的古建筑、水墨树木与左侧题字。



第 9 页另内嵌完整生成片段，约 **19.04 秒、540 × 960**。参考视频也是约 19.04 秒，说明这次示例保留了参考动作的时长。这里记录的是课件文件的实测信息，界面没有给出一组可通用于所有任务的时长或分辨率规则。

下面两帧均截取约 8 秒处：左边是真人参考，右边是生成结果。可以看到下蹲、手臂向两侧摆动的姿态相互对应，而人物外观与场景改成了输入图片的样子。



生成结果没有沿用参考视频顶部的两行标题，后面要在剪映中另加。检查视频时重点回看手指、宽袖、裙摆和脚部：动作幅度变化时，局部边缘可能变形，脚部也可能显得悬浮。确认可用后保存视频，课件没有展开视频下载面板。


## 七、剪映中添加标题与背景音乐

### 7.1 导入视频，添加两行标题

第 10 页列出五个任务：**导入视频、添加字幕、去除水印、更换背景音乐、导出视频**。截图把动作模仿结果放在主视频轨道，上方添加一条文本。

课件实际输入的两行文字是：

```text
早起开心健身操
动作简单又快乐
```



按截图复做时，先导入生成的视频，再新增文本、输入两行标题，放到人物头部上方的留白区域，调整文字持续时间，让它覆盖需要显示标题的片段。

右侧“文本—基础”面板中，字体为 **“古典体”**，颜色为黄色，字间距为 `0`，行间距为 `6`，文本缩放显示 `103%`。预览中的标题带黑色描边，但截图没有展开描边参数，不能据此补出具体宽度。

这里展示的是覆盖在视频上的固定标题，没有展示语音识别字幕、逐句口播或动作名称字幕的制作过程。


### 7.2 更换音乐，并把结尾裁到视频范围

截图主视频文件名为 `2025-04-16_16_39_06.mp4`，片段标记 `00:00:19:01`。它对应约 19 秒的古人跳操片段，文本层也基本覆盖这段画面。视频轨道的扬声器图标显示静音，下方另放了一条《百花香》的 DJ 版本音乐。



需要注意的是，播放器中的工程总时间是 **`00:00:55:02`**。从时间线看，背景音乐明显长于视频，已经延伸到 50 多秒；因此不能直接把工程总长写成“古人视频长 55 秒”。

如果这条成片只保留课件里的一个跳操片段，就把音乐裁到视频结尾，并检查标题结束位置。若要延长画面，需先补足或剪接视频，再重新确认音乐和动作的衔接，不能让长音轨单独留在视频结束后。

更换音乐后完整试听一次，确认没有残留不需要的原声，开头与结尾自然。音乐曲目是截图示例，课件没有规定必须使用这一首。


### 7.3 水印处理与导出检查

本节任务中再次列出“去除水印”，但没有展示视频去水印的具体功能、选区或参数。剪映预览右下角仍能看到“即梦AI”标记，不能把这张工程截图当成已经完成水印处理的成片。

按课件完成字幕和音乐后，导出前重点确认：

1. **画面范围**：人物头部、手臂和裙摆在动作过程中没有被意外裁掉，标题不挡脸。
2. **动作连续性**：看完抬手、下蹲、跳动等段落，检查手脚和衣服是否出现明显异常。
3. **轨道长度**：字幕、视频与音乐结束位置符合实际成片范围，尤其避免 19 秒画面配上 55 秒音乐。
4. **输出文件**：导出后重新播放，确认有完整画面、音乐和标题，没有多余的无画面尾段。

课件没有展示最终导出面板，不能把生图的 `576 × 1024`、参考视频的 `720 × 1280` 或生成片段的 `540 × 960` 直接认作最终导出设置。


## 八、课程小结

第 11 页回顾同一条四步流程：**获取简单易学的健身操视频 → 生成人物全身图并下载 → 即梦动作模仿 → 剪映添加字幕、音乐并导出**。

复做时先得到两个输入文件：清楚的古人全身图、动作连续的真人视频。生成后检查动作与人物是否稳定，再完成标题和音乐的剪接。本课的关键衔接是参考视频提供动作、图片提供角色与场景，以及剪辑阶段让各条轨道覆盖正确的成片范围。

第 12 页没有新增教学内容。
