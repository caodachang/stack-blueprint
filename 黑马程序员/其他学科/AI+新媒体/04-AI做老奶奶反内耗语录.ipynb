# 04 AI做老奶奶反内耗语录


## 资料定位与课程目标

本篇对应课件 **《04-AI做老奶奶反内耗语录》**，共 14 页。课程先展示老奶奶漫画账号，再演示语录批量生成、角色配图、模板排版、小红书图文发布，最后补充蒲公英博主合作的开通入口。第 14 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇使用本地下载的同名课件整理。

开场的“快速涨粉变现搞钱”是课程的账号方向，实际教学以图文制作和合作功能开通为主，没有展示真实广告订单或收益验证。课件所说的“免费 AI”，以工具使用时的额度与收费规则为准。


## 一、先看案例：老奶奶角色与反内耗语录

原课展示了两组老奶奶漫画账号：一组以粉色中式服装、眼镜和夸张手势表现幽默回怼，另一组使用蓝色背景、紫色服装和较温和的表情，围绕情感话题表达观点。

两组案例都把漫画人物与短句放在一起，常用上下两格呈现一句话的前后半段，或一问一答。第一组画面写着“问：你体重多少？”“答：我只接受党和组织的调查”，先提出问题，再用出人意料的回答形成笑点。

后续实操采用“银发、眼镜、粉红色中式外套”的角色形象，再配上语录制作图文。下面保留课件内嵌案例视频的封面帧，便于观察构图与文字形式，未逐句转录视频内容。






## 二、整体流程：从语录到可发布的图文

课件将制作过程分为四步：

| 步骤 | 使用工具 | 操作与产出 |
| --- | --- | --- |
| 1. 批量生产反内耗语录 | 豆包 AI | 提供 10 条参考文案，先总结特点，再生成 10 条类似风格的新文案 |
| 2. 生成老奶奶漫画图 | 即梦 AI | 输入人物提示词，结合参考图生成不同动作的角色，变高清并下载 |
| 3. 套模板设计成品图 | 图怪兽等设计平台 | 搜索模板，将选定语录和人物图片组合成完整图片并导出 |
| 4. 发布完整图文 | 小红书等平台 | 上传成品图，填写标题与话题，检查展示效果后发布 |

本课把“语录生成”和“图片生成”分开处理，文字在设计平台中排版。先确定一条要做的语录，再选合适的动作和构图，才能让上下两格的文字与画面对应。

课件小结提出将成品发布到抖音、小红书、公众号，实际页面演示使用小红书。蒲公英合作功能在图文制作流程之后单独介绍。


## 三、用豆包批量生产反内耗语录

### 3.1 提供参考文案，先分析再仿写

打开豆包：[doubao.com/chat](https://www.doubao.com/chat/)，输入课件中的完整提示词：

~~~text
小红书上有一个网红账号叫“人间清醒柒奶奶”，她是一位情绪疗愈类的博主，发布各种人间清醒、拒绝焦虑的文案，主打精神不内耗，吸引人们点赞，下面我给你她的10篇笔记文案。
不要烦不要烦，工作永远做不完
放心吧，交给我一定会搞砸的!
人生得意须尽欢，到点我是必下班
很喜欢上班，有种牢底坐穿的感觉
不上班带来经济问题，上班带来精神问题
早说了便宜没好货，会百度就不错了
一个星期，总有那么7天不想上班
什么叫万死不辞，就是每天被气死一万次我依然不辞职
奶奶，有人说我自私，因为他没有占到你的便宜
你长的有点土呀，我的土是用来埋你的！
请帮我总结以上人间清醒语录的特点，并按照以上语录格式和特点，帮我继续仿写10条类似风格的文案。
~~~

这条指令先交代账号定位和表达方向，再提供参考，最后同时要求“总结特点”和“仿写 10 条”。原文称参考内容为“10 篇笔记文案”，实际列出的是 10 条短句。

“人间清醒”“拒绝焦虑”“精神不内耗”限定表达主题；已有短句让 AI 看到口语、自嘲、夸张和反转怎样组织。课件没有要求长篇说理，生成结果需要保持短句形式，方便放入漫画画面。


### 3.2 查看 AI 总结的语录特点

原课截图中，豆包先归纳了三个特点：

| 特点 | 截图中的解释 | 参考句怎样体现 |
| --- | --- | --- |
| 语言直白幽默 | 使用通俗易懂、口语化且诙谐的语言，以夸张方式表达想法 | “很喜欢上班，有种牢底坐穿的感觉”用夸张比喻制造笑点 |
| 情感真实强烈 | 表达工作、人际关系等方面的负面情绪，容易引起共鸣 | “不上班带来经济问题，上班带来精神问题”把两难处境放在一起 |
| 句式简洁有力 | 多用简短陈述句或感叹句，节奏明快，便于记忆和传播 | “不要烦不要烦，工作永远做不完”重复开头，再落到一句明确的表达 |

这些特点解释了下一步的排版方式：文案较短，可以把前半句和后半句分放在两格中，用人物表情和动作承接语气。

### 3.3 原课生成的 10 条文案

豆包在特点分析之后返回以下内容：

1. 不要愁不要愁，工资少得眼泪流
2. 放心吧，让我去社交一定会社死的！
3. 生活苦短及时乐，到点我就赶紧撤
4. 超爱熬夜，有种和死神博弈的刺激
5. 不吃垃圾食品没快乐，吃了垃圾食品没健康
6. 早就说过无利不起早，能躺平就很美好
7. 一个月里，总有那么 30 天不想早起
8. 什么叫不离不弃，就是每天被虐千百回我还是不放弃
9. 宝子，有人说我冷漠，因为他没得到我的热情
10. 你说话有点毒呀，我的毒是用来怼你的！

后面的设计演示选择第 4 条，把“超爱熬夜”和“有种和死神博弈的刺激”分别放到上下两格。这里记录的是原课的幽默文案与制作示例。




## 四、用即梦生成老奶奶漫画配图

### 4.1 在图片生成中设置提示词与参考角色

打开即梦：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)，进入“图片生成”。课件第 5 页的输入界面包含提示词和一张参考图片，参考方式显示为“参考角色特征”。

按原课界面操作：

1. 输入银发老奶奶的外貌、服装、动作与画面要求，完整提示词见下一节。
2. 通过“导入参考图”加入角色参考，使用“参考角色特征”帮助保留外观。
3. 选择生图模型、精细度和图片比例。
4. 提交生成，比较返回的多张图片，再选适合当前语录的一张。

| 设置 | 第 5 页截图中的值 |
| --- | --- |
| 生图模型 | `图片 2.0` |
| 精细度 | `5` |
| 图片比例 | `3:4` |
| 参考方式 | 参考角色特征 |

这些数值属于原课当时的界面。右侧结果区的另一批图片标注为 `图片 2.1`、`3:4`，课件未展开模型切换步骤，不能把两张截图理解为始终使用同一模型。






### 4.2 两组生图提示词与人物一致性

课件第 7 页集中给出两组完整提示词和参考图说明，原文如下：

~~~text
AI生图提示词（1）：
银发老奶奶，戴着眼镜，粉红色的中式外套，一手拿酒杯，一手指天，动作自信夸张霸气，上半身，人物背景干净，宫崎骏漫画风格。

AI生图提示词（2）：
银发老奶奶，戴着眼镜，粉红色的中式外套，手拿中式扇子，面容慈祥，动作自信，上半身，人物背景干净，宫崎骏漫画风格。

注：也可加入参考图，生成效果更好，后一张图需要参考前一张图的角色特征、或者人物长相，以保证人物肖像的一致性。
~~~

两条指令共同保留银发、眼镜、粉红色中式外套、上半身和干净背景，区别在于人物动作：第一张拿酒杯、指天，表情和动作更夸张；第二张拿扇子、面容慈祥。

**后一张图参考前一张图的角色特征或长相**，是原课强调的操作。它用于让同一篇作品中的不同动作仍能被识别为同一个角色。生成后还要对照人物脸型、发型和眼镜检查，不能仅因使用了相同文字就认定肖像完全一致。

### 4.3 变高清并下载原图

课件第 6 页强调“AI 变高清、下载原图”：

1. 在结果中找到需要的那张角色图。

### 5.2 将人物与语录排成上下两格

进入编辑器后，把模板中的原有内容替换为本次准备的漫画图和语录。课件画布使用白底，分成上下两个区域：

| 位置 | 人物图片 | 文字 |
| --- | --- | --- |
| 上半部分 | 拿扇子的老奶奶靠左，右侧留白 | 超爱熬夜 |
| 下半部分 | 拿酒杯、指天的老奶奶靠右，左侧留白 | 有种和死神／博弈的刺激 |

按截图中的编辑方式完成：

1. 将下载好的两张图片加入画布，分别放在上方左侧和下方右侧。
2. 使用左侧“文字”中的“添加标题”“添加正文”等入口加入语录。
3. 把一句话分成前后两段，用较醒目的黑色大字放在人物旁边的空白处。
4. 调整图片位置、文字分行和留白，形成两格构图，再导出成品图。

原课编辑状态中，第二张图片仍有较明显的矩形边界；旁边展示的完成稿使用白色背景，并在上下格之间增加横线。以完成稿对照检查文字与人物是否互相遮挡，再保存作品。

模板标题仍显示“趣味2025新年愿望小红书封面”，但画布内容已经换成老奶奶语录，说明这里使用模板的可编辑布局。课件没有规定固定字体名、字号、画布尺寸、抠图参数或导出设置。






## 六、发布完整图文作品

### 6.1 上传成品图，填写标题与话题

进入小红书的“发布图文”页面，上传前一步导出的图片。原课第 10、11 页使用相同的发布截图，这里合并记录。

操作顺序如下：

1. 在图片编辑区域添加成品图。原课上传的是一张包含上下两格的完整图片，截图显示 `1/18`。
2. 填写笔记标题。原课标题为“熬不死我的，终将使我强大！”，标题框显示 `13/20`。
3. 在正文中添加相关话题，原课示例正文主要是一组话题标签。
4. 查看右侧“详情”预览，核对画面文字、标题和正文，再点击“发布”。页面也提供“暂存离开”。

原课话题如下：

~~~text
#被漫画治愈了 #治愈漫画 #女生必看 #女性励志 #情感共鸣 #释放压力焦虑 #人间清醒语录 #拒绝焦虑拒绝内耗 #熬夜最深的夜
~~~

图片张数与字数显示是课件中的界面信息，实际发布按当前页面规则填写。截图下方的活动话题属于当时的页面推荐，不是本课必须添加的内容。




### 6.2 检查双列信息流预览

课件同时展示了双列信息流中的封面效果。原课成品图的上下两格都放在同一张图里，因此缩略图中应能看到拿扇子、拿酒杯的两个动作，以及前后两段文字。

检查缩略图时，重点看“超爱熬夜”的开头是否清楚，第二段文字是否可辨认，人物与文字有没有挤在一起。这里展示的是发布页面的效果预览，不能据此认定作品已经获得真实推荐或互动。

## 七、课程小结

原课第 12 页把本篇制作过程归纳为四步：

1. 使用 **豆包 AI** 批量生产反内耗语录。
2. 使用 **即梦 AI** 生成老奶奶漫画配图。
3. 使用 **AI 设计平台** 套模板，导出成品图。
4. 将完整图文作品发布在抖音、小红书、公众号。

本课示例最终得到一张上下两格的语录漫画：同一类角色用不同动作配合前后两段文字，再附上笔记标题与话题组成完整作品。批量生成得到的是候选语录和候选配图，仍需要选取、排版并检查成品。


## 八、小红书开通蒲公英博主合作

### 8.1 从创作者中心找到“博主合作”

课件在制作流程之后，补充“小红书：如何开通蒲公英平台接广告”。原课展示的是手机端入口：

1. 在小红书打开左上角的三道杠菜单。
2. 选择“创作者中心”。
3. 在创作者中心找到“博主合作”。课件创作者中心首页已显示该入口，也展示了“全部服务”页面中“收益变现”下的同名入口。
4. 点击“博主合作”，进入“开通博主合作”页面。

这里选择的是“博主合作”，对应课件后续的内容合作申请。截图里同时出现的“买手合作”“探店合作”等入口没有在本课展开。
