# 19 AI做个人IP卡通形象口播视频


## 课程目标与制作流程

本篇对应课件 **《19-AI做个人IP卡通形象口播视频》**，共 15 页。课程希望制作一个能反复使用的个人 IP 形象，让固定角色出镜说话，逐步积累账号的辨识度。这里的 IP 指账号持续使用的人物形象与表达特点。

| 阶段 | 工具与操作 | 产出 |
| --- | --- | --- |
| 设计形象 | DeepSeek 写提示词，即梦 AI 生图 | 选定的角色图片 |
| 制作口播 | 即梦“数字人”的“对口型” | 由文本朗读或本地音频驱动的说话视频 |
| 使用自己的声音 | 自己录音，或用 Noiz 克隆音色并生成语音 | 可上传给即梦的配音文件 |
| 后期合成 | 剪映叠放原图与口播视频，调整位置、蒙版与羽化 | 保留原场景的口播成片 |

课件先演示文本朗读，再演示自己录音，之后介绍声音克隆。文本朗读和本地配音是同一角色的两种声音输入方式，可以根据需要选择。声音克隆生成的音频也回到“上传本地配音”这一步。

课件把第二阶段写成“图片转视频”，实际操作截图进入的是 **“数字人” → “对口型”**。本课用一张角色图生成说话视频，没有制作三维角色模型。第 14 页回顾流程，第 15 页没有新增教学内容。


## 一、案例中的个人 IP 口播结构

第 2 页内嵌三段手机录屏，展示不同卡通人物讲解内容。第一段包含戴眼镜的卡通男生与另一位男性角色切换，第二段用穿紫色上衣的男生讲解并叠加信息图，第三段用灰发圆眼镜角色出镜。



共同结构是让人物占据主要画面，用嘴部变化表现说话，再配合字幕或讲解资料传递内容。后续实操固定使用银灰发、黑框眼镜、灰色高领毛衣和米色外套的男性角色。

三段内嵌录屏约为 10.06 秒、10.27 秒和 18.09 秒，属于课件展示片段，不能据此判断被展示作品的完整时长。案例中的宣传标题只用于观察视频呈现方式，本课没有验证这些标题里的产品能力或数据。


## 二、让 DeepSeek 设计固定角色的生图提示词

### 2.1 先确定人物外貌、材质与环境

第 4 页给 DeepSeek 的完整请求是：

```text
我要设计一个ip形象，角色是一个30岁左右，戴黑框眼镜的短发男生，发色有些灰白，上半身特写，采用羊毛毡风格，背景是暖色调的室内，极简科技风，帮我生成一段文生图提示词。
```

人物年龄、眼镜、发型与发色共同构成识别特征，“上半身特写”限定取景，“羊毛毡风格”指定材质方向。暖色室内与极简科技风则描述背景。

### 2.2 保留原课的完整绘画提示词

```text
30岁亚洲男性半身特写，羊毛毡材质（柔软纤维质感，可见细密针脚细节），银灰色短发带自然蓬松纹理，窄边黑色圆框眼镜（哑光材质），眉眼温和带自信神态，身穿浅灰高领毛衣（粗针织质感），肩部搭米色羊毛毡外套。背景为暖橘色极简空间，包含：①弧形线条的悬浮书架（哑光白金属质感）②几何切割的智能台灯（琥珀色灯光）③半透明全息投影屏（显示简约数据图表）④抽象主义装饰画（暖色调色块拼接）。整体照明采用柔光散射，在毛毡表面形成细腻的光泽过渡，数字未来感与手工温度完美融合，8K超清细节，Studio Ghibli风格色调调配，Isometric角度构图。
```

提示词把原始要求展开为人物、服装、场景物件、光线和构图。`Studio Ghibli` 是风格色调参考，`Isometric` 指等距视角构图描述。它们与“羊毛毡材质”共同参与生图，实际结果仍要逐张比较。

“8K超清细节”属于提示词中的质量描述，不等于已经选择 8K 输出。下一步实际设置的图片尺寸为 `1024 × 576`。


## 三、在即梦生成人物图片并选定形象

### 3.1 输入提示词并加入人像参考

第 5 页进入即梦 **“图片生成”**，使用上一节的提示词。输入框还带有一张参考照片，显示 **“参考人像写真”** 标签，说明本例同时使用了文字与参考人像。

| 设置项 | 课件中的选择 |
| --- | --- |
| 生图模型 | 图片 2.0 Pro，`by Seedream 2.0` |
| 图片比例 | `16:9` |
| 图片尺寸 | `1024 × 576` |
| 参考方式 | 参考人像写真 |



输入框下方的 `Deepseek-R1` 按钮和下方“生图模型”分开显示，应分别识别。课件没有展开导入参考图后的选择面板，也没有给出参考强度。

### 3.2 比较多次生成的候选图

第 5 页展示两轮结果，每轮四张。结果标签都是“图片2.0 Pro、16:9、1K”，但画风与构图差别明显：有的偏写实，有的偏卡通，有的把人物不同角度拼成一张图。



本课后续选用的是单人正面半身形象。挑选时检查银灰短发、黑框眼镜、面部特征与服装是否符合预期，同时确认嘴部清楚、没有被手或物件遮挡。多视图拼图虽然能展示角色角度，但不符合后面单个人物正面口播的画面结构。

候选图中并非每张都呈现明显的羊毛毡效果，不能仅凭提示词含有材质词就认定输出完全符合要求。确定一张后保留它，后续对口型与剪映背景使用同一形象。


## 四、对选定图片进行超清处理并保存

第 6 页标注 **“超清下载”**。图片详情页的“编辑”区域中，红框圈出 **“HD 超清”**，详情同时记录模型“图片2.0 Pro”、比例“16:9”和“使用过：超清”。



按本课顺序，先对选定图做超清处理，再保存图片，供数字人口播与后期合成使用。截图展示了超清入口和已使用的记录，没有展开下载菜单、放大倍率或处理后的像素尺寸。



这张图保留完整肩部、服装与室内背景。后面的对口型结果会集中到脸部区域，原图需要一并保留，剪映阶段还会用它恢复较完整的横屏构图。


## 五、用“文本朗读”生成数字人口播

### 5.1 在数字人中上传角色图

第 7 页进入即梦顶部 **“数字人”**，选择 **“对口型”**，在“角色”区域放入选定图。此处旁边还有“动作模仿”，本课没有演示该功能。

| 项目 | 截图中的状态 |
| --- | --- |
| 功能 | 数字人 → 对口型 |
| 角色 | 前面选定的男性半身图 |
| 生成效果 | 生动 |
| 声音输入 | 文本朗读 |
| 说话时长提示 | `12.48s` |



“生成效果”中能看到“标准、生动、大师”三个选项，本页选中“生动”。界面提示 **“可能会对画面进行一定裁剪”**，因此要保留原图，并检查结果里的脸部与头发是否完整。

### 5.2 输入要让人物说出的文案

截图的“文本朗读”区域使用以下故事开头。转录时去掉了开头单独出现的一个双引号，正文保持原图内容：

```text
您听说过‘善心能渡劫’吗？村里有个刘寡妇，丈夫早逝，家徒四壁，连屋顶都漏雨。可就这么个苦命人，最后竟被全村尊为‘活菩萨’
```

课件旁注“适用于固定模特”。本例保持人物、服装和背景基本固定，主要让脸部随朗读变化，适合用同一张角色图继续制作口播。

该页没有展示朗读音色的选择界面，也没有显示最终生成按钮的完整位置。应依据当前页面完成生成，并以实际结果为准检查内容与时长。

### 5.3 查看口播结果

第 8 页内嵌了约 **8.53 秒、512 × 512** 的示例视频。下面是同一视频在开头和约 3 秒处的画面：



输入图是 `16:9` 横图，演示片则为方形脸部构图，符合前一页的裁剪提示。播放时应检查嘴部动作与声音是否对应、眼镜和脸部是否稳定，并确认句子结尾完整。

设置截图显示 `12.48s`，内嵌演示片约为 8.53 秒，两者不是一致的时长记录。笔记分别保留，不把面板提示当成已经验证的成片时长。


## 六、上传自己录制的音频来对口型

第 9 页继续使用同一角色，在“对口型”区域切换到 **“上传本地配音”**。这样人物说话的内容、节奏与音色来自上传文件。

1. 自己录好本次文案，保存为本地音频文件。
2. 确认“角色”仍是前面选定的图片，本页生成效果仍选“生动”。
3. 选择“上传本地配音”，点击“上传音频文件”。
4. 生成后完整播放，检查开头、句间停顿与结尾的声音和口型。



上传区域的原始限制说明是：

| 项目 | 课件截图中的说明 |
| --- | --- |
| 支持格式 | `MP3`、`WAV`、`M4A`、`FLAC`、`AAC`、`OGG` |
| 文件大小 | 最多 `10MB` |
| 时长 | `15s（大师、快速模式）/ 36s（标准模式）` |

这些是课件当时界面的提示，复做时核对当前上传区域的限制。截图没有展示实际上传的文件名、录音软件或录音参数。

第 10 页内嵌本地音频对口型的效果视频，约 **11.42 秒、512 × 512**。人物仍是同一银灰发男性，画面保持方形面部构图。



这里上传的是要让人物说出的完整文案音频。后面 Noiz 用来学习音色的短样本承担不同作用，不能直接把克隆样本当成本次成片文案。


## 七、在 Noiz 克隆自己的音色

第 11 页介绍声音克隆，工具入口为 [Noiz](https://noiz.ai/landing)。声音克隆先根据一小段样本建立可复用音色，下一步再输入新文案生成语音。

### 7.1 在语音库添加声音

进入“语音库”，点击 **“添加声音”**，打开同名弹窗。



弹窗中的项目与要求如下：

| 项目 | 课件中的说明或状态 |
| --- | --- |
| 名称 | 必填，用于区分自己创建的音色 |
| 声音标签 | 每个标签用逗号隔开 |
| 自动降噪 | 截图中已开启 |
| 语音文件 | 必填，按钮写“上传音色（3-10秒纯人声）” |
| 文件大小 | `10M` 以内 |
| 样本长度 | `3—10秒` 内效果最佳，低于3秒或超过10秒会影响生成质量 |



上传自己的纯人声样本，填入名称并核对降噪选项。弹窗还要求确认拥有上传、克隆样本的必要权利，并遵守平台条款。确认这些条件后勾选声明，点击“确认”。

### 7.2 在列表中确认音色已经创建

下一张截图的“语音库”中出现 **“回答我声音克隆”**，声音类型显示 `custom`，同时有声音 ID 与样本文本记录。这是本例创建完成后的可见结果。



列表里的 ID 和样本文本用于辨认这个音色，不是每次都要照抄的配置。创建后应通过试听确认声音，再用它生成后续口播文案。


## 八、用克隆音色生成新文案的音频

### 8.1 选择自己的音色

第 12 页在“音色”标签中，“你的音色”下选中 **“回答我声音克隆”**，右侧有橙色勾选标记。下方公共音色库还有其他声音，但本课此处使用的是刚创建的自定义音色。



### 8.2 输入文案并生成语音

“文本转语音”页面中的完整文案如下，按原截图保留表达：

```text
你天天熬夜刷手机，《内经》说"肝藏血"，子时不睡肝血拿什么养？眼睛模糊发黄还要强撑？回答我！说话！隔壁小李十点闭眼，现在双目如炬，你黑眼圈比墨还浓，是想当国宝还是想瞎？嗯？
```

这段是课件用来展示语音生成的口播台词，其中涉及健康的表述不作为本篇的医学知识。



页面顶部提供 **“插入停顿”** 和 **“标记漏字”** 两个快捷工具，底部“目标语言”选为 **“自识别”**，字数计数显示 `86/500`。输入文案并确认音色后，点击 **“生成语音”**。

本页没有展示快捷工具的具体用法，也没有展开“设置”或“历史”。生成后需要试听音色、读音和停顿，再保存音频，回到即梦“数字人 → 对口型 → 上传本地配音”生成这段新文案的视频。

课件只展示了输入与生成入口，没有展示 Noiz 生成完成后的下载页，也没有给出音频格式、采样率等参数。不要把本节这段新文案当作前面故事口播视频已经使用的配音。


## 九、在剪映叠放原图与口播视频

第 13 页依次列出 **“导入素材、调整视频位置、调整蒙版、羽化值、制作导出”**。素材区能看到两项内容：一张 `16:9` 人物原图，以及一段约9秒的方形口播视频。

### 9.1 先放原图，再把说话视频放到上层

原图保留完整的肩部、外套与室内背景，方形口播片段主要覆盖脸部。把两者叠放，就可以在原场景中保留较完整的人物构图，并让脸部说话。

1. 在剪映“素材”中导入选定角色原图与生成的口播视频。
2. 把原图放在下层，口播视频放在上层，两段起点对齐。
3. 选中上层口播视频，调整位置与大小，让眼睛、眼镜、鼻子和嘴与底图对应。
4. 检查视频带入的声音，确认要保留的文案能够正常播放。



截图中的上层视频标签为 `00:00:08:16`，下层原图更长，草稿总时长显示 `00:00:17:09`。最终应按实际口播范围调整底图与成片时长，避免视频说完后长时间只剩静止图片。

### 9.2 用蒙版和羽化让边缘过渡自然

蒙版用于限定上层视频显示的区域，羽化让蒙版边缘逐渐透明。人物位置对齐后，再调整可见范围与羽化，让说话区域与下层原图衔接，播放检查发际线、眼镜、下巴、衣领和背景边缘。



课件要求调整蒙版与羽化，但截图右侧打开的是“草稿参数”，没有展示蒙版类型、形状、大小、羽化值或缩放坐标。因此本节说明操作目的与先后关系，不补写不存在的固定参数。

### 9.3 完整播放后导出

截图中的草稿帧率是 `30.00帧/秒`，色彩空间为 `Rec.709 SDR`，比例与分辨率都显示“适应”。这些是当前草稿记录，课件没有打开导出设置，不能据此补写最终导出分辨率、码率等参数。

导出前从头到尾检查口型与声音、上下层边缘、画面连续性，以及结尾是否有多余静止段。按实际需要完成导出，再复看保存的视频。课件中的合成预览仍保留“即梦AI”水印，没有展示最终导出完成页。


## 十、课程小结

第 14 页回顾四个阶段：**即梦生成数字人形象、使用图片制作口播、生成或上传个人音频、剪映后期剪辑**。

先用 DeepSeek 把固定人物特征写清楚，再结合参考人像在即梦生成并选定角色图。对这张图做超清处理并保留原图，进入“数字人”的“对口型”，选择文本朗读或本地配音来驱动说话。

需要持续使用自己的声音时，先在 Noiz 上传短样本建立音色，再用新文案生成完整音频，上传给即梦。这里有两个不同文件：克隆音色用的短样本，以及本次人物实际说出的文案音频。

对口型结果可能裁成方形脸部画面，剪映阶段将它与原图叠放，通过位置、蒙版和羽化完成合成。声音、人物与背景衔接后，再修整开头结尾并导出。第 15 页没有新增操作内容。
