# 16 AI变装做个服装穿搭号


## 资料定位与课程目标

本篇对应课件 **《16-AI变装做个服装穿搭号》**，共 13 页。第 2 页展示原图、换装图和短视频，第 3 页说明应用方式，第 4—8 页讲模特试穿，第 9 页更换背景，第 10—11 页生成视频与导出实况，第 12 页小结，第 13 页没有新增教学内容。

本篇以本地 PPTX 的正文、操作截图及 1 段内嵌视频为准。试衣部分使用 **可灵 AI 的“AI试衣—AI换装”**；换背景与运动装视频使用 **即梦 AI**；实况转换使用 **手机剪映**。

课程围绕两类需求展开：商家把商品图穿到官方或自有模特身上，个人博主用固定照片展示不同服装。图片确定后，还可以更换场景、生成短视频，再转换成实况动态图。


## 一、先看换装前后的变化

第 2 页按“网络原图、AI试穿变装、AI试穿视频”排列素材。原图人物穿黄白条纹衬衫，换装后变为浅色方领短袖上衣，人物位置与窗边场景大体延续。



对比时也能看到头发长度、耳饰、下装和背景树木出现变化。因此“固定模特”是输入素材的使用方式，生成后仍要检查脸部、发型、配饰与服装细节。

内嵌试穿视频约 **5.10 秒、736 × 1216、30 帧/秒**，人物从正面略向侧面转动，右下角可见“可灵AI 1.6”标记。下面截取开头与约 3 秒处，观察静态换装图怎样变成短动态。



这段视频对应窗边人物案例。后面即梦的运动装转身视频在课件中以界面截图展示，PPTX 没有内嵌它的完整视频文件。


## 二、三种试穿方式与后续输出

第 3 页把用途分成四项，前三项根据“人物从哪里来”选择，第四项接在图片制作之后。

| 课件方向 | 人物来源 | 服装来源与任务 | 主要产出 |
| --- | --- | --- | --- |
| 官方模特 | 工具内置模特 | 上传自家商品图，单件或多件试穿 | 电商服装展示图 |
| 自定义模特 | 上传固定人物照片 | 为同一人物更换商品服装 | 固定人物的试穿图 |
| 个人穿搭 | 上传个人或固定人物照片 | 更换上衣、下装，必要时调整细节和背景 | 不同穿搭与场景图片 |
| 转换视频、实况 | 前面生成的图片 | 用图生视频生成动作，再到剪映转换 | 短视频、实况动态图 |

换装阶段先准备 **人物图与服装图**。后面换背景、生成视频都以选定的试穿结果继续处理，不能只完成服装上传就跳到视频生成。

总览标注“AI免费生成”，实际试衣截图中有“灵感值”消耗提示。例如 4 张结果显示消耗 20，2 张显示消耗 10。这是课件当时的界面信息，当前额度和费用以实际工具为准。


## 三、使用官方模特试穿商品

### 3.1 单件：更换上衣

第 4 页打开的是 **“AI试衣” → “AI换装”**，人物来源选择“官方模特”，衣服选择“单件”。

1. 在官方模特列表中选择一个适合展示商品的人物。截图选中第一位、米色背景的女性模特。
2. 选择“单件”，上传浅色短袖上衣商品图。
3. 检查服装卡片右上角的类别，截图是 **“上衣”**。
4. 选择生成张数。示例为 `4张`，按钮显示“立即生成”及消耗 `20`。
5. 生成后逐张检查领口、袖型、纽扣与衣摆，选择合适的结果保存。



结果区展示两张候选图，两张已经穿上浅色上衣，但衣摆长短、下摆开口与纽扣细节有所不同。**结果区的两张并排图片都是生成结果**，原始模特和服装输入要回看上面的设置区。



界面还提供“模特规则”“服装规则”和“使用指南”入口，课件没有展开这些说明。上传条件与支持的服装类别应以实际界面要求为准。


### 3.2 多件：上衣与裙子一起搭配

第 5 页把服装选项改为 **“多件”**，同时上传浅色短袖上衣与浅色半身裙，张数仍是 `4张`，按钮显示消耗 `20`。



多件试穿要分别核对每张服装图，检查生成结果中上衣与裙子是否都替换成功。结果区展示同一位深色皮肤、短发模特的两张全身图，裙长、腰部衔接和鞋部表现不同。



本页设置截图仍高亮前一页的第一位官方模特，结果区却换成另一位人物，两张截图不是完全对应的任务状态。可以据此学习多件上传与结果检查，复做时需确认自己实际选中的模特。


## 四、上传自有模特，保持人物来源固定

### 4.1 女性人物：从条纹衬衫换成浅色上衣

第 6 页将人物来源改为 **“自定义”**，上传窗边的黄白条纹衬衫人物照，服装仍是前面的浅色短袖上衣。

1. 在“AI换装”中选择“自定义”，上传准备使用的固定人物照片。
2. 选择“单件”，上传目标上衣，类别设为“上衣”。
3. 选择 `2张`，示例按钮显示消耗 `10`，提交生成。
4. 对比候选图，检查人物外观、领口、衣摆、配饰和背景，再保留合适的一张。



结果图换上目标上衣，但两张的领口、纽扣和下装细节并不完全一致，头发也比输入图短。用于固定人物内容时，应把这些变化一并检查。




### 4.2 男性人物：从西装换成黑色立领上衣

第 7 页沿用“自定义＋单件”的方式，上传穿蓝色西装、双臂交叉的人物照片，再上传黑色立领上衣商品图。类别是“上衣”，张数显示 `2张`。



生成结果保留了灰色背景与交叉手臂的大体姿态，西装变成黑色立领上衣。两张图的手部位置、衣服褶皱和扣子分布仍有差别。



这说明同一套操作可以用于不同人物与商品。手臂遮挡衣服的部位、扣子和领口等结构需要仔细看，不能只凭整体颜色接近就判断服装完全还原。


## 五、个人穿搭：固定人物，更换整套服装

### 5.1 自定义人物与多件搭配

第 8 页上传一张穿紫色运动装的全身人物图，选择 **“多件”**，分别上传蓝色长袖运动上衣和浅色长裤。示例选择 `2张`，按钮显示消耗 `10`。



这一步同时确定新上衣与新下装。相较前面的半身模特，全身图还能检查裤腿、鞋子和整体搭配是否完整。课件标题写“上传个人照”，实操重点是使用同一张人物图反复换衣，照片应使用本人或获准使用的素材。

### 5.2 用局部重绘整理裤装

同页的结果记录名称为 **“局部重绘”**，文字是：

```text
完整裤子，无外搭
```



结果中原来的运动短裤叠穿造型已变为完整浅色长裤，人物仍保持一手叉腰的姿势。这是本页额外展示的局部调整，不能把整张结果都归为一次“多件试穿”的直接输出。

课件没有展示局部重绘入口、涂抹区域与具体参数，因此这里只保留可见的文字与结果。实际调整时，先确认需要修改的位置，再检查裤腰、上衣下摆、手指和鞋部有没有受到影响。


## 六、到即梦更换背景，展示不同场景

### 6.1 导入试穿图作为智能参考

第 9 页切换到即梦的 **“图片生成”**，把蓝色上衣、浅色长裤的人物图导入为“智能参考”。输入框中的文字为：

```text
将图片背景替换为真实的田径场
```



截图“生图模型”选择 **图片 2.0 Pro，`by Seedream 2.0`**。输入框下方另有 `Deepseek-R1` 按钮。图片尺寸栏显示 **`989 × 1360`**，比例选项处于灰显状态，不能把它记成选中了 `9:16`。

### 6.2 生成记录使用的是公园提示词

右侧结果记录里的文字与输入框不同，完整可见文字为：

```text
将图片背景替换为户外公园，真实自然，超高清
```



结果是四张公园场景，记录标签显示 `989:1360`。输入框“田径场”与历史记录“户外公园”对应不同文本，应按想要的场景选择提示词，不能把这组公园结果写成田径场生成成功。

换背景后还要检查人物和商品本身。四张图中的上衣长短、腰部露出程度、领口和拉链并不完全相同，说明参考生成可能连服装细节一起改变。用于展示具体商品时，保留衣服结构准确的一张，再继续做视频。


## 七、把穿搭图生成服装展示视频

第 10 页在即梦中选择 **“视频生成” → “图片生视频”**，上传选定的公园人物图，文件名显示 `ImageInput.png`。动作提示词为：

```text
模特前后展示衣服效果
```

| 设置 | 截图中的选择 |
| --- | --- |
| 输入图片 | 蓝色运动上衣、浅色长裤的公园全身图 |
| 视频模型 | 视频 S2.0，`by Seaweed alpha` |
| 生成时长 | `5s` |
| 视频比例 | 自动匹配 |



生成记录显示“参考首帧”，播放器是 `00:03 / 00:05`，截图中模特已转到背面，可以看到蓝色上衣后背与裤装的整体轮廓。



本页没有完整内嵌视频，只能确认截图中的生成设置和背面展示状态。生成后应完整播放，检查脸部、手部、转身衔接、衣服颜色与线条是否稳定，再保存供手机剪映使用。

另外，只有正面参考图时，模型会推测未提供的背面细节。商品展示应对照实物或已有背面素材核对，避免把生成的后背结构当作商品的真实设计。


## 八、手机剪映导出实况动态图

### 8.1 导入视频，调整比例与构图

第 11 页把运动装展示视频导入手机剪映。底部打开“比例”面板，选中 **`9:16`**，顶部导出类型显示 **“实况”**。

1. 将前面保存的展示视频导入手机剪映。
2. 在“比例”中选择 `9:16`，调整缩放与位置，检查头部、鞋子和转身动作是否完整。
3. 查看实际时长与片段结尾，再在顶部选择“实况”，点击“导出”。



前面的图生视频设置是 5 秒，这张剪映截图的预览则显示 `00:05 / 00:06`。课件没有解释这处时长变化，复做时按自己的时间线检查，尤其注意末尾是否留有多余片段。

### 8.2 保存后检查实况预览

导出结果页显示 **“已保存到相册和草稿”**，下方提供抖音、小红书、微信、朋友圈等分享入口。



后续截图已经进入发布编辑预览，左上角出现 **“LIVE”** 标记，顶部有“选择音乐”，画面提示“live支持声音了，点击可以关闭”。这说明素材在该截图的发布界面中被识别为实况内容。



保存后先检查动态是否正常播放、声音是否需要保留，以及竖屏裁切是否影响人物。课件没有展示正式发布后的页面，也没有给出实况导出的编码、分辨率或设备适配参数。


## 九、课程小结

第 12 页回顾四种用法：**官方模特试穿商品、自有模特试穿商品、个人照片换衣与换背景、图片生成视频并转换实况**。

复做时先在可灵确定人物与商品服装，检查候选图；需要整理裤装时再处理局部细节。随后用即梦更换场景、生成展示动作，最后在手机剪映调整构图并导出实况。每次进入下一步前，都回看人物和服装是否仍与目标一致，避免把前一步的细节变化继续带进视频。

第 13 页没有新增教学内容。
