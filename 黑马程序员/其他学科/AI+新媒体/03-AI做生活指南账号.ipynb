# 03 AI做生活指南账号


## 资料定位与课程目标

本篇对应网盘课件 **《03-AI做生活指南账号》**，共 15 页。课件正文以“健康生活号”为例，讲解简笔画素材生成、封面与内页设计、选题和文案扩写，以及小红书图文发布。第 15 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。

开场的“一条广告到手 1.2W”是课件中的宣传表述，正文没有展开接广告、报价或收益验证。本篇按实际教学内容记录图文制作方法；课件所说的“免费 AI”，也以工具使用时的额度与收费规则为准。


## 一、先看案例：生活建议怎样做成图文

原课先展示两组账号作品：一组用粉色背景、卡通角色和醒目的黑色大字表达早睡、放松、生活习惯等主题；另一组用方格字标题搭配简笔画，把一条生活建议放在封面上。

这种图文通常由两部分组成：**封面用主标题、副标题和插画点出主题，后续内页分条解释具体内容。** 同一账号重复使用相近的字体、颜色和插画风格，形成一组外观统一的作品。下面是课件内嵌案例视频的封面帧，保留其图文形式，未逐句转录视频内容。






## 二、整体流程：从配图素材到完整图文

课件把制作过程分为四步，前三步标注“全程使用国产免费 AI”，最后一步提出同步发布到各大社交平台。实际发布演示使用小红书。

| 步骤 | 使用工具 | 主要操作与产出 |
| --- | --- | --- |
| 1. AI 生成配图素材 | 豆包 AI | 参考已有画面生成卡通简笔画，下载原图 |
| 2. AI 设计封面、配图 | 图怪兽等设计平台 | 搜索封面模板，组合主标题、副标题和插画，制作正文内页 |
| 3. AI 扩写选题、文案 | 豆包 AI | 根据参考选题扩写新选题，参考案例结构生成主标题、副标题和正文 |
| 4. 完整图文作品发布 | 小红书图文发布页面 | 上传封面和内页，填写标题、正文与话题，检查预览后发布 |

第一步得到的是插画素材，第二步才把文字和插画合成可发布的图片。课件先演示设计，再演示文案生成；做新作品时，生成的新文案还需要填回封面和内页，保证文字与图片对应。


## 三、AI 生成配图素材

### 3.1 选择参考图片，用豆包生成类似素材

原课选择“秋天要像鸡一样生活”的图文封面作为参考：白底上方是方格字标题，下方是淡粉色背景的卡通人物。生图时重点描述插画主体、动作和背景，标题与正文留到设计平台编辑。

打开豆包：[doubao.com/chat](https://www.doubao.com/chat/)，提供参考图片，输入截图中的提示词：

~~~text
参照上图，帮我生成一个卡通简笔画人物，甩开手臂大步走路，淡粉色背景
~~~

这条指令用“参照上图”说明参考来源，再给出卡通简笔画风格、甩臂走路的动作，以及淡粉色背景。生成结果是一组外观相近、细节不同的卡通人物，供后续挑选。






### 3.2 选中生成结果，下载原图保存

豆包返回多张图片后，打开需要的那张，在大图查看界面右上角选择“下载原图”。课件截图同时显示“智能编辑”“区域重绘”“扩图”“擦除”等入口，但本课的明确操作是选图并下载，没有展开这些编辑功能。

保存下来的图片用于后续封面设计。检查人物动作、背景颜色是否符合需求，再将原图导入设计平台。

### 4.2 合成完整封面

封面按参考作品组合三部分：

1. **主标题**：上方的大字“秋天要像鸡一样生活”，分成两行，使用方格字样式。
2. **副标题**：主标题下方的小字，补充主题与读者范围。
3. **配图**：把前面下载的卡通人物放在下方，保留淡粉色背景。

在模板编辑器中替换文字、放入图片并调整位置。原课画布中可见“添加标题”“添加正文”“常用文案”等文字入口，选中文字时提示“双击修改文字”。完成后得到一张上文下图的封面。

这里沿用的是“秋天像鸡一样生活”的设计示例，后面生成“冬天像熊一样生活”的新文案后，需要同步更新封面文字及配图。




### 4.3 合成正文配图

封面之后，继续制作承载正文的内页。原课参考内页上方保留标题“秋天像鸡一样生活”，下面按编号分段，部分小标题和重点句使用粉色底色突出。

在设计平台中按这种结构排入正文：先设置页标题，再放编号、小标题和对应段落，让读者顺着序号阅读。内页主要承担解释内容的任务，封面负责引出主题。

> 原课编辑器截图中，多处正文重复“气血虚的人，秋天要像鸡一样生活”，属于排版演示内容，不能作为完整正文直接发布。应把后续确认的文案替换进去，再检查重复文字与排版。

## 五、用豆包扩写选题

### 5.1 用三个参考选题扩写 15 个新选题

课件将这一环节称为“AI 复制爆款选题”。具体做法是给 AI 账号身份、目标人群、参考选题、输出数量和选题方法，让它基于参考方向扩写。

原课提示词如下：

~~~text
身份：假如你是一个【教年轻人如何健康养生的百万粉丝博主】，目标人群是【18-35岁的女性】，请你给我15个关于【健康养生】方向的选题，选题需要围绕以下3个爆款选题扩写，每个选题帮我扩写5个新选题：
1、中医：女生最佳作息时间表
2、原来熬夜的伤害可以被抵消
3、提醒：原来人是这样慢慢变笨的！
要求： 
给出的选题符合小红书爆款选题的风格，一定要通俗易懂，吸引人，抓人眼球，受众面广，且需要运用以下 8 种方法里的任意方法去产出选题。 
8种方法： 
1、能给目标用户带来巨大的价值，对他们有用； 
2、针对目标人群的需求痛点，能解决他们的痛点； 
3、能引起人们的好奇，制造悬念； 
4、能结合目前的热点做选题； 
5、有冲突，有矛盾，能引发用户的评论；
6、蹭一个跟这个选题方向相关的名人、企业家、网红、大佬等； 
7、能迎合人性的某个弱点，比如贪、懒等； 
8、能调动用户情绪，比如激动、兴奋、愤怒、喜悦等。
~~~

这条提示词要求围绕三个方向各扩写 5 个，共 15 个。截图中的返回结果也按三个方向分组，可以逐组核对数量，以及新选题是否符合账号受众。

> 补充：这里的“爆款”是原课的选题表达，不能保证传播效果。健康类标题中的“熬夜伤害可以被抵消”等说法，需要先核实依据，避免将吸引点击的表达直接写成健康结论。

## 六、参考案例结构生成新文案

### 6.1 先整理最近的参考文案

选题之外，课件还要求“批量整理最近火的健康养生文案”，再把案例提供给 AI。示例仍是“秋天像鸡一样生活”，正文依次包含：

1. 用秋意、体感和季节变化引入主题。
2. 引用《黄帝内经》中的文字，连接“秋天”与“鸡的作息”这个比喻。
3. 用“像鸡一样生活，晨起一声啼”展开第一条做法。
4. 用“像鸡一样生活，时刻饮水”展开第二条做法。

这一步让 AI 获得具体的结构、语言风格和篇幅参考。下一个提示词把主题换成“冬天要像熊一样生活”，继续要求输出主标题、副标题和正文。


### 6.2 输入主题、输出格式和案例参考

将新主题与参考文案一起输入豆包。下面保留课件中的完整提示词：

~~~text
我想在我的小红书账号发布一篇主题为「冬天要像熊一样生活」的内容，请你帮我写出文案，要求有主标题、副标题、正文。

以下是5个其他账号的案例参考，请参考它们的内容结构、语言风格、字数。

秋天像鸡一样生活：

1、慢慢已经感受到了秋意，体感清爽了很多，但燥气也起了，到了一年中最需要养阴的季节。
2、秋天最好的养生方式，《黄帝内经》中都给大家说清楚了:秋三月，此谓容平，天气以急，地气以明。早卧早起，与鸡俱兴，使志安宁，以缓秋刑；秋天，学习鸡的作息方式。
3、像鸡一样生活，晨起一声啼:
像鸡一样早起，在天微微亮，鸡鸣声此起彼伏时起床(5:00~7:00)，到空气好的地方深呼吸(到丹田)，放松身体大步走。早晨的新鲜空气能帮助你补充氧气。听着鸟儿声，迎着阳光走一小段路，让阳光拂过身体，既可振奋阳气，又可散湿，来自秋天的疲劳感会少很多。
4、像鸡一样生活，时刻饮水: 
秋天是干燥的季节，保持足够的水分摄入，就像鸡随时都会找水源一样。多喝水有助于维持身体的湿润，促进新陈代谢，减少体内废物。
~~~

原文写“以下是 5 个其他账号的案例参考”，但这一页只实际列出“秋天像鸡一样生活”这一个案例，没有给出另外四篇。跟做时应把案例数量与实际提供的材料对齐。

参考内容结构和表达方式后，仍需为新主题写出自己的正文，检查是否残留原作品句子，以及引用、健康表述是否可靠。


### 6.3 查看生成结果，取出主标题、副标题和正文

课件下一页展示豆包生成的“冬天像熊一样生活”文案：

| 部分 | 截图中的内容 |
| --- | --- |
| 主标题 | 🐻 冬天来了，我们要像熊一样冬眠吗？❄️ |
| 副标题 | `#冬天养生指南 #像熊一样生活 #冬日温暖`，实际返回的是一组话题标签 |
| 开头 | 以气温下降、冬天临近引入，引用《黄帝内经》中的冬季作息描述 |
| 正文小节 1 | 像熊一样生活，冬日养精蓄锐 |
| 正文小节 2 | 像熊一样生活，保暖为主 |
| 正文小节 3 | 像熊一样生活，适量运动 |
| 正文小节 4 | 像熊一样生活，充足睡眠 |
| 结尾 | 冬季生活方式的收束与祝愿，另附 `#冬季保养 #健康生活 #冬眠模式` |

生成结果沿用了季节引入、引用、分条建议和结尾的组织方式。将确认后的标题和正文填回设计稿，按内页的容量分段排版；若需要封面上的一句话副标题，还要把 AI 返回的话题标签改成适合封面的文字。

> 补充：本课用于演示内容生成。动物冬眠的比喻、古籍引用和 AI 给出的饮食、作息建议，应分别核实，不能仅凭生成结果推定适合所有读者。




## 七、完整图文作品发布

### 7.1 上传图片，填写标题、正文和话题

封面与正文内页准备好后，进入小红书的“发布图文”页面。按课件的发布界面完成：

1. 上传封面和内页，检查图片顺序。原课截图已上传 2 张，页面显示 `2/18`。
2. 填写笔记标题。截图中的标题为“气血虚的人，冬天像熊一样生活”。
3. 在正文区填写内容，并通过“#话题”添加与作品有关的话题。
4. 查看图文预览，确认文字、图片和主题一致，再点击“发布”；界面也提供“暂存离开”。

原课正文区截图显示以下话题：

~~~text
#我的健康生活 #健康科普 #健康养生 #养生 #养生日常 #秋季养生 #秋天 #我的养生仪式感 #中医养生
~~~

图片数量、标题和正文字数显示属于原课当时的界面，实际发布按当前平台规则处理。

> 原课素材差异：发布编辑页的标题已经改成“冬天像熊一样生活”，图片缩略图仍是秋天示例，话题也保留“秋季养生”“秋天”。实际制作时，应统一季节、主角、图片文字和话题后再发布。




### 7.2 查看双列预览

课件还展示了“双列”预览，用来观察作品出现在信息流中的样子。此处封面已经改成“冬天要像熊一样生活”，下方显示笔记标题，卡通素材仍沿用前面的走路人物。

通过预览检查大标题是否清楚、分行是否合适，以及缩略图中的插画能否辨认。这是作品展示效果的预览，页面还包含其他笔记的灰色占位卡片，不能将其视为已发布后的真实推荐流量或互动结果。




## 八、课程小结

原课将健康生活号的图文制作归纳为四步：

1. 使用 **豆包 AI** 生成作品的配图素材。
