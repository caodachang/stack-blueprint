## 二、整体制作流程：文案、配图、视频、剪辑

第 3 页把课程拆成四个阶段，并在第 14 页再次归纳为“生成文案—生成配图—生成视频—优化制作”：

| 顺序 | 工具 | 主要操作 | 阶段产出 |
| --- | --- | --- | --- |
| 1 | DeepSeek | 参考爆款文案结构，生成养生食材文案；再为文案生成分镜和 AI 绘画提示词 | 可朗读的短文案、分镜表、绘画提示词 |
| 2 | DeepSeek + 即梦 | 把每个分镜的场景描述转成匹配的食物配图 | 一组风格统一、能对应文案的图片 |
| 3 | 即梦 | 用“图片生视频”让静态配图产生轻微动作或镜头变化 | 可用于剪辑的短视频素材 |
| 4 | 剪映 | 排列素材、添加文本朗读和动画、关键帧、背景音乐、字幕，处理平台水印并导出 | 完整的养生食物短视频 |

课件的制作思路是“对标爆款，优化细节”。这里的对标重点是结构、节奏和画面匹配，不等于复制他人的具体文案、画面或健康结论。

## 三、用 DeepSeek 生成养生食物文案

### 3.1 先给参考文案，再限定输出要求

第 4 页给出的提示词要求：提供 5 条视频文案作为参考，让 DeepSeek 再写 5 个新的养生食物文案；文案开头要吸睛、通俗易懂、总时长不超过 30 秒，并严格参考原文的内容结构、语言风格和字数。

课件中的参考内容包括：蔬菜“冠军”排行、蒸食主食清单、把外貌问题归因于饮食和生活方式、白菜搭配食材的说法，以及“掉秤最快的食物”清单。将课件提示词整理如下：

~~~~text
我需要做食物养生的爆款视频，现在我给你以下5条视频文案参考，请参考后帮我写5个新的养生食物文案。

要求：
1、文案开头吸睛，能够牢牢吸引住用户，并且言简意赅通俗易懂，文案总时长不超过30秒。
2、请严格参考以下文案的内容结构、语言风格、字数。

一：排名第一的蔬菜都有哪些？
一、钙的冠军是油菜。二、维生素A的冠军是苦瓜。三、叶酸的冠军是菠菜。四、维生素E的冠军是茄子。五、维生素B的冠军是芹菜。六、维生素C的冠军是番茄。七、纤维素的冠军是红薯。八、胡萝卜素的冠军是西兰花。九、消化酶的冠军是白萝卜。十、蛋白质的冠军是莲藕。

二：免疫力最喜欢的主食竟然是：一蒸芋头，二蒸紫薯，三蒸板栗，四蒸南瓜，五蒸玉米，六蒸菱角，七蒸豌豆，八蒸毛豆，九蒸马蹄。

三：人丑是吃出来的，原因包括长期饱食、吃的油腻、喝水太多、吃太多甜食、长期躺在床上、压力过大、不注意防晒。

四：白菜到底有多厉害你知道吗？清水煮白菜，不放盐不放油，把生姜去皮放进锅里，加入几根白萝卜、适量花椒和一块豆腐，凑齐这些食材后，连续吃一周，当成一日三餐，饿了就吃，直到吃饱。你会发现上厕所通畅了，体内的垃圾消失了，脸色红润了，浑身变轻松了。

五：掉秤最快的食物是什么？青菜炒香菇。鸡胸肉拌黄瓜。水煮玉米棒。豆芽素炒。西兰花炒蛋。芹菜炒虾仁。豆腐烧菠菜。金针菇拌菠菜。冬瓜做汤。豆腐拌黄瓜。
~~~~

### 3.2 查看输出并做事实审查

第 5 页截图中的输出示例包括“皮肤最爱的蔬菜排行榜”“肠道最怕的刮油食物 Top8”等标题，也出现“抗氧冠军”“排毒冠军”“连吃 15 天素颜都能白到反光”等强功效表达。截图反映的是模型按参考结构生成的结果，不代表课件对这些说法做过营养学验证。

实际使用时，先保留“排行/清单/开头提问”的叙事结构，再逐条核对食材的营养信息、适用人群和食用边界；无法确认的“冠军”和结果承诺要删掉或改成中性描述。



## 四、用 DeepSeek 生成分镜脚本和 AI 绘画提示词

### 4.1 让每句文案对应一个画面

第 6 页在已经生成文案的基础上继续提问，要求输出 10 个分镜头脚本，并让每句文案对应一个 AI 绘画提示词。提示词需要写清绘画风格、人物主体和主体描述、镜头语言、环境描述、光影氛围，合成一段约 50 字的中文提示词。

课件原提示词如下：

~~~~text
文案创作得不错，请将第一个健康食物文案，帮我生成如下内容：
1、10个分镜头脚本
2、每句文案对应一个分镜头的AI绘画提示词
分镜头用表格+序号表示，每个分镜头的提示词需要包含：绘画风格（古风水墨画风格）、人物主体和主体描述、镜头语言、环境描述、光影氛围，并合成一段话。字数50字左右，中文提示词。
3分钟视频约=36个分镜头
~~~~

截图实际显示的是分镜表的第 1–7 行，例如“展开的绢布榜单”“番茄在青瓷碗中旋转”“银耳在琉璃盏中半透明颤动”等；截图下方没有完整显示第 8–10 行，笔记不补写未展示的内容。

“10 个分镜头”和“3 分钟视频约 36 个分镜头”是课件同时给出的两个数量提示：前者是一次生成的表格数量，后者是较长成片的估算。实际数量应按旁白速度、镜头停留时间和成片长度重新安排。

从截图可见的操作线索包括：选择图片生成入口、输入中文提示词、选择候选图片；截图标出了图片模型“图片 2.0 Pro”和 `3:4` 画面比例。这里仅记录截图可见的模型和比例，不把它们写成当前平台必选项，也不补写费用、额度或下载参数。

## 六、用即梦把图片生成短视频素材

### 6.1 图片生视频的操作顺序

第 9–10 页展示即梦的“视频生成—图片生视频”页面。按截图可整理为：

1. 进入即梦的 **视频生成**，选择 **图片生视频**。
2. 上传选中的配图，截图中的文件名显示为 `ImageInput.png`。
3. 输入与画面动作相符的短描述，例如“农妇采摘沾露番茄”或“老医者执笔书写绢布榜单”。
4. 按需要选择视频模型并生成，预览主体、食材和镜头运动是否稳定。
5. 下载合格片段，作为剪映时间线中的画面素材。

课件截图中可见 **视频 S2.0（by Seaweed alpha）**、生成时长 `5s` 和视频比例“自动匹配”。这些是截图当时的界面设置，课件没有说明费用、额度、生成次数、下载格式或更高阶参数，不能据此推导固定的生成成本或成片时长。





## 七、用剪映剪辑完整的养生视频

### 7.1 排列画面，添加朗读和动画

第 12 页要求把生成的视频素材导入剪映并排好顺序，再添加文案、文本朗读和动画。可以按以下顺序复做：

1. 导入各条图片生视频素材，按照旁白的句子顺序排列。
2. 在画面上添加对应字幕或文案，逐句检查食材名称和字幕断句。
3. 使用文本朗读，截图中选择的音色是 **云泽大叔**；试听后检查语速和停顿是否能跟上画面。
4. 添加适量动画，让静态画面和镜头切换更灵动，同时避免动画遮住食材或字幕。

### 7.2 用关键帧、音乐和字幕完成细节优化

第 13 页和剪映截图中可见的细节包括：

| 细节 | 课件截图可见内容 | 作用 |
| --- | --- | --- |
| 关键帧 | 给画面做缩放或位置变化 | 让镜头有重点，避免整段画面完全静止 |
| 背景音乐 | `美丽的神话（笛子版）` | 作为整体氛围音乐，音量要避让旁白 |
| 速度 | 时间线截图出现 `变速 1.6X` | 这是示例工程中的片段设置，不是所有素材的固定值 |
| 字幕 | 逐句调整字幕位置和时间 | 让字幕与朗读、画面同步 |
| 水印 | 去除相关 AI 平台水印 | 发布前检查画面是否仍残留平台标识，并遵守素材授权和平台规则 |

课件没有展示导出分辨率、帧率、码率、音量数值或统一的最终时长，因此只记录截图里出现的剪辑操作。完成后从头播放，重点检查旁白、字幕、食材画面是否一一对应，关键帧是否造成明显抖动，音乐是否压住人声。





## 八、养生视频的合规表达

第 15 页专门讨论“养生视频怎么避免违规”，给出四个方向：**不做功效承诺、把效果写成动态化描述、提前做官方审核、小号提前分发测试**。

课件明确列出的高风险表达包括：

- “根治 / 治愈 / 包治 / 100% 有效”等绝对化承诺。
- “糖尿病必 XX 吃 / 高血压救 XX 星 / 失眠良 XX 药”等把食物写成治疗方案的表达。
- 任何未经证实、面向特定疾病人群的确定性疗效承诺。

更稳妥的改写方向是：把内容放在食材、口味、烹饪方式和个人体验上，使用“有人这样搭配”“个人体验供参考”“体质不同，效果可能存在差异”等限定语；涉及具体症状、疾病、用药或饮食禁忌时，提示用户咨询专业医师或线下正规医疗机构。限定语不能替代事实核验，也不能把虚假的功效说法包装成合规内容。

发布前按课件建议做检查：

1. 用平台官方预审工具检查内容。
2. 使用 **巨量创意—投前检测** 和 **轻抖—违禁词检测** 等工具检查敏感表达。
3. 先用小号分发观察审核和反馈，再决定是否扩大分发。
4. 对食材营养、疾病相关说法和画面中的文字再次核验，保留来源或改成中性表述。



## 九、课程小结

课件最后把“AI 做养生食物爆款视频”的步骤归纳为：

1. **DeepSeek 生成养生食物文案和 AI 绘画提示词**：先给参考结构，再检查每条食材和功效说法。
