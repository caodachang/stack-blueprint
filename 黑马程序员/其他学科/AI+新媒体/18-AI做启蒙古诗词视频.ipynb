# 18 AI做启蒙古诗词视频


## 课程目标与制作流程

本篇对应课件 **《18-AI做启蒙古诗词视频》**，共 14 页。课程面向儿童及家长，用 AI 把诗句拆成分镜，生成统一人物的卡通画面与短视频，再配上朗读、歌曲和字幕，剪成启蒙古诗词视频。

| 阶段 | 工具 | 主要产出 |
| --- | --- | --- |
| 文案与分镜 | DeepSeek | 每句诗对应的画面描述与绘画提示词 |
| 画面与动态 | 即梦 AI | 分镜图片、保持人物特征的后续画面、图生视频 |
| 朗读与音乐 | 剪映、即梦 AI | 诗词朗读音频、古诗词人声歌曲 |
| 剪辑成片 | 电脑剪映 | 按声音调整的视频、背景音乐与字幕 |

课件的文案和画面部分以 **《相思》** 为例，朗读与音乐部分改用 **《望庐山瀑布》**。下面保留各自的示例，复做完整作品时应统一同一首诗的文字、画面、音频与字幕。第 13 页回顾流程，第 14 页没有新增教学内容。


## 一、案例中的启蒙视频结构

第 2 页内嵌两段手机录屏，分别展示《三字经》和《登鹳雀楼》，时长约 **31.73 秒**和 **32.09 秒**。画面把立体卡通人物、古代生活场景、题名与作者、拼音诗句、简短释义组合在一起。



可以观察三个层次：上方保留作品名称和作者，中间用人物行为或风景表现内容，下方随朗读显示拼音、正文与释义。镜头会在家庭、课堂、楼阁和山水等场景之间切换。

案例标题写“0—6岁启蒙古诗”，后面的文案请求则明确指定 **2—6岁儿童及家长**。案例中的拼音和释义有助于理解呈现方式，但本课后续剪映截图只演示诗句字幕，没有展开拼音与释义字幕的制作过程。


## 二、让 DeepSeek 按诗句生成分镜提示词

第 4 页先给 DeepSeek 明确观看人群、诗名、画面结构和固定人物。完整请求如下：

```text
我是一位阅读古诗词视频的创作者，视频的观看人群是2-6岁儿童以及家长，请帮我生成古诗《相思》的分镜画面的AI绘画提示词，要求每句诗都对应一个分镜画面，需要包含：

主体描述、场景描述、风格描述、画面质量词，每个提示词最后都加上“中国古代卡通，明亮的色彩，3D皮克斯风格”。

要求：
1.提示词要简单易读，不要包含复杂的描述
2.不要有形容词和抒情的部分，且不要出现人名
3.每个画面都需要一个人物，始终是一个小女孩，穿着绿色唐装服饰，请描述人物的外貌（固定）、衣着（固定）、行为。
```

这段请求要求每句诗对应一张画面，每张都包含 **主体、场景、风格、画面质量词**。主体要说明人物外貌、衣着和行为，后续只改变诗句对应的动作与场景。

“小女孩、绿色唐装”是全片的人物基础。先确定一张满意的角色图，再将它作为其他分镜的角色参考，比只在每次生成时重复服装颜色更有依据。具体的参考方式在第五节展示。


## 三、《相思》的四组完整分镜提示词

第 5 页按诗句给出四个独立分镜。以下保留原截图中的主体、场景、风格与质量词，方便逐段复制到即梦。

### 分镜1：红豆生南国

```text
主体：女孩站在红豆树下，短发齐刘海，圆脸，穿绿色唐装，抬头看树枝上的红豆。
场景：阳光下的南方庭院，红豆树茂盛，地面有落花。
风格：中国古代卡通，明亮的色彩，3D皮克斯风格。
质量词：4K细节，柔和光影。
```

### 分镜2：春来发几枝

```text
主体：女孩踮脚轻触红豆树新芽，绿色唐装袖口微卷，表情好奇。
场景：春日溪边，背景有薄雾和初开的野花。
风格：中国古代卡通，明亮的色彩，3D皮克斯风格。
质量词：动态植物纹理，浅景深。
```

### 分镜3：愿君多采撷

```text
主体：女孩弯腰用竹篮接红豆，绿色唐装下摆沾泥土，手腕戴红绳。
场景：雨后山坡，红豆散落草丛，蝴蝶环绕。
风格：中国古代卡通，明亮的色彩，3D皮克斯风格。
质量词：水滴反光，细腻材质。
```

### 分镜4：此物最相思

```text
主体：女孩坐窗边写信，红豆堆在砚台旁，绿色唐装领口绣白梅。
场景：古风书房，烛光摇曳，窗外有圆月。
风格：中国古代卡通，明亮的色彩，3D皮克斯风格。
质量词：温暖色调，纸张褶皱质感。
```



原回复最后注明“每句独立分镜，人物外貌/衣着统一，无复杂描述”。实际文字中，只有第一组明确写了“短发齐刘海、圆脸”，后续又加入卷袖、红绳与领口绣白梅等细节。人物固定的要求仍需在生图时继续检查。

请求中的“不要有形容词”和固定风格词本身存在冲突，回复也保留了“柔和、细腻、温暖”等描述，并把质量词放在风格词之后。这里保留原课输出，不能把它视为完全满足全部约束的标准答案。


## 四、在即梦生成第一张分镜图

### 4.1 输入“红豆生南国”的画面描述

第 6 页进入即梦的 **“图片生成”**，复制第一组分镜描述。先生成女孩站在红豆树下、抬头看红豆的画面，为后续统一人物准备参考图。

| 设置项 | 左侧设置截图 |
| --- | --- |
| 生图模型 | 图片 2.0 Pro，`by Seedream 2.0` |
| 图片比例 | `9:16` |
| 图片尺寸 | `576 × 1024` |
| 提示词输入 | 第一组分镜，包含主体、场景、风格与质量词 |



输入框下方有 `Deepseek-R1` 按钮，它与下方的生图模型选择分开显示，不能把按钮名称记成图片生成模型。

### 4.2 从候选图中确定固定角色

结果区展示四张候选图，都是短发女孩穿绿色服饰站在树下，但人物高矮、服装长度、花纹和站位不完全相同。选择一张脸型、发型、衣着和构图合适的图片，再继续生成其他分镜。



同页结果记录的标签是 **“图片3.0、9:16、2K”**，与左侧“图片2.0 Pro、576 × 1024”不一致。这两张截图呈现不同的设置状态，不能把结果标签直接当成左侧设置的实际输出。

提示词中的“4K细节”也是画面描述词，与实际模型、尺寸或结果区的“2K”标签分别核对。


## 五、用“参考角色特征”生成下一张分镜

第 7 页继续生成“春来发几枝”。这次在提示词中加入前面选定的女孩图片，显示为 **“参考角色特征”**，以此延续角色外观。

1. 把选定的女孩图加入参考，核对输入框中的参考缩略图。
2. 输入第二组分镜：女孩踮脚轻触新芽，背景是春日溪边、薄雾与野花。
3. 保持竖屏 `9:16`，本页模型选择 **图片 2.0，`by Seedream 2.0`**，尺寸 `576 × 1024`。
4. 生成后对比脸型、刘海、发髻、袖长与衣服花纹，再保存合适的一张。



结果记录显示 **“参考角色特征、图片2.0、9:16、1K”**，下方同样给出四张候选图。



虽然使用了角色参考，候选图仍出现新增发髻、红豆头饰、袖长变化等差别。应将它们与第一张选定图并排核对，不能仅凭“都是绿色衣服”判断人物完全一致。

后两句可沿用同一角色参考，分别使用第三、第四组描述生成。课件没有展示后两组的完整生图结果，也没有展开参考选择弹窗或参考强度参数。


## 六、把分镜图片转换为短视频

### 6.1 上传分镜图并描述动作

第 8 页进入 **“视频生成” → “图片生视频”**，上传已经选好的树下女孩图片，文件名显示 `ImageInput.jpeg`。动作提示词为：

```text
女孩站在红豆树下，抬头看树枝上的红豆，身体往前略微走动
```

| 设置项 | 截图中的选择 |
| --- | --- |
| 参考输入 | 树下女孩的分镜图 |
| 视频模型 | 视频 S2.0，`by Seaweed alpha` |
| 生成时长 | `5s` |
| 视频比例 | 自动匹配 |



生图提示词已经确定人物和场景，视频阶段重点补充动作变化。结果记录标有“参考首帧”，播放器显示 `00:02 / 00:05`，截图中的女孩抬头并抬起手。



完整播放后检查脸部、手部、步伐与树枝是否稳定，再保存到剪映使用。课件在这里展示的是视频界面截图，没有另外内嵌这段生成视频。

### 6.2 继续制作其他分镜的动态

后面剪映时间线的第二段片段名还能读到：

```text
女孩踮脚轻触红豆树新芽，枝桠随风摆动
```

这条文字用于说明第二段画面的动作方向，来源是剪映片段名称。课件没有展开该段对应的生成设置，不能据此补齐未显示的参数。其余分镜按各自的图与动作继续制作，再统一检查人物连续性。


## 七、在剪映把诗句生成朗读音频

第 9 页使用电脑剪映生成诗词朗读，示例已切换为《望庐山瀑布》。原图中的正文是：

```text
日照香炉生紫烟，遥看瀑布挂前川。
飞流直下三千尺，疑是银河落九天。
```

### 7.1 新建文本并输入诗句

进入顶部 **“文本”**，在“新建文本”中点击 **“默认文本”**，把占位内容替换为诗句，先检查文字和标点。



### 7.2 选择文本朗读音色

选中文本，在右侧进入 **“朗读” → “文本朗读”**。截图展示女声和童声列表，生成后的音轨标签写着 **“文本朗读 乖巧萌妹”**。



选好音色后生成朗读，时间线会出现对应的音频波形。下面这条音轨已经带有“乖巧萌妹”名称。



先试听四句的读音、停顿和节奏，再按需要剪掉多余空白。面板下方还有“朗读跟随文本更新”选项，截图没有勾选。修改诗句后要重新确认音频内容与文本一致。


## 八、在即梦把古诗词生成音乐

第 10 页进入即梦 **“音乐生成”**。本页选中 **“人声歌曲”**，把《望庐山瀑布》作为歌词输入：

```text
日照香炉生紫烟，遥看瀑布挂前川。
飞流直下三千尺，疑是银河落九天。
```

| 设置项 | 截图中的内容 |
| --- | --- |
| 生成类型 | 人声歌曲 |
| 曲风 | 民谣 |
| 音色 | 女声 |
| 心情 | 快乐 |



输入区域还有“一键润色”按钮，截图没有展示它的使用结果。做古诗词作品时，生成前后都要核对歌词是否仍是要使用的原诗。

结果卡片名称是 **《庐山瀑布的欢歌》**，类型标注“人声歌曲”，播放器停在 `00:24`，总时长显示 **`01:27`**。它比单段 5 秒分镜长得多，后续需要选择和剪辑使用范围。



界面同时提供“纯音乐”选项，但本页生成的是带人声的歌曲。作为背景使用时，要试听歌声与朗读的衔接，避免两路人声互相遮盖。


## 九、按音频节奏剪辑画面和音乐

第 11 页的剪辑顺序是：**导入音频并剪辑、导入视频并调整长短、添加背景音乐**。

1. 将朗读音频放到时间线，先检查诗句顺序、句间停顿与总体时长。
2. 导入生成的分镜视频，按诗句顺序排列，让对应画面覆盖各自的朗读区间。
3. 调整片段起止位置和长短，使切镜与句子变化相配合，检查结尾是否仍有画面。
4. 导入音乐，选择需要的段落，调整起点、时长与音量，再完整试听。



本页可见四段《望庐山瀑布》诗句文本、两段红豆女孩视频，以及文件 **`庐山瀑布的欢歌.wav`**。因此它演示的是剪辑方法，画面与诗句还没有统一：红豆庭院画面对应前面的《相思》，不能直接作为瀑布诗句的完整配图。

截图也仍处在素材整理阶段：朗读后面接入歌曲，音乐没有与朗读同时铺底；两段视频只覆盖开头约 10 秒，草稿总时长显示 `00:01:34:19`。实际成片要按选定音频修整时长并补齐对应画面，避免后半段只剩声音。

课件没有展示这里的变速数值、音量数值或最终裁剪范围，不能把这张时间线直接当成已经完成的成片。


## 十、添加字幕，检查画面并导出

第 12 页列出最后三项操作：**添加字幕、去除水印、导出视频**。该页沿用第 11 页的同一张剪映截图。

### 10.1 让字幕对应当前诗句

时间线上已经把四句诗拆成四段文本，预览中的第一句是“日照香炉生紫烟”。字幕需要按每句朗读的开始和结束调整，确认字词、标点、显示位置与当前画面一致。



本例诗句位于画面下部，白色文字叠在浅色背景上。复看时要确认文字可读，且没有遮住人物或关键景物。开场案例包含拼音与释义，本页没有展示它们的添加步骤。

### 10.2 检查水印与完整输出

课件要求去除水印，但预览右下角仍可见“即梦AI”标记，左上角也有 AI 标识。截图没有展示去除入口或处理结果，因此这里只记录本课的检查项，不补写具体工具与操作。

导出前从头到尾播放一遍，检查同一首诗的画面、朗读、歌曲和字幕能否对应，人物是否稳定，有没有缺画面、长尾空白或声音重叠，再导出并复看保存的视频。

课件没有打开导出设置，也没有展示导出完成页，未指定导出分辨率、帧率、码率与字幕样式。前面生图选择了 `9:16`，本页预览也显示竖屏，应在自己的剪映草稿中核对最终构图。


## 十一、课程小结

第 13 页回顾四个阶段：**DeepSeek生成文案与绘画提示词、即梦生成分镜画面与视频、生成诗词朗读与音乐、剪映剪辑导出**。

先按诗句拆出画面，确定角色外貌和衣着，再用同一角色参考连续生成分镜。静态画面确定后，图生视频只补充当前镜头的动作。声音完成后，以实际朗读与歌曲的节奏安排画面和字幕，最后检查整条视频是否覆盖完整、内容是否一致。

本课需要特别核对两处：角色参考后的发型与衣服细节仍可能变化，文案、画面和音频示例也没有始终使用同一首诗。复做完整作品时把这两项检查贯穿各阶段。第 14 页没有新增操作内容。
