## 二、整体流程：文案、图片、对口型与剪辑

第 3、15 页将制作过程分为四步：

| 步骤 | 使用工具 | 操作与产出 |
| --- | --- | --- |
| 1. 生成怼人脚本 | DeepSeek | 参考已有句式，生成“人物 A＋林黛玉回怼”的成组文案 |
| 2. 生成角色配图 | 即梦 AI | 生成写实古装人物近景，选择图片，变高清并下载 |
| 3. 生成角色视频 | 即梦 AI | 分别设置人物图片、朗读文本和音色，生成两个角色的对口型视频 |
| 4. 剪辑成片 | 剪映 | 组合开场、人物对白、字幕、音乐与卡点效果，导出视频 |

原课流程图把第二、三步写为“DeepSeek＋即梦”，但具体截图直接展示即梦的生图和对口型设置，没有另外展示 DeepSeek 生成图片或视频提示词的过程。

学习时需要区分各环节的示例：生成文案展示了 10 组结果，对口型使用“买东西浪费钱”的对话，剪辑成片改用“女孩子读书有什么用”的对话。实际复做一条视频时，应先确定最终台词，再统一人物图片、声音和字幕。


## 三、用 DeepSeek 生成黛玉怼人文案

### 3.1 提示词的组织方式

本课直接给出一条包含示例的对话提示词。它先交代短视频博主的身份与能力，再放入 5 组“人物 A 说话、黛玉回应”的例子，最后要求仿写 10 组。

| 提示词部分 | 本课的具体要求 |
| --- | --- |
| 身份与任务 | 林黛玉怼人赛道的短视频博主，能输出犀利吐槽文案 |
| 风格示例 | 5 组双人对白，示范前一句与后一句的对应关系 |
| 语言要求 | 犀利、毒舌、有反转、使用大白话 |
| 数量与格式 | 新写 10 段，继续按“人物 A／林黛玉回怼”的格式分段列出 |

示例让 AI 看到要模仿的句式，数量和格式则让结果方便逐组挑选。后续每一组对白都能拆成两个角色各自的朗读文本。


### 3.2 原课完整文案提示词

在 DeepSeek 中输入以下原课指令。为便于对照，例句及措辞按课件保留：

~~~text
你是一个林黛玉怼人赛道的短视频博主，擅长做出林黛玉怼人的爆款短视频，并且能够输出林黛玉怼人的犀利吐槽文案，例如下面这5个：
1、人物A：你有病吧？！林黛玉回怼：对啊，还不是你传染的
2、人物A：你是在放屁吧！林黛玉回怼：哟！你闻到了，还合你胃口吗？
3、人物A：你画的妆像鬼一样，吓死我了。林黛玉回怼：你说的话像屎一样，熏死我了！
4、人物A：你们女生穿这么少，不就是为了勾引男人吗。林黛玉回怼：你长这么丑不就是为了报复社会吗！
5、人物A：女孩子读书有什么用，到头来还不是结婚生子。林黛玉回怼：你活那么久有什么用，到头来还不是埋土里！
现在请你根据以上怼人文案的特点，模仿她的文案风格和格式，帮我仿写新的10段怼人文案！要求犀利、毒舌、有反转、且大白话！
输出的10段黛玉怼人短视频文案，请按上述格式分段列出。
~~~

这条指令要求 AI 从已有例句中学习反击方式，再生成新的对话。它没有要求古文腔，也没有要求复述文学原著台词，生成目标是口语化的短视频对白。


### 3.3 原课生成的 10 组参考文案

第 5 页列出的 DeepSeek 生成结果如下，第 15 页小结中的结果截图也展示了这 10 组对白：

~~~text
明白！以下是口语化、直白的林黛玉式毒舌反击，保留犀利反转但降低用词门槛：
1. 人物A：你化妆技术太烂了吧！  林黛玉回怼：你眼珠子糊了浆糊？看不惯就自戳双目去！
2. 人物A：单身狗还挑三拣四？  林黛玉回怼：总比某些烂白菜被猪拱了还沾沾自喜强！
3. 人物A：你P图太过分了吧？  林黛玉回怼：哪儿比得上您真人长得像系统BUG呢？
4. 人物A：女生读研纯粹浪费钱！  林黛玉回怼：你妈生你时脑子忘给装了吧？建议回炉重造！
5. 人物A：穿吊带不就是想被看？  林黛玉回怼：你脑子里养蛆了？看什么都带颜色！
6. 人物A：三十岁没房真失败！  林黛玉回怼：您坟头草都两米高了还操心活人呢？
7. 人物A：整天追星脑残吧？  林黛玉回怼：总比某些人跪舔领导当狗强，至少我追的人真好看！
8. 人物A：你腿粗就别穿短裙！  林黛玉回怼：你嘴臭就别开口！自己照镜子了吗煤气罐成精？
9. 人物A：发际线高还露脑门？  林黛玉回怼：您这地中海配口臭倒是绝配，建议锁死！
10. 人物A：这么胖还吃宵夜？  林黛玉回怼：您又矮又丑还活着，我吃两口怎么了？
~~~

这些是课程提供的原始生成结果，主要用来学习“双人对白＋反转”的输出形式。复做时先从中确定一组，再拆分角色台词，不要将“人物 A：”“林黛玉回怼：”等说明文字一起交给语音朗读。

> 补充：原课部分句子包含粗俗用语和人身攻击。用于自己的作品时，可保留反转结构，改写为针对不合理观点的回应，减少对外貌、身体和家庭的攻击。


## 四、用即梦生成角色配图

### 4.1 黛玉配图提示词与参数

打开即梦：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)，进入“图片生成”。第 6 页输入框的提示词是：

~~~text
林黛玉，少女，写实风格特写，看向镜头，中式风格高清照
~~~

| 设置 | 原课可见值 |
| --- | --- |
| 生图模型 | `图片 2.1` |
| 精细度 | `5` |
| 图片比例 | `9:16` |
| 参考图 | 页面有“导入参考图”入口，截图未显示已导入参考图 |

人物名称用于指定古装角色，写实特写与看向镜头用于约束画面表现，竖屏比例与后续短视频相匹配。截图记录的是课件当时的界面，当前工具中的选项名称可能有所调整。




### 4.2 查看候选图，选定人物形象

第 6 页展示了四张候选图片，人物服装、发饰、脸型和姿态各有差异，整体都是室内古装女子近景。先确定适合账号的人物形象，再做高清处理。

选图时可以对照以下内容：人物是否清楚，脸部与嘴部是否可辨认，手部和饰品是否有明显异常，构图是否方便后续放置字幕。这些是复做时的检查建议。

后面对口型截图中的黛玉采用“坐在桌前、双手托在下巴附近”的另一张图，并非本页四张候选中的原样图片。原课没有展示这张图的完整生成过程，也没有提供男角色的生图提示词，不能仅凭后续人物画面补写成原课指令。




### 4.3 变高清并下载原图

第 7 页强调“AI 变高清、下载原图”：

1. 在候选图中找到选定的人物，调出图片工具栏。
2. 点击 `HD／超清`，进行高清处理。
3. 打开人物大图，检查面部、头发和服饰细节。
4. 点击大图右上方的下载按钮，保存人物图片。

课件左图显示 `HD` 上方的“超清”提示，右图用红色箭头标出下载按钮。两张截图中的人物外观有所变化，本节只据此记录高清与下载入口。

下载的图片是下一步“对口型”的角色素材。准备两个角色时，分别保留人物 A 与黛玉的图片，避免生成时替换错人。






## 五、用即梦分别生成两个角色的对口型视频

### 5.1 黛玉：导入图片、填入回应并设置音色

第 8 页展示的入口为 **“视频生成”下的“对口型”**。页面还提供“图片生视频”“文本生视频”，本课实际选择的是对口型。

按原课可见设置整理：

1. 在角色区域放入准备好的黛玉图片；需要更换时使用“替换”。
2. 在“生成效果”中选择 **标准**，截图中该选项有白色边框。
3. 在对口型区域选择 **文本朗读**。
4. 填入黛玉自己的回应句。
5. 选择音色并试听，确认后提交生成。

| 项目 | 黛玉示例 |
| --- | --- |
| 角色图片 | 古装女子坐在桌前，双手托在下巴附近 |
| 生成效果 | 标准 |
| 输入方式 | 文本朗读 |
| 朗读台词 | `你说这话真浪费唾沫！` |
| 音色 | `黛玉` |
| 页面试听显示 | `0:02` |

页面也有“上传本地配音”，本课选择文本朗读，没有演示本地配音上传流程。这里由即梦完成台词朗读和人物对口型，不需要把两个人的对白一起放进同一个输入框。




### 5.2 提交生成，检查人物说话效果

第 8 页右侧截图显示“参考角色——对口型”的生成任务，进度为 `0%`，并提示“生成排队中，预计等待 27 秒”。这是当时任务的排队状态，实际等待时间随工具情况变化。

生成后播放结果，检查台词是否读全、嘴型与声音是否对应，人物脸部是否稳定。若台词或声音不满意，先修改对应设置，再重新生成。

原课第 10 页随后展示男女角色各自的短视频，尚未合并剪辑。课件内嵌的两段素材分别约 **2.05 秒**和 **2.30 秒**，用于说明对口型片段已经生成。




### 5.3 路人：替换角色与台词，使用另一种音色

第 9 页用相同方式生成路人，也就是“人物 A”的说话片段：

1. 在角色区替换为古装男子图片。
2. 继续使用 **标准** 效果与 **文本朗读**。
3. 填入人物 A 的台词：`你买这东西真浪费钱！`
4. 选择 **阳光青年** 音色，试听后生成。
5. 播放结果，准备与黛玉片段一起导入剪映。

这组对白的对应关系是：

~~~text
人物A：你买这东西真浪费钱！
林黛玉回怼：你说这话真浪费唾沫！
~~~

用不同角色图片和不同音色，观众才能分清谁先说话、谁在回应。每个角色只朗读自己的部分，剪辑时再按对白顺序拼接。

第 9 页结果播放器显示 `00:02／00:02`，它表示这段角色素材的播放状态，不代表后续整条视频只有 2 秒。






## 六、用剪映剪辑完整视频

### 6.1 导入素材，安排开场与双人对白

第 11 页先回看案例，第 12 页再进入剪映，写明“导入视频，添加首句文案，添加字幕特效，画面特效”。预览使用 `9:16` 竖屏。

从时间线可以整理出成片顺序：

1. **开场**：黑底文字“被骂时如何霸气回怼”，旁边搭配爱心效果。时间线中可见相应文字和文本朗读音轨。
2. **人物 A 说话**：放入男角色视频，出现对应字幕。
3. **黛玉回应**：紧接女角色视频，字幕切换成回应内容。
4. **结尾卡点**：在对白后接入短切片段与画面效果，配合音乐节奏。

本页字幕为黄色文字配深色描边，位于人物画面的中下部；右侧打开字幕动画面板，选中 **“爱心弹跳”**，动画时长显示 **2.6 秒**。这些是原课示例状态，课件没有逐项规定字号、位置数值或统一动画时长。

剪辑截图已经改用“女孩子读书有什么用”的对白，角色图片也与第 8、9 页不同。原课未展示这组新台词重新生成对口型的过程，复做时需要准备与最终对白相符的素材。




### 6.2 添加卡点特效，让结尾与音乐对应

第 13 页继续要求“添加卡点特效，匹配音频卡点”。红框中的时间线可以看到：

| 时间线元素 | 原课画面中的用途 |
| --- | --- |
| 多个连续短片段 | 在对白结束后形成快速切换的节奏 |
| `电光眼` 特效 | 在人物眼部叠加明亮的光线效果 |
| `缩放运镜` 特效 | 给结尾画面增加缩放变化 |
| 下方音轨 | 与上方画面切换及特效段对应 |

剪辑时先试听音乐，确定要强化的节奏位置，再调整短片段的切分和特效起止，让结尾视觉变化与声音落在一起。

原课展示了分切后的时间线与效果，没有给出固定切点、音乐名称、逐帧参数或自动卡点的完整菜单路径。截图里较长的工程总时长也不能直接作为本条视频的时长，最终成品另见下一节。




### 6.3 核对字幕，检查画面并导出

原课第 13 页最后写“去除水印，添加字幕导出即可”，但没有展示去水印的具体操作，也没有给出导出分辨率、码率与帧率设置。

对照第 12、13 页，可以把收尾工作理解为：整理最终画面，检查字幕与配音，确认音乐和特效，再导出完整视频。课件中的预览已能看到对白字幕，仍需核对是否漏字、断句是否合适、字幕是否挡住人物。

> 补充检查：从头播放一遍，确认开场可读、两个角色顺序正确、字幕与台词一致，结尾卡点衔接自然，背景音乐没有盖过说话声。涉及工具水印时，优先使用平台提供的合规导出方式，保留要求展示的 AI 标识。导出参数按素材与发布平台要求选择。

## 八、如何起号：找同类内容，持续制作与优化

第 16 页给出三项方向：

1. **挖掘同类账号，借鉴爆款**：观察同类内容使用什么对话主题、人物形象、开场和剪辑节奏。
2. **吐槽时事热点，融入黛玉文案**：将选题与当下讨论的话题结合，继续使用双人对白和反转形式。
3. **循环拍摄、剪辑、发布，优化内容**：持续做内容，再根据实际表现调整选题和制作。

这页属于运营方向提示，没有展开账号注册、发布按钮、更新频率、数据判断标准或具体变现流程。“快速起号”是该页标题中的目标表述，课件未提供可据此推算效果的数据。

> 补充：拆解同类账号时，可以记录开头如何引出话题、对白如何反转、字幕是否易读、结尾是否拖沓，再用于修改自己的脚本与节奏。涉及热点时先核对事实，避免把未经确认的信息写进剧情。


## 九、课程小结

第 15 页将本课方法归纳为四步：

1. 使用 **DeepSeek** 生成黛玉怼人文案。
2. 使用 **即梦 AI** 生成角色配图，选择并下载高清图片。
3. 使用 **即梦 AI 对口型**，分别生成两个角色的说话视频。
4. 使用 **剪映** 完成开场、对白、字幕、音轨和卡点效果，剪辑成片。

本课可复用的是“先确定一组对白，再分别制作两个角色，最后组合成完整短视频”的方法。更换选题时，人物图可以沿用已确认的形象，台词、音频和字幕则要同步更新；原课的不同示例应分开对照。
