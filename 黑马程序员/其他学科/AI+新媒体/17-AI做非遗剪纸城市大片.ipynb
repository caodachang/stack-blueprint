# 17 AI做非遗剪纸城市大片


## 课程目标与制作流程

本篇对应课件 **《17-AI做非遗剪纸城市大片》**，共 16 页。课程把城市地标照片转换成红色立体剪纸风格，再生成向前推进的镜头，让实景逐渐变为剪纸模型，最后在剪映中配乐、卡点并剪成城市宣传短片。

| 阶段 | 使用工具 | 这一阶段要完成什么 |
| --- | --- | --- |
| 准备城市照片 | Pexels | 下载清晰的地标或城市风景图片 |
| 生成剪纸图 | 堆友 D.DESIGN | 用剪纸效果与原图参考保留建筑轮廓 |
| 整理图片 | WatermarkRemover.io | 课件展示下载后的图片去水印处理 |
| 图生视频 | Runway | 把图片生成动态镜头，再用首尾帧控制过渡 |
| 编写运镜提示词 | ChatGPT 或扣子中的“运镜大师” | 根据两张图说明起始画面、镜头运动和最终效果 |
| 剪辑成片 | 电脑剪映 | 导入片段、提取音频、变速卡点、添加转场与字幕并导出 |

以下按照课件顺序整理，提示词和参数来自正文及截图，案例效果另对照 PPTX 中的两段内嵌视频。第 15 页是流程回顾，第 16 页没有新增教学内容。


## 一、案例中的剪纸城市效果

第 2 页先展示一段约 **11.8 秒、1200 × 720、30 帧/秒**的横屏视频。桥梁、街道、楼阁、树木与人物都呈现红白纸雕的轮廓和镂空层次，近景细节清楚，远景逐渐虚化，形成微缩城市的空间感。



本课后面的操作使用天坛和武康大楼演示“真实建筑变成剪纸建筑”。开场的多地标城市视频用于观察纸雕质感与镜头效果，课件没有逐一提供其中所有地标的生图提示词与工程文件。


## 二、准备高清城市地标照片

第 4 页给出的素材入口是 [Pexels 中文站](https://www.pexels.com/zh-cn/)。在图片搜索中查找城市名称或地标名称，选择并下载高清照片，后续作为 AI 生图的参考。

课件以天坛照片为例。建筑位于画面中央，三层屋顶、金色宝顶、台阶和栏杆都能看清，这些轮廓有助于在风格变化后仍辨认出原来的地标。



准备时先确认主体完整、结构清楚，再进入风格生成。后面 Runway 的实景首帧使用另一张竖版天坛照片，构图与本页这张横版照片不同，不能把两者当成同一张原图的直接裁切结果。


## 三、在堆友把地标转换为剪纸风格

### 3.1 查找同款效果并上传参考图

打开 [堆友 D.DESIGN](https://d.design/)。第 5 页展示“万物皆可非遗纸雕”的作品卡片，使用同款效果进入生成界面。操作截图位于 **“AI反应堆”**，左侧是“经典生图”。



在“图片参考”中上传天坛实景图。“图片玩法”选择 **参考边缘**，参考程度设为 `0.8`，用建筑边缘约束生成图的基本形状。

### 3.2 核对风格权重与完整提示词

| 设置位置 | 截图中的选项与数值 |
| --- | --- |
| 增益效果一 | 中国风剪纸，参考程度 `0.7` |
| 效果一界面建议 | `0.6–0.8` |
| 增益效果二 | WB红色主题简洁背景，参考程度 `0.4` |
| 效果二界面建议 | `0.7–1` |
| 图片参考 | 天坛实景照片 |
| 图片玩法 | 参考边缘 |
| 图片参考程度 | `0.8` |

第二项的**实际值是 `0.4`**，与卡片上的建议范围不同。复做本课示例时，应区分实际选择与工具建议。

画面描述完整可见文字为：

```text
中国风纸雕，剪纸风格，跃然纸上，光影，春节，非遗，细节，4K，红色主题，特写，景深（协调人景），
```



这些词共同约束纸雕材质、红色主题、光影与景深。“4K”是提示词中的画面要求，课件没有展示对应的实际导出分辨率。“负面描述”和“高级参数”处于收起状态，不能据此补写未显示的参数。

### 3.3 生成并检查地标结构

点击“立即生成”，本页按钮显示消耗 **10堆豆**。结果区有两张候选图，主体变为红色纸雕，背景变为暖色虚化场景。



挑选结果时，检查宝顶、三层屋檐、门窗、栏杆与台阶能否对应原建筑。图中的匾额文字已经发生变化，不能把生成的小字当作正确的地标文字使用。

课件总览写“AI免费生成”，本页界面同时出现消耗提示，以及“该内容包含会员模型，开通会员，尊享无水印、商用等特权”。额度、下载与商用条件需要按实际平台界面确认。


## 四、下载图片并整理水印

第 6 页在下载后演示 [WatermarkRemover.io 上传页](https://www.watermarkremover.io/zh/upload)。页面支持上传、拖入图片或粘贴图片及 URL，截图标注的最高分辨率是 `5000 × 5000 px`，支持 `png`、`jpeg`、`jpg`、`webp`、`heic`。



按课件操作，上传下载的剪纸图，处理后保存结果。下面左图是带有重复 D.DESIGN 标记的图片，右图是本页提供的处理后图片。



保存后放大查看屋檐、镂空纹样和匾额附近，确认细线条没有被抹掉，再把选定图片用于 Runway。图片处理不会改变前一页生成平台提示的会员与商用授权条件。


## 五、先把剪纸图片生成动态镜头

### 5.1 进入 Runway 视频生成

打开 [Runway](https://runwayml.com/)，进入 **“生成视频”**。第 7 页用红框标出首页新会话中的入口与左侧工具入口，进入后切换到 `Video`，上传准备好的图片。



第 7 页结果区已存在实景天坛的历史记录，第 8 页才展示红色剪纸图的上传状态。这里先学习单张图生成动态，后面再设置实景与剪纸两张图之间的过渡。

### 5.2 上传剪纸图，描述镜头推进

第 8 页正文给出的中文提示词为：

```text
全景摄影画面，向前推近运镜，自然光线下，由宏伟的楼体建筑逐渐向前展现细节。
```

| 参数 | 第 8 页输入截图 |
| --- | --- |
| 输入 | 红色纸雕天坛图 |
| 模型菜单 | `Gen-3 Turbo` |
| 时长 | `5s` |
| 画面尺寸 | `768 × 1280`，竖屏 |



截图中的输入框仍是占位提示，中文运镜词来自本页正文。将文字输入后再提交生成，目标是让镜头逐渐靠近建筑，展开屋檐和镂空细节。

右侧结果记录标为 `Gen-3 Alpha Turbo`，播放器停在约 `0:02 / 0:05`，画面已推进到剪纸天坛的近景。



生成后完整播放，观察推进是否平稳、屋檐和门窗是否扭曲、画面边缘是否出现不需要的裁切。输入截图底部显示 `Upgrade (out of credits)`，表示当时账户额度已用完，不能据本页认定该账户还能继续免费生成。


## 六、用 ChatGPT 为天坛两张图编写过渡运镜

### 6.1 上传实景图与剪纸图

单张图片提示词主要描述镜头运动。要让真实天坛逐渐变为剪纸模型，还需要交代**从什么画面开始、沿什么方向运动、最后变成什么**。

第 9 页打开 [ChatGPT](https://chatgpt.com/)，使用已存在的“运镜大师”GPT。截图侧栏有两个同名入口，课件没有提供它们的具体链接或创建配置。



上传实景天坛与红色剪纸天坛两张图，输入原课请求：

```text
画面放大，过渡运镜，由实景拍摄的天坛公园，自然运镜过渡到现在3D剪纸效果
```



### 6.2 把变化过程写清楚

截图中的回复按“转场镜头设计理由、详细转场镜头描述、英文生成提示词”展开。设计理由是通过逐渐拉近、旋转与过渡，把真实建筑的细节变化连接到纸雕模型，呈现由现实到艺术作品的层次变化。

镜头描述包含四步：

1. 从蓝天白云下的天坛全景开始，缓慢推进，展示建筑与屋顶。
2. 靠近时展开更多细节，可轻微倾斜或转动镜头，延伸空间感。
3. 建筑逐渐变得精致、微缩、色彩鲜明，过渡为 3D 剪纸效果。
4. 最后停留在剪纸建筑模型上，光线变柔和，突出纸雕细节与层次。

这里的“可轻微倾斜或转动”来自中文镜头描述，下面英文提示词着重写缓慢推进，没有明确要求旋转，两段表述不能完全等同。

### 6.3 完整英文视频提示词

```text
A wide shot of the Temple of Heaven in Beijing, with the camera slowly pushing in, capturing the grand architecture under a bright blue sky. As the camera gets closer, the detailed roof and stairs are highlighted, and the background transitions from the expansive sky to a more intimate, creative atmosphere. The scene gradually transforms from the real, monumental structure to a delicate, vibrant 3D paper-cut model of the Temple of Heaven, with intricate details and textures. The lighting becomes softer, emphasizing the artistry and craftsmanship of the paper-cut model.
```

这段提示词同时包含地点、全景起点、推进动作、屋顶和台阶细节、实景到纸雕的材质变化，以及变柔和的光线。下一步将它与 Runway 的两张参考图一起使用。


## 七、用扣子为武康大楼编写同类提示词

### 7.1 在商店中找到运镜工具

第 10 页提供 [扣子](https://www.coze.cn/home) 入口，截图进入的是左侧 **“商店”**，浏览器地址显示 `coze.cn/store/agent?cate_type=recommend`。本页使用已发布的“运镜大师”智能体，课件没有展开智能体的创建过程。



打开该智能体，上传武康大楼的实景图与红色剪纸图，输入：

```text
上海武康路武康大楼的全景镜头，请帮我写一组，镜头缓缓推进，场景逐渐从真实的、宏伟的建筑丝滑的转变为武康大楼精致的、充满活力的3D剪纸模型
```



### 7.2 镜头设计与英文提示词

回复的设计理由是让镜头缓慢推进，给观众足够时间感受真实建筑到 3D 剪纸模型的变化。

中文镜头描述从繁华城市里的武康大楼全景开始，包含蓝天、行人、车辆和照在建筑上的阳光。镜头以舒缓、均匀的速度靠近，建筑细节逐渐清楚，真实材质慢慢变成明亮的剪纸轮廓与线条，最终呈现完整的 3D 剪纸模型。

截图中的完整英文提示词是：

```text
Slowly advancing camera shot of Shanghai Wukang Road's Wukang Building, starting with a wide panoramic view of the real and grand building in the bustling city with pedestrians and vehicles, and the sun shining on it. Then, as the camera moves closer, the building smoothly transforms into a delicate and vibrant 3D paper-cut model.
```

这一示例延续天坛案例的写法，把地标、城市环境和最终模型换成武康大楼。回复开头写“两张关键帧虽未直接呈现”，但输入区确实有两张图，因此不能只凭这段回复判断工具已正确识别图片，仍需核对生成的建筑描述与目标是否一致。


## 八、把运镜提示词与 Runway 首尾帧一起使用

第 11 页回到 Runway，在参考图位置放入两张图片：**实景天坛作为起始画面，红色剪纸天坛作为结束目标**，再将第六节的天坛英文提示词粘贴到文本框。

1. 按先实景、后剪纸的顺序放入两张图，核对缩略图。
2. 复制完整英文提示词，保留“慢慢推进”和“逐渐变为 3D 剪纸模型”等变化关系。
3. 检查画面尺寸、裁切与生成时长，再提交任务。
4. 播放完整结果，确认建筑形态变化与镜头推进能接起来，再保存视频。



该截图模型菜单仍是 `Gen-3 Turbo`，时长 `5s`，尺寸却显示 **`1280 × 768` 横屏**，预览中剪纸图上下有黑色区域。它与第 8 页的 `768 × 1280` 竖屏设置不同。正式生成前要按成片方向检查构图，避免把截图中的横竖差异一并带入最终视频。


## 九、检查过渡效果，并连续衔接多个镜头

### 9.1 对照课件内嵌的过渡视频

第 12 页内嵌一段约 **5.21 秒、768 × 1280、24 帧/秒**的竖屏视频。开头为实景天坛，镜头靠近后逐渐转为红色纸雕，最后继续推进到建筑近处。

下面依次截取 **0 秒、约 1.5 秒、约 4.9 秒**：中间帧还能看到部分白色台阶、栏杆与正在转变的红色建筑，末帧已经形成纸雕近景。



这段视频可用于观察过渡的方向，但不能只看开头和结尾就判断全程顺滑。复看时应留意屋檐、门窗与台阶在中间是否错位，背景与光线是否突变。该内嵌视频是竖屏，不能据此反推第 11 页横屏任务的设置。

### 9.2 用共享图片连接相邻片段

课件小结是：**运镜提示词＋Runway 首尾帧，可以实现视频过渡**。要延长成片，就让前一段的尾帧同时作为下一段的首帧。

课件给出的组合式为：

```text
AB + BC + CD + DE + EF = 成品画面
```

其中 `AB` 表示以图片 A 为首帧、图片 B 为尾帧生成一段视频，`BC` 再从 B 过渡到 C，后续依次重复。共享的 B、C、D、E 是相邻片段衔接的依据。

> 原课同时写“举例用5张图片”，但上式 A—F 实际是 **6张图片、5段视频**。如果只有 A—E 共5张图，应生成 `AB + BC + CD + DE`，共4段。

每一段都要根据本段的起止画面调整提示词，并保持相邻片段的衔接图一致。生成后仍需检查接缝处的构图、推进方向与速度，再进入剪映整理。


## 十、在剪映导入片段、提取音频并变速卡点

第 13 页切换到电脑剪映，按“导入视频素材、下载提取音频、变速卡点”的顺序剪辑。

### 10.1 先排列生成的视频

1. 新建草稿，导入 Runway 生成的实景、剪纸与过渡片段。
2. 把视频拖到主轨道，按设计好的镜头顺序排列。
3. 检查每个片段的画面比例、缩放和位置，让地标在竖屏成片中保持清楚。

素材区可见天坛、实景武康大楼、剪纸武康大楼及开场城市案例。时间线上先排列多段生成视频，后面还接入一段横屏城市案例素材。

### 10.2 提取音频并对齐节奏

课件写“下载提取音频”，截图下方有一条从案例视频分离出的音频波形。把需要使用的音频素材导入后，通过提取或分离音频放到音频轨，让画面切换跟随音乐节奏。

根据节奏点调整片段时长，必要时用“变速”让镜头进入和结束落在合适的位置。截图的音频轨上有多个菱形标记，可结合波形检查卡点位置；课件没有展开变速面板，也没有给出固定倍速或曲线参数。



本页选中的横屏城市素材放在竖屏画布中，上下留有明显黑边。进入下一步前，应按目标构图处理横竖素材的适配，并完整播放检查节奏衔接。


## 十一、添加转场、字幕并导出成片

第 14 页继续列出三项：**添加转场、添加字幕、导出视频**。

1. 在需要连接的相邻片段之间添加转场，结合生成视频本身的变化检查接缝，避免转场破坏已经生成的推进过程。
2. 根据城市和地标补充字幕，检查文字内容、出现时间和位置，避免遮住建筑细节。
3. 完整播放画面与音频，检查卡点、字幕、黑边及结尾，再导出视频并复看导出文件。



本页截图中相邻视频之间能看到转场标记，预览是剪纸天坛，右下角仍有 `runway` 标记。文字要求包含“添加字幕”和“导出视频”，但截图没有展开字幕编辑或导出面板，也没有显示完成导出的结果页，因此不能补写具体字幕样式、转场名称、时长或导出码率。

这张截图的草稿参数显示 `30.00帧/秒`，属于当前草稿信息，课件没有指定它为统一导出参数。最终保存时按实际项目设置检查画面与声音是否正常。


## 十二、课程小结

第 15 页回顾四个阶段：**生成剪纸图片、图片生成视频、用提示词与首尾帧控制过渡、在剪映剪辑成片**。

复做时先准备清晰地标照片，在堆友完成剪纸风格与结构检查，再到 Runway 生成镜头。需要实景到纸雕的变化时，用两张图说明起点与终点，并加入完整运镜提示词。多个片段通过共享首尾图衔接，最后在剪映中按音乐调整时长、完善转场与字幕并导出。

核心检查贯穿每一步：建筑轮廓能否辨认，纸雕细节是否完整，过渡中间是否扭曲，相邻镜头方向和节奏是否接得上。第 16 页没有新增操作内容。
