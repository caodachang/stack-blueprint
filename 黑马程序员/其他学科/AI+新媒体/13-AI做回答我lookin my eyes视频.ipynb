# 13 AI做回答我 lookin my eyes 视频


## 资料定位与课程目标

本篇对应课件 **《13-AI做回答我lookin my eyes视频》**，共 15 页。课程先展示“回答我”养生短视频，再依次讲解文案、音色克隆、人物与背景生图、动作形象替换，最后用剪映合成。第 14 页是小结，第 15 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇以本地 PPTX 的正文、操作截图及 2 段内嵌案例录屏为准。

这节课要学会的是：用新文案和新人物复做原视频的追问节奏与动作，再把人物放进新的养生场景。文案、配音、人物图、动作视频是不同的中间素材，需要在最后的剪辑中对齐。


## 一、案例中的“回答我”视频长什么样

第 2 页展示两段手机竖屏录屏。它们都用古人形象面向观众说话，配合前倾、转头等动作与追问式台词。两段的画风、字幕和选题各有区别：

| 案例 | 录屏可见内容 | 可以观察的表达方法 |
| --- | --- | --- |
| 熬夜主题，账号显示“仲景有话说” | 蓝袍、发髻古人，黄色字幕配红色重点字，顶部有弹幕 | 用生活习惯切入，靠反问、夸张比喻和人物动作强化语气 |
| 早餐主题，账号显示“人民日报” | 灰须古人与药房，顶部标题“今天被‘扁鹊’教育了！‘早上一定要吃早餐’”，下方为白色字幕 | 标题先交代选题，人物面对镜头追问，字幕跟随台词变化 |

案例一的录屏约 49.69 秒，案例二约 32.95 秒。这是课件保存的录屏文件长度，不能据此认定后面要求的“30 秒以内”文案已经达标。案例二录屏可见约 106.4 万点赞，是当时的画面数据；课件标题中的“百万点赞”也只是案例描述。

下面裁取标题或人物讲解区域，去掉大部分手机导航与评论区，保留原画面中的字幕和互动覆盖层。





课件参考台词还反复使用“回答我！说话！”以及“lookin my eyes，tell me why，why baby why？”等口头禅。复做时要把这些停顿和强调与动作、字幕衔接起来，不能只把整段文字一次性铺满画面。


## 二、五步制作流程与素材关系

第 3 页把制作分成五步，操作页展示的实际工具如下：

| 阶段 | 工具 | 具体任务 | 下一步需要的素材 |
| --- | --- | --- | --- |
| 1. 文案 | DeepSeek | 参考样例，生成 3 个养生文案 | 选定并校对的旁白文本 |
| 2. 音频 | 剪映、Noiz | 提取原声，克隆音色，再读出新文案 | 纯人声样本、新旁白及需要保留的原声片段 |
| 3. 人物＋场景 | 即梦 AI | 分别生成李时珍人物图和中药房背景图 | 人物图片、无人物背景图片 |
| 4. 替换形象 | Viggle | 用原视频提供动作，用人物图提供新形象 | 替换人物后的动作视频 |
| 5. 成片 | 剪映 | 组合声音与画面，抠像、处理字幕，选择导出范围 | 完整视频 |

总览页的个别工具标签与后面的操作页不一致：音色克隆实际使用 **Noiz**，动作形象替换实际使用 **Viggle**。课件也没有给出各工具的现行收费与额度，复做时以实际界面为准。

这里有两条素材线：文案通过 Noiz 变成声音；原视频与人物图通过 Viggle 变成人物动作视频。两条线最后在剪映中汇合，药房图作为背景加入。


## 三、用 DeepSeek 生成养生文案

### 3.1 完整提示词

第 4 页先给出一个早餐主题的参考文案，再要求生成 3 个新文案。下面保留原提示词：

```text
参考以下健康养生文案，再帮我写3个新的健康养生文案，适合目标读者是：喜欢关注养生的人群。

要求：
1、文案风格类似，有压迫感，能够牢牢吸引住用户，并且言简意赅，文案时长不超过30秒。
2、请严格参考以下文案的内容结构、语言风格、字数。
你早上不吃饭，那我问你，《内经》讲 "胃不和则卧不安"，你胃气都亏了还想睡个好觉？回答我！说话！
你以为饿一顿没事，阳气晨起如朝阳，你不给他柴火怎么燃起一身正气？回答我！嗯！
你回答我！说话！ lookin my eyes，tell me why，why baby why？
你是不是要让脾胃气血亏虚，寒湿内生？以后吃啥都消化不了，舌苔白的能当墙粉刷？回答我！说话！
老张天天早餐喝粥，现在健步如飞，你胃痛乏力，面色萎黄，你这是要提前见阎王？
```

提示词同时控制了目标人群、语气、参考结构和时长。“30 秒以内”是生成要求，后续还要以实际配音长度检查；停顿、追问和英文口头禅都会占用时间。


### 3.2 三条生成结果

第 5 页分别围绕熬夜、冷饮和久坐生成文案。以下是 **课件中的 AI 原始结果**，用于观察写法；医籍归属、健康推断与“小李”“王姐”“老刘”等对比故事没有在课件中给出依据，不能当作已经核实的健康知识或真实案例。实际使用前应核对引用，并修改失明、疾病等夸张推断。

**① 熬夜刷手机**

```text
你天天熬夜刷手机，《内经》说"肝藏血"，子时不睡，肝血拿什么养？眼睛模糊发黄还要强撑？回答我！说话！隔壁小李十点闭眼，现在双目如炬，你黑眼圈比墨还浓，是想当国宝还是想瞎？嗯？你回答我！说话！ lookin my eyes，tell me why，why baby why？
```

**② 喝冰奶茶**

```text
冰奶茶灌得爽是吧？《金匮要略》讲"脾喜燥恶湿"，你拿寒邪当饭吃？腹痛腹泻还怪天气？回答我！why！王姐三年不碰冷饮，小腹暖得像火炉，你舌苔厚得能种菜还在吨吨灌冰水？
```

**③ 久坐不动**

```text
一坐八小时不动弹？《千金方》早说过"久坐伤肉"，你当自己钢筋铁骨？腰酸背痛才想起来扭两下？回答我！说话！老刘每天站桩半小时，六十岁健如虎，你三十岁就腰椎间盘突出是想坐轮椅参加同学会？
```

三条文案复用了“点名习惯 → 引用或解释 → 追问 → 对比人物”的组织方式，句子短、反问密集，但没有每条都使用相同的英文结尾。可以借鉴这种节奏，内容仍要单独核查。

第 14 页的小结还放了一张 DeepSeek 界面截图，熬夜文案在该图中没有第 5 页末尾的英文段落，因此以本节上面的第 5 页正文作为完整转录。




## 四、提取原声并克隆音色

### 4.1 在剪映中提取音频

第 6 页先把原参考视频放入剪映。截图时间线上可以看到原视频，以及单独的“剩余声音”“整体背景声”轨道，说明声音已经被分离处理。后续克隆需要的是清晰的人声样本，应尽量减少背景音乐和杂音干扰。



导出截图勾选“音频导出”，格式列表中选中 **WAV**。这里记录的是音频素材导出，不是最终视频导出。



截图里的整段音频约 1 分 10 秒、估算大小约 12 MB，而下一步 Noiz 的样本提示是 **3—10 秒、10M 以内**。因此不能把这份整段导出文件直接当作符合要求的克隆样本，应先截取清楚、连续的短人声片段。


### 4.2 在 Noiz 添加克隆声音

第 7 页给出的地址是 [Noiz](https://noiz.ai/landing)。截图中的操作顺序为：

1. 进入“语音库”，点击“添加声音”。
2. 填写名称，可添加声音标签；课件示例后续命名为“回答我声音克隆”。
3. 上传短人声样本。界面写明“3—10 秒纯人声”，文件大小“10M 以内”，并提示过短或过长会影响质量。
4. 检查自动降噪选项，阅读声音权利声明，确认拥有必要权利后点击“确认”。

课件弹窗中自动降噪处于开启状态。声音克隆应使用本人或获准使用的声音，弹窗中的权利勾选也对应这一要求。



创建后，语音库截图出现“回答我声音克隆”“回答我2”“回答我”等自定义条目；下一步在文本转语音中选择相应音色。


### 4.3 用新音色朗读新文案

第 8 页先在“你的音色”中选中 **“回答我声音克隆”**，再把新文案放到文本转语音页面，点击“生成语音”。界面还显示“目标语言：自动识别”“插入停顿”和“标记漏字”等选项。





这张截图只输入了熬夜文案的前一部分，到“嗯？”结束，计数显示 `86/500`，没有输入后面的英文口头禅。后续剪映时间线也同时用了新旁白和原视频的“剩余声音”片段，说明示例音轨包含剪接，不能认定整段台词都由同一次克隆生成。

生成后先试听，检查漏字、停顿与语气，再把新旁白作为独立音频导入剪映。课件没有展示 Noiz 的具体下载或导出面板。


## 五、用即梦生成人物与中药房背景

### 5.1 人物提示词与生成结果

第 9 页给出人物提示词：

```text
生成古人李时珍，正脸像，上半身，古代水墨画人物画风，人物表情端庄，工笔重彩技法，细腻的线条勾勒，衣纹流畅自然。
```

| 提示词要素 | 课件写法 | 作用 |
| --- | --- | --- |
| 人物身份 | 古人李时珍 | 确定角色形象 |
| 构图 | 正脸像、上半身 | 让脸部和上身成为主体，便于后续替换动作 |
| 神态 | 表情端庄 | 约束初始人物表情 |
| 画风与细节 | 古代水墨画、工笔重彩、细腻线条、自然衣纹 | 统一服饰、笔触和画面风格 |

截图结果是古人半身像，人物略向一侧，背后还生成了山水与类似书法的文字。提示词写“正脸像”并不意味着每次都能得到严格正面的结果，仍需挑选脸部清楚、构图合适的一张。




### 5.2 课件里的生图参数

人物与背景截图使用的设置相同：

- **生图模型**：图片 2.0 Pro，界面标注 `by Seedream 2.0`。
- **图片比例**：`3:4`。
- **图片尺寸**：`768 × 1024`。

输入框附近还出现 `Deepseek-R1` 按钮，但生图模型栏明确选择的是“图片 2.0 Pro”，不要把这两处名称混为一谈。



以上是课件当时的示例设置。后面的剪映草稿是 `16:9`，因此竖幅人物图和背景图还要在剪辑时调整缩放与位置，检查横屏裁切后的头部、肩部和背景是否合适。


### 5.3 单独生成无人物的背景

第 10 页正文的完整场景提示词为：

```text
古代水墨画画风，工笔重彩技法，细腻的线条勾勒，中药房间画面，画面无需人物。
```

背景与人物分开生成，可以在剪映中把动作人物叠到药房前面。“画面无需人物”是这个合成关系里的关键约束。



这里需要对照正文与截图：**正文写了“画面无需人物”，但截图输入框和生成记录里的提示词没有这句话**。四张候选图中有几张带人物，而选中放大的第二张是无人药柜场景，所以不能仅凭结果说模型忽略了完整提示词。



复做时补齐无人要求，并挑选确实没有多余人物的背景，保持与前景人物相近的水墨、工笔画风。


## 六、用 Viggle 替换动作人物

### 6.1 原视频提供动作，图片提供新角色

第 11 页给出的地址是 [Viggle Create Mix](https://www.viggle.ai/create-mix)。上传区分为两边：

| 上传位置 | 输入素材 | 在这一步的用途 |
| --- | --- | --- |
| 左侧“运动” | 原参考视频 | 提供说话、转头、身体前倾等动作 |
| 右侧“角色” | 即梦生成的人物图 | 提供新的古人形象 |

上传后点击“生成”。课件截图下方可见“从视频”“V3 测试版”“替换音频”及“去除水印”开关，其中去除水印开关未开启；课件没有展示其他设置的展开面板。



这一步更换的是原视频中的人物形象并迁移动作。课件没有展示把新旁白送入某个对口型功能的过程，因此还不能据此认定新文案的口型已经自动匹配。


### 6.2 输出仍是后续剪辑素材

结果截图中，原人物变成了即梦生成的古人，但原视频的灰色房间背景、弹幕或文字以及平台水印仍在。它还没有变成最终的药房画面。



因此下一步仍要把输出视频放回剪映，处理人物抠像和背景合成。选择片段时也应检查脸部、衣服与动作是否稳定，再与新旁白对齐。


## 七、回剪映组合声音、背景和人物

### 7.1 先建立基础时间线

第 12 页列出三个任务：

1. **导入音频，剪辑音频**：将新旁白放入工程，按台词拆分、调整停顿，并拼接需要保留的原声。
2. **导入背景，调整尺寸**：把中药房图片放到人物下方，铺满所需画面长度，调整缩放与位置。
3. **导入人物画面，匹配口型**：把 Viggle 输出放在背景上方，结合台词安排人物片段。

“匹配口型”是课件列出的任务，PPT 没有展开具体功能、按钮或参数。截图只能确认人物视频和音频都经过分段处理，不能据此补写一套自动对口型操作。



素材区可见原参考视频、新旁白 **“说话回答我.wav”**、人物视频 **“回答我参考视频.mp4”** 和药房图片。预览是古人叠在药柜前，字幕“你天天熬夜刷手机”为黄色、黑色描边。

草稿参数显示 `16:9`、`30.00 帧/秒`、`Rec.709 SDR`，分辨率为“适应”。这些属于截图中的工程参数；课件没有提供最终视频导出设置。


### 7.2 抠像、字幕与 I/O 选区

第 13 页继续列出三项任务，并复用了第 12 页的同一张工程截图：

| 课件原任务 | 在工程中的处理重点 | 截图能确认的内容 |
| --- | --- | --- |
| 视频抠像 | 去掉人物视频原背景，让下层药房图片露出 | 预览已经是古人＋药房；没有展示具体抠像面板与参数 |
| 识别文案，调整颜色 | 根据语音生成字幕，校对文字并调整显示样式 | 时间线上有多段字幕，预览为黄色字、黑色描边 |
| 加 i 点 o 点，导出成片 | 用入点 I、出点 O 选择需要输出的片段范围 | 时间线上出现“取消选定区域”提示；具体快捷键与界面以当前版本为准 |

“i 点 o 点”结合剪辑语境与选区提示，应理解为选择入点、出点，不应当成额外的动画或节奏特效。



局部图中，上层是字幕和多段人物视频，中间是约 22 秒的药房图，下方能看到“说话回答我.wav”与原“剩余声音”片段交替使用。新台词与保留原声需要接得自然，同时检查字幕和嘴部动作有没有明显错位。

整张工程截图的总时间显示约 `00:01:21:15`，但时间线前方还放着原参考素材，不能把它直接写成最终成片时长。设置入点、出点后，应确认输出范围只覆盖需要的合成部分，再导出并播放检查。


## 八、复做时容易混淆的几处

| 容易混淆的地方 | 本课材料应怎样理解 |
| --- | --- |
| 提取了整段音频，就可以拿去克隆 | Noiz 截图提示 3—10 秒、10M 以内，需要先选短人声样本 |
| 文案写了 30 秒以内，配音就一定够短 | 这是提示词要求，需实际试听并检查时长，尤其是追问停顿与英文段落 |
| 动作形象替换等于背景合成和新台词对口型 | Viggle 结果仍有原背景；课件后续另列抠像与口型匹配，但没有展开口型设置 |
| 背景结果里有人，说明模型没听完整要求 | 截图实际输入少了“画面无需人物”，应先检查输入再挑图 |
| 生图是 3:4，成片也一定是 3:4 | 截图生图为 3:4，剪映草稿为 16:9，合成时要调整裁切与构图 |
| 工程总长就是最终视频长度 | 工程里还有原参考素材，应核对入点、出点和实际导出范围 |

完成一版后，重点回看文案时长、声音拼接、人物边缘、字幕与口型是否连贯。涉及健康引用的台词先核对，克隆音色使用有权使用的样本。


## 九、课程小结

第 14 页回顾了同一条五步流程：**DeepSeek 写养生文案 → 剪映提取原声、Noiz 克隆并生成新旁白 → 即梦生成人物与背景 → Viggle 替换动作形象 → 剪映抠像、合成、加字幕并导出**。

复做时可以按中间素材逐项推进：先定一条文案，得到可用的旁白；再选好人物图、无人药房图与动作视频；最后将声音、人物和字幕逐句对齐。把“单项生成成功”和“整条视频剪接自然”分别检查，才能完成课件展示的古人养生“回答我”视频。
