# 05 AI做儿童英语故事


## 资料定位与课程目标

本篇对应课件 **《05-AI做儿童英语故事》**，共 14 页。课程以儿童英语阅读账号为例，演示中英故事脚本生成、卡通配图、英文配音和剪映双语阅读视频制作。第 14 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇使用本地下载的同名课件整理，保留原课提示词与关键操作截图。

开场的“零基础新手也涨粉赚钱”和“爆款视频”属于课程对账号方向的介绍，实际内容集中在视频制作，没有展开收益验证或商业合作流程。课件所说的免费工具，以实际使用时的额度和收费规则为准。


## 一、先看案例：儿童英语阅读视频的画面结构

原课第 2 页展示两个案例：**My Day／我的一天**和 **It's Getting Colder／降温了**。第 10 页在剪辑前再次展示同一组案例，用来对照成片形式。

从案例封面帧可以看到，竖屏画面分为几个区域：

| 区域 | 可见内容 | 在阅读视频中的作用 |
| --- | --- | --- |
| 上方配图 | 起床的小女孩、雪天街道上的小女孩 | 用卡通场景对应短文主题 |
| 标题 | 英文标题与中文标题 | 让读者先知道阅读内容 |
| 正文 | 一整篇英文短文 | 提供连续阅读的文本 |
| 下方字幕 | 带底色的中英对照短句 | 随配音呈现当前读到的内容 |

后续剪映演示沿用“配图、标题、英文全文、双语字幕”的结构。**英文全文用于阅读，底部逐句字幕用于跟读和对照理解**，两者需要分开处理。

## 二、整体流程：脚本、配图、配音与剪辑

课件把儿童英语阅读视频的制作分为四步：

| 步骤 | 使用工具 | 操作与产出 |
| --- | --- | --- |
| 1. 生成儿童故事脚本 | 豆包 AI | 指定读者、主题、篇幅和中英对照要求，生成英文短文及中文译文 |
| 2. 生成儿童故事配图 | 即梦 AI | 用场景提示词生成卡通图，选择图片，变高清并下载 |
| 3. 生成儿童故事音频 | AI 语音工具 | 将英文标题和正文转为英文配音，试听后下载音频 |
| 4. 完整视频剪辑成片 | 剪映 | 组合背景、配图、标题、正文与音频，添加英文字幕及中文翻译 |

故事脚本是后续制作的依据：图片表现故事场景，音频朗读英文内容，剪映负责把这些素材放在同一条时间线上。

原课的示例在不同环节有所切换：豆包一次生成三个主题，即梦示范“小女孩写作业”，配音页面使用 **My Day**，剪映与成片则使用 **National Day Holiday**。实际制作同一条视频时，应统一所用的故事版本。


## 三、用豆包生成儿童英语故事脚本

### 3.1 用 ISRO 模型组织指令

课件先介绍与 AI 对话的 **ISRO 模型**，再把它用于儿童英语短文：

| 部分 | 课件中的含义 | 本课怎样使用 |
| --- | --- | --- |
| I：Identity，身份 | 给 AI 一个身份 | 儿童英语阅读赛道的小红书博主 |
| S：Skill，技能 | 告诉 AI 它会的技能 | 输出儿童英语短文小故事，以及中英文对照 |
| R：Require，要求 | 输出什么结果、数量 | 语义连贯、逻辑通顺，适合指定读者，包含标题与中英对照 |
| O：Output，输出 | 以什么形式展示 | 给出 3 个指定主题的英语＋中文短文，并说明篇幅 |

本课既指定了“零基础以及中国儿童三年级”的读者，也指定了中英文对照形式。主题、读者和输出结构共同决定生成内容，不能只告诉 AI“写一篇英语故事”。


### 3.2 原课完整脚本提示词

打开豆包：[doubao.com/chat](https://www.doubao.com/chat/)，输入以下原课指令：

~~~text
#身份：你是一个儿童英语阅读赛道的小红书博主，你非常擅长制作儿童英语阅读视频。
#技能：你最擅长输出儿童英语短文小故事，以及相应的中英文对照。
#要求：这些短文，语义连贯，逻辑通顺，适合零基础以及中国儿童三年级的学生阅读，且包含标题、中英文对照。
#输出：现在请你帮我输出3段英语+中文的短文，主题分别是：【我的一天】、【国庆假期】、【我喜欢大海】，每段字数在150字左右。
~~~

三个主题分别是日常生活、节假日和自然场景。返回结果采用“中文篇名、英文标题与正文、中文标题与译文”的形式，可分别用于标题、配音和双语字幕校对。

原提示词写的是“每段字数在 150 字左右”，没有明确指英文单词数还是中英总字数。第 8 页截图底部另有“现在输出 3 段英语短文，每段 100 字左右”的输入文字，课件没有说明两次要求如何衔接。下面按截图实际展示的三篇短文记录，不将篇幅视为严格满足 150 个英文单词。




### 3.3 原课短文一：My Day／我的一天

第 5、8 页展示的 **My Day** 如下，这一版英文也被粘贴到后续配音页面。

**英文：**

~~~text
My Day

I get up early in the morning. I brush my teeth and wash my face. Then I have breakfast. After that, I go to school. I have many classes and I learn a lot. At noon, I eat lunch at school. In the afternoon, I play games with my friends. After school, I go home and do my homework. In the evening, I watch TV and then go to bed. This is my day.
~~~

**中文：**

~~~text
我的一天

我早上起得很早。我刷牙洗脸，然后吃早餐。之后，我去上学。我有很多课，也学到了很多东西。中午，我在学校吃午饭。下午，我和朋友们一起玩游戏。放学后，我回家做作业。晚上，我看电视，然后睡觉。这就是我的一天。
~~~

这篇按早晨、上学、中午、下午、放学后和晚上展开，用时间顺序串起日常活动。它与开场案例中以“I wake up early”开头的 **My Day** 是不同版本，后续配音应使用自己最终确定的正文。


### 3.4 原课短文二：National Day Holiday／国庆假期

第 8 页豆包结果中的版本如下：

**英文：**

~~~text
National Day Holiday

During the National Day holiday, I am very happy. My parents and I go to the park. There are many beautiful flowers and trees. We fly kites and have a picnic. We also take photos. Then we go to the zoo to see animals. The pandas are so cute. I like them very much. This holiday is really fun and I enjoy it a lot.
~~~

**中文：**

~~~text
国庆假期

在国庆假期期间，我非常开心。我和爸爸妈妈去公园。那里有许多美丽的花和树。我们放风筝、野餐，还拍照。然后我们去动物园看动物。熊猫非常可爱，我很喜欢它们。这个假期真的很有趣，我很享受。
~~~

这一版围绕去公园、放风筝、野餐、拍照和看动物展开。后面的剪映截图虽使用相同英文标题，正文已经换成“I had a wonderful time”开头的另一版，见第六节。


### 3.5 原课短文三：I Like the Sea／我喜欢大海

**英文：**

~~~text
I Like the Sea

The sea is very beautiful. It is blue and wide. I go to the beach with my family. We play in the sand and build sandcastles. The waves come and go. Sometimes we swim in the sea. The water is cool. We can also see many shells on the beach. I like collecting shells. I really like the sea. It makes me feel relaxed and happy.
~~~

**中文：**

~~~text
我喜欢大海

大海非常美丽。它是蓝色的，而且很宽阔。我和家人去海滩。我们在沙滩上玩耍，堆沙堡。海浪来来去去。有时我们在海里游泳，海水很凉爽。我们还能在海滩上看到很多贝壳。我喜欢收集贝壳。我真的很喜欢大海，它让我感到放松和快乐。
~~~

三篇短文都可以拆出一个主要场景作为配图依据。先选定本次要制作的一篇，再准备对应图片和英文配音，避免将一次生成的三个主题混到同一条视频中。

> 补充：生成后应检查英文拼写、语法、词汇难度和中译是否对应。原课的目标读者描述不等于已完成难度分级，实际使用时仍需按孩子的阅读水平调整。




## 四、用即梦生成儿童故事配图

### 4.1 场景提示词与生图设置

打开即梦：[jimeng.jianying.com/ai-tool/home](https://jimeng.jianying.com/ai-tool/home)，进入“图片生成”。第 6 页截图中的完整提示词为：

~~~text
一位小女孩在床边的书桌上写作业，表情自然，窗外繁星点点，皎洁的明月照在屋里，3D卡通动画风格，色彩明艳，超高清分辨率。
~~~

这条提示词给出人物、动作、环境和画风：小女孩在书桌上写作业，窗外有星星和月亮，采用色彩明艳的 3D 卡通动画风格。这里生成的是故事配图，正文与标题在剪映中另行添加。

| 设置 | 原课截图中的值 |
| --- | --- |
| 生图模型 | `图片 2.1` |
| 精细度 | `5` |
| 图片比例 | `16:9` |
| 参考图 | 页面提供“导入参考图”，截图中未见已导入的参考图 |

这些是课件当时的设置。**配图为 16:9，后续阅读视频为 9:16**：横图放在竖屏画面上方，下方留给故事文字和字幕，两处比例承担不同用途。




### 4.2 查看生成结果，选择适合故事的图片

提交后，原课结果区返回四张“小女孩夜晚写作业”的图片。人物发型、衣服、姿态、灯光和窗外景物有所不同，课件随后选择第一张继续做高清处理。

选图时对照故事中的动作和场景，再检查人物表情、手部、桌面物品与构图。这个示例对应“我的一天”中的写作业活动；后面的“国庆假期”成片换用了游乐园街道图片，课件没有展示那张图的完整生成提示词。

原课这里直接用文字生成卡通配图，没有演示多镜头动画或逐句生成分镜。




### 4.3 变高清并下载原图

课件第 7 页强调“AI 变高清、下载原图”：

1. 在候选结果中找到选中的图片，调出图片工具栏。
2. 点击原课红框标出的 `HD／超清`，等待处理完成。
3. 打开处理后的大图检查效果，通过下载入口保存图片，供剪映导入。

截图中，处理后的结果带有“超清”标记，大图右上角可见下载、收藏等入口。高清处理后仍要检查人物和画面细节。

页面另列出“百度去水印”链接：[image.baidu.com](https://image.baidu.com)，没有展开具体操作，不能把它当作已经演示完成的高清步骤。

## 六、用剪映制作完整阅读视频

### 6.1 导入素材，安排竖屏阅读画面

第 11 页的操作是：**导入音频、图片、故事脚本，添加背景图片**。剪映预览采用 `9:16` 竖屏，示例主题改为 **National Day Holiday**。

按原课可见画面整理：

1. 准备同一版故事的英文正文、对应配图和英文配音，导入剪映。
2. 在竖屏画面上铺背景，原课使用浅粉、浅黄的柔和背景。
3. 将横向卡通图放在上方，原课是带城堡、街道和人群的游乐园画面。
4. 在图片下方放英文标题，再将整篇英文正文排在标题下方。
5. 将配音放入音轨，让图片与阅读文字覆盖朗读所需的时长。

| 画面元素 | 原课样式 |
| --- | --- |
| 背景 | 浅粉、浅黄色调，覆盖整张竖屏画面 |
| 配图 | 上方横图，左右留有边距 |
| 标题 | `National Day Holiday`，居中黑字 |
| 英文正文 | 放在标题下方，红粉色文字，多行排列 |
| 动态字幕 | 在正文下方另留位置，下一步添加 |

时间线上可以看到标题、正文／字幕、图片和配音等轨道。这里按原课画面说明各类素材的用途，课件没有逐项演示背景设置入口，也未给出固定字体、字号和边距数值。




### 6.2 识别英文音频，添加中文字幕

第 12 页继续演示：**识别音频添加字幕，为英文字幕添加中文字幕**。

1. 对导入的英文配音进行字幕识别，得到分段的英文字幕。
2. 对照原始脚本核对识别结果，修正英文单词、标点和断句。
3. 为英文字幕补上相应中文译文，使两种语言指向同一段配音。
4. 调整双语字幕的位置、文字样式和底色，放在英文全文下方。
5. 播放检查，让字幕出现、结束的时间与朗读对应。

原课红框中的一组字幕为：

~~~text
During the National Day Holiday
在国庆假期期间
~~~

字幕使用红粉色圆角底框与白色文字，英文在上、中文在下。它随朗读切换，画面中间的英文全文则作为持续展示的阅读内容。

**正文排版与逐句字幕要分别检查**：正文需要完整可读，字幕需要与声音同步。课件没有展开英文识别的语言选项或中文翻译按钮的完整菜单路径，也没有提供可直接套用的字幕预设参数。




### 6.3 对照成片，统一故事版本

第 13 页展示的成片保留了上方配图、英文标题、英文全文和底部双语字幕。画面中的国庆故事已经换成下面这一版：

~~~text
National Day Holiday

During the National Day Holiday, I had a wonderful time. I went to the park with my dear parents. The park was like a fairyland, filled with numerous beautiful flowers and tall, shady trees. We flew a kite together. The kite, with its colorful tail, soared high into the clear blue sky. It was really exciting.
~~~

与第 8 页相比，这一版主要使用过去时，并增加了 `fairyland`、`numerous`、`soared` 等词。课件未展示改写指令，也没有给出这一版完整的中文译文，不能把前面的中文段落直接当作本版字幕。用于儿童阅读时，还要重新检查词汇难度。

原课内嵌的成片视频约 **5.2 秒**，这里只展示了片段；剪映截图中的时间线总长显示为 `00:00:23:16`，属于时分秒帧格式。课件没有继续演示导出分辨率、码率、帧率选择或平台发布操作。

> 补充：导出前播放检查标题、全文、配音和双语字幕是否属于同一版故事，确认文字没有被截断或重叠，图片与文字持续到朗读结束。导出参数按实际素材与发布要求选择。




## 七、课程小结

原课第 13 页将本篇流程归纳为四步：

1. 使用 **豆包 AI** 生成儿童故事脚本。
2. 使用 **即梦 AI** 生成儿童故事配图。
