# 12 AI做书单号


## 资料定位与课程目标

本篇对应课件 **《12-AI做书单号》**，共 14 页。课程先展示读书解说案例，再用 DeepSeek 写读书文案，通过剪映“图文成片”生成初稿，随后用豆包生成分镜与绘画提示词、用即梦生成配图，最后在剪映中替换素材并优化成片。第 13 页为小结，第 14 页没有新增教学内容。

资料来源：[AI+新媒体搞钱指南](https://pan.baidu.com/s/1XD-zLZ0uqGtOvX5sOtb1aw?pwd=1234)，提取码 `1234`。本篇以本地课件为准，读取正文及操作截图，并抽看 2 段内嵌案例录屏。

课件流程标题多次写“3 分钟读书文案”，第 4 页的完整提示词却要求“5 分钟、约 1200 字”。下文保留这两个口径，复做时先确定目标长度，再用实际朗读时长调整文案。提示词提到的“5 篇参考文案附件”没有出现在本地目录或 PPTX 内嵌附件中，因此保留原要求，不补造参考文章。


## 一、案例中的书单号长什么样

第 2 页展示两段读书视频录屏，第 8 页再次展示同样的两段素材，提醒先观察案例再剪辑。“书单号”在本课中主要指围绕一本书的观点和故事做短视频解读，并用旁白、字幕与配图呈现。

| 案例 | 录屏可见内容 | 可以观察的表达方法 |
| --- | --- | --- |
| 《被讨厌的勇气》 | 开头出现“3 分钟读一本书”，后面用“痛苦是因为过去”“阿德勒的目的论”等文字配简笔画 | 把抽象概念转成目标靶、人物、时钟等直观图像，标题和旁白字幕分层呈现 |
| 《如何度过每天的 24 小时》 | 用上班、休息、时钟等插图讲时间利用，顶部有章节提示 | 用观众熟悉的生活场景引出书，随解说切换图片，以章节提示交代进度 |

两段素材都是手机竖屏录屏，书单内容画面放在中间的横向区域。白色纸张背景、居中插图、顶部关键词或章节和底部字幕构成主要版式。第一段以黑色线条为主并有少量彩色，第二段使用彩色插画，案例风格并不完全相同。

录屏文件分别约 63.06 秒、83.46 秒，只能说明课件截取的示例长度。视频里的“3 分钟”“5 分钟”不能从这两个录屏文件的时长反推。下面裁取真正的讲解画面，去掉大部分手机导航和互动按钮。






## 二、制作流程：先自动成片，再逐段替换素材

第 3 页给出四步流程：

| 阶段 | 工具 | 具体任务 | 交给下一阶段的内容 |
| --- | --- | --- | --- |
| 1. 生成读书文案 | DeepSeek | 分析参考文章，写出有观点、有故事的读书解说 | 核对后的完整旁白 |
| 2. 自动成片 | 剪映 | 将现有文案输入“图文成片”，生成配音、画面与时间线初稿 | 可继续编辑的视频工程 |
| 3. 生成分镜和配图 | 豆包、即梦 | 让分镜、绘画提示词和文案逐句对应，生成统一风格的图片 | 能替换原有画面的配图素材 |
| 4. 替换和优化 | 剪映 | 更换背景、素材、音乐和解说声，调整字幕，添加进度条、动画、特效与音效 | 检查完成后导出的书单视频 |

自动成片先提供旁白和镜头顺序，后续再解决画面风格、内容匹配和观看节奏。课件将前三步标为“AI 免费生成”，但没有说明当时的额度和收费条件，复做时以工具实际界面为准。


## 三、用 DeepSeek 写读书解说文案

### 3.1 一次说清身份、参考材料和输出条件

第 4 页的提示词不是只给书名，而是先指定读书博主身份，再要求分析 5 篇参考文章的开头、结构和语言风格，生成 3 篇新文案。原课还要求内容符合原著，有逻辑、有深度，并列出了目标书籍。

下面保留课件原提示词：

```text
#身份：我是一个读书赛道的短视频博主，我需要制作5分钟精读一本书的爆款短视频。
#需求：下面我会给你5条爆款的读书文案，这些读书文案都十分吸引读者，我需要你分析这5篇文案的风格及特征，然后模仿写出3篇同样风格的文案，并且将现有5篇文章的核心文案结构和风格表达，运用在新生成的3篇文案中。
#要求：
有新开篇：参考5篇原文后，请你给我写3个新的开篇文案，吸引人的开头至关重要；
文章结构：需要你识别5篇原文的文章结构，并且在模仿文案中，也严格按照这个文章结构来撰写；
风格模仿：严格模仿这5篇参考文案的风格、语气，有观点，有故事，有点睛之笔的结尾；
符合原著：撰写的新读书文案，要符合原著，解读视角有逻辑、有深度，且能够吸引自媒体用户；
字数时长：生成的3篇新的读书文案，每一篇时长在5分钟，每一篇文案字数约在1200字左右；
新的3本书，分别是：《曾国藩传》、《骆驼祥子》、《生活的囚徒》，现在根据以上理解，请帮我生成3篇新读书文案。
5篇参考的爆款文案在附件中。
```

实际操作顺序是：准备参考文案，提交提示词和参考材料，得到 3 篇结果，再对照原书审阅。当前资料缺少原提示词所说的 5 篇附件，因而无法复现课程当时的全部输入。

课件要求每篇约 1200 字、时长 5 分钟。字数是写作约束，最终时长还受配音语速和停顿影响，应在生成朗读后检查。


## 四、读懂生成结果的结构，再核对内容

第 5 页展示《曾国藩传》文案的前后两张截图。把它们连接起来，可以看到一篇读书文案如何从吸引注意的开头，推进到故事、观点和现实联系。

| 组成部分 | 截图中的具体写法 | 在解说中的作用 |
| --- | --- | --- |
| 开篇 | 以“考七次秀才落榜、两次投江自杀的‘笨人’，如何逆袭成晚清第一重臣？”制造反差 | 引出人物经历和观众想知道的问题 |
| 引入书籍 | “今天用 5 分钟带大家精读《曾国藩传》”，并介绍张宏杰和人物背景 | 交代书名与解读对象 |
| 第一把钥匙：以拙破巧 | 用读书、带兵、用人等故事解释“笨功夫” | 用故事支撑观点 |
| 第二把钥匙：自虐式修身 | 写日记、“十二条日课”和家信等内容 | 延续分点讲解的结构 |
| 第三把钥匙：屠龙术与生存术 | 用官场经历、奏折和裁撤湘军等情节展开 | 引出人物后期变化 |
| 现实照妖镜 | 把“笨办法”联系到当代追求速成的现象 | 将书中观点连接到观众生活 |
| 金句点睛 | 用“躬身入局，挺膺负责”等话收尾，再回到“笨方法”主题 | 呼应开头并结束解说 |

课件保留的结尾是：

> 最后分享书中这段振聋发聩的话：“天下事，在局外呐喊议论，总是无益，必须躬身入局，挺膺负责，乃有成事之可冀。”当你在为 35 岁危机焦虑时，不妨看看这个笨人如何用最笨的方法，把烂牌打成王炸。

以上是 **模型生成的解说示例**。涉及人物经历、历史事件、引文及“书中记载”的内容，都应回到原著确认出处；文案中加入的现代类比也应与原著内容区分。

第二张截图下方还露出《骆驼祥子》的开头：“他拉车三年攒够买房钱，却在一夜间沦为乞丐……”。这里的“买房钱”与作品中祥子买车的主线存在偏差，说明“要求符合原著”之后仍需人工校对。该文案正文在截图中未完整显示，《生活的囚徒》结果也未展示，不能据此补齐另外两篇。






## 五、用剪映“图文成片”搭建初稿

### 5.1 从已有文案进入自动成片

第 6–7 页依次用红框标出了入口和生成按钮：

1. 打开剪映首页，点击 **图文成片**。
2. 在左上角“已有文案？”区域选择 **自由编辑文案**。
3. 把审阅后的读书文案粘贴到编辑区。
4. 选择解说音色，截图当前选中 **沉稳解说**，再点击 **生成视频**。

使用“自由编辑文案”是因为上一阶段已经写好旁白，需要把确定的文本送入剪辑流程。





截图的粘贴示例讲的是林徽因，右下角显示 `997/20000`，这是当时编辑框的计数，不能当成《曾国藩传》文案的字数。课程借这张图说明操作入口，下一张工程截图才切回《曾国藩传》。




## 六、检查自动成片的时间线和素材

第 7 页的工程截图显示，剪映已经把文案拆成配音片段，画面轨道排列了不同图片和视频，底部有背景音乐 **Last Reunion**。预览中的字幕是“如何逆袭成晚清第一重臣？”，画面使用人物图和红色文字背景。

这一阶段先检查三个对应关系：

- 文案段落是否都进入了时间线，书名、姓名和引文有没有漏读、误读。
- 旁白和字幕是否对应，字幕出现和消失的时间是否合适。
- 自动匹配的素材是否符合正在讲的内容，哪些画面需要重新制作。

截图右侧显示工程为 `16:9`、草稿帧率为 `30.00 帧/秒`，播放器总时码为 `00:03:31:01`。这些反映的是当前示例工程，课件没有展示导出面板，不能把草稿参数当作统一导出要求。

接下来生成配图时，以已经核对的旁白为依据，保留清楚的逐句对应关系，方便回到剪映替换。




## 七、让豆包生成 40 个分镜、绘画提示词和对应文案

### 7.1 用表格固定三者的对应关系

第 9 页的截图要求将《曾国藩传》文案拆成 **40 个分镜头**，并明确每个镜头、绘画提示词对应哪一句旁白。截图中的要求如下：

```text
请将以下曾国藩传的书单文案，帮我输出40个对应的分镜头和AI绘画提示词，要求：

1、每个分镜头和AI绘画提示词要标注对应哪一句短视频文案
2、分镜头+AI绘画提示词+对应的文案用表格展示
3、AI绘画提示词要求统一风格：线条简笔画、黑白色
4、分镜头数量为：40个
```

这段要求后面还要粘贴前面确认的完整读书文案。课件截图继续展示了正文的一部分，笔记将请求区域单独保留，避免把不完整的正文当成完整输入。



### 7.2 查看表格，不把局部截图当作全部结果

输出截图只有两行完整可见，列名是“分镜头”“AI 绘画提示词”“对应的文案”：

| 分镜头 | 画面安排 | 对应关系 |
| --- | --- | --- |
| 考场场景 | 多次考试画面并列，人物失落地走出考场 | 对应开头中考试落榜的叙述 |
| 江边场景 | 人物在江中、岸边有人呼喊 | 仍对应同一句较长的开篇文案 |

这说明一句旁白可以拆成多个镜头。请求中的“40 个”是输出要求，截图没有展示完整 40 行，也没有展示最终验收结果。制作时需要检查整篇文案是否覆盖，是否存在跳句、重复或画面与原著不符。




## 八、用即梦生成统一风格的图片素材

### 8.1 按分镜逐条生成

第 10 页从分镜进入即梦的 **图片生成**。课件截图里的一条输入是：

```text
曾国藩站在河边，纯色底图，线条简笔画，黑白色
```

页面中可见的设置如下：

| 项目 | 截图当前值 |
| --- | --- |
| 生图模型 | 图片 2.0 Pro |
| 精细度 | 5 |
| 图片比例 | 1:1 |
| 提示词风格 | 纯色底图、线条简笔画、黑白色 |

截图中的 `Deepseek-R1` 出现在提示词区按钮上，生图模型栏明确写的是“图片 2.0 Pro”，两处名称需要区分。截图底部“图片尺寸”没有展开数值，因此不补写像素尺寸、生成费用和下载参数。



### 8.2 检查平台提示和实际画风

同页生成记录中，较早的一条“曾国藩跳水……”显示“你输入的文字描述不符合平台规则，请修改后重试”。随后“曾国藩站在河边……”出现了候选图。这个记录只反映当次生成过程，不能据此判断平台的完整审核规则，也不能把改写后的静态江边画面直接描述成原提示词里的动作。

生成结果中有些图保留了蓝色河水和较复杂的背景，即使输入写了“黑白色、纯色底图”，也不保证结果完全满足要求。另一组提示词是：

```text
线条简笔画风格，黑白色，书房场景，张宏杰坐在堆满书籍资料的桌前翻阅资料
```

按旁白逐条生成，选择与当前句子一致的图片，并检查人物特征、服装时代、背景和色彩是否统一。作者书房画面属于 AI 示意图，不能当作作者真实照片。




## 九、回到剪映替换背景和画面素材

第 11 页开始做细节优化，先解决整条视频的画面风格：

1. **更换背景图片**：课件写“纯色底图”，工程中实际展示浅色纸张纹理背景。可以按这种简洁底图安排画面，给插图和字幕留出清楚的区域。
2. **删除原剪映素材，换成 AI 生图**：按分镜对应的文案逐段替换。保留旁白与段落位置，再调整图片的停留时间、缩放和位置，避免删掉画面后破坏解说顺序。
3. **更换背景音乐并调整音量**：试听旁白和音乐同时播放的效果，保证解说能听清。课件没有给出音量数值，也未明确说明替换后的曲名。

截图预览先放书封面与书籍信息，后面切换到与旁白对应的插图。生成图片的 `1:1` 与当前工程的 `16:9` 属于不同阶段，方图可以缩放后放进横向底图中，无需把方图拉伸到变形。



书籍信息卡还需要单独校对。截图写《曾国藩传》作者为“张安杰”，前面的生成文案和书房提示词写“张宏杰”，资料内部已经不一致。截图中的“2019 年（民主与建设出版社版本）”也是示例卡片文字，复做时应按实际使用版本的版权页核对作者、出版年和出版社。




## 十、调整解说、字幕和章节进度，完成导出

第 12 页继续列出后期操作：更换解说声、调整字幕、添加色块进度条、添加动画和特效、添加音效，最后导出。

| 操作 | 本课要解决的问题 | 完成后检查 |
| --- | --- | --- |
| 更换解说声 | 让声音更适合读书解说风格 | 专有名词读音、停顿和语速正确，换音色后仍与字幕对应 |
| 调整字幕 | 让旁白可以清楚阅读 | 文本无错字、断句自然，不遮住插图重点 |
| 添加色块进度条 | 告诉观众当前讲到哪里 | 章节名称与正文一致，高亮随讲解推进 |
| 添加动画、特效 | 让插图切换有节奏 | 动作服务于讲解，画面边缘没有意外裁切 |
| 添加音效 | 在强调或转折处增加提示 | 与旁白衔接自然，不盖住人声 |
| 导出 | 得到可发布的视频文件 | 从头检查内容、声音、字幕及片段衔接 |

预览顶部可见“前言、冲突、成长、高升”等章节文字，左侧有浅黄色色块，时间线上有长条文字轨和色块轨。这些是课件进度提示的可见形态。课程没有展开色块的完整关键帧设置、动画和特效名称，也没有说明各段的时间数值，不能仅凭单张截图认定所有章节切换已经完成。



截图中的作者插图放在画面中间，下方保留字幕，纸张底图贯穿背景。工程底部仍可见 `Last Reunion` 音轨，因此“更换背景音乐”在本课中是一项优化要求，截图不足以确认已换成另一首曲子。

本页没有导出面板，未指定最终分辨率、帧率、码率或文件大小。完成画面和音频检查后，再按实际发布用途选择导出设置。




## 十一、课程小结

第 13 页将全过程归纳为：

1. 用 **DeepSeek** 生成读书文案，先核对原著内容，再确定最终旁白。
2. 用 **剪映图文成片** 搭建初稿，得到可继续编辑的素材与音轨。
3. 用 **AI 生成分镜、绘画提示词和配图**，保留每个镜头与旁白的对应关系。
4. 回到 **剪映替换素材并优化剪辑**，统一背景和画风，调整解说、字幕、进度提示及音效，再导出。

本课的操作重点是先搭好解说时间线，再逐段改进画面。复做时尤其要检查目标时长是否统一、书籍信息与历史情节是否准确、配图是否真正满足风格要求。首页提到“变现”，正文没有展开选品、佣金、挂商品或成交数据，本篇按实际展示的制作流程记录。
