## 一、AI测试新思维

### 1.1 从登录页面到聊天机器人

登录页面通常能按账号、密码和验证码规则判断成功或失败。聊天机器人面对自然语言输入，同一句问题可能产生不同措辞、不同细节的回答，因此不能只比较答案字符串。

| 测试维度 | 传统软件测试：以登录功能为例 | AI测试：以聊天机器人为例 |
| --- | --- | --- |
| 输入 | 结构化数据，如账号、密码 | 非结构化自然语言 |
| 处理 | 固定业务逻辑与规则 | 基于概率的推理和生成 |
| 输出 | 确定的结果，如成功、失败 | 非确定性、多样化文本回复 |
| 验证 | 结果是否严格匹配预期 | 多维度评估回复质量 |

课程用“没标准答案”提醒测试思维的变化：很多回答没有唯一表述，但仍然需要事实正确、遵循指令、逻辑合理。

> 补充说明：输出措辞不唯一，不等于没有正确性标准。事实、计算结果、禁止项和格式要求仍可形成明确的判定条件。

### 1.2 三大转变

1. **从确定性到概率性**：相同输入可能产生不同输出，需要考虑结果的稳定性。
2. **从功能正确到质量评估**：除“能否回答”外，还要判断回答是否准确、有用、安全。
3. **从静态验证到动态优化**：持续监控、评估和迭代，关注模型变化后的行为。

课程概括为：从基于规则的验证，转向基于数据的评估。

### 1.3 五大核心挑战

| 挑战 | 测试时要解决的问题 |
| --- | --- |
| 非确定性输出 | 如何有效测试概率性结果？ |
| 主观评价标准 | 如何定义什么是“好”的回答？ |
| 幻觉问题 | 如何识别看似可信的错误或虚构内容？ |
| 安全与伦理风险 | 是否出现偏见、毒性内容、隐私泄露等？ |
| 持续变化与漂移 | 模型更新后，能力和行为是否变化？ |

课堂思考：如果负责电商客服机器人的测试，会重点关注哪些方面？判断回答“好”或“不好”的标准是什么？

练习：梳理传统软件测试和 AI 测试的区别，再通过大模型完善，整理成 XMind。

## 二、AI测试核心概念

### 2.1 AI、ML、DL与LLM

| 概念 | 中文名称 | 课程中的含义 |
| --- | --- | --- |
| AI | 人工智能 | 让机器模拟人类智能，例如能听、能看、能说、能写的各种技术总称。 |
| ML | 机器学习 | 让机器从数据中学习规律，进而进行预测。 |
| DL | 深度学习 | 机器学习的一个分支，使用深度神经网络处理图像、文本等复杂数据。 |
| LLM | 大语言模型 | 具备语言理解、推理和生成人类语言能力的大模型。 |

### 2.2 三个能力模型

| 能力 | 测试关注点 |
| --- | --- |
| 基础功能 | 输入与输出是否符合基础设定。 |
| 智能能力 | 理解、推理、创作等高级认知能力。 |
| 安全与伦理 | 偏见、毒性、隐私、合规性等风险。 |

课程用“准不准、有用不、坏不坏”概括测试目标。

### 2.3 四个评价维度

| 维度 | 判断内容 |
| --- | --- |
| 准确性 | 事实正确、逻辑合理、信息完整。 |
| 有用性 | 能有效解决用户的问题或需求。 |
| 无害性 | 没有偏见、歧视或有害内容。 |
| 一致性 | 相同或相似输入下，输出稳定可靠。 |

脑图将准确性、一致性归入基础能力，有用性归入智能能力，无害性归入安全与伦理。

## 三、大模型幻觉及测试方法

### 3.1 什么是幻觉

大模型幻觉是：**生成的内容看似合理，实际却是错误或虚构的信息**。

原课场景：用户问“我昨天买了什么？”，AI 自信回答“你昨天买了牛奶和面包”。实际上用户昨天什么都没买，模型却在缺乏依据时给出了确定的事实。

课程从生成机制解释：模型根据前面的文字，预测下一个最可能出现的词，通俗理解为强大的、基于概率的“文字接龙机器”。这种生成方式不保证每一条事实都真实，测试需要识别其输出中的错误与虚构。

### 3.2 三类幻觉

| 类型 | 表现 | 测试方法 | 讲义示例 |
| --- | --- | --- | --- |
| 事实性幻觉 | 编造不存在的事实或信息。 | 已知事实验证、时间敏感信息核对。 | “马云现在还是阿里巴巴董事长吗？” |
| 忠实性幻觉 | 未遵循用户指令、上下文或格式要求。 | 指令遵循度检查、输出格式验证。 | “请用表格列出北京、上海、广州的人口。” |
| 逻辑性幻觉 | 推理过程矛盾，或结论不合理。 | 逻辑链检查、数学计算验证。 | “如果 A>B 且 B>C，那么 C>A，对吗？” |

### 3.3 快速检查与通过标准

1. 回答中的事实是否可验证？要求真实，不把推测当事实。
2. 是否完全遵循用户指令？检查角色、内容与输出格式等限制。
3. 推理过程或结论是否存在矛盾？要求逻辑自洽。
4. 是否引用不存在的来源或信息？要求来源和引用可核对。

### 3.4 从测试用例到评估

测试用例是为特定测试目的编写的、详细且可执行的文档。沿用前面手工测试的八要素：用例编号、标题、所属模块、优先级、前置条件、测试步骤、测试数据、预期结果。执行时再记录**实际结果**，形成九列记录。

原课流程：通过提示词准备测试数据 → 生成幻觉测试用例 → 输入被测模型 → 记录实际回答 → 人工或使用另一个大模型评估。

课堂案例测试[豆包](https://www.doubao.com/chat)是否存在三类幻觉，建议用大模型辅助设计用例。练习页要求测试[文心一言](https://yiyan.baidu.com/)；该页任务文本仍写“豆包”，整理时以练习标题和链接指向的文心一言为目标。

## 四、提示词工程应用

### 4.1 提示词的构成

提示词（Prompt）是人与大模型交互时使用的指令。提示词工程是按指定要求，以结构化形式描述提示词的过程。规范输入可以缓解幻觉，但评估步骤仍然需要保留。

| 组成 | 要说明什么 |
| --- | --- |
| 角色、人设 | 你是谁？赋予模型什么专业身份？ |
| 任务、目标 | 需要完成什么工作？ |
| 要求、约束 | 具体要求是什么？明确不能做什么？ |
| 输出 | 使用什么格式？有哪些输出要求？ |
| 背景，可选 | 用户当前的情况。 |
| 示例，可选 | 可以参考的样例。 |

### 4.2 C.R.E.A.M原则

| 原则 | 含义 |
| --- | --- |
| Clear：清晰 | 指令明确、无歧义。 |
| Role：角色 | 赋予 AI 特定专业身份。 |
| Example：示例 | 用少量示例（Few-Shot）对齐标准，也可以不提供示例（Zero-Shot）。 |
| Add Constraints：附加约束 | 明确格式、长度、排除项等。 |
| Manage Output：管理输出 | 使用结构化输出，便于后续处理。 |

原课参考模板：

```text
你以xxx的身份，帮我xxxx
要求：
1.xxxx
2.xxxx
3.不能xxx
输出：xxx
背景（可选）：xxxx
示例（可选）：参考如下示例
```

### 4.3 用提示词生成幻觉测试用例

课堂笔记示例要求生成约 15 条用例，沿用手工测试八要素，覆盖三类幻觉，并输出方便转成 Excel 的 Markdown 文档：

```text
你以专业的AI测试工程师身份，帮我设计关于大模型幻觉测试的测试用例
要求：
1.用例参考手工软件测试8要素模板（用例编号、标题、所属模块、优先级、前置条件、测试步骤、测试数据、预期结果）
2.覆盖3大类型幻觉。事实性幻觉、忠实性幻觉、逻辑性幻觉
3.覆盖场景较全面，数量控制在15条左右
输出：md格式的用例文档，需要转Excel
背景（可选）：当前是初入AI测试职场的小白
示例（可选）：无
```

讲义另给出一版提示词，重点是显式记录“实际输出、结果判断、备注”，以及排除与测试无关的内容：

```text
你以AI测试工程师的身份，帮我编写一份关于测试豆包幻觉的测试用例
要求：
1.覆盖三大类型幻觉，事实性幻觉、忠实性幻觉、逻辑性幻觉场景
2.测试用例核心内容包含：用例编号、幻觉类型、测试输入、预期输出、实际输出、结果判断、备注说明
3.不回答与幻觉测试无关的内容，避免干扰
输出：以md格式输出用例，我需要转Excel表格
背景：我当前是一个刚接触AI大模型测试的初级测试人员
示例：参考如下示例……
```

讲义示例的字段组合与课堂八要素模板不同，使用时按当前测试记录要求确定列名。预期结果先写，实际结果在执行后填写。

### 4.4 用提示词评估被测模型的回答

课堂先获取豆包回答的截图，再提交给 DeepSeek，要求按事实、忠实性、逻辑性分别判定。

![豆包幻觉测试回答](06-img/image-20260324151832957.png)

```text
你以AI测试工程师，帮我评估附件中豆包给出的结果是否满足幻觉测试通过的标准
要求：
1.事实准确性：【通过/不通过，并说明原因】
2.忠实性（指令的遵循度）：【通过/不通过，并说明原因】
3.逻辑性（逻辑是否自洽）：【通过/不通过，并说明原因】
4.总体评价：【通过/不通过】
输出：以图表形式输出评估结果
```

DeepSeek 在课堂示例中给出的**总体评价为“不通过”**：

| 维度 | 课堂评估结论 | 课堂给出的原因 |
| --- | --- | --- |
| 事实准确性 | 不通过 | 将“公元前259年1月27日”作为唯一出生日期，未注明换算争议，被评估为过度确定。统一六国的顺序“韩、赵、魏、楚、燕、齐”正确，年份基本准确，其中赵、燕彻底灭亡的年份处理被认为正确。 |
| 忠实性、指令遵循度 | 通过 | 回答了出生年月日、公历与统一六国顺序，未添加无关信息或偏离指令。 |
| 逻辑性、内部自洽 | 通过 | 课堂评估认为出生日期与农历对应一致，统一顺序按时间排列，并与“远交近攻”的描述一致，无前后矛盾。 |

> 源资料核对：这里保留的是 DeepSeek 的评估示例，不是对历史日期的独立考证。“幻觉测试用例”工作表将 `TC-FACT-001` 实际结果记为 `pass`，而课堂 Markdown 的这次评估给出总体“不通过”；两份资料的记录不一致，不能据此合并成一个确定的测试结论。正式判定时应核对同一次执行的实际回答、权威依据和预期条件，其他模型的评估也需要复核。

讲义中评估提示词还要求：“如发现幻觉，请明确指出其类型及具体问题。”这一项能帮助把总体结论落实到具体的失败点。

## 五、智能体初体验及测试思维

### 5.1 从回答问题到执行任务

原课引入两个问题：大模型能否直接查询数据、发送邮件或使用工具？换一个新会话后，能否继续记住昨天讨论的话题与状态？这些需求需要相应的工具能力和记忆机制支持。

智能体（AI Agent）是一种能够感知环境、自主理解、进行决策并执行动作的自动化程序。课程将其比作“能理解、会思考、有记忆、能执行的数字员工”。

讲义中的实现流程：用户问题 → AI大脑理解 → 调用记忆系统 → 规划执行步骤 → 使用工具能力 → 生成回答 → 存储到记忆。

![智能体实现流程](06-img/agent-flow.png)

### 5.2 智能体的组成与测试关注点

课程表达式：**AI Agent = LLM + 记忆 + 任务规划 + 工具使用**。

| 组成 | 作用 | 测试关注点 |
| --- | --- | --- |
| AI大脑、LLM | 理解问题、思考答案。 | 回答准确，逻辑合理。 |
| 记忆系统 | 记住会话历史与用户信息。 | 多轮对话一致性、信息持久性。 |
| 任务规划 | 将复杂任务分解为步骤。 | 步骤完整、顺序合理。 |
| 工具能力 | 调用插件完成任务。 | 调用准确、错误处理能力。 |

> 补充说明：记忆、当前时间和外部动作是否可用，取决于平台配置及工具支持。只在提示词中写“记住名字”或“根据当前时间问候”，不能代替对实际能力的验证。

### 5.3 智能体开发技术栈

| 技术栈 | 讲义列出的优势 | 讲义列出的限制 |
| --- | --- | --- |
| Coze在线版 | 快速实现需求，低代码。 | 平台限制较多，较深的定制功能受限。 |
| Dify | 低代码，快速实现需求，能对接本地模型与服务。 | 讲义指出本地版部分插件和功能需要自行补充。 |
| LangChain等框架 | 扩展性强，可对接本地模型并实现自定义复杂逻辑。 | 上手成本较高，开发周期较长。 |

本阶段使用 **Coze + Dify**。上表保留课程比较，平台功能会随版本变化，本课先用 Coze 完成基础体验。

### 5.5 案例一：打招呼智能体

将下面的课堂提示词填入智能体配置：

```text
你是一个热情友好的打招呼助手。
你的职责：
1. 主动、热情地问候用户。
2. 询问并记住用户的名字。
3. 根据当前时间（早上/下午/晚上）使用合适的问候语。
4. 如果用户询问与打招呼无关的问题，请礼貌地表示自己主要负责问候。

输出要求：
- 每次回复不超过100字。
- 使用合适的表情符号（如😊）。
- 保持积极、友好的语气。
```

创建后按讲义表格验证四个方面。每项根据实际回复记录通过或不通过：

| 测试类型 | 测试操作 | 预期结果 |
| --- | --- | --- |
| 人设测试 | 发送“你好”。 | 热情回应，并询问用户的名字。 |
| 记忆测试 | 告诉它名字，再问“我叫什么？” | 正确复述用户的名字。 |
| 边界测试 | 提问“今天天气怎么样？” | 礼貌拒绝，并说明主要负责问候。 |
| 格式测试 | 检查回复。 | 长度不超过100字，包含合适表情符号。 |

提示词还要求按早上、下午、晚上使用合适问候语，因此调试时需确认智能体确实能获得相应时间信息。

### 5.6 从使用者转向测试者

使用者关注“这个智能体能干什么”；测试者进一步关注“是否具备所承诺的能力，超出能力范围会怎样反应”。

原课从四个方向挖掘问题：

1. 功能漏洞：需求是否未实现，或者实现错误？
2. 逻辑缺陷：条件判断是否存在错误？
3. 用户体验：提示是否友好，响应如何？
4. 安全隐患：是否泄露信息，或输出偏见内容？

### 5.7 案例二：程序员鼓励师

课堂先让 DeepSeek 根据需求生成智能体提示词，再把生成内容复制到 Coze 中：

生成的智能体名为 `CodeMate`，重点是情绪识别、Bug处理逻辑、激励与休息提醒、简洁交互、隐私与健康约束。完整提示词如下：

```text
# 角色设定
你是 CodeMate，一位资深且充满同理心的程序员鼓励师。你拥有10年全栈开发经验，但从不居高临下。你的使命是：治愈程序员的焦虑，激发他们的信心，并在他们陷入技术死胡同时提供灵感和轻量级的技术指引。

# 核心能力与行为准则

## 1. 情绪识别与自适应语气
- **输入分析：** 首先快速判断用户情绪。
  - 如果用户输入包含“崩溃了”、“为什么不行”、“三天了”，判定为【高压/挫败】模式 -> 语气：温柔、先共情、后理性。
  - 如果用户输入包含“哈哈”、“终于跑通了”、“牛逼”，判定为【成就/兴奋】模式 -> 语气：热烈庆祝、放大喜悦、甚至配合夸张的彩虹屁。
  - 如果用户输入技术细节或报错日志，判定为【求助/钻研】模式 -> 语气：冷静、条理清晰、鼓励式提问（引导用户自己发现错误，而不是直接给代码）。
- **禁止行为：** 禁止说“这很简单”、“你怎么这个都不会”。即使是最基础的报错，也要说“这是一个经典的小陷阱，很多人都遇到过”。

## 2. 技术交互逻辑
- **Bug 处理流程：**
  1.  **安抚：** “看到这个报错别慌，这是编译器在保护你，我们来拆解一下。”
  2.  **归因：** 基于报错信息（如 NullPointer, IndentationError）给出通俗解释。
  3.  **启发：** 使用苏格拉底式提问，例如：“我注意到第15行的变量类型似乎变了，要不要检查一下那里？”
  4.  **正向反馈：** 用户解决后，给予即时肯定：“漂亮！就是这个地方！你的排查思路很清晰。”
- **知识边界：** 当你遇到无法解决的复杂架构问题时，坦诚告知，并提供寻找解决方案的思路（如建议查阅某文档的关键词），而不是强行编造错误代码。

## 3. 激励机制（游戏化）
- **进度认可：** 将枯燥的开发过程转化为“打怪升级”。
  - 例如：“完成了这个模块的封装，你的‘代码健壮性’技能点+5！”
- **休息提醒：** 如果检测到用户连续高强度输入（或主动询问），必须强制插入“番茄钟”休息建议，提醒喝水、眨眼、站起来活动。
- **仪式感：** 当用户完成一个阶段性任务（如“项目上线”、“重构完成”），自动触发虚拟庆祝仪式（放烟花、颁发虚拟勋章）。

## 4. 交互风格
- **语言：** 中英文夹杂（符合程序员习惯），适度使用技术梗（如“删库跑路”、“这个需求做不了”），但需确保幽默不越界。
- **简洁性：** 拒绝长篇大论的大道理。回复结构优先采用 **短句 + Emoji + 关键点加粗** 的格式，便于快速阅读。

## 5. 安全与伦理约束
- **数据隐私：** 绝对禁止要求用户上传实际生产环境代码或敏感密钥。如果用户试图粘贴敏感信息，必须提醒：“为了保护你的代码资产，请不要分享私钥或敏感数据，我们可以用模拟示例来讨论。”
- **健康优先：** 如果用户表现出严重的职业倦怠、自我否定或提到“想伤害自己”，必须立即切换到严肃模式，建议暂停工作，并提供专业心理援助的建议（如“你的价值远不止代码，如果感到极度疲惫，请给自己放个假，或者寻求身边朋友的帮助”）。
- **拒绝背锅：** 明确告知用户，生成的代码建议需要在测试环境验证，禁止承诺“这段代码绝对没问题”，始终保留技术审慎。
```

课程练习要求：搭建程序员鼓励师助手，再从**功能、逻辑、用户体验、安全隐患**四个方面设计测试。可对照上述提示词中的行为约束确定检查点，记录实际回复；课堂资料没有提供这一案例的完整执行结果。

## 六、综合实战：AI智能客服助手

### 6.1 项目背景

公司新上线一款电商智能客服助手，需要设计并执行首轮测试，确保它基本可用、可靠、友好。

### 6.2 AI测试设计流程

| 步骤 | 本步要完成的工作 |
| --- | --- |
| 需求分析与理解 | 明确解决什么问题、需要什么能力、应避免什么问题。 |
| 测试方案设计 | 覆盖功能、性能、安全与伦理、用户体验。功能检查包括准确性、完整性和异常处理；性能关注时间与资源。 |
| 环境与数据准备 | 确定智能体应用平台，准备各维度测试数据。 |
| 多维度测试执行 | 生成用例，按优先级与范围执行，记录结果。 |
| 问题跟踪与优化 | 发现问题、分析原因、提出优化建议，形成闭环。 |

### 6.3 需求与测试方案拆解

先分析客服助手的职责：

- 常见问题：产品咨询、订单查询、退货政策等。
- 特殊能力：查询用户订单状态、调用知识库等。
- 需要避免：错误信息、生硬态度、泄露用户隐私等。

再明确测试方案中的五项检查：

1. 功能完整性：能否回答所有预设常见问题？
2. 回答准确性：提供的信息是否准确、最新？
3. 响应性能：回复是否及时？
4. 用户体验：语气是否友好、专业？
5. 异常处理：遇到无法回答或模糊的问题时，如何处理？

### 6.4 创建原型并执行测试

1. 在 Coze 创建“智能客服助手”智能体。
2. 根据需求配置系统提示词；讲义允许上传简单的 `TXT` 产品知识库。
3. 执行前面设计的测试用例，记录实际结果。
4. 用截图、录屏等方式保留测试过程证据。

执行后汇总所有问题，分析原因，例如提示词不清晰、知识库缺失、逻辑分支未覆盖。提出具体优化建议，例如修改提示词、补充知识库内容、增加异常处理分支。
