## 一、认识软件测试

**软件测试**：通过技术手段运行软件，检查软件是否满足需求的过程。

测试的目的，是减少软件缺陷（bug），保障软件质量。

## 二、软件测试技能

| 技能 | 说明 |
| --- | --- |
| **功能测试** | 主要验证程序的功能是否满足需求。 |
| **自动化测试** | 使用代码或工具代替人工验证项目功能。 |
| **接口测试** | 针对模块与模块、系统与系统之间数据交互的测试。 |
| **性能测试** | 模拟多人使用软件，查找服务器缺陷。 |

## 三、软件测试分类

> **应用场景**：这些分类常用于面试中被面试官问到或提及时的回答。

### 3.1 按照阶段划分

| 阶段 | 说明 | 通常负责角色 |
| --- | --- | --- |
| **单元测试** | 对开发的源代码进行测试。 | 开发人员 |
| **集成测试** | 也叫接口测试，测试系统与系统、模块与模块之间的数据交互是否正常。 | 测试人员 |
| **系统测试** | 也叫功能测试，测试整个软件产品的功能是否满足需求，也包含兼容性、文档等测试。 | 测试人员 |
| **验收测试** | 模拟用户验证软件是否满足用户需求。 | 用户或用户代表 |

验收测试按发布范围还可以分为：

- **内测**：alpha 测试（α）。
- **公测**：beta 测试（β）。
- **候选版**：gamma 测试（γ）。

### 3.2 按照代码可见度

| 类型 | 说明 |
| --- | --- |
| **黑盒测试** | 看不到源代码，进行功能级别的测试。 |
| **灰盒测试** | 部分代码可见，相当于做接口测试。 |
| **白盒测试** | 通过源代码测试源代码，通常对应单元测试。 |

## 五、AI 应用

### 5.1 使用步骤

1. 选择工具。
2. 编写提示词。
3. 优化结果。

### 5.2 AI 提示词模板

指令可以按下面的典型格式组织：

| 要素 | 要回答的问题 | 测试场景示例 |
| --- | --- | --- |
| **角色（身份）** | 你是谁？ | 你是测试工程师。 |
| **指示（任务）** | 要做什么？ | 分析登录功能需求。 |
| **输入（要求）** | 有什么具体要求？ | 账号必填，使用已注册的合法手机号；密码必填，且与注册密码一致。 |
| **背景（当前状况，可选）** | 当前情况是什么？ | 当前用户对软件测试不熟悉。 |
| **输出（结果）** | 希望得到什么结果？ | 将补充完善后的需求以 Markdown 形式输出。 |
| **参考（示例，可选）** | 是否需要模仿示例？ | 按照提供的示例输出，使结果更贴近要求。 |

> 提示词的基本思路是：先明确角色和任务，再给出输入约束；必要时补充背景、输出格式和参考示例。

## 六、测试设计

### 6.1 等价类划分法

#### 使用场景

适用于**表单类型**的页面。

#### 步骤

1. 明确需求：确定测试目的和条件。
2. 按每个条件划分等价类：有效类、无效类。
3. 根据有效类和无效类的组合梳理测试点。

#### 注意事项

- **有效**：所有条件都有效，测试点数看单条件有效类中的最大数。
- **无效**：只要有一个条件或一个项无效即可，测试点数是每个无效类数量之和。

### 6.2 边界值分析法

#### 使用场景

适用于针对**边界范围**数据进行测试的场景。

#### 步骤

1. 明确需求：确定测试目的和条件。
2. 按每个条件划分等价类：有效类、无效类。
3. 确定边界范围值：上点、离点、内点；这一步可以和等价类划分合并。
4. 根据有效类和无效类的组合梳理测试点。

#### 注意事项

- 使用边界值分析法必须存在边界范围。
- 离点可以优化：开区间选择内部离点测试，闭区间选择外部离点测试。
