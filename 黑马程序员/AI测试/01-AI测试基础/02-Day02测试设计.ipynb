# Day02课堂笔记

## 1. 测试用例

编写测试用例的前提，是已经在 XMind 中梳理完测试点。先确定要验证哪些条件，再把测试点整理成可以执行的用例。

用例可以由人工编写，也可以用 AI 辅助编写。使用 AI 时，把已有测试点和参考示例一起提供给它，让输出覆盖这些测试点，并采用需要的文档格式。

### 1.1 用 AI 编写测试用例

原课以电商项目为例：让 AI 以软件测试工程师的身份，依据注册页面和测试点编写用例，先输出 Markdown 文档，最终转换成 Excel 表格。

> 原笔记的引导文字写的是“用户登录测试用例”，但实际提示词和测试点针对的是**注册页面**。下面按提示词中的注册场景保留内容。

```text
你以软件测试工程师身份，帮我编写电商注册页面测试用例（参考附件截图）
要求：覆盖如下的测试点
商城注册测试点
    注册成功
        设置密码16位字母+数字+符号，推荐人手机为空，其他条件都满足
        设置密码6位字母，推荐人手机已注册，其他条件都满足
        设置密码6~16位数字，推荐人手机为空，其他条件都满足
        .....
    注册失败
        手机号已注册
        手机号为空
        手机号格式错误
        验证码错误
        验证码过期
        验证码为空
        ......
输出md格式的文档，最终转为Excel表格
参考示例如附件demo
```

提示词中的省略号来自原课，表示这里展示的是部分测试点；实际使用时，应填入自己已经梳理好的完整测试点。

### 1.2 参考用例格式

原课提供的 `demo` 截图，是一条 QQ 登录成功用例，用于说明输出表格的格式。

![测试用例参考示例](02-demo.jpg)

| 字段 | 原课示例 |
| --- | --- |
| 用例编号 | `qq_login_001` |
| 用例标题 | 登录成功（6 位长度账号 + 密码都有效） |
| 模块/项目 | QQ 登录 |
| 优先级 | `P0` |
| 预置条件 | 账号已注册 |
| 测试步骤 | 1. 打开 QQ 软件；2. 输入 QQ 账号和密码；3. 点击登录。 |
| 测试数据 | QQ：`654456`；密码：`Aa123456`（正确密码）。 |
| 预期结果 | 登录成功 |

参考图说明的是用例字段和组织方式；生成注册用例时，具体步骤、数据和预期结果应对应前面提供的注册测试点。

## 2. 测试执行

### 2.1 执行准备

开始执行前，需要满足两个条件：

1. 已经完成测试用例。
2. 已经有可运行的测试环境。原课提到环境由运维搭建，同时提出思考：测试人员是否具备搭建环境的能力？

### 2.2 执行顺序与结果记录

1. 先对**核心业务流程**进行冒烟测试，执行正向用例。冒烟测试用于先确认核心流程能够正常运行。
2. 再执行其他功能模块和业务的测试用例，并做好记录。

| 执行结果 | 含义 |
| --- | --- |
| `pass` | 通过 |
| `fail` | 失败 |

### 2.3 扩展：测试环境

| 环境 | 原课说明 |
| --- | --- |
| 本地测试环境 | 测试人员自己使用。 |
| 预生产环境 | 上线前所有测试人员都能使用。 |
