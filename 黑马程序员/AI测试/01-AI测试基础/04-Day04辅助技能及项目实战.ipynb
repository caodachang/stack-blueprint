# Day04辅助技能及项目实战

## 课程内容

本节围绕测试执行中的抓包辅助技能，以及电商项目的测试设计展开：

1. 抓包涉及的 URL、HTTP 请求与响应。
2. 使用 Charles 定位前后端问题、模拟弱网、修改请求做服务端异常测试。
3. 按项目测试流程推进测试工作。
4. 用业务流程图法、测试设计方法和质量模型设计测试点，再借助 AI Agent 生成 Excel 用例。

## 一、测试执行辅助：抓包应用

### 1.1 网络知识

#### URL

URL 表示资源在网络中的存储位置，课程给出的结构为：

```text
协议://域名或IP:端口号/资源路径?查询参数
```

原脑图将协议、域名和端口列为必填部分，主机位置也可以使用 IP 地址。

> 补充：使用协议默认端口时，可以省略端口号，例如 HTTP 默认端口 `80`、HTTPS 默认端口 `443`。

#### HTTP 请求

HTTP 是超文本传输协议，原课概括为主要传输超文本信息的数据协议。抓包时需要理解请求与响应的构成。

| 请求的组成 | 内容 |
| --- | --- |
| 请求行 | 请求方法、请求 URL、协议版本。 |
| 请求头 | 客户端属性信息。 |
| 请求体 | 客户端发送请求时携带的数据。 |

课程以增删改查说明请求方法：

| 方法 | 课程中的用途 |
| --- | --- |
| `GET` | 查询。 |
| `POST` | 新增。 |
| `PUT` | 修改。 |
| `DELETE` | 删除。

#### HTTP 响应

HTTP 响应由响应行、响应头和响应体组成：

| 部分 | 作用 |
| --- | --- |
| 响应行 | 包含协议和版本、响应状态码、结果描述。 |
| 响应头 | 服务端属性信息。 |
| 响应体 | 服务器返回给客户端的数据，例如图片、HTML、JSON。 |

状态码按范围理解：

| 范围 | 含义 |
| --- | --- |
| `2xx` | 请求成功。 |
| `3xx` | 重定向。 |
| `4xx` | 客户端错误。 |
| `5xx` | 服务端错误。

### 1.2 抓包目的

#### 定位前后端 bug

通过抓包观察请求和响应，帮助开发快速处理 bug。课程给出的判断顺序是：

1. 先确认请求是否有问题。请求有问题，定位到前端。
2. 再确认响应是否有问题。响应有问题，定位到后端。
3. 请求、响应都没有问题，但页面显示错误，定位到前端。

![抓包定位前后端 bug](04-img/抓包定位前后端bug.png)

#### 模拟弱网测试

目标是模拟软件系统在不同网络带宽下能否正常使用。

1. 在 Charles 中设置网络带宽，根据需求选择 `3G`、`4G` 等网络条件。
2. 开启弱网开关，操作客户端进行验证，找出耗流量最大的页面或接口。
3. 查看数据包对应的延迟时间和客户端表现：观察 `Overview -> Duration` 的值，并检查页面是否显示不全等异常。

#### 绕过界面做服务端测试（异常数据）

通过断点修改请求，构造页面无法直接提交的异常数据，核心目的是验证服务端逻辑。课程步骤如下：

1. 在 Charles 的 `Proxy` 中找到 `Breakpoints` 设置。
2. 添加需要设置断点的 URL，并勾选 `Request`。
3. 在浏览器打开指定 URL，先输入正确数据。
4. 在 Charles 中开启断点开关。
5. 点击浏览器操作按钮，请求被拦截后自动跳转到 Charles。
6. 在 Charles 的请求编辑菜单中，将数据改成待测的异常数据，点击 `Execute`；后续关联的 URL 直接点击 `Execute`。
7. 执行完成后查看页面结果，确认是否符合预期。
8. 测试结束后关闭断点。

原脑图提到断点设置操作可参考配套 PDF。课程还指出抓包可以用于 **Web 和 App**。

### 1.3 抓包原理与设置

在客户端和服务端之间设置网络代理，由抓包工具观察双方交互的数据包。

客户端（浏览器）的代理配置为：

| 配置项 | 课程设置 |
| --- | --- |
| 代理服务器 | 安装了抓包工具的电脑 IP 地址。 |
| 端口号 | Charles 默认端口 `8888`，应与工具实际配置一致。 |

完成代理设置后，开启抓包工具，通过客户端访问服务器，再到工具的会话列表中找到对应数据包。

## 二、项目测试流程

项目测试流程用于指导测试人员开展具体测试工作，课程给出的顺序是：

1. 需求分析与评审。
2. 编写测试计划。
3. 测试设计（用例）。
4. 测试执行（用例）。
5. 缺陷管理。
6. 测试报告。

## 三、测试设计思路

### 3.1 核心业务测试设计

有业务流程图时，使用**流程图法**进行功能测试。

没有业务流程图时，可以向产品经理获取，也可以自己绘制：先熟悉整个系统的业务，再使用工具画出业务流程图。

### 3.2 单功能测试设计

1. 熟悉需求文档和 UI 原型图，确定测试依据。
2. 结合测试设计方法与质量模型设计测试点，可以用 AI 辅助。
3. 将设计好的测试点转换成用例，可以使用 AI Agent 生成 Excel。

第二步从以下方向展开：

| 维度 | 课程中的测试方向 |
| --- | --- |
| 功能：显示 | 数据正确；页面布局与原型图一致。 |
| 功能：操作 | 使用等价类、边界值、判定表等测试设计方法。 |
| 非功能 | 兼容性、易用性、安全性、性能。

## 四、电商项目：核心业务测试设计

课堂资料中的 `电商项目测试.xmind` 将核心业务分为购物车下单、商家发货和售后业务。

### 4.1 购物车下单购买业务

| 类型 | 流程或测试场景 |
| --- | --- |
| 正向 | 搜索商品 → 加入购物车 → 结算 → 下单 → 支付 → 待发货，验证下单购买成功。 |
| 逆向 | 搜索商品 → 加入购物车 → 库存不足。 |
| 逆向 | 搜索商品 → 加入购物车 → 结算 → 登录失败。 |
| 逆向 | 搜索商品 → 加入购物车 → 结算 → 下单 → 商品不存在（已下架）。 |
| 逆向 | 搜索商品 → 加入购物车 → 结算 → 下单 → 支付 → 余额不足。

### 4.2 商家发货业务

商家发货的正向流程是：**商家登录 → 确认订单 → 填写必填项 → 发货成功**。

逆向场景包括：

- 商家登录后确认订单无效，发货失败。
- 商家登录并确认订单后，必填项为空，发货失败。

![商家发货业务流程图](04-img/发货业务流程.jpg)

### 4.3 售后业务

原脑图的这一分支用于熟悉售后业务。登录后申请售后，可以选择**仅退款、退货退款、换货**。其中仅退款和换货只列出分支，退货退款展开了以下步骤和判断：

1. 填写必填项，提交并判断是否成功。
2. 商家审核，判断审核是否通过。
3. 用户发货，填写必填项，提交并判断是否成功。
4. 判断用户是否取消。
5. 商家收货。
6. 商家退款，检查账户是否正常。
7. 退款成功。

这些判断节点是梳理流程分支的依据；原脑图没有逐条展开所有逆向用例。

## 五、电商项目：单模块测试设计

原脑图用登录模块展示功能与非功能测试点，条数写作 `x条`，没有给出固定总数。

### 5.1 功能测试点

**显示**：页面信息显示正确，页面布局与 UI 原型图一致。

**操作**：

| 类型 | 原脑图中的测试点 |
| --- | --- |
| 登录成功 | 账号为已注册、格式正确的手机号，其他条件都满足；账号为已注册、格式正确的邮箱，其他条件都满足。 |
| 登录失败 | 未注册手机号、未注册邮箱、手机号格式错误等。原图使用省略号，没有列出完整清单。 |
| 其他操作 | 能跳转到忘记密码页面；能使用微信授权登录；能使用 QQ 授权登录；能跳转到立即注册页面；点击图片验证码能够更新。

### 5.2 非功能测试点

原脑图列出四个方向：兼容性、性能、安全、易用性，未继续展开具体测试点。

### 5.3 单模块分析：登录输入项

原脑图继续将账号、密码、验证码分别划分为有效类和无效类：

| 输入项 | 有效类 | 无效类 |
| --- | --- | --- |
| 账号 | 已注册且格式正确的手机号；已注册且格式正确的邮箱。 | 未注册手机号；未注册邮箱；手机号格式错误；邮箱格式错误；空。 |
| 密码 | 与账号对应一致的密码。 | 错误密码；空。 |
| 验证码 | 正确且未过期。 | 错误；过期；空。 |

手机号格式错误包括：

- 小于 `11` 位。
- 大于 `11` 位。
- 非数字（原脑图写作“非自然数”）。
- 第一位不是 `1`。
- 第二位小于 `3`。

邮箱格式错误包括：用户名为空、没有 `@`、没有域名、多个 `@`。

> 补充：这些格式条件按课程项目保留，实际设计时应以被测项目的账号校验规则为依据。

### 6.2 电商下单用例

#### 正向：`order_001`

- **标题**：购物车下单购买成功（搜索商品 → 加入购物车 → 结算 → 下单 → 支付 → 待发货）。
- **模块、优先级**：下单业务，`P0`。
- **前置条件**：库存充足、账号密码正确、商品已存在、余额充足。
- **测试数据**：商品“小米手机”。

测试步骤：

1. 进入首页。
2. 搜索指定商品。
3. 进入商品详情，添加购物车。
4. 进入购物车进行结算。
5. 提交订单。
6. 选择在线支付（微信），支付成功。

**预期结果**：下单成功，订单状态是“待发货”。

#### 逆向：`order_002`、`order_003`

两条用例所属模块均为“下单业务”，优先级均为 `P1`，测试商品均为“小米手机”。

| 编号 | 前置条件 | 测试步骤 | 预期结果 |
| --- | --- | --- | --- |
| `order_002` | 库存不足。 | 进入首页 → 搜索指定商品 → 进入商品详情，添加购物车失败。 | 下单失败，库存不足。 |
| `order_003` | 库存充足，账号密码错误。 | 进入首页 → 搜索指定商品 → 进入商品详情并添加购物车 → 进入购物车进行结算失败。 | 下单失败，登录失败。 |

> 原表将 `order_003` 的标题写为“购物车下单购买成功的业务”，但其流程终止于“登录失败”，前置条件与预期结果也都对应失败。这里按实际场景整理为逆向用例。

### 6.3 文章发布演示用例

#### 业务用例：`article_001`

标题为“发布文章成功（登录 → 发布文章 → 审核通过）”，模块为“文章发布业务”，优先级 `P0`，前置条件在原表中填为 `/`。

测试步骤：

1. 进入自媒体系统登录页面。
2. 输入登录账号和密码，点击登录。
3. 进入文章发布页面，提交文章内容信息。
4. 进入后台管理系统登录页面。
5. 输入后台管理登录信息并登录。
6. 进入内容审核，选中对应文章并审核通过。

测试数据包括自媒体账号、系统管理员账号及各自密码，以及文章标题、正文、标签、频道、时间、是否有封面图。课程示例中两种账号名均为 `demo`，执行时使用与测试环境对应的账号密码。

**预期结果**：发布文章审核通过；发布内容列表显示已审核的文章，内容审核列表显示已发布文章。
