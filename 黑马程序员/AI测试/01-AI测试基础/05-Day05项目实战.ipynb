# Day05项目实战

## 一、项目介绍

### 1.1 项目是什么

本节以 **TPSHOP 电商项目**为例开展项目微实战。项目属于电商类型，课堂资料同时从 B/S 和 C/S 两种形态介绍用户入口：

- **B/S 架构**：普通用户端、后台管理端通过浏览器访问。
- **C/S 架构**：用户端以客户端或 App 形式提供服务。

项目系统由 **Web 系统和 App 系统**组成。

### 1.2 项目给谁用

项目主要服务三类角色：

| 角色 | 典型职责 |
| --- | --- |
| 普通用户 | 浏览和搜索商品、加入购物车、下单、支付、申请售后。 |
| 商家用户 | 管理商品、处理订单、发货、处理售后申请。 |
| 系统管理员 | 管理用户、审核商家、配置系统、查看全局数据。 |

### 1.3 核心功能

#### 业务线

- 购物下单流程。
- 商家发货流程。
- 用户售后流程。

#### 单功能模块

- 登录。
- 购物车。
- 下单。
- 支付。
- 以及其他业务模块。

项目测试需要同时关注跨角色的业务链路和单页面、单接口的功能。

## 二、项目测试

### 2.1 测试设计

#### 设计思路

1. **熟悉需求**：阅读需求文档和 UI 原型，先理解产品目标和业务规则，保证对需求的理解一致。
2. **明确测试目的和依据**：确认当前功能要解决什么问题，哪些文档和原型是测试依据。
3. **整理测试点**：
   - 功能测试包括显示和操作。
   - 操作测试按照等价类、边界值、判定表等用例设计方法继续细化。
   - 非功能测试包括性能、兼容性、易用性和安全。
4. **编写测试用例**：可以按模板手写，也可以使用 AI 工具把测试点转换成测试用例。课程将 AI 工具生成用例作为首选辅助方式，但生成结果仍要人工检查。

### 2.2 测试执行

测试执行的前提是先完成冒烟测试：

1. 优先执行核心业务中的正向用例，确认主流程可以运行。
2. 冒烟测试通过后，再全面执行其他测试用例。
3. 记录每条用例的执行结果：
   - `pass`：执行通过。
   - `fail`：执行失败，需要结合缺陷管理流程处理。

### 2.3 缺陷管理

测试执行发现问题后，按缺陷生命周期推进：

1. **提 Bug**：记录模块、标题、前置条件、重现步骤、期望结果、实际结果和附件。
2. **验证 Bug**：开发修复后，测试人员按原步骤复现，确认修复是否有效，并检查是否引入回归问题。
3. **关闭或重新打开**：验证通过则关闭；仍能复现或修复不符合预期则重新打开。

缺陷状态必须与实际验证结果一致，不能仅根据开发标记的“已修复”直接关闭。

## 三、测试报告

### 3.1 作用和编写人

测试报告的标志是**测试活动结束**。它用于对产品质量进行详细说明和总结，也是对测试结果进行评估的依据。

课程给出的协作方式是：领导提供报告模板，其他测试人员填写具体内容，填写过程可以借助 AI。

### 3.2 编写前提

开始编写测试报告前，应确认：

- 测试用例已经全部执行完毕。
- Bug 已经全部修改完毕，并且没有遗留问题。

如果仍有未执行用例或未闭环缺陷，应在报告中如实反映，不能直接给出“可以发布”的结论。

### 3.3 报告内容

课堂脑图将报告分成四个核心部分：

| 部分 | 应包含的内容 |
| --- | --- |
| 项目概述 | 项目类型、系统架构、用户角色、核心业务和测试范围。 |
| 过程回顾 | 测试时间、人员、资源和承担的任务。 |
| 统计分析 | 用例总数、执行数、通过与失败数量、通过率，以及 Bug 数量和严重级别分布。 |
| 结果确认 | 判断测试目标是否达成、是否存在遗留风险、是否符合上线条件。 |
| 总结改进 | 保留做得好的实践，针对问题提出后续改进。 |

## 四、使用 AI 生成测试报告

### 4.1 提示词示例

课程给出的提示词要求 AI 以软件测试工程师身份，根据项目背景和测试数据输出 Markdown 报告：

```text
你以软件测试工程师的身份，帮我编写电商项目的测试报告
要求：
1.报告内容：项目概述、过程回顾、统计分析、结果确认、总结改进
2.项目概述：基于B/S架构的电商项目，主要用户有普通用户、商家、管理员等；其中核心业务包含下单购买业务、商家发货业务、用户售后业务等；包含的核心模块有登录、搜索、购物车、下单、支付、发货等
3.过程回顾：总共3个测试人员，持续时间3个月，电脑3台，手机若干；主要负责测试功能
4.统计分析：总共编写800条用例，发现bug 24条，其中中等级别占比约70%
5.结果确认：能否正常发布，是否符合上线条件
6.总结改进：过程中好的地方保留，不好的地方改进
输出一份md格式的文档
```

使用时，测试人员要把真实项目数据替换提示词中的示例数据，并在输出后核对统计数字、上线结论和未闭环风险。

### 4.2 AI 输出后的检查

重点检查以下内容：

1. 报告章节是否完整，是否覆盖项目概述、过程回顾、统计分析、结果确认和总结改进。
2. 用例数、执行数、通过数、失败数和通过率是否能够相互对应。
3. Bug 总数与严重级别分布之和是否一致。
4. 报告中的“允许发布”或“不能发布”是否有测试结果和遗留风险支撑。
5. 总结中的优点和改进项是否来自真实测试过程，而不是泛泛生成。

## 五、课堂报告示例：电商平台 V2.0

配套资料 `电商平台V2.0 系统测试报告.md` 是一份完整的 Markdown 报告示例。报告属性如下：

| 属性 | 示例内容 |
| --- | --- |
| 报告名称 | 电商平台 V2.0 系统测试报告 |
| 项目名称 | 电商平台升级项目 |
| 测试版本 | V2.0 |
| 报告日期 | 2023-10-27 |
| 编写人 | 软件测试部 |

> 以上日期和项目名称是课堂示例中的内容，整理到实际项目时应替换为真实信息。

### 5.1 示例项目概述

报告示例将系统描述为基于 B/S 架构的电商平台升级版本，服务普通用户、商家用户和平台管理员。

测试范围覆盖：

- 用户登录与注册。
- 商品搜索与筛选。
- 购物车管理。
- 订单创建与结算。
- 在线支付（模拟）。
- 用户侧和商家侧订单管理。
- 发货与物流跟踪。
- 售后申请与处理。

核心业务围绕下单购买、商家发货和用户售后展开。

### 5.2 示例过程回顾

报告中的团队与资源：

- 3 名软件测试工程师。
- 测试周期 3 个月，包含测试计划、用例设计、执行和回归测试。
- 3 台测试 PC，覆盖不同操作系统和浏览器。
- 若干移动设备，用于兼容性和移动端专项测试。

测试策略以功能测试为核心，使用黑盒测试方法，依据需求规格说明书和设计文档设计用例，并覆盖功能测试、集成测试、UI 测试、兼容性测试和回归测试。下单、支付、发货等核心链路进行了端到端测试。

### 5.3 示例统计分析

#### 测试用例执行统计

| 用例类型 | 用例数量 | 执行数量 | 通过数量 | 失败数量 | 通过率 |
| --- | ---: | ---: | ---: | ---: | ---: |
| 功能测试用例 | 800 | 800 | 788 | 12 | 98.5% |
| 总计 | 800 | 800 | 788 | 12 | 98.5% |

#### 缺陷严重级别

| 严重等级 | 数量 | 占比 | 示例定义 |
| --- | ---: | ---: | --- |
| 致命（Critical） | 0 | 0% | 系统崩溃、数据丢失或核心功能完全失效。 |
| 严重（Major） | 2 | 8.3% | 主要功能错误，影响核心业务流程，但系统仍可运行。 |
| 中等（Medium） | 17 | 70.8% | 次要功能错误，不影响主体流程，但影响用户体验。 |
| 轻微（Minor） | 5 | 20.8% | 界面、提示信息等瑕疵，不影响功能使用。 |
| 总计 | 24 | 100% |  |

#### 缺陷闭环和模块分布

示例报告记录：

- 已修复 24 个缺陷，占 100%。
- 已验证关闭 24 个缺陷，占 100%。
- 重新打开 0 个。
- 缺陷最多的模块前三名：订单与支付 9 个（37.5%）、商家后台管理 7 个（29.2%）、用户前台交互 5 个（20.8%）。

统计分析既要展示数量，也要说明缺陷集中在哪些模块和严重级别，从而支持风险判断。

### 5.4 示例结果确认

报告示例的结果确认包含三层：

1. 所有计划测试用例已 100% 执行。
2. 24 个缺陷已全部修复并验证，没有遗留缺陷；登录、搜索、加购、下单、支付、发货、售后等核心业务流程验证通过。
3. 没有致命级别缺陷，严重级别缺陷已经修复，中等及以下级别问题主要是交互优化和边界条件问题，风险可控。

据此，示例报告给出的结论是：V2.0 功能稳定、核心流程顺畅，符合既定上线质量标准，建议发布上线。这个结论只适用于示例中的数据，实际项目必须重新评估。

### 5.5 示例总结与改进

#### 需要保留的实践

- **用例评审机制**：开发、产品和测试共同澄清需求歧义，提升用例覆盖率和准确性。
- **规范的缺陷管理流程**：遵循“提交 → 分配 → 修复 → 验证 → 关闭”，并使用 Jira 跟踪。
- **核心链路端到端测试**：对购物、支付、发货进行跨角色测试，发现模块间数据传递和状态同步问题。

#### 后续改进

- **提升测试深度**：示例中约 70% 的缺陷属于中等级别，后续加强异常流、边界值、并发和数据一致性用例，并考虑错误推测法和探索性测试。
- **优化回归策略**：建立核心功能回归用例集，研究使用 Selenium、Playwright 等自动化框架覆盖登录、下单等高复用流程。
- **测试左移**：在需求评审和设计阶段让测试人员更早介入，从可测试性角度提出建议。

## 六、复习要点

- 项目测试先理解项目、角色、系统和核心业务，再设计测试点。
- 测试执行以冒烟测试为入口，核心正向流程通过后再全面执行。
- 缺陷必须经过提交、修复、验证和关闭或重新打开的闭环。
- 测试报告不是测试用例的简单汇总，而是对测试活动和产品质量的评估依据。
- AI 适合生成报告结构和初稿，测试人员必须核对统计关系、风险和发布结论。
