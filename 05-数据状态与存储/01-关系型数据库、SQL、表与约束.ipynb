# 关系型数据库、SQL、表与约束

## 关系型数据库

关系型数据库是用“表”和“关系”组织数据的数据库。

## 表、行、列

表是同一类数据的集合，行是一条记录，列是记录里的一个字段。

例如用户表可以长这样：

| id | name | email |
| --- | --- | --- |
| 1 | 小林 | lin@example.com |
| 2 | 小周 | zhou@example.com |

`users` 是表，`id`、`name`、`email` 是列，每一行代表一个用户。

- **主键**用来唯一标识一行数据。
- **外键**用来表达两张表之间的关系。

## SQL

SQL 是关系型数据库常用的操作语言

## 约束

约束是数据库对数据合法性的限制。常见约束包括：

- `PRIMARY KEY`：主键，唯一标识一行
- `FOREIGN KEY`：外键，限制表之间的引用关系
- `NOT NULL`：不能为空
- `UNIQUE`：不能重复
- `CHECK`：必须满足某个条件
