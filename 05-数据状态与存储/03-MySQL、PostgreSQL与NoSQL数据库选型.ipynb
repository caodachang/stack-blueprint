# MySQL、PostgreSQL与NoSQL数据库选型

## MySQL 与 PostgreSQL

MySQL 是使用非常广泛的开源关系型数据库。

PostgreSQL 也是开源关系型数据库，功能更完整、SQL 能力更强。

MongoDB 是文档数据库，常用 JSON 类似的文档保存数据。

Redis 是常见的内存型键值数据库。

## NoSQL

NoSQL 是一类不以传统关系表为中心的数据库。

| 类型 | 代表 | 适合场景 |
| --- | --- | --- |
| 键值数据库 | Redis | 缓存、计数器、会话 |
| 文档数据库 | MongoDB | 结构灵活的 JSON 文档 |
| 列式数据库 | Cassandra | 大规模写入和查询 |
| 图数据库 | Neo4j | 关系网络、路径查询 |
