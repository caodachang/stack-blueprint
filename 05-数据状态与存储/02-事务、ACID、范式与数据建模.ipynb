# 事务、ACID、范式与数据建模

## 事务与 ACID

事务是一组必须一起成功或一起失败的数据库操作

ACID 是事务的四个核心性质。

| 性质 | 含义 |
| --- | --- |
| Atomicity 原子性 | 要么全部成功，要么全部失败 |
| Consistency 一致性 | 事务前后都满足数据规则 |
| Isolation 隔离性 | 并发事务之间不能随意互相干扰 |
| Durability 持久性 | 提交后的结果要可靠保存 |

## 范式

范式是关系型数据库设计表结构时的一组规范，目标是减少重复数据，降低更新出错的概率。

反范式是为了查询性能或历史快照，故意保留一些重复字段

## 数据建模

数据建模是把业务对象、对象属性、对象关系设计成可保存的数据结构。
