# 项目结构


## 项目根目录

| 文件 | 是什么 |
|------|------|
| `frontend/` | React 前端 |
| `backend/` | Python 后端 |
| `e2e/` | Playwright 端到端测试 |
| `.gitignore` | 告诉 Git 哪些文件不跟踪（`node_modules/`、`.env` 等） |
| `.env` | 环境变量（密钥、数据库地址），不提交到 Git |
| `.env.example` | 环境变量模板，告诉别人需要填什么 |
| `README.md` | 项目说明 |


## 前端（React + TypeScript + Vite）

```
frontend/
├── src/
│   ├── components/          # 可复用的 UI 组件
│   ├── pages/               # 每个路由对应的页面
│   ├── hooks/               # 自定义 Hook（封装状态逻辑）
│   ├── stores/              # 全局状态管理
│   ├── types/               # TypeScript 类型定义
│   ├── utils/               # 工具函数
│   ├── App.tsx              # 根组件
│   └── main.tsx             # 入口文件，挂载 React 到页面
├── public/                  # 静态文件，直接复制到最终产物，不经过编译
├── index.html               # HTML 壳，Vite 注入 JS 到这里
├── package.json             # 项目名、脚本、依赖声明
├── pnpm-lock.yaml           # 锁定每个依赖的精确版本
├── tsconfig.json            # TypeScript 编译规则
├── vite.config.ts           # Vite 构建配置
└── vitest.config.ts         # Vitest 测试配置
```

| 文件 | 是什么 |
|------|------|
| `package.json` | 项目的身份证——名字、版本、启动脚本、依赖列表。`npm install` / `pnpm install` 读取它决定装什么 |
| `pnpm-lock.yaml` | 锁定依赖的精确版本。保证你、同事、CI 服务器装的是完全同一版依赖 |
| `tsconfig.json` | 告诉 TypeScript 怎么检查代码（严格程度、目标浏览器版本、路径别名） |
| `vite.config.ts` | Vite 的配置——开发服务器端口、代理后端 API、构建输出目录 |
| `index.html` | 浏览器的入口 HTML。Vite 会自动把编译后的 JS 和 CSS 注入这里 |
| `public/` | 放不需要编译的文件——favicon、robots.txt、字体文件。直接复制到输出目录 |
| `src/main.tsx` | JS 的入口。ReactDOM 把 `<App />` 挂到 `index.html` 的 `<div id="root">` 上 |
| `src/components/` | 可复用的 UI 零件——按钮、弹窗、表格。每个组件一个 `.tsx` 文件 |
| `src/pages/` | 每个路由对应一个页面组件。`/login` → `LoginPage.tsx` |
| `src/hooks/` | 把组件共用的状态逻辑抽出来。`useAuth`、`useFetch` |
| `src/types/` | 接口和类型定义。`User`、`Order`、`ApiResponse` 的 TS 类型放这里 |


## 后端（Python + FastAPI + uv）

```
backend/
├── app/
│   ├── api/                 # 路由——URL 路径对应的处理函数
│   ├── models/              # SQLAlchemy 数据库表定义
│   ├── schemas/             # Pydantic 校验模型（请求和响应的格式）
│   ├── services/            # 业务逻辑（不和 HTTP 耦合）
│   ├── core/                # 配置、数据库连接、依赖注入
│   └── main.py              # FastAPI 应用入口
├── tests/                   # pytest 测试
├── alembic/
│   └── versions/            # 数据库迁移脚本
├── alembic.ini              # Alembic 数据库连接配置
├── pyproject.toml           # 项目元数据、依赖、工具配置
└── uv.lock                  # 锁定依赖版本
```

| 文件 | 是什么 |
|------|------|
| `pyproject.toml` | Python 项目的身份证。元数据（名字、作者），依赖声明（dependencies），工具配置（pytest、ruff、mypy）——全在这一个文件里 |
| `uv.lock` | uv 生成的锁文件，锁定每个依赖包和它的子依赖的精确版本。同 `pnpm-lock.yaml` |
| `app/main.py` | FastAPI 应用实例。创建 `app = FastAPI()`，注册路由，配置中间件 |
| `app/api/` | 路由层——`app/api/users.py` 里定义 `GET /users`、`POST /users`。只负责接收请求和返回响应，不写业务逻辑 |
| `app/models/` | 数据库表定义。`User` 类映射到 `users` 表，字段类型映射到列类型 |
| `app/schemas/` | 请求/响应的格式校验。`UserCreate` 校验创建用户时前端传来的数据，`UserResponse` 定义返回给前端的格式 |
| `app/services/` | 真正的业务逻辑。API 层调用它，它调用数据库模型。解耦——换框架时只改 API 层 |
| `app/core/` | 数据库连接（`database.py`）、应用配置（`config.py`）、全局依赖注入（`dependencies.py`） |
| `alembic/` | 数据库迁移工具。改完模型后运行 `alembic revision --autogenerate` 自动生成迁移脚本，`alembic upgrade head` 执行到数据库 |
| `alembic.ini` | Alembic 的数据库连接地址和基本设置 |


## 测试

| 文件 | 是什么 |
|------|------|
| `backend/tests/` | pytest 测试。测试 API 接口和业务逻辑，不依赖真实数据库 |
| `frontend/ vitest.config.ts` | Vitest 测试配置。单元测试组件和 Hook |
| `e2e/` | Playwright 端到端测试。模拟真实用户在浏览器里操作，验证前后端完整流程 |

三层测试的区别：
- **单元测试**（pytest / Vitest）：测单个函数或组件，快，不依赖外部服务
- **集成测试**：测多个模块一起工作
- **端到端测试**（Playwright）：打开真实浏览器，模拟点击、输入，验证整个应用


## 不提交到 Git 的

这些目录由工具自动生成，在 `.gitignore` 里排除：

| 目录 | 谁生成的 | 为什么不能提交 |
|------|---------|-------------|
| `node_modules/` | `pnpm install` | 几十万个文件，体积巨大 |
| `.venv/` | `uv venv` | Python 虚拟环境，和电脑绑定的 |
| `dist/` / `build/` | `pnpm build` | 编译产物，源码改了就得重新生成 |
| `__pycache__/` | Python 运行时 | `.pyc` 字节码缓存，自动生成 |
| `.env` | 你自己创建 | 包含密钥和密码，泄露等于裸奔 |
