# Git概述与基础操作

## Git的三个本地状态

| 位置 | 含义 |
| --- | --- |
| 工作区 | 你正在编辑的文件 |
| 暂存区 | 准备放进下一次 commit 的修改 |
| 本地仓库 | 已经形成 commit 的历史 |

## 从零开始的最小命令

```bash
git init   #初始化 Git 文件夹
git status #查看文件夹状态
```

## 一次本地提交怎么做

一次最小提交通常是这几步：

```bash
git status
git diff      # 查看修改了什么
git add README.md
git commit -m "更新 README 说明"

## 查看历史

```bash
git log           # 查看提交历史
git log --oneline # 友好查看
```

## 本地和远程怎么同步

```bash
git pull # 把远程的新提交拉到本地
git push # 把本地的新提交推到远程
```
