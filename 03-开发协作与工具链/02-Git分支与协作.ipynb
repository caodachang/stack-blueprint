# Git分支与协作

## branch是什么

| 概念 | 含义 |
| --- | --- |
| main / master | 常见主分支，通常代表稳定主线 |
| feature branch | 功能分支，用来开发新功能 |
| bugfix branch | 修复分支，用来修 bug |
| merge | 把一个分支的修改合进另一个分支 |
| conflict | 两边改了同一处，需要人工解决 |

## 查看和创建分支

查看当前有哪些分支：

```bash
git branch
```

创建并切换到新分支 feature/feedback-form ：

```bash
git switch -c feature/feedback-form
```

切换分支：

```bash
git switch main
```

## 在分支上提交修改

切到功能分支后，仍然按普通 Git 流程提交。

```bash
git status
git diff
git add src/feedback.ts
git commit -m "增加反馈提交逻辑"
```

## merge是什么

merge 是把一个分支的修改合进当前分支（团队协作里通常会做评审后再合并）。

常见流程：

```bash
git switch main
git merge feature/feedback-form
```

含义是：

- 先切回 `main`
- 再把 `feature/feedback-form` 的修改合进 `main`

merge 会保留“这个功能从分支合进来”的历史。

## rebase是什么

rebase 把当前分支的提交“基线”对齐到另一条分支的**最新提交**后面。

常见命令：

```bash
git switch feature/feedback-form
git rebase main
```
main 和 feature/feedback-form 依然是两个分支  
不要随便 rebase 已经推给别人共同使用的分支。它会改写提交历史，可能影响队友。

## 远程分支怎么用

把本地新分支推到远程：

```bash
git push -u origin feature/feedback-form
```

之后通常可以直接：

```bash
git push
git pull
```

| 命令 | 作用 |
| --- | --- |
| git push -u origin 分支名 | 第一次把本地分支推到远程，并建立跟踪关系 |
| git push | 推送当前分支的新提交 |
| git pull | 拉取当前分支的远程更新 |

## 一条常见协作流程

一个功能从开始到合并，通常是这样：

```bash
git switch main
git pull
git switch -c feature/feedback-form  # 新建专属功能分支并修改
git status
git add .
git commit -m "增加反馈表单"
git push -u origin feature/feedback-form
```

然后在 GitHub 上创建 PR（提交合并到 main 分支的申请），让别人评审，再合并到主分支。
