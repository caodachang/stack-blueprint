# Git与GitHub协作基础

## Git 的三个区域

工作区 -> git add -> 暂存区 -> git commit -> 本地仓库

## 初始化与查看状态

初始化为 Git 仓库

```bash
git init
```

查看仓库当前状态：

```bash
git status
```

## 暂存与提交

暂存一个文件：

```bash
git add README.md
```

暂存当前目录下所有改动：

```bash
git add .
```

把暂存区里的改动提交到本地仓库：

```bash
git commit -m "更新 README 说明"
```

## 查看历史与差异

查看提交历史：

```bash
git log
git log --oneline
```

查看文件具体改了哪里：

```bash
git diff
git diff --staged
```

## 分支

```bash
git branch
git switch -c feature-login
git switch main
git merge feature-login
```

| 命令 | 作用 |
| --- | --- |
| `git branch` | 查看本地分支 |
| `git switch -c feature-login` | 创建并切换到新分支 |
| `git switch main` | 切回 main 分支 |
| `git merge feature-login` | 把 feature-login 分支合并进当前分支 |

分支合并时产生的冲突需要人工解决

## 远程仓库

远程仓库是放在网络平台上的仓库，例如 GitHub、GitLab、Gitee。

常见命令：

```bash
git clone <仓库地址>
git remote -v
git push
git pull
```

| 命令 | 作用 |
| --- | --- |
| `git clone` | 把远程仓库复制到本地 |
| `git remote -v` | 查看本地仓库关联的远程地址 |
| `git push` | 把本地提交推送到远程仓库 |
| `git pull` | 从远程仓库拉取最新提交 |

## 第一次关联 GitHub 仓库

第一次把本地项目推到 GitHub，通常先在 GitHub 网页上创建一个空仓库，再在本地项目根目录执行：

```bash
git init
git add .
git commit -m "初始化项目"
git branch -M main
git remote add origin <GitHub仓库地址>  # 把本地仓库和 GitHub 仓库关联起来
git push -u origin main   # 对 main 推送一次后，就可以直接 git push 了
```

`origin` 是远程仓库的常用默认名字。  
`git push -u origin main` 会把本地 `main` 分支推送到远程，并建立默认上游分支。

## GitHub 协作流

一个常见的 GitHub 协作流程是：

clone 仓库 -> 新建分支 -> 修改代码 -> commit -> push -> PR（Pull Request） -> review -> merge

推送后，在 GitHub 页面上创建 PR，请别人 review，确认后再 merge。

| GitHub 功能 | 作用 |
| --- | --- |
| Repository | 在线仓库 |
| Pull Request | 提交改动并请求合并 |
| Review | 代码评审 |
| Issue | 记录任务、问题或需求 |
| Actions | 自动运行测试、构建或部署流程 |
