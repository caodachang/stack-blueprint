# uv、npm、pnpm与项目依赖管理

## uv 是什么

`uv` 是 Python 生态里的项目和依赖管理工具

- 创建或使用 Python 虚拟环境。
- 根据 `pyproject.toml` 和 `uv.lock` 安装依赖。
- 内置 Python 解释器管理，一键下载、切换多版本 Python

## uv 常见命令

| 命令 | 作用 |
| --- | --- |
| `uv sync` | 根据锁文件和项目配置同步依赖环境 |
| `uv add requests` | 添加一个运行依赖 |
| `uv remove requests` | 删除一个依赖 |
| `uv run python main.py` | 在 uv 管理的环境里运行 Python 文件 |
| `uv run pytest` | 在 uv 管理的环境里运行测试 |

## npm 是什么

`npm` 是 Node.js 自带的包管理器

- 根据 `package.json` 安装依赖。
- 生成或更新 `package-lock.json`。
- 运行 `package.json` 里定义的脚本。

Node.js：服务端 JS 的运行环境，突破 JS 只能在浏览器运行的限制

## npm 常见命令

| 命令 | 作用 |
| --- | --- |
| `npm install` | 根据 `package.json` 和锁文件安装依赖 |
| `npm install react` | 添加依赖 |
| `npm uninstall react` | 删除依赖 |
| `npm run dev` | 运行 `package.json` 里名为 dev 的脚本 |
| `npm run build` | 运行构建脚本 |

## pnpm 是什么

`pnpm` 也是 Node.js 和前端项目常用的包管理器，常更节省磁盘空间，也更强调依赖结构的严格性

## pnpm 常见命令

| 命令 | 作用 |
| --- | --- |
| `pnpm install` | 根据 `package.json` 和锁文件安装依赖 |
| `pnpm add react` | 添加依赖 |
| `pnpm remove react` | 删除依赖 |
| `pnpm dev` | 运行 `package.json` 里名为 dev 的脚本 |
| `pnpm build` | 运行构建脚本 |

## package.json 里的 scripts

`package.json` 里经常有一个 `scripts` 字段，用来给常用命令起名字。

例子：

```json
{
  "scripts": {
    "dev": "vite",
    "build": "vite build",
    "test": "vitest"
  }
}
```

这时可以执行：

```bash
npm run dev
npm run build
npm run test
```

如果使用 pnpm，很多脚本可以写成：

```bash
pnpm dev
pnpm build
pnpm test
```
