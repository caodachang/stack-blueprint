## 运行时是什么

运行时，指程序运行时依赖的一组环境和机制。

它可能包括：

- 执行代码的引擎或虚拟机。
- 标准库。
- 内存管理。
- 模块加载。
- 异步任务、线程或事件循环。
- 与操作系统交互的能力。

| 语言或生态 | 常见运行时 |
| --- | --- |
| Python | CPython |
| JavaScript 浏览器端 | 浏览器 + JavaScript 引擎 |
| JavaScript 服务端 | Node.js |
| Java | JVM |
| C# | CLR |

## 源代码到运行的大致过程

代码变为程序行为，中间通常会经历一些步骤。

源代码 -> 检查/转换 -> 执行 -> 运行时管理资源

不同语言的路径不同：

| 类型 | 大致过程 | 例子 |
| --- | --- | --- |
| 编译后运行 | 源代码先变成目标代码或字节码，再运行 | Go、Rust、Java |
| 解释执行 | 运行时读取代码并执行 | Python 常见理解 |
| JIT | 运行时边执行边把经常被执行的热点代码优化编译 | JavaScript 引擎、JVM |

现实里的实现经常是混合的。

## 静态类型、动态类型，强类型、弱类型

- 静态/动态：类型检查发生在运行前还是运行时。
- 强/弱：语言是否允许宽松的隐式类型转换。

## GC

GC （Garbage Collection）垃圾回收，作用是自动找出不再使用的内存，并把它回收。
