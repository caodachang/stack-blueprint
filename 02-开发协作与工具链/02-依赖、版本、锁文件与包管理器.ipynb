# 依赖、版本、锁文件与包管理器

## 依赖

| 类型 | 例子 | 作用 |
| --- | --- | --- |
| 运行依赖 | Web 框架、数据库驱动、UI 组件库 | 项目运行时需要 |
| 开发依赖 | 测试工具、格式化工具、打包工具 | 开发或构建时需要 |
| 间接依赖 | 依赖的依赖 | 由其他包继续引入 |

## 依赖声明文件

| 生态 | 常见依赖声明文件 |
| --- | --- |
| Python | `pyproject.toml`、`requirements.txt` |
| Node.js / 前端 | `package.json` |
| Rust | `Cargo.toml` |
| Java | `pom.xml`、`build.gradle` |

## 版本

| 写法 | 大致含义 |
| --- | --- |
| `1.4.2` | 指定一个明确版本 |
| `>=1.4.0` | 至少需要 1.4.0 |
| `<2.0.0` | 不能到 2.0.0 或更高 |
| `^1.4.2` | 允许兼容范围内的新版本 |
| `~1.4.2` | 允许较小范围内的新版本 |

## 锁文件

记录一次安装时最终解析出来的精确依赖版本。

| 生态 | 常见锁文件 |
| --- | --- |
| Python / uv | `uv.lock` |
| npm | `package-lock.json` |
| pnpm | `pnpm-lock.yaml` |
| Yarn | `yarn.lock` |
| Rust | `Cargo.lock` |

## 包管理器

| 生态 | 常见包管理器 |
| --- | --- |
| Python | `pip`、`uv`、`poetry` |
| Node.js / 前端 | `npm`、`pnpm`、`yarn` |
| Rust | `cargo` |
| Java | `maven`、`gradle` |

## 哪些文件不应该提交到 Git

| 文件或目录 | 原因 |
| --- | --- |
| `.venv/` | 本地虚拟环境，可重新生成 |
| `node_modules/` | 本地依赖目录，体积大，可重新安装 |
| `.env`、`env.local` | 本地环境变量文件，包含密钥、密码等敏感信息 |
| `__pycache__/`、`*.pyc` | Python 运行自动生成的缓存字节码文件 |
| `*.log`、`logs/` | 程序运行日志，动态生成 |
