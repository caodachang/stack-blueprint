# 把想法拆成页面、接口、数据和流程

## 四个拆解对象

| 对象 | 主要问题 |
| --- | --- |
| 页面 | 用户看到什么、点什么、填写什么 |
| 数据 | 系统需要保存什么 |
| 接口 | 页面、后端或外部系统之间交换什么能力 |
| 流程 | 用户和系统按什么顺序完成任务 |

## 以反馈功能为例

| 对象 | 拆解结果 |
| --- | --- |
| 页面 | 反馈提交页、我的反馈列表 |
| 数据 | Feedback：id、user_id、type、title、description、contact、created_at |
| 接口 | 提交反馈、获取我的反馈列表 |
| 流程 | 打开页面 -> 填写表单 -> 校验 -> 提交 -> 成功或失败 |

## 在规格里怎么写

| 模块 | 内容 |
| --- | --- |
| 页面 | 反馈提交页；包含类型、标题、描述、联系方式和提交按钮 |
| 数据 | Feedback 对象，保存提交人、类型、标题、描述、联系方式和创建时间 |
| 接口 | 提交反馈；获取当前用户的反馈列表 |
| 流程 | 用户填写表单并提交，系统校验后保存，返回成功或失败状态 |
| 权限 | 用户必须登录；只能查看自己的反馈 |
