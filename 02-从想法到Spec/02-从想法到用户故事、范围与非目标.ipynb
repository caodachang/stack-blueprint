# 从想法到用户故事、范围与非目标

## 三步收敛

| 步骤 | 作用 | 问题 |
| --- | --- | --- |
| 用户故事 | 把功能放回用户目标 | 谁为了什么要做什么？ |
| 功能范围 | 定义当前版本能力 | 这一版具体做什么？ |
| 非目标 | 排除当前版本不做的内容 | 哪些合理功能先不做？ |

## 第一步：写用户故事

常见格式：

> 作为某类用户，我希望做某件事，以便获得某个结果。

| 功能名 | 用户故事 |
| --- | --- |
| 搜索 | 作为资料库用户，我希望按关键词搜索文章，以便快速找到内容 |
| 收藏 | 作为学习者，我希望收藏重要文章，以便之后复习 |
| 提交反馈 | 作为登录用户，我希望提交产品反馈，以便团队了解问题和建议 |

## 第二步：写当前范围

范围不要停在功能名，要写到能判断是否属于当前版本。

| 用户故事 | 当前范围 |
| --- | --- |
| 用户希望搜索文章 | 支持按标题和正文关键词搜索，结果按更新时间倒序显示 |
| 用户希望收藏文章 | 登录用户可以收藏和取消收藏，详情页显示收藏状态 |
| 用户希望提交反馈 | 填写类型、标题、描述、联系方式并提交，系统保存反馈 |

## 第三步：写非目标

非目标要明确，不要写成“高级功能以后再说”。

| 当前范围 | 非目标 |
| --- | --- |
| 提交文字反馈 | 不支持图片上传 |
| 保存反馈记录 | 不发送邮件通知 |
| 查看我的反馈 | 不做管理员后台 |
| 按状态筛选 | 不做导出和统计图 |

## MVP、V1和后续版本

MVP 是最小可行产品；V1 是第一个正式版本。

| 版本 | 重点 | 反馈系统例子 |
| --- | --- | --- |
| MVP | 验证核心流程 | 用户提交文字反馈并看到成功提示 |
| V1 | 补齐基础可用性 | 登录关联、类型选择、失败提示、反馈列表 |
| 后续 | 扩展效率和管理能力 | 图片上传、邮件通知、统计、工单流转 |
