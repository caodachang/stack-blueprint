# 验收标准、边界条件与完成定义

## 验收标准

验收标准要具体、可检查。

| 模糊标准 | 可验收标准 |
| --- | --- |
| 搜索正常 | 输入关键词后，只显示标题或正文包含关键词的文章 |
| 登录可用 | 邮箱和密码正确时登录成功，错误时显示“邮箱或密码错误” |
| 数据保存成功 | 提交后保存 user_id、type、title、description、created_at |

## 边界条件和状态

| 类型 | 要问什么 |
| --- | --- |
| 空状态 | 没有数据时显示什么？ |
| 错误状态 | 保存失败、网络失败时怎么办？ |
| 输入边界 | 为空、超长、格式错误怎么办？ |
| 操作边界 | 重复提交、刷新、返回怎么办？ |
| 权限边界 | 未登录、权限不足、访问别人数据怎么办？ |

## 完成定义

- 范围内功能已实现
- 验收标准逐条通过
- 空状态、错误状态和边界条件已处理
- 权限规则符合规格
- 没有无关改动或无关依赖
- 必要文档、配置示例已更新

## 在规格里怎么写

| 项目 | 内容 |
| --- | --- |
| 验收标准 | 合法输入可以提交；成功后显示确认信息；失败有提示 |
| 空状态 | 没有历史反馈时显示“暂无反馈” |
| 错误状态 | 提交失败时保留输入，允许重试 |
| 输入边界 | 标题和描述必填；描述最多 2000 字 |
| 完成定义 | 功能、边界、权限、文档检查通过 |
