# 如何写一个AI可执行的Spec

## 为什么需要它

前面几篇分别解决了：概念边界、用户故事和范围、验收和边界、页面接口数据流程。

这一篇把它们组装成一份可以交给 AI 执行的 Spec。

## 推荐结构

| 顺序 | 小节 | 作用 |
| --- | --- | --- |
| 1 | 目标 | 说明解决什么问题 |
| 2 | 用户故事、范围、非目标 | 收敛当前版本边界 |
| 3 | 页面、数据、接口、流程 | 拆成可实现对象 |
| 4 | 边界条件和状态 | 补充空、错、加载、权限不足等情况 |
| 5 | 验收标准 | 说明怎样判断完成 |
| 6 | 实现约束 | 说明沿用什么、不要改什么 |

## 示例：反馈提交功能

| 模块 | 内容 |
| --- | --- |
| 目标 | 收集登录用户对产品的问题和建议 |
| 用户故事 | 作为登录用户，我希望提交反馈，以便团队了解问题 |
| 范围 | 提交文字反馈，保存记录，显示成功或失败状态 |
| 非目标 | 不做图片上传、后台管理、邮件通知 |
| 页面 | 类型、标题、描述、联系方式、提交按钮 |
| 数据 | user_id、type、title、description、contact、created_at |
| 接口 | 提交反馈、获取我的反馈列表 |
| 边界 | 必填为空不能提交；失败保留输入；提交中按钮禁用 |
| 验收 | 合法输入可提交；失败有提示；不会重复提交 |
| 约束 | 沿用现有项目结构，不新增无关依赖 |
