# 需求、PRD、Spec与SDD

## 四个词的边界

| 词 | 主要回答 | 典型内容 |
| --- | --- | --- |
| 需求 | 为什么做，谁需要 | 用户、问题、目标 |
| PRD | 产品做什么，用户怎么用 | 功能、页面、流程、版本范围 |
| Spec | 系统怎样表现，怎样验收 | 数据、接口、权限、状态、验收标准 |
| SDD | 如何用规格驱动开发 | 先写清规格，再拆任务、实现、检查 |

SDD 有时也指 Software Design Document，也就是软件设计文档。

这本手册里主要用 **Spec-Driven Development，规格驱动开发** 这个意思。

## 更准确的过程

更准确的说法是：

想法 -> 需求 -> 产品表达 -> 可执行规格 -> 按规格实现和验收

| 阶段 | 变化 |
| --- | --- |
| 想法 | 做一个待办工具 |
| 需求 | 学习者想集中记录任务，避免漏做 |
| 产品表达 | 可以创建待办、标记完成、查看今日任务 |
| 可执行规格 | 待办字段、页面状态、筛选规则、验收标准都写清 |
| 按规格实现 | 人或 AI 按规格拆任务、写代码、检查结果 |

## 面试常见问法

### PRD 和 Spec 有什么区别？

PRD 偏产品表达，说明做什么和怎么用；Spec 偏可实现和可验收，说明系统行为、边界和标准。
