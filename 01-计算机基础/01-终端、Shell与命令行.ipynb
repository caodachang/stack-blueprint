# 终端、Shell与命令行


## 三者是什么关系

| | 是什么 | 例子 |
|------|------|------|
| **终端** | 一个窗口程序，负责显示文字和接收键盘输入 | Windows 终端、iTerm2、VS Code 的 Terminal 面板 |
| **Shell** | 终端里面跑的程序，负责理解你输入的命令并执行 | cmd、 bash、zsh、PowerShell |
| **命令行** | 你跟 Shell 交流的方式——用文字输入命令，而不是点鼠标 | `git commit`、`ls -la` |

你打开终端 → 终端自动启动一个 Shell → 你在 Shell 里用命令行输入指令。终端是壳，Shell 是大脑，命令行是语言。

你看到的那行 `$` 或 `>` 叫提示符——Shell 在等你输入。你输入命令、按回车、Shell 执行、输出结果、再显示提示符。这个循环叫 REPL（Read → Evaluate → Print → Loop）。

常见的 Shell：

| Shell | 提示符 | 在哪 |
|------|------|------|
| bash | `$` | Linux/macOS 默认，Windows 的 Git Bash |
| zsh | `%` 或 `$` | macOS 新默认 |
| PowerShell | `PS>` | Windows 自带 |
| cmd | `>` | Windows 老终端 |


## 两个基础命令

`echo`——把后面的文字打印出来。

```bash
echo hello          # 输出 hello
```

`which`——查一个命令在磁盘上的位置。

```bash
which python        # 输出 /usr/bin/python
```

Shell 里 `$环境变量名` 的意思是：把环境变量名替换成它的值。

```bash
echo $HOME       # HOME 是家目录，输出 /home/user
```


## PATH 环境变量

`PATH` 是一个环境变量，存着一串文件夹路径，用 `:` 分隔。Shell 按顺序在这些文件夹里搜，找到第一个就执行。

```bash
echo $PATH
# /usr/local/bin:/usr/bin:/bin  PATH里有三个路径

which python
# /usr/bin/python
```

如果没有 `PATH`，每次都得敲完整路径：

```bash
/usr/bin/python script.py   # 不用 PATH，直接指定位置
python script.py            # Shell 通过 PATH 找到 /usr/bin/python
```

环境变量不止 `PATH`。还有`HOME` 是你家目录，`USER` 是当前用户名等：

```bash
echo $HOME    # /home/user
echo $USER    # user
```


## 重定向：把输出换个地方

每个命令运行时有三个通道：

| 通道 | 名字 | 默认连到 |
|------|------|---------|
| stdin | 标准输入 | 键盘 |
| stdout | 标准输出 | 屏幕 |
| stderr | 标准错误 | 屏幕 |

重定向就是把这些通道接到别的地方。符号含义：

| 符号 | 作用 |
|------|------|
| `>` | 把 stdout 写到文件（覆盖） |
| `>>` | 把 stdout 追加到文件末尾 |
| `<` | 从文件读取作为 stdin |
| `2>` | 把 stderr 写到文件 |
| `\|` | 把左边命令的 stdout 变成右边命令的 stdin |

```bash
echo hello > file.txt       # stdout 写到 file.txt（覆盖原内容）
echo world >> file.txt      # stdout 追加到 file.txt 末尾
cat < file.txt              # 把 file.txt 的内容作为 cat 的输入
ls nonexist 2> error.log    # 这个命令会报错，错误信息存到 error.log
```

管道把多个命令串成流水线，上一个命令的输出直接变成下一个命令的输入：

```bash
cat log.txt | grep ERROR | wc -l
# cat 读日志 → grep 筛出含 ERROR 的行 → wc -l 数有多少行含 ERROR
```

## 脚本：把命令存下来复用

每次敲一串命令太麻烦。把命令写进一个文件，一次性执行——这就是脚本。

脚本第一行叫 shebang，告诉系统用哪个 Shell 执行：

```bash
#!/bin/bash
echo "开始备份..."
cp -r ~/project ~/backup/project-$(date +%Y%m%d)
echo "备份完成"
```

给脚本加执行权限然后运行：

```bash
chmod +x backup.sh   # 加执行权限
./backup.sh          # ./ 告知这个命令是当前目录下的，不用去 PATH
```
