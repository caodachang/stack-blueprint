大量软件依赖这些运行库。删了哪个都可能导致某个软件打不开。全部保留。

| 名称 |
|------|
| Microsoft Visual C++ 2008 Redistributable x64 |
| Microsoft Visual C++ 2015-2022 Redistributable x86 |
| Microsoft Visual C++ v14 Redistributable x64 |
