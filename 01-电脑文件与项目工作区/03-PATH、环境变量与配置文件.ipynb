# PATH、环境变量与配置文件

## 环境变量

环境变量是程序运行时能看到的外部配置。

常见环境变量：

| 名称 | 常见含义 |
| --- | --- |
| `PATH` | 命令搜索路径 |
| `HOME` | 用户主目录 |
| `TEMP` | 临时文件目录 |

程序可以根据环境变量决定行为。

>开发环境（开发者本地或测试用环境）和生产环境（真实用户使用的线上环境）连接不同的数据库

## PATH 

`PATH` 重要的环境变量之一，用来告诉 Shell 去哪些目录里寻找命令。

## 配置文件

配置文件是保存程序设置的文件。

| 格式 | 特点 |
| --- | --- |
| `.json` |  结构严格，前端常见 |
| `.yml` / `.yaml` | 层级清晰，缩进敏感 |
| `.toml` | Python 生态常见，适合项目配置 |
| `.env` | 特殊配置文件，存放项目本地运行需要的的环境变量 |
