# 终端、Shell与命令行

## 终端，Shell 是什么

终端是承载命令行交互的窗口，Shell 解释并执行命令的程序。

| Shell | 常见位置 |
| --- | --- |
| PowerShell | Windows 常用，也可跨平台 |
| CMD | Windows 传统命令解释器 |
| Bash | Linux/macOS 常见 |
| Zsh | macOS 新版本常见 |

## CLI是什么

- **CLI**：命令行界面
- **GLI**：图形用户界面

## 命令输出与错误

命令执行后的输出大致分三类：

- 标准输出（stdout）命令正常输出的内容
- 标准错误（stderr）错误、警告等诊断信息
- 退出码（exit code）命令是否成功的状态码
