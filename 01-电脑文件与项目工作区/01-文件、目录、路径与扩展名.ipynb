# 文件、目录、路径与扩展名

## 路径写法差异

- Windows 有盘符，常用反斜杠 `\`
- macOS/Linux 以及大多数开发工具和配置文件使用正斜杠 `/`。

## 扩展名

扩展名用来提示文件类型，影响系统默认用什么程序打开

| 类型 | 扩展名 | 常见含义 |
| --- | --- | --- |
| 网页 | `.html` | HTML 页面结构 |
| 样式 | `.css` | CSS 样式 |
| JavaScript | `.js` | JavaScript 脚本 |
| TypeScript | `.ts` | TypeScript 源码 |
| React | `.jsx`、`.tsx` | React 组件文件 |
| Python | `.py` | Python 源码 |
| Notebook | `.ipynb` | Jupyter Notebook |
| JSON | `.json` | 结构化数据或配置 |
| YAML | `.yml`、`.yaml` | 配置文件 |
| TOML | `.toml` | 配置文件 |
| 环境变量 | `.env` | 本地环境配置 |
| Markdown | `.md` | 文档 |
| 文本 | `.txt` | 普通文本 |
| 图片 | `.png`、`.jpg`、`.jpeg`、`.gif`、`.webp`、`.svg` | 图片资源 |
| 字体 | `.ttf`、`.otf`、`.woff`、`.woff2` | 字体文件 |
| 压缩包 | `.zip`、`.tar`、`.gz`、`.tar.gz` | 压缩或打包文件 |
| 数据库 | `.db`、`.sqlite` | 本地数据库文件 |
| 日志 | `.log` | 程序运行日志 |
| 可执行文件 | `.exe`、`.bat`、`.sh` | 可运行脚本或程序 |
| 锁文件 | `.lock` | 依赖版本锁定文件 |
