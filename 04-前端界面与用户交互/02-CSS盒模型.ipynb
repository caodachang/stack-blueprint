# CSS盒模型

## CSS 

CSS 是层叠样式表。用来做视觉样式

层叠：多个样式规则可能同时作用在同一个元素上，浏览器会根据来源、选择器优先级、书写顺序等规则决定最终样式。

```css
.card {
  color: #222;
  background: white;
  padding: 16px;
  border: 1px solid #ddd;
}
```

`.` 是选择器，表示选择 class 为 `card` 的元素。

## 盒模型

浏览器会把很多元素看成一个盒子。这个盒子从里到外包括内容、内边距、边框和外边距。

| 部分 | CSS 属性 | 含义 |
| --- | --- | --- |
| 内容 | `width`、`height` | 元素内容区域 |
| 内边距 | `padding` | 内容和边框之间的距离 |
| 边框 | `border` | 盒子的边线 |
| 外边距 | `margin` | 当前盒子和其他盒子之间的距离 |

## box-sizing

默认情况下，`width` 不一定包含 `padding` 和 `border`。这会让尺寸计算变得不直观。

很多项目会设置：

```css
* {
  box-sizing: border-box;
}
```
- *：全局通配符，所有标签统一开启 border-box
- `border-box` ：元素最终宽度包含内容、内边距和边框。

**Flexbox** 是弹性盒布局 **Grid** 是网格布局 **响应式设计**指页面能适应不同屏幕尺寸
