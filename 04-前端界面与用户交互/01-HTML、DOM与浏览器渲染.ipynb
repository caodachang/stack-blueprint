# HTML、DOM与浏览器渲染

## HTML 超文本标记语言

一个很小的 HTML 页面：

```html
<!doctype html>
<html lang="zh-CN">
  <head>
    <meta charset="utf-8">
    <title>示例页面</title>
  </head>
  <body>
    <h1>欢迎</h1>
    <p>这是一个段落。</p>
    <a href="https://example.com">访问示例网站</a>
  </body>
</html>
```

`head` 放页面元信息，例如标题、字符编码、样式引用。`body` 放用户能看到的主要内容。

## 标签和元素

标签包住内容后，会形成元素。

```html
<p>这是一个段落。</p>
```

`<p>` 是开始标签，`</p>` 是结束标签，中间的文字是内容。整个结构叫一个段落元素。

有些元素没有成对结束标签，例如图片：

```html
<img src="avatar.png" alt="用户头像">
```

`src` 和 `alt` 是属性。属性用来补充元素的信息。

语义化 HTML 指用合适的标签表达内容含义

## DOM

DOM 是文档对象模型。

HTML 是写在文件里的标记，DOM 是浏览器解析后在内存里形成的页面结构。

## DOM 树

这段 HTML：

```html
<body>
  <h1>标题</h1>
  <p>一段文字</p>
</body>
```

浏览器会把它理解成类似这样的层级：body 下面有 h1 和 p，h1 里有文本“标题”，p 里有文本“一段文字”。

这种层级结构就是 DOM 树

## 浏览器渲染

浏览器渲染指浏览器把 HTML、CSS、图片、字体等资源处理成用户在屏幕上看到的页面。

CSS 是层叠样式表，用来控制页面样式

## HTML、DOM 和渲染的关系

浏览器解析 HTML 得到 DOM，浏览器再结合 CSS 和其他资源，把 DOM 对应的页面画到屏幕上。

JavaScript 是浏览器里最重要的编程语言。浏览器给 JS 提供了 web API

JavaScript 修改页面时，是直接修改当前页面内存里的 DOM

HTML 描述结构，CSS 控制样式，JavaScript 负责让页面产生行为

JS 也能通过 Node.js 运行在服务器和命令行环境里
