# Tailwind与前端CSS工程化

## CSS 工程化

CSS 工程化是一组实践：命名规范、组件化、设计令牌、预处理器、CSS Modules、原子化 CSS、构建压缩等

## Tailwind

Tailwind 是一个 CSS 工程化框架

它鼓励先提供很多小而明确的工具类，直接组合在 HTML 或组件里。

```html
<button class="rounded bg-blue-600 px-4 py-2 text-white">
  保存
</button>
```

`bg-blue-600` 控制背景色，`px-4` 控制左右内边距，`py-2` 控制上下内边距，`text-white` 控制文字颜色。

## 配置和设计约束

Tailwind 项目通常会有配置文件，用来扩展主题、扫描文件、配置插件。

```js
export default {
  content: ["./index.html", "./src/**/*.{js,ts,jsx,tsx}"],
  theme: {
    extend: {}
  },
  plugins: []
}
```

`content` 告诉 Tailwind 去哪些文件里找 class。`theme` 用来定义颜色、间距、字体等设计规则。

## 前端 CSS 工程化还包括什么

除了 Tailwind，前端项目还可能使用下列 CSS 组织方式。

| 方式 | 作用 |
| --- | --- |
| BEM | 用命名规范降低类名混乱 |
| Sass / Less | 提供变量、嵌套、函数等能力 |
| CSS Modules | 让 CSS 类名局部化，减少冲突 |
| CSS-in-JS | 在 JavaScript 里组织组件样式 |
| Design Tokens | 把颜色、字号、间距等设计值抽成统一变量 |
